# FILE HANDLING + INPUT / OUTPUT IN PYTHON
# COMPLETE LEARNING GUIDE

```python
# ============================================================
# FILE HANDLING + INPUT / OUTPUT IN PYTHON
# COMPLETE LEARNING GUIDE
# ============================================================
#
# This guide builds a practical mental model of:
#
#     input → process → output
#
# and then extends it to:
#
#     program
#        ↓
#     input
#        ↓
#     memory
#        ↓
#     file / data source
#        ↓
#     processing
#        ↓
#     output
#
# The goal is not merely to memorize file methods.
# The goal is to understand how Python moves data between:
#
#     keyboard
#     screen
#     variables
#     text files
#     binary files
#     CSV
#     JSON
#     directories
#     programs
#     external data
#
# This is especially important for:
#
#     Data Analysis
#     Data Cleaning
#     Automation
#     Machine Learning
#     ETL
#     Research
#     Logging
#     Configuration
#     Report generation
#
#
# ============================================================
# MASTER MAP
# ============================================================
#
#                    PYTHON PROGRAM
#                         │
#             ┌───────────┴───────────┐
#             ↓                       ↓
#          INPUT                    OUTPUT
#             │                       │
#     ┌───────┼────────┐      ┌───────┼────────┐
#     ↓       ↓        ↓      ↓       ↓        ↓
# keyboard  file    another  screen  file    another
#                    program
#
#
# FILE HANDLING:
#
#     Path
#       ↓
#     open()
#       ↓
#     file object
#       ↓
#     read / write / seek / tell
#       ↓
#     close()
#
#
# SAFE FILE WORK:
#
#     with open(...) as file:
#                 │
#                 ↓
#          use the file
#                 │
#                 ↓
#          automatic close
#
#
# ============================================================
# 1. THE MOST IMPORTANT MENTAL MODEL
# ============================================================
#
# Think of a file as persistent storage.
#
# A normal Python variable exists while the program is running:
#
#     name = "Ali"
#
# A file can preserve data after the program ends:
#
#     name = "Ali"
#     ↓
#     save to file
#     ↓
#     program ends
#     ↓
#     program starts again
#     ↓
#     read file
#     ↓
#     "Ali"
#
#
# MEMORY VS FILE
# ------------------------------------------------------------
#
#     RAM / Python memory
#
#     name ─────→ "Ali"
#
#     Temporary during program execution
#
#
#     FILE
#
#     name.txt
#     ┌──────────────┐
#     │ Ali          │
#     └──────────────┘
#
#     Persistent storage
#
#
# The central difference:
#
#     variable → temporary program state
#     file     → persistent external state
#
#
# ============================================================
# 2. INPUT → PROCESS → OUTPUT
# ============================================================
#
# Almost every program can be understood as:
#
#              INPUT
#                ↓
#             PROCESS
#                ↓
#              OUTPUT
#
#
# Example:
#
#     keyboard
#        ↓
#     age = input(...)
#        ↓
#     convert to int
#        ↓
#     calculate age next year
#        ↓
#     print result
#
#
# Data-analysis version:
#
#     CSV file
#         ↓
#     read data
#         ↓
#     clean data
#         ↓
#     calculate statistics
#         ↓
#     save results
#
#
# ============================================================
# 3. STANDARD INPUT / OUTPUT / ERROR
# ============================================================
#
# Python programs traditionally work with three standard streams:
#
#     stdin   → standard input
#     stdout  → standard output
#     stderr  → standard error
#
#
#                    PROGRAM
#                       │
#            ┌──────────┼──────────┐
#            ↓          ↓          ↓
#          stdin      stdout     stderr
#            ↓          ↓          ↓
#        keyboard     screen     screen
#
#
# Common Python functions:
#
#     input()   → stdin
#     print()   → stdout
#
# Errors/exceptions are separate from normal program output.
#
#
# ============================================================
# 4. input()
# ============================================================
#
# Basic syntax:
#
#     input()
#
# or:
#
#     input("Prompt: ")
#
#
# IMPORTANT:
# input() ALWAYS returns a string.
#
#     age = input("Age: ")
#
# If the user enters:
#
#     35
#
# Python receives:
#
#     "35"
#
# NOT:
#
#     35
#
#
# Therefore:
#
#     age = int(input("Age: "))
#
#
# FLOW:
#
#     keyboard
#        ↓
#     "35"
#        ↓
#     input()
#        ↓
#     int()
#        ↓
#     35
#
#
# ============================================================
# 5. input() EXAMPLES
# ============================================================
#
# name = input("Enter your name: ")
#
# age = int(input("Enter your age: "))
#
# height = float(input("Enter height in meters: "))
#
# print(name)
# print(age)
# print(height)
#
#
# ============================================================
# 6. input() AND CLEANING
# ============================================================
#
# User input may contain spaces.
#
#     name = input("Name: ").strip()
#
# Example:
#
#     input:
#         "   Yasin   "
#
#     result:
#         "Yasin"
#
#
# Numeric input:
#
#     value = input("Enter value: ").strip()
#
#     try:
#         value = float(value)
#     except ValueError:
#         value = None
#
#
# This is a basic data-cleaning pattern:
#
#     RAW INPUT
#         ↓
#     strip()
#         ↓
#     validate
#         ↓
#     convert
#         ↓
#     usable value
#
#
# ============================================================
# 7. print()
# ============================================================
#
# Basic syntax:
#
#     print(value)
#
# Examples:
#
#     print("Hello")
#     print(25)
#     print(3.14)
#     print(True)
#
#
# Multiple values:
#
#     print("Age:", 35)
#
# Output:
#
#     Age: 35
#
#
# ============================================================
# 8. print() PARAMETERS
# ============================================================
#
# Important parameters:
#
#     print(*objects, sep=' ', end='\n', file=None, flush=False)
#
#
# The most useful:
#
#     sep
#     end
#     file
#     flush
#
#
# ------------------------------------------------------------
# sep
# ------------------------------------------------------------
#
#     print("A", "B", "C")
#
# Output:
#
#     A B C
#
# Default separator:
#
#     sep=" "
#
#
# Custom:
#
#     print("A", "B", "C", sep=",")
#
# Output:
#
#     A,B,C
#
#
# Another:
#
#     print("2026", "09", "10", sep="-")
#
# Output:
#
#     2026-09-10
#
#
# ------------------------------------------------------------
# end
# ------------------------------------------------------------
#
# Default:
#
#     end="\n"
#
# "\n" means newline.
#
#
#     print("Hello")
#     print("World")
#
# Output:
#
#     Hello
#     World
#
#
# Custom:
#
#     print("Hello", end=" ")
#     print("World")
#
# Output:
#
#     Hello World
#
#
# ============================================================
# 9. PRINTING TABLE-LIKE OUTPUT
# ============================================================
#
# Example:
#
#     print("Name", "Score", sep="\t")
#     print("Ali", 88, sep="\t")
#     print("Sara", 94, sep="\t")
#
#
# Better for controlled formatting:
#
#     print(f"{'Name':<10}{'Score':>10}")
#     print(f"{'Ali':<10}{88:>10}")
#     print(f"{'Sara':<10}{94:>10}")
#
#
# ============================================================
# 10. f-STRINGS
# ============================================================
#
# f-strings are one of the most useful output tools.
#
#     name = "Yasin"
#     age = 35
#
#     print(f"My name is {name}.")
#     print(f"I am {age} years old.")
#
#
# Numeric formatting:
#
#     score = 91.4567
#
#     print(f"{score:.2f}")
#
# Output:
#
#     91.46
#
#
# Percentage:
#
#     accuracy = 0.9134
#     print(f"{accuracy:.2%}")
#
# Output:
#
#     91.34%
#
#
# Thousands separator:
#
#     sales = 1250000
#     print(f"{sales:,}")
#
# Output:
#
#     1,250,000
#
#
# ============================================================
# 11. FILES: THE CORE IDEA
# ============================================================
#
# Python does not normally manipulate the physical file directly.
#
# Instead:
#
#     path
#       ↓
#     open()
#       ↓
#     file object
#       ↓
#     methods
#
#
# Example:
#
#     file = open("data.txt", "r")
#
#
# Here:
#
#     "data.txt" → path
#     "r"        → mode
#     file       → file object
#
#
# ============================================================
# 12. open()
# ============================================================
#
# Basic syntax:
#
#     open(file, mode="r", encoding=None, ...)
#
#
# Most important modes:
#
#     r   read
#     w   write / overwrite
#     a   append
#     x   create new file
#
#
# Binary:
#
#     rb
#     wb
#     ab
#
#
# Main table:
#
#     +------+-------------------------------------------+
#     | Mode | Meaning                                   |
#     +------+-------------------------------------------+
#     | r    | Read existing file                        |
#     | w    | Write; overwrite/create                   |
#     | a    | Append; create if needed                  |
#     | x    | Create; fail if already exists            |
#     | rb   | Read binary                               |
#     | wb   | Write binary                              |
#     | ab   | Append binary                             |
#     +------+-------------------------------------------+
#
#
# VERY IMPORTANT:
#
#     "w" can destroy existing contents.
#
#
# ============================================================
# 13. FILE OPENING FLOW
# ============================================================
#
#     open("data.txt", "r")
#              │
#              ↓
#        operating system
#              │
#              ↓
#          file object
#              │
#       ┌──────┼──────┐
#       ↓      ↓      ↓
#     read   seek   close
#
#
# ============================================================
# 14. ALWAYS PREFER with open()
# ============================================================
#
# Less safe:
#
#     file = open("data.txt", "r")
#     data = file.read()
#     file.close()
#
#
# Safer:
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         data = file.read()
#
#
# Mental model:
#
#     with open(...)
#             ↓
#       acquire resource
#             ↓
#       use resource
#             ↓
#       leave block
#             ↓
#       automatically close
#
#
# This is called a context manager.
#
#
# ============================================================
# 15. WHY with MATTERS
# ============================================================
#
# If an exception occurs:
#
#     with open(...) as file:
#         risky_operation()
#
# Python still performs the cleanup when leaving the context.
#
# General principle:
#
#     acquire
#       ↓
#     use
#       ↓
#     release
#
#
# Examples beyond files:
#
#     with open(...)
#     with lock
#     with database connection
#     with some_resource
#
#
# ============================================================
# 16. FILE OBJECT METHODS — MASTER TABLE
# ============================================================
#
#     +----------------+--------------------------------+
#     | Method         | Purpose                        |
#     +----------------+--------------------------------+
#     | read()         | Read text                      |
#     | readline()     | Read one line                  |
#     | readlines()    | Read all lines into a list      |
#     | write()        | Write a string                 |
#     | writelines()   | Write multiple strings         |
#     | seek()         | Move file position             |
#     | tell()         | Get current position            |
#     | flush()        | Push buffered output           |
#     | close()        | Close file                     |
#     +----------------+--------------------------------+
#
#
# Useful properties:
#
#     file.name
#     file.mode
#     file.closed
#
#
# ============================================================
# 17. read()
# ============================================================
#
# Read the entire file:
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         text = file.read()
#
#     print(text)
#
#
# Mental model:
#
#     FILE
#     ┌──────────────────────┐
#     │ line 1               │
#     │ line 2               │
#     │ line 3               │
#     └──────────────────────┘
#                ↓
#             read()
#                ↓
#        one large string
#
#
# Good for:
#
#     small/medium text files
#
# Potential problem:
#
#     huge file → entire content loaded into memory
#
#
# ============================================================
# 18. read(size)
# ============================================================
#
# You can request a limited number of characters/bytes.
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         chunk = file.read(100)
#
#
# Concept:
#
#     FILE
#     ┌──────────────────────────────┐
#     │ ABCDEFGHIJKLMNOPQRSTUVWXYZ...│
#     └──────────────────────────────┘
#                ↓
#             read(5)
#                ↓
#               ABCDE
#
#
# Useful for:
#
#     large files
#     chunk processing
#     streaming-style work
#
#
# ============================================================
# 19. readline()
# ============================================================
#
# Read one line at a time.
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         line1 = file.readline()
#         line2 = file.readline()
#
#
# If file contains:
#
#     Ali
#     Sara
#     Reza
#
# First call:
#
#     "Ali\n"
#
# Second call:
#
#     "Sara\n"
#
#
# ============================================================
# 20. readlines()
# ============================================================
#
# Reads all lines and returns a list.
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         lines = file.readlines()
#
# Result:
#
#     [
#         "Ali\n",
#         "Sara\n",
#         "Reza\n"
#     ]
#
#
# WARNING:
#
#     readlines() loads all lines into memory.
#
# For large files, prefer:
#
#     for line in file:
#         ...
#
#
# ============================================================
# 21. THE BEST BASIC WAY TO READ A TEXT FILE
# ============================================================
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         for line in file:
#             print(line.strip())
#
#
# Why this is powerful:
#
#     file
#       ↓
#     one line
#       ↓
#     process
#       ↓
#     next line
#       ↓
#     process
#       ↓
#     ...
#
#
# You do not need to load the whole file into memory.
#
#
# ============================================================
# 22. NEWLINE PROBLEM
# ============================================================
#
# A text file often stores:
#
#     "Ali\n"
#
# "\n" represents a newline.
#
# Therefore:
#
#     print(line)
#
# can produce an extra blank line because:
#
#     line already contains "\n"
#
#
# Use:
#
#     print(line.strip())
#
# or:
#
#     print(line, end="")
#
#
# Difference:
#
#     strip()
#         removes whitespace at both ends
#
#     rstrip("\n")
#         specifically removes newline characters
#
#
# For data cleaning, be careful:
#
#     strip()
#
# removes spaces/tabs as well as newline characters.
#
#
# ============================================================
# 23. WRITE MODE
# ============================================================
#
#     with open("output.txt", "w", encoding="utf-8") as file:
#         file.write("Hello\n")
#         file.write("World\n")
#
#
# Result:
#
#     Hello
#     World
#
#
# IMPORTANT:
#
#     "w" replaces existing content.
#
#
# Diagram:
#
#     old file
#     ┌─────────────────┐
#     │ old data        │
#     └─────────────────┘
#             ↓
#          mode "w"
#             ↓
#     ┌─────────────────┐
#     │ new data        │
#     └─────────────────┘
#
#
# ============================================================
# 24. write()
# ============================================================
#
# write() expects a string in text mode.
#
#     file.write("Hello")
#
# Correct:
#
#     file.write(str(123))
#
# Or use f-strings:
#
#     file.write(f"Score: {score}\n")
#
#
# IMPORTANT:
#
# write() does NOT automatically add "\n".
#
#     file.write("A")
#     file.write("B")
#
# gives:
#
#     AB
#
# while:
#
#     file.write("A\n")
#     file.write("B\n")
#
# gives:
#
#     A
#     B
#
#
# ============================================================
# 25. writelines()
# ============================================================
#
#     lines = [
#         "Ali\n",
#         "Sara\n",
#         "Reza\n"
#     ]
#
#     with open("names.txt", "w", encoding="utf-8") as file:
#         file.writelines(lines)
#
#
# IMPORTANT:
#
# writelines() does NOT add newline characters.
#
# Therefore:
#
#     ["Ali", "Sara", "Reza"]
#
# produces:
#
#     AliSaraReza
#
# unless newline characters are included.
#
#
# ============================================================
# 26. APPEND MODE
# ============================================================
#
#     with open("log.txt", "a", encoding="utf-8") as file:
#         file.write("New event\n")
#
#
# "a" means:
#
#     keep existing content
#          ↓
#     move to end
#          ↓
#     write new content
#
#
# Diagram:
#
#     existing:
#
#     A
#     B
#     C
#
#          ↓ append "D"
#
#     A
#     B
#     C
#     D
#
#
# Common applications:
#
#     logs
#     history
#     audit records
#     incremental results
#
#
# ============================================================
# 27. CREATE MODE
# ============================================================
#
#     with open("new_file.txt", "x", encoding="utf-8") as file:
#         file.write("Created")
#
#
# "x" means:
#
#     create file
#     BUT
#     fail if it already exists
#
#
# Useful when you do NOT want accidental overwriting.
#
#
# ============================================================
# 28. FILE MODE COMPARISON
# ============================================================
#
#     +------+----------+-------------------------------+
#     | Mode | Exists?  | Behavior                      |
#     +------+----------+-------------------------------+
#     | r    | yes      | read                          |
#     | r    | no       | FileNotFoundError             |
#     | w    | yes      | overwrite                     |
#     | w    | no       | create                        |
#     | a    | yes      | append                        |
#     | a    | no       | create                        |
#     | x    | yes      | FileExistsError               |
#     | x    | no       | create                        |
#     +------+----------+-------------------------------+
#
#
# ============================================================
# 29. FILE POSITION / CURSOR
# ============================================================
#
# A file object maintains a current position.
#
# Imagine:
#
#     A B C D E F G
#     ^
#     position = 0
#
# After reading 3 characters:
#
#     A B C | D E F G
#             ^
#             position = 3
#
#
# The file cursor determines where the next operation occurs.
#
#
# ============================================================
# 30. tell()
# ============================================================
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         print(file.tell())
#         file.read(5)
#         print(file.tell())
#
#
# tell() reports the current stream position.
#
#
# Conceptually:
#
#     tell()
#        ↓
#     "Where am I?"
#
#
# ============================================================
# 31. seek()
# ============================================================
#
#     file.seek(position)
#
# It changes the current position.
#
#
# Example:
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         file.read(5)
#         file.seek(0)
#         text = file.read()
#
#
# Diagram:
#
#     A B C D E F G
#     ─────────────
#     ^
#     0
#
#     read(5)
#
#     A B C D E | F G
#               ^
#
#     seek(0)
#
#     A B C D E F G
#     ^
#
#
# ============================================================
# 32. seek() COMMON USES
# ============================================================
#
#     file.seek(0)
#
# → return to beginning.
#
#
# Useful for:
#
#     rereading
#     random access
#     binary files
#     controlled file processing
#
#
# For ordinary line-by-line data processing, you often do not
# need seek().
#
#
# ============================================================
# 33. close()
# ============================================================
#
# Explicit:
#
#     file.close()
#
# Check:
#
#     file.closed
#
#
# But normally:
#
#     with open(...) as file:
#         ...
#
# is preferable.
#
#
# ============================================================
# 34. FILE PROPERTIES
# ============================================================
#
#     with open("data.txt", "r", encoding="utf-8") as file:
#         print(file.name)
#         print(file.mode)
#         print(file.closed)
#
#
# Example:
#
#     data.txt
#     r
#     False
#
# After leaving the with block:
#
#     file.closed
#
# becomes:
#
#     True
#
#
# ============================================================
# 35. ENCODING
# ============================================================
#
# Text is stored as bytes.
#
# Python converts between:
#
#     text
#       ↕
#     encoding
#       ↕
#     bytes
#
#
# Common encoding:
#
#     UTF-8
#
#
# Recommended:
#
#     open(
#         "data.txt",
#         "r",
#         encoding="utf-8"
#     )
#
#
# Why specify encoding?
#
# Because text may contain:
#
#     Persian
#     Arabic
#     Chinese
#     Japanese
#     accented characters
#     symbols
#
#
# Example:
#
#     text = "سلام"
#
#
# ============================================================
# 36. TEXT VS BYTES
# ============================================================
#
# TEXT MODE:
#
#     "hello"
#        ↓
#     str
#
#
# BINARY MODE:
#
#     b"hello"
#        ↓
#     bytes
#
#
# Diagram:
#
#     human-readable text
#            ↓
#        encoding
#            ↓
#           bytes
#            ↓
#         storage
#
#
# Text:
#
#     open("file.txt", "r", encoding="utf-8")
#
# Binary:
#
#     open("file.bin", "rb")
#
#
# ============================================================
# 37. BINARY FILES
# ============================================================
#
# Binary files include:
#
#     images
#     audio
#     video
#     PDF files
#     executable files
#     compressed files
#     many scientific formats
#
#
# Example:
#
#     with open("image.jpg", "rb") as file:
#         data = file.read()
#
#
# data is:
#
#     bytes
#
#
# ============================================================
# 38. COPYING A BINARY FILE
# ============================================================
#
#     with open("source.bin", "rb") as source:
#         data = source.read()
#
#     with open("copy.bin", "wb") as destination:
#         destination.write(data)
#
#
# For very large files, process in chunks:
#
#     with open("source.bin", "rb") as source:
#         with open("copy.bin", "wb") as destination:
#             while chunk := source.read(1024 * 1024):
#                 destination.write(chunk)
#
#
# This is a useful real-world pattern:
#
#     large file
#         ↓
#     read chunk
#         ↓
#     write chunk
#         ↓
#     read next chunk
#         ↓
#     ...
#
#
# ============================================================
# 39. PATHS
# ============================================================
#
# A path tells Python where a file is located.
#
# Relative path:
#
#     "data.csv"
#
#     "data/results.csv"
#
#
# Absolute path:
#
#     "C:/Users/Yasin/Documents/data.csv"
#
#
# On Windows:
#
#     "C:\\Users\\Yasin\\Documents\\data.csv"
#
#
# Better:
#
#     "C:/Users/Yasin/Documents/data.csv"
#
# or pathlib:
#
#     Path("C:/Users/Yasin/Documents/data.csv")
#
#
# ============================================================
# 40. pathlib — MODERN PATH HANDLING
# ============================================================
#
# Import:
#
#     from pathlib import Path
#
#
# Create path:
#
#     path = Path("data") / "results.csv"
#
#
# This is better than manually constructing paths:
#
#     "data" + "/" + "results.csv"
#
#
# Diagram:
#
#     Path("data")
#          +
#     "results.csv"
#          ↓
#     data/results.csv
#
#
# pathlib works across operating systems.
#
#
# ============================================================
# 41. pathlib MASTER METHODS / PROPERTIES
# ============================================================
#
#     Path("data.txt")
#
# Useful:
#
#     path.exists()
#     path.is_file()
#     path.is_dir()
#     path.name
#     path.stem
#     path.suffix
#     path.parent
#     path.parts
#
#
# Example:
#
#     path = Path("data/results.csv")
#
#     path.name
#         → "results.csv"
#
#     path.stem
#         → "results"
#
#     path.suffix
#         → ".csv"
#
#     path.parent
#         → Path("data")
#
#
# ============================================================
# 42. PATH TABLE
# ============================================================
#
#     +------------------+--------------------------------+
#     | Expression       | Meaning                        |
#     +------------------+--------------------------------+
#     | exists()         | Does it exist?                 |
#     | is_file()        | Is it a file?                  |
#     | is_dir()         | Is it a directory?             |
#     | name             | Final name                     |
#     | stem             | Name without extension         |
#     | suffix           | Extension                      |
#     | parent           | Parent directory               |
#     | parts            | Path components                |
#     +------------------+--------------------------------+
#
#
# ============================================================
# 43. CREATE DIRECTORIES
# ============================================================
#
#     from pathlib import Path
#
#     output_dir = Path("output")
#     output_dir.mkdir(exist_ok=True)
#
#
# Nested directories:
#
#     output_dir.mkdir(parents=True, exist_ok=True)
#
#
# Meaning:
#
#     parents=True
#         create missing parent directories.
#
#     exist_ok=True
#         do not fail if directory already exists.
#
#
# ============================================================
# 44. LIST FILES
# ============================================================
#
#     from pathlib import Path
#
#     folder = Path("data")
#
#     for path in folder.iterdir():
#         print(path)
#
#
# Filter files:
#
#     for path in folder.iterdir():
#         if path.is_file():
#             print(path)
#
#
# ============================================================
# 45. glob()
# ============================================================
#
# Find matching files:
#
#     list(folder.glob("*.csv"))
#
#
# Examples:
#
#     "*.csv"
#     "*.txt"
#     "*.json"
#     "data_*.csv"
#
#
# Recursive:
#
#     folder.rglob("*.csv")
#
#
# Diagram:
#
#     data/
#     ├── a.csv
#     ├── b.csv
#     ├── c.txt
#     └── sub/
#         └── d.csv
#
#     glob("*.csv")
#         ↓
#     a.csv
#     b.csv
#
#     rglob("*.csv")
#         ↓
#     a.csv
#     b.csv
#     d.csv
#
#
# ============================================================
# 46. DELETE / RENAME / MOVE
# ============================================================
#
# Path operations:
#
#     path.unlink()
#         → delete file
#
#     path.rename(new_path)
#         → rename/move
#
#
# Directory:
#
#     path.rmdir()
#
# Only use deletion methods when you are certain.
#
# Safer pattern:
#
#     if path.exists():
#         path.unlink()
#
#
# ============================================================
# 47. FILE READING — COMPLETE PATTERN
# ============================================================
#
#     from pathlib import Path
#
#     path = Path("data.txt")
#
#     with path.open("r", encoding="utf-8") as file:
#         text = file.read()
#
#
# pathlib also provides:
#
#     text = path.read_text(encoding="utf-8")
#
#
# and:
#
#     path.write_text(
#         "Hello",
#         encoding="utf-8"
#     )
#
#
# These are convenient for simple text files.
#
#
# ============================================================
# 48. read_text() AND write_text()
# ============================================================
#
# Simple read:
#
#     from pathlib import Path
#
#     text = Path("data.txt").read_text(
#         encoding="utf-8"
#     )
#
#
# Simple write:
#
#     Path("output.txt").write_text(
#         "Hello\nWorld",
#         encoding="utf-8"
#     )
#
#
# Use these when you want:
#
#     entire text ↔ file
#
# For more control:
#
#     open()
#
# is still fundamental.
#
#
# ============================================================
# 49. FILE ITERATION
# ============================================================
#
# A file object is iterable.
#
# Therefore:
#
#     with open("data.txt", encoding="utf-8") as file:
#         for line in file:
#             print(line.strip())
#
#
# Mental model:
#
#     file
#       ↓
#     iterator
#       ↓
#     line 1
#       ↓
#     line 2
#       ↓
#     line 3
#       ↓
#     ...
#
#
# This is one reason Python file handling works naturally
# with loops.
#
#
# ============================================================
# 50. READING ONLY VALID LINES
# ============================================================
#
#     with open("data.txt", encoding="utf-8") as file:
#         for line in file:
#             line = line.strip()
#
#             if not line:
#                 continue
#
#             print(line)
#
#
# Pattern:
#
#     read
#       ↓
#     clean
#       ↓
#     validate
#       ↓
#     process
#
#
# ============================================================
# 51. WRITING MULTIPLE RECORDS
# ============================================================
#
#     names = ["Ali", "Sara", "Reza"]
#
#     with open("names.txt", "w", encoding="utf-8") as file:
#         for name in names:
#             file.write(name + "\n")
#
#
# Data flow:
#
#     Python list
#          ↓
#     loop
#          ↓
#     one record
#          ↓
#     string
#          ↓
#     file
#
#
# ============================================================
# 52. CSV — WHY IT MATTERS
# ============================================================
#
# CSV = Comma-Separated Values.
#
# Example:
#
#     name,age,score
#     Ali,25,88
#     Sara,31,94
#     Reza,29,81
#
#
# Conceptually:
#
#     CSV file
#          ↓
#     rows
#          ↓
#     columns
#
#
# CSV is extremely important in:
#
#     Data Analysis
#     Machine Learning
#     Research
#     Excel workflows
#     ETL
#     reporting
#
#
# ============================================================
# 53. DO NOT PARSE REAL CSV WITH split(",")
# ============================================================
#
# You may see:
#
#     line.split(",")
#
# It can work for very simple data.
#
# But real CSV can contain:
#
#     "Smith, John"
#
# inside a quoted field.
#
# Therefore:
#
#     split(",")
#
# is NOT a complete CSV parser.
#
# Use:
#
#     csv
#
# from the standard library.
#
#
# ============================================================
# 54. csv MODULE
# ============================================================
#
#     import csv
#
#
# Main tools:
#
#     csv.reader
#     csv.writer
#     csv.DictReader
#     csv.DictWriter
#
#
# Master map:
#
#     CSV
#      │
#      ├── reader      → rows
#      ├── DictReader  → dictionaries
#      ├── writer      → rows
#      └── DictWriter  → dictionaries
#
#
# ============================================================
# 55. csv.reader
# ============================================================
#
#     import csv
#
#     with open(
#         "data.csv",
#         "r",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         reader = csv.reader(file)
#
#         for row in reader:
#             print(row)
#
#
# Each row is usually a list of strings.
#
#
# CSV:
#
#     Ali,25,88
#
# becomes:
#
#     ["Ali", "25", "88"]
#
#
# IMPORTANT:
#
#     "25" is still a string.
#
# You must convert it:
#
#     age = int(row[1])
#
#
# ============================================================
# 56. DictReader
# ============================================================
#
# This is often more convenient.
#
# CSV:
#
#     name,age,score
#     Ali,25,88
#
#
# Code:
#
#     import csv
#
#     with open(
#         "data.csv",
#         "r",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         reader = csv.DictReader(file)
#
#         for row in reader:
#             print(row)
#
#
# A row behaves approximately like:
#
#     {
#         "name": "Ali",
#         "age": "25",
#         "score": "88"
#     }
#
#
# ============================================================
# 57. CSV DATA CONVERSION
# ============================================================
#
# Remember:
#
#     CSV = text
#
# Therefore:
#
#     "25" → string
#
# not:
#
#     25 → integer
#
#
# Example:
#
#     age = int(row["age"])
#     score = float(row["score"])
#
#
# Data pipeline:
#
#     CSV
#       ↓
#     strings
#       ↓
#     clean
#       ↓
#     convert
#       ↓
#     numeric/date/category values
#
#
# ============================================================
# 58. csv.writer
# ============================================================
#
#     import csv
#
#     rows = [
#         ["name", "age", "score"],
#         ["Ali", 25, 88],
#         ["Sara", 31, 94],
#     ]
#
#     with open(
#         "output.csv",
#         "w",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         writer = csv.writer(file)
#         writer.writerows(rows)
#
#
# ============================================================
# 59. writerow() VS writerows()
# ============================================================
#
#     writer.writerow(row)
#
# → one row.
#
#
#     writer.writerows(rows)
#
# → multiple rows.
#
#
# Example:
#
#     writer.writerow(["Ali", 25, 88])
#
#
# ============================================================
# 60. DictWriter
# ============================================================
#
#     import csv
#
#     fieldnames = [
#         "name",
#         "age",
#         "score"
#     ]
#
#     with open(
#         "output.csv",
#         "w",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         writer = csv.DictWriter(
#             file,
#             fieldnames=fieldnames
#         )
#
#         writer.writeheader()
#
#         writer.writerow({
#             "name": "Ali",
#             "age": 25,
#             "score": 88
#         })
#
#
# ============================================================
# 61. JSON
# ============================================================
#
# JSON = JavaScript Object Notation.
#
# Very common for:
#
#     APIs
#     configuration
#     web applications
#     machine-learning metadata
#     nested data
#
#
# Example:
#
#     {
#         "name": "Ali",
#         "age": 25,
#         "skills": ["Python", "SQL"]
#     }
#
#
# Python:
#
#     {
#         "name": "Ali",
#         "age": 25,
#         "skills": ["Python", "SQL"]
#     }
#
#
# ============================================================
# 62. JSON MAPPING
# ============================================================
#
#     JSON                 Python
#
#     object      ───────→ dict
#     array       ───────→ list
#     string      ───────→ str
#     number      ───────→ int / float
#     true        ───────→ True
#     false       ───────→ False
#     null        ───────→ None
#
#
# ============================================================
# 63. json MODULE
# ============================================================
#
#     import json
#
#
# Main functions:
#
#     json.load()
#     json.loads()
#     json.dump()
#     json.dumps()
#
#
# Very important distinction:
#
#     load   → file → Python
#     loads  → string → Python
#
#     dump   → Python → file
#     dumps  → Python → string
#
#
# ============================================================
# 64. JSON MASTER DIAGRAM
# ============================================================
#
#     JSON FILE
#         │
#         │ json.load()
#         ↓
#     Python object
#
#     Python object
#         │
#         │ json.dump()
#         ↓
#     JSON FILE
#
#
#     JSON STRING
#         │
#         │ json.loads()
#         ↓
#     Python object
#
#     Python object
#         │
#         │ json.dumps()
#         ↓
#     JSON STRING
#
#
# ============================================================
# 65. json.load()
# ============================================================
#
#     import json
#
#     with open(
#         "data.json",
#         "r",
#         encoding="utf-8"
#     ) as file:
#
#         data = json.load(file)
#
#     print(data)
#
#
# ============================================================
# 66. json.dump()
# ============================================================
#
#     import json
#
#     data = {
#         "name": "Ali",
#         "age": 25
#     }
#
#     with open(
#         "data.json",
#         "w",
#         encoding="utf-8"
#     ) as file:
#
#         json.dump(
#             data,
#             file,
#             ensure_ascii=False,
#             indent=2
#         )
#
#
# indent=2 makes the file human-readable.
#
#
# ============================================================
# 67. json.dumps()
# ============================================================
#
#     import json
#
#     data = {
#         "name": "Ali",
#         "score": 88
#     }
#
#     text = json.dumps(
#         data,
#         ensure_ascii=False
#     )
#
# print(text)
#
#
# Important:
#
#     dumps → string
#
#     dump → file
#
#
# ============================================================
# 68. JSON NESTED DATA
# ============================================================
#
#     data = {
#         "student": {
#             "name": "Ali",
#             "scores": [88, 91, 95]
#         }
#     }
#
#
# Access:
#
#     data["student"]["name"]
#
#     data["student"]["scores"]
#
#
#     data["student"]["scores"][0]
#
#
# Nested data is common in API responses.
#
#
# ============================================================
# 69. FILE ERRORS
# ============================================================
#
# File operations can fail.
#
# Common exceptions:
#
#     FileNotFoundError
#     FileExistsError
#     PermissionError
#     IsADirectoryError
#     NotADirectoryError
#     UnicodeDecodeError
#     UnicodeEncodeError
#     OSError
#
#
# Example:
#
#     try:
#         with open(
#             "missing.txt",
#             encoding="utf-8"
#         ) as file:
#             data = file.read()
#     except FileNotFoundError:
#         print("File does not exist.")
#
#
# ============================================================
# 70. NEVER USE BARE except FOR NORMAL FILE WORK
# ============================================================
#
# Avoid:
#
#     try:
#         ...
#     except:
#         ...
#
#
# Better:
#
#     except FileNotFoundError:
#         ...
#
#
# Or:
#
#     except (ValueError, TypeError):
#         ...
#
#
# Specific exceptions make debugging much easier.
#
#
# ============================================================
# 71. try / except / else / finally
# ============================================================
#
# Structure:
#
#     try:
#         risky operation
#
#     except SomeError:
#         handle error
#
#     else:
#         runs if no exception
#
#     finally:
#         runs whether error or not
#
#
# File example:
#
#     try:
#         with open(
#             "data.txt",
#             encoding="utf-8"
#         ) as file:
#             data = file.read()
#
#     except FileNotFoundError:
#         print("Missing file.")
#
#     else:
#         print("File loaded successfully.")
#
#
# ============================================================
# 72. VALIDATING A FILE BEFORE READING
# ============================================================
#
#     from pathlib import Path
#
#     path = Path("data.csv")
#
#     if not path.exists():
#         print("File does not exist.")
#
#     elif not path.is_file():
#         print("Path is not a file.")
#
#     else:
#         print("Ready to read.")
#
#
# Mental model:
#
#     path
#       ↓
#     exists?
#       ↓
#     file?
#       ↓
#     open
#
#
# ============================================================
# 73. FILE ENCODING ERRORS
# ============================================================
#
# If a file was not encoded as UTF-8, this may fail:
#
#     open(
#         "data.txt",
#         encoding="utf-8"
#     )
#
#
# Possible issue:
#
#     UnicodeDecodeError
#
#
# Do not blindly change encoding.
# First determine how the source file was actually encoded.
#
#
# ============================================================
# 74. NEWLINE PARAMETER FOR CSV
# ============================================================
#
# Recommended:
#
#     open(
#         "data.csv",
#         "r",
#         encoding="utf-8",
#         newline=""
#     )
#
#
# and:
#
#     open(
#         "data.csv",
#         "w",
#         encoding="utf-8",
#         newline=""
#     )
#
#
# The csv module is designed to manage newline handling itself.
#
#
# ============================================================
# 75. DATA CLEANING FROM A TEXT FILE
# ============================================================
#
# Suppose a file contains:
#
#     Ali
#     Sara
#
#     Reza
#       Yasin
#
#
# Basic cleaning:
#
#     names = []
#
#     with open(
#         "names.txt",
#         encoding="utf-8"
#     ) as file:
#
#         for line in file:
#             name = line.strip()
#
#             if name:
#                 names.append(name)
#
#
# Pipeline:
#
#     file
#       ↓
#     line
#       ↓
#     strip
#       ↓
#     remove empty values
#       ↓
#     list
#
#
# ============================================================
# 76. READING NUMERIC DATA
# ============================================================
#
# File:
#
#     10
#     20
#     30
#
#
# Code:
#
#     numbers = []
#
#     with open(
#         "numbers.txt",
#         encoding="utf-8"
#     ) as file:
#
#         for line in file:
#             line = line.strip()
#
#             if not line:
#                 continue
#
#             numbers.append(float(line))
#
#
# Result:
#
#     [10.0, 20.0, 30.0]
#
#
# ============================================================
# 77. SAFE NUMERIC PARSING
# ============================================================
#
# Real data may contain:
#
#     10
#     20
#     ?
#     N/A
#     30
#
#
# A robust parser:
#
#     def to_float(value):
#         value = value.strip()
#
#         if value in {"", "?", "N/A", "NA"}:
#             return None
#
#         try:
#             return float(value)
#         except ValueError:
#             return None
#
#
# This turns:
#
#     "10"  → 10.0
#     "20"  → 20.0
#     "?"   → None
#     "N/A" → None
#
#
# ============================================================
# 78. PRESERVE RAW DATA
# ============================================================
#
# Good data workflow:
#
#     raw file
#         ↓
#     read
#         ↓
#     cleaned representation
#         ↓
#     output file
#
#
# Avoid:
#
#     raw data
#        ↓
#     overwrite immediately
#
#
# Better:
#
#     raw_data.csv
#          ↓
#     cleaned_data.csv
#
#
# This gives you:
#
#     reproducibility
#     auditability
#     debugging ability
#     ability to compare before/after
#
#
# ============================================================
# 79. SIMPLE CSV CLEANING PIPELINE
# ============================================================
#
#     import csv
#
#     cleaned_rows = []
#
#     with open(
#         "raw.csv",
#         "r",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         reader = csv.DictReader(file)
#
#         for row in reader:
#
#             name = row["name"].strip()
#
#             age_text = row["age"].strip()
#
#             try:
#                 age = int(age_text)
#             except ValueError:
#                 age = None
#
#             cleaned_rows.append({
#                 "name": name,
#                 "age": age
#             })
#
#
# ============================================================
# 80. WRITE CLEANED CSV
# ============================================================
#
#     fieldnames = ["name", "age"]
#
#     with open(
#         "cleaned.csv",
#         "w",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         writer = csv.DictWriter(
#             file,
#             fieldnames=fieldnames
#         )
#
#         writer.writeheader()
#         writer.writerows(cleaned_rows)
#
#
# Complete flow:
#
#     raw.csv
#        ↓
#     DictReader
#        ↓
#     clean
#        ↓
#     convert
#        ↓
#     validate
#        ↓
#     cleaned_rows
#        ↓
#     DictWriter
#        ↓
#     cleaned.csv
#
#
# ============================================================
# 81. READING AND WRITING JSON DATA
# ============================================================
#
#     import json
#
#     with open(
#         "input.json",
#         "r",
#         encoding="utf-8"
#     ) as file:
#         data = json.load(file)
#
#
#     # Process data here
#
#
#     with open(
#         "output.json",
#         "w",
#         encoding="utf-8"
#     ) as file:
#         json.dump(
#             data,
#             file,
#             ensure_ascii=False,
#             indent=2
#         )
#
#
# ============================================================
# 82. TEXT FILE VS CSV VS JSON
# ============================================================
#
#     +----------+----------------------+----------------------+
#     | Format   | Best for             | Structure            |
#     +----------+----------------------+----------------------+
#     | TXT      | simple text          | mostly unstructured  |
#     | CSV      | tables               | rows + columns       |
#     | JSON     | nested records/APIs  | hierarchical         |
#     +----------+----------------------+----------------------+
#
#
# Think:
#
#     TXT  → text
#     CSV  → table
#     JSON → nested objects
#
#
# ============================================================
# 83. FILE FORMAT DECISION TABLE
# ============================================================
#
#     Need simple notes?
#         → TXT
#
#     Need spreadsheet-like rows?
#         → CSV
#
#     Need nested records?
#         → JSON
#
#     Need image/audio/video?
#         → binary mode / specialized library
#
#     Need large analytical tables?
#         → CSV for interchange,
#           then pandas/Parquet/database as appropriate
#
#
# ============================================================
# 84. FLUSH
# ============================================================
#
# Python may buffer output.
#
#     file.flush()
#
# asks Python to push buffered output onward.
#
#
# Usually:
#
#     with open(...) as file:
#         ...
#
# is sufficient.
#
#
# flush() becomes more relevant for:
#
#     logs
#     long-running programs
#     interactive output
#     streaming
#
#
# ============================================================
# 85. print() TO A FILE
# ============================================================
#
# print() can write to a file using:
#
#     file=
#
#
# Example:
#
#     with open(
#         "log.txt",
#         "w",
#         encoding="utf-8"
#     ) as file:
#
#         print("Program started", file=file)
#         print("Processing...", file=file)
#         print("Done", file=file)
#
#
# This is often cleaner than manually adding "\n".
#
#
# ============================================================
# 86. LOGGING IDEA
# ============================================================
#
# For serious applications, Python's:
#
#     logging
#
# module is better than many scattered:
#
#     print()
#
# calls.
#
#
# Concept:
#
#     program
#       │
#       ├── INFO
#       ├── WARNING
#       ├── ERROR
#       └── DEBUG
#             ↓
#           log file
#
#
# Basic example:
#
#     import logging
#
#     logging.basicConfig(
#         filename="app.log",
#         level=logging.INFO
#     )
#
#     logging.info("Program started")
#
#
# ============================================================
# 87. STANDARD INPUT FROM A FILE
# ============================================================
#
# Many command-line programs can receive redirected input:
#
#     python program.py < input.txt
#
#
# Conceptually:
#
#     input.txt
#         ↓
#       stdin
#         ↓
#     input()
#
#
# Python can also use:
#
#     import sys
#
#     sys.stdin
#     sys.stdout
#     sys.stderr
#
#
# ============================================================
# 88. sys.stdin / stdout / stderr
# ============================================================
#
#     import sys
#
#     for line in sys.stdin:
#         print(line.strip())
#
#
# Output:
#
#     sys.stdout
#
#
# Error output:
#
#     print(
#         "Something went wrong",
#         file=sys.stderr
#     )
#
#
# Useful for:
#
#     command-line tools
#     pipelines
#     Unix-style data processing
#     automation
#
#
# ============================================================
# 89. os MODULE VS pathlib
# ============================================================
#
# Older/common style:
#
#     import os
#
# Modern path-oriented style:
#
#     from pathlib import Path
#
#
# For paths and files, generally prefer:
#
#     pathlib
#
# Examples:
#
#     Path.cwd()
#     Path.home()
#     Path("data").glob("*.csv")
#
#
# os is still important for:
#
#     environment variables
#     process-level operations
#     lower-level operating-system interaction
#
#
# ============================================================
# 90. CURRENT WORKING DIRECTORY
# ============================================================
#
#     from pathlib import Path
#
#     print(Path.cwd())
#
#
# CWD = Current Working Directory.
#
#
# Relative paths are interpreted relative to the CWD.
#
#
# Diagram:
#
#     Current directory
#            │
#       ┌────┼────┐
#       ↓    ↓    ↓
#     data  output script
#
#
# If:
#
#     open("data.csv")
#
# Python looks for:
#
#     data.csv
#
# relative to the current working directory.
#
#
# ============================================================
# 91. ABSOLUTE VS RELATIVE PATH
# ============================================================
#
# RELATIVE:
#
#     "data/file.csv"
#
# Means:
#
#     starting from current directory.
#
#
# ABSOLUTE:
#
#     "C:/Users/Yasin/project/data/file.csv"
#
# Means:
#
#     exact location.
#
#
# Prefer relative paths for portable projects when possible.
#
#
# ============================================================
# 92. PROJECT DIRECTORY STRUCTURE
# ============================================================
#
# A practical data project:
#
#     project/
#     │
#     ├── data/
#     │   ├── raw/
#     │   └── processed/
#     │
#     ├── output/
#     │
#     ├── notebooks/
#     │
#     ├── src/
#     │
#     └── main.py
#
#
# Data flow:
#
#     data/raw
#          ↓
#       cleaning
#          ↓
#     data/processed
#          ↓
#       analysis
#          ↓
#       output
#
#
# ============================================================
# 93. SAFE DATA PROJECT PRINCIPLE
# ============================================================
#
#     RAW
#      ↓
#     NEVER DESTROY
#      ↓
#     PROCESS
#      ↓
#     CLEANED
#      ↓
#     ANALYSIS
#      ↓
#     OUTPUT
#
#
# This is much safer than repeatedly editing the original file.
#
#
# ============================================================
# 94. FILE HANDLING + FUNCTIONS
# ============================================================
#
# Instead of writing one huge script:
#
#     open file
#     read
#     clean
#     validate
#     calculate
#     write
#
# divide the workflow:
#
#     def read_data():
#         ...
#
#     def clean_data():
#         ...
#
#     def validate_data():
#         ...
#
#     def save_data():
#         ...
#
#
# Diagram:
#
#     main()
#       │
#       ├── read_data()
#       │
#       ├── clean_data()
#       │
#       ├── validate_data()
#       │
#       └── save_data()
#
#
# ============================================================
# 95. A REUSABLE TEXT READER
# ============================================================
#
#     def read_text_file(path):
#         with open(
#             path,
#             "r",
#             encoding="utf-8"
#         ) as file:
#             return file.read()
#
#
# Usage:
#
#     text = read_text_file("data.txt")
#
#
# Important idea:
#
#     function handles file mechanics
#     caller handles the business logic
#
#
# ============================================================
# 96. A REUSABLE LINE READER
# ============================================================
#
#     def read_lines(path):
#         with open(
#             path,
#             "r",
#             encoding="utf-8"
#         ) as file:
#             return [
#                 line.strip()
#                 for line in file
#                 if line.strip()
#             ]
#
#
# Result:
#
#     ["Ali", "Sara", "Reza"]
#
#
# ============================================================
# 97. A REUSABLE CSV READER
# ============================================================
#
#     import csv
#
#     def read_csv(path):
#         with open(
#             path,
#             "r",
#             encoding="utf-8",
#             newline=""
#         ) as file:
#             return list(csv.DictReader(file))
#
#
# Then:
#
#     rows = read_csv("data.csv")
#
#
# This separates:
#
#     file handling
#
# from:
#
#     data analysis
#
#
# ============================================================
# 98. DATA PIPELINE ARCHITECTURE
# ============================================================
#
#     ┌───────────────┐
#     │ Raw data      │
#     │ CSV / JSON    │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Read          │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Clean         │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Validate      │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Transform     │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Analyze       │
#     └───────┬───────┘
#             ↓
#     ┌───────────────┐
#     │ Save output   │
#     └───────────────┘
#
#
# ============================================================
# 99. FILE HANDLING AND DATA ANALYSIS
# ============================================================
#
# Imagine:
#
#     sales.csv
#
#     name,department,sales
#     Ali,IT,1200
#     Sara,HR,900
#     Reza,IT,1500
#
#
# Read:
#
#     import csv
#
#     with open(
#         "sales.csv",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         rows = list(csv.DictReader(file))
#
#
# Calculate total:
#
#     total = sum(
#         float(row["sales"])
#         for row in rows
#     )
#
#
# ============================================================
# 100. GROUPING DATA
# ============================================================
#
# File handling is often only the first step.
#
# Once data is loaded:
#
#     rows
#       ↓
#     filter
#       ↓
#     group
#       ↓
#     aggregate
#
#
# Example:
#
#     from collections import defaultdict
#
#     groups = defaultdict(list)
#
#     for row in rows:
#         groups[row["department"]].append(row)
#
#
# Result conceptually:
#
#     IT → [row1, row3]
#     HR → [row2]
#
#
# ============================================================
# 101. COUNTER FOR FILE DATA
# ============================================================
#
#     from collections import Counter
#
#     departments = [
#         row["department"]
#         for row in rows
#     ]
#
#     counts = Counter(departments)
#
#
# Result:
#
#     Counter({
#         "IT": 2,
#         "HR": 1
#     })
#
#
# ============================================================
# 102. FILE HANDLING + DATA VALIDATION
# ============================================================
#
# A robust data workflow should ask:
#
#     Does file exist?
#             ↓
#     Is it readable?
#             ↓
#     Does it have expected columns?
#             ↓
#     Are values valid?
#             ↓
#     Are required fields present?
#             ↓
#     Are numeric fields numeric?
#             ↓
#     Are ranges reasonable?
#
#
# Example:
#
#     required_columns = {
#         "name",
#         "age",
#         "score"
#     }
#
#     actual_columns = set(rows[0])
#
#     if not required_columns <= actual_columns:
#         raise ValueError(
#             "Required columns are missing."
#         )
#
#
# ============================================================
# 103. VALIDATING NUMERIC RANGES
# ============================================================
#
# Example:
#
#     score = float(row["score"])
#
#     if not 0 <= score <= 100:
#         raise ValueError(
#             "Score must be between 0 and 100."
#         )
#
#
# Important:
#
#     outlier
#
# does not automatically mean:
#
#     invalid
#
# For example:
#
#     age = 101
#
# might be suspicious.
#
# It should be:
#
#     flagged / investigated
#
# rather than automatically deleted without context.
#
#
# ============================================================
# 104. DATA CLEANING: SENTINEL VALUES
# ============================================================
#
# Real files may use:
#
#     ""
#     "NA"
#     "N/A"
#     "null"
#     "None"
#     "?"
#     "-"
#
# to mean missing.
#
#
# Standardize them:
#
#     MISSING = {
#         "",
#         "NA",
#         "N/A",
#         "null",
#         "None",
#         "?"
#     }
#
#
#     value = value.strip()
#
#     if value in MISSING:
#         value = None
#
#
# ============================================================
# 105. CASE NORMALIZATION
# ============================================================
#
# Input:
#
#     male
#     Male
#     MALE
#     m
#
#
# A cleaning rule could map:
#
#     gender_map = {
#         "male": "Male",
#         "m": "Male",
#         "female": "Female",
#         "f": "Female"
#     }
#
#
# Then:
#
#     normalized = gender_map.get(
#         value.strip().lower()
#     )
#
#
# ============================================================
# 106. FILE HANDLING + DATE DATA
# ============================================================
#
# File:
#
#     date,sales
#     2026-09-01,100
#     2026-09-02,120
#
#
# CSV provides:
#
#     "2026-09-01"
#
# Convert:
#
#     from datetime import datetime
#
#     date = datetime.strptime(
#         row["date"],
#         "%Y-%m-%d"
#     ).date()
#
#
# Then dates can be:
#
#     compared
#     sorted
#     filtered
#     subtracted
#
#
# ============================================================
# 107. FILE HANDLING + EXCEL
# ============================================================
#
# Excel files:
#
#     .xlsx
#
# are NOT plain text CSV files.
#
# Python can work with them using external libraries such as:
#
#     openpyxl
#
# or:
#
#     pandas
#
#
# Concept:
#
#     Excel
#        ↓
#     specialized library
#        ↓
#     Python structures
#
#
# Do not try to parse .xlsx with:
#
#     open(..., "r")
#
# as ordinary text.
#
#
# ============================================================
# 108. FILE HANDLING + PANDAS
# ============================================================
#
# Once Python foundations are solid, pandas simplifies tabular work.
#
# Typical:
#
#     import pandas as pd
#
#     df = pd.read_csv("data.csv")
#
#
# Then:
#
#     df.head()
#     df.info()
#     df.describe()
#
#
# Write:
#
#     df.to_csv(
#         "cleaned.csv",
#         index=False
#     )
#
#
# The important conceptual connection:
#
#     csv module
#          ↓
#     low-level Python understanding
#          ↓
#     pandas
#          ↓
#     high-level tabular analysis
#
#
# Understanding file handling makes pandas much easier to reason about.
#
#
# ============================================================
# 109. FILE HANDLING + SQL
# ============================================================
#
# Similar conceptual pipeline:
#
#     CSV
#       ↓
#     read
#       ↓
#     Python
#
#
# SQL:
#
#     database
#       ↓
#     query
#       ↓
#     Python
#
#
# Both are data sources.
#
#
#     FILE / DATABASE
#           ↓
#        EXTRACT
#           ↓
#        TRANSFORM
#           ↓
#        ANALYZE
#           ↓
#         OUTPUT
#
#
# This is the foundation of ETL.
#
#
# ============================================================
# 110. FILE HANDLING + ETL
# ============================================================
#
# ETL:
#
#     Extract
#     Transform
#     Load
#
#
# Example:
#
#     raw.csv
#        ↓
#     EXTRACT
#        ↓
#     Python
#        ↓
#     TRANSFORM
#        ↓
#     cleaned.csv
#        ↓
#     LOAD
#        ↓
#     database / BI system
#
#
# This is one of the most important real-world applications
# of Python file handling.
#
#
# ============================================================
# 111. STREAMING LARGE FILES
# ============================================================
#
# Suppose:
#
#     data.csv = 20 GB
#
# Bad approach:
#
#     data = file.read()
#
# because it attempts to load the entire file.
#
#
# Better:
#
#     with open("large.txt", encoding="utf-8") as file:
#         for line in file:
#             process(line)
#
#
# Or CSV:
#
#     import csv
#
#     with open(
#         "large.csv",
#         encoding="utf-8",
#         newline=""
#     ) as file:
#
#         reader = csv.DictReader(file)
#
#         for row in reader:
#             process(row)
#
#
# Memory:
#
#     entire file:
#
#     █████████████████████████████
#
#     streaming:
#
#     ██ → ██ → ██ → ██ → ██
#
#
# ============================================================
# 112. GENERATORS + FILES
# ============================================================
#
# A generator can create a memory-efficient processing pipeline.
#
#     def clean_lines(path):
#         with open(
#             path,
#             encoding="utf-8"
#         ) as file:
#             for line in file:
#                 line = line.strip()
#
#                 if line:
#                     yield line
#
#
# Usage:
#
#     for line in clean_lines("data.txt"):
#         print(line)
#
#
# Pipeline:
#
#     file
#       ↓
#     generator
#       ↓
#     clean line
#       ↓
#     consumer
#
#
# ============================================================
# 113. CHUNK PROCESSING
# ============================================================
#
# For large binary/text data, process chunks:
#
#     CHUNK_SIZE = 1024 * 1024
#
#     with open(
#         "large.bin",
#         "rb"
#     ) as file:
#
#         while True:
#             chunk = file.read(CHUNK_SIZE)
#
#             if not chunk:
#                 break
#
#             process(chunk)
#
#
# Meaning:
#
#     not:
#
#         all data → memory
#
#     but:
#
#         chunk → process
#         chunk → process
#         chunk → process
#
#
# ============================================================
# 114. walrus OPERATOR IN FILE READING
# ============================================================
#
# Python allows:
#
#     while chunk := file.read(1024):
#         process(chunk)
#
#
# This combines:
#
#     assignment
#     +
#     condition
#
#
# But the explicit version is often easier for beginners:
#
#     while True:
#         chunk = file.read(1024)
#
#         if not chunk:
#             break
#
#         process(chunk)
#
#
# Learn the explicit version first.
#
#
# ============================================================
# 115. TEMPORARY FILES
# ============================================================
#
# For temporary data, Python provides:
#
#     tempfile
#
# Examples:
#
#     TemporaryFile
#     NamedTemporaryFile
#     TemporaryDirectory
#
#
# Concept:
#
#     temporary data
#          ↓
#     use
#          ↓
#     cleanup
#
#
# Useful for:
#
#     intermediate processing
#     tests
#     downloads
#     generated data
#
#
# ============================================================
# 116. FILE LOCKING / CONCURRENCY — CONCEPT
# ============================================================
#
# If multiple processes write to the same file simultaneously,
# problems can occur.
#
#     Process A ──┐
#                 ├──→ same file
#     Process B ──┘
#
#
# Possible problems:
#
#     corrupted output
#     interleaved writes
#     lost updates
#
#
# Serious concurrent applications may need:
#
#     locks
#     databases
#     atomic file replacement
#
#
# For ordinary single-process scripts, this is usually not an issue.
#
#
# ============================================================
# 117. ATOMIC-STYLE OUTPUT PATTERN
# ============================================================
#
# For important generated files, a safer conceptual pattern is:
#
#     create temporary output
#             ↓
#     successfully finish writing
#             ↓
#     replace target
#
#
# This reduces the chance that a failed program leaves a
# partially written final file.
#
#
# ============================================================
# 118. COMMON FILE HANDLING MISTAKES
# ============================================================
#
# (1) Forgetting encoding
#
#     open("data.txt")
#
# Better:
#
#     open(
#         "data.txt",
#         encoding="utf-8"
#     )
#
#
# (2) Forgetting with
#
#     file = open(...)
#
# Better:
#
#     with open(...) as file:
#         ...
#
#
# (3) Using "w" accidentally
#
#     w = overwrite
#
#
# (4) Assuming CSV values are already numeric
#
#     "25" != 25
#
#
# (5) Using split(",") as a full CSV parser
#
#     use csv module
#
#
# (6) Loading huge files completely
#
#     use iteration/chunks
#
#
# (7) Forgetting newline="" for csv
#
#     use newline=""
#
#
# (8) Ignoring missing files
#
#     handle FileNotFoundError
#
#
# (9) Destroying raw data
#
#     keep raw and processed files separate
#
#
# (10) Mixing file handling with all business logic
#
#     use functions
#
#
# ============================================================
# 119. COMMON PATH MISTAKES
# ============================================================
#
# Relative paths depend on CWD.
#
# If:
#
#     open("data.csv")
#
# fails, ask:
#
#     Where is the program running?
#
#
# Check:
#
#     from pathlib import Path
#
#     print(Path.cwd())
#
#
# Also:
#
#     path = Path("data.csv")
#     print(path.resolve())
#
#
# ============================================================
# 120. COMMON CSV MISTAKES
# ============================================================
#
# Mistake:
#
#     age = row["age"] + 1
#
# if:
#
#     row["age"] == "25"
#
# This fails because:
#
#     "25" + 1
#
# is invalid.
#
#
# Correct:
#
#     age = int(row["age"])
#     age += 1
#
#
# ============================================================
# 121. COMMON JSON MISTAKES
# ============================================================
#
# Confusing:
#
#     json.load
#     json.loads
#
#
# Remember:
#
#     load
#         → file
#
#     loads
#         → string
#
#
#     dump
#         → file
#
#     dumps
#         → string
#
#
# Memory trick:
#
#     s = string
#
#
# ============================================================
# 122. FILE METHOD VS FUNCTION
# ============================================================
#
# FUNCTION:
#
#     open("data.txt")
#
# METHOD:
#
#     file.read()
#
#
# Mental model:
#
#     function
#         ↓
#     open(file)
#
#
#     object
#         ↓
#     file.read()
#
#
# Python file handling combines both.
#
#
# ============================================================
# 123. MASTER FILE METHOD TABLE
# ============================================================
#
#     +------------------+----------------------------------+
#     | Tool             | Main purpose                      |
#     +------------------+----------------------------------+
#     | open()           | Open a file                       |
#     | read()           | Read content                      |
#     | readline()       | Read one line                     |
#     | readlines()      | Read all lines                    |
#     | write()          | Write one string                  |
#     | writelines()     | Write many strings                |
#     | seek()           | Move cursor                       |
#     | tell()           | Get cursor position               |
#     | flush()          | Flush buffer                      |
#     | close()          | Close file                        |
#     +------------------+----------------------------------+
#
#
# pathlib:
#
#     +------------------+----------------------------------+
#     | Tool             | Main purpose                      |
#     +------------------+----------------------------------+
#     | exists()         | Check existence                  |
#     | is_file()        | Check file                        |
#     | is_dir()         | Check directory                   |
#     | mkdir()          | Create directory                  |
#     | glob()           | Find matching files               |
#     | rglob()          | Recursive file search             |
#     | unlink()         | Delete file                       |
#     | rename()         | Rename/move                       |
#     | read_text()      | Read text                         |
#     | write_text()     | Write text                        |
#     +------------------+----------------------------------+
#
#
# ============================================================
# 124. MASTER INPUT / OUTPUT TABLE
# ============================================================
#
#     +----------------------+------------------------------+
#     | Need                 | Use                          |
#     +----------------------+------------------------------+
#     | User input           | input()                      |
#     | Screen output        | print()                      |
#     | Read text            | open(..., "r")              |
#     | Write text           | open(..., "w")              |
#     | Append text          | open(..., "a")              |
#     | Create safely        | open(..., "x")              |
#     | Binary read          | open(..., "rb")             |
#     | Binary write         | open(..., "wb")             |
#     | CSV read             | csv.reader / DictReader     |
#     | CSV write            | csv.writer / DictWriter     |
#     | JSON read             | json.load / loads           |
#     | JSON write            | json.dump / dumps           |
#     | Paths                | pathlib.Path                |
#     | Logs                 | logging                     |
#     +----------------------+------------------------------+
#
#
# ============================================================
# 125. WHICH TOOL SHOULD I USE?
# ============================================================
#
#     Need keyboard input?
#         ↓
#       input()
#
#     Need screen output?
#         ↓
#       print()
#
#     Need simple text file?
#         ↓
#       open()
#
#     Need table?
#         ↓
#       csv
#
#     Need nested structured data?
#         ↓
#       json
#
#     Need path manipulation?
#         ↓
#       pathlib
#
#     Need many files?
#         ↓
#       pathlib.glob()
#
#     Need huge file?
#         ↓
#       iteration / chunks
#
#     Need logs?
#         ↓
#       logging
#
#     Need serious tabular analysis?
#         ↓
#       pandas
#
#
# ============================================================
# 126. COMPLETE END-TO-END EXAMPLE
#     RAW CSV → CLEAN → VALIDATE → ANALYZE → SAVE
# ============================================================
#
# The following is intentionally based on standard Python.
# No pandas is required.
#
# Example raw CSV:
#
#     name,department,age,score
#      Ali ,IT,25,88
#     Sara,HR,?,94
#      Reza,IT,29,91
#     Mina,HR,31,N/A
#
#
# ------------------------------------------------------------
# STEP 1 — IMPORT TOOLS
# ------------------------------------------------------------
#
# import csv
# from pathlib import Path
#
#
# ------------------------------------------------------------
# STEP 2 — DEFINE PATHS
# ------------------------------------------------------------
#
# input_path = Path("raw_students.csv")
# output_path = Path("cleaned_students.csv")
#
#
# ------------------------------------------------------------
# STEP 3 — READ
# ------------------------------------------------------------
#
# rows = []
#
# with input_path.open(
#     "r",
#     encoding="utf-8",
#     newline=""
# ) as file:
#
#     reader = csv.DictReader(file)
#
#     for row in reader:
#         rows.append(row)
#
#
# ------------------------------------------------------------
# STEP 4 — CLEAN
# ------------------------------------------------------------
#
# missing_values = {
#     "",
#     "?",
#     "NA",
#     "N/A",
#     "null",
#     "None"
# }
#
# cleaned_rows = []
#
# for row in rows:
#
#     name = row["name"].strip()
#
#     department = row["department"].strip().title()
#
#     age_text = row["age"].strip()
#     score_text = row["score"].strip()
#
#     age = None
#     score = None
#
#     if age_text not in missing_values:
#         try:
#             age = int(age_text)
#         except ValueError:
#             age = None
#
#     if score_text not in missing_values:
#         try:
#             score = float(score_text)
#         except ValueError:
#             score = None
#
#     cleaned_rows.append({
#         "name": name,
#         "department": department,
#         "age": age,
#         "score": score
#     })
#
#
# ------------------------------------------------------------
# STEP 5 — VALIDATE
# ------------------------------------------------------------
#
# valid_rows = []
#
# for row in cleaned_rows:
#
#     if not row["name"]:
#         continue
#
#     if row["age"] is not None:
#         if not 0 <= row["age"] <= 120:
#             continue
#
#     if row["score"] is not None:
#         if not 0 <= row["score"] <= 100:
#             continue
#
#     valid_rows.append(row)
#
#
# ------------------------------------------------------------
# STEP 6 — ANALYZE
# ------------------------------------------------------------
#
# scores = [
#     row["score"]
#     for row in valid_rows
#     if row["score"] is not None
# ]
#
# if scores:
#     average_score = sum(scores) / len(scores)
# else:
#     average_score = None
#
#
# ------------------------------------------------------------
# STEP 7 — SORT
# ------------------------------------------------------------
#
# valid_rows.sort(
#     key=lambda row: (
#         row["score"] is None,
#         -(row["score"] or 0)
#     )
# )
#
#
# ------------------------------------------------------------
# STEP 8 — WRITE CLEANED DATA
# ------------------------------------------------------------
#
# output_path.parent.mkdir(
#     parents=True,
#     exist_ok=True
# )
#
# fieldnames = [
#     "name",
#     "department",
#     "age",
#     "score"
# ]
#
# with output_path.open(
#     "w",
#     encoding="utf-8",
#     newline=""
# ) as file:
#
#     writer = csv.DictWriter(
#         file,
#         fieldnames=fieldnames
#     )
#
#     writer.writeheader()
#     writer.writerows(valid_rows)
#
#
# ------------------------------------------------------------
# STEP 9 — REPORT
# ------------------------------------------------------------
#
# print("Rows loaded:", len(rows))
# print("Rows valid:", len(valid_rows))
#
# if average_score is not None:
#     print(
#         f"Average score: {average_score:.2f}"
#     )
#
#
# ============================================================
# 127. THE SAME PIPELINE AS A DIAGRAM
# ============================================================
#
#     raw_students.csv
#             │
#             ↓
#       Path.open()
#             │
#             ↓
#       DictReader
#             │
#             ↓
#        raw strings
#             │
#             ↓
#          clean
#             │
#             ↓
#         convert
#             │
#             ↓
#         validate
#             │
#             ↓
#       valid records
#             │
#        ┌────┴────┐
#        ↓         ↓
#     analyze     sort
#        │         │
#        └────┬────┘
#             ↓
#       DictWriter
#             ↓
#     cleaned_students.csv
#
#
# ============================================================
# 128. A BETTER ARCHITECTURE USING FUNCTIONS
# ============================================================
#
# For larger programs:
#
#     import csv
#     from pathlib import Path
#
#
#     def read_csv(path):
#         with Path(path).open(
#             "r",
#             encoding="utf-8",
#             newline=""
#         ) as file:
#             return list(
#                 csv.DictReader(file)
#             )
#
#
#     def clean_row(row):
#         missing = {
#             "",
#             "?",
#             "NA",
#             "N/A"
#         }
#
#         age_text = row["age"].strip()
#
#         try:
#             age = (
#                 None
#                 if age_text in missing
#                 else int(age_text)
#             )
#         except ValueError:
#             age = None
#
#         return {
#             "name": row["name"].strip(),
#             "age": age
#         }
#
#
#     def clean_data(rows):
#         return [
#             clean_row(row)
#             for row in rows
#         ]
#
#
#     def write_csv(path, rows, fieldnames):
#         with Path(path).open(
#             "w",
#             encoding="utf-8",
#             newline=""
#         ) as file:
#
#             writer = csv.DictWriter(
#                 file,
#                 fieldnames=fieldnames
#             )
#
#             writer.writeheader()
#             writer.writerows(rows)
#
#
# This architecture is easier to:
#
#     test
#     debug
#     reuse
#     extend
#     maintain
#
#
# ============================================================
# 129. FILE HANDLING + OBJECT-ORIENTED THINKING
# ============================================================
#
# A file object has:
#
#     state
#     +
#     behavior
#
#
# Example:
#
#     file.name
#     file.mode
#     file.closed
#
# and:
#
#     file.read()
#     file.write()
#     file.seek()
#
#
# This is object-oriented programming in everyday Python.
#
#
# Mental model:
#
#     file object
#          │
#     ┌────┴─────────┐
#     ↓              ↓
#    state         methods
#     │              │
#   name           read()
#   mode           write()
#   position       seek()
#
#
# ============================================================
# 130. FILES ARE ITERABLE
# ============================================================
#
#     with open("data.txt", encoding="utf-8") as file:
#         iterator = iter(file)
#
#     line = next(iterator)
#
#
# This connects file handling to the Python iteration model:
#
#     iterable
#         ↓
#     iter()
#         ↓
#     iterator
#         ↓
#     next()
#         ↓
#     value
#
#
# You normally do not need to call iter()/next() manually,
# because:
#
#     for line in file:
#
# handles the iteration naturally.
#
#
# ============================================================
# 131. CONTEXT MANAGER MENTAL MODEL
# ============================================================
#
#     with RESOURCE as x:
#         USE x
#
#
# Conceptually:
#
#     enter
#       ↓
#     resource available
#       ↓
#     work
#       ↓
#     exit
#       ↓
#     cleanup
#
#
# For files:
#
#     open
#       ↓
#     read/write
#       ↓
#     close
#
#
# ============================================================
# 132. READING A FILE AS A LIST
# ============================================================
#
#     from pathlib import Path
#
#     lines = Path(
#         "data.txt"
#     ).read_text(
#         encoding="utf-8"
#     ).splitlines()
#
#
# splitlines() removes line-boundary markers cleanly.
#
#
# Example:
#
#     "Ali\nSara\nReza"
#
# becomes:
#
#     ["Ali", "Sara", "Reza"]
#
#
# ============================================================
# 133. splitlines()
# ============================================================
#
# String method:
#
#     text.splitlines()
#
#
# Useful when the entire file is already in memory.
#
#
# Compare:
#
#     file iteration
#
# → memory efficient.
#
#
#     read().splitlines()
#
# → entire file first, then split.
#
#
# ============================================================
# 134. WRITING TEXT WITH pathlib
# ============================================================
#
#     from pathlib import Path
#
#     output = Path("report.txt")
#
#     output.write_text(
#         "Analysis complete\n",
#         encoding="utf-8"
#     )
#
#
# Read:
#
#     report = output.read_text(
#         encoding="utf-8"
#     )
#
#
# ============================================================
# 135. FILE EXTENSIONS
# ============================================================
#
# Extension often gives a clue:
#
#     .txt   → text
#     .csv   → comma-separated table
#     .json  → JSON
#     .xlsx  → Excel
#     .pdf   → PDF
#     .jpg   → image
#     .png   → image
#     .bin   → binary data
#
#
# But:
#
#     extension does not guarantee actual file content.
#
# Do not assume a file is valid merely because its name ends
# with ".csv".
#
#
# ============================================================
# 136. FILE SIZE
# ============================================================
#
#     from pathlib import Path
#
#     path = Path("data.csv")
#
#     print(path.stat().st_size)
#
#
# Size is usually reported in bytes.
#
#
# Convert:
#
#     size_mb = path.stat().st_size / (
#         1024 ** 2
#     )
#
#
# Useful for deciding:
#
#     read all?
#     stream?
#     chunk?
#
#
# ============================================================
# 137. FILE METADATA
# ============================================================
#
#     stat = path.stat()
#
# Useful information includes:
#
#     st_size
#     modification time
#     access time
#     creation-related platform information
#
#
# Exact metadata semantics can vary by operating system.
#
#
# ============================================================
# 138. ENCODING + DECODING MENTAL MODEL
# ============================================================
#
#     Python str
#         │
#       encode()
#         ↓
#       bytes
#         │
#      storage
#         │
#      read bytes
#         ↓
#      decode()
#         ↓
#     Python str
#
#
# Example:
#
#     text = "Hello"
#
#     raw = text.encode("utf-8")
#
#     restored = raw.decode("utf-8")
#
#
# ============================================================
# 139. TEXT FILE = CHARACTER INTERFACE
# ============================================================
#
# In text mode:
#
#     read()
#
# returns:
#
#     str
#
#
# In binary mode:
#
#     read()
#
# returns:
#
#     bytes
#
#
# Table:
#
#     +------------+----------------+
#     | Mode       | read() result  |
#     +------------+----------------+
#     | "r"        | str            |
#     | "rb"       | bytes          |
#     +------------+----------------+
#
#
# ============================================================
# 140. FILE HANDLING AND MEMORY
# ============================================================
#
# Think about data size.
#
# SMALL:
#
#     file → read() → memory
#
# LARGE:
#
#     file → line/chunk → process
#
#
# Decision:
#
#     Is the complete file small?
#             │
#       ┌─────┴─────┐
#       ↓           ↓
#      yes          no
#       ↓           ↓
#    read all    stream/chunk
#
#
# ============================================================
# 141. PERFORMANCE BASICS
# ============================================================
#
# Roughly:
#
#     read entire file
#         → O(n) data processed
#         → O(n) memory
#
#
#     line-by-line processing
#         → O(n) data processed
#         → much lower working memory
#
#
# The computational work is still proportional to the amount
# of data read; streaming primarily changes memory usage.
#
#
# ============================================================
# 142. WHEN TO USE read()
# ============================================================
#
# Good:
#
#     small configuration file
#     small text document
#     source code
#     small JSON
#
#
# Less appropriate:
#
#     huge log file
#     multi-GB data
#     massive CSV
#
#
# ============================================================
# 143. WHEN TO USE LINE ITERATION
# ============================================================
#
# Good for:
#
#     logs
#     large text files
#     line-oriented datasets
#     filtering records
#     counting events
#
#
# Pattern:
#
#     with open(...) as file:
#         for line in file:
#             ...
#
#
# ============================================================
# 144. WHEN TO USE CHUNKS
# ============================================================
#
# Good for:
#
#     binary files
#     very large files
#     copying
#     hashing
#     streaming transformations
#
#
# Pattern:
#
#     while chunk := file.read(CHUNK_SIZE):
#         process(chunk)
#
#
# ============================================================
# 145. DATA INTEGRITY
# ============================================================
#
# File handling is not only:
#
#     "Can Python read this?"
#
# It is also:
#
#     "Did I read the data correctly?"
#
#
# Check:
#
#     expected columns
#     number of rows
#     missing values
#     numeric conversion
#     duplicates
#     valid ranges
#     encoding
#     file size
#
#
# Example:
#
#     print("Rows:", len(rows))
#
#
# ============================================================
# 146. CHECKSUM IDEA
# ============================================================
#
# For important files, a checksum/hash can help identify whether
# file content changed.
#
# Python provides:
#
#     hashlib
#
#
# Concept:
#
#     file
#       ↓
#     hash function
#       ↓
#     fingerprint
#
#
# Common algorithms include:
#
#     SHA-256
#
# This is useful for:
#
#     reproducibility
#     downloads
#     data integrity
#     duplicate detection
#
#
# ============================================================
# 147. EXAMPLE: SHA-256 OF A FILE
# ============================================================
#
#     import hashlib
#
#     sha256 = hashlib.sha256()
#
#     with open(
#         "data.bin",
#         "rb"
#     ) as file:
#
#         for chunk in iter(
#             lambda: file.read(1024 * 1024),
#             b""
#         ):
#             sha256.update(chunk)
#
#     digest = sha256.hexdigest()
#
#     print(digest)
#
#
# This demonstrates an important advanced pattern:
#
#     file → chunks → processing
#
#
# ============================================================
# 148. FILES + RESEARCH DATA
# ============================================================
#
# For research:
#
#     raw/
#       ↓
#     imported/
#       ↓
#     cleaned/
#       ↓
#     analyzed/
#       ↓
#     results/
#
#
# Keep:
#
#     original source
#     cleaning script
#     cleaned data
#     analysis script
#     generated results
#
#
# This supports reproducibility.
#
#
# ============================================================
# 149. FILES + MACHINE LEARNING
# ============================================================
#
# Typical ML workflow:
#
#     dataset.csv
#          ↓
#     read
#          ↓
#     inspect
#          ↓
#     clean
#          ↓
#     validate
#          ↓
#     feature engineering
#          ↓
#     train/test split
#          ↓
#     model
#          ↓
#     predictions
#          ↓
#     save predictions
#
#
# File handling is therefore not a separate skill from ML.
#
# It is part of the data pipeline surrounding the model.
#
#
# ============================================================
# 150. FILES + MODEL OUTPUT
# ============================================================
#
# Example:
#
#     predictions = [
#         {"id": 1, "prediction": 0},
#         {"id": 2, "prediction": 1}
#     ]
#
#
# Save as JSON or CSV depending on the use case.
#
#
# CSV:
#
#     good for tables / BI / Excel
#
#
# JSON:
#
#     good for nested application/API structures
#
#
# ============================================================
# 151. FILE HANDLING + POWER BI
# ============================================================
#
# A common workflow:
#
#     Python
#        ↓
#     clean CSV
#        ↓
#     Power BI
#        ↓
#     dashboard
#
#
# Python can perform:
#
#     cleaning
#     validation
#     preprocessing
#     feature engineering
#
# Power BI can perform:
#
#     visualization
#     dashboarding
#     business reporting
#
#
# ============================================================
# 152. FILE HANDLING + AUTOMATION
# ============================================================
#
# Example automation:
#
#     every morning
#          ↓
#     find newest CSV
#          ↓
#     read
#          ↓
#     clean
#          ↓
#     calculate KPIs
#          ↓
#     save report
#          ↓
#     log result
#
#
# Python file handling is the foundation of this type of
# automation.
#
#
# ============================================================
# 153. FINDING THE NEWEST FILE
# ============================================================
#
#     from pathlib import Path
#
#     files = list(
#         Path("data").glob("*.csv")
#     )
#
#     newest = max(
#         files,
#         key=lambda path: path.stat().st_mtime
#     )
#
#     print(newest)
#
#
# This is a practical automation pattern.
#
#
# ============================================================
# 154. BATCH PROCESSING FILES
# ============================================================
#
#     from pathlib import Path
#
#     for path in Path("data").glob("*.csv"):
#         print("Processing:", path.name)
#
#
# Extend:
#
#     for path in Path("data").glob("*.csv"):
#         rows = read_csv(path)
#         cleaned = clean_data(rows)
#         save_results(path, cleaned)
#
#
# Diagram:
#
#     a.csv ──→ process
#     b.csv ──→ process
#     c.csv ──→ process
#     d.csv ──→ process
#
#
# ============================================================
# 155. FILE NAMING
# ============================================================
#
# Good filenames should communicate:
#
#     what
#     when
#     status/version
#
#
# Examples:
#
#     raw_sales_2026-09-10.csv
#     cleaned_sales_2026-09-10.csv
#     model_predictions_2026-09-10.csv
#
#
# Avoid ambiguous:
#
#     final.csv
#     final2.csv
#     final_new.csv
#     final_really_final.csv
#
#
# ============================================================
# 156. OUTPUT DIRECTORIES
# ============================================================
#
# A useful pattern:
#
#     output/
#     ├── cleaned/
#     ├── reports/
#     ├── predictions/
#     └── logs/
#
#
# This separates outputs by purpose.
#
#
# ============================================================
# 157. FILE HANDLING CHECKLIST
# ============================================================
#
# Before reading:
#
#     [ ] Does the path exist?
#     [ ] Is it a file?
#     [ ] Is the format what I expect?
#     [ ] Is the encoding correct?
#
#
# While reading:
#
#     [ ] Are rows parsed correctly?
#     [ ] Are values strings?
#     [ ] Are missing values represented consistently?
#     [ ] Are columns present?
#
#
# During processing:
#
#     [ ] Convert types
#     [ ] Validate ranges
#     [ ] Detect duplicates
#     [ ] Track rejected records
#
#
# Before writing:
#
#     [ ] Correct output path?
#     [ ] Correct format?
#     [ ] Do I want overwrite or append?
#     [ ] Is raw data preserved?
#
#
# After writing:
#
#     [ ] Does output exist?
#     [ ] Can it be reopened?
#     [ ] Does row count make sense?
#
#
# ============================================================
# 158. MASTER DECISION TABLE
# ============================================================
#
#     TASK
#         ↓
#
#     User enters data?
#         → input()
#
#     Display data?
#         → print()
#
#     Small text file?
#         → read() / write()
#
#     Large text file?
#         → for line in file
#
#     Huge binary file?
#         → read chunks
#
#     Table?
#         → csv
#
#     Nested structured data?
#         → json
#
#     Paths?
#         → pathlib
#
#     Many files?
#         → glob / rglob
#
#     Logs?
#         → logging
#
#     Temporary data?
#         → tempfile
#
#     Integrity verification?
#         → hashlib
#
#     Tabular analysis?
#         → pandas
#
#
# ============================================================
# 159. MOST IMPORTANT SYNTAX TO MEMORIZE
# ============================================================
#
# # User input
# value = input("Enter value: ")
#
#
# # Output
# print(value)
#
#
# # Read complete text file
# with open(
#     "data.txt",
#     "r",
#     encoding="utf-8"
# ) as file:
#     text = file.read()
#
#
# # Read line by line
# with open(
#     "data.txt",
#     "r",
#     encoding="utf-8"
# ) as file:
#     for line in file:
#         print(line.strip())
#
#
# # Write
# with open(
#     "output.txt",
#     "w",
#     encoding="utf-8"
# ) as file:
#     file.write("Hello\n")
#
#
# # Append
# with open(
#     "log.txt",
#     "a",
#     encoding="utf-8"
# ) as file:
#     file.write("New event\n")
#
#
# # Path
# from pathlib import Path
#
# path = Path("data") / "file.csv"
#
#
# # CSV
# import csv
#
# with open(
#     "data.csv",
#     encoding="utf-8",
#     newline=""
# ) as file:
#     reader = csv.DictReader(file)
#
#     for row in reader:
#         print(row)
#
#
# # JSON
# import json
#
# with open(
#     "data.json",
#     encoding="utf-8"
# ) as file:
#     data = json.load(file)
#
#
# ============================================================
# 160. THE DEEPEST "PYTHON FEELING"
# ============================================================
#
# Do not think:
#
#     "I need to memorize 20 file methods."
#
#
# Think:
#
#     "I have a stream of data."
#
#
# Then ask:
#
#     Where does the data come from?
#         ↓
#     How is it represented?
#         ↓
#     How much data is there?
#         ↓
#     How should I access it?
#         ↓
#     How should I transform it?
#         ↓
#     Where should the result go?
#
#
# ============================================================
# 161. THE CORE FILE-HANDLING MENTAL MODEL
# ============================================================
#
#                    DATA
#                     │
#          ┌──────────┼───────────┐
#          ↓          ↓           ↓
#       keyboard     file       network
#          │          │           │
#          └──────────┼───────────┘
#                     ↓
#                  Python
#                     │
#              ┌──────┼──────┐
#              ↓      ↓      ↓
#           clean   analyze transform
#              │      │      │
#              └──────┼──────┘
#                     ↓
#                   output
#                     │
#          ┌──────────┼───────────┐
#          ↓          ↓           ↓
#        screen      CSV        JSON
#
#
# ============================================================
# 162. LEARNING ORDER
# ============================================================
#
# Learn these in this order:
#
# ------------------------------------------------------------
# LEVEL 1 — BASIC I/O
# ------------------------------------------------------------
#
#     input()
#     print()
#     f-strings
#     sep
#     end
#
# MUST BE ABLE TO:
#
#     [ ] ask for input
#     [ ] convert input
#     [ ] print formatted results
#
#
# ------------------------------------------------------------
# LEVEL 2 — TEXT FILES
# ------------------------------------------------------------
#
#     open()
#     with
#     read()
#     readline()
#     readlines()
#     write()
#     writelines()
#
# MUST BE ABLE TO:
#
#     [ ] read a text file
#     [ ] write a text file
#     [ ] append to a file
#     [ ] process lines
#
#
# ------------------------------------------------------------
# LEVEL 3 — PATHS
# ------------------------------------------------------------
#
#     pathlib.Path
#     exists()
#     is_file()
#     is_dir()
#     mkdir()
#     glob()
#
# MUST BE ABLE TO:
#
#     [ ] construct paths
#     [ ] check files
#     [ ] create directories
#     [ ] find groups of files
#
#
# ------------------------------------------------------------
# LEVEL 4 — CSV
# ------------------------------------------------------------
#
#     csv.reader
#     csv.DictReader
#     csv.writer
#     csv.DictWriter
#
# MUST BE ABLE TO:
#
#     [ ] read a table
#     [ ] understand that values start as strings
#     [ ] convert types
#     [ ] write cleaned CSV
#
#
# ------------------------------------------------------------
# LEVEL 5 — JSON
# ------------------------------------------------------------
#
#     json.load
#     json.loads
#     json.dump
#     json.dumps
#
# MUST BE ABLE TO:
#
#     [ ] read JSON
#     [ ] navigate nested dictionaries/lists
#     [ ] write JSON
#
#
# ------------------------------------------------------------
# LEVEL 6 — ERRORS
# ------------------------------------------------------------
#
#     try
#     except
#     FileNotFoundError
#     ValueError
#     PermissionError
#
# MUST BE ABLE TO:
#
#     [ ] handle missing files
#     [ ] handle invalid data
#     [ ] avoid bare except
#
#
# ------------------------------------------------------------
# LEVEL 7 — LARGE DATA
# ------------------------------------------------------------
#
#     file iteration
#     generators
#     chunks
#
# MUST BE ABLE TO:
#
#     [ ] avoid loading huge files unnecessarily
#     [ ] process data incrementally
#
#
# ------------------------------------------------------------
# LEVEL 8 — DATA PIPELINES
# ------------------------------------------------------------
#
#     extract
#     clean
#     validate
#     transform
#     analyze
#     save
#
# MUST BE ABLE TO:
#
#     [ ] build a complete reusable data pipeline
#
#
# ============================================================
# 163. PRACTICE LADDER
# ============================================================
#
# Do not jump directly into huge datasets.
#
#     LEVEL 1
#     ───────
#     input → print
#
#     LEVEL 2
#     ───────
#     write text file
#     read text file
#
#     LEVEL 3
#     ───────
#     read lines
#     clean lines
#
#     LEVEL 4
#     ───────
#     create CSV
#     read CSV
#
#     LEVEL 5
#     ───────
#     clean CSV
#     convert types
#
#     LEVEL 6
#     ───────
#     validate CSV
#
#     LEVEL 7
#     ───────
#     JSON
#
#     LEVEL 8
#     ───────
#     process multiple files
#
#     LEVEL 9
#     ───────
#     large-file streaming
#
#     LEVEL 10
#     ───────
#     complete ETL project
#
#
# ============================================================
# 164. PRACTICE PROJECTS
# ============================================================
#
# PROJECT 1 — PERSONAL NOTES
#
#     input()
#       ↓
#     notes.txt
#       ↓
#     read
#       ↓
#     display
#
#
# PROJECT 2 — STUDENT SCORES
#
#     names + scores
#       ↓
#     CSV
#       ↓
#     calculate average
#       ↓
#     save report
#
#
# PROJECT 3 — EXPENSE TRACKER
#
#     date
#     category
#     amount
#       ↓
#     expenses.csv
#       ↓
#     totals
#       ↓
#     report.csv
#
#
# PROJECT 4 — LOG ANALYZER
#
#     application.log
#       ↓
#     line-by-line reading
#       ↓
#     count ERROR
#       ↓
#     report
#
#
# PROJECT 5 — DATA CLEANER
#
#     messy.csv
#       ↓
#     whitespace cleaning
#       ↓
#     missing values
#       ↓
#     type conversion
#       ↓
#     validation
#       ↓
#     clean.csv
#
#
# PROJECT 6 — BATCH PROCESSOR
#
#     folder/
#       ├── file1.csv
#       ├── file2.csv
#       ├── file3.csv
#       └── file4.csv
#
#           ↓
#
#     process every CSV
#           ↓
#     combine results
#
#
# PROJECT 7 — RESEARCH DATA PIPELINE
#
#     raw inspection data
#          ↓
#     CSV/Excel import
#          ↓
#     cleaning
#          ↓
#     validation
#          ↓
#     feature engineering
#          ↓
#     processed dataset
#          ↓
#     ML analysis
#
#
# ============================================================
# 165. FINAL CHEAT SHEET
# ============================================================
#
# INPUT:
#
#     input()
#
#
# OUTPUT:
#
#     print()
#
#
# OPEN:
#
#     open(path, mode, encoding="utf-8")
#
#
# SAFE OPEN:
#
#     with open(...) as file:
#         ...
#
#
# READ:
#
#     file.read()
#     file.readline()
#     file.readlines()
#
#
# WRITE:
#
#     file.write()
#     file.writelines()
#
#
# POSITION:
#
#     file.tell()
#     file.seek()
#
#
# PATH:
#
#     Path(...)
#     path.exists()
#     path.is_file()
#     path.is_dir()
#     path.mkdir()
#     path.glob()
#
#
# CSV:
#
#     csv.reader
#     csv.DictReader
#     csv.writer
#     csv.DictWriter
#
#
# JSON:
#
#     json.load
#     json.loads
#     json.dump
#     json.dumps
#
#
# ERRORS:
#
#     try
#     except
#
#
# LARGE DATA:
#
#     for line in file
#     chunks
#     generators
#
#
# LOGGING:
#
#     logging
#
#
# ============================================================
# 166. THE ONE-PAGE MEMORY MAP
# ============================================================
#
#     USER
#      │
#      │ input()
#      ↓
#     str
#      │
#      │ conversion
#      ↓
#     int / float / ...
#
#
#     FILE
#      │
#      │ open()
#      ↓
#     file object
#      │
#      ├── read()
#      ├── readline()
#      ├── write()
#      ├── seek()
#      └── tell()
#
#
#     with
#      │
#      ↓
#     automatic cleanup
#
#
#     PATH
#      │
#      ↓
#     pathlib.Path
#      │
#      ├── exists()
#      ├── is_file()
#      ├── mkdir()
#      └── glob()
#
#
#     TABLE
#      │
#      ↓
#     csv
#      │
#      ├── reader
#      ├── DictReader
#      ├── writer
#      └── DictWriter
#
#
#     NESTED DATA
#      │
#      ↓
#     json
#      │
#      ├── load / loads
#      └── dump / dumps
#
#
#     RAW DATA
#        ↓
#      READ
#        ↓
#      CLEAN
#        ↓
#     CONVERT
#        ↓
#     VALIDATE
#        ↓
#    TRANSFORM
#        ↓
#     ANALYZE
#        ↓
#      SAVE
#
#
# ============================================================
# 167. THE MOST IMPORTANT RULES TO REMEMBER
# ============================================================
#
# (1) input() returns str.
#
# (2) print() writes normal output.
#
# (3) Use with open(...) for files.
#
# (4) Explicitly specify encoding for text files when practical.
#
# (5) "w" overwrites.
#
# (6) "a" appends.
#
# (7) "x" creates and refuses to overwrite an existing file.
#
# (8) CSV values initially come in as strings.
#
# (9) Use csv instead of split(",") for real CSV parsing.
#
# (10) JSON:
#
#      load  = file → Python
#      loads = string → Python
#      dump  = Python → file
#      dumps = Python → string
#
# (11) Use pathlib for modern path manipulation.
#
# (12) Large files should usually be processed incrementally.
#
# (13) Keep raw data separate from cleaned data.
#
# (14) File handling is part of data engineering, not merely
#      a beginner programming topic.
#
# (15) The important question is always:
#
#          Where does the data come from?
#                  ↓
#             How is it represented?
#                  ↓
#             How do I process it?
#                  ↓
#             Where should the result go?
#
#
# ============================================================
# 168. FINAL "PYTHON FEELING"
# ============================================================
#
# If you truly understand file handling, you should start
# thinking like this:
#
#
#     DATA IS A FLOW.
#
#
#     SOURCE
#       ↓
#     READ
#       ↓
#     REPRESENT
#       ↓
#     CLEAN
#       ↓
#     VALIDATE
#       ↓
#     TRANSFORM
#       ↓
#     ANALYZE
#       ↓
#     WRITE
#       ↓
#     RESULT
#
#
# Python gives you several levels of control:
#
#     input()
#         ↓
#     basic interaction
#
#     open()
#         ↓
#     raw file access
#
#     pathlib
#         ↓
#     filesystem organization
#
#     csv / json
#         ↓
#     structured data
#
#     generators / chunks
#         ↓
#     scalable processing
#
#     pandas / databases
#         ↓
#     high-level data analysis
#
#
# So the real foundation is not:
#
#     "memorize read(), write(), seek()..."
#
#
# It is:
#
#     UNDERSTAND THE DATA FLOW.
#
#
#     input
#       ↓
#     data
#       ↓
#     Python objects
#       ↓
#     processing
#       ↓
#     output
#
#
# Once this mental model becomes natural, CSV, JSON, pandas,
# SQL, ETL, automation, Power BI pipelines, and ML datasets
# become much easier to understand.
#
#
# ============================================================
# END OF FILE HANDLING + INPUT / OUTPUT COMPLETE GUIDE
# ============================================================
```
