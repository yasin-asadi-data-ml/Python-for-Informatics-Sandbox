# DATA ANALYTICS EXAMPLES


```python
# ============================================================
# 46. DATA ANALYTICS — CLASSIFYING VALUES
# ============================================================

# This example classifies each salary into a category.
#
# Basic idea:
# salary → compare with thresholds → choose category → print


salaries = [30000, 45000, 70000, 90000, 120000]

# salaries is the collection being processed.
# The for loop takes one value at a time and stores it in salary.

for salary in salaries:

    # salary receives the current item from salaries.
    # Each iteration uses a different salary.

    if salary < 40000:
        # < compares two values and produces True or False.
        # If True, this block executes.

        category = "Low"

    elif salary < 70000:
        # elif is checked only if the previous condition was False.
        # This classifies values from 40000 up to, but not including, 70000.

        category = "Medium"

    else:
        # else executes when both previous conditions are False.
        # Therefore, salary >= 70000 is classified as "High".

        category = "High"

    # print() displays the current salary and its category.
    # print() displays values; it does not return the displayed values.

    print(salary, category)


# ============================================================
# HOW IT WORKS
# ============================================================

# Python processes one salary at a time:
#
# salary
#   ↓
# salary < 40000?
#   ↓
# True → "Low"
# False
#   ↓
# salary < 70000?
#   ↓
# True → "Medium"
# False → "High"
#   ↓
# print result
#   ↓
# next salary


# ============================================================
# EXECUTION TRACE
# ============================================================

# Step | salary | Decision                  | category
# -----|--------|---------------------------|----------
#  1   | 30000  | 30000 < 40000 → True     | Low
#  2   | 45000  | 45000 < 70000 → True     | Medium
#  3   | 70000  | both conditions → False  | High
#  4   | 90000  | both conditions → False  | High
#  5   | 120000 | both conditions → False  | High
#
# Only ONE branch of the if/elif/else chain executes
# for each salary.


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# salaries contains five values.
#
# DURING:
# Take one salary → evaluate conditions → assign category
# → print the result.
#
# AFTER:
# The iteration finishes and Python moves to the next salary.


# ============================================================
# STATE CHANGES
# ============================================================

# WHAT CHANGES?
# salary changes on each iteration.
# category receives the classification for the current salary.
#
# WHAT STAYS THE SAME?
# salaries and the classification rules remain unchanged.


# ============================================================
# MENTAL MODEL
# ============================================================

# Think of if / elif / else as a decision machine:
#
# value
#   ↓
# check condition
#   ↓
# True / False
#   ↓
# choose ONE branch
#   ↓
# produce a result
#
# The same decision process is repeated for every item.


# ============================================================
# WHY THIS WORKS
# ============================================================

# Conditions are checked from top to bottom.
#
# 30000 < 40000 → True → "Low"
#
# 45000 < 40000 → False
# 45000 < 70000 → True → "Medium"
#
# 70000 < 40000 → False
# 70000 < 70000 → False
# → else → "High"


# ============================================================
# SYNTAX → MEANING
# ============================================================

# for salary in salaries:
# for      → starts the loop
# salary   → loop variable receiving the current item
# in       → takes items from the collection one at a time
# salaries → collection being processed
#
# if / elif / else:
# if       → checks the first condition
# elif     → checks another condition if the previous one was False
# else     → executes when all previous conditions were False


# ============================================================
# CORE IDEA:
# ============================================================

# A loop processes each value.
#
# if / elif / else decides what to do with the current value.
#
# collection → current item → condition → decision → result


# ============================================================
# COMMON MISTAKE
# ============================================================

# A common mistake is thinking every if/elif/else block runs.
#
# Only the FIRST True branch runs.
#
# For 45000:
# 45000 < 40000 → False
# 45000 < 70000 → True
# → "Medium"
# → else is skipped.


# ============================================================
# OUTPUT
# ============================================================

# The program displays:
#
# 30000 Low
# 45000 Medium
# 70000 High
# 90000 High
# 120000 High
#
# Each output is the current salary followed by
# the category selected by the conditions.
```

```python
# ============================================================
# 47. DATA ANALYTICS — FILTERING
# ============================================================

# This example filters a list and keeps only sales
# that are greater than or equal to 500.
#
# Basic idea:
# sales → check each value → keep matching values → new list


sales = [100, 250, 500, 750, 1200]

# sales is the original collection.
# The values in sales will be checked one at a time.

high_sales = []

# Start with an empty list.
# Matching sales will be added to this list during the loop.


for sale in sales:

    # sale receives the current value from sales.

    if sale >= 500:
        # >= checks whether sale is greater than or equal to 500.
        # The result is True or False.
        #
        # If True, the current sale is kept.

        high_sales.append(sale)
        # append() adds sale to the end of high_sales.
        #
        # Example:
        # high_sales = [500]
        # then → [500, 750]
        # then → [500, 750, 1200]


print(high_sales)

# print() displays the final filtered list.
# It does not return the displayed list.


# ============================================================
# HOW IT WORKS
# ============================================================

# Python processes each sale:
#
# sale
#   ↓
# sale >= 500?
#   ↓
# True → append sale to high_sales
# False → skip it
#   ↓
# next sale
#   ↓
# repeat


# ============================================================
# EXECUTION TRACE
# ============================================================

# Step | sale | sale >= 500 | high_sales
# -----|------|-------------|-------------------
# Start|  -   |      -      | []
#  1   | 100  | False       | []
#  2   | 250  | False       | []
#  3   | 500  | True        | [500]
#  4   | 750  | True        | [500, 750]
#  5   | 1200 | True        | [500, 750, 1200]


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# sales contains all sales.
# high_sales is empty.
#
# DURING:
# Take one sale → check the condition
# → append it only if the condition is True.
#
# AFTER:
# high_sales contains only sales >= 500.


# ============================================================
# MENTAL MODEL
# ============================================================

# Filtering means:
#
# collection
#    ↓
# examine each item
#    ↓
# condition
#    ↓
# keep or skip
#    ↓
# filtered collection
#
# The original sales list is not changed.
# A separate list stores the values that pass the condition.


# ============================================================
# WHAT CHANGES / WHAT STAYS THE SAME
# ============================================================

# WHAT CHANGES?
# sale changes on each iteration.
# high_sales grows when a sale satisfies the condition.
#
# WHAT STAYS THE SAME?
# sales remains unchanged.
# The filtering rule sale >= 500 remains unchanged.


# ============================================================
# CORE IDEA:
# ============================================================

# Filtering means selecting items from a collection
# based on a condition.
#
# Here:
#
# sale >= 500 → keep
# sale < 500  → skip


# ============================================================
# COMMON MISTAKE
# ============================================================

# Do not confuse append() with replacing the list.
#
# high_sales.append(sale)
# adds one item to the existing list.
#
# It does NOT create a new list each time.


# ============================================================
# OUTPUT
# ============================================================

# The output is:
#
# [500, 750, 1200]
#
# These are the only values in sales that satisfy:
#
# sale >= 500
```

```python
# ============================================================
# 48. DATA ANALYTICS — CALCULATING TOTAL SALES
# ============================================================

# This example calculates the total of all sales.
#
# Basic idea:
# sales → take each value → add it to total → final total


sales = [100, 250, 500, 750, 1200]

# sales is the collection containing the values to be added.


total_sales = 0

# Start the accumulator at 0.
# The total will grow as each sale is processed.


for sale in sales:

    # sale receives the current value from sales.

    total_sales += sale
    # += takes the current value of total_sales,
    # adds sale to it, and stores the new value back.
    #
    # This is equivalent to:
    # total_sales = total_sales + sale


print("Total sales:", total_sales)

# print() displays the final accumulated value.
# It does not return the displayed value.


# ============================================================
# HOW IT WORKS
# ============================================================

# Python processes one sale at a time:
#
# sale
#   ↓
# add sale to total_sales
#   ↓
# update total_sales
#   ↓
# next sale
#   ↓
# repeat
#   ↓
# final total


# ============================================================
# EXECUTION TRACE
# ============================================================

# Step | sale | total_sales
# -----|------|------------
# Start|  -   | 0
#  1   | 100  | 0 + 100 = 100
#  2   | 250  | 100 + 250 = 350
#  3   | 500  | 350 + 500 = 850
#  4   | 750  | 850 + 750 = 1600
#  5   | 1200 | 1600 + 1200 = 2800


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# total_sales = 0
#
# DURING:
# Take each sale → add it to total_sales → update the total.
#
# AFTER:
# total_sales = 2800


# ============================================================
# STATE CHANGES
# ============================================================

# WHAT CHANGES?
# sale changes on each iteration.
# total_sales increases after every sale.
#
# WHAT STAYS THE SAME?
# sales remains unchanged.
#
# State:
# 0 → 100 → 350 → 850 → 1600 → 2800


# ============================================================
# MENTAL MODEL
# ============================================================

# Think of total_sales as a running total:
#
# starting value
#      ↓
# add current item
#      ↓
# updated total
#      ↓
# add next item
#      ↓
# repeat
#
# This pattern is called accumulation.


# ============================================================
# WHY THIS WORKS
# ============================================================

# total_sales += sale means:
#
# current total
#      +
# current sale
#      ↓
# new total
#
# Because total_sales keeps its updated value,
# each new sale is added to the previous total.


# ============================================================
# CORE IDEA:
# ============================================================

# An accumulator stores a running result while a loop
# processes a collection.
#
# Here:
#
# total_sales = 0
#      ↓
# add each sale
#      ↓
# total_sales = 2800


# ============================================================
# COMMON MISTAKE
# ============================================================

# Do not initialize total_sales inside the loop:
#
# for sale in sales:
#     total_sales = 0
#     total_sales += sale
#
# That would reset the accumulator on every iteration.
#
# It must be initialized BEFORE the loop so its value
# can carry from one iteration to the next.


# ============================================================
# OUTPUT
# ============================================================

# The output is:
#
# Total sales: 2800
#
# The result comes from:
# 100 + 250 + 500 + 750 + 1200 = 2800
```

```python
# ============================================================
# 49. DATA ANALYTICS — AVERAGE SALES
# ============================================================

# This example calculates the average of all sales.
#
# Basic idea:
# sales → calculate total → count values → divide total by count
#
# Average = total ÷ number of values


sales = [100, 250, 500, 750, 1200]

# sales is the collection containing the values to analyze.


total_sales = 0

# Start the accumulator at 0.
# It will store the running total of the sales.


for sale in sales:
    # sale receives one value from sales on each iteration.

    total_sales += sale

    # Add the current sale to the running total.
    #
    # This is equivalent to:
    # total_sales = total_sales + sale


average_sales = total_sales / len(sales)

# len(sales) returns the number of items in the list.
#
# len(sales) → 5
# total_sales → 2800
#
# Therefore:
# average_sales = 2800 / 5
# average_sales = 560.0
#
# / performs true division, so the result is a float.


print("Average:", average_sales)

# print() displays the calculated average.
# It does not return the displayed value.


# ============================================================
# HOW IT WORKS
# ============================================================

# Python first calculates the total:
#
# 0 → 100 → 350 → 850 → 1600 → 2800
#
# Then:
#
# total_sales = 2800
#       ↓
# len(sales) = 5
#       ↓
# 2800 / 5
#       ↓
# average_sales = 560.0
#       ↓
# print result


# ============================================================
# EXECUTION TRACE
# ============================================================

# Step | sale | total_sales
# -----|------|------------
# Start|  -   | 0
#  1   | 100  | 100
#  2   | 250  | 350
#  3   | 500  | 850
#  4   | 750  | 1600
#  5   | 1200 | 2800
#
# After the loop:
# len(sales) = 5
# 2800 / 5 = 560.0


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# sales contains 5 values.
# total_sales = 0.
#
# DURING:
# Add every sale to total_sales.
#
# AFTER:
# total_sales = 2800
# number of values = 5
# average_sales = 560.0


# ============================================================
# MENTAL MODEL
# ============================================================

# Think of an average as:
#
# TOTAL
#   ÷
# NUMBER OF VALUES
#   ↓
# AVERAGE
#
# Here:
#
# 2800 ÷ 5 = 560.0


# ============================================================
# STATE CHANGES
# ============================================================

# WHAT CHANGES?
# sale changes on each loop iteration.
# total_sales increases as each sale is added.
#
# WHAT STAYS THE SAME?
# sales remains unchanged.
#
# After the loop, total_sales contains the final total.


# ============================================================
# WHY THIS WORKS
# ============================================================

# An arithmetic average is calculated by dividing
# the sum of the values by the number of values.
#
# total_sales
#     ↓
# sum of all sales
#
# len(sales)
#     ↓
# number of sales
#
# division
#     ↓
# average_sales


# ============================================================
# CORE IDEA:
# ============================================================

# To calculate an average:
#
# 1. Find the total.
# 2. Count the values.
# 3. Divide the total by the count.
#
# In this example:
# 2800 ÷ 5 = 560.0


# ============================================================
# COMMON MISTAKE
# ============================================================

# Do not divide by a fixed number if the list can change.
#
# len(sales) is used because it automatically counts
# how many values are currently in the list.
#
# Important edge case:
# If sales were empty, len(sales) would be 0,
# and dividing by zero would cause ZeroDivisionError.


# ============================================================
# OUTPUT
# ============================================================

# The output is:
#
# Average: 560.0
#
# The .0 appears because / produces a floating-point result.
```

```python
# ============================================================
# 50. DATA ANALYTICS — FINDING THE BEST SALE
# ============================================================

# This example finds the largest value in a list of sales.
#
# Basic idea:
# first sale → assume it is the best → compare each sale
# → update best_sale when a larger value is found


sales = [100, 250, 500, 750, 1200]

# sales is the collection being analyzed.


best_sale = sales[0]

# Start by using the first sale as the current best sale.
#
# sales[0] means:
# "Get the item at index 0."
#
# Index 0 is the first item in a Python list.
#
# Therefore:
# best_sale = 100


for sale in sales:

    # sale receives one value from sales on each iteration.
    #
    # Python compares the current sale with the best value
    # found so far.

    if sale > best_sale:

        # > checks whether sale is greater than best_sale.
        # The comparison produces True or False.
        #
        # If True, a new largest value has been found.

        best_sale = sale

        # Replace the previous best_sale with the new larger value.


print("Best sale:", best_sale)

# print() displays the final largest sale.
# It does not return the displayed value.


# ============================================================
# HOW IT WORKS
# ============================================================

# Python keeps a "best value so far":
#
# first value
#     ↓
# best_sale
#     ↓
# compare next sale
#     ↓
# larger? → update best_sale
#     ↓
# repeat
#     ↓
# final best_sale


# ============================================================
# EXECUTION TRACE
# ============================================================

# Start:
# best_sale = sales[0] = 100
#
# Step | sale | sale > best_sale | best_sale
# -----|------|------------------|----------
#  1   | 100  | False            | 100
#  2   | 250  | True             | 250
#  3   | 500  | True             | 500
#  4   | 750  | True             | 750
#  5   | 1200 | True             | 1200
#
# Final:
# best_sale = 1200


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# sales contains the values.
# best_sale starts as the first value: 100.
#
# DURING:
# Compare each sale with best_sale.
# If the current sale is larger, update best_sale.
#
# AFTER:
# best_sale contains the largest value: 1200.


# ============================================================
# MENTAL MODEL
# ============================================================

# Think of best_sale as a record holder.
#
# Current record = 100
#     ↓
# 250 is larger → new record = 250
#     ↓
# 500 is larger → new record = 500
#     ↓
# 750 is larger → new record = 750
#     ↓
# 1200 is larger → new record = 1200
#
# The variable always stores the best value found so far.


# ============================================================
# STATE CHANGES
# ============================================================

# WHAT CHANGES?
# sale changes on every iteration.
# best_sale changes only when a larger sale is found.
#
# WHAT STAYS THE SAME?
# sales remains unchanged.
#
# best_sale changes:
# 100 → 250 → 500 → 750 → 1200


# ============================================================
# WHY THIS WORKS
# ============================================================

# The first value provides a valid starting point.
#
# Each new sale is compared with the largest value found so far.
#
# If:
# sale > best_sale
#
# then sale must be a better candidate,
# so best_sale is updated.
#
# After every value has been checked,
# no remaining value can be larger than best_sale.


# ============================================================
# CORE IDEA:
# ============================================================

# Finding a maximum manually follows this pattern:
#
# 1. Start with a candidate.
# 2. Compare each value with the candidate.
# 3. Replace the candidate when a larger value is found.
# 4. The final candidate is the largest value.


# ============================================================
# COMMON MISTAKE
# ============================================================

# Do not start with:
#
# best_sale = 0
#
# if the data could contain negative values.
#
# Using sales[0] makes the algorithm work correctly
# regardless of whether the sales are positive or negative,
# as long as the list is not empty.
#
# Important edge case:
# sales[0] causes IndexError if sales is an empty list.


# ============================================================
# OUTPUT
# ============================================================

# The output is:
#
# Best sale: 1200
#
# 1200 is the largest value in sales.
```

```python
# ============================================================
# 51. DATA ANALYTICS — DATA QUALITY CHECK
# ============================================================

# This example checks whether age values look valid.
#
# Each age is classified into one of three categories:
#
# age < 0    → Invalid
# age > 120  → Suspicious
# otherwise  → Valid
#
# Basic idea:
# data → check each value → apply rules → classify value


ages = [25, 30, -5, 40, 150, 35]

# ages is the collection being checked.
# The list contains both normal and potentially problematic values.


for age in ages:

    # age receives one value from ages during each iteration.
    # Python checks the values one at a time.


    if age < 0:
        # Compare age with 0.
        # If age is negative, the condition is True.
        #
        # Negative ages are treated as invalid data.

        print(age, "Invalid")


    elif age > 120:
        # This condition is checked only if age < 0 was False.
        #
        # Values above 120 are not automatically declared invalid;
        # they are marked "Suspicious" so they can be investigated.

        print(age, "Suspicious")


    else:
        # else executes when both previous conditions are False.
        #
        # Therefore:
        # age >= 0 AND age <= 120
        #
        # The value is considered valid by these rules.

        print(age, "Valid")


# ============================================================
# HOW IT WORKS
# ============================================================

# Each age passes through the same decision process:
#
# age
#  ↓
# age < 0?
#  ↓
# Yes → "Invalid"
# No
#  ↓
# age > 120?
#  ↓
# Yes → "Suspicious"
# No → "Valid"
#  ↓
# print result
#  ↓
# next age


# ============================================================
# EXECUTION TRACE
# ============================================================

# Step | age | First check | Second check | Result
# -----|-----|-------------|--------------|----------
#  1   | 25  | False       | False        | Valid
#  2   | 30  | False       | False        | Valid
#  3   | -5  | True        | not checked  | Invalid
#  4   | 40  | False       | False        | Valid
#  5   | 150 | False       | True         | Suspicious
#  6   | 35  | False       | False        | Valid
#
# "not checked" is important:
# Once an if condition is True, Python skips the remaining
# elif/else branches for that iteration.


# ============================================================
# BEFORE → DURING → AFTER
# ============================================================

# BEFORE:
# ages contains the data to be checked.
#
# DURING:
# Take one age → check the rules → print its classification.
#
# AFTER:
# Every age has been checked and classified.


# ============================================================
# MENTAL MODEL
# ============================================================

# Think of the if/elif/else structure as a data-quality filter:
#
#                 age
#                  ↓
#             check rule 1
#              /       \
#           True       False
#            ↓           ↓
#         Invalid     check rule 2
#                       /      \
#                    True       False
#                     ↓           ↓
#                Suspicious     Valid


# ============================================================
# WHY THIS WORKS
# ============================================================

# Conditions produce True or False.
#
# For age = -5:
# -5 < 0 → True
# → "Invalid"
#
# For age = 150:
# 150 < 0 → False
# 150 > 120 → True
# → "Suspicious"
#
# For age = 40:
# 40 < 0 → False
# 40 > 120 → False
# → else → "Valid"


# ============================================================
# STATE CHANGES
# ============================================================

# WHAT CHANGES?
# age changes on each iteration.
#
# WHAT STAYS THE SAME?
# ages remains unchanged.
# The three data-quality rules remain unchanged.
#
# This example classifies data without modifying
# the original list.


# ============================================================
# CORE IDEA:
# ============================================================

# Data quality checks use rules to identify values
# that are valid, invalid, or worth investigating.
#
# Here:
#
# negative → Invalid
# unusually high → Suspicious
# otherwise → Valid


# ============================================================
# COMMON MISTAKE
# ============================================================

# "Suspicious" is different from "Invalid".
#
# The program does not say that every value above 120
# is definitely wrong.
#
# It only flags the value for further investigation.
#
# Also, because the conditions are checked from top to bottom,
# the order of the rules can affect the classification.


# ============================================================
# OUTPUT
# ============================================================

# The output is:
#
# 25 Valid
# 30 Valid
# -5 Invalid
# 40 Valid
# 150 Suspicious
# 35 Valid
#
# Each line contains the original age followed by
# the category produced by the data-quality rules.
```

```python
# ============================================================
# 52. DATA ANALYTICS — MISSING VALUES
# ============================================================

# MISSING VALUES
#
# A missing value means that a piece of data is not available,
# not recorded, or intentionally left empty.
#
# In Python, None is commonly used to represent "no value".
#
# In data analytics, detecting missing values is important because
# missing data can affect calculations, statistics, models, and
# the reliability of conclusions drawn from a dataset.
#
# This example demonstrates a simple and reusable pattern:
#
#     COLLECTION
#          ↓
#     examine each item
#          ↓
#     check whether the item is missing
#          ↓
#     count missing items
#          ↓
#     report the final count


values = [10, 20, None, 40, None, 60]

# values is the SOURCE COLLECTION.
#
# Actual contents:
#
#     position →  0    1     2    3     4    5
#     value    → 10   20   None   40   None  60
#
# There are 6 items in total.
#
# Two of those items are None.
#
# IMPORTANT:
# None is not the same thing as 0.
#
#     0     → an actual numeric value
#     None  → absence of a value
#
# Therefore, None should not automatically be treated as zero.


missing_count = 0

# missing_count is a COUNTER.
#
# It stores how many missing values have been found so far.
#
# BEFORE the loop:
#
#     missing_count = 0
#
# This is called initializing the counter.
#
# The counter starts at zero because Python has not examined
# any values yet, so no missing values have been counted.
#
# The state will change as the loop finds None values.
#
# Conceptually:
#
#     INITIAL STATE
#     missing_count = 0
#
#             ↓
#
#     inspect values
#
#             ↓
#
#     find missing value
#
#             ↓
#
#     missing_count increases by 1


for value in values:

    # The for loop processes the collection ONE ITEM AT A TIME.
    #
    # Each iteration:
    #
    #     values
    #       ↓
    #     next item
    #       ↓
    #     value receives that item
    #       ↓
    #     execute the loop body
    #       ↓
    #     move to the next item
    #
    # Actual iterations:
    #
    #     Iteration 1 → value = 10
    #     Iteration 2 → value = 20
    #     Iteration 3 → value = None
    #     Iteration 4 → value = 40
    #     Iteration 5 → value = None
    #     Iteration 6 → value = 60
    #
    # The variable value therefore represents the CURRENT ITEM,
    # not the entire values list.

    if value is None:

        # Python evaluates:
        #
        #     value is None
        #
        # The operator "is" checks object identity.
        #
        # In this particular situation, "is None" is the standard
        # and appropriate way to test whether a variable contains
        # the special None object.
        #
        # The condition therefore asks:
        #
        #     "Is the current value actually None?"
        #
        # The result is a Boolean:
        #
        #     True  → the current value is missing
        #     False → the current value is not None
        #
        # IMPORTANT:
        # "is None" is different from:
        #
        #     value == 0
        #
        # because zero is a real numeric value, while None represents
        # the absence of a value.

        missing_count += 1

        # This is a COUNTER UPDATE.
        #
        # Conceptually:
        #
        #     missing_count += 1
        #
        # means:
        #
        #     missing_count = missing_count + 1
        #
        # Python:
        #
        #     1. retrieves the current missing_count
        #     2. adds 1
        #     3. produces the new value
        #     4. assigns the new value back to missing_count
        #
        # Example:
        #
        #     Before finding first None:
        #         missing_count = 0
        #
        #     0 + 1
        #         ↓
        #         1
        #
        #     After:
        #         missing_count = 1
        #
        # When the second None is found:
        #
        #     1 + 1
        #         ↓
        #         2
        #
        #     missing_count = 2
        #
        # Notice the important relationship:
        #
        #     value
        #       ↓
        #     check for None
        #       ↓
        #     True
        #       ↓
        #     increase counter
        #       ↓
        #     updated missing_count
        #
        # If the condition is False, this statement is skipped.


print("Missing values:", missing_count)

# After the loop has processed every item,
# missing_count contains the FINAL number of missing values.
#
# ============================================================
# EXECUTION TRACE
# ============================================================
#
#     Step | value | value is None | missing_count
#     -----|-------|---------------|---------------
#     Start|   -   |       -       |       0
#       1  |  10   |     False     |       0
#       2  |  20   |     False     |       0
#       3  | None  |      True     |       1
#       4  |  40   |     False     |       1
#       5  | None  |      True     |       2
#       6  |  60   |     False     |       2
#
# The counter changes ONLY when the current value is None.
#
# ============================================================
# BEFORE → DURING → AFTER
# ============================================================
#
# BEFORE:
#
#     values = [10, 20, None, 40, None, 60]
#     missing_count = 0
#
# DURING:
#
#     10   → not missing → count stays 0
#     20   → not missing → count stays 0
#     None → missing     → count becomes 1
#     40   → not missing → count stays 1
#     None → missing     → count becomes 2
#     60   → not missing → count stays 2
#
# AFTER:
#
#     missing_count = 2
#
# ============================================================
# VALUE FLOW
# ============================================================
#
#     values
#       ↓
#     current item
#       ↓
#     value
#       ↓
#     value is None
#       ↓
#     True / False
#       ↓
#     if True → increase missing_count
#       ↓
#     final count
#       ↓
#     print()
#
# ============================================================
# MENTAL MODEL
# ============================================================
#
# Think of this program as a person inspecting a list one item
# at a time while keeping a tally:
#
#     "Is this item missing?"
#
#     No  → continue
#     Yes → add 1 to the tally
#
# After every item has been inspected, the tally tells us
# how many missing values were found.
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# This is a very common COUNTING PATTERN:
#
#     counter = 0
#
#     for item in collection:
#         if condition:
#             counter += 1
#
# In general:
#
#     initialize counter
#          ↓
#     inspect each item
#          ↓
#     test a condition
#          ↓
#     if condition is True
#          ↓
#     increase counter
#
# The condition can change depending on the problem.
#
# For example:
#
#     if value is None:
#         → count missing values
#
#     if value < 0:
#         → count negative values
#
#     if value > 100:
#         → count unusually large values
#
#     if value == 0:
#         → count zero values
#
# The STRUCTURE remains the same.
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# Do not confuse:
#
#     None
#
# with:
#
#     0
#
# or:
#
#     ""
#
# or:
#
#     False
#
# These represent different values/concepts.
#
# In this example, only None is considered a missing value
# because that is the condition explicitly being tested.
#
# ============================================================
# OUTPUT
# ============================================================
#
# The two None values are detected:
#
#     [10, 20, None, 40, None, 60]
#             ↑             ↑
#          missing       missing
#
# Therefore:
#
#     missing_count = 2
#
# Output:
#
#     Missing values: 2
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A loop can inspect every item in a collection, while a counter
# can maintain state across iterations.
#
# Here, each item is tested for None, and the counter increases
# only when a missing value is found.
#
# The reusable pattern is:
#
#     examine each item
#          ↓
#     test a condition
#          ↓
#     count the items that satisfy the condition
```

```python
# ============================================================
# 53. DATA ANALYTICS — IGNORING MISSING VALUES
# ============================================================

# This example demonstrates an important data-analysis pattern:
# calculating a statistic while safely ignoring missing values.
#
# In this dataset:
#
# 10, 20, None, 40, None, 60
#
# None represents a missing value.
#
# The goal is NOT to treat None as a number.
# Instead, Python should skip it and calculate the average
# using only the available numerical values.

values = [10, 20, None, 40, None, 60]

# total = running sum of valid numerical values.
# count = number of valid numerical values processed.
#
# Together, they allow us to calculate:
#
# average = total / count
#
# These are called ACCUMULATORS because their values
# are updated as the loop processes the data.

total = 0
count = 0

# ============================================================
# CONCEPTUAL FLOW
# ============================================================
#
# values
#   ↓
# examine each value
#   ↓
# Is value missing?
#   ↓
# YES → skip it with continue
# NO  → add it to total and increase count
#   ↓
# after the loop:
# total / count
#   ↓
# average
#
# The important idea is:
#
# "Process valid data, ignore missing data."

for value in values:

    # value is the CURRENT item from the list.
    #
    # None is a special Python value commonly used
    # to represent the absence of a value.
    #
    # "is None" checks whether value is specifically
    # the None object.
    #
    # It is preferable here to:
    #
    #     value is None
    #
    # rather than:
    #
    #     value == None
    #
    # because "is" checks object identity and is the
    # standard way to test for None.

    if value is None:

        # continue immediately skips the rest of the
        # current loop iteration.
        #
        # Therefore, None is NOT added to total
        # and count is NOT increased.
        #
        # Then Python moves to the next item in values.

        continue

    # ========================================================
    # MICRO-EXECUTION
    # ========================================================
    #
    # total += value
    #
    # means conceptually:
    #
    #     total = total + value
    #
    # Example:
    #
    # total = 30
    # value = 40
    #
    # 30 + 40
    #   ↓
    # 70
    #   ↓
    # total = 70
    #
    # So total maintains the RUNNING SUM.

    total += value

    # count tracks how many VALID numerical values
    # have been included in the calculation.
    #
    # It increases only when value is not None.
    #
    # This is important because dividing by the total
    # number of list elements would incorrectly include
    # the missing values.

    count += 1

# ============================================================
# EXECUTION TRACE
# ============================================================
#
# Start:
# total = 0
# count = 0
#
# value = 10:
# total = 10
# count = 1
#
# value = 20:
# total = 30
# count = 2
#
# value = None:
# skipped
# total = 30
# count = 2
#
# value = 40:
# total = 70
# count = 3
#
# value = None:
# skipped
# total = 70
# count = 3
#
# value = 60:
# total = 130
# count = 4
#
# Final state:
#
# total = 130
# count = 4
#
# The two None values never affect either accumulator.

# ============================================================
# CALCULATING THE AVERAGE
# ============================================================
#
# Mathematical idea:
#
# average = sum of valid values / number of valid values
#
# Unicode notation:
#
# average = 130 ÷ 4 = 32.5
#
# Python:
#
# average = total / count
#
# At this point:
#
# total = 130
# count = 4
#
# Therefore:
#
# 130 / 4
#   ↓
# 32.5

average = total / count

# print() displays the final value.
#
# It does not calculate the average itself.
# The calculation has already happened above.

print("Average:", average)

# ============================================================
# HOW IT WORKS
# ============================================================
#
# Python processes the collection one item at a time.
#
# Numeric value
#     ↓
# include it
#     ↓
# update total
#     ↓
# update count
#
# None
#     ↓
# skip with continue
#     ↓
# move to next item
#
# After all items are processed:
#
# running total = 130
# valid-value count = 4
#     ↓
# 130 / 4
#     ↓
# average = 32.5
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# This pattern is widely useful in data analytics:
#
# collection
#     ↓
# inspect each item
#     ↓
# identify invalid/missing values
#     ↓
# skip invalid values
#     ↓
# process valid values
#     ↓
# maintain accumulators
#     ↓
# calculate the final statistic
#
# The key pattern is:
#
#     if value is missing:
#         skip it
#     otherwise:
#         process it
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# Do NOT simply add None to total.
#
#     total += None
#
# would raise a TypeError because None cannot be added
# to an integer.
#
# Also, count should represent VALID values, not all
# values in the list.
#
# Here there are 6 list elements but only 4 valid numbers.
#
# ============================================================
# EDGE CASE
# ============================================================
#
# If every value were None, count would remain 0.
# Then:
#
#     total / count
#
# would cause a ZeroDivisionError.
#
# A production data-analysis program would normally
# check whether count > 0 before calculating the average.
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Missing data should be handled explicitly.
#
# The loop filters out None values while simultaneously
# maintaining the sum and count of valid observations.
#
# Then:
#
#     average = valid_sum / valid_count
#
# This is a reusable pattern for calculating statistics
# from incomplete datasets.
#
# OUTPUT:
#
# Average: 32.5
```

```python
# ============================================================
# 54. DATA ANALYTICS — OUTLIER DETECTION
# ============================================================

values = [10, 12, 15, 14, 13, 100]

mean = sum(values) / len(values)

for value in values:

    if value > mean * 2:
        print("Potential outlier:", value)

```

```python
# ============================================================
# 55. DATA ANALYTICS — LOOPING THROUGH COLUMNS
# ============================================================

import pandas as pd

df = pd.DataFrame({
    "Age": [25, 30, 35],
    "Salary": [30000, 50000, 70000],
    "Experience": [2, 5, 8]
})

for column in df.columns:
    print(column)

```

```python
# ============================================================
# 56. LOOPING THROUGH DATAFRAME ROWS
# ============================================================

# Loop through each row in the DataFrame using iterrows() method
# iterrows() returns an iterator that yields index and row data
for index, row in df.iterrows():

    # Print the index of the current row
    print(index)
    # Print the data in the current row as a pandas Series
    print(row)

```

```python
# ============================================================
# 57. LOOPING THROUGH DATAFRAME ROW VALUES
# ============================================================

# Iterate through each row of the DataFrame using iterrows()
# This method returns an iterator yielding index and row data
for index, row in df.iterrows():

    # Print the values of 'Age' and 'Salary' columns for each row
    # The values are accessed using column names as keys
    print(
        row["Age"],  # Access the value in the 'Age' column
        row["Salary"]  # Access the value in the 'Salary' column
    )
```

```python
# ============================================================
# 58. APPLYING CONDITIONS TO DATAFRAME ROWS
# ============================================================
#
# Use iterrows() to process a DataFrame row by row.
#
#     index → current row index
#     row   → current row
#
# Pattern:
#
#     for index, row in df.iterrows():
#         if condition:
#             action
#
# ============================================================

for index, row in df.iterrows():

    # Check if the current row has a salary above 50,000
    if row["Salary"] > 50000:

        # Print the age and a message for matching rows
        print(row["Age"], "High salary")


# ============================================================
# PYTHON FEELING
# ============================================================
#
#     for row → examine each row
#     if      → check a condition
#     action  → do something if True
#
# Think:
#
#     LOOP → CHECK → ACT
#
# ============================================================
#
# DATA SCIENCE TIP:
#
# For simple filtering, prefer Pandas Boolean masking:
#
#     df.loc[df["Salary"] > 50000, "Age"]
#
# This is generally more efficient than iterrows().
#
# ============================================================
```

```python
# ============================================================
# 59. LOOPING THROUGH COLUMNS FOR DATA QUALITY
# ============================================================
#
# Before analyzing or building a Machine Learning model, we
# need to check the quality of our data.
#
# One important data-quality check is finding MISSING VALUES.
#
# Pandas provides:
#
#     .isnull()
#
# to identify missing values.
#
#     .sum()
#
# then counts how many missing values exist.
#
# The pattern is:
#
#     df[column].isnull().sum()
#
# Think:
#
#     DataFrame
#         ↓
#     One column at a time
#         ↓
#     Find missing values
#         ↓
#     Count them
#
# ============================================================


# ============================================================
# LOOP THROUGH COLUMNS
# ============================================================
#
#     df.columns
#
# provides the names of all columns in the DataFrame.
#
# The loop processes each column one at a time.
#
# ============================================================

for column in df.columns:


    # ========================================================
    # COUNT MISSING VALUES
    # ========================================================
    #
    #     df[column]
    #
    # selects the current column.
    #
    #     .isnull()
    #
    # checks every value and returns:
    #
    #     True  → value is missing
    #     False → value is not missing
    #
    #     .sum()
    #
    # counts the True values because:
    #
    #     True  = 1
    #     False = 0
    #
    # Therefore:
    #
    #     df[column].isnull().sum()
    #
    # gives the total number of missing values.
    #
    # ========================================================

    missing = df[column].isnull().sum()


    # ========================================================
    # PRINT RESULT
    # ========================================================
    #
    # Display the column name together with its number of
    # missing values.
    #
    # ========================================================

    print(
        column,
        "missing values:",
        missing
    )


# ============================================================
# EXAMPLE
# ============================================================
#
# Suppose we have:
#
#     Name      → 0 missing values
#     Age       → 2 missing values
#     Salary    → 5 missing values
#     Department → 1 missing value
#
# The output might be:
#
#     Name missing values: 0
#     Age missing values: 2
#     Salary missing values: 5
#     Department missing values: 1
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every column, check something."
#
# Think:
#
#     for column in df.columns:
#         ...
#
# Then access the current column:
#
#     df[column]
#
# In this example:
#
#     df[column].isnull().sum()
#
# means:
#
#     "For the current column, find and count its missing
#      values."
#
# ============================================================


# ============================================================
# DATA SCIENCE TIP
# ============================================================
#
# Pandas can check ALL columns at once:
#
#     df.isnull().sum()
#
# This is usually simpler and faster than explicitly writing
# a Python loop.
#
# The loop is useful when you want to add custom logic, such as:
#
#     • reporting only columns with missing values
#     • calculating missing-value percentages
#     • applying different rules to different columns
#     • automatically deciding how to handle missing data
#
# ============================================================
```

```python
# ============================================================
# 60. LOOPING THROUGH COLUMNS FOR UNIQUE VALUES
# ============================================================
#
# In Data Science, we often want to inspect a DataFrame
# column by column.
#
# One useful question is:
#
# "How many different values does each column contain?"
#
# Pandas provides:
#
#     .nunique()
#
# to count the number of DISTINCT values in a Series/column.
#
# ============================================================
# CONCEPTUAL FLOW
# ============================================================
#
# DataFrame
#     ↓
# df.columns
#     ↓
# get one column name
#     ↓
# df[column]
#     ↓
# select the current column
#     ↓
# .nunique()
#     ↓
# count distinct values
#     ↓
# print result
#
# ============================================================
# COMPONENT ROLES
# ============================================================
#
# df
#     → the Pandas DataFrame being analyzed
#
# df.columns
#     → collection-like object containing the column labels
#
# column
#     → loop variable containing the CURRENT column name
#
# df[column]
#     → selects the column whose name is currently stored
#       in column
#
# .nunique()
#     → counts distinct values in the selected column
#
# ============================================================


# ============================================================
# LOOP THROUGH COLUMNS
# ============================================================
#
# The for loop takes each column name from df.columns,
# one at a time.
#
# LOOP PATTERN:
#
# df.columns
#     ↓
# column = first column name
#     ↓
# process it
#     ↓
# column = next column name
#     ↓
# process it
#     ↓
# repeat until no columns remain
#
# ============================================================

for column in df.columns:

    # ========================================================
    # COUNT UNIQUE VALUES
    # ========================================================
    #
    # First:
    #
    #     df[column]
    #
    # selects the CURRENT column.
    #
    # If:
    #
    #     column = "Gender"
    #
    # then:
    #
    #     df[column]
    #
    # conceptually becomes:
    #
    #     df["Gender"]
    #
    # Then:
    #
    #     df[column].nunique()
    #
    # counts the distinct values in that column.
    #
    # ========================================================
    #
    # VALUE FLOW:
    #
    # "Gender"
    #     ↓
    # df["Gender"]
    #     ↓
    # selected Series
    #     ↓
    # .nunique()
    #     ↓
    # number of distinct values
    #
    # ========================================================

    print(
        column,
        "unique values:",
        df[column].nunique()
    )


# ============================================================
# HOW PYTHON EXECUTES THIS
# ============================================================
#
# Suppose:
#
#     df.columns = ["Gender", "Age", "City"]
#
# Python effectively performs:
#
# Iteration 1:
#     column = "Gender"
#     df["Gender"].nunique()
#     print the result
#
# Iteration 2:
#     column = "Age"
#     df["Age"].nunique()
#     print the result
#
# Iteration 3:
#     column = "City"
#     df["City"].nunique()
#     print the result
#
# No more column names
#     ↓
# loop ends
#
# ============================================================
# EXAMPLE
# ============================================================
#
# Suppose the columns contain:
#
# Gender → Male, Female
# Age    → 25, 26, 27, ... 59
# City   → Tehran, Karaj, Shiraz
#
# Then the number of distinct values might be:
#
# Gender → 2
# Age    → 35
# City   → 3
#
# Possible output:
#
# Gender unique values: 2
# Age unique values: 35
# City unique values: 3
#
# The exact numbers depend on the actual DataFrame.
#
# ============================================================
# MENTAL MODEL
# ============================================================
#
# Think:
#
# "For every column, perform the same analysis."
#
#     for column in df.columns:
#         analyze(df[column])
#
# Here, the analysis is:
#
#     df[column].nunique()
#
# So the reusable Python/Data Science pattern is:
#
#     for column in df.columns:
#         perform_operation_on(df[column])
#
# ============================================================
# WHAT CHANGES / WHAT STAYS THE SAME
# ============================================================
#
# WHAT CHANGES?
#
#     column
#
# changes on every iteration because it receives
# a different column name.
#
# WHAT STAYS THE SAME?
#
#     df
#
# remains the same DataFrame.
#
# The operation:
#
#     df[column].nunique()
#
# is also conceptually the same operation each time,
# but it operates on a different column.
#
# ============================================================
# DATA SCIENCE TIP
# ============================================================
#
# Pandas can perform this particular operation directly:
#
#     df.nunique()
#
# This calculates the number of unique values for
# every column at once.
#
# Therefore:
#
#     df.nunique()
#
# is usually simpler when you only need the counts.
#
# The loop becomes especially useful when each column
# needs additional customized processing.
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# Remember that:
#
#     column
#
# contains the COLUMN NAME, not the column's data.
#
# For example:
#
#     column = "Age"
#
# while:
#
#     df[column]
#
# gives the actual Age column.
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# COLLECTION
#     ↓
# get each item
#     ↓
# store current item in loop variable
#     ↓
# perform an operation
#     ↓
# repeat
#
# For a DataFrame:
#
# df.columns
#     ↓
# each column name
#     ↓
# df[column]
#     ↓
# operation on that column
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A DataFrame can be processed column by column by iterating
# through df.columns and using the current column name
# to select the corresponding column.
#
# The key pattern is:
#
#     for column in df.columns:
#         operation(df[column])
#
# This pattern is useful whenever the same analysis or
# transformation must be applied independently to every
# column.
#
# ============================================================
```

```python
# ============================================================
# 61. LOOPING THROUGH COLUMNS FOR DATA TYPES
# ============================================================
#
# In Pandas, a DataFrame is a table made of rows and columns.
#
# Sometimes we want to perform the same operation on EVERY
# column rather than writing separate code for each column.
#
# The general pattern is:
#
#     for column in df.columns:
#         ...
#
# Conceptual flow:
#
#     DataFrame
#         ↓
#     df.columns
#         ↓
#     column names
#         ↓
#     one column name at a time
#         ↓
#     df[column]
#         ↓
#     current column
#         ↓
#     .dtype
#         ↓
#     data type
#
# ============================================================
# CONCEPTUAL BREAKDOWN
# ============================================================
#
# CONCEPT:
#     Iterating through DataFrame columns.
#
# PURPOSE:
#     Inspect or process every column using the same logic.
#
# COMPONENTS:
#
#     df.columns
#         → collection-like Index containing the column names
#
#     column
#         → loop variable containing the CURRENT column name
#
#     df[column]
#         → selects the column whose name is currently stored
#           in column
#
#     .dtype
#         → gives the data type of the selected Pandas Series
#
# RELATIONSHIP:
#
#     df.columns
#         ↓
#     current column name
#         ↓
#     df[column]
#         ↓
#     selected Series
#         ↓
#     .dtype
#         ↓
#     column's data type
#
# ============================================================
# LOOP THROUGH COLUMNS
# ============================================================
#
# df.columns provides the column names.
#
# Python takes those names one at a time and assigns each
# name to the loop variable "column".
#
# For every iteration:
#
#     1. column receives a column name.
#     2. df[column] selects that column.
#     3. .dtype retrieves its data type.
#     4. print() displays the information.
#
# ============================================================
# LOOP CYCLE
# ============================================================
#
#     df.columns
#         ↓
#     get next column name
#         ↓
#     column = current name
#         ↓
#     df[column]
#         ↓
#     current column
#         ↓
#     .dtype
#         ↓
#     print result
#         ↓
#     more columns?
#       ↙       ↘
#     YES        NO
#      ↓          ↓
#   repeat      exit loop
#
# ============================================================
# VALUE FLOW
# ============================================================
#
# Suppose:
#
#     df.columns = ["Age", "Salary", "Name"]
#
# First iteration:
#
#     column = "Age"
#         ↓
#     df["Age"]
#         ↓
#     Age column
#         ↓
#     .dtype
#         ↓
#     int64
#
# Second iteration:
#
#     column = "Salary"
#         ↓
#     df["Salary"]
#         ↓
#     Salary column
#         ↓
#     .dtype
#         ↓
#     float64
#
# The loop variable changes on every iteration.
#
# The DataFrame itself is NOT changed by this code.
#
# ============================================================
#
# NOW THE ACTUAL CODE:
# ============================================================

for column in df.columns:

    # "column" contains the name of the current column.
    #
    # df[column] uses that name to select the corresponding
    # column from the DataFrame.
    #
    # Then .dtype asks Pandas:
    #
    #     "What data type does this column contain?"
    #
    # print() displays the column name and its data type.

    print(column,"type:",df[column].dtype)


# ============================================================
# HOW PYTHON + PANDAS EXECUTE IT
# ============================================================
#
# Python first evaluates:
#
#     df.columns
#
# This gives the available column names.
#
# Then the loop begins.
#
# Example:
#
#     column = "Age"
#         ↓
#     df["Age"].dtype
#         ↓
#     int64
#         ↓
#     print("Age", "type:", "int64")
#
# Then Python moves to the next column:
#
#     column = "Salary"
#         ↓
#     df["Salary"].dtype
#         ↓
#     float64
#         ↓
#     print("Salary", "type:", "float64")
#
# This continues until there are no more column names.
#
# ============================================================
# EXAMPLE
# ============================================================
#
# Suppose df contains:
#
#     Age       → int64
#     Salary    → float64
#     Name      → object
#     Employed  → bool
#
# The loop produces:
#
#     Age type: int64
#     Salary type: float64
#     Name type: object
#     Employed type: bool
#
# The exact dtype names depend on the actual DataFrame
# and Pandas version.
#
# ============================================================
# WHAT CHANGES / WHAT STAYS THE SAME
# ============================================================
#
# WHAT CHANGES?
#
#     column
#         → changes to the next column name on each iteration.
#
# WHAT STAYS THE SAME?
#
#     df
#         → the DataFrame itself is not modified.
#
# The code only INSPECTS the columns and prints information.
#
# ============================================================
# MENTAL MODEL
# ============================================================
#
# Think:
#
#     "Give me every column name,
#      one at a time,
#      and let me do something with that column."
#
# In Python/Pandas:
#
#     for column in df.columns:
#         ...
#
# Then:
#
#     df[column]
#
# means:
#
#     "Select the column whose name I currently have."
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you hear:
#
#     "Do something with every column."
#
# Think:
#
#     for column in df.columns:
#         ...
#
# Then use the current column:
#
#     df[column]
#
# Examples:
#
#     for column in df.columns:
#         print(column)
#
#     for column in df.columns:
#         print(df[column].dtype)
#
#     for column in df.columns:
#         print(df[column].isnull().sum())
#
# The STRUCTURE remains the same:
#
#     get each column
#         ↓
#     perform an operation
#
# Only the operation changes.
#
# ============================================================
# DATA SCIENCE TIP
# ============================================================
#
# Pandas already provides a convenient way to inspect all
# column data types:
#
#     df.dtypes
#
# Therefore, the loop is not necessarily the shortest way
# to obtain this information.
#
# The loop is valuable when you need CUSTOM processing for
# each column.
#
# For example:
#
#     for column in df.columns:
#         ...
#
# could be used to check missing values, apply transformations,
# validate data, or perform different analysis for each column.
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# This is a general "process every column" pattern:
#
#     DataFrame
#         ↓
#     df.columns
#         ↓
#     current column name
#         ↓
#     df[column]
#         ↓
#     perform operation
#
# The same pattern appears repeatedly in Data Science:
#
#     for column in df.columns:
#         inspect column
#
#     for column in df.columns:
#         clean column
#
#     for column in df.columns:
#         calculate statistic
#
#     for column in df.columns:
#         validate column
#
# ============================================================
# CORE IDEA
# ============================================================
#
# The important pattern is:
#
#     for column in df.columns:
#
# It means:
#
#     "Iterate through the DataFrame's column names,
#      one at a time."
#
# Then:
#
#     df[column]
#
# connects the CURRENT column name to the actual column
# in the DataFrame.
#
# This gives you a reusable way to apply the same processing
# logic to every column in a Pandas DataFrame.
# ============================================================
```
