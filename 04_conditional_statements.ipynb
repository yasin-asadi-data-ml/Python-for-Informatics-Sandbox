#             PYTHON CONDITIONAL STATEMENTS


```python
# ============================================================
#             PYTHON CONDITIONAL STATEMENTS
# ============================================================
#
# Conditional statements allow Python to make decisions.
#
# Basic structure:
#
# if condition:
#     code
#
# elif another_condition:
#     code
#
# else:
#     code
#
# ============================================================


# ============================================================
# 1. BASIC IF STATEMENT
# ============================================================

age = 25

if age >= 18:
    print("You are an adult.")
```

```python
# ============================================================
# 2. IF + ELSE
# ============================================================

# Assigning value to variable 'age'
age = 16

# Checking if age is greater than or equal to 18
if age >= 18:
    # This block executes if condition is True
    print("You are an adult.")
else:
    # This block executes if condition is False
    print("You are a minor.")
```

```python
# ============================================================
# 3. IF + ELIF + ELSE
# ============================================================

age = 25

if age < 13:
    print("Child")

elif age < 18:
    print("Teenager")

else:
    print("Adult")
```

```python
# ===========================================================================
# 4. MULTIPLE CONDITIONS

# This section demonstrates how to evaluate multiple conditions in an if statement
# ============================================================

age = 30          # Variable storing the person's age
has_license = True # Boolean variable indicating if the person has a license

# Check if the person is 18 or older AND has a valid license
if age >= 18 and has_license:
    print("You can drive.")  # Print message if both conditions are true
else:
    print("You cannot drive.") # Print message if either condition is false

```

```python
# ============================================================
# 5. COMPARISON OPERATORS
# ============================================================

x = 10
y = 20
print("COMPARISON OPERATORS:\n")
print (f"is {x} == {y} ? : {x == y}") # equal to
print (f"is {x} != {y} ? : {x != y}")  # not equal to
print (f"is {x} > {y} ? : {x > y}")    # greater than
print (f"is {x} < {y} ? : {x < y}")    # less than
print (f"is {x} >= {y} ? : {x >= y}") # greater than or equal to
print (f"is {x} <= {y} ? : {x <= y}") # less than or equal to
```

```python
# ============================================================
# 6. USING COMPARISON OPERATORS IN IF
# ============================================================

score = 85  # Assign a value of 85 to the variable 'score'



# Check if the score is greater than or equal to 50
if score >= 50:  # If the condition is true (score is 85 which is >= 50)
    print("Pass")  # Print "Pass" if the condition is true
else:
    print("Fail")  # Print "Fail" if the condition is false

```

```python
# ============================================================
# 7. AND
# ============================================================

age = 25          # Variable storing the age value
salary = 50000

# Check if both conditions are true:
# 1. Age is greater than or equal to 18
# 2. Salary is greater than or equal to 30000
if age >= 18 and salary >= 30000:
    print("Condition satisfied.")

```

```python
# ============================================================
# 8. OR
# ============================================================

# Assign the value "IT" to the variable department
department = "IT"

# Check if department is either "IT" or "Data Science"
# If condition is true, print Technology-related department
if department == "IT" or department == "Data Science":
    print("Technology-related department.")

```

```python
# ============================================================
# 9. NOT
# This section demonstrates the use of the NOT logical operator
# in Python, which negates a boolean value.
# ============================================================

# Initialize a boolean variable to store student status
is_student = False

# Check if the person is NOT a student using the NOT operator
if not is_student:
    # Print message if the condition is True (person is not a student)
    print("The person is not a student.")
```

```python
# ============================================================
# 10. COMBINING AND / OR / NOT
# ============================================================



# Initialize variables with candidate information
age = 30                    # Candidate's age
experience = 6              # Years of professional experience
has_degree = True           # Whether candidate has a degree

if (age >= 25 and experience >= 5) or has_degree:
    print("Candidate satisfies the requirement.")

```

```python
# ============================================================
# 11. NESTED IF
# This program demonstrates nested if-else statements to check driving eligibility
# based on age and possession of a driving license.
# ============================================================

age = 30          # Variable to store the person's age
has_license = True  # Boolean variable indicating if the person has a license

# First condition check: if the person is 18 years or older
if age >= 18:

    # Nested condition: if the person has a license
    if has_license:
        print("You can drive.")  # Message if both conditions are met

    # Alternative for nested condition: if the person doesn't have a license
    else:
        print("You need a driving licence.")  # Message when age is sufficient but no license

# Alternative for first condition: if the person is under 18
else:
    print("You are too young to drive.")  # Message when age requirement is not met

```

```python
# ============================================================

# 24. TERNARY CONDITIONAL EXPRESSION

# ============================================================

age = 25  # Assigning the value 25 to the variable 'age'

# A ternary conditional expression is a short way to write
# a simple if-else decision in one line.

# Using a ternary conditional expression to determine the status
# If age is greater than or equal to 18, status is "Adult"
# Otherwise, status is "Minor"

status = "Adult" if age >= 18 else "Minor"

# Read this as:
# "Use 'Adult' if age >= 18, otherwise use 'Minor'."

# The equivalent regular if-else statement would be:
#
# if age >= 18:
#     status = "Adult"
# else:
#     status = "Minor"

# Ternary expression to set the status

print(status)  # Printing the status value
```

```python
# ============================================================

# 35. DATA ANALYTICS — OUTLIER LOGIC

# ============================================================

salary = 250000

# Q1 (first quartile): 25% of the data is below this value.
Q1 = 40000

# Q3 (third quartile): 75% of the data is below this value.
Q3 = 80000

# IQR (Interquartile Range) is the distance between Q1 and Q3.
# It represents the middle 50% of the data.
IQR = Q3 - Q1

# Calculate the lower and upper boundaries for potential outliers.
# The common rule uses 1.5 × IQR beyond Q1 and Q3.

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# A value is considered a potential outlier if it is:
# - below the lower limit, OR
# - above the upper limit.

if salary < lower_limit or salary > upper_limit:

    # The salary falls outside the expected range.
    print("Potential outlier.")

else:

    # The salary falls within the expected range.
    print("Within expected range.")
```

```python
# ============================================================
# 39. USING CONDITIONS WITH LISTS
# ============================================================

scores = [85, 42, 91, 67, 55]

for score in scores:

    if score >= 50:
        print(score, "Pass")

    else:
        print(score, "Fail")


# ============================================================
# 40. CONDITIONS INSIDE A LOOP
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]

for number in numbers:

    if number % 2 == 0:
        print(number, "Even")

    else:
        print(number, "Odd")


# ============================================================
# 41. FILTERING DATA WITH A CONDITION
# ============================================================

numbers = [10, 25, 40, 55, 70]

for number in numbers:

    if number > 40:
        print(number)


# ============================================================
# 42. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [10, 20, 30, 40, 50]

large_numbers = [
    number
    for number in numbers
    if number > 25
]

print(large_numbers)


# ============================================================
# 43. LIST COMPREHENSION WITH IF-ELSE
# ============================================================

numbers = [10, 20, 30, 40, 50]

categories = [
    "High" if number >= 30 else "Low"
    for number in numbers
]

print(categories)


# ============================================================
# 44. CONDITIONAL LOGIC WITH PANDAS
# ============================================================

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Yasin", "John"],
    "Age": [25, 32, 41, 19],
    "Salary": [35000, 55000, 90000, 28000]
})

print(df)


# ============================================================
# 45. PANDAS BOOLEAN FILTERING
# ============================================================

print(df[df["Age"] >= 30])

print(df[df["Salary"] > 50000])


# ============================================================
# 46. PANDAS AND CONDITION
# ============================================================

print(
    df[
        (df["Age"] >= 30) &
        (df["Salary"] > 50000)
    ]
)


# ============================================================
# 47. PANDAS OR CONDITION
# ============================================================

print(
    df[
        (df["Age"] < 25) |
        (df["Salary"] > 80000)
    ]
)


# ============================================================
# 48. PANDAS NOT CONDITION
# ============================================================

print(
    df[
        ~(df["Age"] >= 30)
    ]
)


# ============================================================
# 49. PANDAS np.where()
# ============================================================

df["Age_Group"] = np.where(
    df["Age"] >= 30,
    "30+",
    "Under 30"
)

print(df)


# ============================================================
# 50. PANDAS MULTIPLE CONDITIONS WITH np.select()
# ============================================================

conditions = [
    df["Salary"] < 40000,
    df["Salary"] < 70000,
    df["Salary"] >= 70000
]

choices = [
    "Low",
    "Medium",
    "High"
]

df["Salary_Level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df)


# ============================================================
# 51. PRACTICAL ANALYTICS EXAMPLE
# ============================================================

df["Performance"] = np.where(
    df["Salary"] >= 50000,
    "High",
    "Low"
)

print(df)


# ============================================================
# 52. PRACTICAL DATA QUALITY EXAMPLE
# ============================================================

df["Age_Status"] = np.where(
    (df["Age"] >= 18) & (df["Age"] <= 100),
    "Valid",
    "Check"
)

print(df)


# ============================================================
# 53. CONDITIONS WITH FUNCTIONS
# ============================================================

def classify_salary(salary):

    if salary < 40000:
        return "Low"

    elif salary < 70000:
        return "Medium"

    else:
        return "High"


print(classify_salary(35000))
print(classify_salary(60000))
print(classify_salary(90000))


# ============================================================
# 54. APPLYING A FUNCTION TO A DATAFRAME
# ============================================================

df["Salary_Level"] = df["Salary"].apply(
    classify_salary
)

print(df)


# ============================================================
# 55. PRACTICAL ML METRIC CLASSIFICATION
# ============================================================

precision = 0.82
recall = 0.74

if precision >= 0.80 and recall >= 0.80:
    print("Strong performance on both metrics.")

elif precision >= 0.80 and recall < 0.80:
    print("Good precision but recall needs improvement.")

elif precision < 0.80 and recall >= 0.80:
    print("Good recall but precision needs improvement.")

else:
    print("Both metrics need improvement.")


# ============================================================
# 56. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================

failure_probability = 0.73

if failure_probability >= 0.80:
    action = "Immediate inspection"

elif failure_probability >= 0.60:
    action = "Schedule inspection"

elif failure_probability >= 0.40:
    action = "Monitor"

else:
    action = "Normal operation"

print(action)


# ============================================================
# 57. COMBINING MULTIPLE FEATURES
# ============================================================

pipe_age = 35
pressure = 90
failure_history = True

if (
    pipe_age > 30
    and pressure > 80
    and failure_history
):
    risk = "High"

elif (
    pipe_age > 30
    or pressure > 80
    or failure_history
):
    risk = "Medium"

else:
    risk = "Low"

print(risk)


# ============================================================
# 58. MATCH-CASE
# ============================================================

department = "IT"

match department:

    case "IT":
        print("Information Technology")

    case "HR":
        print("Human Resources")

    case "Finance":
        print("Finance")

    case _:
        print("Unknown department")


# ============================================================
# 59. MATCH-CASE WITH MULTIPLE OPTIONS
# ============================================================

department = "AI"

match department:

    case "AI" | "ML" | "Data Science":
        print("Data/AI field")

    case "Network":
        print("Networking field")

    case "Finance":
        print("Finance field")

    case _:
        print("Other field")


# ============================================================
# 60. PRACTICAL DECISION-MAKING EXAMPLE
# ============================================================

age = 35
experience = 8
python_skill = True
data_analysis_skill = True

if (
    age >= 18
    and experience >= 3
    and python_skill
    and data_analysis_skill
):
    print("Candidate meets the requirements.")

else:
    print("Candidate does not meet all requirements.")
```
