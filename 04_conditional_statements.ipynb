```python
# ============================================================
#             ANY() AND ALL() FUNCTIONS
# ============================================================
#
# any() returns True if at least one item in an iterable is truthy.
#
# all() returns True if every item in an iterable is truthy.
#
# For strings:
#     ""       → False (empty string)
#     "Ali"    → True  (non-empty string)
#
# ============================================================


# ============================================================
# 1. CHECK IF ANY VALUE IS NOT EMPTY
# ============================================================

email = "Ali"              # Variable to store email address
phone = "0923456789"       # Variable to store phone number
username = "Ali001"        # Variable to store username

print("If any of email, phone, or username is not empty, print True:")
print(any([email, phone, username]))


# ============================================================
# 2. CHECK IF ALL VALUES ARE NOT EMPTY
# ============================================================

print("If all of email, phone, and username are not empty, print True:")
print(all([email, phone, username]))
```

```python
# ============================================================
#             ANY() AND ALL() WITH DIFFERENT DATA TYPES
# ============================================================
#
# any() → True if AT LEAST ONE item is truthy.
#
# all() → True if ALL items are truthy.
#
# They can work with:
#     - strings
#     - numbers
#     - booleans
#     - lists
#     - tuples
#     - sets
#     - dictionaries
#     - conditions / comparisons
#
# ============================================================


# ============================================================
# 1. WITH STRINGS
# ============================================================

names = ["Ali", "", "Yasin"]

print(any(names))   # True  → at least one string is not empty
print(all(names))   # False → not all strings are non-empty

```

```python
# ============================================================
# 2. WITH NUMBERS
# ============================================================

numbers = [0, 0, 5, 0]

print(any(numbers))   # True  → at least one number is non-zero
print(all(numbers))   # False → not all numbers are non-zero
```

```python
# ============================================================
# 3. WITH BOOLEANS
# ============================================================

values = [False, False, True]

print(any(values))   # True  → at least one value is True
print(all(values))   # False → not all values are True

```

```python
# ============================================================
# 4. WITH CONDITIONS
# ============================================================

age = 25
score = 85
balance = 100

print(any([
    age >= 18,
    score >= 90,
    balance > 500
]))
# True → age >= 18 is True


print(all([
    age >= 18,
    score >= 80,
    balance > 50
]))
# True → all three conditions are True
```

```python
# ============================================================
# 5. WITH A LIST
# ============================================================

items = ["Python", "SQL", "Power BI"]

print(any(items))   # True → at least one item exists
print(all(items))   # True → all items are non-empty
```

```python
# ============================================================
# 6. WITH AN EMPTY LIST
# ============================================================

empty_list = []

print(any(empty_list))   # False
print(all(empty_list))   # True
#
# IMPORTANT:
# any([]) → False
# all([]) → True
#
# This is a special rule of Python.
# ============================================================
```

```python
# ============================================================
# 7. WITH A DICTIONARY
# ============================================================

user = {
    "email": "ali@example.com",
    "phone": "",
    "username": "Ali001"
}

print(any(user.values()))
# True → at least one value is non-empty

print(all(user.values()))
# False → phone is empty

```

```python
# ============================================================
# 8. WITH CONDITIONS IN A LIST COMPREHENSION
# ============================================================

numbers = [10, 20, 30, 40, 50]

print(any(number > 40 for number in numbers))
# True → 50 is greater than 40

print(all(number > 5 for number in numbers))
# True → every number is greater than 5

```

```python
# ============================================================
# QUICK SUMMARY
# ============================================================
#
# any() → "Is there AT LEAST ONE True value?"
#
# all() → "Are ALL values True?"
#
# Examples:
#
# any([False, False, True]) → True
# any([False, False, False]) → False
#
# all([True, True, True])   → True
# all([True, False, True])  → False
#
# ============================================================
```

```python
# ============================================================
#                    isinstance()
# ============================================================
#
# isinstance() checks whether a value belongs to a specific
# data type (or class).
#
# Basic structure:
#
# isinstance(value, type)
#
# It returns:
#     True  → if the value is the specified type
#     False → if the value is not the specified type
#
# ============================================================


# ============================================================
# 1. CHECK IF A VALUE IS AN INTEGER
# ============================================================

age = 25

print(isinstance(age, int))
# True

```

```python
# ============================================================
# 2. CHECK IF A VALUE IS A STRING
# ============================================================

name = "Ali"

print(isinstance(name, str))
# True
```

```python
# ============================================================
# 3. CHECK IF A VALUE IS A FLOAT
# ============================================================

price = 19.99

print(isinstance(price, float))
# True
```

```python
# ============================================================
# 4. CHECK IF A VALUE IS A BOOLEAN
# ============================================================

is_logged_in = True

print(isinstance(is_logged_in, bool))
# True
```

```python
# ============================================================
# 5. CHECKING THE WRONG TYPE
# ============================================================

age = 25

print(isinstance(age, str))
# False

print(isinstance(age, float))
# False

```

```python
# ============================================================
# 6. CHECK IF A VALUE IS A LIST
# ============================================================

students = ["Ali", "Yasin", "Sina"]

print(isinstance(students, list))
# True
```

```python
# ============================================================
# 7. CHECK MULTIPLE TYPES
# ============================================================
#
# You can provide a tuple of types.
#
# isinstance(value, (type1, type2, ...))
#
# It returns True if the value matches ANY of the types.
# ============================================================

value = 25

print(isinstance(value, (int, float)))
# True → value is an int


value = 19.5

print(isinstance(value, (int, float)))
# True → value is a float


value = "25"

print(isinstance(value, (int, float)))
# False → value is a string

```

```python
# ============================================================
# 8. USING isinstance() WITH IF
# ============================================================

value = 100

if isinstance(value, int):
    print("The value is an integer.")
else:
    print("The value is not an integer.")

```

###             PYTHON CONDITIONAL STATEMENTS


```python
# ============================================================
#             PYTHON CONDITIONAL STATEMENTS
# ============================================================
#
# Conditional statements allow Python to make decisions.
#
# Basic structure:
#
# if condition:
#     code
#
# elif another_condition:
#     code
#
# else:
#     code
#
# ============================================================


# ============================================================
# 1. BASIC IF STATEMENT
# ============================================================

age = 25

if age >= 18:
    print("You are an adult.")
```

```python
# ============================================================
# 2. IF + ELSE
# ============================================================

# Assigning value to variable 'age'
age = 16

# Checking if age is greater than or equal to 18
if age >= 18:
    # This block executes if condition is True
    print("You are an adult.")
else:
    # This block executes if condition is False
    print("You are a minor.")
```

```python
# ============================================================
# 3. IF + ELIF + ELSE
# ============================================================

age = 25

if age < 13:
    print("Child")

elif age < 18:
    print("Teenager")

else:
    print("Adult")
```

```python
# ===========================================================================
# 4. MULTIPLE CONDITIONS

# This section demonstrates how to evaluate multiple conditions in an if statement
# ============================================================

age = 30          # Variable storing the person's age
has_license = True # Boolean variable indicating if the person has a license

# Check if the person is 18 or older AND has a valid license
if age >= 18 and has_license:
    print("You can drive.")  # Print message if both conditions are true
else:
    print("You cannot drive.") # Print message if either condition is false

```

```python
# ============================================================
# 5. COMPARISON OPERATORS
# ============================================================

x = 10
y = 20
print("COMPARISON OPERATORS:\n")
print (f"is {x} == {y} ? : {x == y}") # equal to
print (f"is {x} != {y} ? : {x != y}")  # not equal to
print (f"is {x} > {y} ? : {x > y}")    # greater than
print (f"is {x} < {y} ? : {x < y}")    # less than
print (f"is {x} >= {y} ? : {x >= y}") # greater than or equal to
print (f"is {x} <= {y} ? : {x <= y}") # less than or equal to
```

```python
# ============================================================
# 6. USING COMPARISON OPERATORS IN IF
# ============================================================

score = 85  # Assign a value of 85 to the variable 'score'



# Check if the score is greater than or equal to 50
if score >= 50:  # If the condition is true (score is 85 which is >= 50)
    print("Pass")  # Print "Pass" if the condition is true
else:
    print("Fail")  # Print "Fail" if the condition is false

```

```python
# ============================================================
# 7. AND
# ============================================================

age = 25          # Variable storing the age value
salary = 50000

# Check if both conditions are true:
# 1. Age is greater than or equal to 18
# 2. Salary is greater than or equal to 30000
if age >= 18 and salary >= 30000:
    print("Condition satisfied.")

```

```python
# ============================================================
# 11. NESTED IF
# This program demonstrates nested if-else statements to check driving eligibility
# based on age and possession of a driving license.
# ============================================================

age = 30          # Variable to store the person's age
has_license = True  # Boolean variable indicating if the person has a license

# First condition check: if the person is 18 years or older
if age >= 18:

    # Nested condition: if the person has a license
    if has_license:
        print("You can drive.")  # Message if both conditions are met

    # Alternative for nested condition: if the person doesn't have a license
    else:
        print("You need a driving licence.")  # Message when age is sufficient but no license

# Alternative for first condition: if the person is under 18
else:
    print("You are too young to drive.")  # Message when age requirement is not met

```

```python
# ============================================================
#                    is AND is not
# ============================================================
#
# Python has two different concepts:
#
# ==      → checks if two values are equal
#
# is      → checks if two variables refer to the SAME object
#
# is not  → checks if two variables do NOT refer to the
#           same object
#
# ============================================================


# ============================================================
# 1. ==  CHECKS VALUE EQUALITY
# ============================================================

x = 10
y = 10

print(x == y)
# True
#
# Both variables contain the value 10.
```

```python
# ============================================================
# 2. is  CHECKS OBJECT IDENTITY
# ============================================================

x = 10
y = x

print(x is y)
# True
#
# y = x makes y refer to the same object as x.
#
# In other words:
#
# x ─────┐
#        ↓
#       10
#        ↑
# y ─────┘
```

```python
# ============================================================
# 3. is NOT THE SAME AS ==
# ============================================================

x = [1, 2, 3]
y = [1, 2, 3]

print(x == y)
# True
#
# The lists contain the same values.


print(x is y)
# False
#
# They are two different list objects.
#
# x → [1, 2, 3]
# y → [1, 2, 3]
#
# Same contents, different objects.

```

```python
# ============================================================
# 4. ASSIGNING ONE VARIABLE TO ANOTHER
# ============================================================

x = [1, 2, 3]

y = x

print(x == y)
# True

print(x is y)
# True
#
# y = x does NOT create a new list.
#
# Both variables refer to the same object.
#
# x ─────┐
#        ↓
#    [1, 2, 3]
#        ↑
# y ─────┘
```

```python
# ============================================================
# 5. is NOT
# ============================================================

x = [1, 2, 3]
y = [1, 2, 3]

print(x is not y)
# True
#
# x and y are different objects.


x = [1, 2, 3]
y = x

print(x is not y)
# False
#
# x and y refer to the same object.

```

```python
# ============================================================
# 6. is NOT USED FOR NORMAL VALUE COMPARISON
# ============================================================
#
# Usually:
#
# Use ==  → to compare values
# Use is  → to compare object identity
#
# Correct:
# ============================================================

name = "Ali"

if name == "Ali":
    print("The name is Ali.")

```

```python
# ============================================================
# 7. CHECKING FOR None
# ============================================================
#
# None means "no value" / "no object".
#
# When checking whether something is None,
# use:
#
#     is None
#
# or:
#
#     is not None
#
# ============================================================


email = None

print(email is None)
# True


print(email is not None)
# False

```

```python
# ============================================================
# 11. == VS is WITH None
# ============================================================
#
# For None, Python convention is:
#
#     is None
#     is not None
#
# NOT:
#
#     == None
#     != None
#
# ============================================================

email = None

print(email is None)
# True


print(email is not None)
# False
```

```python
# ============================================================
# QUICK SUMMARY
# ============================================================
#
# ==      → Are the VALUES equal?
#
# is      → Are they the SAME OBJECT?
#
# is not  → Are they DIFFERENT OBJECTS?
#
# None    → Represents the absence of a value.
#
#
# Examples:
#
# x = [1, 2, 3]
# y = [1, 2, 3]
#
# x == y       → True
# x is y       → False
#
#
# x = [1, 2, 3]
# y = x
#
# x == y       → True
# x is y       → True
#
#
# email = None
#
# email is None      → True
# email is not None  → False
#
# ============================================================
```

```python
# ============================================================

# 24. TERNARY CONDITIONAL EXPRESSION

# ============================================================

age = 25  # Assigning the value 25 to the variable 'age'

# A ternary conditional expression is a short way to write
# a simple if-else decision in one line.

# Using a ternary conditional expression to determine the status
# If age is greater than or equal to 18, status is "Adult"
# Otherwise, status is "Minor"

status = "Adult" if age >= 18 else "Minor"

# Read this as:
# "Use 'Adult' if age >= 18, otherwise use 'Minor'."

# The equivalent regular if-else statement would be:
#
# if age >= 18:
#     status = "Adult"
# else:
#     status = "Minor"

# Ternary expression to set the status

print(status)  # Printing the status value
```

```python
# ============================================================
#          ONE-LINE IF VS INDENTED IF / ELSE
# ============================================================
#
# Python allows simple if/else decisions to be written
# in one line.
#
# Normal if/else:
#
# if condition:
#     code
# else:
#     code
#
#
# One-line if/else:
#
# value_if_true if condition else value_if_false
#
# ============================================================


# ============================================================
# 1. NORMAL INDENTED IF / ELSE
# ============================================================

score = 75

if score >= 50:
    result = "Passed"
else:
    result = "Failed"

print(result)

# Output:
# Passed


# ============================================================
# 2. ONE-LINE IF / ELSE
# ============================================================

score = 75

result = "Passed" if score >= 50 else "Failed"

print(result)

# Output:
# Passed

```

```python
# ============================================================
# 3. COMPARE THEM
# ============================================================
#
# Normal:
#
# if score >= 50:
#     result = "Passed"
# else:
#     result = "Failed"
#
#
# One line:
#
# result = "Passed" if score >= 50 else "Failed"
#
# Both do the same thing.


# ============================================================
# 4. ANOTHER EXAMPLE
# ============================================================

age = 20

if age >= 18:
    status = "Adult"
else:
    status = "Minor"

print(status)


# One-line version:

status = "Adult" if age >= 18 else "Minor"

print(status)
```

```python
# ============================================================
# 5. ONE-LINE IF WITHOUT ELSE
# ============================================================
#
# There is also a short form when you only need an action
# if the condition is True:
#
#     if condition: code
#
# Example:
# ============================================================

score = 90

if score >= 50: print("Passed")


# This is valid Python.
#
# However, for anything more complicated,
# use the normal indented form.
```

```python
# ============================================================
# 6. WHEN TO USE EACH ONE
# ============================================================
#
# Use the normal indented version when:
#
#     - The code is more complicated
#     - You have multiple statements
#     - You want better readability
#
#
# Use the one-line version when:
#
#     - The decision is very simple
#     - You are assigning a value
#
# ============================================================


# ============================================================
# QUICK SUMMARY
# ============================================================
#
# NORMAL:
#
# if condition:
#     value = A
# else:
#     value = B
#
#
# ONE-LINE:
#
# value = A if condition else B
#
#
# Example:
#
# result = "Pass" if score >= 50 else "Fail"
#
# ============================================================
```

Normal:

IF condition is True
    → do this
ELSE
    → do that


One line:

- do_this IF condition ELSE do_that



Read it almost like English:

- result is "Pass" if score is at least 50, otherwise "Fail".

One important terminology point: 
- result = "Pass" if ... else ... is usually called a conditional expression 
- (often informally called a ternary operator in Python).

### Independent IFs (Sepatare ifs)

```python
# ============================================================
#                    INDEPENDENT IF STATEMENTS
# ============================================================
#
# Independent if statements are multiple separate if statements.
#
# Each if is checked separately.
#
# IMPORTANT:
#
# All conditions are tested,
# even if one condition is already True.
#
# ============================================================


# ============================================================
# 1. BASIC EXAMPLE
# ============================================================

age = 25

if age >= 18:
    print("You are an adult.")

if age >= 21:
    print("You can legally drive in some countries.")

if age >= 65:
    print("You are a senior.")


# Output:
#
# You are an adult.
# You can legally drive in some countries.
#
# Each if was checked separately.

```

```python
# ============================================================
# 2. MULTIPLE CONDITIONS CAN BE TRUE
# ============================================================

score = 95

if score >= 50:
    print("You passed.")

if score >= 70:
    print("Good score.")

if score >= 90:
    print("Excellent score.")


# Output:
#
# You passed.
# Good score.
# Excellent score.
#
# ALL three conditions are True,
# so ALL three blocks execute.

```

```python
# ============================================================
# 3. ALL CONDITIONS ARE TESTED
# ============================================================

x = 10

if x > 5:
    print("Condition 1 is True.")

if x > 8:
    print("Condition 2 is True.")

if x > 20:
    print("Condition 3 is True.")


# Output:
#
# Condition 1 is True.
# Condition 2 is True.
#
# Python still checks Condition 3,
# even though the previous conditions were True.

```

```python
# ============================================================
# 4. INDEPENDENT if VS if / elif / else
# ============================================================

score = 95


# ------------------------------------------------------------
# Independent if statements
# ------------------------------------------------------------

if score >= 50:
    print("Passed.")

if score >= 70:
    print("Good.")

if score >= 90:
    print("Excellent.")


# Output:
#
# Passed.
# Good.
# Excellent.
# ALL block execute.
```

```python
# ------------------------------------------------------------
# if / elif / else
# ------------------------------------------------------------

if score >= 90:
    print("Excellent.")
elif score >= 70:
    print("Good.")
elif score >= 50:
    print("Passed.")
else:
    print("Failed.")


# Output:
#
# Excellent.
#
# Only ONE block executes.
#
# Once Python finds the first True condition,
# it skips the remaining elif/else conditions.

```

#### A good rule to put in your notes is:

- Use separate ifs when more than one condition can be True.
- Use if/elif/else when only one condition should be True.

```python
# Independent ifs
# ────────────────────────────────
# if A → check
# if B → check
# if C → check

# A, B, and C can ALL execute.


# if / elif / else
# ────────────────────────────────
# if A → check
#    ↓ False
# elif B → check
#    ↓ True
# elif C → SKIP
# else → SKIP

# Only ONE branch executes.
```

```python

# ============================================================
# 5. IMPORTANT DIFFERENCE
# ============================================================
#
# Independent if:
#
#     if condition_1:
#         code
#
#     if condition_2:
#         code
#
#     if condition_3:
#         code
#
# Multiple conditions can execute.
#
#
# if / elif / else:
#
#     if condition_1:
#         code
#     elif condition_2:
#         code
#     elif condition_3:
#         code
#     else:
#         code
#
# Only ONE branch executes.
#
# ============================================================


# ============================================================
# QUICK SUMMARY
# ============================================================
#
# INDEPENDENT ifs:
#
#     Each if is checked separately.
#     All conditions are tested.
#     Multiple conditions can be True.
#     Multiple blocks can execute.
#
#
# if / elif / else:
#
#     Conditions are checked from top to bottom.
#     Once a condition is True, the remaining branches
#     are skipped.
#     Only one branch executes.
#
# ============================================================
```

```python
# ============================================================

# 35. DATA ANALYTICS — OUTLIER LOGIC

# ============================================================

salary = 250000

# Q1 (first quartile): 25% of the data is below this value.
Q1 = 40000

# Q3 (third quartile): 75% of the data is below this value.
Q3 = 80000

# IQR (Interquartile Range) is the distance between Q1 and Q3.
# It represents the middle 50% of the data.
IQR = Q3 - Q1

# Calculate the lower and upper boundaries for potential outliers.
# The common rule uses 1.5 × IQR beyond Q1 and Q3.

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# A value is considered a potential outlier if it is:
# - below the lower limit, OR
# - above the upper limit.

if salary < lower_limit or salary > upper_limit:

    # The salary falls outside the expected range.
    print("Potential outlier.")

else:

    # The salary falls within the expected range.
    print("Within expected range.")
```

```python
# ============================================================
# 39. USING CONDITIONS WITH LISTS
# ============================================================

scores = [85, 42, 91, 67, 55]

for score in scores:

    if score >= 50:
        print(score, "Pass")

    else:
        print(score, "Fail")


# ============================================================
# 40. CONDITIONS INSIDE A LOOP
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]

for number in numbers:

    if number % 2 == 0:
        print(number, "Even")

    else:
        print(number, "Odd")


# ============================================================
# 41. FILTERING DATA WITH A CONDITION
# ============================================================

numbers = [10, 25, 40, 55, 70]

for number in numbers:

    if number > 40:
        print(number)


# ============================================================
# 42. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [10, 20, 30, 40, 50]

large_numbers = [
    number
    for number in numbers
    if number > 25
]

print(large_numbers)


# ============================================================
# 43. LIST COMPREHENSION WITH IF-ELSE
# ============================================================

numbers = [10, 20, 30, 40, 50]

categories = [
    "High" if number >= 30 else "Low"
    for number in numbers
]

print(categories)


# ============================================================
# 44. CONDITIONAL LOGIC WITH PANDAS
# ============================================================

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Yasin", "John"],
    "Age": [25, 32, 41, 19],
    "Salary": [35000, 55000, 90000, 28000]
})

print(df)


# ============================================================
# 45. PANDAS BOOLEAN FILTERING
# ============================================================

print(df[df["Age"] >= 30])

print(df[df["Salary"] > 50000])


# ============================================================
# 46. PANDAS AND CONDITION
# ============================================================

print(
    df[
        (df["Age"] >= 30) &
        (df["Salary"] > 50000)
    ]
)


# ============================================================
# 47. PANDAS OR CONDITION
# ============================================================

print(
    df[
        (df["Age"] < 25) |
        (df["Salary"] > 80000)
    ]
)


# ============================================================
# 48. PANDAS NOT CONDITION
# ============================================================

print(
    df[
        ~(df["Age"] >= 30)
    ]
)


# ============================================================
# 49. PANDAS np.where()
# ============================================================

df["Age_Group"] = np.where(
    df["Age"] >= 30,
    "30+",
    "Under 30"
)

print(df)


# ============================================================
# 50. PANDAS MULTIPLE CONDITIONS WITH np.select()
# ============================================================

conditions = [
    df["Salary"] < 40000,
    df["Salary"] < 70000,
    df["Salary"] >= 70000
]

choices = [
    "Low",
    "Medium",
    "High"
]

df["Salary_Level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df)


# ============================================================
# 51. PRACTICAL ANALYTICS EXAMPLE
# ============================================================

df["Performance"] = np.where(
    df["Salary"] >= 50000,
    "High",
    "Low"
)

print(df)


# ============================================================
# 52. PRACTICAL DATA QUALITY EXAMPLE
# ============================================================

df["Age_Status"] = np.where(
    (df["Age"] >= 18) & (df["Age"] <= 100),
    "Valid",
    "Check"
)

print(df)


# ============================================================
# 53. CONDITIONS WITH FUNCTIONS
# ============================================================

def classify_salary(salary):

    if salary < 40000:
        return "Low"

    elif salary < 70000:
        return "Medium"

    else:
        return "High"


print(classify_salary(35000))
print(classify_salary(60000))
print(classify_salary(90000))


# ============================================================
# 54. APPLYING A FUNCTION TO A DATAFRAME
# ============================================================

df["Salary_Level"] = df["Salary"].apply(
    classify_salary
)

print(df)


# ============================================================
# 55. PRACTICAL ML METRIC CLASSIFICATION
# ============================================================

precision = 0.82
recall = 0.74

if precision >= 0.80 and recall >= 0.80:
    print("Strong performance on both metrics.")

elif precision >= 0.80 and recall < 0.80:
    print("Good precision but recall needs improvement.")

elif precision < 0.80 and recall >= 0.80:
    print("Good recall but precision needs improvement.")

else:
    print("Both metrics need improvement.")


# ============================================================
# 56. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================

failure_probability = 0.73

if failure_probability >= 0.80:
    action = "Immediate inspection"

elif failure_probability >= 0.60:
    action = "Schedule inspection"

elif failure_probability >= 0.40:
    action = "Monitor"

else:
    action = "Normal operation"

print(action)


# ============================================================
# 57. COMBINING MULTIPLE FEATURES
# ============================================================

pipe_age = 35
pressure = 90
failure_history = True

if (
    pipe_age > 30
    and pressure > 80
    and failure_history
):
    risk = "High"

elif (
    pipe_age > 30
    or pressure > 80
    or failure_history
):
    risk = "Medium"

else:
    risk = "Low"

print(risk)


# ============================================================
# 58. MATCH-CASE
# ============================================================

department = "IT"

match department:

    case "IT":
        print("Information Technology")

    case "HR":
        print("Human Resources")

    case "Finance":
        print("Finance")

    case _:
        print("Unknown department")


# ============================================================
# 59. MATCH-CASE WITH MULTIPLE OPTIONS
# ============================================================

department = "AI"

match department:

    case "AI" | "ML" | "Data Science":
        print("Data/AI field")

    case "Network":
        print("Networking field")

    case "Finance":
        print("Finance field")

    case _:
        print("Other field")


# ============================================================
# 60. PRACTICAL DECISION-MAKING EXAMPLE
# ============================================================

age = 35
experience = 8
python_skill = True
data_analysis_skill = True

if (
    age >= 18
    and experience >= 3
    and python_skill
    and data_analysis_skill
):
    print("Candidate meets the requirements.")

else:
    print("Candidate does not meet all requirements.")
```
