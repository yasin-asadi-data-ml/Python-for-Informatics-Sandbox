```python
# ============================================================
#             PYTHON CONDITIONAL STATEMENTS
# ============================================================
#
# Conditional statements allow Python to make decisions.
#
# Basic structure:
#
# if condition:
#     code
#
# elif another_condition:
#     code
#
# else:
#     code
#
# ============================================================


# ============================================================
# 1. BASIC IF STATEMENT
# ============================================================

age = 25

if age >= 18:
    print("You are an adult.")


# ============================================================
# 2. IF + ELSE
# ============================================================

age = 16

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")


# ============================================================
# 3. IF + ELIF + ELSE
# ============================================================

age = 25

if age < 13:
    print("Child")

elif age < 18:
    print("Teenager")

else:
    print("Adult")


# ============================================================
# 4. MULTIPLE CONDITIONS
# ============================================================

age = 30
has_license = True

if age >= 18 and has_license:
    print("You can drive.")
else:
    print("You cannot drive.")


# ============================================================
# 5. COMPARISON OPERATORS
# ============================================================

x = 10
y = 20

print(x == y)       # equal to
print(x != y)       # not equal to
print(x > y)        # greater than
print(x < y)        # less than
print(x >= y)       # greater than or equal to
print(x <= y)       # less than or equal to


# ============================================================
# 6. USING COMPARISON OPERATORS IN IF
# ============================================================

score = 85

if score >= 50:
    print("Pass")
else:
    print("Fail")


# ============================================================
# 7. AND
# ============================================================

age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Condition satisfied.")


# ============================================================
# 8. OR
# ============================================================

department = "IT"

if department == "IT" or department == "Data Science":
    print("Technology-related department.")


# ============================================================
# 9. NOT
# ============================================================

is_student = False

if not is_student:
    print("The person is not a student.")


# ============================================================
# 10. COMBINING AND / OR / NOT
# ============================================================

age = 30
experience = 6
has_degree = True

if (age >= 25 and experience >= 5) or has_degree:
    print("Candidate satisfies the requirement.")


# ============================================================
# 11. NESTED IF
# ============================================================

age = 30
has_license = True

if age >= 18:

    if has_license:
        print("You can drive.")

    else:
        print("You need a driving licence.")

else:
    print("You are too young to drive.")


# ============================================================
# 12. MULTIPLE ELIF CONDITIONS
# ============================================================

score = 87

if score >= 90:
    grade = "A"

elif score >= 80:
    grade = "B"

elif score >= 70:
    grade = "C"

elif score >= 60:
    grade = "D"

else:
    grade = "F"

print(grade)


# ============================================================
# 13. CHECKING RANGES
# ============================================================

age = 25

if 18 <= age <= 65:
    print("Working-age adult.")


# ============================================================
# 14. RANGE CONDITIONS
# ============================================================

temperature = 25

if temperature < 0:
    print("Freezing")

elif 0 <= temperature < 15:
    print("Cold")

elif 15 <= temperature < 25:
    print("Moderate")

elif 25 <= temperature < 35:
    print("Warm")

else:
    print("Hot")


# ============================================================
# 15. MEMBERSHIP WITH IN
# ============================================================

department = "IT"

if department in ["IT", "Data Science", "AI"]:
    print("Technology field.")


# ============================================================
# 16. MEMBERSHIP WITH NOT IN
# ============================================================

country = "Iran"

if country not in ["USA", "UK", "Canada"]:
    print("Country is not in the selected list.")


# ============================================================
# 17. STRINGS WITH CONDITIONS
# ============================================================

name = "Yasin"

if name == "Yasin":
    print("Hello Yasin!")


# Case-insensitive comparison
name = "YASIN"

if name.lower() == "yasin":
    print("Name matched.")


# ============================================================
# 18. STRING METHODS IN CONDITIONS
# ============================================================

email = "yasin@example.com"

if email.endswith(".com"):
    print("This is a .com address.")


if "@" in email:
    print("This looks like an email address.")


# ============================================================
# 19. CHECKING EMPTY STRINGS
# ============================================================

name = ""

if name:
    print("Name exists.")
else:
    print("Name is empty.")


# ============================================================
# 20. TRUTHY AND FALSY VALUES
# ============================================================

# These values are generally FALSY:
#
# False
# None
# 0
# 0.0
# ""
# []
# {}
# ()
# set()


value = 0

if value:
    print("Truthy")
else:
    print("Falsy")


# ============================================================
# 21. CHECKING NONE
# ============================================================

value = None

if value is None:
    print("No value was provided.")
else:
    print("A value exists.")


# ============================================================
# 22. IS NOT NONE
# ============================================================

value = 100

if value is not None:
    print("Value exists.")


# ============================================================
# 23. CONDITIONAL ASSIGNMENT
# ============================================================

age = 25

if age >= 18:
    status = "Adult"
else:
    status = "Minor"

print(status)


# ============================================================
# 24. TERNARY CONDITIONAL EXPRESSION
# ============================================================

age = 25

status = "Adult" if age >= 18 else "Minor"

print(status)


# ============================================================
# 25. TERNARY WITH NUMBERS
# ============================================================

number = 10

result = "Positive" if number > 0 else "Not positive"

print(result)


# ============================================================
# 26. CHECKING POSITIVE / NEGATIVE / ZERO
# ============================================================

number = -5

if number > 0:
    print("Positive")

elif number < 0:
    print("Negative")

else:
    print("Zero")


# ============================================================
# 27. EVEN OR ODD
# ============================================================

number = 17

if number % 2 == 0:
    print("Even")

else:
    print("Odd")


# ============================================================
# 28. DIVISIBILITY
# ============================================================

number = 30

if number % 5 == 0:
    print("Divisible by 5.")

else:
    print("Not divisible by 5.")


# ============================================================
# 29. MULTIPLE DIVISIBILITY CONDITIONS
# ============================================================

number = 30

if number % 3 == 0 and number % 5 == 0:
    print("Divisible by both 3 and 5.")

elif number % 3 == 0:
    print("Divisible by 3.")

elif number % 5 == 0:
    print("Divisible by 5.")

else:
    print("Not divisible by 3 or 5.")


# ============================================================
# 30. DATA ANALYTICS — CLASSIFYING VALUES
# ============================================================

salary = 75000

if salary < 40000:
    category = "Low"

elif salary < 70000:
    category = "Medium"

else:
    category = "High"

print(category)


# ============================================================
# 31. DATA ANALYTICS — CUSTOMER SEGMENTATION
# ============================================================

purchase = 2500

if purchase < 500:
    customer_type = "Low Value"

elif purchase < 2000:
    customer_type = "Medium Value"

else:
    customer_type = "High Value"

print(customer_type)


# ============================================================
# 32. DATA ANALYTICS — RISK CLASSIFICATION
# ============================================================

risk_score = 72

if risk_score < 30:
    risk = "Low Risk"

elif risk_score < 70:
    risk = "Medium Risk"

else:
    risk = "High Risk"

print(risk)


# ============================================================
# 33. DATA ANALYTICS — MISSING VALUE LOGIC
# ============================================================

value = None

if value is None:
    print("Missing value.")

else:
    print("Value available.")


# ============================================================
# 34. DATA ANALYTICS — DATA QUALITY
# ============================================================

age = -5

if age < 0:
    print("Invalid age.")

elif age > 120:
    print("Suspicious age.")

else:
    print("Age appears valid.")


# ============================================================
# 35. DATA ANALYTICS — OUTLIER LOGIC
# ============================================================

salary = 250000

Q1 = 40000
Q3 = 80000

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

if salary < lower_limit or salary > upper_limit:
    print("Potential outlier.")

else:
    print("Within expected range.")


# ============================================================
# 36. DATA ANALYTICS — PERFORMANCE CLASSIFICATION
# ============================================================

accuracy = 0.87

if accuracy >= 0.90:
    performance = "Excellent"

elif accuracy >= 0.80:
    performance = "Good"

elif accuracy >= 0.70:
    performance = "Acceptable"

else:
    performance = "Poor"

print(performance)


# ============================================================
# 37. MACHINE LEARNING — CLASSIFICATION LOGIC
# ============================================================

probability = 0.82

if probability >= 0.5:
    prediction = 1

else:
    prediction = 0

print(prediction)


# ============================================================
# 38. MACHINE LEARNING — RISK PREDICTION
# ============================================================

failure_probability = 0.78

if failure_probability >= 0.80:
    prediction = "Very High Risk"

elif failure_probability >= 0.60:
    prediction = "High Risk"

elif failure_probability >= 0.40:
    prediction = "Medium Risk"

else:
    prediction = "Low Risk"

print(prediction)


# ============================================================
# 39. USING CONDITIONS WITH LISTS
# ============================================================

scores = [85, 42, 91, 67, 55]

for score in scores:

    if score >= 50:
        print(score, "Pass")

    else:
        print(score, "Fail")


# ============================================================
# 40. CONDITIONS INSIDE A LOOP
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]

for number in numbers:

    if number % 2 == 0:
        print(number, "Even")

    else:
        print(number, "Odd")


# ============================================================
# 41. FILTERING DATA WITH A CONDITION
# ============================================================

numbers = [10, 25, 40, 55, 70]

for number in numbers:

    if number > 40:
        print(number)


# ============================================================
# 42. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [10, 20, 30, 40, 50]

large_numbers = [
    number
    for number in numbers
    if number > 25
]

print(large_numbers)


# ============================================================
# 43. LIST COMPREHENSION WITH IF-ELSE
# ============================================================

numbers = [10, 20, 30, 40, 50]

categories = [
    "High" if number >= 30 else "Low"
    for number in numbers
]

print(categories)


# ============================================================
# 44. CONDITIONAL LOGIC WITH PANDAS
# ============================================================

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Yasin", "John"],
    "Age": [25, 32, 41, 19],
    "Salary": [35000, 55000, 90000, 28000]
})

print(df)


# ============================================================
# 45. PANDAS BOOLEAN FILTERING
# ============================================================

print(df[df["Age"] >= 30])

print(df[df["Salary"] > 50000])


# ============================================================
# 46. PANDAS AND CONDITION
# ============================================================

print(
    df[
        (df["Age"] >= 30) &
        (df["Salary"] > 50000)
    ]
)


# ============================================================
# 47. PANDAS OR CONDITION
# ============================================================

print(
    df[
        (df["Age"] < 25) |
        (df["Salary"] > 80000)
    ]
)


# ============================================================
# 48. PANDAS NOT CONDITION
# ============================================================

print(
    df[
        ~(df["Age"] >= 30)
    ]
)


# ============================================================
# 49. PANDAS np.where()
# ============================================================

df["Age_Group"] = np.where(
    df["Age"] >= 30,
    "30+",
    "Under 30"
)

print(df)


# ============================================================
# 50. PANDAS MULTIPLE CONDITIONS WITH np.select()
# ============================================================

conditions = [
    df["Salary"] < 40000,
    df["Salary"] < 70000,
    df["Salary"] >= 70000
]

choices = [
    "Low",
    "Medium",
    "High"
]

df["Salary_Level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df)


# ============================================================
# 51. PRACTICAL ANALYTICS EXAMPLE
# ============================================================

df["Performance"] = np.where(
    df["Salary"] >= 50000,
    "High",
    "Low"
)

print(df)


# ============================================================
# 52. PRACTICAL DATA QUALITY EXAMPLE
# ============================================================

df["Age_Status"] = np.where(
    (df["Age"] >= 18) & (df["Age"] <= 100),
    "Valid",
    "Check"
)

print(df)


# ============================================================
# 53. CONDITIONS WITH FUNCTIONS
# ============================================================

def classify_salary(salary):

    if salary < 40000:
        return "Low"

    elif salary < 70000:
        return "Medium"

    else:
        return "High"


print(classify_salary(35000))
print(classify_salary(60000))
print(classify_salary(90000))


# ============================================================
# 54. APPLYING A FUNCTION TO A DATAFRAME
# ============================================================

df["Salary_Level"] = df["Salary"].apply(
    classify_salary
)

print(df)


# ============================================================
# 55. PRACTICAL ML METRIC CLASSIFICATION
# ============================================================

precision = 0.82
recall = 0.74

if precision >= 0.80 and recall >= 0.80:
    print("Strong performance on both metrics.")

elif precision >= 0.80 and recall < 0.80:
    print("Good precision but recall needs improvement.")

elif precision < 0.80 and recall >= 0.80:
    print("Good recall but precision needs improvement.")

else:
    print("Both metrics need improvement.")


# ============================================================
# 56. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================

failure_probability = 0.73

if failure_probability >= 0.80:
    action = "Immediate inspection"

elif failure_probability >= 0.60:
    action = "Schedule inspection"

elif failure_probability >= 0.40:
    action = "Monitor"

else:
    action = "Normal operation"

print(action)


# ============================================================
# 57. COMBINING MULTIPLE FEATURES
# ============================================================

pipe_age = 35
pressure = 90
failure_history = True

if (
    pipe_age > 30
    and pressure > 80
    and failure_history
):
    risk = "High"

elif (
    pipe_age > 30
    or pressure > 80
    or failure_history
):
    risk = "Medium"

else:
    risk = "Low"

print(risk)


# ============================================================
# 58. MATCH-CASE
# ============================================================

department = "IT"

match department:

    case "IT":
        print("Information Technology")

    case "HR":
        print("Human Resources")

    case "Finance":
        print("Finance")

    case _:
        print("Unknown department")


# ============================================================
# 59. MATCH-CASE WITH MULTIPLE OPTIONS
# ============================================================

department = "AI"

match department:

    case "AI" | "ML" | "Data Science":
        print("Data/AI field")

    case "Network":
        print("Networking field")

    case "Finance":
        print("Finance field")

    case _:
        print("Other field")


# ============================================================
# 60. PRACTICAL DECISION-MAKING EXAMPLE
# ============================================================

age = 35
experience = 8
python_skill = True
data_analysis_skill = True

if (
    age >= 18
    and experience >= 3
    and python_skill
    and data_analysis_skill
):
    print("Candidate meets the requirements.")

else:
    print("Candidate does not meet all requirements.")
```
