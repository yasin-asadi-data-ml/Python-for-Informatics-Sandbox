```python
# ============================================================
#                    PYTHON FUNCTIONS
# ============================================================
#
# A function is a reusable block of code.
#
# Basic structure:
#
# def function_name(parameters):
#     code
#
# ============================================================


# ============================================================
# 1. SIMPLE FUNCTION
# ============================================================

def say_hello():
    print("Hello!")


say_hello()


# ============================================================
# 2. FUNCTION WITH A PARAMETER
# ============================================================

def greet(name):
    print("Hello", name)


greet("Ali")
greet("Sara")
greet("Yasin")


# ============================================================
# 3. MULTIPLE PARAMETERS
# ============================================================

def introduce(name, age):
    print("Name:", name)
    print("Age:", age)


introduce("Yasin", 35)


# ============================================================
# 4. FUNCTION WITH THREE PARAMETERS
# ============================================================

def introduce_person(name, age, field):

    print("Name:", name)
    print("Age:", age)
    print("Field:", field)


introduce_person(
    "Yasin",
    35,
    "Data Science"
)


# ============================================================
# 5. RETURN
# ============================================================
#
# print() displays a value.
#
# return sends a value back to the caller.
#
# ============================================================

def add(a, b):

    result = a + b

    return result


answer = add(10, 20)

print(answer)


# ============================================================
# 6. RETURN VS PRINT
# ============================================================

def add_with_print(a, b):

    print(a + b)


def add_with_return(a, b):

    return a + b


add_with_print(10, 20)

result = add_with_return(10, 20)

print(result)


# ============================================================
# 7. RETURN MULTIPLE VALUES
# ============================================================

def calculate(a, b):

    total = a + b
    difference = a - b
    product = a * b

    return total, difference, product


result = calculate(10, 5)

print(result)


# ============================================================
# 8. UNPACKING RETURNED VALUES
# ============================================================

def calculate(a, b):

    total = a + b
    difference = a - b
    product = a * b

    return total, difference, product


total, difference, product = calculate(10, 5)

print(total)
print(difference)
print(product)


# ============================================================
# 9. DEFAULT PARAMETERS
# ============================================================

def greet(name="User"):

    print("Hello", name)


greet()

greet("Yasin")


# ============================================================
# 10. MULTIPLE DEFAULT PARAMETERS
# ============================================================

def introduce(
    name="Unknown",
    age=0,
    field="Unknown"
):

    print(name)
    print(age)
    print(field)


introduce()

introduce(
    "Yasin",
    35,
    "Data Science"
)


# ============================================================
# 11. POSITIONAL ARGUMENTS
# ============================================================

def student(name, age, field):

    print(name)
    print(age)
    print(field)


student(
    "Yasin",
    35,
    "Data Science"
)


# ============================================================
# 12. KEYWORD ARGUMENTS
# ============================================================

student(
    name="Yasin",
    age=35,
    field="Data Science"
)


# ============================================================
# 13. KEYWORD ARGUMENTS IN DIFFERENT ORDER
# ============================================================

student(
    field="Data Science",
    name="Yasin",
    age=35
)


# ============================================================
# 14. MIXING POSITIONAL AND KEYWORD ARGUMENTS
# ============================================================

student(
    "Yasin",
    age=35,
    field="Data Science"
)


# ============================================================
# 15. *ARGS
# ============================================================
#
# *args allows a function to receive
# an arbitrary number of positional arguments.
#
# ============================================================

def add_numbers(*numbers):

    total = 0

    for number in numbers:

        total += number

    return total


print(add_numbers(10, 20))
print(add_numbers(10, 20, 30))
print(add_numbers(10, 20, 30, 40, 50))


# ============================================================
# 16. *ARGS IS A TUPLE
# ============================================================

def show_values(*values):

    print(values)
    print(type(values))


show_values(10, 20, 30)


# ============================================================
# 17. **KWARGS
# ============================================================
#
# **kwargs allows arbitrary keyword arguments.
#
# ============================================================

def show_information(**information):

    print(information)


show_information(
    name="Yasin",
    age=35,
    field="Data Science"
)


# ============================================================
# 18. **KWARGS IS A DICTIONARY
# ============================================================

def show_information(**information):

    print(type(information))

    for key, value in information.items():

        print(key, ":", value)


show_information(
    name="Yasin",
    age=35,
    field="Data Science"
)


# ============================================================
# 19. *ARGS AND **KWARGS TOGETHER
# ============================================================

def example(*args, **kwargs):

    print("args:", args)
    print("kwargs:", kwargs)


example(
    10,
    20,
    30,
    name="Yasin",
    field="AI"
)


# ============================================================
# 20. FUNCTION WITH A LIST
# ============================================================

def calculate_sum(numbers):

    total = 0

    for number in numbers:

        total += number

    return total


numbers = [10, 20, 30, 40]

print(calculate_sum(numbers))


# ============================================================
# 21. FUNCTION WITH A TUPLE
# ============================================================

def calculate_average(numbers):

    total = sum(numbers)

    average = total / len(numbers)

    return average


numbers = (10, 20, 30, 40)

print(calculate_average(numbers))


# ============================================================
# 22. FUNCTION WITH A DICTIONARY
# ============================================================

def show_student(student):

    print("Name:", student["name"])
    print("Age:", student["age"])
    print("Field:", student["field"])


student = {
    "name": "Yasin",
    "age": 35,
    "field": "Data Science"
}

show_student(student)


# ============================================================
# 23. FUNCTION WITH A SET
# ============================================================

def count_unique(values):

    return len(set(values))


numbers = [10, 10, 20, 30, 30, 40]

print(count_unique(numbers))


# ============================================================
# 24. FUNCTION + CONDITION
# ============================================================

def classify_age(age):

    if age < 18:
        return "Minor"

    elif age < 65:
        return "Adult"

    else:
        return "Senior"


print(classify_age(25))
print(classify_age(70))


# ============================================================
# 25. FUNCTION + LOOP
# ============================================================

def calculate_total(numbers):

    total = 0

    for number in numbers:

        total += number

    return total


numbers = [10, 20, 30, 40]

print(calculate_total(numbers))


# ============================================================
# 26. FUNCTION + LOOP + CONDITION
# ============================================================

def count_even_numbers(numbers):

    count = 0

    for number in numbers:

        if number % 2 == 0:

            count += 1

    return count


numbers = [1, 2, 3, 4, 5, 6]

print(count_even_numbers(numbers))


# ============================================================
# 27. FUNCTION FOR MINIMUM VALUE
# ============================================================

def find_minimum(numbers):

    minimum = numbers[0]

    for number in numbers:

        if number < minimum:

            minimum = number

    return minimum


print(find_minimum([10, 5, 30, 2, 20]))


# ============================================================
# 28. FUNCTION FOR MAXIMUM VALUE
# ============================================================

def find_maximum(numbers):

    maximum = numbers[0]

    for number in numbers:

        if number > maximum:

            maximum = number

    return maximum


print(find_maximum([10, 50, 30, 90, 20]))


# ============================================================
# 29. FUNCTION FOR AVERAGE
# ============================================================

def calculate_average(numbers):

    if len(numbers) == 0:

        return None

    return sum(numbers) / len(numbers)


print(calculate_average([10, 20, 30, 40]))


# ============================================================
# 30. FUNCTION FOR MEDIAN
# ============================================================

def calculate_median(numbers):

    numbers = sorted(numbers)

    n = len(numbers)

    middle = n // 2

    if n % 2 == 1:

        return numbers[middle]

    else:

        return (
            numbers[middle - 1]
            + numbers[middle]
        ) / 2


print(calculate_median([10, 20, 30]))
print(calculate_median([10, 20, 30, 40]))


# ============================================================
# 31. FUNCTION FOR RANGE
# ============================================================

def calculate_range(numbers):

    return max(numbers) - min(numbers)


print(calculate_range([10, 20, 30, 50]))


# ============================================================
# 32. FUNCTION FOR PERCENTAGE
# ============================================================

def calculate_percentage(part, total):

    if total == 0:

        return 0

    return (part / total) * 100


print(calculate_percentage(25, 100))


# ============================================================
# 33. DATA ANALYTICS — CLASSIFY SALES
# ============================================================

def classify_sales(sales):

    if sales < 1000:

        return "Low"

    elif sales < 5000:

        return "Medium"

    else:

        return "High"


print(classify_sales(750))
print(classify_sales(2500))
print(classify_sales(8000))


# ============================================================
# 34. DATA ANALYTICS — CLEAN A STRING
# ============================================================

def clean_text(text):

    text = text.strip()

    text = text.lower()

    return text


print(clean_text("   PYTHON   "))


# ============================================================
# 35. DATA ANALYTICS — CLEAN A LIST OF STRINGS
# ============================================================

def clean_names(names):

    cleaned_names = []

    for name in names:

        name = name.strip()
        name = name.title()

        cleaned_names.append(name)

    return cleaned_names


names = [
    " ali ",
    "SARA",
    " yasin"
]

print(clean_names(names))


# ============================================================
# 36. DATA ANALYTICS — REMOVE MISSING VALUES
# ============================================================

def remove_missing(values):

    cleaned = []

    for value in values:

        if value is not None:

            cleaned.append(value)

    return cleaned


values = [10, None, 20, None, 30]

print(remove_missing(values))


# ============================================================
# 37. DATA ANALYTICS — COUNT MISSING VALUES
# ============================================================

def count_missing(values):

    count = 0

    for value in values:

        if value is None:

            count += 1

    return count


values = [10, None, 20, None, 30]

print(count_missing(values))


# ============================================================
# 38. DATA ANALYTICS — MISSING VALUE RATE
# ============================================================

def missing_rate(values):

    if len(values) == 0:

        return 0

    missing = 0

    for value in values:

        if value is None:

            missing += 1

    return missing / len(values) * 100


values = [10, None, 20, None, 30]

print(missing_rate(values))


# ============================================================
# 39. DATA ANALYTICS — FILTER VALUES
# ============================================================

def filter_large_values(values, threshold):

    result = []

    for value in values:

        if value > threshold:

            result.append(value)

    return result


numbers = [10, 20, 30, 40, 50]

print(
    filter_large_values(
        numbers,
        25
    )
)


# ============================================================
# 40. DATA ANALYTICS — TRANSFORM VALUES
# ============================================================

def add_tax(prices, tax_rate):

    result = []

    for price in prices:

        new_price = price * (
            1 + tax_rate
        )

        result.append(new_price)

    return result


prices = [100, 200, 300]

print(
    add_tax(
        prices,
        0.20
    )
)


# ============================================================
# 41. DATA ANALYTICS — Z-SCORE
# ============================================================

def z_scores(values):

    mean = sum(values) / len(values)

    squared_differences = []

    for value in values:

        squared_difference = (
            value - mean
        ) ** 2

        squared_differences.append(
            squared_difference
        )

    variance = (
        sum(squared_differences)
        / len(values)
    )

    standard_deviation = variance ** 0.5

    result = []

    for value in values:

        z = (
            value - mean
        ) / standard_deviation

        result.append(z)

    return result


values = [10, 20, 30, 40, 50]

print(z_scores(values))


# ============================================================
# 42. DATA ANALYTICS — SIMPLE OUTLIER CHECK
# ============================================================

def find_large_values(values, threshold):

    outliers = []

    for value in values:

        if value > threshold:

            outliers.append(value)

    return outliers


values = [
    10,
    12,
    14,
    15,
    100
]

print(
    find_large_values(
        values,
        50
    )
)


# ============================================================
# 43. FUNCTION RETURNING A DICTIONARY
# ============================================================

def descriptive_statistics(values):

    result = {

        "count": len(values),

        "mean": sum(values) / len(values),

        "minimum": min(values),

        "maximum": max(values)
    }

    return result


values = [10, 20, 30, 40, 50]

stats = descriptive_statistics(values)

print(stats)


# ============================================================
# 44. FUNCTION RETURNING MULTIPLE STATISTICS
# ============================================================

def basic_statistics(values):

    count = len(values)

    mean = sum(values) / count

    minimum = min(values)

    maximum = max(values)

    return (
        count,
        mean,
        minimum,
        maximum
    )


count, mean, minimum, maximum = (
    basic_statistics(
        [10, 20, 30, 40]
    )
)

print(count)
print(mean)
print(minimum)
print(maximum)


# ============================================================
# 45. FUNCTION + LIST COMPREHENSION
# ============================================================

def square_numbers(numbers):

    return [
        number ** 2
        for number in numbers
    ]


print(
    square_numbers(
        [1, 2, 3, 4, 5]
    )
)


# ============================================================
# 46. FUNCTION + DICTIONARY COMPREHENSION
# ============================================================

def create_square_dictionary(numbers):

    return {
        number: number ** 2
        for number in numbers
    }


print(
    create_square_dictionary(
        [1, 2, 3, 4]
    )
)


# ============================================================
# 47. LAMBDA FUNCTIONS
# ============================================================
#
# Lambda = small anonymous function.
#
# ============================================================

square = lambda x: x ** 2

print(square(5))


# ============================================================
# 48. LAMBDA WITH TWO PARAMETERS
# ============================================================

add = lambda a, b: a + b

print(add(10, 20))


# ============================================================
# 49. MAP()
# ============================================================

numbers = [1, 2, 3, 4, 5]

squares = list(
    map(
        lambda x: x ** 2,
        numbers
    )
)

print(squares)


# ============================================================
# 50. FILTER()
# ============================================================

numbers = [10, 15, 20, 25, 30]

large_numbers = list(
    filter(
        lambda x: x > 20,
        numbers
    )
)

print(large_numbers)


# ============================================================
# 51. SORTING WITH A FUNCTION
# ============================================================

students = [

    {"name": "Ali", "score": 75},

    {"name": "Sara", "score": 90},

    {"name": "Yasin", "score": 85}
]

students_sorted = sorted(
    students,
    key=lambda student: student["score"],
    reverse=True
)

print(students_sorted)


# ============================================================
# 52. NESTED FUNCTION
# ============================================================

def outer_function():

    def inner_function():

        print("Inside inner function")

    inner_function()


outer_function()


# ============================================================
# 53. SCOPE
# ============================================================

def test():

    x = 10

    print(x)


test()

# x does not exist outside
# the function.


# ============================================================
# 54. LOCAL VARIABLE
# ============================================================

def calculate():

    result = 10 + 20

    return result


print(calculate())


# ============================================================
# 55. GLOBAL VARIABLE
# ============================================================

tax_rate = 0.20


def calculate_tax(price):

    return price * tax_rate


print(calculate_tax(100))


# ============================================================
# 56. AVOID UNNECESSARY GLOBAL VARIABLES
# ============================================================
#
# Better approach:
#
# Pass information into functions.
#
# ============================================================

def calculate_tax(price, tax_rate):

    return price * tax_rate


print(
    calculate_tax(
        100,
        0.20
    )
)


# ============================================================
# 57. DOCSTRING
# ============================================================

def calculate_average(numbers):
    """
    Calculate the arithmetic mean
    of a collection of numbers.
    """

    return sum(numbers) / len(numbers)


print(
    calculate_average(
        [10, 20, 30]
    )
)


# ============================================================
# 58. TYPE HINTS
# ============================================================

def add(a: float, b: float) -> float:

    return a + b


print(add(10, 20))


# ============================================================
# 59. TYPE HINTS WITH LIST
# ============================================================

def calculate_average(
    numbers: list[float]
) -> float:

    return sum(numbers) / len(numbers)


print(
    calculate_average(
        [10.0, 20.0, 30.0]
    )
)


# ============================================================
# 60. PRACTICAL ML EXAMPLE
# ============================================================

def classify_probability(probability):

    if probability >= 0.80:

        return "High Risk"

    elif probability >= 0.50:

        return "Medium Risk"

    else:

        return "Low Risk"


probabilities = [
    0.20,
    0.55,
    0.82,
    0.91
]

for probability in probabilities:

    print(
        probability,
        "->",
        classify_probability(
            probability
        )
    )


# ============================================================
# 61. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================

def recommend_pipeline_action(
    failure_probability
):

    if failure_probability >= 0.80:

        return "Immediate inspection"

    elif failure_probability >= 0.60:

        return "Schedule inspection"

    elif failure_probability >= 0.40:

        return "Monitor"

    else:

        return "Normal operation"


probabilities = [
    0.15,
    0.42,
    0.73,
    0.91,
    0.31
]

for probability in probabilities:

    action = recommend_pipeline_action(
        probability
    )

    print(
        probability,
        "->",
        action
    )


# ============================================================
# 62. PRACTICAL ML METRICS FUNCTION
# ============================================================

def evaluate_model(
    accuracy,
    precision,
    recall,
    f1
):

    return {

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1": f1
    }


metrics = evaluate_model(
    0.91,
    0.84,
    0.76,
    0.80
)

print(metrics)


# ============================================================
# 63. FUNCTION FOR DATA QUALITY
# ============================================================

def data_quality_report(values):

    total = len(values)

    missing = 0

    for value in values:

        if value is None:

            missing += 1

    missing_percentage = (
        missing / total
    ) * 100

    return {

        "Total": total,

        "Missing": missing,

        "Missing %": missing_percentage
    }


data = [
    10,
    20,
    None,
    40,
    None,
    60
]

print(
    data_quality_report(data)
)


# ============================================================
# 64. FUNCTIONS CALLING OTHER FUNCTIONS
# ============================================================

def calculate_total(values):

    return sum(values)


def calculate_average(values):

    total = calculate_total(values)

    return total / len(values)


values = [10, 20, 30, 40]

print(
    calculate_average(values)
)


# ============================================================
# 65. MAIN FUNCTION
# ============================================================

def main():

    numbers = [10, 20, 30, 40]

    average = calculate_average(
        numbers
    )

    print("Average:", average)


main()


# ============================================================
# 66. __name__ == "__main__"
# ============================================================

def main():

    print("Program started")


if __name__ == "__main__":

    main()
```
