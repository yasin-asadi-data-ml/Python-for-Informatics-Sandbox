## Python Functions


```python
# ============================================================
#                    PYTHON FUNCTIONS
# ============================================================
#
# A function is a reusable block of code that performs
# a specific task.
#
# Functions help you:
#   - Organize your code
#   - Avoid repeating the same code
#   - Make programs easier to read and maintain
#   - Reuse the same logic whenever needed
#
# A function is defined using the "def" keyword.
#
# Basic structure:
#
# def function_name(parameters):
#     code
#
# "def"        -> Defines a new function
# "function_name" -> The name of the function
# "parameters" -> Optional input values
# ":"          -> Starts the function body
# "code"       -> Instructions executed by the function
#
# A function runs only when it is called.
#
# Example:
#
# def greet():
#     print("Hello!")
#
# greet()
#
# ============================================================
```

```python
# ============================================================
# 1. SIMPLE FUNCTION
# ============================================================
#
# A function is a reusable block of code designed to
# perform a specific task.
#
# In this example, the function simply prints "Hello!".
#
# ============================================================


# Define a function named "say_hello".
#
# The "def" keyword tells Python that we are creating
# a new function.
#
# "say_hello" is the name of the function.
#
# The parentheses "()" are used for parameters.
# There are no parameters in this function, so they are empty.
#
# The colon ":" marks the beginning of the function body.
def say_hello():

    # This is a docstring.
    # A docstring describes what the function does.
    #
    # This function prints a greeting message.
    """
    Print a simple greeting message.
    This function does not require any input
    and does not return a value.
    """

    # Print "Hello!" to the console.
    #
    # This line is part of the function because it is
    # indented underneath the function definition.
    print("Hello!")


# ============================================================
# FUNCTION CALL
# ============================================================
#
# Defining a function does not execute its code.
#
# We must call the function to tell Python to run it.
#
# Writing "say_hello()" calls the function.
# Python then executes the code inside the function.
# ============================================================

say_hello()


# ============================================================
# OUTPUT
# ============================================================
#
# Hello!
#
# ============================================================
```

```python
# ============================================================
# 2. FUNCTION WITH A PARAMETER
# ============================================================
#
# A parameter allows a function to receive input.
# The value passed to the function is called an argument.
#
# In this example, the "name" parameter receives a different
# name each time the function is called.
# ============================================================

# Define a function named "greet" with one parameter: "name".
def greet(name):

    # Print a greeting message using the value of "name".
    print("Hello", name)


# Call the function with different arguments.
# Each argument is passed to the "name" parameter.
greet("Ali")       # Pass "Ali" as the argument
greet("Sara")      # Pass "Sara" as the argument
greet("Yasin")     # Pass "Yasin" as the argument
```

```python
# ============================================================
# 2. FUNCTION WITH A PARAMETER
# ============================================================
#
# A parameter allows a function to receive input.
# Here, "name" is the parameter that receives a name
# when the function is called.
# ============================================================

# Define a function named "greet" with one parameter: "name".
def greet(name):

    # Print a greeting using the value stored in "name".
    print("Hello", name)


# Call the function and pass different arguments to "name".
greet("Ali")       # "Ali" is passed to the "name" parameter
greet("Sara")      # "Sara" is passed to the "name" parameter
greet("Yasin")     # "Yasin" is passed to the "name" parameter
```

```python
# ============================================================
# 4. FUNCTION WITH THREE PARAMETERS
# ============================================================
#
# A function can accept multiple parameters.
# In this example, the function receives three pieces of
# information: a person's name, age, and field.
#
# The arguments are passed to the parameters in the same
# order in which the parameters are defined.
# ============================================================

# Define a function with three parameters:
# "name", "age", and "field".
def introduce_person(name, age, field):

    # Display the person's name.
    print("Name:", name)

    # Display the person's age.
    print("Age:", age)

    # Display the person's field of study or work.
    print("Field:", field)


# Call the function and provide three arguments.
# "Yasin" -> name
# 35      -> age
# "Data Science" -> field
introduce_person(
    "Yasin",
    35,
    "Data Science"
)
```

```python
# ============================================================
# 5. RETURN
# ============================================================
#
# print() displays a value on the screen.
#
# return sends a value from the function back to the caller.
#
# The returned value can then be stored in a variable,
# used in another calculation, or passed to another function.
# ============================================================

# Define a function named "add" with two parameters: "a" and "b".
def add(a, b):

    # Add "a" and "b" and store the result in "result".
    result = a + b

    # Return the result to the code that called the function.
    return result


# Call the function with 10 and 20 as arguments.
# Store the returned value in the variable "answer".
answer = add(10, 20)


# Display the value stored in "answer".
print(answer)  # Output: 30
```

```python
# ============================================================
# 6. RETURN VS PRINT
# ============================================================
#
# print() displays a value on the screen.
# return sends a value back to the caller.
#
# A printed value cannot be directly stored or reused by the
# calling code, while a returned value can be stored in a
# variable and used later.
# ============================================================


# Define a function that calculates and prints the sum directly.
def add_with_print(a, b):

    # Calculate the sum and display it on the screen.
    print(a + b)


# Define a function that calculates and returns the sum.
def add_with_return(a, b):

    # Calculate the sum and send the result back to the caller.
    return a + b


# Call the function with 10 and 20.
# The result is displayed directly by print().
add_with_print(10, 20)  # Output: 30


# Call the function with 10 and 20.
# The returned value is stored in the variable "result".
result = add_with_return(10, 20)


# Display the value stored in "result".
print(result)  # Output: 30
```

```python
# ============================================================
# 7. RETURN MULTIPLE VALUES
# ============================================================
#
# A function can return more than one value.
#
# In this example, the function calculates the sum,
# difference, and product of two numbers and returns
# all three results.
# ============================================================


# Define a function with two parameters: "a" and "b".
def calculate(a, b):

    # Calculate the sum of a and b.
    total = a + b

    # Calculate the difference between a and b.
    difference = a - b

    # Calculate the product of a and b.
    product = a * b

    # Return all three values.
    # Python automatically packs them into a tuple.
    return total, difference, product


# ============================================================
# CALLING THE FUNCTION
# ============================================================

# Call the function with 10 and 5 as arguments.
# The three returned values are stored together in "result".
result = calculate(10, 5)


# Display the returned tuple.
print(result)  # Output: (15, 5, 50)
```

```python
# Multiple values can be returned by separating them with commas.
# return total, difference, product

# Python automatically returns them as a tuple:
# (total, difference, product)
```

```python
# ============================================================
# 8. UNPACKING RETURNED VALUES
# ============================================================
#
# A function can return multiple values as a tuple.
# Python also allows us to unpack those values directly
# into separate variables.
# ============================================================


# Define a function with two parameters: "a" and "b".
def calculate(a, b):

    # Calculate the sum of a and b.
    total = a + b

    # Calculate the difference between a and b.
    difference = a - b

    # Calculate the product of a and b.
    product = a * b

    # Return the three values as a tuple.
    return total, difference, product


# ============================================================
# UNPACKING RETURNED VALUES
# ============================================================

# The function returns the tuple: (15, 5, 50).
# Python unpacks the three values into three separate variables.
#
# First value  -> total
# Second value -> difference
# Third value  -> product
total, difference, product = calculate(10, 5)


# Display each returned value separately.
print(total)       # Output: 15
print(difference)  # Output: 5
print(product)     # Output: 50


# Store all returned values in a single variable.
# "X" contains the complete tuple: (15, 5, 50).
X = calculate(10, 5)

# Display the tuple stored in "X".
print(X)           # Output: (15, 5, 50)
```

```python
# ============================================================
# 9. DEFAULT PARAMETERS
# ============================================================
#
# A parameter can have a default value.
# The default value is used when no argument is provided
# for that parameter when the function is called.
#
# Default parameters make arguments optional.
# ============================================================


# Define a function with one parameter: "name".
# If no argument is provided, "name" automatically gets "User".
def greet(name="User"):

    # Print a greeting message using the value of "name".
    print("Hello", name)


# ============================================================
# CALLING WITHOUT AN ARGUMENT
# ============================================================

# No argument is provided, so Python uses the default value "User".
greet()  # Output: Hello User


# ============================================================
# CALLING WITH AN ARGUMENT
# ============================================================

# "Yasin" is provided as an argument.
# The default value "User" is replaced by "Yasin".
greet("Yasin")  # Output: Hello Yasin
```

```python
# ============================================================
# 10. DEFAULT PARAMETERS: ORDER MATTERS
# ============================================================
#
# Parameters without default values are required.
# Parameters with default values are optional.
#
# A required parameter MUST come before a parameter
# with a default value.
# ============================================================


# ❌ WRONG
# A required parameter ("age") cannot come after a
# parameter with a default value ("name").
#
# This causes a SyntaxError.
#
# def bad_example(name="User", age):
#     print(name, age)


# ✅ CORRECT
# The required parameter "age" comes first.
# The optional parameter "name" comes after it.
def good_example(age, name="User"):

    # Display the name and age.
    print(name, "is", age, "years old")


# Provide only the required argument.
# "name" is not provided, so the default value "User" is used.
good_example(25)  # Output: User is 25 years old


# Provide both arguments.
# "Yasin" replaces the default value "User".
good_example(25, "Yasin")  # Output: Yasin is 25 years old


# ============================================================
# KEYWORD ARGUMENTS
# ============================================================
#
# A keyword argument passes a value using the parameter's name.
# This makes the function call more explicit and readable.
# ============================================================


# Pass the value "Yasin" directly to the "name" parameter.
greet(name="Yasin")  # Output: Hello Yasin
```

```python
# ============================================================
# THE CORE RULE
# ============================================================
#
# Required parameters must come before parameters
# with default values.
#
# Required parameter -> first
# Default parameter  -> after required parameters
#
# Example:
#
# def function(required, optional="default"):
#     ...
#
# This is because Python does not allow a required parameter
# to appear after a parameter that already has a default value.
# ============================================================
```

```python
# ============================================================
# 10. MULTIPLE DEFAULT PARAMETERS
# ============================================================
#
# A function can have multiple parameters with default values.
# Each parameter can have its own default value.
#
# When no argument is provided, Python uses the corresponding
# default value for each parameter.
# ============================================================


# Define a function with three parameters.
# All three parameters have default values, so all are optional.
def introduce(
    name="Unknown",    # Default value for the "name" parameter
    age=0,             # Default value for the "age" parameter
    field="Unknown"    # Default value for the "field" parameter
):

    # Display the person's name.
    print(name)

    # Display the person's age.
    print(age)

    # Display the person's field.
    print(field)


# ============================================================
# THE CORE RULE
# ============================================================
#
# Required parameters must come before parameters
# with default values.
#
# Required parameter -> first
# Default parameter  -> after required parameters
#
# Example:
#
# def function(required, optional="default"):
#     ...
#
# Multiple default parameters can appear after the required
# parameters, and each one can have its own default value.
# ============================================================
```

```python
# ============================================================
# CALL WITH NO ARGUMENTS
# ============================================================
#
# When no arguments are provided, Python uses the default
# value for every parameter.
#
# The function uses:
#   name  = "Unknown"
#   age   = 0
#   field = "Unknown"
# ============================================================

# Call the function without providing any arguments.
introduce()

# Output:
# Unknown
# 0
# Unknown
```

```python
# ============================================================
# CALL WITH ARGUMENTS
# ============================================================
#
# When arguments are provided, they replace the corresponding
# default values.
#
# The arguments are matched to the parameters in order:
#   1st argument -> name
#   2nd argument -> age
#   3rd argument -> field
#
# Since values are provided for all three parameters, none of
# the default values are used.
# ============================================================

# Call the function with three arguments.
introduce(
    "Yasin",        # 1st argument -> name
    35,             # 2nd argument -> age
    "Data Science"  # 3rd argument -> field
)

# Output:
# Yasin
# 35
# Data Science
```

```python
# ============================================================
# 11. PARTIAL OVERRIDING WITH KEYWORD ARGUMENTS
# ============================================================
#
# Default parameters allow us to provide only the values
# we want to change.
#
# Positional arguments are matched from left to right.
# Keyword arguments allow us to specify exactly which
# parameter we want to change.
# ============================================================


# ============================================================
# PARTIAL POSITIONAL ARGUMENTS
# ============================================================

# We can provide fewer positional arguments than the function
# has parameters.
#
# The provided values are assigned from left to right:
#   "Yasin" -> name
#
# The remaining parameters use their default values:
#   age   -> 0
#   field -> "Unknown"
print("--- Partial positional ---")

introduce("Yasin")

# Output:
# Yasin
# 0
# Unknown


# ============================================================
# KEYWORD ARGUMENTS: SKIP A PARAMETER
# ============================================================
#
# Positional arguments cannot skip a parameter in the middle.
#
# Keyword arguments solve this problem because we can specify
# the parameter by its name.
#
# Here, "name" and "field" are provided, while "age" is skipped.
# Therefore, "age" keeps its default value of 0.
print("--- Keyword arguments (skip age) ---")

introduce(
    name="Yasin",
    field="Data Science"
)

# Output:
# Yasin
# 0
# Data Science


# ============================================================
# MIXING POSITIONAL AND KEYWORD ARGUMENTS
# ============================================================
#
# Positional and keyword arguments can be used together.
#
# CORE RULE:
# All positional arguments must come BEFORE keyword arguments.
#
# Here:
#   "Yasin" -> name (positional)
#   age     -> default value: 0
#   field   -> "Data Science" (keyword)
print("--- Mixed (positional + keyword) ---")

introduce(
    "Yasin",
    field="Data Science"
)

# Output:
# Yasin
# 0
# Data Science
```

```python
# ============================================================
# 12. CRITICAL ORDER RULE
# ============================================================
#
# Parameters with default values MUST come after parameters
# without default values.
#
# This rule applies whether the function has one default
# parameter or multiple default parameters.
# ============================================================


# ✅ CORRECT
# Required parameters come first.
# Optional parameters with default values come afterward.
def correct_order(
    required_param,
    optional1="default1",
    optional2="default2"
):
    pass


# ❌ WRONG
# A required parameter cannot come after a parameter
# that already has a default value.
#
# This causes a SyntaxError.
#
# def wrong_order(
#     optional1="default1",
#     required_param,
#     optional2="default2"
# ):
#     pass


# ============================================================
# QUICK SUMMARY OF CALLING STYLES
# ============================================================
#
# 1. No arguments
#    -> All parameters use their default values.
#
# 2. All arguments
#    -> All default values are replaced by the provided values.
#
# 3. Some positional arguments
#    -> Values are assigned from left to right.
#    -> Remaining parameters use their default values.
#
# 4. Keyword arguments
#    -> Specific parameters can be changed by name.
#    -> Parameters not provided use their default values.
#
# 5. Mixed arguments
#    -> Positional arguments must come BEFORE keyword arguments.
#
# ============================================================
# CORE RULES TO REMEMBER
# ============================================================
#
# Function definition:
# Required parameters -> Default parameters
#
# Function call:
# Positional arguments -> Keyword arguments
#
# ============================================================
```

```python
# ============================================================
# 11. POSITIONAL ARGUMENTS
# ============================================================
#
# Positional arguments are matched to parameters by their
# position (order).
#
# The first argument is assigned to the first parameter,
# the second argument to the second parameter, and so on.
#
# Parameters without default values are REQUIRED.
# Therefore, the correct number of arguments must be provided.
# ============================================================


# Define a function with three required parameters.
def student(name, age, field):

    # Display the student's name.
    print(name)

    # Display the student's age.
    print(age)

    # Display the student's field.
    print(field)


# ============================================================
# CALLING WITH POSITIONAL ARGUMENTS
# ============================================================

# Arguments are assigned to parameters from left to right:
#   1st argument -> name
#   2nd argument -> age
#   3rd argument -> field
student("Yasin", 35, "Data Science")

# Output:
# Yasin
# 35
# Data Science

```

```python
# ============================================================
# ORDER MATTERS
# ============================================================
#
# Changing the order of the arguments changes which parameter
# receives each value.
# ============================================================

student(35, "Data Science", "Yasin")

# Output:
# 35
# Data Science
# Yasin

```

```python


# ============================================================
# POSITIONAL VS KEYWORD ARGUMENTS
# ============================================================
#
# Positional argument:
#     greet(25, "Yasin")
#
# Keyword argument:
#     greet(age=25, name="Yasin")
#
# Positional arguments are assigned by ORDER.
# Keyword arguments are assigned by PARAMETER NAME.
#
# Keyword arguments can make function calls clearer,
# especially when a function has many parameters.
#
# Keyword arguments will be covered in the next section.
# ============================================================
```

```python
# ============================================================
# EXACT NUMBER OF ARGUMENTS
# ============================================================
#
# When all parameters are required, the function call must
# provide exactly the same number of arguments as parameters.
#
# Too few arguments  -> TypeError
# Too many arguments -> TypeError
# ============================================================

# ❌ Too few arguments: the required "field" parameter is missing.
# student("Yasin", 35)

# ❌ Too many arguments: an extra argument is provided.
# student("Yasin", 35, "Data Science", "Extra")


# ============================================================
# POSITIONAL ARGUMENTS WITH DEFAULT PARAMETERS
# ============================================================
#
# A function can combine required parameters with parameters
# that have default values.
#
# Required parameters MUST come before default parameters.
#
# When calling the function:
#   - Required parameters must receive a value.
#   - Default parameters may use their default value.
# ============================================================


# "age" is a required parameter.
# "name" is an optional parameter with a default value.
def greet(age, name="User"):

    # Display the name and age.
    print(name, age)


# Provide only the required argument.
# "name" is not provided, so Python uses the default value "User".
greet(25)  # Output: User 25


# Provide both arguments positionally.
# 25 -> age
# "Yasin" -> name
# The provided name overrides the default value "User".
greet(25, "Yasin")  # Output: Yasin 25
```

```python
# ============================================================
# 12. KEYWORD ARGUMENTS
# ============================================================
#
# Keyword arguments are passed using the parameter's name.
#
# Unlike positional arguments, the order does NOT matter.
# Python matches each argument to the parameter by its name.
# ============================================================

# Each value is explicitly assigned to its parameter.
student(
    name="Yasin",
    age=35,
    field="Data Science"
)
```

```python
# ============================================================
# 13. KEYWORD ARGUMENTS IN DIFFERENT ORDER
# ============================================================
#
# Because the arguments are identified by parameter name,
# they can be provided in any order.
#
# The result is the same as the previous function call.
# ============================================================

student(
    field="Data Science",
    name="Yasin",
    age=35
)
```

```python
# ============================================================
# 15. *ARGS
# ============================================================
#
# *args allows a function to receive an arbitrary number
# of positional arguments.
#
# The arguments are collected into a tuple inside the function.
#
# The name "args" is a convention. You can use another name,
# but the * is what tells Python to collect the arguments.
# ============================================================


# Define a function that can accept any number of arguments.
def add_numbers(*numbers):

    # Start the total at 0.
    total = 0

    # Loop through each number received by the function.
    for number in numbers:

        # Add the current number to the total.
        total += number

    # Return the final sum.
    return total


# Pass two positional arguments to the function.
print("Pass two positional arguments to the function.(10, 20)")
print(add_numbers(10, 20))  # Output: 30


# Pass three positional arguments to the function.
print("Pass three positional arguments to the function.(10, 20, 30)")
print(add_numbers(10, 20, 30))  # Output: 60


# Pass five positional arguments to the function.
print("Pass five positional arguments to the function.(10, 20, 30, 40, 50)")
print(add_numbers(10, 20, 30, 40, 50))  # Output: 150


# ============================================================
# CORE IDEA
# ============================================================
#
# *numbers collects all positional arguments into a tuple.
#
# add_numbers(10, 20)
# -> numbers = (10, 20)
#
# add_numbers(10, 20, 30)
# -> numbers = (10, 20, 30)
#
# add_numbers(10, 20, 30, 40, 50)
# -> numbers = (10, 20, 30, 40, 50)
#
# This allows the function to work with any number of
# positional arguments.
# ============================================================
```

```python
# ============================================================
# 16. *ARGS IS A TUPLE
# ============================================================
#
# *args collects all positional arguments into a tuple.
#
# In this example, "values" will contain all the arguments
# passed to the function.
# ============================================================


# Define a function that accepts any number of positional arguments.
def show_values(*values):

    # Display the tuple containing all received arguments.
    print(values)

    # Display the data type of "values".
    print(type(values))


# Call the function with three positional arguments.
show_values(10, 20, 30)

# Output:
# (10, 20, 30)
# <class 'tuple'>


# ============================================================
# CORE IDEA
# ============================================================
#
# *values -> collects positional arguments
# values  -> is a tuple
#
# show_values(10, 20, 30)
#       ↓
# values = (10, 20, 30)
#
# Therefore, you can use normal tuple operations with *args.
# ============================================================
```

```python
# ============================================================
# 17. **KWARGS
# ============================================================
#
# **kwargs allows a function to receive an arbitrary number
# of keyword arguments.
#
# The arguments are collected into a dictionary inside
# the function.
#
# The name "kwargs" is a convention. The ** is what tells
# Python to collect the keyword arguments.
# ============================================================


# Define a function that accepts any number of keyword arguments.
def show_information(**information):

    # Display the dictionary containing all received arguments.
    print(information)


# Call the function with three keyword arguments.
show_information(
    name="Yasin",
    age=35,
    field="Data Science"
)

# Output:
# {'name': 'Yasin', 'age': 35, 'field': 'Data Science'}


# ============================================================
# CORE IDEA
# ============================================================
#
# **information -> collects keyword arguments
# information   -> is a dictionary
#
# show_information(name="Yasin", age=35, field="Data Science")
#                         ↓
# information = {
#     "name": "Yasin",
#     "age": 35,
#     "field": "Data Science"
# }
#
# Therefore, you can use normal dictionary operations with
# **kwargs.
# ============================================================
```

```python
show_information(n =1123,b=12312,x=1231,c=1231, xc=123123)
```

```python
# ============================================================
# 18. **KWARGS IS A DICTIONARY
# ============================================================
#
# **kwargs collects all keyword arguments into a dictionary.
#
# Each keyword argument becomes a key-value pair:
#   parameter name -> key
#   parameter value -> value
#
# This means we can use normal dictionary operations such as
# .items(), .keys(), and .values().
# ============================================================


# Define a function that accepts any number of keyword arguments.
def show_information(**information):

    # Display the data type of "information".
    # **information is always stored as a dictionary.
    print(type(information))

    # Loop through each key-value pair in the dictionary.
    for key, value in information.items():

        # Display the key and its corresponding value.
        print(key, ":", value)


# Call the function with three keyword arguments.
show_information(
    name="Yasin",
    age=35,
    field="Data Science"
)

# Output:
# <class 'dict'>
# name : Yasin
# age : 35
# field : Data Science


# ============================================================
# CORE IDEA
# ============================================================
#
# **information -> collects keyword arguments
# information   -> is a dictionary
#
# show_information(name="Yasin", age=35, field="Data Science")
#                         ↓
# information = {
#     "name": "Yasin",
#     "age": 35,
#     "field": "Data Science"
# }
#
# .items() returns each key-value pair:
#
# "name", "Yasin"
# "age", 35
# "field", "Data Science"
# ============================================================
```

```python
# ============================================================
# 19. *ARGS AND **KWARGS TOGETHER
# ============================================================
#
# A function can use *args and **kwargs at the same time.
#
# *args  -> collects arbitrary positional arguments into a tuple
# **kwargs -> collects arbitrary keyword arguments into a dictionary
#
# This allows a function to accept both types of arguments
# without knowing in advance how many will be provided.
# ============================================================


# Define a function that accepts both positional and
# keyword arguments.
def example(*args, **kwargs):

    # Display all positional arguments.
    # "args" is stored as a tuple.
    print("args:", args)

    # Display all keyword arguments.
    # "kwargs" is stored as a dictionary.
    print("kwargs:", kwargs)


# Call the function with both positional and keyword arguments.
example(
    10,                 # Positional argument -> collected by *args
    20,                 # Positional argument -> collected by *args
    30,                 # Positional argument -> collected by *args
    name="Yasin",       # Keyword argument -> collected by **kwargs
    field="AI"          # Keyword argument -> collected by **kwargs
)

# Output:
# args: (10, 20, 30)
# kwargs: {'name': 'Yasin', 'field': 'AI'}


# ============================================================
# CORE IDEA
# ============================================================
#
# Positional arguments:
#     10, 20, 30
#           ↓
#     args = (10, 20, 30)
#
# Keyword arguments:
#     name="Yasin", field="AI"
#           ↓
#     kwargs = {
#         "name": "Yasin",
#         "field": "AI"
#     }
#
# IMPORTANT:
# *args must come before **kwargs in the function definition.
#
# def example(*args, **kwargs):
#     ...
# ============================================================
```

```python
# ============================================================
# 20. FUNCTION WITH A LIST
# ============================================================
#
# A list can be passed to a function as an argument.
#
# The function can then work with the elements of the list
# just like it would with a list defined inside the function.
# ============================================================


# Define a function that accepts a list of numbers.
def calculate_sum(numbers):

    # Initialize the total to 0.
    total = 0

    # Loop through each number in the list.
    for number in numbers:

        # Add the current number to the total.
        total += number

    # Return the final sum.
    return total


# Create a list containing four numbers.
numbers = [10, 20, 30, 40]


# Pass the list to the function and print the returned result.
print(calculate_sum(numbers))  # Output: 100
```

```python
# ============================================================
# 21. FUNCTION WITH A TUPLE
# ============================================================
#
# A tuple can be passed to a function as an argument.
#
# Tuples are similar to lists, but their elements cannot be
# changed after the tuple is created.
#
# In this example, the function calculates the average
# of all numbers in the tuple.
# ============================================================


# Define a function that accepts a tuple of numbers.
def calculate_average(numbers):

    # Calculate the sum of all numbers in the tuple.
    total = sum(numbers)

    # Calculate the average by dividing the total
    # by the number of elements in the tuple.
    average = total / len(numbers)

    # Return the calculated average.
    return average


# Create a tuple containing four numbers.
numbers = (10, 20, 30, 40)


# Pass the tuple to the function and print the returned average.
print(calculate_average(numbers))  # Output: 25.0
```

```python
# ============================================================
# 24. FUNCTION + CONDITION
# ============================================================
#
# A function can contain conditional statements.
#
# In this example, the function checks the value of "age"
# and returns a different result depending on the condition.
#
# The return statement immediately ends the function and
# sends the selected value back to the caller.
# ============================================================


# Define a function with one parameter: "age".
def classify_age(age):

    # Check whether the age is less than 18.
    if age < 18:
        return "Minor"

    # If the first condition is false, check whether
    # the age is less than 65.
    elif age < 65:
        return "Adult"

    # If both previous conditions are false,
    # the age is 65 or greater.
    else:
        return "Senior"


# Call the function with an age of 25.
# 25 is less than 65, so the function returns "Adult".
print(classify_age(25))  # Output: Adult


# Call the function with an age of 70.
# 70 is not less than 18 or 65, so the function returns "Senior".
print(classify_age(70))  # Output: Senior


# ============================================================
# CORE IDEA
# ============================================================
#
# Function + condition:
#
# Input -> condition -> return a result
#
# classify_age(25)
#       ↓
# age < 18  -> False
# age < 65  -> True
#       ↓
# "Adult"
#
# classify_age(70)
#       ↓
# age < 18  -> False
# age < 65  -> False
#       ↓
# "Senior"
# ============================================================
```

```python
# ============================================================
# 25. FUNCTION + LOOP
# ============================================================
#
# A function can contain a loop to process multiple values.
#
# In this example, the function receives a list of numbers
# and uses a "for" loop to calculate their total.
# ============================================================

# Define a function that accepts a list of numbers.
def calculate_total(numbers):

    # Start the total at 0.
    total = 0

    # Loop through each number in the list.
    for number in numbers:

        # Add the current number to the total.
        total += number

    # Return the final total to the caller.
    return total


# Create a list containing four numbers.
numbers = [10, 20, 30, 40]


# Pass the list to the function and print the returned total.
print(calculate_total(numbers))  # Output: 100


# ============================================================
# HOW IT WORKS
# ============================================================
#
# numbers = [10, 20, 30, 40]
#
# The loop processes each number:
#
# 1. total = 0  + 10 -> 10
# 2. total = 10 + 20 -> 30
# 3. total = 30 + 30 -> 60
# 4. total = 60 + 40 -> 100
#
# The function then returns 100.
# ============================================================
```

```python
# ============================================================
# 26. FUNCTION + LOOP + CONDITION
# ============================================================
#
# A function can combine a loop and a condition to process
# multiple values and select values that meet a condition.
#
# In this example, the function counts the even numbers
# in a list.
# ============================================================

# Define a function that accepts a list of numbers.
def count_even_numbers(numbers):

    # Start the counter at 0.
    count = 0

    # Loop through each number in the list.
    for number in numbers:

        # Check if the number is even.
        # % returns the remainder of a division.
        # An even number has a remainder of 0 when divided by 2.
        if number % 2 == 0:

            # Increase the counter by 1.
            count += 1

    # Return the number of even values found.
    return count


# Create a list containing odd and even numbers.
numbers = [1, 2, 3, 4, 5, 6]


# Pass the list to the function and print the returned result.
print(count_even_numbers(numbers))  # Output: 3


# ============================================================
# HOW IT WORKS
# ============================================================
#
# number   condition          count
#   1      1 % 2 == 0  -> False
#   2      2 % 2 == 0  -> True   -> 1
#   3      3 % 2 == 0  -> False
#   4      4 % 2 == 0  -> True   -> 2
#   5      5 % 2 == 0  -> False
#   6      6 % 2 == 0  -> True   -> 3
#
# The function returns 3 because there are three even numbers:
# 2, 4, and 6.
# ============================================================
```

```python
# ============================================================
# 27. FUNCTION FOR MINIMUM VALUE
# ============================================================
#
# A function can use a loop and a condition to find the
# smallest value in a collection of numbers.
#
# In this example, the function compares each number with
# the current minimum and updates it when a smaller number
# is found.
# ============================================================

# Define a function that accepts a list of numbers.
def find_minimum(numbers):

    # Assume the first number is the minimum value.
    minimum = numbers[0]

    # Loop through each number in the list.
    for number in numbers:

        # Check if the current number is smaller
        # than the current minimum.
        if number < minimum:

            # Update the minimum with the smaller number.
            minimum = number

    # Return the smallest value found.
    return minimum


# Call the function with a list of numbers
# and print the returned minimum value.
print(find_minimum([10, 5, 30, 2, 20]))  # Output: 2


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Start:
# minimum = 10
#
# Check each number:
#
# 10 < 10 -> False -> minimum = 10
#  5 < 10 -> True  -> minimum = 5
# 30 <  5 -> False -> minimum = 5
#  2 <  5 -> True  -> minimum = 2
# 20 <  2 -> False -> minimum = 2
#
# The function returns 2 because it is the smallest value.
# ============================================================
```

```python
# ============================================================
# 28. FUNCTION FOR MAXIMUM VALUE
# ============================================================
#
# A function can use a loop and a condition to find the
# largest value in a collection of numbers.
#
# In this example, the function compares each number with
# the current maximum and updates it when a larger number
# is found.
# ============================================================

# Define a function that accepts a list of numbers.
def find_maximum(numbers):

    # Assume the first number is the maximum value.
    maximum = numbers[0]

    # Loop through each number in the list.
    for number in numbers:

        # Check if the current number is larger
        # than the current maximum.
        if number > maximum:

            # Update the maximum with the larger number.
            maximum = number

    # Return the largest value found.
    return maximum


# Call the function with a list of numbers
# and print the returned maximum value.
print(find_maximum([10, 50, 30, 90, 20]))  # Output: 90


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Start:
# maximum = 10
#
# Check each number:
#
# 10 > 10 -> False -> maximum = 10
# 50 > 10 -> True  -> maximum = 50
# 30 > 50 -> False -> maximum = 50
# 90 > 50 -> True  -> maximum = 90
# 20 > 90 -> False -> maximum = 90
#
# The function returns 90 because it is the largest value.
# ============================================================
```

```python
# ============================================================
# 29. FUNCTION FOR AVERAGE
# ============================================================
#
# The average (arithmetic mean) is calculated by dividing
# the sum of all values by the number of values.
#
# This function also checks whether the list is empty before
# calculating the average.
# ============================================================

# Define a function that accepts a list of numbers.
def calculate_average(numbers):

    # Check if the list is empty.
    # len(numbers) == 0 means there are no values to calculate.
    if len(numbers) == 0:

        # Return None because an empty list has no average.
        return None

    # Calculate the average:
    # sum(numbers) -> adds all values
    # len(numbers) -> counts the number of values
    #
    # Return the calculated average to the caller.
    return sum(numbers) / len(numbers)


# Call the function with four numbers
# and print the returned average.
print(calculate_average([10, 20, 30, 40]))  # Output: 25.0


# ============================================================
# HOW IT WORKS
# ============================================================
#
# numbers = [10, 20, 30, 40]
#
# Sum of values:
# 10 + 20 + 30 + 40 = 100
#
# Number of values:
# len(numbers) = 4
#
# Average:
# 100 / 4 = 25.0
#
# If the list is empty:
# [] -> return None
#
# The empty-list check prevents division by zero.
# ============================================================
```

```python
# ============================================================
# 30. FUNCTION FOR MEDIAN
# ============================================================
#
# The median is the middle value of a sorted collection.
#
# Rules:
#   - Odd number of values -> the middle value is the median.
#   - Even number of values -> the median is the average of
#     the two middle values.
#
# The values are sorted first so that the middle position
# can be identified correctly.
# ============================================================

# Define a function that accepts a list of numbers.
def calculate_median(numbers):

    # Sort the numbers from smallest to largest.
    # sorted() creates a new sorted list.
    numbers = sorted(numbers)

    # Store the number of values in the list.
    n = len(numbers)

    # Find the middle index using integer division.
    # // 2 gives the floor division result.
    middle = n // 2

    # Check whether the number of values is odd.
    if n % 2 == 1:

        # For an odd number of values, return the
        # value at the middle index.
        return numbers[middle]

    else:

        # For an even number of values, there are two
        # middle values.
        #
        # numbers[middle - 1] -> first middle value
        # numbers[middle]     -> second middle value
        #
        # Calculate and return their average.
        return (
            numbers[middle - 1]
            + numbers[middle]
        ) / 2


# ============================================================
# ODD NUMBER OF VALUES
# ============================================================
#
# [10, 20, 30]
#
# The list is already sorted.
# There are 3 values, so the middle value is 20.
#
# Output: 20
# ============================================================

print(calculate_median([10, 20, 30]))


# ============================================================
# EVEN NUMBER OF VALUES
# ============================================================
#
# [10, 20, 30, 40]
#
# There are 4 values, so there are two middle values:
#   20 and 30
#
# Median = (20 + 30) / 2 = 25.0
#
# Output: 25.0
# ============================================================

print(calculate_median([10, 20, 30, 40]))


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Odd:
# [10, 20, 30]
#       ↑
#     middle
#       ↓
#      20
#
# Even:
# [10, 20, 30, 40]
#       ↑   ↑
#     middle values
#       ↓   ↓
#      20  30
#
# Median = (20 + 30) / 2 = 25.0
# ============================================================
```

```python
# ============================================================
# 31. FUNCTION FOR RANGE
# ============================================================
#
# The range measures the difference between the largest
# and smallest values in a collection.
#
# Formula:
# Range = Maximum value - Minimum value
#
# In Python:
#   max(numbers) -> finds the largest value
#   min(numbers) -> finds the smallest value
# ============================================================

# Define a function that accepts a list of numbers.
def calculate_range(numbers):

    # Find the maximum and minimum values,
    # then return their difference.
    return max(numbers) - min(numbers)


# Call the function with a list of numbers
# and print the returned range.
print(calculate_range([10, 20, 30, 50]))  # Output: 40


# ============================================================
# HOW IT WORKS
# ============================================================
#
# numbers = [10, 20, 30, 50]
#
# Maximum value:
# max(numbers) = 50
#
# Minimum value:
# min(numbers) = 10
#
# Range:
# 50 - 10 = 40
#
# The function returns 40.
# ============================================================
```

```python
# ============================================================
# 32. FUNCTION FOR PERCENTAGE
# ============================================================
#
# A percentage shows how much one value represents
# compared with another value.
#
# Formula:
# Percentage = (Part / Total) × 100
#
# In this example:
#   part  -> the value we want to compare
#   total -> the total or reference value
#
# The function also checks if "total" is 0 to avoid
# a ZeroDivisionError.
# ============================================================

# Define a function with two parameters:
# "part" and "total".
def calculate_percentage(part, total):

    # Check if the total is zero.
    if total == 0:

        # Return 0 instead of attempting division by zero.
        return 0

    # Calculate the percentage and return the result.
    return (part / total) * 100


# Call the function with:
# part = 25
# total = 100
print(calculate_percentage(25, 100))  # Output: 25.0


# ============================================================
# HOW IT WORKS
# ============================================================
#
# part  = 25
# total = 100
#
# Percentage:
# (25 / 100) × 100
# = 0.25 × 100
# = 25.0
#
# Therefore, 25 is 25% of 100.
#
# If total = 0:
# calculate_percentage(25, 0)
# -> returns 0
#
# This prevents division by zero.
# ============================================================
```

```python
# ============================================================
# 33. DATA ANALYTICS — CLASSIFY SALES
# ============================================================
#
# Functions can be used to classify numerical data into
# categories based on predefined conditions.
#
# In this example, sales values are classified into three
# categories:
#
#   Sales < 1000       -> Low
#   Sales < 5000       -> Medium
#   Sales >= 5000      -> High
#
# This type of classification is commonly used in data
# analysis to convert numerical values into meaningful
# categories.
# ============================================================

# Define a function that accepts a sales value.
def classify_sales(sales):

    # Check if sales are less than 1000.
    if sales < 1000:

        # Return the "Low" category.
        return "Low"

    # If sales are 1000 or higher, check if they are
    # less than 5000.
    elif sales < 5000:

        # Return the "Medium" category.
        return "Medium"

    # If both conditions are false, sales are 5000 or higher.
    else:

        # Return the "High" category.
        return "High"


# Test the function with a sales value below 1000.
print(classify_sales(750))  # Output: Low


# Test the function with a sales value between 1000 and 4999.
print(classify_sales(2500))  # Output: Medium


# Test the function with a sales value of 5000 or higher.
print(classify_sales(8000))  # Output: High
```

```python
# ============================================================
# 34. DATA ANALYTICS — CLEAN A STRING
# ============================================================
#
# Functions can be used to clean and standardize text data.
#
# In this example, the function:
#   1. Removes unnecessary spaces from the beginning and end.
#   2. Converts all letters to lowercase.
#
# These operations are commonly used during data cleaning
# and preprocessing.
# ============================================================

# Define a function that accepts a text string.
def clean_text(text):

    # Remove leading and trailing whitespace.
    # strip() does not remove spaces between words.
    text = text.strip()

    # Convert all letters to lowercase.
    text = text.lower()

    # Return the cleaned text.
    return text


# Pass a string containing extra spaces and uppercase letters.
print(clean_text("   PYTHON   "))  # Output: python


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original:
# "   PYTHON   "
#
# After strip():
# "PYTHON"
#
# After lower():
# "python"
#
# Final result:
# "python"
# ============================================================
```

```python
# ============================================================
# 35. DATA ANALYTICS — CLEAN A LIST OF STRINGS
# ============================================================
#
# Functions can be used to clean and standardize multiple
# text values at once.
#
# In this example, each name is:
#   1. Stripped of leading and trailing whitespace.
#   2. Converted to title case.
#   3. Added to a new cleaned list.
#
# This is a common text-cleaning step in data preprocessing.
# ============================================================

# Define a function that accepts a list of names.
def clean_names(names):

    # Create an empty list to store the cleaned names.
    cleaned_names = []

    # Loop through each name in the original list.
    for name in names:

        # Remove leading and trailing whitespace.
        name = name.strip()

        # Convert the name to title case.
        # For example: "SARA" -> "Sara"
        name = name.title()

        # Add the cleaned name to the new list.
        cleaned_names.append(name)

    # Return the complete list of cleaned names.
    return cleaned_names


# Create a list containing names with inconsistent formatting.
names = [
    " ali ",
    "SARA",
    " yasin"
]


# Pass the list to the function and print the cleaned names.
print(clean_names(names))
# Output: ['Ali', 'Sara', 'Yasin']


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original:
# [" ali ", "SARA", " yasin"]
#
# After strip() and title():
#
# " ali "  -> "Ali"
# "SARA"   -> "Sara"
# " yasin" -> "Yasin"
#
# Final result:
# ["Ali", "Sara", "Yasin"]
# ============================================================
```

```python
# ============================================================
# 36. DATA ANALYTICS — REMOVE MISSING VALUES
# ============================================================
#
# Missing values are common in real-world datasets.
#
# In this example, None represents a missing value.
# The function removes all None values from the list and
# returns a new list containing only valid values.
#
# This is a basic example of data cleaning.
# ============================================================

# Define a function that accepts a list of values.
def remove_missing(values):

    # Create an empty list to store the cleaned values.
    cleaned = []

    # Loop through each value in the original list.
    for value in values:

        # Check if the current value is not missing.
        # "is not None" checks specifically for the None value.
        if value is not None:

            # Add the valid value to the cleaned list.
            cleaned.append(value)

    # Return the list after removing the missing values.
    return cleaned


# Create a list containing numbers and missing values.
values = [10, None, 20, None, 30]


# Pass the list to the function and print the cleaned result.
print(remove_missing(values))
# Output: [10, 20, 30]


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original:
# [10, None, 20, None, 30]
#
# Check each value:
#
# 10   -> not None -> keep
# None -> missing  -> remove
# 20   -> not None -> keep
# None -> missing  -> remove
# 30   -> not None -> keep
#
# Final result:
# [10, 20, 30]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# None represents the absence of a value.
#
# if value is not None:
#     -> keep the value
#
# if value is None:
#     -> skip the value
#
# ============================================================
```

```python
# ============================================================
# 37. DATA ANALYTICS — COUNT MISSING VALUES
# ============================================================
#
# Missing values are common in real-world datasets.
#
# In this example, None represents a missing value.
# The function counts how many None values are present
# in the list.
#
# Unlike the previous example, the missing values are not
# removed. Instead, their total count is returned.
# ============================================================

# Define a function that accepts a list of values.
def count_missing(values):

    # Start the counter at 0.
    count = 0

    # Loop through each value in the list.
    for value in values:

        # Check if the current value is missing.
        if value is None:

            # Increase the counter by 1.
            count += 1

    # Return the total number of missing values.
    return count


# Create a list containing numbers and missing values.
values = [10, None, 20, None, 30]


# Pass the list to the function and print the returned count.
print(count_missing(values))  # Output: 2


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original:
# [10, None, 20, None, 30]
#
# Check each value:
#
# 10   -> not missing -> count = 0
# None -> missing     -> count = 1
# 20   -> not missing -> count = 1
# None -> missing     -> count = 2
# 30   -> not missing -> count = 2
#
# Final result:
# 2 missing values
# ============================================================
```

```python
# ============================================================
# 38. DATA ANALYTICS — MISSING VALUE RATE
# ============================================================
#
# The missing value rate shows the percentage of values
# that are missing in a collection.
#
# Formula:
# Missing Rate = (Number of Missing Values / Total Values) × 100
#
# In this example, None represents a missing value.
# ============================================================

# Define a function that accepts a list of values.
def missing_rate(values):

    # Check if the list is empty.
    if len(values) == 0:

        # Return 0 because there are no values to calculate
        # a missing value rate.
        return 0

    # Start the missing-value counter at 0.
    missing = 0

    # Loop through each value in the list.
    for value in values:

        # Check if the current value is missing.
        if value is None:

            # Increase the missing-value counter by 1.
            missing += 1

    # Calculate the missing value rate as a percentage
    # and return the result.
    return missing / len(values) * 100


# Create a list containing numbers and missing values.
values = [10, None, 20, None, 30]


# Pass the list to the function and print the missing rate.
print(missing_rate(values))  # Output: 40.0


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original:
# [10, None, 20, None, 30]
#
# Total values:
# len(values) = 5
#
# Missing values:
# None, None -> 2
#
# Missing rate:
# (2 / 5) × 100 = 40.0%
#
# Therefore, 40% of the values are missing.
# ============================================================
```

```python
# ============================================================
# 39. DATA ANALYTICS — FILTER VALUES
# ============================================================
#
# Filtering is the process of selecting values that satisfy
# a specific condition.
#
# In this example, the function selects only the values
# that are greater than a given threshold.
#
# This is a common operation in data analysis when we need
# to work with only a subset of the data.
# ============================================================

# Define a function with two parameters:
# "values" contains the data to filter.
# "threshold" defines the minimum value to select.
def filter_large_values(values, threshold):

    # Create an empty list to store the filtered values.
    result = []

    # Loop through each value in the list.
    for value in values:

        # Check if the current value is greater than
        # the specified threshold.
        if value > threshold:

            # Add the value to the result list.
            result.append(value)

    # Return the list containing only the values
    # that satisfy the condition.
    return result


# Create a list of numbers to filter.
numbers = [10, 20, 30, 40, 50]


# Call the function with:
# numbers  -> the list to filter
# 25       -> the threshold
#
# Only values greater than 25 are returned.
print(
    filter_large_values(
        numbers,
        25
    )
)

# Output:
# [30, 40, 50]


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Threshold = 25
#
# Check each value:
#
# 10 > 25 -> False -> skip
# 20 > 25 -> False -> skip
# 30 > 25 -> True  -> keep
# 40 > 25 -> True  -> keep
# 50 > 25 -> True  -> keep
#
# Final result:
# [30, 40, 50]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Input data
#     ↓
# Loop through values
#     ↓
# Check condition
#     ↓
# Keep matching values
#     ↓
# Return filtered list
#
# This pattern is fundamental in data analysis and
# data preprocessing.
# ============================================================
```

```python
# ============================================================
# 40. DATA ANALYTICS — TRANSFORM VALUES
# ============================================================
#
# Transformation means changing values according to a
# specific rule or calculation.
#
# In this example, the function adds a tax to each price.
#
# Formula:
# New Price = Price × (1 + Tax Rate)
#
# The tax rate is provided as a decimal:
#   0.10 -> 10%
#   0.20 -> 20%
#   0.25 -> 25%
# ============================================================

# Define a function with two parameters:
# "prices"    -> the list of prices to transform
# "tax_rate"  -> the tax rate as a decimal
def add_tax(prices, tax_rate):

    # Create an empty list to store the transformed prices.
    result = []

    # Loop through each price in the list.
    for price in prices:

        # Calculate the new price after adding tax.
        new_price = price * (
            1 + tax_rate
        )

        # Add the transformed price to the result list.
        result.append(new_price)

    # Return the list of transformed prices.
    return result


# Create a list containing the original prices.
prices = [100, 200, 300]


# Call the function with:
# prices   -> the original prices
# 0.20     -> a 20% tax rate
print(
    add_tax(
        prices,
        0.20
    )
)

# Output:
# [120.0, 240.0, 360.0]


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Tax rate = 0.20 (20%)
#
# 100 × (1 + 0.20) = 120.0
# 200 × (1 + 0.20) = 240.0
# 300 × (1 + 0.20) = 360.0
#
# Final result:
# [120.0, 240.0, 360.0]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Input values
#     ↓
# Loop through each value
#     ↓
# Apply a transformation
#     ↓
# Store the new value
#     ↓
# Return the transformed list
#
# This pattern is commonly used in data cleaning,
# feature engineering, and data preprocessing.
# ============================================================
```

```python
# ============================================================
# 41. DATA ANALYTICS — Z-SCORE
# ============================================================
#
# A z-score measures how far a value is from the mean,
# expressed in units of standard deviation.
#
# Formula:
#
# z = (value - mean) / standard deviation
#
# Interpretation:
#   z = 0   -> value is equal to the mean
#   z > 0   -> value is above the mean
#   z < 0   -> value is below the mean
#
# In this example, the function calculates the z-score
# for every value in the list.
# ============================================================

# Define a function that accepts a list of values.
def z_scores(values):

    # Calculate the arithmetic mean of the values.
    mean = sum(values) / len(values)

    # Create an empty list to store the squared differences
    # from the mean.
    squared_differences = []

    # Loop through each value in the list.
    for value in values:

        # Calculate the difference between the value and the mean,
        # then square the difference.
        squared_difference = (
            value - mean
        ) ** 2

        # Store the squared difference in the list.
        squared_differences.append(
            squared_difference
        )

    # Calculate the variance.
    # This version uses the population variance:
    # sum of squared differences / number of values.
    variance = (
        sum(squared_differences)
        / len(values)
    )

    # Calculate the standard deviation.
    # The square root of a number can be calculated using ** 0.5.
    standard_deviation = variance ** 0.5

    # Create an empty list to store the z-scores.
    result = []

    # Loop through the values again to calculate each z-score.
    for value in values:

        # Calculate the z-score of the current value.
        z = (
            value - mean
        ) / standard_deviation

        # Add the calculated z-score to the result list.
        result.append(z)

    # Return the complete list of z-scores.
    return result


# Create a list of numerical values.
values = [10, 20, 30, 40, 50]


# Calculate the z-score for each value and print the result.
print(z_scores(values))

# Output:
# [-1.414213562373095, -0.7071067811865475, 0.0,
#   0.7071067811865475, 1.414213562373095]


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Values:
# [10, 20, 30, 40, 50]
#
# Mean:
# (10 + 20 + 30 + 40 + 50) / 5 = 30
#
# The value 30 is the mean, so its z-score is 0.
#
# Values below 30 have negative z-scores.
# Values above 30 have positive z-scores.
#
# The farther a value is from the mean, the larger
# the absolute value of its z-score.
# ============================================================
#
# Example:
#
# value = 50
# mean  = 30
#
# The z-score is positive because 50 is above the mean.
#
# value = 10
# mean  = 30
#
# The z-score is negative because 10 is below the mean.
# ============================================================
#
# IMPORTANT:
#
# This function calculates the POPULATION standard deviation.
# For a sample, the variance calculation uses (n - 1)
# instead of n.
# ============================================================
```

```python
# ============================================================
# 42. DATA ANALYTICS — SIMPLE OUTLIER CHECK
# ============================================================
#
# An outlier is a value that is unusually different from
# the other values in a dataset.
#
# In this simple example, a fixed threshold is used to identify
# unusually large values.
#
# Any value greater than the threshold is treated as an
# outlier candidate.
#
# IMPORTANT:
# This is a simple rule-based check, not a statistical
# outlier detection method such as IQR or z-score.
# ============================================================

# Define a function with two parameters:
# "values"    -> the data to check
# "threshold" -> the value used as the cutoff
def find_large_values(values, threshold):

    # Create an empty list to store values
    # that are greater than the threshold.
    outliers = []

    # Loop through each value in the dataset.
    for value in values:

        # Check if the current value is greater
        # than the specified threshold.
        if value > threshold:

            # Add the value to the outliers list.
            outliers.append(value)

    # Return all values that exceeded the threshold.
    return outliers


# Create a dataset containing one unusually large value.
values = [
    10,
    12,
    14,
    15,
    100
]


# Call the function with a threshold of 50.
# Any value greater than 50 will be returned.
print(
    find_large_values(
        values,
        50
    )
)

# Output:
# [100]


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Threshold = 50
#
# Check each value:
#
# 10  > 50 -> False -> skip
# 12  > 50 -> False -> skip
# 14  > 50 -> False -> skip
# 15  > 50 -> False -> skip
# 100 > 50 -> True  -> keep
#
# Final result:
# [100]
# ============================================================
#
# CORE IDEA:
#
# Data
#   ↓
# Check each value
#   ↓
# value > threshold?
#   ↓
# Yes -> add to outliers
# No  -> skip
#   ↓
# Return detected values
# ============================================================
```

```python
# ============================================================
# 43. FUNCTION RETURNING A DICTIONARY
# ============================================================
#
# A function can return a dictionary containing multiple
# related results.
#
# In data analytics, dictionaries are useful for grouping
# descriptive statistics under meaningful names.
#
# In this example, the function calculates:
#   - Count
#   - Mean
#   - Minimum
#   - Maximum
# ============================================================

# Define a function that accepts a list of numerical values.
def descriptive_statistics(values):

    # Create a dictionary containing the calculated statistics.
    result = {

        # Count the total number of values.
        "count": len(values),

        # Calculate the arithmetic mean.
        "mean": sum(values) / len(values),

        # Find the smallest value.
        "minimum": min(values),

        # Find the largest value.
        "maximum": max(values)
    }

    # Return the dictionary to the caller.
    return result


# Create a list of numerical values.
values = [10, 20, 30, 40, 50]


# Call the function and store the returned dictionary
# in the variable "stats".
stats = descriptive_statistics(values)


# Display the complete dictionary.
print(stats)

# Output:
# {
#     'count': 5,
#     'mean': 30.0,
#     'minimum': 10,
#     'maximum': 50
# }


# ============================================================
# ACCESSING INDIVIDUAL RESULTS
# ============================================================
#
# Because "stats" is a dictionary, individual statistics
# can be accessed using their keys.
# ============================================================

print("Count is ",stats["count"])    # Output: 5
print("Mean is ",stats["mean"])     # Output: 30.0
print("Minimum is ",stats["minimum"])  # Output: 10
print("Maximum is ",stats["maximum"])  # Output: 50


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Input:
# [10, 20, 30, 40, 50]
#
# The function calculates:
#
# count   -> 5
# mean    -> 30.0
# minimum -> 10
# maximum -> 50
#
# These results are grouped into one dictionary:
#
# {
#     "count": 5,
#     "mean": 30.0,
#     "minimum": 10,
#     "maximum": 50
# }
#
# The dictionary is then returned by the function.
# ============================================================
#
# CORE IDEA:
#
# Multiple results
#       ↓
# Store them in a dictionary
#       ↓
# return dictionary
#       ↓
# Access results using their keys
#
# This pattern is useful when a function needs to return
# several named results together.
# ============================================================
```

```python
# ============================================================
# 44. FUNCTION RETURNING MULTIPLE STATISTICS
# ============================================================
#
# A function can return multiple values.
#
# When multiple values are returned using commas, Python
# automatically packs them into a tuple.
#
# For example:
#
# return count, mean, minimum, maximum
#
# is equivalent to:
#
# return (count, mean, minimum, maximum)
#
# The returned tuple can then be unpacked into separate
# variables.
# ============================================================


# Define a function that calculates basic statistics.
def basic_statistics(values):

    # Count the number of values.
    count = len(values)

    # Calculate the mean (average).
    mean = sum(values) / count

    # Find the smallest value.
    minimum = min(values)

    # Find the largest value.
    maximum = max(values)

    # Return multiple values.
    #
    # Python packs these values into a tuple:
    #
    # (count, mean, minimum, maximum)
    #
    return (
        count,
        mean,
        minimum,
        maximum
    )


# Call the function.
#
# The function returns one tuple containing four values.
#
# The tuple is unpacked into four separate variables:
#
# count, mean, minimum, maximum
#
count, mean, minimum, maximum = (
    basic_statistics(
        [10, 20, 30, 40]
    )
)


# Print each returned value.
print(count)
print(mean)
print(minimum)
print(maximum)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# The function returns:
#
# (4, 25.0, 10, 40)
#
# This is a tuple.
#
# Tuple positions:
#
# 1st value -> count   = 4
# 2nd value -> mean    = 25.0
# 3rd value -> minimum = 10
# 4th value -> maximum = 40
#
# Tuple unpacking assigns each value to its variable:
#
# count, mean, minimum, maximum = (4, 25.0, 10, 40)
#
# ============================================================
#
# CORE IDEA:
#
# return a, b, c
#
#        ↓
#
# (a, b, c)    <- tuple
#
#        ↓
#
# x, y, z = (a, b, c)
#
#        ↓
#
# x = a
# y = b
# z = c
#
# ============================================================
```

```python
# ============================================================
# 45. FUNCTION + LIST COMPREHENSION
# ============================================================
#
# A function can use a list comprehension to create a new list
# from an existing list.
#
# A list comprehension provides a short and readable way to
# perform an operation on every item in a list.
#
# Basic structure:
#
# [expression for item in iterable]
#
# "expression" -> What to do with each item
# "item"       -> Current item
# "iterable"   -> List or other collection to loop through
# ============================================================


# Define a function that squares every number in a list.
def square_numbers(numbers):

    # The list comprehension loops through each number
    # and calculates its square.
    #
    # This:
    #
    # [number ** 2 for number in numbers]
    #
    # is similar to:
    #
    # result = []
    #
    # for number in numbers:
    #     result.append(number ** 2)
    #
    # return result
    #
    # The list comprehension creates and returns a new list.
    return [
        number ** 2
        for number in numbers
    ]


# Call the function with a list of numbers.
#
# Each number is squared:
#
# 1 ** 2 -> 1
# 2 ** 2 -> 4
# 3 ** 2 -> 9
# 4 ** 2 -> 16
# 5 ** 2 -> 25
#
# The function returns:
#
# [1, 4, 9, 16, 25]
print(
    square_numbers(
        [1, 2, 3, 4, 5]
    )
)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# List comprehension:
#
# [number ** 2 for number in numbers]
#
#                 ↓
#         Loop through numbers
#
#                 ↓
#       Take each current number
#
#                 ↓
#          Calculate number ** 2
#
#                 ↓
#          Put the result in a list
#
#                 ↓
#        Return the new list
#
# ============================================================
#
# CORE IDEA:
#
# List comprehension = compact way to create a new list
#                      from an existing iterable.
#
# Function + list comprehension = reusable data transformation
#                                 in a compact form.
#
# ============================================================
```

```python
# ============================================================
# 46. FUNCTION + DICTIONARY COMPREHENSION
# ============================================================
#
# A function can use a dictionary comprehension to create
# a new dictionary from an existing iterable.
#
# A dictionary comprehension provides a short and readable
# way to create key-value pairs.
#
# Basic structure:
#
# {key: value for item in iterable}
#
# "key"      -> The key for each dictionary item
# "value"    -> The value associated with the key
# "item"     -> The current item
# "iterable" -> List or other collection to loop through
# ============================================================


# Define a function that creates a dictionary containing
# each number and its square.
def create_square_dictionary(numbers):

    # The dictionary comprehension loops through each number.
    #
    # The number becomes the dictionary key.
    # The square of the number becomes the dictionary value.
    #
    # This:
    #
    # {
    #     number: number ** 2
    #     for number in numbers
    # }
    #
    # is similar to:
    #
    # result = {}
    #
    # for number in numbers:
    #     result[number] = number ** 2
    #
    # return result
    #
    # The dictionary comprehension creates and returns
    # a new dictionary.
    return {
        number: number ** 2
        for number in numbers
    }


# Call the function with a list of numbers.
#
# Each number becomes a key,
# and its square becomes the corresponding value.
#
# 1 -> 1
# 2 -> 4
# 3 -> 9
# 4 -> 16
#
# The function returns:
#
# {1: 1, 2: 4, 3: 9, 4: 16}
print(
    create_square_dictionary(
        [1, 2, 3, 4]
    )
)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Dictionary comprehension:
#
# {
#     number: number ** 2
#     for number in numbers
# }
#
#                  ↓
#          Loop through numbers
#
#                  ↓
#        Take the current number
#
#                  ↓
#       Use number as the KEY
#
#                  ↓
#       Calculate number ** 2
#
#                  ↓
#       Use the result as the VALUE
#
#                  ↓
#          Create a dictionary
#
# ============================================================
#
# CORE IDEA:
#
# Dictionary comprehension = compact way to create a new
#                            dictionary from an iterable.
#
# Example:
#
# numbers = [1, 2, 3, 4]
#
#         ↓
#
# {
#     1: 1,
#     2: 4,
#     3: 9,
#     4: 16
# }
#
# Each dictionary item has:
#
# KEY   -> number
# VALUE -> number ** 2
#
# ============================================================
```

### Lambda and Other Useful Function

```python
# ============================================================
# 47. LAMBDA FUNCTIONS
# ============================================================
#
# A lambda function is a small anonymous function.
#
# "Anonymous" means that the function does not have a
# traditional name defined with the "def" keyword.
#
# Lambda functions are usually used for short, simple
# operations.
#
# Basic structure:
#
# lambda parameters: expression
#
# "lambda"     -> Creates the function
# "parameters" -> Input values
# ":"          -> Separates parameters from the expression
# "expression" -> The value returned by the function
#
# A lambda function automatically returns the result
# of its expression.
#
# Example:
#
# lambda x: x ** 2
#
# means:
#
# Take x and return x squared.
# ============================================================


# Create a lambda function that squares a number.
#
# x is the parameter.
# x ** 2 is the expression.
#
# The lambda function returns x ** 2.
square = lambda x: x ** 2


# Call the lambda function with the argument 5.
#
# 5 ** 2 = 25
print(square(5))


# ============================================================
# HOW IT WORKS
# ============================================================
#
# This lambda:
#
# lambda x: x ** 2
#
# is similar to this regular function:
#
# def square(x):
#     return x ** 2
#
# Both perform the same operation.
#
# The main difference is that a lambda function is written
# in a compact form and is commonly used for simple,
# short-lived operations.
#
# ============================================================
#
# CORE IDEA:
#
# Regular function:
#
# def square(x):
#     return x ** 2
#
# Lambda function:
#
# square = lambda x: x ** 2
#
# Both can be called as:
#
# square(5)
#
# Result:
#
# 25
#
# ============================================================
```

```python
# ============================================================
# 48. LAMBDA WITH TWO PARAMETERS
# ============================================================
#
# A lambda function can accept more than one parameter.
#
# Parameters are separated by commas.
#
# Basic structure:
#
# lambda parameter1, parameter2: expression
#
# The expression is automatically returned.
# ============================================================


# Create a lambda function with two parameters.
#
# "a" and "b" are the parameters.
# "a + b" is the expression.
#
# The function takes two values and returns their sum.
add = lambda a, b: a + b


# Call the lambda function with two arguments.
#
# a = 10
# b = 20
#
# 10 + 20 = 30
print(add(10, 20))


# ============================================================
# HOW IT WORKS
# ============================================================
#
# This:
#
# add = lambda a, b: a + b
#
# is similar to:
#
# def add(a, b):
#     return a + b
#
# Both functions:
#
# 1. Accept two parameters.
# 2. Add the two values.
# 3. Return the result.
#
# ============================================================
#
# CORE IDEA:
#
# Lambda with one parameter:
#
# lambda x: x ** 2
#
# Lambda with two parameters:
#
# lambda a, b: a + b
#
# Multiple parameters are separated by commas.
#
# The result of the expression is automatically returned.
#
# ============================================================
```

```python
# ============================================================
# 49. MAP()
# ============================================================
#
# map() applies a function to every item in an iterable.
#
# It is commonly used to transform each value in a list.
#
# Basic structure:
#
# map(function, iterable)
#
# "function"  -> Operation to perform on each item
# "iterable"  -> Collection of items to process
#
# map() returns a map object.
#
# To get a list of the results, use list():
#
# list(map(function, iterable))
#
# map() is often used together with lambda functions.
# ============================================================


# Create a list of numbers.
numbers = [1, 2, 3, 4, 5]


# Use map() to square every number.
#
# The lambda function:
#
# lambda x: x ** 2
#
# takes one number and returns its square.
#
# map() applies this lambda function to every item
# in the "numbers" list.
#
# list() converts the resulting map object into a list.
squares = list(
    map(
        lambda x: x ** 2,
        numbers
    )
)


# Print the new list.
print(squares)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# numbers:
#
# [1, 2, 3, 4, 5]
#
#                 ↓
#
# map(lambda x: x ** 2, numbers)
#
#                 ↓
#
# Apply the function to every number:
#
# 1 ** 2 -> 1
# 2 ** 2 -> 4
# 3 ** 2 -> 9
# 4 ** 2 -> 16
# 5 ** 2 -> 25
#
#                 ↓
#
# map object
#
#                 ↓
#
# list(...)
#
#                 ↓
#
# [1, 4, 9, 16, 25]
#
# ============================================================
#
# CORE IDEA:
#
# map() = apply the same function to every item.
#
# Example:
#
# list(
#     map(
#         lambda x: x ** 2,
#         numbers
#     )
# )
#
# means:
#
# "Take every number in numbers, square it,
#  and create a list of the results."
#
# ============================================================
#
# EQUIVALENT FOR LOOP:
#
# squares = []
#
# for x in numbers:
#     squares.append(x ** 2)
#
# map() provides a more compact way to perform
# this type of transformation.
#
# ============================================================
```

```python
# ============================================================
# 50. FILTER()
# ============================================================
#
# filter() selects items from an iterable based on a condition.
#
# It keeps only the items for which the function returns True.
#
# Basic structure:
#
# filter(function, iterable)
#
# "function"  -> Condition used to test each item
# "iterable"  -> Collection of items to check
#
# filter() returns a filter object.
#
# To get a list of the filtered results, use list():
#
# list(filter(function, iterable))
#
# filter() is often used together with lambda functions.
# ============================================================


# Create a list of numbers.
numbers = [10, 15, 20, 25, 30]


# Use filter() to keep only numbers greater than 20.
#
# The lambda function:
#
# lambda x: x > 20
#
# checks whether each number is greater than 20.
#
# If the condition is True:
#     The value is kept.
#
# If the condition is False:
#     The value is excluded.
#
# list() converts the resulting filter object into a list.
large_numbers = list(
    filter(
        lambda x: x > 20,
        numbers
    )
)


# Print the filtered list.
print(large_numbers)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# numbers:
#
# [10, 15, 20, 25, 30]
#
#                 ↓
#
# filter(lambda x: x > 20, numbers)
#
#                 ↓
#
# Check each number:
#
# 10 > 20 -> False -> Exclude 10
# 15 > 20 -> False -> Exclude 15
# 20 > 20 -> False -> Exclude 20
# 25 > 20 -> True  -> Keep 25
# 30 > 20 -> True  -> Keep 30
#
#                 ↓
#
# filter object
#
#                 ↓
#
# list(...)
#
#                 ↓
#
# [25, 30]
#
# ============================================================
#
# CORE IDEA:
#
# filter() = keep items that satisfy a condition.
#
# map()    = transform every item.
# filter() = select some items.
#
# Example:
#
# map:
#     [1, 2, 3] -> [1, 4, 9]
#
# filter:
#     [10, 15, 25, 30] -> [25, 30]
#
# ============================================================
#
# EQUIVALENT FOR LOOP:
#
# large_numbers = []
#
# for x in numbers:
#     if x > 20:
#         large_numbers.append(x)
#
# filter() provides a compact way to perform
# this type of selection.
#
# ============================================================
```

```python
# ============================================================
# 51. SORTING WITH A FUNCTION
# ============================================================
#
# sorted() can use a function to decide how items should
# be sorted.
#
# The "key" parameter specifies the value that should be
# used for sorting.
#
# Basic structure:
#
# sorted(iterable, key=function, reverse=False)
#
# "iterable" -> Collection of items to sort
# "key"      -> Function that determines the sorting value
# "reverse"  -> False = ascending, True = descending
#
# A lambda function is often used with "key" because it
# provides a short way to specify the sorting rule.
#
# sorted() returns a NEW sorted list.
# The original list is not changed.
# ============================================================


# Create a list of dictionaries.
#
# Each dictionary represents one student and contains:
#     "name"  -> Student's name
#     "score" -> Student's score
students = [

    {"name": "Ali", "score": 75},

    {"name": "Sara", "score": 90},

    {"name": "Yasin", "score": 85}
]


# Sort the students by their scores.
students_sorted = sorted(
    students,

    # The lambda function tells sorted() which value to use.
    #
    # "student" represents one dictionary at a time.
    #
    # student["score"] gets the student's score.
    #
    # Therefore, students are sorted by their scores.
    key=lambda student: student["score"],

    # reverse=True sorts from largest to smallest.
    #
    # This means the highest score comes first.
    reverse=True
)


# Print the sorted list.
print(students_sorted)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Original scores:
#
# Ali   -> 75
# Sara  -> 90
# Yasin -> 85
#
# The key function:
#
# lambda student: student["score"]
#
# gets the score from each dictionary.
#
# sorted() then compares these values:
#
# 75
# 90
# 85
#
# Because reverse=True:
#
# 90 -> Sara
# 85 -> Yasin
# 75 -> Ali
#
# Result:
#
# [
#     {"name": "Sara", "score": 90},
#     {"name": "Yasin", "score": 85},
#     {"name": "Ali", "score": 75}
# ]
#
# ============================================================
#
# CORE IDEA:
#
# key = function that tells sorted() WHAT VALUE to sort by.
#
# In this example:
#
# key=lambda student: student["score"]
#
# means:
#
# "Sort each student using their score."
#
# ============================================================
#
# ASCENDING ORDER:
#
# reverse=False
#
# 75 -> 85 -> 90
#
# DESCENDING ORDER:
#
# reverse=True
#
# 90 -> 85 -> 75
#
# ============================================================
#
# EQUIVALENT FUNCTION:
#
# Instead of a lambda, we could define a regular function:
#
# def get_score(student):
#     return student["score"]
#
# Then:
#
# students_sorted = sorted(
#     students,
#     key=get_score,
#     reverse=True
# )
#
# The lambda version is simply more compact.
#
# ============================================================
```

```python
# ============================================================
# 52. NESTED FUNCTION
# ============================================================
#
# A nested function is a function defined inside another
# function.
#
# The outer function contains the inner function.
#
# Basic structure:
#
# def outer_function():
#
#     def inner_function():
#         code
#
#     inner_function()
#
# The inner function is normally available only inside
# the outer function.
# ============================================================


# Define the outer function.
def outer_function():

    # Define a function inside the outer function.
    #
    # This is called a nested function.
    def inner_function():

        # Code inside the inner function.
        print("Inside inner function")


    # Call the inner function.
    #
    # The inner function is called from inside
    # the outer function.
    inner_function()


# Call the outer function.
#
# This starts the execution of outer_function().
#
# The outer function then:
#
# 1. Defines inner_function()
# 2. Calls inner_function()
# 3. Prints "Inside inner function"
outer_function()


# ============================================================
# HOW IT WORKS
# ============================================================
#
# outer_function()
#        |
#        ↓
# Define inner_function()
#        |
#        ↓
# Call inner_function()
#        |
#        ↓
# Print "Inside inner function"
#
# ============================================================
#
# CORE IDEA:
#
# A nested function is simply a function defined inside
# another function.
#
# Outer function:
#
# def outer_function():
#
#     ...
#
#     def inner_function():
#         ...
#
# ============================================================
#
# IMPORTANT:
#
# The inner function is local to the outer function.
#
# Therefore, this works:
#
# def outer_function():
#
#     def inner_function():
#         print("Hello")
#
#     inner_function()
#
#
# But this does NOT work outside the outer function:
#
# inner_function()
#
# because "inner_function" was defined inside
# "outer_function".
#
# ============================================================
#
# Nested functions are useful when a helper function is needed
# only inside one specific function.
#
# ============================================================
```

```python
# ============================================================
# 53. SCOPE
# ============================================================
#
# Scope determines where a variable can be accessed in a
# Python program.
#
# A variable created inside a function normally has
# LOCAL SCOPE.
#
# A local variable can be used inside the function where
# it was created, but it cannot normally be accessed
# from outside that function.
#
# ============================================================


# Define a function.
def test():

    # Create a local variable.
    #
    # "x" exists only inside the test() function.
    x = 10

    # x can be accessed inside the function.
    print(x)


# Call the function.
test()


# ============================================================
# OUTSIDE THE FUNCTION
# ============================================================
#
# x does not exist in the scope outside test().
#
# Therefore, this would cause a NameError:
#
# print(x)
#
# because x is a LOCAL VARIABLE of test().
#
# ============================================================
#
# HOW IT WORKS:
#
# test()
#   |
#   +-- x = 10
#   |
#   +-- print(x)  -> 10
#
# After test() finishes, its local variable x is no longer
# accessible from outside the function.
#
# ============================================================
#
# CORE IDEA:
#
# LOCAL SCOPE:
#
# A variable created inside a function belongs to that
# function's local scope.
#
# Example:
#
# def test():
#     x = 10
#
#     print(x)       # Works
#
# test()
#
# print(x)           # NameError
#
# ============================================================
#
# Python has several levels of variable scope.
#
# The main ones are:
#
# Local    -> Inside the current function
# Enclosing -> Inside an outer function
# Global   -> At the module/program level
# Built-in -> Names provided by Python
#
# These are commonly remembered as:
#
# L -> Local
# E -> Enclosing
# G -> Global
# B -> Built-in
#
# ============================================================
```

```python
# ============================================================
# 54. LOCAL VARIABLE
# ============================================================
#
# A local variable is a variable created inside a function.
#
# It belongs to the local scope of that function and can
# normally be accessed only inside that function.
#
# Local variables are commonly used to store temporary
# values needed while the function is running.
# ============================================================


# Define a function.
def calculate():

    # Create a local variable.
    #
    # "result" exists inside calculate().
    result = 10 + 20

    # Return the value stored in the local variable.
    return result


# Call the function and print the returned value.
#
# The function calculates 10 + 20,
# stores the result in the local variable,
# and returns that value.
print(calculate())


# ============================================================
# HOW IT WORKS
# ============================================================
#
# calculate()
#      |
#      ↓
# result = 10 + 20
#      |
#      ↓
# result = 30
#      |
#      ↓
# return result
#      |
#      ↓
# print(30)
#
# Output:
#
# 30
#
# ============================================================
#
# IMPORTANT:
#
# "result" is a LOCAL VARIABLE.
#
# It can be used inside calculate():
#
# def calculate():
#     result = 10 + 20
#     print(result)       # Works
#
#
# But this would cause a NameError:
#
# print(result)
#
# because "result" was created inside calculate()
# and is not available in the outer scope.
#
# ============================================================
#
# CORE IDEA:
#
# Local variable:
#
# A variable created inside a function and normally
# accessible only within that function.
#
# The value of a local variable can still be made
# available outside the function by returning it:
#
# return result
#
# Then the caller can store the returned value:
#
# answer = calculate()
#
# ============================================================
```

```python
# ============================================================
# 55. GLOBAL VARIABLE
# ============================================================
#
# A global variable is a variable created outside of a
# function.
#
# A global variable can normally be accessed from inside
# functions.
#
# Global variables belong to the global scope of the program.
# ============================================================


# Create a global variable.
#
# "tax_rate" is defined outside the function,
# so it has global scope.
tax_rate = 0.20


# Define a function that calculates tax.
def calculate_tax(price):

    # The function can read the global variable "tax_rate".
    #
    # 0.20 means a 20% tax rate.
    return price * tax_rate


# Call the function with a price of 100.
#
# 100 * 0.20 = 20.0
print(calculate_tax(100))


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Global variable:
#
# tax_rate = 0.20
#
#        ↓
#
# calculate_tax(100)
#        ↓
#
# price * tax_rate
#        ↓
#
# 100 * 0.20
#        ↓
#
# 20.0
#
# ============================================================
#
# IMPORTANT:
#
# A function can READ a global variable:
#
# tax_rate = 0.20
#
# def calculate_tax(price):
#     return price * tax_rate
#
# This works because tax_rate exists in the global scope.
#
# ============================================================
#
# LOCAL vs GLOBAL:
#
# Local variable:
#
# def calculate():
#     result = 30
#
# "result" exists inside the function.
#
#
# Global variable:
#
# tax_rate = 0.20
#
# def calculate_tax(price):
#     return price * tax_rate
#
# "tax_rate" exists outside the function and can be
# read from inside the function.
#
# ============================================================
#
# CORE IDEA:
#
# GLOBAL SCOPE
#     ↓
# Variables defined outside functions.
#
# LOCAL SCOPE
#     ↓
# Variables defined inside functions.
#
# ============================================================
#
# NOTE:
#
# Although functions can read global variables, relying heavily
# on global variables is generally discouraged because it can
# make programs harder to understand and maintain.
#
# Passing values as parameters is often clearer:
#
# def calculate_tax(price, tax_rate):
#     return price * tax_rate
#
# ============================================================
```

```python
# ============================================================
# 56. AVOID UNNECESSARY GLOBAL VARIABLES
# ============================================================
#
# Global variables can be useful, but using too many of them
# can make a program harder to understand and maintain.
#
# A better approach is to pass information into functions
# through parameters.
#
# This makes the function more independent and reusable.
#
# Instead of:
#
# tax_rate = 0.20
#
# def calculate_tax(price):
#     return price * tax_rate
#
# Pass tax_rate directly to the function:
#
# def calculate_tax(price, tax_rate):
#     return price * tax_rate
#
# ============================================================


# Define a function with two parameters.
#
# "price" is the value being taxed.
# "tax_rate" is the tax rate to apply.
#
# Both values are provided by the caller.
def calculate_tax(price, tax_rate):

    # Calculate and return the tax amount.
    return price * tax_rate


# Call the function and provide both arguments.
#
# price = 100
# tax_rate = 0.20
#
# 100 * 0.20 = 20.0
print(
    calculate_tax(
        100,
        0.20
    )
)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# calculate_tax(
#     100,
#     0.20
# )
#
#        ↓
#
# price    = 100
# tax_rate = 0.20
#
#        ↓
#
# price * tax_rate
#
#        ↓
#
# 100 * 0.20
#
#        ↓
#
# 20.0
#
# ============================================================
#
# WHY THIS APPROACH IS BETTER:
#
# The function does not depend on a specific global variable.
#
# The same function can be reused with different tax rates:
#
# calculate_tax(100, 0.20)   -> 20.0
# calculate_tax(100, 0.10)   -> 10.0
# calculate_tax(500, 0.15)   -> 75.0
#
# ============================================================
#
# CORE IDEA:
#
# Avoid unnecessary global variables.
#
# Prefer:
#
# def calculate_tax(price, tax_rate):
#     return price * tax_rate
#
# This makes the function:
#
#   - More reusable
#   - Easier to test
#   - Easier to understand
#   - Less dependent on outside variables
#
# ============================================================
```

```python
# ============================================================
# 57. DOCSTRING
# ============================================================
#
# A docstring is a string used to document a function,
# class, or module.
#
# A function's docstring is written immediately after
# the function definition.
#
# Docstrings explain:
#   - What the function does
#   - What its parameters represent
#   - What the function returns
#
# A docstring is written inside triple quotes:
#
# """
# Documentation goes here.
# """
#
# Python stores the docstring as part of the function's
# documentation and makes it available through __doc__
# and the help() function.
# ============================================================


# Define a function that calculates the arithmetic mean.
def calculate_average(numbers):

    # Docstring:
    # Documents the purpose of the function.
    """
    Calculate the arithmetic mean
    of a collection of numbers.
    """

    # Calculate and return the arithmetic mean.
    return sum(numbers) / len(numbers)


# Call the function and print the returned result.
#
# sum([10, 20, 30]) = 60
# len([10, 20, 30]) = 3
# 60 / 3 = 20.0
print(
    calculate_average(
        [10, 20, 30]
    )
)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# The docstring:
#
# """
# Calculate the arithmetic mean
# of a collection of numbers.
# """
#
# is placed immediately after:
#
# def calculate_average(numbers):
#
# This tells Python that the string is the function's
# documentation.
#
# ============================================================
#
# YOU CAN ACCESS THE DOCSTRING:
#
# print(calculate_average.__doc__)
#
# This displays the documentation stored inside the function.
#
# You can also use:
#
# help(calculate_average)
#
# to display information about the function.
#
# ============================================================
#
# CORE IDEA:
#
# Docstring = documentation written inside Python code.
#
# It explains what a function is intended to do and can be
# viewed by programmers and documentation tools.
#
# ============================================================
#
# IMPORTANT:
#
# A docstring is NOT the same as a regular comment.
#
# Comment:
#
# # This is a comment.
#
# Docstring:
#
# """
# This is a docstring.
# """
#
# A docstring is specifically used to document Python
# functions, classes, and modules.
#
# ============================================================
```

```python
# ============================================================
# 58. TYPE HINTS
# ============================================================
#
# Type hints are optional annotations that describe the
# expected types of variables, parameters, and return values.
#
# They make code easier to read, understand, and maintain.
#
# Type hints do NOT normally enforce the type at runtime.
# Python will still execute the function unless the code
# itself causes an error.
#
# Basic structure:
#
# def function_name(parameter: type) -> return_type:
#     code
#
# "parameter: type" -> Expected type of the parameter
# "-> return_type"  -> Expected type returned by the function
# ============================================================


# Define a function with type hints.
#
# "a: float" means a is expected to be a float.
# "b: float" means b is expected to be a float.
# "-> float" means the function is expected to return a float.
def add(a: float, b: float) -> float:

    # Add the two values and return the result.
    return a + b


# Call the function.
#
# Python accepts 10 and 20 here even though they are integers.
# Type hints are annotations; they do not automatically convert
# the values into floats.
print(add(10, 20))


# ============================================================
# HOW IT WORKS
# ============================================================
#
# This:
#
# def add(a: float, b: float) -> float:
#
# tells the reader:
#
# a       -> expected to be a float
# b       -> expected to be a float
# return  -> expected to be a float
#
# The annotations provide information about the intended
# types of the inputs and output.
#
# ============================================================
#
# WITHOUT TYPE HINTS:
#
# def add(a, b):
#     return a + b
#
#
# WITH TYPE HINTS:
#
# def add(a: float, b: float) -> float:
#     return a + b
#
# The second version communicates the expected types more
# clearly.
#
# ============================================================
#
# IMPORTANT:
#
# Type hints do NOT force Python to use a specific type.
#
# For example:
#
# def add(a: float, b: float) -> float:
#     return a + b
#
# Python does not automatically reject:
#
# add(10, 20)
#
# because 10 and 20 are integers.
#
# Type hints mainly provide information for:
#
#   - Programmers
#   - Code editors
#   - IDEs
#   - Static type checkers
#   - Documentation tools
#
# ============================================================
#
# CORE IDEA:
#
# Type hints describe the EXPECTED type of data.
#
# parameter:
#
# a: float
#
# means:
#
# "a is expected to be a float."
#
# return annotation:
#
# -> float
#
# means:
#
# "This function is expected to return a float."
#
# ============================================================
```

```python
# ============================================================
# 59. TYPE HINTS WITH LIST
# ============================================================
#
# Type hints can also describe the type of items inside
# a collection such as a list.
#
# Basic structure:
#
# list[type]
#
# For example:
#
# list[float]
#
# means:
#
# "A list containing float values."
#
# This allows us to describe not only that a parameter is
# a list, but also what type of values the list should contain.
# ============================================================


# Define a function with a list type hint.
#
# numbers: list[float]
#     -> "numbers" is expected to be a list of float values.
#
# -> float
#     -> The function is expected to return a float.
def calculate_average(
    numbers: list[float]
) -> float:

    # Calculate and return the arithmetic mean.
    return sum(numbers) / len(numbers)


# Call the function with a list of float values.
print(
    calculate_average(
        [10.0, 20.0, 30.0]
    )
)


# ============================================================
# HOW IT WORKS
# ============================================================
#
# This type hint:
#
# numbers: list[float]
#
# tells us:
#
# numbers
#    ↓
# list
#    ↓
# each item is expected to be a float
#
# Example:
#
# [10.0, 20.0, 30.0]
#
#       ↓
#
# list[float]
#
# ============================================================
#
# The return type:
#
# -> float
#
# tells us that the function is expected to return
# a floating-point number.
#
# For example:
#
# 20.0
#
# ============================================================
#
# IMPORTANT:
#
# list[float] describes the expected type.
# It does NOT automatically convert values to float.
#
# Type hints are annotations and normally do not enforce
# types at runtime.
#
# ============================================================
#
# MORE EXAMPLES:
#
# list[int]
#     -> List of integers
#
# list[str]
#     -> List of strings
#
# list[bool]
#     -> List of Boolean values
#
# list[float]
#     -> List of floating-point numbers
#
# ============================================================
#
# CORE IDEA:
#
# list[type]
#
# lets us specify:
#
# "This parameter should be a list containing values
#  of this type."
#
# Example:
#
# numbers: list[float]
#
# means:
#
# "numbers should be a list of float values."
#
# ============================================================
```

```python
# ============================================================
# 60. PRACTICAL ML EXAMPLE
# ============================================================

def classify_probability(probability):

    if probability >= 0.80:

        return "High Risk"

    elif probability >= 0.50:

        return "Medium Risk"

    else:

        return "Low Risk"


probabilities = [
    0.20,
    0.55,
    0.82,
    0.91
]

for probability in probabilities:

    print(
        probability,
        "->",
        classify_probability(
            probability
        )
    )


# ============================================================
# 61. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================

def recommend_pipeline_action(
    failure_probability
):

    if failure_probability >= 0.80:

        return "Immediate inspection"

    elif failure_probability >= 0.60:

        return "Schedule inspection"

    elif failure_probability >= 0.40:

        return "Monitor"

    else:

        return "Normal operation"


probabilities = [
    0.15,
    0.42,
    0.73,
    0.91,
    0.31
]

for probability in probabilities:

    action = recommend_pipeline_action(
        probability
    )

    print(
        probability,
        "->",
        action
    )


# ============================================================
# 62. PRACTICAL ML METRICS FUNCTION
# ============================================================

def evaluate_model(
    accuracy,
    precision,
    recall,
    f1
):

    return {

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1": f1
    }


metrics = evaluate_model(
    0.91,
    0.84,
    0.76,
    0.80
)

print(metrics)


# ============================================================
# 63. FUNCTION FOR DATA QUALITY
# ============================================================

def data_quality_report(values):

    total = len(values)

    missing = 0

    for value in values:

        if value is None:

            missing += 1

    missing_percentage = (
        missing / total
    ) * 100

    return {

        "Total": total,

        "Missing": missing,

        "Missing %": missing_percentage
    }


data = [
    10,
    20,
    None,
    40,
    None,
    60
]

print(
    data_quality_report(data)
)


# ============================================================
# 64. FUNCTIONS CALLING OTHER FUNCTIONS
# ============================================================

def calculate_total(values):

    return sum(values)


def calculate_average(values):

    total = calculate_total(values)

    return total / len(values)


values = [10, 20, 30, 40]

print(
    calculate_average(values)
)


# ============================================================
# 65. MAIN FUNCTION
# ============================================================

def main():

    numbers = [10, 20, 30, 40]

    average = calculate_average(
        numbers
    )

    print("Average:", average)


main()


# ============================================================
# 66. __name__ == "__main__"
# ============================================================

def main():

    print("Program started")


if __name__ == "__main__":

    main()
```
