# PYTHON CONDITIONAL STATEMENTS


```python
# ============================================================
# PYTHON CONDITIONAL STATEMENTS
# COMPLETE LEARNING GUIDE
# ============================================================
#
# Conditional statements let Python MAKE DECISIONS.
#
# The central idea is:
#
#     CONDITION
#         ↓
#     True or False
#         ↓
#     DECISION
#         ↓
#     execute some code OR skip it
#
#
# The main conditional tools in Python are:
#
#     if
#     elif
#     else
#
# They work together with:
#
#     comparison operators
#     logical operators
#     membership operators
#     identity operators
#     truthiness
#     conditional expressions
#     nested conditions
#     match/case
#
# ============================================================


# ============================================================
# 1. THE BASIC IDEA OF A CONDITION
# ============================================================
#
# A condition is an expression that produces:
#
#     True
#     False
#
# Example:
#
#     age = 25
#     age >= 18
#
# Result:
#
#     True
#
#
# Mental model:
#
#     VALUE
#       ↓
#     COMPARISON
#       ↓
#     True / False
#       ↓
#     DECISION
#
# ------------------------------------------------------------
# Example
# ------------------------------------------------------------

age = 25

print(age >= 18)


# Output:
#
# True


# ============================================================
# 2. BOOLEAN VALUES
# ============================================================
#
# Python has two Boolean values:
#
#     True
#     False
#
# Their type is:
#
#     bool
#
# ------------------------------------------------------------

is_student = True
has_job = False

print(is_student)
print(has_job)

print(type(is_student))


# Output:
#
# True
# False
# <class 'bool'>


# ============================================================
# 3. bool()
# ============================================================
#
# bool() converts a value into its Boolean interpretation.
#
# Pattern:
#
#     bool(value)
#
#
# Examples:
#
#     bool(True)       → True
#     bool(False)      → False
#     bool(10)         → True
#     bool(0)          → False
#     bool("Hello")    → True
#     bool("")         → False
#     bool([1, 2])     → True
#     bool([])         → False
#
#
# Mental model:
#
#     value
#       ↓
#     bool(value)
#       ↓
#     True / False
#
# ------------------------------------------------------------

print(bool(10))
print(bool(0))

print(bool("Python"))
print(bool(""))

print(bool([1, 2, 3]))
print(bool([]))


# ============================================================
# 4. TRUTHINESS
# ============================================================
#
# Python does not require a condition to literally be:
#
#     True
#     False
#
# Many values can be evaluated as True or False.
#
#
# Common FALSE values:
#
#     False
#     None
#     0
#     0.0
#     ""
#     []
#     ()
#     {}
#     set()
#
#
# Most other values are TRUE.
#
#
# Mental model:
#
#     value
#       ↓
#     "Is this truthy?"
#       ↓
#     True / False
#
# ------------------------------------------------------------

value = 10

if value:
    print("Value is truthy.")


# ============================================================
# 5. if STATEMENT
# ============================================================
#
# The simplest conditional statement:
#
#     if condition:
#         action
#
#
# Meaning:
#
#     IF the condition is True
#         → execute the indented block
#
#     IF the condition is False
#         → skip the block
#
# ------------------------------------------------------------

age = 25

if age >= 18:
    print("Adult.")


# ============================================================
# 6. if FLOW
# ============================================================
#
#                    START
#                      ↓
#                 Check condition
#                      ↓
#               ┌──────┴──────┐
#             True           False
#               ↓               ↓
#          execute block       skip
#               ↓               ↓
#               └──────┬────────┘
#                      ↓
#                    NEXT
#
#
# The condition is evaluated BEFORE the body runs.
#
# ============================================================


# ============================================================
# 7. INDENTATION
# ============================================================
#
# Indentation tells Python which statements belong to the if.
#
# Correct:
#
#     if age >= 18:
#         print("Adult.")
#
#
# Incorrect:
#
#     if age >= 18:
#     print("Adult.")
#
#
# Python uses indentation as part of its syntax.
#
# ------------------------------------------------------------

age = 25

if age >= 18:
    print("Adult.")
    print("Can continue.")


# Both print statements belong to the if block.


# ============================================================
# 8. COMPARISON OPERATORS
# ============================================================
#
# Conditional statements commonly use comparison operators.
#
# ┌────────┬─────────────────────────────┐
# │ ==     │ equal to                    │
# │ !=     │ not equal to                │
# │ >      │ greater than                │
# │ <      │ less than                   │
# │ >=     │ greater than or equal to    │
# │ <=     │ less than or equal to       │
# └────────┴─────────────────────────────┘
#
#
# Every comparison produces True or False.
#
# ------------------------------------------------------------

x = 10

print(x == 10)   # True
print(x != 10)   # False
print(x > 5)     # True
print(x < 5)     # False
print(x >= 10)   # True
print(x <= 10)   # True


# ============================================================
# 9. == VS =
# ============================================================
#
# =   means ASSIGN
#
#     age = 30
#
#
# ==  means COMPARE
#
#     age == 30
#
#
# Mental model:
#
#     =   → put a value into a variable
#
#     ==  → ask "are these equal?"
#
# ------------------------------------------------------------

age = 30

if age == 30:
    print("Age is 30.")


# ============================================================
# 10. if WITH DIFFERENT COMPARISONS
# ============================================================

score = 85

if score > 80:
    print("High score.")

if score == 85:
    print("Score is exactly 85.")

if score != 50:
    print("Score is not 50.")


# ============================================================
# 11. INDEPENDENT if STATEMENTS
# ============================================================
#
# IMPORTANT:
#
# Separate if statements are independent.
#
# Python checks EVERY if.
#
#
#     if A → check
#     if B → check
#     if C → check
#
#
# A, B, and C can ALL execute.
#
# ------------------------------------------------------------

score = 95

if score >= 50:
    print("Passed.")

if score >= 80:
    print("Good score.")

if score >= 90:
    print("Excellent score.")


# Output:
#
# Passed.
# Good score.
# Excellent score.


# ============================================================
# 12. MEMORY + DECISION MODEL
# ============================================================
#
#     score = 95
#
# Memory:
#
#     score ─────→ 95
#
#
# First if:
#
#     95 >= 50
#          ↓
#        True
#          ↓
#      print()
#
#
# Second if:
#
#     95 >= 80
#          ↓
#        True
#          ↓
#      print()
#
#
# Third if:
#
#     95 >= 90
#          ↓
#        True
#          ↓
#      print()
#
#
# Each if performs its OWN decision.
#
# ============================================================


# ============================================================
# 13. if / else
# ============================================================
#
# Use else when there are TWO alternatives.
#
#
#     if condition:
#         action_A
#     else:
#         action_B
#
#
# Meaning:
#
#     True  → A
#     False → B
#
# ------------------------------------------------------------

age = 16

if age >= 18:
    print("Adult.")
else:
    print("Minor.")


# ============================================================
# 14. if / else FLOW
# ============================================================
#
#                    START
#                      ↓
#                 condition?
#                      ↓
#              ┌───────┴───────┐
#            True             False
#              ↓                 ↓
#         IF block          ELSE block
#              ↓                 ↓
#              └────────┬────────┘
#                       ↓
#                      NEXT
#
#
# Exactly ONE branch executes.
#
# ============================================================


# ============================================================
# 15. if / elif / else
# ============================================================
#
# Use elif when there are MULTIPLE possible conditions.
#
#
#     if condition_A:
#         action_A
#     elif condition_B:
#         action_B
#     elif condition_C:
#         action_C
#     else:
#         default_action
#
#
# Python checks from TOP to BOTTOM.
#
# The FIRST True condition wins.
#
# ------------------------------------------------------------

score = 75

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")


# Output:
#
# C


# ============================================================
# 16. if / elif / else FLOW
# ============================================================
#
#                 START
#                   ↓
#               condition A?
#                /       \
#             True       False
#              ↓           ↓
#          action A     condition B?
#                          /    \
#                       True    False
#                        ↓        ↓
#                    action B  condition C?
#                                  /   \
#                               True   False
#                                ↓       ↓
#                            action C   ELSE
#
#
# Only ONE branch executes.
#
# ============================================================


# ============================================================
# 17. FIRST TRUE CONDITION WINS
# ============================================================
#
# This is one of the MOST IMPORTANT rules of elif.
#
# ------------------------------------------------------------

score = 95

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")


# Output:
#
# A
#
# Python does NOT continue checking the remaining elif branches.
#
#
# Flow:
#
#     95 >= 90
#          ↓
#        True
#          ↓
#         "A"
#          ↓
#       STOP CHAIN
#
# ============================================================


# ============================================================
# 18. ORDER MATTERS IN elif
# ============================================================
#
# This is WRONG for grading:
#
#     if score >= 50:
#         print("Pass")
#     elif score >= 90:
#         print("Excellent")
#
#
# Why?
#
# For score = 95:
#
#     95 >= 50 → True
#
# Python stops immediately.
#
#
# Better:
#
#     if score >= 90:
#         print("Excellent")
#     elif score >= 50:
#         print("Pass")
#
#
# More specific conditions should generally come FIRST
# when the conditions overlap.
#
# ============================================================


# ============================================================
# 19. elif WITHOUT else
# ============================================================
#
# else is OPTIONAL.
#
# ------------------------------------------------------------

temperature = 25

if temperature > 30:
    print("Hot.")
elif temperature < 10:
    print("Cold.")


# If neither condition is True:
#
#     nothing happens.
#
#
# Flow:
#
#     if True  → execute
#     elif True → execute
#     neither → skip entire chain
#
# ============================================================


# ============================================================
# 20. else WITHOUT elif
# ============================================================
#
# Yes.
#
# Basic:
#
#     if condition:
#         ...
#     else:
#         ...
#
# elif is not required.
#
# ============================================================


# ============================================================
# 21. NESTED if STATEMENTS
# ============================================================
#
# An if can contain another if.
#
#
#     if condition_A:
#         if condition_B:
#             action
#
#
# Mental model:
#
#     condition A?
#          ↓
#        True
#          ↓
#     condition B?
#          ↓
#        True
#          ↓
#       action
#
# ------------------------------------------------------------

age = 25
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed.")


# ============================================================
# 22. NESTED CONDITION FLOW
# ============================================================
#
#     age >= 18?
#          │
#      ┌───┴───┐
#    False    True
#      │        │
#     STOP   has_id?
#               │
#           ┌───┴───┐
#         False    True
#           │        │
#          STOP    ALLOW
#
#
# The inner condition is only checked if the outer
# condition is True.
#
# ============================================================


# ============================================================
# 23. MULTIPLE NESTED CONDITIONS
# ============================================================

age = 25
has_id = True
is_member = True

if age >= 18:
    if has_id:
        if is_member:
            print("Access granted.")


# Mental model:
#
#     age >= 18
#         ↓
#     has_id
#         ↓
#     is_member
#         ↓
#     Access granted
#
#
# All three conditions must be True.
#
# ============================================================


# ============================================================
# 24. LOGICAL OPERATORS
# ============================================================
#
# Python provides:
#
#     and
#     or
#     not
#
#
# These combine or modify conditions.
#
# ┌────────┬──────────────────────────────┐
# │ and    │ BOTH conditions must be True │
# │ or     │ AT LEAST ONE must be True   │
# │ not    │ reverses True/False         │
# └────────┴──────────────────────────────┘
#
# ============================================================


# ============================================================
# 25. and
# ============================================================
#
#     A and B
#
# True only when BOTH A and B are True.
#
#
# Truth table:
#
#     A       B       A and B
#     ------------------------
#     False   False   False
#     False   True    False
#     True    False   False
#     True    True    True
#
# ------------------------------------------------------------

age = 25
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")


# ============================================================
# 26. and VISUAL MODEL
# ============================================================
#
#     CONDITION A ───┐
#                    ├──→ AND ──→ result
#     CONDITION B ───┘
#
#
#     True  + True  → True
#     True  + False → False
#     False + True  → False
#     False + False → False
#
#
# Mental model:
#
#     and = "BOTH must be true"
#
# ============================================================


# ============================================================
# 27. or
# ============================================================
#
#     A or B
#
# True when AT LEAST ONE condition is True.
#
#
# Truth table:
#
#     A       B       A or B
#     -----------------------
#     False   False   False
#     False   True    True
#     True    False   True
#     True    True    True
#
# ------------------------------------------------------------

is_admin = False
is_manager = True

if is_admin or is_manager:
    print("Access allowed.")


# ============================================================
# 28. or VISUAL MODEL
# ============================================================
#
#     CONDITION A ───┐
#                    ├──→ OR ──→ result
#     CONDITION B ───┘
#
#
#     True  + True  → True
#     True  + False → True
#     False + True  → True
#     False + False → False
#
#
# Mental model:
#
#     or = "AT LEAST ONE must be true"
#
# ============================================================


# ============================================================
# 29. not
# ============================================================
#
# not reverses a Boolean result.
#
#
#     not True  → False
#     not False → True
#
# ------------------------------------------------------------

is_raining = False

if not is_raining:
    print("Go outside.")


# ============================================================
# 30. not VISUAL MODEL
# ============================================================
#
#     True  ──→ not ──→ False
#
#     False ──→ not ──→ True
#
#
# Mental model:
#
#     not = "reverse the Boolean result"
#
# ============================================================


# ============================================================
# 31. COMBINING and / or / not
# ============================================================

age = 25
has_id = True
is_banned = False

if age >= 18 and has_id and not is_banned:
    print("Access allowed.")


# Read it conceptually as:
#
#     age >= 18
#          AND
#     has_id
#          AND
#     NOT is_banned
#
# All must evaluate to True.
#
# ============================================================


# ============================================================
# 32. OPERATOR PRECEDENCE
# ============================================================
#
# For logical operators:
#
#     not
#     and
#     or
#
# are evaluated in that order.
#
#
# Example:
#
#     A or B and C
#
# is interpreted as:
#
#     A or (B and C)
#
#
# Use parentheses when clarity matters.
#
# ------------------------------------------------------------

a = True
b = False
c = True

result = a or (b and c)

print(result)


# ============================================================
# 33. PARENTHESES MAKE CONDITIONS CLEAR
# ============================================================
#
# Instead of writing a complicated expression:
#
#     if age >= 18 and has_id or is_admin:
#
# Prefer clear grouping when the logic matters:
#
#     if (age >= 18 and has_id) or is_admin:
#
#
# Parentheses make the intended logical structure visible.
#
# ============================================================


# ============================================================
# 34. MEMBERSHIP OPERATORS: in / not in
# ============================================================
#
# Membership asks:
#
#     "Is this value inside this collection?"
#
#
#     in
#     not in
#
# ------------------------------------------------------------

names = ["Ali", "Sara", "Yasin"]

if "Yasin" in names:
    print("Yasin is in the list.")


if "John" not in names:
    print("John is not in the list.")


# ============================================================
# 35. MEMBERSHIP FLOW
# ============================================================
#
#     collection
#          ↓
#     membership test
#          ↓
#     True / False
#          ↓
#       decision
#
#
# Example:
#
#     "Yasin" in names
#             ↓
#           True
#             ↓
#          execute
#
# ============================================================


# ============================================================
# 36. MEMBERSHIP WITH STRINGS
# ============================================================

text = "Python programming"

if "Python" in text:
    print("Found Python.")


if "Java" not in text:
    print("Java is not present.")


# Membership with strings is case-sensitive:
#
#     "Python" in "Python programming" → True
#     "python" in "Python programming" → False
#
# ============================================================


# ============================================================
# 37. MEMBERSHIP WITH DICTIONARIES
# ============================================================
#
# For a dictionary:
#
#     key in dictionary
#
# checks KEYS by default.
#
# ------------------------------------------------------------

student = {
    "name": "Yasin",
    "age": 30
}

if "name" in student:
    print("Name key exists.")


# To check values:
#
#     value in dictionary.values()
#
# ------------------------------------------------------------

if "Yasin" in student.values():
    print("Yasin is a value.")


# ============================================================
# 38. IDENTITY OPERATORS: is / is not
# ============================================================
#
# Identity asks whether two references point to the SAME
# OBJECT.
#
#
#     is
#     is not
#
#
# IMPORTANT:
#
#     ==  → same value?
#     is  → same object?
#
# ------------------------------------------------------------

a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a == b)   # True
print(a == c)   # True

print(a is b)   # True
print(a is c)   # False


# ============================================================
# 39. == VS is
# ============================================================
#
#     a == c
#
# asks:
#
#     "Do they contain equal values?"
#
#
#     a is c
#
# asks:
#
#     "Are they the exact same object?"
#
#
# Visual:
#
#     a ───────────→ [1, 2, 3]
#                      ↑
#     b ───────────────┘
#
#
#     c ───────────→ [1, 2, 3]
#
#
# a and b → same object
# a and c → different objects
#
# ============================================================


# ============================================================
# 40. is / is not WITH None
# ============================================================
#
# The standard way to check for None is:
#
#     is None
#
# or:
#
#     is not None
#
# ------------------------------------------------------------

value = None

if value is None:
    print("No value.")


if value is not None:
    print("A value exists.")


# Prefer:
#
#     value is None
#
# rather than:
#
#     value == None
#
# ============================================================


# ============================================================
# 41. CHAINED COMPARISONS
# ============================================================
#
# Python allows:
#
#     18 <= age <= 65
#
# instead of:
#
#     age >= 18 and age <= 65
#
# ------------------------------------------------------------

age = 30

if 18 <= age <= 65:
    print("Age is within the range.")


# Mental model:
#
#     18 <= age <= 65
#
# means:
#
#     18 <= age
#        AND
#     age <= 65
#
# ============================================================


# ============================================================
# 42. MORE CHAINED COMPARISONS
# ============================================================

score = 85

if 80 <= score < 90:
    print("Score is in the 80s.")


# ============================================================
# 43. COMPARISON CHAIN VISUAL
# ============================================================
#
#     18 <= age <= 65
#
#       ↓
#
#     18 <= age
#          AND
#     age <= 65
#
#          ↓
#
#       True / False
#
# ============================================================


# ============================================================
# 44. NEGATIVE CONDITIONS
# ============================================================
#
# Conditions can describe what is NOT true.
#
# ------------------------------------------------------------

age = 15

if not age >= 18:
    print("Not an adult.")


# Often this is clearer:
#
#     if age < 18:
#
# instead of:
#
#     if not age >= 18:
#
#
# Prefer the simplest expression that communicates the idea.
#
# ============================================================


# ============================================================
# 45. CONDITIONAL EXPRESSIONS
# ============================================================
#
# Python has a one-line conditional expression.
#
#
# Pattern:
#
#     value_if_true if condition else value_if_false
#
#
# Example:
#
#     status = "Adult" if age >= 18 else "Minor"
#
# ------------------------------------------------------------

age = 25

status = "Adult" if age >= 18 else "Minor"

print(status)


# ============================================================
# 46. CONDITIONAL EXPRESSION FLOW
# ============================================================
#
#     condition?
#       /     \
#    True    False
#      ↓        ↓
#   value A  value B
#      \        /
#       \      /
#        result
#
#
# It is an EXPRESSION because it produces a value.
#
# ============================================================


# ============================================================
# 47. if STATEMENT VS CONDITIONAL EXPRESSION
# ============================================================
#
# Statement:
#
#     if age >= 18:
#         status = "Adult"
#     else:
#         status = "Minor"
#
#
# Expression:
#
#     status = "Adult" if age >= 18 else "Minor"
#
#
# Use the conditional expression for simple value selection.
#
# Use normal if/else when the logic or actions become larger.
#
# ============================================================


# ============================================================
# 48. CONDITIONAL EXPRESSION WITH NUMBERS
# =============================================================

a = 10
b = 20

maximum = a if a > b else b

print(maximum)


# Equivalent:
#
#     if a > b:
#         maximum = a
#     else:
#         maximum = b
#
# ============================================================


# ============================================================
# 49. MULTIPLE CONDITIONS IN A CONDITIONAL EXPRESSION
# ============================================================

age = 25
has_id = True

message = (
    "Allowed"
    if age >= 18 and has_id
    else "Denied"
)

print(message)


# ============================================================
# 50. AVOID OVERLY COMPLICATED ONE-LINERS
# ============================================================
#
# This is technically possible:
#
#     result = "A" if score >= 90 else "B" if score >= 80 else "C"
#
# But it can become difficult to read.
#
# Prefer:
#
#     if score >= 90:
#         result = "A"
#     elif score >= 80:
#         result = "B"
#     else:
#         result = "C"
#
# Readability matters.
#
# ============================================================


# ============================================================
# 51. EMPTY COLLECTIONS IN CONDITIONS
# ============================================================
#
# Empty collections are False.
#
# Non-empty collections are True.
#
# ------------------------------------------------------------

numbers = []

if numbers:
    print("List contains values.")
else:
    print("List is empty.")


# ------------------------------------------------------------

numbers = [10, 20]

if numbers:
    print("List contains values.")
else:
    print("List is empty.")


# ============================================================
# 52. EMPTY STRING IN CONDITIONS
# ============================================================

name = ""

if name:
    print("Name was provided.")
else:
    print("Name is empty.")


# This is often cleaner than:
#
#     if name != "":
#
# ============================================================


# ============================================================
# 53. NONE IN CONDITIONS
# ============================================================

result = None

if result is None:
    print("No result.")


# ============================================================
# 54. ZERO IN CONDITIONS
# ============================================================

count = 0

if count:
    print("There are items.")
else:
    print("Count is zero.")


# ============================================================
# 55. TRUE/FALSE TABLE FOR COMMON VALUES
# ============================================================
#
#     Value          Boolean meaning
#     --------------------------------
#     True           True
#     False          False
#     None           False
#     0              False
#     1              True
#     ""             False
#     "Python"       True
#     []             False
#     [1]            True
#     ()             False
#     (1,)           True
#     {}             False
#     {"a": 1}       True
#     set()          False
#     {1}            True
#
# ============================================================


# ============================================================
# 56. SHORT-CIRCUIT EVALUATION
# ============================================================
#
# Python can stop evaluating logical expressions early.
#
#
# AND:
#
#     False and anything
#
# The result is already known to be false.
#
#
# OR:
#
#     True or anything
#
# The result is already known to be true.
#
# ============================================================


# ============================================================
# 57. and SHORT-CIRCUIT
# ============================================================
#
#     A and B
#
# If A is False:
#
#     B does not need to be evaluated.
#
#
# Flow:
#
#     A?
#      ↓
#    False
#      ↓
#    STOP
#
# ------------------------------------------------------------

x = 0

if x != 0 and 10 / x > 2:
    print("Condition is True.")


# The second condition is not evaluated because:
#
#     x != 0
#         ↓
#       False
#
# Therefore Python avoids:
#
#     10 / 0
#
# ============================================================


# ============================================================
# 58. or SHORT-CIRCUIT
# ============================================================
#
#     A or B
#
# If A is True:
#
#     B does not need to be evaluated.
#
# ------------------------------------------------------------

is_admin = True

if is_admin or expensive_check():
    print("Access allowed.")


# If is_admin is True, Python does not need to evaluate
# expensive_check().
#
# NOTE:
# This example assumes expensive_check() exists.
# It is shown only to demonstrate evaluation behavior.
#
# ============================================================


# ============================================================
# 59. SHORT-CIRCUIT MEMORY MODEL
# ============================================================
#
#     A and B
#
#     A = False
#        ↓
#     result already determined
#        ↓
#     B skipped
#
#
#     A or B
#
#     A = True
#        ↓
#     result already determined
#        ↓
#     B skipped
#
# ============================================================


# ============================================================
# 60. USING if WITH FUNCTIONS
# ============================================================
#
# Functions can return values that become conditions.
#
# ------------------------------------------------------------

def is_adult(age):
    return age >= 18


age = 25

if is_adult(age):
    print("Adult.")


# Flow:
#
#     age
#      ↓
#     is_adult(age)
#      ↓
#     True / False
#      ↓
#     if
#
# ============================================================


# ============================================================
# 61. FUNCTIONS RETURNING BOOLEAN VALUES
# ============================================================

def is_even(number):
    return number % 2 == 0


number = 10

if is_even(number):
    print("Even number.")


# This is a useful Python pattern:
#
#     if function(...):
#         ...
#
# ============================================================


# ============================================================
# 62. if WITH LISTS
# ============================================================

numbers = [10, 20, 30]

if numbers:
    print("Numbers exist.")


if 20 in numbers:
    print("20 exists.")


# ============================================================
# 63. if WITH TUPLES
# ============================================================

numbers = (10, 20, 30)

if 20 in numbers:
    print("20 exists.")


# ============================================================
# 64. if WITH SETS
# ============================================================

numbers = {10, 20, 30}

if 20 in numbers:
    print("20 exists.")


# ============================================================
# 65. if WITH DICTIONARIES
# ============================================================

student = {
    "name": "Yasin",
    "age": 30
}

if "age" in student:
    print("Age exists.")


if student["age"] >= 18:
    print("Adult.")


# ============================================================
# 66. SAFE DICTIONARY CONDITIONAL WITH get()
# ============================================================
#
# Direct:
#
#     student["city"]
#
# can raise KeyError if the key does not exist.
#
#
# Safer:
#
#     student.get("city")
#
# ------------------------------------------------------------

student = {
    "name": "Yasin",
    "age": 30
}

city = student.get("city")

if city is not None:
    print(city)
else:
    print("City is not available.")


# ============================================================
# 67. if / elif FOR CATEGORIZATION
# ============================================================
#
# Conditional chains are useful for converting continuous
# values into categories.
#
# ------------------------------------------------------------

score = 82

if score >= 90:
    category = "Excellent"
elif score >= 75:
    category = "Good"
elif score >= 50:
    category = "Pass"
else:
    category = "Fail"

print(category)


# ============================================================
# 68. DECISION TABLE
# ============================================================
#
#     score       condition       category
#     --------------------------------------
#     95          >= 90           Excellent
#     82          >= 75           Good
#     60          >= 50           Pass
#     30          otherwise       Fail
#
#
# The first matching condition determines the category.
#
# ============================================================


# ============================================================
# 69. RANGE CONDITIONS
# ============================================================

temperature = 22

if temperature < 10:
    print("Cold.")
elif 10 <= temperature < 25:
    print("Moderate.")
else:
    print("Warm.")


# ============================================================
# 70. MULTIPLE CONDITIONS FOR VALIDATION
# ============================================================

age = 25
has_id = True
has_ticket = True

if age >= 18 and has_id and has_ticket:
    print("Entry allowed.")
else:
    print("Entry denied.")


# ============================================================
# 71. ANY() WITH CONDITIONS
# ============================================================
#
# any(iterable) returns True if AT LEAST ONE element is truthy.
#
#
#     any([False, False, True])
#                     ↓
#                   True
#
# ------------------------------------------------------------

values = [False, False, True]

if any(values):
    print("At least one condition is True.")


# ============================================================
# 72. ALL() WITH CONDITIONS
# ============================================================
#
# all(iterable) returns True if EVERY element is truthy.
#
#
#     all([True, True, True])
#               ↓
#             True
#
# ------------------------------------------------------------

values = [True, True, True]

if all(values):
    print("All conditions are True.")


# ============================================================
# 73. any() VS all()
# ============================================================
#
#     any()
#       ↓
#     "Is AT LEAST ONE true?"
#
#
#     all()
#       ↓
#     "Are ALL true?"
#
#
# Visual:
#
#     [False, False, True]
#                    ↓
#                  any → True
#
#
#     [True, True, True]
#                  ↓
#                all → True
#
# ============================================================


# ============================================================
# 74. any() WITH COMPARISONS
# ============================================================

scores = [45, 60, 72, 91]

if any(score >= 90 for score in scores):
    print("At least one excellent score exists.")


# ============================================================
# 75. all() WITH COMPARISONS
# ============================================================

scores = [60, 70, 80, 90]

if all(score >= 50 for score in scores):
    print("Everyone passed.")


# ============================================================
# 76. CONDITIONS INSIDE A FOR LOOP
# ============================================================
#
# This combines:
#
#     for → selects the current item
#     if  → decides what to do with it
#
# ------------------------------------------------------------

numbers = [10, 15, 20, 25, 30]

for number in numbers:
    if number > 20:
        print(number)


# Mental model:
#
#     LIST
#       ↓
#     take item
#       ↓
#     assign to number
#       ↓
#     check condition
#       ↓
#     True → action
#     False → skip
#       ↓
#     next item
#
#
# FOR provides the ITEMS.
# IF makes the DECISION.
#
# ============================================================


# ============================================================
# 77. ITERATION + CONDITION TRACKING
# ============================================================
#
#     numbers = [10, 15, 20, 25, 30]
#
#     Step    number    number > 20    Action
#     ------------------------------------------------
#       1       10         False        skip
#       2       15         False        skip
#       3       20         False        skip
#       4       25         True         print(25)
#       5       30         True         print(30)
#
#
# The condition is checked independently for every item.
#
# ============================================================


# ============================================================
# 78. if + else INSIDE A LOOP
# ============================================================

numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")


# ============================================================
# 79. if / elif / else INSIDE A LOOP
# ============================================================

scores = [95, 82, 68, 40]

for score in scores:
    if score >= 90:
        print("A")
    elif score >= 75:
        print("B")
    elif score >= 50:
        print("C")
    else:
        print("F")


# Mental model:
#
#     next score
#         ↓
#     if condition
#         ↓
#     elif condition
#         ↓
#     elif condition
#         ↓
#     else
#         ↓
#     next score
#
# ============================================================


# ============================================================
# 80. FILTERING WITH if
# ============================================================
#
# Filtering means:
#
#     KEEP items that satisfy a condition.
#
# ------------------------------------------------------------

numbers = [10, 25, 40, 55, 70]

large_numbers = []

for number in numbers:
    if number > 40:
        large_numbers.append(number)

print(large_numbers)


# Result:
#
# [55, 70]
#
#
# Pattern:
#
#     source
#       ↓
#     current item
#       ↓
#     condition?
#      /    \
#   True   False
#     ↓       ↓
#   keep    discard
#
# ============================================================


# ============================================================
# 81. TRANSFORMATION + CONDITION
# ============================================================
#
# Conditions can determine how an item is transformed.
#
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number % 2 == 0:
        result = number * 2
    else:
        result = number * 3

    print(number, "→", result)


# ============================================================
# 82. CONDITIONAL LIST COMPREHENSION
# ============================================================
#
# Filter:
#
#     [item for item in items if condition]
#
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [
    number
    for number in numbers
    if number % 2 == 0
]

print(even_numbers)


# ============================================================
# 83. CONDITIONAL EXPRESSION INSIDE A COMPREHENSION
# ============================================================
#
# Pattern:
#
#     [
#         value_if_true if condition else value_if_false
#         for item in iterable
#     ]
#
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5]

labels = [
    "even" if number % 2 == 0 else "odd"
    for number in numbers
]

print(labels)


# ============================================================
# 84. TWO DIFFERENT USES OF if IN COMPREHENSIONS
# ============================================================
#
# FILTER:
#
#     [x for x in values if condition]
#
#     condition
#         ↓
#     keep / discard
#
#
# CONDITIONAL VALUE:
#
#     [A if condition else B for x in values]
#
#     condition
#         ↓
#     choose A or B
#
#
# These are NOT the same thing.
#
# ============================================================


# ============================================================
# 85. COMMON CONDITIONAL PATTERN: PASS/FAIL
# ============================================================

score = 72

if score >= 50:
    print("Pass")
else:
    print("Fail")


# ============================================================
# 86. COMMON CONDITIONAL PATTERN: POSITIVE/NEGATIVE
# ============================================================

number = -10

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")


# ============================================================
# 87. COMMON CONDITIONAL PATTERN: EVEN/ODD
# ============================================================

number = 17

if number % 2 == 0:
    print("Even")
else:
    print("Odd")


# ============================================================
# 88. COMMON CONDITIONAL PATTERN: MAXIMUM
# ============================================================

a = 25
b = 40

if a > b:
    maximum = a
else:
    maximum = b

print(maximum)


# ============================================================
# 89. COMMON CONDITIONAL PATTERN: MINIMUM
# ============================================================

a = 25
b = 40

if a < b:
    minimum = a
else:
    minimum = b

print(minimum)


# ============================================================
# 90. COMMON CONDITIONAL PATTERN: EMPTY DATA
# ============================================================

data = []

if data:
    print("Data available.")
else:
    print("No data available.")


# ============================================================
# 91. COMMON CONDITIONAL PATTERN: VALIDATION
# ============================================================

username = "Yasin"
password = "12345"

if username and password:
    print("Both fields were provided.")
else:
    print("Missing information.")


# ============================================================
# 92. COMMON CONDITIONAL PATTERN: DATA CLEANING
# ============================================================

name = "  Yasin  "

if name:
    name = name.strip()

print(name)


# ============================================================
# 93. CONDITIONAL STRING COMPARISON
# ============================================================

name = "Yasin"

if name == "Yasin":
    print("Correct name.")


# String comparison is case-sensitive:
#
#     "Yasin" == "Yasin" → True
#     "Yasin" == "yasin" → False
#
# ============================================================


# ============================================================
# 94. CASE-INSENSITIVE CONDITIONAL CHECK
# ============================================================

name = "YASIN"

if name.lower() == "yasin":
    print("Name matches.")


# ============================================================
# 95. MULTIPLE ACCEPTED VALUES
# ============================================================

day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("Weekend.")


# More readable for many alternatives:
#
#     if day in ("Saturday", "Sunday"):
#
# ------------------------------------------------------------

if day in ("Saturday", "Sunday"):
    print("Weekend.")


# ============================================================
# 96. CONDITIONS WITH USER INPUT
# ============================================================
#
# input() always returns a string.
#
# Convert it when numerical comparison is needed.
#
# ------------------------------------------------------------

age = int(input("Enter your age: "))

if age >= 18:
    print("Adult.")
else:
    print("Minor.")


# ============================================================
# 97. INPUT VALIDATION
# ============================================================

age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age.")
elif age < 18:
    print("Minor.")
else:
    print("Adult.")


# ============================================================
# 98. CONDITIONS AND EXCEPTIONS
# ============================================================
#
# Conditions and exceptions solve different problems.
#
# if:
#     "I know how to decide based on a value."
#
# try/except:
#     "An operation may fail unexpectedly."
#
#
# Example:
#
#     if age >= 18:
#         ...
#
# is a normal decision.
#
# ============================================================


# ============================================================
# 99. match / case
# ============================================================
#
# Python also provides structural pattern matching:
#
#     match value:
#         case pattern:
#             action
#
#
# It is useful when selecting behavior based on patterns
# or specific structures.
#
# ------------------------------------------------------------

command = "start"

match command:
    case "start":
        print("Starting.")
    case "stop":
        print("Stopping.")
    case "pause":
        print("Pausing.")
    case _:
        print("Unknown command.")


# ============================================================
# 100. match / case FLOW
# ============================================================
#
#             value
#               ↓
#            match
#               ↓
#       ┌───────┼────────┐
#       ↓       ↓        ↓
#     case A  case B   case C
#       ↓       ↓        ↓
#     action  action   action
#
#     case _ = default
#
# ============================================================


# ============================================================
# 101. case _ AS DEFAULT
# ============================================================
#
# The underscore:
#
#     _
#
# can act as a catch-all pattern.
#
# ------------------------------------------------------------

status = "unknown"

match status:
    case "success":
        print("Success.")
    case "error":
        print("Error.")
    case _:
        print("Other status.")


# ============================================================
# 102. match / case VS if / elif
# ============================================================
#
# if / elif:
#
#     if condition_A:
#         ...
#     elif condition_B:
#         ...
#
#
# match / case:
#
#     match value:
#         case pattern_A:
#             ...
#         case pattern_B:
#             ...
#
#
# Use if/elif when conditions are naturally Boolean
# comparisons or logical expressions.
#
# Use match/case when matching a value or structure against
# patterns is clearer.
#
# ============================================================


# ============================================================
# 103. MATCHING MULTIPLE VALUES
# ============================================================

day = "Saturday"

match day:
    case "Saturday" | "Sunday":
        print("Weekend.")
    case _:
        print("Weekday.")


# ============================================================
# 104. MATCHING STRUCTURED DATA
# ============================================================
#
# match/case can go beyond simple strings.
#
# Example with a tuple:
#
# ------------------------------------------------------------

point = (0, 5)

match point:
    case (0, 0):
        print("Origin.")
    case (0, y):
        print("On the y-axis.")
    case (x, 0):
        print("On the x-axis.")
    case (x, y):
        print("General point.")


# ============================================================
# 105. CONDITIONAL ASSIGNMENT WITH DEFAULT VALUES
# ============================================================
#
# A common pattern:
#
#     value = something if condition else default
#
# ------------------------------------------------------------

score = None

result = score if score is not None else 0

print(result)


# ============================================================
# 106. if NOT TO USE
# ============================================================
#
# Avoid unnecessary conditions.
#
# Instead of:
#
#     if is_valid == True:
#         print("Valid")
#
# Prefer:
#
#     if is_valid:
#         print("Valid")
#
#
# Instead of:
#
#     if is_valid == False:
#         print("Invalid")
#
# Prefer:
#
#     if not is_valid:
#         print("Invalid")
#
# ============================================================


# ============================================================
# 107. BOOLEAN COMPARISON CLEANUP
# ============================================================
#
# Avoid:
#
#     if x == True:
#
# Prefer:
#
#     if x:
#
#
# Avoid:
#
#     if x == False:
#
# Prefer:
#
#     if not x
#
#
# EXCEPTION:
#
# Sometimes explicit comparison is useful when you truly need
# to distinguish Boolean True/False from other truthy values.
#
# ============================================================


# ============================================================
# 108. AVOID UNNECESSARY NESTING
# ============================================================
#
# Instead of:
#
#     if age >= 18:
#         if has_id:
#             print("Allowed")
#
#
# You can often write:
#
#     if age >= 18 and has_id:
#         print("Allowed")
#
#
# Both are valid.
#
# Use the form that makes the logic easiest to understand.
#
# ============================================================


# ============================================================
# 109. GUARD CONDITIONS
# ============================================================
#
# A guard condition checks for an invalid or special case
# early.
#
# ------------------------------------------------------------

age = -5

if age < 0:
    print("Invalid age.")
else:
    print("Valid age.")


# In functions, guard clauses are especially useful:
#
#     if invalid:
#         return
#
# They prevent deeply nested code.
#
# ============================================================


# ============================================================
# 110. CONDITIONS WITH return
# ============================================================

def check_age(age):

    if age < 0:
        return "Invalid age."

    if age >= 18:
        return "Adult."

    return "Minor."


print(check_age(25))


# Mental model:
#
#     function starts
#          ↓
#     invalid?
#       ↓
#     no
#       ↓
#     adult?
#       ↓
#     yes
#       ↓
#     return "Adult."
#          ↓
#     function ends
#
# ============================================================


# ============================================================
# 111. CONDITIONS WITH break
# ============================================================
#
# In a loop:
#
#     if condition:
#         break
#
# means:
#
#     if condition becomes True
#         → leave the loop immediately
#
# ------------------------------------------------------------

numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        break

    print(number)


# Output:
#
# 10
# 20
#
#
# Flow:
#
#     10 → condition False → print
#     20 → condition False → print
#     30 → condition True  → break
#
# ============================================================


# ============================================================
# 112. CONDITIONS WITH continue
# ============================================================
#
#     if condition:
#         continue
#
# means:
#
#     skip the rest of the CURRENT iteration.
#
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number % 2 == 0:
        continue

    print(number)


# Output:
#
# 1
# 3
# 5
#
#
# Condition True:
#
#     even → continue → skip print
#
# Condition False:
#
#     odd → print
#
# ============================================================


# ============================================================
# 113. CONDITIONAL CONTROL FLOW MAP
# ============================================================
#
#                         LOOP
#                           ↓
#                     get current item
#                           ↓
#                        if condition
#                       /            \
#                    True           False
#                     ↓                ↓
#                  action           skip
#                     ↓                ↓
#                  continue?       next code
#                     ↓
#                  break?
#                     ↓
#                next iteration
#
# ============================================================


# ============================================================
# 114. COMMON CONDITIONAL OPERATORS SUMMARY
# ============================================================
#
# COMPARISON:
#
#     ==      equal
#     !=      not equal
#     >       greater
#     <       less
#     >=      greater/equal
#     <=      less/equal
#
#
# LOGICAL:
#
#     and
#     or
#     not
#
#
# MEMBERSHIP:
#
#     in
#     not in
#
#
# IDENTITY:
#
#     is
#     is not
#
#
# ============================================================


# ============================================================
# 115. CONDITION TYPE SUMMARY
# ============================================================
#
# ┌───────────────────────────────┬─────────────────────────┐
# │ Condition                     │ Question                │
# ├───────────────────────────────┼─────────────────────────┤
# │ x == y                        │ Same value?             │
# │ x != y                        │ Different value?        │
# │ x > y                         │ x greater?              │
# │ x < y                         │ x smaller?              │
# │ x >= y                        │ x at least y?           │
# │ x <= y                        │ x at most y?            │
# │ x in collection               │ Present?                │
# │ x not in collection           │ Absent?                 │
# │ x is None                     │ Exactly None?            │
# │ A and B                       │ Both true?              │
# │ A or B                        │ At least one true?      │
# │ not A                         │ Reverse A?              │
# │ bool(x)                       │ Truthy?                 │
# └───────────────────────────────┴─────────────────────────┘
#
# ============================================================


# ============================================================
# 116. if / elif / else SUMMARY
# ============================================================
#
#     if
#     ↓
#     first condition
#
#     elif
#     ↓
#     another condition
#
#     else
#     ↓
#     everything else
#
#
# RULE:
#
#     if     → checked first
#     elif   → checked only if previous conditions were False
#     else   → runs if every previous condition was False
#
#
# In an if/elif/else chain:
#
#     maximum ONE branch executes.
#
# ============================================================


# ============================================================
# 117. INDEPENDENT if VS if/elif/else
# ============================================================
#
# INDEPENDENT if:
#
#     if A:
#         ...
#
#     if B:
#         ...
#
#     if C:
#         ...
#
#     → A, B, and C can ALL execute.
#
#
# if/elif/else:
#
#     if A:
#         ...
#     elif B:
#         ...
#     elif C:
#         ...
#     else:
#         ...
#
#     → ONLY ONE branch executes.
#
#
# This distinction is CRITICAL.
#
# ============================================================


# ============================================================
# 118. DECISION TABLE: INDEPENDENT if
# ============================================================
#
#     score = 95
#
#     Condition       Result
#     -----------------------
#     >= 50           True
#     >= 80           True
#     >= 90           True
#
#     All three execute.
#
#
# ============================================================
# 119. DECISION TABLE: if / elif / else
# ============================================================
#
#     score = 95
#
#     Condition       Result
#     -----------------------
#     >= 90           True  ← FIRST TRUE
#     >= 80           NOT CHECKED
#     >= 70           NOT CHECKED
#     else            NOT USED
#
#
# Only the first matching branch executes.
#
# ============================================================


# ============================================================
# 120. NESTED VS LOGICAL CONDITIONS
# ============================================================
#
# NESTED:
#
#     if A:
#         if B:
#             action
#
#
# COMBINED:
#
#     if A and B:
#         action
#
#
# Conceptually both mean:
#
#     A must be True
#     AND
#     B must be True
#
#
# But they have different control-flow structure.
#
# ============================================================


# ============================================================
# 121. CONDITIONAL DECISION TREE
# ============================================================
#
# Example:
#
#     score = 83
#
#                     score >= 90?
#                       /       \
#                    Yes         No
#                    ↓            ↓
#                    A        score >= 75?
#                              /       \
#                           Yes         No
#                           ↓            ↓
#                           B        score >= 50?
#                                      /      \
#                                   Yes        No
#                                    ↓          ↓
#                                    C          F
#
#
# For 83:
#
#     83 >= 90 → False
#     83 >= 75 → True
#     result = B
#
# ============================================================


# ============================================================
# 122. MEMORY MODEL OF A CONDITIONAL
# ============================================================
#
# Example:
#
#     age = 25
#
#
# MEMORY:
#
#     age ─────→ 25
#
#
# CONDITION:
#
#     age >= 18
#
#         ↓
#
#     25 >= 18
#
#         ↓
#
#       True
#
#         ↓
#
#     execute if block
#
#
# A condition does NOT change age.
#
# It reads values and produces a Boolean decision.
#
# ============================================================


# ============================================================
# 123. CONDITION = QUESTION
# ============================================================
#
# A useful mental model:
#
#     if condition:
#
# means:
#
#     "Ask this question."
#
#
# Example:
#
#     if age >= 18:
#
# reads naturally as:
#
#     "Is age at least 18?"
#
#
# Example:
#
#     if "Yasin" in names:
#
# reads as:
#
#     "Is Yasin inside names?"
#
#
# Example:
#
#     if score >= 90 and has_project:
#
# reads as:
#
#     "Is the score at least 90 AND is there a project?"
#
# ============================================================


# ============================================================
# 124. CONDITION → BOOLEAN → ACTION
# ============================================================
#
# This is the fundamental conditional pattern:
#
#
#     VARIABLES
#         ↓
#     EXPRESSION
#         ↓
#     True / False
#         ↓
#     if
#         ↓
#     ACTION
#
#
# Example:
#
#     score = 85
#
#     score >= 50
#          ↓
#        True
#          ↓
#     print("Pass")
#
# ============================================================


# ============================================================
# 125. CONDITIONALS + VARIABLES
# ============================================================
#
# Variables hold the information.
#
# Conditions inspect the information.
#
# Actions respond to the result.
#
#
#     variable
#        ↓
#     condition
#        ↓
#     decision
#        ↓
#     action
#
#
# Example:
#
#     temperature = 35
#
#     temperature > 30
#              ↓
#            True
#              ↓
#         "Hot"
#
# ============================================================


# ============================================================
# 126. CONDITIONALS + FUNCTIONS
# ============================================================
#
# A function can:
#
#     receive data
#          ↓
#     evaluate conditions
#          ↓
#     return a decision/result
#
#
# Example:
#
#     def classify_score(score):
#         if score >= 90:
#             return "A"
#         elif score >= 75:
#             return "B"
#         else:
#             return "C"
#
#
# This turns a decision process into a reusable function.
#
# ============================================================


# ============================================================
# 127. CONDITIONALS + LOOPS + COLLECTIONS
# ============================================================
#
# These three concepts work together constantly in Python.
#
#
# COLLECTION
#     ↓
# FOR LOOP
#     ↓
# CURRENT ITEM
#     ↓
# IF CONDITION
#     ↓
# ACTION
#     ↓
# NEXT ITEM
#
#
# Example:
#
#     numbers = [10, 20, 30, 40]
#
#     for number in numbers:
#         if number > 20:
#             print(number)
#
#
# FOR:
#     "Which item am I processing?"
#
# IF:
#     "Should I do something with this item?"
#
# ============================================================


# ============================================================
# 128. CONDITIONALS + DATA ANALYSIS
# ============================================================
#
# Conditions are everywhere in data work.
#
#
# Examples:
#
#     if age >= 18
#         → adult
#
#     if revenue > target
#         → above target
#
#     if failure_probability > threshold
#         → high risk
#
#     if value is None
#         → missing value
#
#     if score >= 0.8
#         → high confidence
#
#     if category in allowed_categories
#         → valid
#
#
# A huge amount of data processing is:
#
#     READ DATA
#        ↓
#     CHECK CONDITION
#        ↓
#     CLASSIFY / FILTER / TRANSFORM
#        ↓
#     STORE RESULT
#
# ============================================================


# ============================================================
# 129. DATA-QUALITY EXAMPLE
# ============================================================

age = 35

if age < 0:
    print("Invalid age.")
elif age > 120:
    print("Suspicious age.")
else:
    print("Age appears valid.")


# ============================================================
# 130. RISK CLASSIFICATION EXAMPLE
# ============================================================

failure_probability = 0.82

if failure_probability >= 0.80:
    risk = "High"
elif failure_probability >= 0.50:
    risk = "Medium"
else:
    risk = "Low"

print(risk)


# ============================================================
# 131. MISSING-VALUE DECISION
# ============================================================

value = None

if value is None:
    print("Missing.")
else:
    print("Available.")


# ============================================================
# 132. THRESHOLD DECISION
# ============================================================

accuracy = 0.91

if accuracy >= 0.90:
    print("Meets target.")
else:
    print("Below target.")


# ============================================================
# 133. MULTIPLE BUSINESS CONDITIONS
# ============================================================

revenue = 120000
target = 100000
customers = 150

if revenue >= target and customers >= 100:
    print("Target achieved with sufficient customers.")
elif revenue >= target:
    print("Revenue target achieved.")
else:
    print("Revenue target not achieved.")


# ============================================================
# 134. PRACTICAL DECISION PIPELINE
# ============================================================
#
#     DATA
#       ↓
#     READ VALUE
#       ↓
#     CHECK CONDITION
#       ↓
#     ┌───────────────┐
#     │               │
#   TRUE            FALSE
#     │               │
#     ↓               ↓
#   ACTION          OTHER ACTION
#     │               │
#     └───────┬───────┘
#             ↓
#          NEXT STEP
#
# ============================================================


# ============================================================
# 135. COMMON MISTAKES
# ============================================================
#
# MISTAKE 1:
#
#     if age = 18:
#
# WRONG
#
# Use:
#
#     if age == 18:
#
#
# MISTAKE 2:
#
#     if age > 18
#
# WRONG
#
# Missing colon.
#
# Correct:
#
#     if age > 18:
#
#
# MISTAKE 3:
#
#     if age >= 18:
#     print("Adult")
#
# WRONG indentation.
#
#
# MISTAKE 4:
#
#     if age >= 18:
#         print("Adult")
#     else age < 18:
#
# WRONG.
#
# Use:
#
#     else:
#
#
# MISTAKE 5:
#
#     if x == True:
#
# Usually unnecessary.
#
# Prefer:
#
#     if x:
#
# ============================================================


# ============================================================
# 136. COMMON LOGIC MISTAKE: WRONG elif ORDER
# ============================================================
#
# WRONG:
#
#     if score >= 50:
#         print("Pass")
#     elif score >= 90:
#         print("Excellent")
#
#
# For score = 95:
#
#     95 >= 50 → True
#
# So "Pass" executes.
#
#
# CORRECT:
#
#     if score >= 90:
#         print("Excellent")
#     elif score >= 50:
#         print("Pass")
#
# ============================================================


# ============================================================
# 137. COMMON LOGIC MISTAKE: USING and WHEN or IS NEEDED
# ============================================================
#
# Suppose:
#
#     day = "Saturday"
#
# WRONG:
#
#     if day == "Saturday" and day == "Sunday":
#
# A single day cannot simultaneously equal both strings.
#
#
# Correct:
#
#     if day == "Saturday" or day == "Sunday":
#
# Or:
#
#     if day in ("Saturday", "Sunday"):
#
# ============================================================


# ============================================================
# 138. COMMON LOGIC MISTAKE: CHAINING VALUES INCORRECTLY
# ============================================================
#
# Avoid:
#
#     if x == 1 or 2:
#
# This does NOT mean:
#
#     x == 1 OR x == 2
#
#
# Correct:
#
#     if x == 1 or x == 2:
#
# Better:
#
#     if x in (1, 2):
#
# ============================================================


# ============================================================
# 139. COMMON LOGIC MISTAKE: is VS ==
# ============================================================
#
# Usually:
#
#     == → compare values
#
#     is → compare object identity
#
#
# Use:
#
#     if name == "Yasin":
#
# not:
#
#     if name is "Yasin":
#
#
# Use:
#
#     if value is None:
#
# for None checks.
#
# ============================================================


# ============================================================
# 140. COMMON LOGIC MISTAKE: EMPTY LIST COMPARISON
# ============================================================
#
# Instead of:
#
#     if numbers == []:
#
# you can often write:
#
#     if not numbers:
#
#
# To check that data exists:
#
#     if numbers:
#
# ============================================================


# ============================================================
# 141. COMMON LOGIC MISTAKE: TOO MANY NESTED ifs
# ============================================================
#
# Hard to read:
#
#     if A:
#         if B:
#             if C:
#                 if D:
#                     action
#
#
# Often clearer:
#
#     if A and B and C and D:
#         action
#
#
# Or use guard clauses/functions when the logic is complex.
#
# ============================================================


# ============================================================
# 142. SIMPLE CONDITIONAL DESIGN RULE
# ============================================================
#
# Ask:
#
#     "What decision am I making?"
#
# Then identify:
#
#     1. Input/value
#     2. Condition
#     3. True action
#     4. False/default action
#
#
# Example:
#
#     Input:
#         score
#
#     Condition:
#         score >= 50
#
#     True:
#         Pass
#
#     False:
#         Fail
#
# ============================================================


# ============================================================
# 143. CONDITIONAL DECISION TEMPLATE
# ============================================================
#
#     value = ...
#
#     if condition:
#         action
#     elif another_condition:
#         another_action
#     else:
#         default_action
#
#
# Think:
#
#     INPUT
#       ↓
#     QUESTION
#       ↓
#     DECISION
#       ↓
#     ACTION
#
# ============================================================


# ============================================================
# 144. MASTER CONDITIONAL PATTERNS
# ============================================================
#
# TWO-WAY DECISION:
#
#     if condition:
#         A
#     else:
#         B
#
#
# MULTI-WAY DECISION:
#
#     if A:
#         X
#     elif B:
#         Y
#     else:
#         Z
#
#
# INDEPENDENT DECISIONS:
#
#     if A:
#         X
#
#     if B:
#         Y
#
#
# COMBINED CONDITIONS:
#
#     if A and B:
#         X
#
#
# ALTERNATIVE CONDITIONS:
#
#     if A or B:
#         X
#
#
# NEGATED CONDITION:
#
#     if not A:
#         X
#
#
# MEMBERSHIP:
#
#     if x in collection:
#         X
#
#
# IDENTITY:
#
#     if x is None:
#         X
#
#
# ONE-LINE VALUE SELECTION:
#
#     result = A if condition else B
#
# ============================================================


# ============================================================
# 145. MASTER TRUTH TABLE
# ============================================================
#
# AND:
#
#     A       B       A and B
#     ------------------------
#     F       F       F
#     F       T       F
#     T       F       F
#     T       T       T
#
#
# OR:
#
#     A       B       A or B
#     -----------------------
#     F       F       F
#     F       T       T
#     T       F       T
#     T       T       T
#
#
# NOT:
#
#     A       not A
#     ------------
#     F       T
#     T       F
#
# ============================================================


# ============================================================
# 146. CONDITIONAL OPERATOR MAP
# ============================================================
#
#                    CONDITIONS
#                        │
#       ┌────────────────┼─────────────────┐
#       ↓                ↓                 ↓
#   COMPARISON       MEMBERSHIP         IDENTITY
#       │                │                 │
#   == != > <       in / not in       is / is not
#   >= <=
#       │                │                 │
#       └────────────────┼─────────────────┘
#                        ↓
#                   True / False
#                        ↓
#                    LOGIC
#                        │
#              ┌─────────┼─────────┐
#              ↓         ↓         ↓
#             and        or        not
#              │         │         │
#              └─────────┼─────────┘
#                        ↓
#                     if / elif
#                        ↓
#                      else
#                        ↓
#                      ACTION
#
# ============================================================


# ============================================================
# 147. CONDITIONAL LEARNING MAP
# ============================================================
#
# STEP 1
#     True / False
#
#        ↓
#
# STEP 2
#     Comparison operators
#
#        ↓
#
# STEP 3
#     if
#
#        ↓
#
# STEP 4
#     if / else
#
#        ↓
#
# STEP 5
#     if / elif / else
#
#        ↓
#
# STEP 6
#     Logical operators
#     and / or / not
#
#        ↓
#
# STEP 7
#     Membership
#     in / not in
#
#        ↓
#
# STEP 8
#     Identity
#     is / is not
#
#        ↓
#
# STEP 9
#     Truthiness
#
#        ↓
#
# STEP 10
#     Nested conditions
#
#        ↓
#
# STEP 11
#     Conditions + loops
#
#        ↓
#
# STEP 12
#     Filtering / transformation
#
#        ↓
#
# STEP 13
#     Conditional expressions
#
#        ↓
#
# STEP 14
#     any() / all()
#
#        ↓
#
# STEP 15
#     match / case
#
#        ↓
#
# STEP 16
#     Real-world decision systems
#
# ============================================================


# ============================================================
# 148. CONDITIONALS: WHAT PYTHON IS REALLY DOING
# ============================================================
#
# Python does not "understand" the condition as a human does.
#
# Conceptually:
#
#     expression
#         ↓
#     evaluate expression
#         ↓
#     Boolean/truth value
#         ↓
#     choose control-flow path
#
#
# Example:
#
#     age >= 18
#
#         ↓
#
#     compare 25 with 18
#
#         ↓
#
#     True
#
#         ↓
#
#     execute if block
#
# ============================================================


# ============================================================
# 149. THE PYTHON FEELING
# ============================================================
#
# When you see:
#
#     if condition:
#         action
#
# Think:
#
#     "ASK A QUESTION."
#
#     condition?
#         ↓
#     True or False?
#         ↓
#     If True → do this.
#
#
# When you see:
#
#     if condition:
#         A
#     else:
#         B
#
# Think:
#
#     "CHOOSE BETWEEN TWO PATHS."
#
#
# When you see:
#
#     if A:
#         X
#     elif B:
#         Y
#     else:
#         Z
#
# Think:
#
#     "CHECK OPTIONS FROM TOP TO BOTTOM.
#      FIRST TRUE OPTION WINS."
#
#
# When you see:
#
#     if A and B:
#
# Think:
#
#     "BOTH must be true."
#
#
# When you see:
#
#     if A or B:
#
# Think:
#
#     "AT LEAST ONE must be true."
#
#
# When you see:
#
#     if not A:
#
# Think:
#
#     "REVERSE the condition."
#
#
# When you see:
#
#     if x in collection:
#
# Think:
#
#     "Is x present?"
#
#
# When you see:
#
#     if x is None:
#
# Think:
#
#     "Is x exactly the None object?"
#
#
# ============================================================


# ============================================================
# 150. FINAL MENTAL MODEL
# ============================================================
#
#
#                 DATA / VARIABLES
#                        ↓
#                   CONDITION
#                        ↓
#                  TRUE / FALSE
#                        ↓
#              ┌─────────┴─────────┐
#              ↓                   ↓
#            TRUE                FALSE
#              ↓                   ↓
#          IF BLOCK          ELSE / SKIP
#              ↓                   ↓
#              └─────────┬─────────┘
#                        ↓
#                     NEXT STEP
#
#
# The deepest idea:
#
#     CONDITIONALS CONTROL THE PATH
#     THAT PYTHON TAKES THROUGH YOUR PROGRAM.
#
#
# FOR LOOP:
#
#     "Which item should I process?"
#
#
# IF:
#
#     "What should I do with this item/value?"
#
#
# ELIF:
#
#     "What other condition should I check?"
#
#
# ELSE:
#
#     "What should happen if none of the conditions matched?"
#
#
# AND:
#
#     "Both."
#
#
# OR:
#
#     "At least one."
#
#
# NOT:
#
#     "The opposite."
#
#
# IN:
#
#     "Is it contained?"
#
#
# IS:
#
#     "Is it the same object?"
#
#
# CONDITIONAL EXPRESSION:
#
#     "Choose one value based on a condition."
#
#
# MATCH / CASE:
#
#     "Which pattern does this value match?"
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
#     CONDITION
#         ↓
#     True / False
#         ↓
#     DECISION
#         ↓
#     CONTROL FLOW
#         ↓
#     ACTION
#
#
# Python feeling:
#
#     "Evaluate → Decide → Execute → Continue."
#
# ============================================================
```
