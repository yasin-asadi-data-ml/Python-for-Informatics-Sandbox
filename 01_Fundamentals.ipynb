```python
# ------------------------------------------------------------
# Searching
# ------------------------------------------------------------

# This code demonstrates string methods for finding and counting substrings in Python
sentence = "Python is fun. Python is powerful."

# Using find() to get the index of the first occurrence of "Python"
print(sentence.find("Python"))      # first occurrence

# Using rfind() to get the index of the last occurrence of "Python"
print(sentence.rfind("Python"))     # last occurrence


# Using index() to get the index of the first occurrence of "Python"
# This method raises an error if the substring is not found (as noted in the comment)
print(sentence.index("Python"))     # like find(), but raises error if not found WOW!!!!!!

# Using count() to count the total occurrences of "Python" in the string
print(sentence.count("Python"))     # count occurrences
```

```python
# ------------------------------------------------------------
# maketrans() + translate()
# This script demonstrates the use of Python's string maketrans() and translate() methods
# for character replacement in strings.
# ------------------------------------------------------------

# Define a string to be translated
text = "abc123"

# Create a translation table using str.maketrans method
# The table maps specific characters to their replacements:
# lowercase letters a, b, c are mapped to uppercase A, B, C
# digits 1, 2, 3 are mapped to special characters !, @, #
table = str.maketrans({
    "a": "A",
    "b": "B",
    "c": "C",
    "1": "!",
    "2": "@",
    "3": "#"
})

# Apply the translation table to the input string
# This will replace each character in text according to the mapping in table
# Print the text and table variables
print("text: ",text)  # Display original string
print("table: ",table)  # Display the translation table object

# Print the Unicode values of characters 'a', 'b', and 'c'

# This shows the ASCII values of these characters
print("Unicode of \"a\" :",ord("a"))  # Unicode code point of 'a'
print("Unicode of \"b\" :",ord("b"))  # Unicode code point of 'b'
print("Unicode of \"c\" :",ord("c"))  # Unicode code point of 'c'

"""
Replace each character in the string using the given translation table.
Translation table, which must be a mapping of Unicode ordinals to
Unicode ordinals, strings, or None.    
"""

# Apply the translation to the original string and display the result
print("text.translate: ",text.translate(table))
# Show translated string
```

```python
# ------------------------------------------------------------
# Removing characters with translate()
# ------------------------------------------------------------

# Define a string containing text and numbers
text = "Machine Learning 2026"

# Create a translation table that removes all digits
# The first two arguments are empty because we're not replacing characters
# The third argument contains all digits to be removed
table = str.maketrans("", "", "0123456789")
print(type(table)) # Print the type of the translation table

# Apply the translation table to remove digits from the text
# and print the result
print(text)

print(text.translate(table))# translate the text using the translation table.
```
