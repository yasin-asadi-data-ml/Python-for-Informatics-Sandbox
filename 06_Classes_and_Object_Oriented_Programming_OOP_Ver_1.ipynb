```python
# ============================================================
#       PYTHON CLASSES & OBJECT-ORIENTED PROGRAMMING
# ============================================================
#
# OOP = Object-Oriented Programming
#
# The four major ideas commonly associated with OOP are:
#
# 1. Encapsulation
# 2. Inheritance
# 3. Polymorphism
# 4. Abstraction
#
# A CLASS is a blueprint.
# An OBJECT is an instance created from a class.
#
# ============================================================


# ============================================================
# 1. SIMPLE CLASS
# ============================================================

class Student:
    pass


# Create an object
student1 = Student()

print(student1)


# ============================================================
# 2. CLASS WITH ATTRIBUTES
# ============================================================

class Student:

    name = "Yasin"
    age = 35
    field = "Data Science"


student1 = Student()

print(student1.name)
print(student1.age)
print(student1.field)


# ============================================================
# 3. CREATING MULTIPLE OBJECTS
# ============================================================

class Student:

    name = "Unknown"
    age = 0


student1 = Student()
student2 = Student()

student1.name = "Ali"
student1.age = 25

student2.name = "Sara"
student2.age = 30

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)


# ============================================================
# 4. __init__()
# ============================================================
#
# __init__() is called automatically when an object is created.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age


student1 = Student("Yasin", 35)

print(student1.name)
print(student1.age)


# ============================================================
# 5. UNDERSTANDING self
# ============================================================
#
# self refers to the current object.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age


student1 = Student("Ali", 25)
student2 = Student("Sara", 30)

print(student1.name)
print(student2.name)


# ============================================================
# 6. MULTIPLE ATTRIBUTES
# ============================================================

class Researcher:

    def __init__(
        self,
        name,
        field,
        experience
    ):

        self.name = name
        self.field = field
        self.experience = experience


researcher1 = Researcher(
    "Yasin",
    "Machine Learning",
    8
)

print(researcher1.name)
print(researcher1.field)
print(researcher1.experience)


# ============================================================
# 7. METHODS
# ============================================================
#
# A method is a function defined inside a class.
#
# ============================================================

class Student:

    def __init__(self, name):

        self.name = name

    def introduce(self):

        print(
            "My name is",
            self.name
        )


student1 = Student("Yasin")

student1.introduce()


# ============================================================
# 8. METHOD WITH PARAMETERS
# ============================================================

class Calculator:

    def add(self, a, b):

        return a + b

    def subtract(self, a, b):

        return a - b


calculator = Calculator()

print(
    calculator.add(10, 20)
)

print(
    calculator.subtract(20, 10)
)


# ============================================================
# 9. OBJECT ATTRIBUTES
# ============================================================

class Person:

    def __init__(self, name, age):

        self.name = name
        self.age = age


person = Person(
    "Yasin",
    35
)

print(person.name)
print(person.age)


# ============================================================
# 10. MODIFY OBJECT ATTRIBUTES
# ============================================================

person.age = 36

print(person.age)


# ============================================================
# 11. ADD NEW ATTRIBUTE
# ============================================================

person.city = "Karaj"

print(person.city)


# ============================================================
# 12. DELETE ATTRIBUTE
# ============================================================

del person.city

# print(person.city)
# This would produce an AttributeError.


# ============================================================
# 13. DELETE OBJECT
# ============================================================

student = Student("Ali")

del student

# The object no longer exists.


# ============================================================
# 14. CHECK OBJECT TYPE
# ============================================================

class Student:
    pass


student = Student()

print(type(student))


# ============================================================
# 15. isinstance()
# ============================================================

class Student:
    pass


student = Student()

print(
    isinstance(
        student,
        Student
    )
)


# ============================================================
# 16. CLASS ATTRIBUTES
# ============================================================
#
# A class attribute belongs to the class.
#
# ============================================================

class Employee:

    company = "Data Analytics Lab"

    def __init__(self, name):

        self.name = name


employee1 = Employee("Ali")
employee2 = Employee("Sara")

print(employee1.company)
print(employee2.company)


# ============================================================
# 17. INSTANCE ATTRIBUTES
# ============================================================

class Employee:

    company = "Data Analytics Lab"

    def __init__(self, name, salary):

        self.name = name
        self.salary = salary


employee1 = Employee(
    "Ali",
    50000
)

employee2 = Employee(
    "Sara",
    60000
)

print(employee1.name)
print(employee1.salary)

print(employee2.name)
print(employee2.salary)


# ============================================================
# 18. CLASS ATTRIBUTE VS INSTANCE ATTRIBUTE
# ============================================================

class Employee:

    company = "AI Lab"

    def __init__(self, name):

        self.name = name


employee1 = Employee("Ali")

print(employee1.company)

Employee.company = "Data Science Lab"

print(employee1.company)


# ============================================================
# 19. METHOD USING OBJECT ATTRIBUTES
# ============================================================

class Rectangle:

    def __init__(self, width, height):

        self.width = width
        self.height = height

    def area(self):

        return self.width * self.height


rectangle = Rectangle(
    10,
    5
)

print(rectangle.area())


# ============================================================
# 20. MULTIPLE METHODS
# ============================================================

class Rectangle:

    def __init__(self, width, height):

        self.width = width
        self.height = height

    def area(self):

        return self.width * self.height

    def perimeter(self):

        return (
            2 * self.width
            + 2 * self.height
        )


rectangle = Rectangle(
    10,
    5
)

print(rectangle.area())
print(rectangle.perimeter())


# ============================================================
# 21. METHOD MODIFYING AN ATTRIBUTE
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.balance = balance

    def deposit(self, amount):

        self.balance += amount


account = BankAccount(1000)

account.deposit(500)

print(account.balance)


# ============================================================
# 22. METHOD WITH CONDITION
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.balance = balance

    def withdraw(self, amount):

        if amount <= self.balance:

            self.balance -= amount

            return "Withdrawal successful."

        else:

            return "Insufficient balance."


account = BankAccount(1000)

print(
    account.withdraw(300)
)

print(account.balance)


# ============================================================
# 23. METHOD WITH LOOP
# ============================================================

class DataAnalyzer:

    def __init__(self, values):

        self.values = values

    def calculate_total(self):

        total = 0

        for value in self.values:

            total += value

        return total


analyzer = DataAnalyzer(
    [10, 20, 30, 40]
)

print(
    analyzer.calculate_total()
)


# ============================================================
# 24. DATA ANALYTICS CLASS
# ============================================================

class DataAnalyzer:

    def __init__(self, values):

        self.values = values

    def mean(self):

        return (
            sum(self.values)
            / len(self.values)
        )

    def minimum(self):

        return min(self.values)

    def maximum(self):

        return max(self.values)

    def count(self):

        return len(self.values)


data = DataAnalyzer(
    [10, 20, 30, 40, 50]
)

print("Mean:", data.mean())
print("Minimum:", data.minimum())
print("Maximum:", data.maximum())
print("Count:", data.count())


# ============================================================
# 25. DATA ANALYTICS CLASS — DESCRIPTIVE STATISTICS
# ============================================================

class Statistics:

    def __init__(self, values):

        self.values = values

    def mean(self):

        return (
            sum(self.values)
            / len(self.values)
        )

    def range(self):

        return (
            max(self.values)
            - min(self.values)
        )

    def summary(self):

        return {
            "count": len(self.values),
            "mean": self.mean(),
            "minimum": min(self.values),
            "maximum": max(self.values),
            "range": self.range()
        }


data = Statistics(
    [10, 20, 30, 40, 50]
)

print(data.summary())


# ============================================================
# 26. __str__()
# ============================================================
#
# __str__() controls how an object is represented
# when print() is used.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

    def __str__(self):

        return (
            f"Student: {self.name}, "
            f"Age: {self.age}"
        )


student = Student(
    "Yasin",
    35
)

print(student)


# ============================================================
# 27. __repr__()
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

    def __repr__(self):

        return (
            f"Student("
            f"name='{self.name}', "
            f"age={self.age})"
        )


student = Student(
    "Yasin",
    35
)

print(student)


# ============================================================
# 28. __len__()
# ============================================================

class Dataset:

    def __init__(self, data):

        self.data = data

    def __len__(self):

        return len(self.data)


dataset = Dataset(
    [10, 20, 30, 40]
)

print(len(dataset))


# ============================================================
# 29. __getitem__()
# ============================================================
#
# Allows object[index].
#
# ============================================================

class Dataset:

    def __init__(self, data):

        self.data = data

    def __getitem__(self, index):

        return self.data[index]


dataset = Dataset(
    [10, 20, 30, 40]
)

print(dataset[0])
print(dataset[2])


# ============================================================
# 30. ENCAPSULATION
# ============================================================
#
# Encapsulation means controlling how internal
# data is accessed or modified.
#
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self._balance = balance

    def get_balance(self):

        return self._balance


account = BankAccount(1000)

print(
    account.get_balance()
)


# ============================================================
# 31. PRIVATE-STYLE ATTRIBUTE
# ============================================================
#
# __name uses name mangling.
#
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.__balance = balance

    def get_balance(self):

        return self.__balance


account = BankAccount(1000)

print(
    account.get_balance()
)


# ============================================================
# 32. PROPERTY
# ============================================================
#
# @property allows a method to behave like an attribute.
#
# ============================================================

class Employee:

    def __init__(self, salary):

        self._salary = salary

    @property
    def salary(self):

        return self._salary


employee = Employee(70000)

print(employee.salary)


# ============================================================
# 33. PROPERTY WITH SETTER
# ============================================================

class Employee:

    def __init__(self, salary):

        self._salary = salary

    @property
    def salary(self):

        return self._salary

    @salary.setter
    def salary(self, value):

        if value >= 0:

            self._salary = value

        else:

            raise ValueError(
                "Salary cannot be negative."
            )


employee = Employee(70000)

employee.salary = 80000

print(employee.salary)


# ============================================================
# 34. INHERITANCE
# ============================================================
#
# A child class can inherit from a parent class.
#
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name

    def introduce(self):

        return f"My name is {self.name}."


class Student(Person):

    pass


student = Student("Yasin")

print(student.name)
print(student.introduce())


# ============================================================
# 35. CHILD CLASS WITH ADDITIONAL ATTRIBUTE
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name


class Student(Person):

    def __init__(
        self,
        name,
        field
    ):

        super().__init__(name)

        self.field = field


student = Student(
    "Yasin",
    "Machine Learning"
)

print(student.name)
print(student.field)


# ============================================================
# 36. super()
# ============================================================
#
# super() accesses functionality from the parent class.
#
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name


class Researcher(Person):

    def __init__(
        self,
        name,
        field
    ):

        super().__init__(name)

        self.field = field


researcher = Researcher(
    "Yasin",
    "Artificial Intelligence"
)

print(researcher.name)
print(researcher.field)


# ============================================================
# 37. METHOD OVERRIDING
# ============================================================

class Animal:

    def sound(self):

        return "Some sound"


class Dog(Animal):

    def sound(self):

        return "Bark"


class Cat(Animal):

    def sound(self):

        return "Meow"


dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound())


# ============================================================
# 38. POLYMORPHISM
# ============================================================
#
# Different objects can respond to the same method name
# in different ways.
#
# ============================================================

animals = [
    Dog(),
    Cat()
]

for animal in animals:

    print(animal.sound())


# ============================================================
# 39. POLYMORPHISM — DATA ANALYTICS EXAMPLE
# ============================================================

class CSVData:

    def load(self):

        return "Loading CSV data"


class SQLData:

    def load(self):

        return "Loading SQL data"


class APIData:

    def load(self):

        return "Loading API data"


data_sources = [
    CSVData(),
    SQLData(),
    APIData()
]

for source in data_sources:

    print(source.load())


# ============================================================
# 40. ABSTRACT BASE CLASS
# ============================================================
#
# An abstract class defines a common interface.
#
# ============================================================

from abc import ABC, abstractmethod


class DataSource(ABC):

    @abstractmethod
    def load(self):

        pass


class CSVData(DataSource):

    def load(self):

        return "CSV data loaded"


class SQLData(DataSource):

    def load(self):

        return "SQL data loaded"


csv = CSVData()
sql = SQLData()

print(csv.load())
print(sql.load())


# ============================================================
# 41. CLASS METHOD
# ============================================================
#
# @classmethod receives cls instead of self.
#
# ============================================================

class Employee:

    company = "AI Lab"

    def __init__(self, name):

        self.name = name

    @classmethod
    def change_company(
        cls,
        new_company
    ):

        cls.company = new_company


print(Employee.company)

Employee.change_company(
    "Data Science Lab"
)

print(Employee.company)


# ============================================================
# 42. STATIC METHOD
# ============================================================
#
# A static method does not need self or cls.
#
# ============================================================

class Calculator:

    @staticmethod
    def add(a, b):

        return a + b


print(
    Calculator.add(
        10,
        20
    )
)


# ============================================================
# 43. CLASS METHOD VS STATIC METHOD
# ============================================================

class Example:

    class_variable = 100

    def __init__(self, value):

        self.value = value

    @classmethod
    def show_class_variable(cls):

        return cls.class_variable

    @staticmethod
    def multiply(a, b):

        return a * b


example = Example(50)

print(
    example.show_class_variable()
)

print(
    example.multiply(
        10,
        20
    )
)


# ============================================================
# 44. COMPOSITION
# ============================================================
#
# One class can contain an object of another class.
#
# ============================================================

class Engine:

    def start(self):

        return "Engine started"


class Car:

    def __init__(self):

        self.engine = Engine()

    def start(self):

        return self.engine.start()


car = Car()

print(car.start())


# ============================================================
# 45. DATA ANALYTICS — DATASET CLASS
# ============================================================

class Dataset:

    def __init__(self, name, data):

        self.name = name
        self.data = data

    def count(self):

        return len(self.data)

    def mean(self):

        return (
            sum(self.data)
            / len(self.data)
        )

    def minimum(self):

        return min(self.data)

    def maximum(self):

        return max(self.data)


dataset = Dataset(
    "Sales",
    [100, 200, 300, 400, 500]
)

print("Dataset:", dataset.name)
print("Count:", dataset.count())
print("Mean:", dataset.mean())
print("Minimum:", dataset.minimum())
print("Maximum:", dataset.maximum())


# ============================================================
# 46. DATA ANALYTICS — DATA CLEANING CLASS
# ============================================================

class DataCleaner:

    def __init__(self, data):

        self.data = data

    def remove_missing(self):

        self.data = [
            value
            for value in self.data
            if value is not None
        ]

        return self.data

    def remove_duplicates(self):

        self.data = list(
            set(self.data)
        )

        return self.data


data = [
    10,
    20,
    None,
    20,
    30,
    None
]

cleaner = DataCleaner(data)

print(
    cleaner.remove_missing()
)

print(
    cleaner.remove_duplicates()
)


# ============================================================
# 47. DATA ANALYTICS — DATASET SUMMARY CLASS
# ============================================================

class DatasetSummary:

    def __init__(self, values):

        self.values = values

    def summary(self):

        return {

            "count": len(self.values),

            "mean": (
                sum(self.values)
                / len(self.values)
            ),

            "minimum": min(self.values),

            "maximum": max(self.values)
        }


summary = DatasetSummary(
    [10, 20, 30, 40, 50]
)

print(
    summary.summary()
)


# ============================================================
# 48. MACHINE LEARNING MODEL CLASS
# ============================================================

class SimpleModel:

    def __init__(self, name):

        self.name = name
        self.is_trained = False

    def train(self, X, y):

        print(
            "Training",
            self.name
        )

        self.is_trained = True

    def predict(self, X):

        if not self.is_trained:

            raise ValueError(
                "Model has not been trained."
            )

        return [
            0
            for _ in X
        ]


model = SimpleModel(
    "My ML Model"
)

X_train = [
    [1, 2],
    [3, 4],
    [5, 6]
]

y_train = [
    0,
    1,
    1
]

model.train(
    X_train,
    y_train
)

predictions = model.predict(
    [
        [7, 8],
        [9, 10]
    ]
)

print(predictions)


# ============================================================
# 49. MACHINE LEARNING — MODEL INHERITANCE
# ============================================================

class Model:

    def train(self, X, y):

        raise NotImplementedError


class ClassificationModel(Model):

    def train(self, X, y):

        return "Classification model trained"


class RegressionModel(Model):

    def train(self, X, y):

        return "Regression model trained"


classification = ClassificationModel()
regression = RegressionModel()

print(
    classification.train(
        X_train,
        y_train
    )
)

print(
    regression.train(
        X_train,
        y_train
    )
)


# ============================================================
# 50. OOP DATA ANALYTICS PROJECT STRUCTURE
# ============================================================

class DataPipeline:

    def __init__(self, data):

        self.data = data

    def clean(self):

        # Cleaning operations
        return self

    def transform(self):

        # Transformation operations
        return self

    def analyze(self):

        # Analysis operations
        return self

    def report(self):

        return "Analysis completed"


pipeline = DataPipeline(
    [10, 20, 30, 40]
)

result = (
    pipeline
    .clean()
    .transform()
    .analyze()
    .report()
)

print(result)
```

```python
# ============================================================
# PYTHON CLASSES & OBJECT-ORIENTED PROGRAMMING (OOP)
# COMPLETE LEARNING GUIDE
# ============================================================
#
# This guide covers:
#
#     Classes
#     Objects
#     Instances
#     Attributes
#     Methods
#     self
#     __init__
#     instance state
#     class attributes
#     instance methods
#     class methods
#     static methods
#     properties
#     getters/setters
#     encapsulation
#     inheritance
#     method overriding
#     super()
#     polymorphism
#     abstraction
#     abstract base classes
#     composition
#     aggregation
#     association
#     dataclasses
#     special/dunder methods
#     operator overloading
#     protocols
#     iterators
#     context managers
#     callable objects
#     class variables
#     class namespaces
#     object identity
#     mutability
#     inheritance design
#     multiple inheritance
#     MRO
#     mixins
#     dependency injection
#     OOP design patterns
#     data-analysis applications
#     ML applications
#     common mistakes
#     practice roadmap
#
#
# The central idea:
#
#     CLASS
#       ↓
#     blueprint / definition
#       ↓
#     OBJECT
#       ↓
#     actual instance
#
#
# Python feeling:
#
#     "A class describes what something IS
#      and what that thing CAN DO."
#
#
# ============================================================
# 1. THE BIG PICTURE
# ============================================================
#
# OOP organizes a program around OBJECTS.
#
# An object usually contains:
#
#     STATE
#       +
#     BEHAVIOR
#
#
# STATE:
#
#     data belonging to the object
#
#
# BEHAVIOR:
#
#     operations the object can perform
#
#
# Example:
#
#     Student
#
#     STATE:
#         name
#         age
#         university
#         scores
#
#     BEHAVIOR:
#         study()
#         calculate_average()
#         introduce()
#
#
# Diagram:
#
#                 STUDENT OBJECT
#          ┌─────────────────────────┐
#          │ STATE                   │
#          │                         │
#          │ name = "Yasin"          │
#          │ age = 30                │
#          │ scores = [80, 90, 95]   │
#          │                         │
#          │ BEHAVIOR                │
#          │                         │
#          │ study()                 │
#          │ average()               │
#          │ introduce()             │
#          └─────────────────────────┘
#
#
# ============================================================
# 2. WHY OOP?
# ============================================================
#
# Without OOP, a large program can become:
#
#     data
#     ↓
#     many unrelated variables
#     ↓
#     many functions
#     ↓
#     difficult relationships
#
#
# OOP groups related data and behavior:
#
#     OBJECT
#        ├── data
#        └── behavior
#
#
# OOP can provide:
#
#     organization
#     encapsulation
#     reuse
#     abstraction
#     polymorphism
#     extensibility
#     maintainability
#
#
# Important:
#
#     OOP is a tool.
#
#     Not every Python problem needs a class.
#
# Often:
#
#     simple data → list/dict
#     simple transformation → function
#     complex state + behavior → class
#
#
# ============================================================
# 3. CLASS VS OBJECT
# ============================================================
#
# CLASS:
#
#     definition / blueprint
#
#
# OBJECT:
#
#     actual instance created from the class
#
#
# Analogy:
#
#     CLASS
#       ↓
#     house blueprint
#
#     OBJECT
#       ↓
#     actual house
#
#
# Another analogy:
#
#     CLASS = type of thing
#     OBJECT = particular thing
#
#
# ============================================================
# 4. YOUR FIRST CLASS
# ============================================================

class Student:
    pass


# Student is now a class.
#
#
# Create an object:
#
student1 = Student()
student2 = Student()


# Diagram:
#
#              Student CLASS
#                   │
#          ┌────────┴────────┐
#          ↓                 ↓
#      student1          student2
#      OBJECT            OBJECT
#
#
# They are separate objects created
# from the same class.


# ============================================================
# 5. isinstance()
# ============================================================

print(isinstance(student1, Student))
print(isinstance(student2, Student))


# isinstance(object, class)
#
# asks:
#
#     "Is this object an instance of this class?"
#
#
# Result:
#
#     True
#     True


# ============================================================
# 6. type()
# ============================================================

print(type(student1))
print(type(student2))


# type(student1)
#
# tells us the object's type/class.


# ============================================================
# 7. OBJECT IDENTITY
# ============================================================

print(id(student1))
print(id(student2))


# Each object has an identity.
#
# The exact number is an implementation detail.
#
# What matters:
#
#     student1 is student2
#
# is False because they are different objects.


print(student1 is student2)


# ============================================================
# 8. CLASS ATTRIBUTES
# ============================================================

class Student:
    university = "IAU"


student1 = Student()
student2 = Student()

print(student1.university)
print(student2.university)


# university is a CLASS ATTRIBUTE.
#
#
# Diagram:
#
#              Student CLASS
#                   │
#          university = "IAU"
#                   │
#             ┌─────┴─────┐
#             ↓           ↓
#         student1     student2
#             │           │
#             └─────┬─────┘
#                   ↓
#            can access university
#
#
# Class attributes are shared conceptually
# through the class.


# ============================================================
# 9. INSTANCE ATTRIBUTES
# ============================================================

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Yasin", 30)
student2 = Student("Ali", 25)

print(student1.name)
print(student2.name)


# name and age are INSTANCE ATTRIBUTES.
#
#
# Diagram:
#
#     student1
#     ┌──────────────┐
#     │ name = Yasin │
#     │ age = 30     │
#     └──────────────┘
#
#
#     student2
#     ┌──────────────┐
#     │ name = Ali   │
#     │ age = 25     │
#     └──────────────┘
#
#
# Each object has its own instance state.


# ============================================================
# 10. __init__()
# ============================================================
#
# __init__ is an initializer method.
#
# It runs after an instance has been created
# and initializes its state.
#
#
# Common beginner wording:
#
#     "__init__ creates the object"
#
# More precisely:
#
#     object creation is handled by __new__()
#     initialization is handled by __init__()
#
#
# For normal Python programming,
# __init__ is where you usually initialize attributes.


# ============================================================
# 11. self
# ============================================================

class Student:

    def __init__(self, name):
        self.name = name


student = Student("Yasin")


# self means:
#
#     "this particular instance"
#
#
# When we write:
#
#     student.name
#
# we are accessing the name belonging to student.
#
#
# Diagram:
#
#     Student("Yasin")
#            ↓
#        self = student
#            ↓
#     self.name = "Yasin"


# ============================================================
# 12. self IS NOT A SPECIAL KEYWORD
# ============================================================
#
# self is a naming convention.
#
# Python does not require the parameter to literally
# be named self.
#
# But:
#
#     ALWAYS use self
#
# for normal instance methods.
#
#
# Correct style:
#
#     def method(self):
#         ...
#
#
# Technically this:
#
#     def method(instance):
#         ...
#
# can work,
# but it is strongly against normal Python convention.


# ============================================================
# 13. INSTANCE METHOD
# ============================================================

class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}"


student = Student("Yasin")

print(student.introduce())


# introduce() is an INSTANCE METHOD.
#
#
# Call:
#
#     student.introduce()
#
#
# Conceptually Python supplies the instance:
#
#     Student.introduce(student)
#
#
# Therefore:
#
#     self → student


# ============================================================
# 14. METHODS WITH MORE PARAMETERS
# ============================================================

class Calculator:

    def add(self, a, b):
        return a + b


calculator = Calculator()

print(calculator.add(10, 20))


# Parameters:
#
#     self
#     a
#     b
#
#
# Arguments supplied explicitly:
#
#     10
#     20
#
#
# self is supplied automatically through
# normal instance method binding.


# ============================================================
# 15. OBJECT STATE
# ============================================================

class BankAccount:

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount


account = BankAccount("Yasin", 1000)

account.deposit(500)
account.withdraw(200)

print(account.balance)


# State:
#
#     owner
#     balance
#
#
# Behavior:
#
#     deposit()
#     withdraw()
#
#
# This is one of the main reasons classes are useful:
#
#     data + operations that logically belong together


# ============================================================
# 16. OBJECT STATE DIAGRAM
# ============================================================
#
#     BankAccount
#     ┌────────────────────────┐
#     │ owner = "Yasin"        │
#     │ balance = 1300         │
#     ├────────────────────────┤
#     │ deposit()              │
#     │ withdraw()             │
#     └────────────────────────┘
#
#
# deposit(500):
#
#     1000 → 1500
#
#
# withdraw(200):
#
#     1500 → 1300


# ============================================================
# 17. MULTIPLE OBJECTS
# ============================================================

class BankAccount:

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


account1 = BankAccount("Yasin", 1000)
account2 = BankAccount("Ali", 500)

account1.deposit(200)

print(account1.balance)
print(account2.balance)


# account1 changes.
# account2 does not.
#
#
# Diagram:
#
#     account1 ──→ balance = 1200
#
#     account2 ──→ balance = 500


# ============================================================
# 18. CLASS ATTRIBUTE VS INSTANCE ATTRIBUTE
# ============================================================

class Student:

    university = "IAU"

    def __init__(self, name):
        self.name = name


student = Student("Yasin")

print(student.university)
print(student.name)


# university:
#
#     class-level
#
#
# name:
#
#     instance-level


# ============================================================
# 19. CLASS ATTRIBUTE SHARING
# ============================================================

class Student:

    university = "IAU"


student1 = Student()
student2 = Student()

print(student1.university)
print(student2.university)


# Both can access the same class attribute.


# ============================================================
# 20. INSTANCE ATTRIBUTE CAN OVERRIDE LOOKUP
# ============================================================

class Student:

    university = "IAU"


student1 = Student()
student2 = Student()

student1.university = "Other University"

print(student1.university)
print(student2.university)
print(Student.university)


# Result conceptually:
#
#     student1.university → "Other University"
#     student2.university → "IAU"
#     Student.university   → "IAU"
#
#
# student1 now has its own instance attribute
# with the same name.
#
#
# Lookup:
#
#     student1
#        ↓
#     instance attribute?
#        ↓
#     yes → use it
#
# If not:
#
#     class
#        ↓
#     class attribute


# ============================================================
# 21. __dict__
# ============================================================

class Student:

    university = "IAU"

    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Yasin", 30)

print(student.__dict__)
print(Student.__dict__)


# __dict__ exposes the namespace mapping
# for many normal Python objects/classes.
#
#
# student.__dict__ approximately contains:
#
#     {
#         "name": "Yasin",
#         "age": 30
#     }
#
#
# Class __dict__ contains class-level definitions
# and metadata.


# ============================================================
# 22. ATTRIBUTE LOOKUP
# ============================================================
#
# When evaluating:
#
#     student.name
#
# Python performs attribute lookup through
# the object's class machinery.
#
#
# Simplified mental model:
#
#     student
#       ↓
#     instance attributes
#       ↓
#     class
#       ↓
#     base classes
#       ↓
#     inherited attributes
#
#
# Python's actual lookup rules are more sophisticated,
# especially with descriptors, properties, and inheritance.
#
# For learning, the above model is the useful starting point.


# ============================================================
# 23. METHODS ARE ATTRIBUTES TOO
# ============================================================

class Student:

    def greet(self):
        return "Hello"


student = Student()

print(student.greet)


# greet is an attribute that resolves to a bound method.
#
#
# Calling:
#
#     student.greet()
#
# invokes it.


# ============================================================
# 24. BOUND METHOD
# ============================================================

class Student:

    def greet(self):
        return "Hello"


student = Student()

method = student.greet

print(method())


# student.greet is a BOUND METHOD.
#
# It remembers the instance.
#
#
# Conceptually:
#
#     student.greet
#          ↓
#     bound method
#          ↓
#     self = student


# ============================================================
# 25. EXPLICIT METHOD CALL
# ============================================================

class Calculator:

    def add(self, a, b):
        return a + b


calculator = Calculator()

print(calculator.add(10, 20))

print(Calculator.add(calculator, 10, 20))


# These produce the same result.
#
#
# Normal style:
#
#     calculator.add(10, 20)
#
#
# Lower-level explicit form:
#
#     Calculator.add(calculator, 10, 20)
#
#
# The first form is the normal Python style.


# ============================================================
# 26. METHODS THAT MODIFY STATE
# ============================================================

class Counter:

    def __init__(self):
        self.value = 0

    def increase(self):
        self.value += 1

    def decrease(self):
        self.value -= 1


counter = Counter()

counter.increase()
counter.increase()
counter.decrease()

print(counter.value)


# Methods can change instance state.


# ============================================================
# 27. METHODS THAT RETURN VALUES
# ============================================================

class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


rectangle = Rectangle(10, 5)

print(rectangle.area())


# area() does not need to modify state.
#
# It reads state and calculates a result.


# ============================================================
# 28. METHODS THAT BOTH MODIFY AND RETURN
# ============================================================

class Account:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance


account = Account(100)

new_balance = account.deposit(50)

print(new_balance)


# A method can:
#
#     modify state
#     return a value
#
# But make the behavior clear.


# ============================================================
# 29. CLASS METHODS
# ============================================================

class Student:

    university = "IAU"

    @classmethod
    def get_university(cls):
        return cls.university


print(Student.get_university())


# @classmethod changes how the method is bound.
#
# It receives:
#
#     cls
#
# instead of:
#
#     self
#
#
# cls refers to the class.


# ============================================================
# 30. cls
# ============================================================

class Student:

    university = "IAU"

    @classmethod
    def show_class(cls):
        print(cls)
        print(cls.university)


Student.show_class()


# cls → Student class.


# ============================================================
# 31. WHEN TO USE CLASSMETHOD
# ============================================================
#
# Useful for:
#
#     class-level behavior
#     alternative constructors
#     operations using class state
#     factory methods
#
#
# Typical pattern:
#
#     @classmethod
#     def from_something(cls, data):
#         return cls(...)
#
#
# This is one of the most useful classmethod patterns.


# ============================================================
# 32. ALTERNATIVE CONSTRUCTOR
# ============================================================

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, text):

        name, age = text.split(",")

        return cls(name, int(age))


student = Student.from_string("Yasin,30")

print(student.name)
print(student.age)


# Normal constructor:
#
#     Student("Yasin", 30)
#
#
# Alternative constructor:
#
#     Student.from_string("Yasin,30")
#
#
# from_string() creates the same kind of object
# using another input format.


# ============================================================
# 33. STATIC METHODS
# ============================================================

class MathTools:

    @staticmethod
    def square(number):
        return number ** 2


print(MathTools.square(5))


# A static method receives neither:
#
#     self
#     cls
#
# automatically.
#
#
# It is essentially a utility function grouped
# inside a class namespace.


# ============================================================
# 34. INSTANCE / CLASS / STATIC METHOD TABLE
# ============================================================
#
#     Type             Automatic argument
#     -----------------------------------------
#     instance         self
#     classmethod      cls
#     staticmethod     none
#
#
#     Instance method:
#
#         def method(self):
#
#
#     Class method:
#
#         @classmethod
#         def method(cls):
#
#
#     Static method:
#
#         @staticmethod
#         def method():
#
#
# Decision:
#
#     needs object state?
#         → instance method
#
#     needs class state / constructor?
#         → classmethod
#
#     logically belongs to class but needs
#     neither object nor class state?
#         → staticmethod


# ============================================================
# 35. PROPERTY
# ============================================================

class Circle:

    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14159 * self.radius ** 2


circle = Circle(10)

print(circle.area)


# Notice:
#
#     circle.area
#
# not:
#
#     circle.area()
#
#
# @property lets a method behave like
# a calculated attribute.


# ============================================================
# 36. PROPERTY MENTAL MODEL
# ============================================================
#
# Without property:
#
#     circle.calculate_area()
#
#
# With property:
#
#     circle.area
#
#
# The interface looks like data,
# while the implementation performs computation.


# ============================================================
# 37. PROPERTY WITH SETTER
# ============================================================

class Person:

    def __init__(self, age):
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):

        if value < 0:
            raise ValueError("Age cannot be negative")

        self._age = value


person = Person(30)

print(person.age)

person.age = 31

print(person.age)


# Property gives us controlled access.
#
#
#     person.age = 31
#             ↓
#        age.setter
#             ↓
#        validation
#             ↓
#        self._age = 31


# ============================================================
# 38. PROPERTY GETTER / SETTER
# ============================================================
#
# Conceptual flow:
#
#     READ
#
#     person.age
#         ↓
#     @property
#         ↓
#     getter
#
#
#     WRITE
#
#     person.age = 31
#         ↓
#     @age.setter
#         ↓
#     validation
#         ↓
#     internal storage
#
#
# This is useful when an attribute requires
# validation or computed behavior.


# ============================================================
# 39. UNDERSCORE CONVENTIONS
# ============================================================
#
#     name
#         normal public attribute
#
#     _name
#         internal-use convention
#
#     __name
#         name mangling inside classes
#
#     __name__
#         special/dunder name
#
#
# Important:
#
# Python does not have traditional strict private fields
# like some languages.
#
#     _name
#
# mostly communicates:
#
#     "This is an internal implementation detail."


# ============================================================
# 40. NAME MANGLING
# ============================================================

class Example:

    def __init__(self):
        self.__value = 100


example = Example()

print(example.__dict__)


# __value is name-mangled approximately to:
#
#     _Example__value
#
#
# This is mainly useful to reduce accidental name collisions
# in subclasses.
#
# It is NOT a security mechanism.


# ============================================================
# 41. ENCAPSULATION
# ============================================================
#
# Encapsulation means organizing and controlling
# access to an object's internal state and behavior.
#
#
# Example:
#
#     BankAccount
#
# Instead of allowing arbitrary invalid state:
#
#     balance = -999999
#
# methods/properties can enforce rules.
#
#
# Diagram:
#
#     external code
#          ↓
#     public interface
#          ↓
#     validation / rules
#          ↓
#     internal state


# ============================================================
# 42. INHERITANCE
# ============================================================
#
# Inheritance allows a class to derive from another class.
#
#
#     Parent
#       │
#       ├── shared behavior
#       └── shared attributes
#              ↓
#          Child class
#
#
# Example:
#
#     Animal
#       ↓
#     Dog
#
#
# Dog is an Animal.


# ============================================================
# 43. BASIC INHERITANCE
# ============================================================

class Animal:

    def speak(self):
        return "Some sound"


class Dog(Animal):
    pass


dog = Dog()

print(dog.speak())


# Dog inherits speak() from Animal.


# ============================================================
# 44. INHERITANCE DIAGRAM
# ============================================================
#
#             Animal
#               │
#          ┌────┴────┐
#          ↓         ↓
#         Dog       Cat
#
#
# Animal:
#
#     speak()
#
#
# Dog inherits speak()
# Cat inherits speak()


# ============================================================
# 45. ADDING CHILD-SPECIFIC METHODS
# ============================================================

class Animal:

    def eat(self):
        return "Eating"


class Dog(Animal):

    def bark(self):
        return "Woof"


dog = Dog()

print(dog.eat())
print(dog.bark())


# Dog has:
#
#     inherited behavior
#     own behavior


# ============================================================
# 46. METHOD OVERRIDING
# ============================================================

class Animal:

    def speak(self):
        return "Some sound"


class Dog(Animal):

    def speak(self):
        return "Woof"


dog = Dog()

print(dog.speak())


# Dog overrides the inherited speak() implementation.


# ============================================================
# 47. OVERRIDING DIAGRAM
# ============================================================
#
#             Animal
#               │
#           speak()
#           "Some sound"
#               │
#               ↓
#             Dog
#               │
#           speak()
#             "Woof"
#
#
# When dog.speak() is called:
#
#     Python finds Dog.speak first.


# ============================================================
# 48. super()
# ============================================================

class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed


dog = Dog("Max", "Labrador")

print(dog.name)
print(dog.breed)


# super() gives access to behavior
# from the next class in the method resolution order.
#
#
# Here:
#
#     super().__init__(name)
#
# initializes the inherited part of the object.


# ============================================================
# 49. WHY super() IS USEFUL
# ============================================================
#
# Without super():
#
#     child may duplicate parent initialization logic.
#
#
# With super():
#
#     parent initialization
#          ↓
#     child initialization
#
#
# This reduces duplication and works correctly
# with cooperative multiple inheritance when designed properly.


# ============================================================
# 50. INHERITANCE + ADDITIONAL STATE
# ============================================================

class Person:

    def __init__(self, name):
        self.name = name


class Student(Person):

    def __init__(self, name, university):
        super().__init__(name)
        self.university = university


student = Student("Yasin", "IAU")

print(student.name)
print(student.university)


# Person:
#
#     name
#
#
# Student adds:
#
#     university


# ============================================================
# 51. POLYMORPHISM
# ============================================================
#
# Polymorphism means:
#
#     "different objects can respond
#      to the same operation in different ways."
#
#
# Example:
#
#     dog.speak()
#     cat.speak()
#
# Same interface:
#
#     speak()
#
# Different behavior.


# ============================================================
# 52. POLYMORPHISM EXAMPLE
# ============================================================

class Dog:

    def speak(self):
        return "Woof"


class Cat:

    def speak(self):
        return "Meow"


animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())


# The loop does not need:
#
#     if isinstance(animal, Dog):
#         ...
#
#     elif isinstance(animal, Cat):
#         ...
#
#
# It simply asks:
#
#     animal.speak()
#
#
# This is a major Python concept.


# ============================================================
# 53. DUCK TYPING
# ============================================================
#
# Python often cares less about:
#
#     "What exact class is this?"
#
# and more about:
#
#     "Can this object perform the operation I need?"
#
#
# Famous mental model:
#
#     "If it behaves like a duck,
#      use it like a duck."
#
#
# Example:

class FileLike:

    def write(self, text):
        print("Writing:", text)


class ConsoleLike:

    def write(self, text):
        print("Console:", text)


def send_message(writer):
    writer.write("Hello")


send_message(FileLike())
send_message(ConsoleLike())


# send_message() does not need a specific class.
#
# It needs an object with:
#
#     write()


# ============================================================
# 54. DUCK TYPING VS EXPLICIT TYPE CHECKING
# ============================================================
#
# Often avoid:
#
#     if isinstance(x, SomeClass):
#         ...
#
# when all you need is:
#
#     x.some_method()
#
#
# Prefer behavior-based design when appropriate.
#
#
# This is one reason Python's OOP can be flexible.


# ============================================================
# 55. ABSTRACT BASE CLASSES
# ============================================================
#
# Sometimes we want to define an interface:
#
#     "Every subclass must implement this behavior."
#
#
# Python provides:
#
#     abc
#     ABC
#     abstractmethod


# ============================================================
# 56. BASIC ABSTRACT CLASS
# ============================================================

from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def speak(self):
        pass


class Dog(Animal):

    def speak(self):
        return "Woof"


dog = Dog()

print(dog.speak())


# Animal defines the required interface.
#
# Dog supplies the implementation.


# ============================================================
# 57. ABSTRACT CLASS DIAGRAM
# ============================================================
#
#             Animal
#          ABSTRACT CLASS
#               │
#          ┌────┴────┐
#          ↓         ↓
#         Dog       Cat
#          │         │
#       speak()   speak()
#
#
# Animal says:
#
#     "You must provide speak()."


# ============================================================
# 58. WHY ABSTRACT CLASSES?
# ============================================================
#
# Useful when:
#
#     several classes share a common interface
#     you want enforceable structure
#     you are designing a framework/API
#     polymorphism is central
#
#
# For small scripts,
# ABCs may be unnecessary.


# ============================================================
# 59. COMPOSITION
# ============================================================
#
# Composition means:
#
#     an object contains another object.
#
#
# Example:
#
#     Car
#       contains
#     Engine
#
#
# Diagram:
#
#     Car
#      │
#      └────→ Engine
#
#
# This represents:
#
#     HAS-A
#
# rather than:
#
#     IS-A


# ============================================================
# 60. COMPOSITION EXAMPLE
# ============================================================

class Engine:

    def start(self):
        return "Engine started"


class Car:

    def __init__(self):
        self.engine = Engine()

    def start(self):
        return self.engine.start()


car = Car()

print(car.start())


# Car HAS an Engine.
#
#
# Composition:
#
#     Car
#      │
#      ↓
#     Engine
#
#
# This is often more flexible than inheritance.


# ============================================================
# 61. INHERITANCE VS COMPOSITION
# ============================================================
#
# INHERITANCE:
#
#     IS-A
#
#     Dog IS-A Animal
#
#
# COMPOSITION:
#
#     HAS-A
#
#     Car HAS-A Engine
#
#
# Table:
#
#     Relationship     Example
#     --------------------------------
#     IS-A              Dog → Animal
#     HAS-A             Car → Engine
#
#
# A useful design question:
#
#     "Is this genuinely a subtype,
#      or does this object simply use another object?"


# ============================================================
# 62. AGGREGATION
# ============================================================
#
# Aggregation is a weaker "has-a" relationship.
#
# The contained object can exist independently.
#
#
# Example:
#
#     Department
#         ↓
#       Teachers
#
# A Teacher can exist independently of a Department.


# ============================================================
# 63. AGGREGATION EXAMPLE
# ============================================================

class Teacher:

    def __init__(self, name):
        self.name = name


class Department:

    def __init__(self, teachers):
        self.teachers = teachers


teacher1 = Teacher("Ali")
teacher2 = Teacher("Sara")

department = Department(
    [teacher1, teacher2]
)


# Department uses Teacher objects,
# but does not necessarily own their entire lifetime.


# ============================================================
# 64. ASSOCIATION
# ============================================================
#
# Association simply means:
#
#     objects know about or interact with each other.
#
#
# Example:
#
#     Student
#         ↕
#     Course
#
#
# Neither necessarily contains the other as a component.


# ============================================================
# 65. RELATIONSHIP TABLE
# ============================================================
#
#     Relationship     Meaning
#     ---------------------------------------
#     Inheritance      IS-A
#     Composition      HAS-A, strong ownership
#     Aggregation      HAS-A, weaker ownership
#     Association      interacts with
#
#
# Do not obsess over rigid terminology.
# Focus on the actual object relationship.


# ============================================================
# 66. DATACLASSES
# ============================================================
#
# For classes that primarily store data,
# dataclasses can eliminate repetitive boilerplate.


from dataclasses import dataclass


@dataclass
class Student:

    name: str
    age: int
    score: float


student = Student("Yasin", 30, 92.5)

print(student)


# Dataclass automatically provides useful methods
# such as a readable __repr__ and __init__.


# ============================================================
# 67. DATACLASS MENTAL MODEL
# ============================================================
#
# Traditional:
#
#     class Student:
#         def __init__(...):
#             ...
#         def __repr__(...):
#             ...
#
#
# Dataclass:
#
#     @dataclass
#     class Student:
#         name: str
#         age: int
#         score: float
#
#
# Use dataclasses when the primary purpose
# is structured data with lightweight behavior.


# ============================================================
# 68. DATACLASS DEFAULT VALUES
# ============================================================

@dataclass
class Student:

    name: str
    age: int
    active: bool = True


student = Student("Yasin", 30)

print(student)


# Simple immutable defaults are straightforward.
#
# For mutable defaults, use field(default_factory=...).


# ============================================================
# 69. DATACLASS WITH default_factory
# ============================================================

from dataclasses import dataclass, field


@dataclass
class Student:

    name: str
    scores: list[float] = field(
        default_factory=list
    )


student1 = Student("Ali")
student2 = Student("Sara")

student1.scores.append(90)

print(student1)
print(student2)


# Each object receives its own list.
#
# This avoids shared mutable default state.


# ============================================================
# 70. SPECIAL METHODS / DUNDER METHODS
# ============================================================
#
# "Dunder" means:
#
#     double underscore
#
#
# Examples:
#
#     __init__
#     __str__
#     __repr__
#     __len__
#     __iter__
#     __next__
#     __getitem__
#     __setitem__
#     __contains__
#     __eq__
#     __lt__
#     __add__
#     __enter__
#     __exit__
#     __call__
#
#
# They allow objects to participate naturally
# in Python syntax and protocols.


# ============================================================
# 71. __str__
# ============================================================

class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"


student = Student("Yasin")

print(student)


# print(student)
#
# uses __str__() for human-readable output.


# ============================================================
# 72. __repr__
# ============================================================

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return (
            f"Student(name={self.name!r}, "
            f"age={self.age!r})"
        )


student = Student("Yasin", 30)

print(repr(student))


# __repr__ is intended to provide
# an unambiguous/developer-oriented representation.
#
#
# Good rule:
#
#     __str__ → user-friendly
#     __repr__ → developer/debugging-friendly


# ============================================================
# 73. __len__
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __len__(self):
        return len(self.rows)


dataset = Dataset([1, 2, 3, 4])

print(len(dataset))


# len(dataset)
#
# invokes the length protocol.


# ============================================================
# 74. __getitem__
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __getitem__(self, index):
        return self.rows[index]


dataset = Dataset([10, 20, 30])

print(dataset[0])
print(dataset[1:])


# This allows:
#
#     dataset[0]
#     dataset[1:]
#
# to work naturally.


# ============================================================
# 75. __setitem__
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __getitem__(self, index):
        return self.rows[index]

    def __setitem__(self, index, value):
        self.rows[index] = value


dataset = Dataset([10, 20, 30])

dataset[0] = 100

print(dataset[0])


# Enables:
#
#     dataset[index] = value


# ============================================================
# 76. __contains__
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __contains__(self, value):
        return value in self.rows


dataset = Dataset([10, 20, 30])

print(20 in dataset)
print(99 in dataset)


# Enables:
#
#     value in object


# ============================================================
# 77. __eq__
# ============================================================

class Student:

    def __init__(self, name):
        self.name = name

    def __eq__(self, other):

        if not isinstance(other, Student):
            return NotImplemented

        return self.name == other.name


student1 = Student("Yasin")
student2 = Student("Yasin")

print(student1 == student2)


# Without suitable equality behavior,
# two separate objects are not automatically equal
# merely because their attributes match.


# ============================================================
# 78. == VS is
# ============================================================
#
#     ==
#         value equality
#
#     is
#         object identity
#
#
# Example:
#
#     a = Student("Yasin")
#     b = Student("Yasin")
#
#     a == b
#         depends on __eq__
#
#     a is b
#         False
#
#
# "is" asks:
#
#     "Are these the exact same object?"
#
#
# "==" asks:
#
#     "Do these objects compare equal?"


# ============================================================
# 79. __lt__
# ============================================================

class Student:

    def __init__(self, score):
        self.score = score

    def __lt__(self, other):
        return self.score < other.score


students = [
    Student(90),
    Student(70),
    Student(85)
]

students.sort()

for student in students:
    print(student.score)


# __lt__ defines:
#
#     <
#
#
# Similar rich comparison methods include:
#
#     __lt__  <
#     __le__  <=
#     __eq__  ==
#     __ne__  !=
#     __gt__  >
#     __ge__  >=


# ============================================================
# 80. OPERATOR OVERLOADING
# ============================================================
#
# Python operators can be connected to special methods.
#
#
#     a + b
#         ↓
#     __add__()
#
#
#     a == b
#         ↓
#     __eq__()
#
#
#     len(a)
#         ↓
#     __len__()
#
#
#     a[0]
#         ↓
#     __getitem__()
#
#
# This allows user-defined objects to feel like
# normal Python objects.


# ============================================================
# 81. __add__
# ============================================================

class Vector:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(
            self.x + other.x,
            self.y + other.y
        )


v1 = Vector(1, 2)
v2 = Vector(3, 4)

v3 = v1 + v2

print(v3.x, v3.y)


# v1 + v2
#
# invokes __add__.


# ============================================================
# 82. __call__
# ============================================================
#
# __call__ makes an object callable.
#
#
# Normally:
#
#     function()
#
#
# With __call__:
#
#     object()
#
# can also work.


class Multiplier:

    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor


double = Multiplier(2)

print(double(10))


# double is an object,
# but behaves like a callable.


# ============================================================
# 83. CALLABLE OBJECT
# ============================================================
#
#     object
#        ↓
#     __call__()
#        ↓
#     object(...)
#
#
# This is useful for:
#
#     configurable operations
#     callbacks
#     stateful functions
#     machine-learning components
#     reusable transformations


# ============================================================
# 84. CUSTOM ITERABLE
# ============================================================

class Numbers:

    def __init__(self, values):
        self.values = values

    def __iter__(self):
        return iter(self.values)


numbers = Numbers([10, 20, 30])

for number in numbers:
    print(number)


# __iter__ allows:
#
#     for number in numbers
#
#
# The object becomes iterable.


# ============================================================
# 85. CUSTOM ITERATOR
# ============================================================

class Counter:

    def __init__(self, limit):
        self.current = 0
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):

        if self.current >= self.limit:
            raise StopIteration

        value = self.current

        self.current += 1

        return value


counter = Counter(3)

for number in counter:
    print(number)


# Protocol:
#
#     __iter__()
#         ↓
#     iterator
#         ↓
#     __next__()
#         ↓
#     value
#         ↓
#     StopIteration


# ============================================================
# 86. CONTEXT MANAGERS
# ============================================================
#
# Context managers power:
#
#     with ...
#
#
# Common example:
#
#     with open(...) as file:
#         ...
#
#
# Custom context managers implement:
#
#     __enter__()
#     __exit__()


# ============================================================
# 87. CUSTOM CONTEXT MANAGER
# ============================================================

class Resource:

    def __enter__(self):
        print("Resource opened")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Resource closed")


with Resource():
    print("Working")


# Flow:
#
#     with Resource()
#          ↓
#     __enter__()
#          ↓
#     body
#          ↓
#     __exit__()
#
#
# This pattern is useful for controlled setup
# and cleanup.


# ============================================================
# 88. MULTIPLE INHERITANCE
# ============================================================
#
# Python allows a class to inherit
# from multiple classes.
#
#
#     Class C
#       ↙   ↘
#      A     B
#
#
# C(A, B)


# ============================================================
# 89. MULTIPLE INHERITANCE EXAMPLE
# ============================================================

class Logger:

    def log(self, message):
        print("LOG:", message)


class Validator:

    def validate(self, value):
        return value is not None


class Processor(Logger, Validator):
    pass


processor = Processor()

processor.log("Starting")
print(processor.validate(10))


# Processor inherits from both classes.


# ============================================================
# 90. MRO — METHOD RESOLUTION ORDER
# ============================================================
#
# When multiple inheritance exists,
# Python needs an order for searching methods.
#
# This is the MRO.
#
#
# You can inspect it:
#
#     ClassName.mro()
#
# or:
#
#     ClassName.__mro__


class A:
    pass


class B(A):
    pass


print(B.mro())


# Simplified idea:
#
#     Python searches classes according to
#     the method resolution order.


# ============================================================
# 91. MRO DIAGRAM
# ============================================================
#
#     C
#     ↓
#     A
#     ↓
#     object
#
#
# With multiple inheritance:
#
#         C
#        / \
#       A   B
#        \ /
#       object
#
#
# Python computes a consistent MRO using
# the C3 linearization algorithm.


# ============================================================
# 92. super() AND MRO
# ============================================================
#
# Important:
#
#     super()
#
# does NOT simply mean:
#
#     "call my immediate parent"
#
#
# More accurately:
#
#     "continue method lookup according to the MRO."
#
#
# This distinction becomes important with
# multiple inheritance and cooperative classes.


# ============================================================
# 93. MIXINS
# ============================================================
#
# A mixin is a small class designed to add
# a focused capability to another class.
#
#
# Example:
#
#     LoggingMixin
#     ValidationMixin
#     SerializationMixin
#
#
# Mixins usually are not intended to represent
# standalone domain objects.


# ============================================================
# 94. MIXIN EXAMPLE
# ============================================================

class LoggingMixin:

    def log(self, message):
        print("LOG:", message)


class User(LoggingMixin):

    def __init__(self, name):
        self.name = name


user = User("Yasin")

user.log("User created")


# User gains logging capability
# through the mixin.


# ============================================================
# 95. COMPOSITION OF BEHAVIOR
# ============================================================
#
# Instead of:
#
#     class HugeClass:
#         logging
#         validation
#         storage
#         networking
#         calculations
#
#
# prefer small components:
#
#     Logger
#     Validator
#     Storage
#     Calculator
#
#
# and compose them where useful.


# ============================================================
# 96. OOP + FUNCTIONS
# ============================================================
#
# Classes and functions are not opposites.
#
#
# A class often contains:
#
#     methods
#
# and methods are functions associated
# with the class/object.
#
#
# Good OOP programs still use ordinary functions
# when a task does not require object state.


# ============================================================
# 97. WHEN SHOULD I USE A CLASS?
# ============================================================
#
# Consider a class when you have:
#
#     persistent state
#     several related operations
#     multiple instances
#     meaningful object identity
#     domain concepts
#     reusable components
#     lifecycle/state transitions
#
#
# Example:
#
#     BankAccount
#         balance
#         deposit()
#         withdraw()
#
#
# A simple calculation:
#
#     def calculate_average(numbers):
#
# may not need a class.


# ============================================================
# 98. CLASS VS FUNCTION DECISION TABLE
# ============================================================
#
#     Situation                         Use
#     --------------------------------------------
#     Simple calculation                function
#     Stateless transformation          function
#     Small utility                     function
#     Persistent state                  class
#     Multiple related operations       class
#     Many similar instances            class
#     Domain entity                     class
#     Data-only record                  dataclass
#     External resource lifecycle       class/context manager
#
#
# Do not use classes merely because
# "professional Python uses OOP."


# ============================================================
# 99. OBJECT LIFECYCLE
# ============================================================
#
# Simplified:
#
#     class definition
#          ↓
#     object construction
#          ↓
#     __new__()
#          ↓
#     __init__()
#          ↓
#     object used
#          ↓
#     object eventually becomes unreachable
#          ↓
#     cleanup / garbage collection
#
#
# Most everyday classes only need __init__.
#
# __new__ is an advanced customization point.


# ============================================================
# 100. __new__ — ADVANCED
# ============================================================

class Example:

    def __new__(cls):
        print("Creating instance")
        return super().__new__(cls)

    def __init__(self):
        print("Initializing instance")


example = Example()


# Output conceptually:
#
#     Creating instance
#     Initializing instance
#
#
# __new__ creates/returns the instance.
# __init__ initializes it.


# ============================================================
# 101. OBJECT DESTRUCTION
# ============================================================
#
# Python has:
#
#     __del__()
#
# but it should generally NOT be treated
# as a reliable resource-management mechanism.
#
#
# For files, locks, connections, etc.:
#
#     use context managers
#
# rather than relying on __del__.


# ============================================================
# 102. CLASS NAMESPACE
# ============================================================

class Student:

    university = "IAU"

    def greet(self):
        return "Hello"


print(Student.__dict__)


# A class has a namespace containing:
#
#     attributes
#     methods
#     descriptors
#     metadata
#
#
# Creating an instance then gives it
# its own instance namespace.


# ============================================================
# 103. SETTING ATTRIBUTES DYNAMICALLY
# ============================================================

class Student:
    pass


student = Student()

student.name = "Yasin"
student.age = 30

print(student.name)
print(student.age)


# Python objects can often receive attributes dynamically,
# depending on their class design.


# ============================================================
# 104. getattr()
# ============================================================

class Student:

    def __init__(self, name):
        self.name = name


student = Student("Yasin")

print(getattr(student, "name"))

print(getattr(student, "age", None))


# getattr(object, attribute_name, default)
#
# Useful when the attribute name is dynamic.


# ============================================================
# 105. setattr()
# ============================================================

student = Student("Yasin")

setattr(student, "age", 30)

print(student.age)


# Equivalent conceptually:
#
#     student.age = 30
#
#
# setattr() is useful when the attribute name
# comes from a variable.


# ============================================================
# 106. hasattr()
# ============================================================

student = Student("Yasin")

print(hasattr(student, "name"))
print(hasattr(student, "age"))


# Useful when checking whether an attribute exists.
#
# But often prefer directly attempting the required
# operation or using a well-defined interface.


# ============================================================
# 107. delattr()
# ============================================================

student = Student("Yasin")

student.age = 30

delattr(student, "age")

print(hasattr(student, "age"))


# Equivalent:
#
#     del student.age


# ============================================================
# 108. PROPERTY VS RAW ATTRIBUTE
# ============================================================
#
# Raw attribute:
#
#     student.age = 30
#
# direct state storage.
#
#
# Property:
#
#     student.age = 30
#
# may execute validation logic.
#
#
# The external interface can remain the same
# while the implementation becomes safer.


# ============================================================
# 109. DATA VALIDATION WITH OOP
# ============================================================

class Student:

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_valid(self):
        return 0 <= self.score <= 100


student = Student("Yasin", 92)

print(student.is_valid())


# Object combines:
#
#     data
#     validation behavior


# ============================================================
# 110. DATA ANALYSIS OBJECT
# ============================================================

class Dataset:

    def __init__(self, values):
        self.values = values

    def count(self):
        return len(self.values)

    def total(self):
        return sum(self.values)

    def average(self):

        if not self.values:
            return None

        return self.total() / self.count()


dataset = Dataset([10, 20, 30, 40])

print(dataset.count())
print(dataset.total())
print(dataset.average())


# This is a simple example of
# encapsulating related data operations.


# ============================================================
# 111. DATASET OBJECT DIAGRAM
# ============================================================
#
#     Dataset
#     ┌──────────────────────────┐
#     │ values = [10,20,30,40]   │
#     ├──────────────────────────┤
#     │ count()                  │
#     │ total()                  │
#     │ average()                │
#     └──────────────────────────┘
#
#
# Instead of:
#
#     len(values)
#     sum(values)
#     sum(values) / len(values)
#
# the object exposes domain-specific operations.
#
# But for a simple list,
# the built-in functions are often already enough.
#
# The class becomes more valuable as the domain logic grows.


# ============================================================
# 112. OOP FOR DATA CLEANING
# ============================================================

class DataCleaner:

    def clean_name(self, value):
        return value.strip().title()

    def clean_score(self, value):

        if value is None:
            return None

        try:
            return float(value)

        except (TypeError, ValueError):
            return None


cleaner = DataCleaner()

print(cleaner.clean_name(" yasin "))
print(cleaner.clean_score("92.5"))


# A class can group related cleaning operations.
#
# However, if these functions do not need state,
# ordinary functions may be simpler.


# ============================================================
# 113. STATEFUL DATA PROCESSOR
# ============================================================

class RunningAverage:

    def __init__(self):
        self.total = 0
        self.count = 0

    def add(self, value):
        self.total += value
        self.count += 1

    def average(self):

        if self.count == 0:
            return None

        return self.total / self.count


average = RunningAverage()

average.add(10)
average.add(20)
average.add(30)

print(average.average())


# This is a strong class use case:
#
#     the object maintains state
#     across multiple operations.


# ============================================================
# 114. STATE TRANSITION
# ============================================================
#
#     RunningAverage
#
#     Initial:
#         total = 0
#         count = 0
#
#     add(10):
#         total = 10
#         count = 1
#
#     add(20):
#         total = 30
#         count = 2
#
#     add(30):
#         total = 60
#         count = 3
#
#     average():
#         60 / 3 = 20


# ============================================================
# 115. OOP FOR ML COMPONENTS
# ============================================================
#
# Many ML libraries use classes for components such as:
#
#     models
#     transformers
#     preprocessors
#     pipelines
#
#
# Conceptually:
#
#     model.fit(X, y)
#          ↓
#     model state learned
#          ↓
#     model.predict(X_new)
#
#
# This is a perfect example of state + behavior.


# ============================================================
# 116. SIMPLE MODEL-LIKE OBJECT
# ============================================================

class ThresholdModel:

    def __init__(self, threshold):
        self.threshold = threshold

    def predict(self, values):

        return [
            1 if value >= self.threshold else 0
            for value in values
        ]


model = ThresholdModel(0.5)

print(model.predict([0.2, 0.8, 0.6, 0.1]))


# State:
#
#     threshold
#
#
# Behavior:
#
#     predict()


# ============================================================
# 117. FIT / PREDICT STYLE
# ============================================================

class SimpleMeanModel:

    def __init__(self):
        self.mean_ = None

    def fit(self, values):

        self.mean_ = sum(values) / len(values)

        return self

    def predict(self, values):

        if self.mean_ is None:
            raise ValueError("Model is not fitted")

        return [
            self.mean_
            for _ in values
        ]


model = SimpleMeanModel()

model.fit([10, 20, 30])

print(model.predict([1, 2, 3]))


# Notice:
#
#     fit()
#         ↓
#     object gains learned state
#         ↓
#     predict()
#         ↓
#     uses learned state
#
#
# This is the fundamental OOP idea behind
# many estimator APIs.


# ============================================================
# 118. METHOD CHAINING
# ============================================================

class Counter:

    def __init__(self, value=0):
        self.value = value

    def add(self, amount):
        self.value += amount
        return self

    def subtract(self, amount):
        self.value -= amount
        return self


counter = Counter()

counter.add(10).add(20).subtract(5)

print(counter.value)


# Returning self enables:
#
#     object.method().method().method()
#
#
# Use chaining carefully.
# Readability matters more than cleverness.


# ============================================================
# 119. IMMUTABLE-STYLE OBJECTS
# ============================================================
#
# Sometimes methods should return a NEW object
# instead of modifying the existing object.
#
#
# Mutable style:
#
#     object.change()
#
# modifies object.
#
#
# Immutable-style:
#
#     new_object = object.changed()
#
# leaves original unchanged.
#
#
# Both styles are useful.


# ============================================================
# 120. COPY OF OBJECT STATE
# ============================================================

from copy import copy, deepcopy


class Student:

    def __init__(self, name, scores):
        self.name = name
        self.scores = scores


student1 = Student(
    "Yasin",
    [80, 90]
)

student2 = copy(student1)

student2.name = "Ali"

print(student1.name)
print(student2.name)


# Shallow copy:
#
#     new outer object
#     shared nested references
#
#
# For nested mutable state,
# deepcopy() may be necessary.


# ============================================================
# 121. SHALLOW COPY DIAGRAM
# ============================================================
#
#     student1
#        │
#        ├── name ──→ "Yasin"
#        │
#        └── scores ──→ [80, 90]
#
#
#     student2
#        │
#        ├── name ──→ "Yasin"
#        │
#        └── scores ─────────┘
#
#
# Outer objects are different.
# Nested list can still be shared.


# ============================================================
# 122. INHERITANCE WITH PROPERTIES
# ============================================================

class Person:

    def __init__(self, name):
        self.name = name

    @property
    def description(self):
        return f"Person: {self.name}"


class Student(Person):

    @property
    def description(self):
        return f"Student: {self.name}"


student = Student("Yasin")

print(student.description)


# A child class can override properties too.


# ============================================================
# 123. POLYMORPHIC DATA PROCESSING
# ============================================================

class CSVExporter:

    def export(self, data):
        return "CSV DATA"


class JSONExporter:

    def export(self, data):
        return "JSON DATA"


def save(exporter, data):
    return exporter.export(data)


print(save(CSVExporter(), [1, 2, 3]))
print(save(JSONExporter(), [1, 2, 3]))


# save() does not care about the exact exporter class.
#
# It only needs:
#
#     export()


# ============================================================
# 124. PROTOCOL THINKING
# ============================================================
#
# Instead of thinking:
#
#     "I need a CSVExporter."
#
# think:
#
#     "I need something that can export()."
#
#
# This is protocol-oriented thinking
# and fits Python's duck typing.


# ============================================================
# 125. TYPING PROTOCOL
# ============================================================
#
# Python's typing module can formally describe
# behavior-based interfaces.


from typing import Protocol


class Exporter(Protocol):

    def export(self, data) -> str:
        ...


class CSVExporter:

    def export(self, data) -> str:
        return "CSV"


def save(exporter: Exporter, data):
    return exporter.export(data)


print(save(CSVExporter(), [1, 2, 3]))


# Protocols are mainly useful for:
#
#     static type checking
#     documenting interfaces
#     flexible architecture


# ============================================================
# 126. DATACLASS + METHODS
# ============================================================

@dataclass
class Student:

    name: str
    scores: list[float]

    def average(self):

        if not self.scores:
            return None

        return sum(self.scores) / len(self.scores)


student = Student(
    "Yasin",
    [80, 90, 95]
)

print(student.average())


# Dataclass does not prevent methods.
#
# It simply automates common data-container code.


# ============================================================
# 127. CLASS VARIABLES FOR SHARED STATE
# ============================================================

class User:

    count = 0

    def __init__(self, name):
        self.name = name
        User.count += 1


User("Ali")
User("Sara")
User("Yasin")

print(User.count)


# count belongs to the class.
#
# All instances contribute to it.


# ============================================================
# 128. CLASS ATTRIBUTE MUTATION TRAP
# ============================================================

class Team:

    members = []


team1 = Team()
team2 = Team()

team1.members.append("Ali")

print(team1.members)
print(team2.members)


# Both see the same list.
#
#
# Why?
#
#     members is one shared class-level list.
#
#
# This is often NOT what you want.


# ============================================================
# 129. CORRECT INSTANCE LIST
# ============================================================

class Team:

    def __init__(self):
        self.members = []


team1 = Team()
team2 = Team()

team1.members.append("Ali")

print(team1.members)
print(team2.members)


# Now each instance has its own list.


# ============================================================
# 130. CLASS ATTRIBUTE DECISION
# ============================================================
#
# Use class attributes for data genuinely shared:
#
#     DEFAULT_TIMEOUT = 30
#     organization_name = "IAU"
#
#
# Use instance attributes for per-object state:
#
#     self.name
#     self.age
#     self.balance
#     self.scores
#
#
# Warning:
#
#     mutable class attributes
#
# can accidentally create shared state.


# ============================================================
# 131. CONSTANTS IN CLASSES
# ============================================================

class Temperature:

    ABSOLUTE_ZERO_C = -273.15


print(Temperature.ABSOLUTE_ZERO_C)


# Uppercase communicates:
#
#     "Treat this as a constant."
#
# Python does not enforce true immutability
# merely because a name is uppercase.


# ============================================================
# 132. CLASS CONFIGURATION
# ============================================================

class DataProcessor:

    DEFAULT_MISSING_VALUE = None
    DEFAULT_THRESHOLD = 0.5

    def __init__(self, threshold=None):

        if threshold is None:
            threshold = self.DEFAULT_THRESHOLD

        self.threshold = threshold


processor = DataProcessor()

print(processor.threshold)


# Class attributes can provide defaults/configuration.


# ============================================================
# 133. FACTORY METHODS
# ============================================================

class User:

    def __init__(self, name, role):
        self.name = name
        self.role = role

    @classmethod
    def admin(cls, name):
        return cls(name, "admin")

    @classmethod
    def researcher(cls, name):
        return cls(name, "researcher")


user1 = User.admin("Yasin")
user2 = User.researcher("Ali")

print(user1.role)
print(user2.role)


# Factory methods provide readable ways
# to construct specialized objects.


# ============================================================
# 134. ENCAPSULATING COMPLEX LOGIC
# ============================================================

class Pipeline:

    def __init__(self, data):
        self.data = data

    def clean(self):
        self.data = [
            value.strip()
            for value in self.data
        ]

        return self

    def remove_empty(self):
        self.data = [
            value
            for value in self.data
            if value
        ]

        return self

    def get_data(self):
        return self.data


pipeline = Pipeline(
    [" Ali ", "", " Sara "]
)

result = (
    pipeline
    .clean()
    .remove_empty()
    .get_data()
)

print(result)


# This is an example of a fluent interface.
#
# Again:
#
#     use when it improves readability,
#     not merely because it looks clever.


# ============================================================
# 135. OOP PIPELINE DIAGRAM
# ============================================================
#
#     raw data
#        ↓
#     Pipeline object
#        ↓
#     clean()
#        ↓
#     remove_empty()
#        ↓
#     get_data()
#        ↓
#     clean data


# ============================================================
# 136. OOP AND SEPARATION OF RESPONSIBILITIES
# ============================================================
#
# A large system can contain:
#
#     DataLoader
#     DataCleaner
#     DataValidator
#     FeatureEngineer
#     Model
#     Evaluator
#     Reporter
#
#
# Example:
#
#     DataLoader
#          ↓
#     DataCleaner
#          ↓
#     DataValidator
#          ↓
#     FeatureEngineer
#          ↓
#     Model
#          ↓
#     Evaluator


# ============================================================
# 137. SIMPLE OOP DATA PIPELINE
# ============================================================

class DataCleaner:

    def clean_names(self, names):
        return [
            name.strip().title()
            for name in names
        ]


class DataAnalyzer:

    def average(self, values):
        return sum(values) / len(values)


names = [" yasin ", "ALI", " sara "]
scores = [90, 85, 95]

cleaner = DataCleaner()
analyzer = DataAnalyzer()

clean_names = cleaner.clean_names(names)
average_score = analyzer.average(scores)

print(clean_names)
print(average_score)


# This separates responsibilities.
#
# But again:
#
# if the classes contain no state,
# simple functions may be even better.


# ============================================================
# 138. DEPENDENCY INJECTION
# ============================================================
#
# Instead of a class creating all of its dependencies internally,
# pass dependencies into it.
#
#
# Example:
#
#     Reporter
#        ↓
#     receives exporter
#
#
# This makes components easier to:
#
#     test
#     replace
#     configure


# ============================================================
# 139. DEPENDENCY INJECTION EXAMPLE
# ============================================================

class CSVExporter:

    def export(self, data):
        return "CSV"


class JSONExporter:

    def export(self, data):
        return "JSON"


class Report:

    def __init__(self, exporter):
        self.exporter = exporter

    def generate(self, data):
        return self.exporter.export(data)


report = Report(CSVExporter())

print(report.generate([1, 2, 3]))


# Replace dependency:
#
#     report = Report(JSONExporter())
#
# without changing Report.generate().


# ============================================================
# 140. OOP TESTABILITY
# ============================================================
#
# Good object design makes behavior testable.
#
#
# Example:
#
#     account.deposit(100)
#
# then verify:
#
#     account.balance
#
#
# Or:
#
#     model.fit(data)
#
# then:
#
#     model.predict(...)
#
#
# Small objects with clear interfaces
# are easier to test.


# ============================================================
# 141. SINGLE RESPONSIBILITY
# ============================================================
#
# A class should have a focused responsibility.
#
#
# Bad:
#
#     DataSystem
#         ↓
#     load files
#     clean data
#     train model
#     send email
#     create plots
#     connect database
#
#
# Better:
#
#     DataLoader
#     DataCleaner
#     ModelTrainer
#     Reporter
#     EmailSender
#
#
# Do not interpret this as:
#
#     "every class must contain exactly one method."
#
# The principle concerns responsibility,
# not arbitrary line counts.


# ============================================================
# 142. OPEN/CLOSED IDEA
# ============================================================
#
# A useful design goal:
#
#     open for extension
#     closed for unnecessary modification
#
#
# Example:
#
#     save(exporter)
#
# can accept new exporters
# without changing save().
#
#
# This works naturally with polymorphism/protocols.


# ============================================================
# 143. LISKOV SUBSTITUTION IDEA
# ============================================================
#
# If:
#
#     Dog is an Animal
#
# code expecting an Animal should generally
# be able to use Dog without violating the expected contract.
#
#
# If subclass behavior fundamentally breaks
# assumptions made by the parent abstraction,
# inheritance may be the wrong design.


# ============================================================
# 144. INTERFACE SEGREGATION IDEA
# ============================================================
#
# Prefer small focused interfaces over giant interfaces.
#
#
# Instead of:
#
#     object must:
#         read
#         write
#         delete
#         export
#         train
#         predict
#
# perhaps separate capabilities:
#
#     Reader
#     Writer
#     Exporter
#     Predictor
#
#
# This works particularly well with Python protocols.


# ============================================================
# 145. DEPENDENCY INVERSION IDEA
# ============================================================
#
# High-level logic should depend on abstractions/interfaces
# rather than hard-coded concrete implementations.
#
#
# Example:
#
#     Report
#        ↓
#     Exporter interface
#        ↑
#     ┌────┴────┐
#     CSV      JSON
#
#
# This is the deeper idea behind dependency injection.


# ============================================================
# 146. OOP DESIGN PRINCIPLES — PRACTICAL TABLE
# ============================================================
#
#     Principle       Practical question
#     ---------------------------------------------
#     SRP             Does this class have one focus?
#     OCP             Can behavior be extended cleanly?
#     LSP             Can subclasses honor the contract?
#     ISP             Is the interface unnecessarily large?
#     DIP             Can dependencies be replaced?
#
#
# These principles are useful,
# but do not turn them into rigid rules for tiny scripts.


# ============================================================
# 147. CLASS INHERITANCE TREE
# ============================================================

class Employee:
    pass


class Lecturer(Employee):
    pass


class Researcher(Employee):
    pass


class LecturerResearcher(Lecturer, Researcher):
    pass


# Diagram:
#
#                Employee
#                /      \
#               /        \
#        Lecturer      Researcher
#               \        /
#                \      /
#          LecturerResearcher
#
#
# Multiple inheritance can become complex.
# Use it deliberately.


# ============================================================
# 148. COMPOSITION ALTERNATIVE TO MULTIPLE INHERITANCE
# ============================================================

class Lecturer:

    def teach(self):
        return "Teaching"


class Researcher:

    def research(self):
        return "Researching"


class Academic:

    def __init__(self):
        self.lecturer = Lecturer()
        self.researcher = Researcher()

    def teach(self):
        return self.lecturer.teach()

    def research(self):
        return self.researcher.research()


academic = Academic()

print(academic.teach())
print(academic.research())


# Instead of inheriting from two classes,
# Academic contains two components.
#
# This can sometimes be easier to reason about.


# ============================================================
# 149. "PREFER COMPOSITION OVER INHERITANCE"
# ============================================================
#
# This is a useful design heuristic,
# not an absolute law.
#
#
# Inheritance is good when:
#
#     genuine subtype relationship
#     shared interface
#     meaningful substitutability
#
#
# Composition is often better when:
#
#     behavior can vary independently
#     components can be swapped
#     relationships are HAS-A
#     inheritance creates deep hierarchies


# ============================================================
# 150. OOP WITH ENUMS
# ============================================================
#
# Enumerations can represent a fixed set of named states.


from enum import Enum


class Status(Enum):

    PENDING = "pending"
    SUCCESS = "success"
    FAILED = "failed"


status = Status.SUCCESS

print(status)
print(status.value)


# Useful for:
#
#     statuses
#     categories
#     modes
#     finite state choices


# ============================================================
# 151. OBJECT STATE MACHINE
# ============================================================
#
# Some objects have valid states:
#
#     PENDING
#        ↓
#     PROCESSING
#        ↓
#     SUCCESS
#
# or:
#
#     PROCESSING
#        ↓
#     FAILED
#
#
# OOP can encapsulate valid transitions.


# ============================================================
# 152. SIMPLE STATE MACHINE
# ============================================================

class Job:

    def __init__(self):
        self.status = Status.PENDING

    def start(self):
        self.status = Status.PROCESSING

    def succeed(self):
        self.status = Status.SUCCESS

    def fail(self):
        self.status = Status.FAILED


# A more complete design would validate
# whether each transition is legal.


# ============================================================
# 153. METHODS AS COMMANDS VS QUERIES
# ============================================================
#
# A useful mental distinction:
#
#     COMMAND
#         changes state
#
#     QUERY
#         returns information
#
#
# Example:
#
#     account.deposit(100)
#         → command
#
#     account.get_balance()
#         → query
#
#
# This is not a Python requirement,
# but can improve API design.


# ============================================================
# 154. PROPERTY AS QUERY
# ============================================================

class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height


rectangle = Rectangle(10, 5)

print(rectangle.area)


# area behaves naturally as derived state.


# ============================================================
# 155. SPECIAL METHODS MASTER TABLE
# ============================================================
#
#     Syntax / operation       Special method
#     ----------------------------------------
#     object(...)              __call__
#     str(object)              __str__
#     repr(object)             __repr__
#     len(object)              __len__
#     object[index]            __getitem__
#     object[index] = value    __setitem__
#     value in object          __contains__
#     a == b                   __eq__
#     a != b                   __ne__
#     a < b                    __lt__
#     a <= b                   __le__
#     a > b                    __gt__
#     a >= b                   __ge__
#     a + b                    __add__
#     a - b                    __sub__
#     a * b                    __mul__
#     a / b                    __truediv__
#     for x in object          __iter__
#     next(object)             __next__
#     with object              __enter__/__exit__
#     object initialization    __init__
#     object creation          __new__
#
#
# You do NOT need to memorize all of these immediately.
#
# Learn them by protocol/use case.


# ============================================================
# 156. OBJECT PROTOCOLS
# ============================================================
#
# Python often asks objects:
#
#     Can you be iterated?
#     Can you be indexed?
#     Can you be called?
#     Can you be compared?
#     Can you be used with with?
#     Can you report length?
#
#
# Special methods implement these protocols.


# ============================================================
# 157. OOP + ITERATION
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __iter__(self):
        return iter(self.rows)

    def __len__(self):
        return len(self.rows)


dataset = Dataset(
    ["row1", "row2", "row3"]
)

print(len(dataset))

for row in dataset:
    print(row)


# The class now behaves naturally
# like a collection.


# ============================================================
# 158. OOP + INDEXING
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = rows

    def __getitem__(self, index):
        return self.rows[index]

    def __len__(self):
        return len(self.rows)


dataset = Dataset([10, 20, 30])

print(dataset[0])
print(dataset[1:])


# This creates a collection-like object.


# ============================================================
# 159. OOP + DATA RECORD
# ============================================================

@dataclass
class Record:

    name: str
    age: int
    score: float

    @property
    def passed(self):
        return self.score >= 50


record = Record(
    "Yasin",
    30,
    92
)

print(record.passed)


# This is a very practical combination:
#
#     dataclass
#     attributes
#     property
#     type hints


# ============================================================
# 160. OOP + DATA VALIDATION
# ============================================================

@dataclass
class Measurement:

    value: float
    unit: str

    def __post_init__(self):

        if not isinstance(self.value, (int, float)):
            raise TypeError(
                "value must be numeric"
            )

        if not self.unit:
            raise ValueError(
                "unit cannot be empty"
            )


measurement = Measurement(25.5, "C")

print(measurement)


# __post_init__ is called by dataclasses
# after generated initialization.


# ============================================================
# 161. OOP + DATA CLEANING OBJECT
# ============================================================

class Cleaner:

    MISSING = {"", "NA", "N/A", "?", "-"}

    def clean_text(self, value):

        if value is None:
            return None

        value = value.strip()

        if value in self.MISSING:
            return None

        return value


cleaner = Cleaner()

print(cleaner.clean_text(" Ali "))
print(cleaner.clean_text("N/A"))


# Class attribute MISSING contains configuration shared
# by all Cleaner instances.


# ============================================================
# 162. CONFIGURABLE CLEANER
# ============================================================

class Cleaner:

    def __init__(self, missing_values=None):

        if missing_values is None:
            missing_values = {
                "",
                "NA",
                "N/A",
                "?"
            }

        self.missing_values = set(missing_values)

    def clean_text(self, value):

        if value is None:
            return None

        value = value.strip()

        if value in self.missing_values:
            return None

        return value


cleaner = Cleaner(
    missing_values={"", "NA", "unknown"}
)

print(cleaner.clean_text("unknown"))


# Now the object stores configuration as state.


# ============================================================
# 163. OOP + FILE PROCESSING
# ============================================================

class TextFile:

    def __init__(self, filename):
        self.filename = filename

    def read(self):

        with open(
            self.filename,
            "r",
            encoding="utf-8"
        ) as file:

            return file.read()


# Object state:
#
#     filename
#
#
# Behavior:
#
#     read()
#
#
# A real application might also provide:
#
#     write()
#     append()
#     exists()
#     line_count()


# ============================================================
# 164. CONTEXT MANAGER CLASS FOR RESOURCES
# ============================================================

class ManagedFile:

    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def __enter__(self):

        self.file = open(
            self.filename,
            "r",
            encoding="utf-8"
        )

        return self.file

    def __exit__(
        self,
        exc_type,
        exc_value,
        traceback
    ):

        if self.file is not None:
            self.file.close()


# Usage:
#
#     with ManagedFile("data.txt") as file:
#         data = file.read()
#
#
# This illustrates how classes implement
# resource-management protocols.


# ============================================================
# 165. CLASS-BASED API DESIGN
# ============================================================
#
# A good class API should make valid operations easy:
#
#     model.fit(data)
#     model.predict(data)
#
#
# and invalid operations difficult or explicit:
#
#     model.predict(...)
#
# before fit may raise a clear error.
#
#
# The interface should communicate:
#
#     what the object knows
#     what it can do
#     what state it requires


# ============================================================
# 166. OBJECT INVARIANTS
# ============================================================
#
# An invariant is a condition that should remain true
# for a valid object.
#
#
# BankAccount:
#
#     balance >= 0
#
#
# Student:
#
#     0 <= score <= 100
#
#
# Dataset:
#
#     rows is a list of records
#
#
# Methods should preserve important invariants.


# ============================================================
# 167. INVARIANT EXAMPLE
# ============================================================

class Score:

    def __init__(self, value):
        self.value = value

    def set_value(self, value):

        if not 0 <= value <= 100:
            raise ValueError(
                "Score must be between 0 and 100"
            )

        self.value = value


score = Score(90)

score.set_value(95)

print(score.value)


# The object attempts to maintain:
#
#     0 <= value <= 100


# ============================================================
# 168. ABSTRACTING DOMAIN CONCEPTS
# ============================================================
#
# Compare:
#
#     balance += amount
#
# with:
#
#     account.deposit(amount)
#
#
# The second expression communicates domain meaning.
#
#
# OOP can turn raw operations into
# meaningful domain actions.


# ============================================================
# 169. DOMAIN MODELING
# ============================================================
#
# For a water-pipeline application:
#
#     Pipe
#       ├── diameter
#       ├── age
#       ├── material
#       ├── pressure
#       ├── condition
#       ├── calculate_age()
#       ├── is_high_risk()
#       └── maintenance_due()
#
#
# For an ML system:
#
#     Model
#       ├── parameters
#       ├── fit()
#       ├── predict()
#       └── evaluate()
#
#
# The class represents a domain concept.


# ============================================================
# 170. OOP FOR ENGINEERING DATA
# ============================================================

@dataclass
class Pipe:

    pipe_id: str
    age: float
    pressure: float
    diameter: float

    @property
    def pressure_ratio(self):

        if self.diameter == 0:
            raise ValueError(
                "diameter cannot be zero"
            )

        return self.pressure / self.diameter

    def is_old(self, threshold=50):
        return self.age >= threshold


pipe = Pipe(
    "P001",
    55,
    8.2,
    200
)

print(pipe.pressure_ratio)
print(pipe.is_old())


# This demonstrates domain-oriented objects.


# ============================================================
# 171. OOP FOR A DATA RECORD COLLECTION
# ============================================================

class PipeDataset:

    def __init__(self, pipes):
        self.pipes = list(pipes)

    def count(self):
        return len(self.pipes)

    def old_pipes(self, threshold=50):
        return [
            pipe
            for pipe in self.pipes
            if pipe.is_old(threshold)
        ]

    def average_age(self):

        if not self.pipes:
            return None

        return (
            sum(pipe.age for pipe in self.pipes)
            / len(self.pipes)
        )


pipes = [
    Pipe("P001", 40, 7.5, 200),
    Pipe("P002", 60, 8.1, 150),
    Pipe("P003", 55, 7.8, 250)
]

dataset = PipeDataset(pipes)

print(dataset.count())
print(dataset.average_age())
print(dataset.old_pipes())


# Dataset object:
#
#     owns/organizes records
#     provides domain operations


# ============================================================
# 172. OBJECT COMPOSITION TREE
# ============================================================
#
#     PipeDataset
#          │
#          ├── Pipe
#          │    ├── pipe_id
#          │    ├── age
#          │    └── pressure
#          │
#          ├── Pipe
#          │    ├── pipe_id
#          │    ├── age
#          │    └── pressure
#          │
#          └── Pipe
#
#
# This is composition of objects into a larger model.


# ============================================================
# 173. OOP + POLYMORPHIC MODELS
# ============================================================

class BaseModel:

    def fit(self, X, y):
        raise NotImplementedError

    def predict(self, X):
        raise NotImplementedError


class ThresholdModel(BaseModel):

    def fit(self, X, y):
        self.threshold = sum(y) / len(y)
        return self

    def predict(self, X):
        return [
            int(value >= self.threshold)
            for value in X
        ]


class AlwaysZeroModel(BaseModel):

    def fit(self, X, y):
        return self

    def predict(self, X):
        return [0 for _ in X]


models = [
    ThresholdModel(),
    AlwaysZeroModel()
]

for model in models:
    model.fit([1, 2, 3], [0, 1, 1])
    print(model.predict([1, 2, 3]))


# Polymorphic interface:
#
#     fit()
#     predict()
#
# Different implementations,
# same external interface.


# ============================================================
# 174. ABSTRACT MODEL INTERFACE
# ============================================================

from abc import ABC, abstractmethod


class Model(ABC):

    @abstractmethod
    def fit(self, X, y):
        ...

    @abstractmethod
    def predict(self, X):
        ...


class DummyModel(Model):

    def fit(self, X, y):
        self.value = 0
        return self

    def predict(self, X):
        return [self.value for _ in X]


model = DummyModel()

model.fit([1, 2], [0, 1])

print(model.predict([3, 4]))


# The abstract class documents/enforces the interface.


# ============================================================
# 175. CLASS METHOD VS STATIC METHOD VS FUNCTION
# ============================================================
#
# Suppose we need:
#
#     calculate_distance(a, b)
#
# If it has no relation to class state:
#
#     ordinary function may be best.
#
#
# If it is logically grouped under a class:
#
#     staticmethod may be reasonable.
#
#
# If it needs class information:
#
#     classmethod.
#
#
# If it needs object state:
#
#     instance method.


# ============================================================
# 176. DECISION TABLE — WHAT SHOULD THIS BE?
# ============================================================
#
#     Does it need self?
#         YES → instance method
#
#     Does it need cls?
#         YES → classmethod
#
#     Does it need neither?
#         ↓
#     Could it simply be a normal function?
#         YES → ordinary function may be clearer
#
#     Is it naturally part of class's namespace?
#         YES → staticmethod can be reasonable


# ============================================================
# 177. INHERITANCE DECISION TREE
# ============================================================
#
# Ask:
#
#     Is Child genuinely a subtype of Parent?
#              │
#          ┌───┴───┐
#         YES      NO
#          ↓        ↓
#     inheritance  composition/
#                  association
#
#
# Then ask:
#
#     Does Child satisfy Parent's contract?
#              │
#             NO
#              ↓
#     inheritance may be wrong


# ============================================================
# 178. COMMON OOP MISTAKES
# ============================================================
#
# (1) Forgetting self:
#
#     def greet():
#
#     instead of:
#
#     def greet(self):
#
#
# (2) Using self incorrectly:
#
#     self = ...
#
#
# (3) Confusing class and instance attributes.
#
#
# (4) Accidentally sharing mutable class state.
#
#
# (5) Overusing inheritance.
#
#
# (6) Creating classes for every tiny operation.
#
#
# (7) Huge classes with dozens of responsibilities.
#
#
# (8) Using getters/setters everywhere without need.
#
#
# (9) Using private-style name mangling as if it were security.
#
#
# (10) Calling super() incorrectly in multiple inheritance.
#
#
# (11) Forgetting that methods are attributes.
#
#
# (12) Confusing == and is.
#
#
# (13) Mutating object state unexpectedly.
#
#
# (14) Deep inheritance hierarchies.
#
#
# (15) Using inheritance merely for code reuse.


# ============================================================
# 179. CLASS ATTRIBUTE TRAP
# ============================================================
#
# Dangerous:
#
#     class Student:
#         scores = []
#
#
# Better:
#
#     class Student:
#
#         def __init__(self):
#             self.scores = []
#
#
# Rule:
#
#     per-object mutable data
#         → instance attribute


# ============================================================
# 180. MUTABLE DEFAULT + CLASS STATE
# ============================================================
#
# Two related problems:
#
#     def __init__(self, items=[]):
#         self.items = items
#
#
# and:
#
#     class Example:
#         items = []
#
#
# Both can create unintended sharing.
#
#
# Safe pattern:
#
#     def __init__(self, items=None):
#         if items is None:
#             items = []
#         self.items = items


# ============================================================
# 181. DEEP INHERITANCE WARNING
# ============================================================
#
# Avoid unnecessarily deep trees:
#
#     A
#      ↓
#     B
#      ↓
#     C
#      ↓
#     D
#      ↓
#     E
#      ↓
#     F
#
#
# Every level increases:
#
#     lookup complexity
#     coupling
#     maintenance cost
#     conceptual burden
#
#
# Prefer simple hierarchies unless the domain
# genuinely requires them.


# ============================================================
# 182. OOP DEBUGGING
# ============================================================
#
# When an object behaves unexpectedly, inspect:
#
#     type(obj)
#     isinstance(obj, Class)
#     obj.__dict__
#     Class.__dict__
#     obj.method
#     Class.mro()
#     repr(obj)
#
#
# Useful:
#
#     dir(obj)
#
# to inspect available attributes/methods.


# ============================================================
# 183. dir()
# ============================================================

class Student:

    def greet(self):
        return "Hello"


student = Student()

print(dir(student))


# dir() gives a useful list of attributes available
# through the object.
#
# It includes inherited and special attributes too.


# ============================================================
# 184. help()
# ============================================================

help(Student)


# help() provides documentation about an object/class.
#
# In notebooks/interactive learning,
# this is extremely useful.


# ============================================================
# 185. inspect MODULE
# ============================================================

import inspect


class Student:

    def greet(self, name):
        return f"Hello {name}"


print(inspect.signature(Student.greet))
print(inspect.getsource(Student.greet))


# inspect is useful for exploring Python objects
# during development and learning.


# ============================================================
# 186. FUNCTION SIGNATURE OF METHODS
# ============================================================
#
#     Student.greet
#
# may show:
#
#     (self, name)
#
#
#     student.greet
#
# behaves as a bound method where self is already supplied.


# ============================================================
# 187. OBJECT REPRESENTATION
# ============================================================

@dataclass
class Student:

    name: str
    age: int


student = Student("Yasin", 30)

print(str(student))
print(repr(student))


# Dataclass automatically provides useful representation behavior.


# ============================================================
# 188. HASHABILITY
# ============================================================
#
# Some objects can be used as dictionary keys or set elements.
#
#
# Requirements generally involve:
#
#     equality
#     stable hash
#
#
# Mutable objects are often not suitable as hash keys
# when their equality-relevant state can change.
#
#
# Example:
#
#     strings
#     integers
#     tuples containing hashable values
#
# are commonly hashable.


# ============================================================
# 189. __hash__
# ============================================================
#
# Python can use:
#
#     __hash__()
#
# to support hashing.
#
#
# Be careful when implementing:
#
#     __eq__()
#     __hash__()
#
# yourself.
#
# If objects compare equal,
# their hashes must be equal.


# ============================================================
# 190. FROZEN DATACLASS
# ============================================================

@dataclass(frozen=True)
class Point:

    x: int
    y: int


point = Point(10, 20)

print(point)


# frozen=True prevents normal attribute reassignment:
#
#     point.x = 50
#
# would raise an error.
#
#
# Frozen dataclasses can be useful for immutable-style
# value objects.


# ============================================================
# 191. VALUE OBJECT VS ENTITY
# ============================================================
#
# VALUE OBJECT:
#
#     identity is mainly its values.
#
#     Point(10, 20)
#
#
# ENTITY:
#
#     identity matters independently of current values.
#
#     Student(id=1001, ...)
#
#
# This distinction can help domain modeling.


# ============================================================
# 192. OBJECT IDENTITY IN DATA SYSTEMS
# ============================================================
#
# Consider:
#
#     Student(id=101, name="Ali")
#
# and:
#
#     Student(id=101, name="Ali")
#
#
# They may represent the same domain entity
# even if they are separate Python objects.
#
#
# Therefore:
#
#     Python identity
#
# and:
#
#     domain identity
#
# are different concepts.


# ============================================================
# 193. CLASS DESIGN FOR DATABASE RECORDS
# ============================================================

@dataclass
class Customer:

    customer_id: int
    name: str
    email: str


customer = Customer(
    1001,
    "Ali",
    "ali@example.com"
)

print(customer)


# Dataclasses are often convenient for
# representing structured records.


# ============================================================
# 194. OOP + JSON CONCEPT
# ============================================================
#
# An object can provide methods such as:
#
#     to_dict()
#     from_dict()
#
# to translate between object state
# and serializable data.


# ============================================================
# 195. SERIALIZATION METHODS
# ============================================================

@dataclass
class Student:

    name: str
    age: int
    score: float

    def to_dict(self):
        return {
            "name": self.name,
            "age": self.age,
            "score": self.score
        }

    @classmethod
    def from_dict(cls, data):
        return cls(
            data["name"],
            data["age"],
            data["score"]
        )


student = Student("Yasin", 30, 92.5)

data = student.to_dict()

student2 = Student.from_dict(data)

print(data)
print(student2)


# This is a useful OOP/data bridge.


# ============================================================
# 196. OOP + CSV CONCEPT
# ============================================================
#
# An object can also represent one row:
#
#     Student
#       ↓
#     one record
#
#
# A separate loader can convert:
#
#     CSV row
#        ↓
#     Student object
#
#
# Keeping loading logic separate from the record class
# can maintain clean responsibilities.


# ============================================================
# 197. OOP ARCHITECTURE FOR DATA
# ============================================================
#
#     CSV / JSON
#          ↓
#     DataLoader
#          ↓
#     Records / dataclasses
#          ↓
#     Validator
#          ↓
#     Analyzer
#          ↓
#     Model
#          ↓
#     Reporter
#
#
# Each component has a focused role.


# ============================================================
# 198. OOP + COLLECTION PROTOCOL
# ============================================================
#
# If a custom Dataset supports:
#
#     __len__()
#     __iter__()
#     __getitem__()
#
# it can behave naturally like a collection.
#
#
# This lets users write:
#
#     len(dataset)
#     dataset[0]
#     for row in dataset:
#         ...
#
#
# Instead of forcing:
#
#     dataset.get_length()
#     dataset.get_item(0)
#     dataset.iterate()
#
#
# Python protocols make APIs feel natural.


# ============================================================
# 199. COMPLETE COLLECTION-LIKE DATASET
# ============================================================

class Dataset:

    def __init__(self, rows):
        self.rows = list(rows)

    def __len__(self):
        return len(self.rows)

    def __getitem__(self, index):
        return self.rows[index]

    def __iter__(self):
        return iter(self.rows)

    def __contains__(self, value):
        return value in self.rows


dataset = Dataset([10, 20, 30])

print(len(dataset))
print(dataset[0])
print(20 in dataset)

for row in dataset:
    print(row)


# This object now speaks several Python protocols.


# ============================================================
# 200. OOP MASTER PROTOCOL MAP
# ============================================================
#
#     Want len(object)?
#         → __len__
#
#     Want object[index]?
#         → __getitem__
#
#     Want object[index] = value?
#         → __setitem__
#
#     Want value in object?
#         → __contains__
#
#     Want for x in object?
#         → __iter__
#
#     Want next(object)?
#         → __next__
#
#     Want object(...)?
#         → __call__
#
#     Want print(object)?
#         → __str__
#
#     Want repr(object)?
#         → __repr__
#
#     Want object1 + object2?
#         → __add__
#
#     Want object1 == object2?
#         → __eq__
#
#     Want with object?
#         → __enter__/__exit__


# ============================================================
# 201. CLASS DECORATORS
# ============================================================
#
# Python decorators are not limited to functions.
#
# A decorator can modify a class too.
#
#
# Example:
#
#     @dataclass
#     class Student:
#         ...
#
#
# dataclass is effectively a class transformation tool.


# ============================================================
# 202. __slots__
# ============================================================
#
# __slots__ can restrict instance attributes
# and can reduce per-instance memory overhead
# in appropriate cases.
#
#
# Example:

class Point:

    __slots__ = (
        "x",
        "y"
    )

    def __init__(self, x, y):
        self.x = x
        self.y = y


point = Point(10, 20)

print(point.x)
print(point.y)


# __slots__ is an advanced feature.
#
# Do not use it automatically.
#
# It changes normal attribute-storage behavior
# and has trade-offs.


# ============================================================
# 203. ABSTRACT PROPERTY
# ============================================================

from abc import ABC, abstractmethod


class Shape(ABC):

    @property
    @abstractmethod
    def area(self):
        ...


class Rectangle(Shape):

    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height


rectangle = Rectangle(10, 5)

print(rectangle.area)


# Abstract classes can require properties as well as methods.


# ============================================================
# 204. MULTIPLE INHERITANCE WITH super()
# ============================================================

class A:

    def process(self):
        print("A")


class B(A):

    def process(self):
        print("B")
        super().process()


class C(B):

    def process(self):
        print("C")
        super().process()


c = C()

c.process()


# Output:
#
#     C
#     B
#     A
#
#
# super() continues according to MRO.


# ============================================================
# 205. OOP AND ERROR HANDLING
# ============================================================

class BankAccount:

    def __init__(self, balance=0):
        self.balance = balance

    def withdraw(self, amount):

        if amount < 0:
            raise ValueError(
                "Amount cannot be negative"
            )

        if amount > self.balance:
            raise ValueError(
                "Insufficient balance"
            )

        self.balance -= amount


account = BankAccount(100)

account.withdraw(50)

print(account.balance)


# The class protects its invariants
# by validating operations.


# ============================================================
# 206. CUSTOM EXCEPTIONS
# ============================================================

class InsufficientBalanceError(Exception):
    pass


class BankAccount:

    def __init__(self, balance=0):
        self.balance = balance

    def withdraw(self, amount):

        if amount > self.balance:
            raise InsufficientBalanceError(
                "Insufficient balance"
            )

        self.balance -= amount


account = BankAccount(100)

try:
    account.withdraw(200)

except InsufficientBalanceError as error:
    print(error)


# Custom exceptions can make domain errors clearer.


# ============================================================
# 207. OOP + LOGGING
# ============================================================
#
# A class can encapsulate operational state:
#
#     logger
#     configuration
#     counters
#
#
# But use Python's logging module for real application
# logging rather than relying on print().


# ============================================================
# 208. OBJECT STATE + LOGGING COUNTER
# ============================================================

class Processor:

    def __init__(self):
        self.processed = 0

    def process(self, value):

        self.processed += 1

        return value * 2


processor = Processor()

print(processor.process(10))
print(processor.process(20))

print(processor.processed)


# Object state records historical information
# across method calls.


# ============================================================
# 209. OOP AND CACHING
# ============================================================

class Calculator:

    def __init__(self):
        self.cache = {}

    def square(self, value):

        if value not in self.cache:
            self.cache[value] = value ** 2

        return self.cache[value]


calculator = Calculator()

print(calculator.square(10))
print(calculator.square(10))

print(calculator.cache)


# The object stores cached state.


# ============================================================
# 210. OOP VS MODULES
# ============================================================
#
# A module can already organize functions:
#
#     data_cleaning.py
#         clean_name()
#         clean_score()
#         normalize()
#
#
# A class becomes useful when operations
# share object state/configuration:
#
#     cleaner = Cleaner(...)
#
#     cleaner.clean_name(...)
#     cleaner.clean_score(...)
#
#
# Ask:
#
#     "Do these operations need shared state?"


# ============================================================
# 211. OOP VS DICTIONARY
# ============================================================
#
# Dictionary:
#
#     student = {
#         "name": "Yasin",
#         "age": 30
#     }
#
#
# Class:
#
#     student.name
#     student.average()
#
#
# Use a dictionary when:
#
#     data is simple
#     schema is flexible
#     behavior is minimal
#
#
# Use a class/dataclass when:
#
#     structure matters
#     behavior belongs to the data
#     validation matters
#     type information helps
#     object identity/state matters


# ============================================================
# 212. DICTIONARY → DATACLASS
# ============================================================

@dataclass
class Student:

    name: str
    age: int
    score: float

    def passed(self):
        return self.score >= 50


student = Student(
    name="Yasin",
    age=30,
    score=92
)

print(student.passed())


# This gives structure and behavior to the record.


# ============================================================
# 213. OOP + TYPE HINTS
# ============================================================

class Calculator:

    def add(
        self,
        a: float,
        b: float
    ) -> float:

        return a + b


calculator = Calculator()

print(calculator.add(10, 20))


# Type hints make class interfaces easier to understand.


# ============================================================
# 214. SELF-REFERENTIAL TYPE HINT
# ============================================================

from __future__ import annotations


class Counter:

    def __init__(self, value=0):
        self.value = value

    def add(self, amount: int) -> Counter:
        self.value += amount
        return self


counter = Counter()

counter.add(10).add(20)

print(counter.value)


# Returning Counter allows chaining.
#
# from __future__ import annotations
# helps postpone evaluation of annotations.


# ============================================================
# 215. GENERIC DATA CLASS
# ============================================================

from dataclasses import dataclass
from typing import Generic, TypeVar


T = TypeVar("T")


@dataclass
class Box(Generic[T]):

    value: T

    def get(self) -> T:
        return self.value


number_box = Box(10)
text_box = Box("Python")

print(number_box.get())
print(text_box.get())


# This is an advanced typing concept:
#
#     Box[int]
#     Box[str]
#
# Conceptually, the class can contain different types
# while preserving type information for static analysis.


# ============================================================
# 216. CLASS DECISION — DATA MODEL
# ============================================================
#
# If the object mainly represents data:
#
#     @dataclass
#
#
# If it represents state + behavior:
#
#     class
#
#
# If it is only a transformation:
#
#     function
#
#
# If it is a configurable transformation:
#
#     class or callable object
#
#
# If it is an interface:
#
#     ABC / Protocol


# ============================================================
# 217. COMMON OOP DESIGN PATTERNS
# ============================================================
#
# You do not need to memorize design-pattern catalogs.
#
# Understand the recurring ideas:
#
#
# FACTORY:
#
#     object creation through a method/function.
#
#
# STRATEGY:
#
#     interchangeable algorithms/behaviors.
#
#
# ADAPTER:
#
#     make one interface look like another.
#
#
# OBSERVER:
#
#     notify interested objects when state changes.
#
#
# CONTEXT MANAGER:
#
#     controlled resource lifecycle.
#
#
# DEPENDENCY INJECTION:
#
#     provide dependencies from outside.
#
#
# COMPOSITION:
#
#     combine small objects.


# ============================================================
# 218. STRATEGY PATTERN
# ============================================================

class Addition:

    def calculate(self, a, b):
        return a + b


class Multiplication:

    def calculate(self, a, b):
        return a * b


class Calculator:

    def __init__(self, strategy):
        self.strategy = strategy

    def calculate(self, a, b):
        return self.strategy.calculate(a, b)


calculator = Calculator(Addition())

print(calculator.calculate(10, 5))

calculator.strategy = Multiplication()

print(calculator.calculate(10, 5))


# Strategy:
#
#     Calculator
#         ↓
#     interchangeable algorithm
#         ↓
#     Addition / Multiplication


# ============================================================
# 219. ADAPTER PATTERN
# ============================================================

class OldSystem:

    def old_method(self):
        return "Old result"


class Adapter:

    def __init__(self, old_system):
        self.old_system = old_system

    def new_method(self):
        return self.old_system.old_method()


adapter = Adapter(OldSystem())

print(adapter.new_method())


# Adapter converts:
#
#     old interface
#          ↓
#     adapter
#          ↓
#     expected interface


# ============================================================
# 220. OBSERVER IDEA
# ============================================================
#
# A subject changes state:
#
#     Subject
#        │
#        ├── Observer 1
#        ├── Observer 2
#        └── Observer 3
#
#
# Subject notifies observers.
#
# This is useful in:
#
#     event systems
#     GUIs
#     reactive applications
#     monitoring


# ============================================================
# 221. SIMPLE OBSERVER
# ============================================================

class Subject:

    def __init__(self):
        self.observers = []

    def subscribe(self, observer):
        self.observers.append(observer)

    def notify(self, message):

        for observer in self.observers:
            observer.update(message)


class Observer:

    def update(self, message):
        print("Received:", message)


subject = Subject()

subject.subscribe(Observer())

subject.notify("Data updated")


# This illustrates object-to-object communication.


# ============================================================
# 222. OOP ARCHITECTURE FOR A DATA PROJECT
# ============================================================
#
# A larger project might look like:
#
#
#     ┌──────────────────┐
#     │    DataLoader    │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │    DataCleaner   │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │    Validator     │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │ FeatureEngineer  │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │      Model       │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │    Evaluator     │
#     └────────┬─────────┘
#              ↓
#     ┌──────────────────┐
#     │     Reporter     │
#     └──────────────────┘
#
#
# Not every project needs all of these classes.
#
# The architecture should follow actual complexity.


# ============================================================
# 223. MINI PROJECT — PIPELINE OBJECT MODEL
# ============================================================

@dataclass
class Record:

    name: str
    age: int
    score: float


class RecordCleaner:

    def clean_name(self, value):
        return value.strip().title()

    def clean_age(self, value):

        try:
            return int(value)

        except (TypeError, ValueError):
            return None

    def clean_score(self, value):

        try:
            return float(value)

        except (TypeError, ValueError):
            return None


class RecordValidator:

    def validate(self, record):

        if not record.name:
            return False

        if record.age is None:
            return False

        if record.score is None:
            return False

        if not 0 <= record.age <= 120:
            return False

        if not 0 <= record.score <= 100:
            return False

        return True


raw_records = [
    {
        "name": " yasin ",
        "age": "30",
        "score": "92"
    },
    {
        "name": " sara",
        "age": "25",
        "score": "88"
    },
    {
        "name": "ali ",
        "age": "?",
        "score": "91"
    }
]

cleaner = RecordCleaner()
validator = RecordValidator()

records = []

for row in raw_records:

    record = Record(
        name=cleaner.clean_name(row["name"]),
        age=cleaner.clean_age(row["age"]),
        score=cleaner.clean_score(row["score"])
    )

    if validator.validate(record):
        records.append(record)


for record in records:
    print(record)


# Architecture:
#
#     raw dictionary
#           ↓
#     RecordCleaner
#           ↓
#     Record
#           ↓
#     RecordValidator
#           ↓
#     valid records


# ============================================================
# 224. WHY THIS PROJECT IS OOP
# ============================================================
#
# Record:
#
#     represents the data.
#
#
# RecordCleaner:
#
#     owns cleaning operations.
#
#
# RecordValidator:
#
#     owns validation rules.
#
#
# This separates concerns.
#
#
# But for a tiny dataset,
# functions could be simpler.
#
# The point is learning how OOP can model
# a growing system.


# ============================================================
# 225. OBJECT COLLABORATION
# ============================================================
#
# OOP is not just:
#
#     "put functions inside classes."
#
#
# A deeper idea is:
#
#     objects collaborate.
#
#
#     Loader
#       ↓
#     Cleaner
#       ↓
#     Validator
#       ↓
#     Analyzer
#
#
# Each object provides services to another object.


# ============================================================
# 226. MESSAGE-PASSING MENTAL MODEL
# ============================================================
#
#     object A
#         │
#         │ method call
#         ↓
#     object B
#         │
#         │ result
#         ↓
#     object A
#
#
# Example:
#
#     analyzer.calculate(cleaner.clean(data))
#
#
# or:
#
#     cleaned = cleaner.clean(data)
#     result = analyzer.calculate(cleaned)
#
#
# The second is often easier to debug.


# ============================================================
# 227. OBJECT BOUNDARIES
# ============================================================
#
# A useful OOP question:
#
#     "Who should own this behavior?"
#
#
# Example:
#
#     pipe.is_old()
#
# makes sense because the rule concerns one Pipe.
#
#
#     dataset.old_pipes()
#
# makes sense because selecting many old pipes
# concerns the dataset.
#
#
# This is domain-oriented design.


# ============================================================
# 228. INSTANCE METHOD OR EXTERNAL FUNCTION?
# ============================================================
#
# Ask:
#
#     Does this operation naturally belong
#     to one object's state?
#
# YES:
#
#     pipe.is_old()
#
#
# NO:
#
#     calculate_correlation(x, y)
#
# may be better as:
#
#     calculate_correlation(x, y)
#
# rather than forcing:
#
#     SomeClass.calculate_correlation(x, y)
#
#
# Do not turn every function into a method.


# ============================================================
# 229. OOP AND FUNCTIONAL STYLE CAN COEXIST
# ============================================================
#
# A Python program can combine:
#
#     classes
#     functions
#     comprehensions
#     generators
#     modules
#     standard-library tools
#
#
# Example:
#
#     class Dataset:
#         ...
#
#     def normalize(values):
#         ...
#
#     cleaned = [
#         ...
#     ]
#
#
# Python is multi-paradigm.


# ============================================================
# 230. OOP LEARNING ORDER
# ============================================================
#
# LEVEL 1 — ESSENTIAL
#
#     [ ] class
#     [ ] object
#     [ ] instance
#     [ ] attributes
#     [ ] methods
#     [ ] self
#     [ ] __init__
#
#
# LEVEL 2 — CORE
#
#     [ ] instance state
#     [ ] class attributes
#     [ ] instance methods
#     [ ] classmethod
#     [ ] staticmethod
#     [ ] properties
#
#
# LEVEL 3 — OOP RELATIONSHIPS
#
#     [ ] inheritance
#     [ ] overriding
#     [ ] super()
#     [ ] polymorphism
#     [ ] duck typing
#     [ ] composition
#
#
# LEVEL 4 — PYTHON OBJECT MODEL
#
#     [ ] __str__
#     [ ] __repr__
#     [ ] __len__
#     [ ] __getitem__
#     [ ] __iter__
#     [ ] __call__
#     [ ] protocols
#
#
# LEVEL 5 — ADVANCED
#
#     [ ] ABC
#     [ ] Protocol
#     [ ] multiple inheritance
#     [ ] MRO
#     [ ] mixins
#     [ ] descriptors/properties
#     [ ] __new__
#     [ ] __slots__
#
#
# LEVEL 6 — DESIGN
#
#     [ ] composition
#     [ ] dependency injection
#     [ ] separation of responsibilities
#     [ ] polymorphic interfaces
#     [ ] testable objects
#     [ ] domain modeling


# ============================================================
# 231. PRACTICE LADDER
# ============================================================
#
# Practice in this order:
#
#     1. Empty class
#          ↓
#     2. Object creation
#          ↓
#     3. __init__
#          ↓
#     4. Instance attributes
#          ↓
#     5. Instance methods
#          ↓
#     6. Class attributes
#          ↓
#     7. classmethod
#          ↓
#     8. staticmethod
#          ↓
#     9. property
#          ↓
#    10. inheritance
#          ↓
#    11. overriding
#          ↓
#    12. super
#          ↓
#    13. polymorphism
#          ↓
#    14. composition
#          ↓
#    15. dataclasses
#          ↓
#    16. special methods
#          ↓
#    17. protocols
#          ↓
#    18. larger architecture


# ============================================================
# 232. PRACTICE PROJECT 1 — COUNTER
# ============================================================
#
# Build:
#
#     Counter
#
# Requirements:
#
#     value
#     increment()
#     decrement()
#     reset()
#     get_value()
#
#
# Then add:
#
#     __str__()
#     __repr__()
#
#
# Then:
#
#     __iadd__()
#
# for:
#
#     counter += 5


# ============================================================
# 233. PRACTICE PROJECT 2 — STUDENT
# ============================================================
#
# Build:
#
#     Student
#
# Attributes:
#
#     name
#     age
#     scores
#
#
# Methods:
#
#     add_score()
#     average()
#     highest_score()
#     lowest_score()
#     passed()
#
#
# Property:
#
#     status
#
#
# Then add:
#
#     __str__()
#     __repr__()


# ============================================================
# 234. PRACTICE PROJECT 3 — DATASET
# ============================================================
#
# Build:
#
#     Dataset
#
# Requirements:
#
#     rows
#     count()
#     filter()
#     sort()
#     average()
#     summary()
#
#
# Then implement:
#
#     __len__()
#     __getitem__()
#     __iter__()
#     __contains__()
#
#
# Goal:
#
#     make your Dataset object feel like
#     a normal Python collection.


# ============================================================
# 235. PRACTICE PROJECT 4 — BANK ACCOUNT
# ============================================================
#
# Build:
#
#     BankAccount
#
# State:
#
#     owner
#     balance
#
#
# Methods:
#
#     deposit()
#     withdraw()
#     transfer()
#
#
# Validation:
#
#     no negative deposits
#     no overdrawing
#
#
# Custom exception:
#
#     InsufficientBalanceError
#
#
# This project teaches:
#
#     state
#     methods
#     validation
#     exceptions
#     object collaboration


# ============================================================
# 236. PRACTICE PROJECT 5 — ML MODEL INTERFACE
# ============================================================
#
# Create:
#
#     BaseModel
#
# with:
#
#     fit()
#     predict()
#
#
# Then create:
#
#     MeanModel
#     ThresholdModel
#
#
# Make both obey the same interface.
#
#
# This teaches:
#
#     abstract classes
#     polymorphism
#     inheritance
#     interfaces
#     model state


# ============================================================
# 237. PRACTICE PROJECT 6 — PIPELINE
# ============================================================
#
# Create:
#
#     DataLoader
#     DataCleaner
#     DataValidator
#     DataAnalyzer
#
#
# Pipeline:
#
#     raw CSV
#       ↓
#     loader
#       ↓
#     cleaner
#       ↓
#     validator
#       ↓
#     analyzer
#
#
# Then replace one component
# without changing the others.
#
# This teaches dependency injection and interfaces.


# ============================================================
# 238. OOP CHEAT SHEET — BASIC SYNTAX
# ============================================================
#
# CLASS:
#
#     class Student:
#         ...
#
#
# OBJECT:
#
#     student = Student()
#
#
# INITIALIZER:
#
#     def __init__(self, ...):
#
#
# INSTANCE ATTRIBUTE:
#
#     self.name = name
#
#
# INSTANCE METHOD:
#
#     def greet(self):
#
#
# CLASS ATTRIBUTE:
#
#     university = "IAU"
#
#
# CLASS METHOD:
#
#     @classmethod
#     def method(cls):
#
#
# STATIC METHOD:
#
#     @staticmethod
#     def method():
#
#
# PROPERTY:
#
#     @property
#     def value(self):
#
#
# INHERITANCE:
#
#     class Dog(Animal):
#
#
# PARENT/MRO ACCESS:
#
#     super()
#
#
# ABSTRACT METHOD:
#
#     @abstractmethod
#
#
# DATACLASS:
#
#     @dataclass
#
#
# SPECIAL METHOD:
#
#     def __str__(self):
#


# ============================================================
# 239. OOP CHEAT SHEET — RELATIONSHIPS
# ============================================================
#
#     IS-A
#         → inheritance
#
#     HAS-A
#         → composition / aggregation
#
#     USES
#         → association / dependency
#
#     SAME INTERFACE
#         → polymorphism
#
#     REQUIRED INTERFACE
#         → ABC / Protocol


# ============================================================
# 240. OOP CHEAT SHEET — METHOD TYPES
# ============================================================
#
#     instance method
#         self
#         object state
#
#
#     classmethod
#         cls
#         class state / alternative constructors
#
#
#     staticmethod
#         no automatic self/cls
#         utility logically grouped with class
#
#
#     property
#         attribute-like interface
#
#
#     special method
#         connects object to Python protocols


# ============================================================
# 241. OOP CHEAT SHEET — PYTHON PROTOCOLS
# ============================================================
#
#     len(x)
#         → __len__
#
#     x[i]
#         → __getitem__
#
#     for item in x
#         → __iter__
#
#     next(x)
#         → __next__
#
#     x(...)
#         → __call__
#
#     str(x)
#         → __str__
#
#     repr(x)
#         → __repr__
#
#     x == y
#         → __eq__
#
#     x + y
#         → __add__
#
#     with x
#         → __enter__ / __exit__


# ============================================================
# 242. OOP CHEAT SHEET — BUILT-IN INSPECTION TOOLS
# ============================================================
#
#     type(obj)
#         → object's type
#
#     isinstance(obj, Class)
#         → instance check
#
#     id(obj)
#         → object identity
#
#     dir(obj)
#         → available attributes
#
#     obj.__dict__
#         → instance namespace when available
#
#     Class.__dict__
#         → class namespace mapping
#
#     Class.mro()
#         → method resolution order
#
#     callable(obj)
#         → whether callable


# ============================================================
# 243. OOP MASTER DIAGRAM
# ============================================================
#
#
#                         CLASS
#                           │
#                 ┌─────────┴─────────┐
#                 │                   │
#              STATE              BEHAVIOR
#                 │                   │
#          attributes              methods
#                 │                   │
#          ┌──────┴──────┐      ┌─────┴─────┐
#          ↓             ↓      ↓           ↓
#       instance       class   instance    class
#       attributes   attributes methods    methods
#                                  │
#                         ┌────────┼────────┐
#                         ↓        ↓        ↓
#                       normal   property  static
#
#
#                         CLASS
#                           │
#                      inheritance
#                           ↓
#                      SUBCLASS
#                           │
#                 ┌─────────┼─────────┐
#                 ↓         ↓         ↓
#              override   extend   super()
#
#
#                    OBJECT RELATIONSHIPS
#                           │
#             ┌─────────────┼─────────────┐
#             ↓             ↓             ↓
#           IS-A          HAS-A          USES
#             ↓             ↓             ↓
#       inheritance    composition    association
#
#
# ============================================================
# 244. THE COMPLETE OOP MENTAL MODEL
# ============================================================
#
#     CLASS
#       ↓
#     describes a kind of object
#       ↓
#     INSTANCE
#       ↓
#     stores state
#       ↓
#     methods operate on state
#       ↓
#     properties control derived/accessed state
#       ↓
#     inheritance can reuse/extend behavior
#       ↓
#     polymorphism lets different objects
#     share an interface
#       ↓
#     composition lets objects collaborate
#       ↓
#     protocols make objects integrate naturally
#     with Python syntax
#
#
# ============================================================
# 245. THE "PYTHON FEELING" OF OOP
# ============================================================
#
# When you see:
#
#     class Student:
#
# think:
#
#     "I am defining a new kind of object."
#
#
# When you see:
#
#     student = Student(...)
#
# think:
#
#     "I am creating one actual object."
#
#
# When you see:
#
#     self.name
#
# think:
#
#     "This object's own state."
#
#
# When you see:
#
#     def method(self):
#
# think:
#
#     "Something this object can do."
#
#
# When you see:
#
#     @classmethod
#
# think:
#
#     "Behavior associated with the class itself."
#
#
# When you see:
#
#     @staticmethod
#
# think:
#
#     "A utility grouped with the class."
#
#
# When you see:
#
#     @property
#
# think:
#
#     "An attribute-like interface backed by logic."
#
#
# When you see:
#
#     class Child(Parent):
#
# think:
#
#     "Child is a specialized form of Parent."
#
#
# When you see:
#
#     super()
#
# think:
#
#     "Continue behavior through the inheritance MRO."
#
#
# When you see:
#
#     obj.method()
#
# think:
#
#     "Send a request to this object."
#
#
# When you see:
#
#     object1 + object2
#
# think:
#
#     "Python may be invoking an operator protocol."
#
#
# When you see:
#
#     for x in object:
#
# think:
#
#     "This object implements an iteration protocol."


# ============================================================
# 246. FINAL DECISION TREE
# ============================================================
#
# START
#   │
#   ↓
# Do I just need reusable computation?
#   │
#   ├── YES → function
#   │
#   └── NO
#        ↓
# Does the concept have persistent state?
#        │
#        ├── NO → function/module/dataclass as appropriate
#        │
#        └── YES
#             ↓
#       Does it represent a domain object?
#             │
#             ├── YES → class
#             │
#             └── NO
#                  ↓
#          Is it primarily structured data?
#                  │
#                  ├── YES → dataclass
#                  │
#                  └── NO → consider class/composition
#
#
# Then:
#
#     Need reuse across types?
#         → polymorphism/protocol
#
#     Genuine IS-A relationship?
#         → inheritance
#
#     HAS-A relationship?
#         → composition
#
#     Need validation?
#         → property / method / constructor
#
#     Need alternative construction?
#         → classmethod
#
#     Need Python syntax integration?
#         → special methods/protocols


# ============================================================
# 247. WHAT YOU SHOULD BE ABLE TO BUILD AFTER THIS
# ============================================================
#
# You should eventually be able to build:
#
#     [ ] Student class
#     [ ] BankAccount class
#     [ ] Dataset class
#     [ ] DataCleaner
#     [ ] DataValidator
#     [ ] File/resource manager
#     [ ] ML model-like class
#     [ ] interchangeable model implementations
#     [ ] dataclass-based records
#     [ ] iterable custom datasets
#     [ ] callable objects
#     [ ] context managers
#     [ ] simple class hierarchies
#     [ ] composition-based systems
#     [ ] protocol-based APIs
#
#
# More importantly, you should understand:
#
#     WHY a class exists
#
# rather than simply knowing:
#
#     HOW to write class syntax.


# ============================================================
# 248. FINAL OOP LEARNING CHECKLIST
# ============================================================
#
# FOUNDATION
#
#     [ ] class
#     [ ] object
#     [ ] instance
#     [ ] type
#     [ ] identity
#     [ ] isinstance
#
#
# STATE
#
#     [ ] instance attributes
#     [ ] class attributes
#     [ ] self
#     [ ] __init__
#     [ ] __dict__
#     [ ] mutable state
#
#
# METHODS
#
#     [ ] instance methods
#     [ ] classmethod
#     [ ] staticmethod
#     [ ] property
#     [ ] setter
#     [ ] method binding
#
#
# INHERITANCE
#
#     [ ] inheritance
#     [ ] overriding
#     [ ] super
#     [ ] polymorphism
#     [ ] duck typing
#     [ ] ABC
#
#
# COMPOSITION
#
#     [ ] composition
#     [ ] aggregation
#     [ ] association
#     [ ] dependency injection
#
#
# PYTHON OBJECT MODEL
#
#     [ ] __str__
#     [ ] __repr__
#     [ ] __len__
#     [ ] __getitem__
#     [ ] __setitem__
#     [ ] __contains__
#     [ ] __eq__
#     [ ] __lt__
#     [ ] __add__
#     [ ] __call__
#     [ ] __iter__
#     [ ] __next__
#     [ ] __enter__
#     [ ] __exit__
#
#
# ADVANCED
#
#     [ ] dataclasses
#     [ ] Protocol
#     [ ] multiple inheritance
#     [ ] MRO
#     [ ] mixins
#     [ ] __new__
#     [ ] __slots__
#     [ ] hashing
#
#
# DESIGN
#
#     [ ] single responsibility
#     [ ] composition vs inheritance
#     [ ] clear interfaces
#     [ ] object invariants
#     [ ] testability
#     [ ] domain modeling
#
#
# ============================================================
# 249. THE ONE-PAGE OOP SUMMARY
# ============================================================
#
#     CLASS
#         ↓
#     defines a type
#
#     OBJECT
#         ↓
#     instance of that type
#
#     ATTRIBUTE
#         ↓
#     data/state or class information
#
#     METHOD
#         ↓
#     behavior associated with object/class
#
#     self
#         ↓
#     current instance
#
#     cls
#         ↓
#     current class
#
#     __init__
#         ↓
#     initialize instance state
#
#     property
#         ↓
#     attribute-like controlled behavior
#
#     inheritance
#         ↓
#     IS-A relationship
#
#     composition
#         ↓
#     HAS-A relationship
#
#     polymorphism
#         ↓
#     same interface, different behavior
#
#     duck typing
#         ↓
#     behavior matters more than exact type
#
#     ABC / Protocol
#         ↓
#     define expected interfaces
#
#     dataclass
#         ↓
#     convenient structured data class
#
#     special methods
#         ↓
#     integrate objects with Python syntax
#
#
# ============================================================
# 250. THE MOST IMPORTANT IDEA
# ============================================================
#
# OOP is NOT:
#
#     "Put everything into classes."
#
#
# OOP IS:
#
#     "Model related state and behavior as objects,
#      define clear interfaces between those objects,
#      and choose relationships that represent
#      the actual problem."
#
#
# The progression to remember:
#
#
#     VARIABLE
#        ↓
#     stores a value
#
#     FUNCTION
#        ↓
#     stores reusable behavior
#
#     CLASS
#        ↓
#     defines a kind of object
#
#     OBJECT
#        ↓
#     combines state + behavior
#
#     COMPOSITION
#        ↓
#     combines objects
#
#     POLYMORPHISM
#        ↓
#     lets different objects share interfaces
#
#     PROTOCOLS
#        ↓
#     let objects integrate naturally
#     with Python's way of doing things
#
#
# ============================================================
# 251. FINAL PYTHON FEELING
# ============================================================
#
#     DATA
#       +
#     BEHAVIOR
#       ↓
#     OBJECT
#
#
#     OBJECT
#       +
#     DEFINED INTERFACE
#       ↓
#     COMPONENT
#
#
#     COMPONENT
#       +
#     OTHER COMPONENTS
#       ↓
#     SYSTEM
#
#
# For data science and machine learning,
# this becomes:
#
#     raw data
#        ↓
#     data objects / records
#        ↓
#     cleaning components
#        ↓
#     validation components
#        ↓
#     feature-processing components
#        ↓
#     model object
#        ↓
#     fit()
#        ↓
#     learned state
#        ↓
#     predict()
#        ↓
#     evaluation
#
#
# And the most important practical rule:
#
#     SIMPLE PROBLEM
#         → simple function/data structure
#
#     COMPLEX STATE + BEHAVIOR
#         → class
#
#     MANY COLLABORATING CONCEPTS
#         → objects + composition
#
#     SAME INTERFACE, DIFFERENT IMPLEMENTATIONS
#         → polymorphism
#
#     PYTHON-NATIVE BEHAVIOR
#         → protocols / special methods
#
#
# ============================================================
# END
# ============================================================
```
