```python
# ============================================================
# PYTHON STANDARD LIBRARY
# COMPLETE PRACTICAL LEARNING GUIDE
#
# json, csv, datetime, pathlib, os
#
# Goal:
#   Learn these libraries conceptually, not by memorizing syntax.
#
# Main idea:
#
#   JSON  → structured data ↔ text
#   CSV   → table ↔ text
#   datetime → dates + times + durations
#   pathlib → filesystem paths as Python objects
#   os    → operating-system interaction
#
# These five modules are extremely useful for:
#   - Data Analysis
#   - Data Cleaning
#   - Automation
#   - File Processing
#   - Research
#   - Machine Learning pipelines
#   - Reading/writing datasets
#   - Creating project folders
#   - Logging
#   - Working with dates
#   - Building reproducible workflows
# ============================================================


# ============================================================
# 0. MASTER MAP
# ============================================================

# Think of the five libraries as solving different problems:
#
#
#                 YOUR PYTHON PROGRAM
#                         |
#          +--------------+--------------+
#          |              |              |
#          v              v              v
#       DATA FILES      TIME/DATE      COMPUTER
#          |              |              |
#      +---+---+          |          +---+---+
#      |       |          |          |       |
#      v       v          v          v       v
#     JSON    CSV      datetime    pathlib   os
#
#
# JSON
# ────
# Python objects ↔ JSON text
#
# CSV
# ───
# Rows/columns ↔ text file
#
# datetime
# ─────────
# date + time + duration + timezone
#
# pathlib
# ────────
# Paths and filesystem operations using objects
#
# os
# ──
# Operating-system environment and lower-level filesystem operations


# ============================================================
# 1. FIRST: IMPORTING THE LIBRARIES
# ============================================================

import json
import csv
import datetime
from pathlib import Path
import os


# ============================================================
# 2. WHEN SHOULD I USE WHICH LIBRARY?
# ============================================================

# ┌──────────────┬─────────────────────────────────────────────┐
# │ Problem      │ Best starting point                         │
# ├──────────────┼─────────────────────────────────────────────┤
# │ JSON data    │ json                                        │
# │ Spreadsheet- │ csv                                         │
# │ like data    │                                             │
# │ Dates/times  │ datetime                                    │
# │ File paths   │ pathlib                                     │
# │ OS settings  │ os                                          │
# │ Environment  │ os.environ                                 │
# │ Current      │ datetime.datetime.now()                     │
# │ date         │                                             │
# │ Create file  │ pathlib.Path                                │
# │ Rename file  │ pathlib.Path.rename()                       │
# │ List files   │ Path.iterdir() / Path.glob()                │
# │ Make folder  │ Path.mkdir()                                │
# │ Delete file  │ Path.unlink()                               │
# └──────────────┴─────────────────────────────────────────────┘
#
#
# IMPORTANT PRACTICAL RULE:
#
# For modern Python code:
#
#     pathlib
#
# is usually the clearest choice for working with paths.
#
# os is still extremely important for:
#     environment variables
#     operating-system information
#     process/environment interaction
#     some lower-level filesystem operations
#
#
# Simple feeling:
#
#     pathlib → "Where is the file and what should I do with it?"
#     os      → "What is the operating system/environment doing?"


# ============================================================
# 3. JSON — THE BASIC IDEA
# ============================================================

# JSON = JavaScript Object Notation
#
# JSON is a TEXT FORMAT.
#
# It is not the same thing as a Python dictionary.
#
#
# Python dictionary
#
#     {
#         "name": "Ali",
#         "age": 25
#     }
#
#
# JSON representation
#
#     {
#         "name": "Ali",
#         "age": 25
#     }
#
# They look almost identical, but one is a Python object
# and the other is serialized text/data format.
#
#
# The fundamental transformation:
#
#
#       Python object
#            |
#            | json.dumps()
#            v
#       JSON string
#            |
#            | json.loads()
#            v
#       Python object
#
#
# File version:
#
#       Python object
#            |
#            | json.dump()
#            v
#       .json file
#            |
#            | json.load()
#            v
#       Python object


# ============================================================
# 4. JSON ↔ PYTHON TYPE MAPPING
# ============================================================

# ┌──────────────────┬──────────────────────┐
# │ Python           │ JSON                 │
# ├──────────────────┼──────────────────────┤
# │ dict             │ object               │
# │ list             │ array                │
# │ str              │ string               │
# │ int              │ number               │
# │ float            │ number               │
# │ True             │ true                 │
# │ False            │ false                │
# │ None             │ null                 │
# └──────────────────┴──────────────────────┘


# ============================================================
# 5. JSON: dumps() — PYTHON OBJECT → JSON STRING
# ============================================================

data = {
    "name": "Ali",
    "age": 25,
    "scores": [85, 91, 78],
    "active": True,
    "address": None
}

json_text = json.dumps(data)

print(json_text)

# json.dumps()
#
#     Python object
#           ↓
#     JSON-formatted string
#
#
# IMPORTANT:
#
# dumps = dump string
#
# Think:
#
#     s = string
#
#     dumps() → produces a string
#
#
# dump() → writes to a file


# ============================================================
# 6. JSON: loads() — JSON STRING → PYTHON OBJECT
# ============================================================

json_text = '{"name": "Sara", "age": 28, "active": true}'

data = json.loads(json_text)

print(data)
print(type(data))

# Result:
#
#     {'name': 'Sara', 'age': 28, 'active': True}
#
#     <class 'dict'>
#
#
# loads:
#
#     JSON string
#          ↓
#     Python object
#
#
# Again:
#
#     loads → load string
#     load  → load file


# ============================================================
# 7. JSON: dump() — WRITE JSON TO A FILE
# ============================================================

data = {
    "name": "Yasin",
    "skills": ["Python", "SQL", "Power BI"],
    "experience": 7
}

with open("person.json", "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4)

# Resulting file:
#
# {
#     "name": "Yasin",
#     "skills": [
#         "Python",
#         "SQL",
#         "Power BI"
#     ],
#     "experience": 7
# }


# ============================================================
# 8. JSON: load() — READ JSON FROM A FILE
# ============================================================

with open("person.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print(data)

# Flow:
#
#     person.json
#          |
#          v
#     json.load()
#          |
#          v
#     Python dictionary


# ============================================================
# 9. JSON: dumps() vs dump() vs loads() vs load()
# ============================================================

# ┌────────────┬──────────────────────────────┐
# │ Function   │ Meaning                      │
# ├────────────┼──────────────────────────────┤
# │ dumps()    │ Python → JSON string         │
# │ dump()     │ Python → JSON file           │
# │ loads()    │ JSON string → Python         │
# │ load()     │ JSON file → Python           │
# └────────────┴──────────────────────────────┘
#
#
# MEMORY TRICK:
#
#     S = String
#
#     dumps  → string
#     loads  → string
#
#     dump   → file
#     load   → file


# ============================================================
# 10. JSON FORMATTING
# ============================================================

data = {
    "name": "Ali",
    "age": 25,
    "skills": ["Python", "SQL"]
}

print(json.dumps(data))

print(json.dumps(data, indent=4))

# indent=4 makes JSON human-readable.
#
# Useful for:
#     configuration files
#     debugging
#     research datasets
#     API responses
#     project settings


# ============================================================
# 11. JSON sort_keys
# ============================================================

data = {
    "z": 10,
    "a": 20,
    "m": 30
}

print(json.dumps(data, indent=4, sort_keys=True))

# sort_keys=True
#
# alphabetically sorts dictionary keys in the output.
#
# Useful for:
#     reproducible files
#     readable configuration
#     comparing JSON files


# ============================================================
# 12. JSON ensure_ascii
# ============================================================

data = {
    "name": "یاسین",
    "city": "کرج"
}

print(json.dumps(data))

print(json.dumps(data, ensure_ascii=False))

# ensure_ascii=False
#
# preserves Unicode characters directly.
#
# Very useful for:
#     Persian
#     Arabic
#     Chinese
#     Japanese
#     other Unicode text


# ============================================================
# 13. JSON COMMON PRACTICAL STRUCTURE
# ============================================================

students = [
    {
        "name": "Ali",
        "age": 25,
        "score": 88
    },
    {
        "name": "Sara",
        "age": 27,
        "score": 94
    }
]

with open("students.json", "w", encoding="utf-8") as file:
    json.dump(students, file, indent=4, ensure_ascii=False)

# This is a very common pattern:
#
#     list
#       |
#       +--- dictionary
#       +--- dictionary
#       +--- dictionary
#
# This structure is especially useful when records have
# different fields or nested information.


# ============================================================
# 14. JSON NESTED DATA
# ============================================================

company = {
    "name": "Example Company",
    "employees": [
        {
            "name": "Ali",
            "department": "IT",
            "skills": ["Python", "SQL"]
        },
        {
            "name": "Sara",
            "department": "Data",
            "skills": ["Python", "Power BI"]
        }
    ]
}

# Access:
print(company["name"])
print(company["employees"][0]["name"])
print(company["employees"][1]["skills"][0])

#
# Mental model:
#
# company
#   |
#   +── name
#   |
#   +── employees
#         |
#         +── [0]
#         |     |
#         |     +── name
#         |     +── department
#         |     +── skills
#         |
#         +── [1]
#               |
#               +── name
#               +── department
#               +── skills


# ============================================================
# 15. JSON ERROR HANDLING
# ============================================================

bad_json = '{"name": "Ali", "age": 25'

try:
    data = json.loads(bad_json)
except json.JSONDecodeError as error:
    print("Invalid JSON:", error)

# JSONDecodeError means:
#
#     Python tried to interpret JSON
#     but the JSON syntax was invalid.


# ============================================================
# 16. JSON WHAT CANNOT BE DIRECTLY SERIALIZED?
# ============================================================

# JSON naturally supports:
#
#     dict
#     list
#     str
#     int
#     float
#     bool
#     None
#
# But arbitrary Python objects are not automatically JSON.
#
# Example:
#
#     from datetime import datetime
#     datetime.now()
#
# cannot normally be passed directly to json.dumps().
#
# Convert it to a suitable representation first.
#
# Common strategy:
#
#     datetime → ISO string
#     Path     → string
#     custom object → dictionary


# ============================================================
# 17. JSON PRACTICAL: CONFIGURATION FILE
# ============================================================

# config.json might contain:
#
# {
#     "input_folder": "data",
#     "output_folder": "results",
#     "threshold": 0.75,
#     "debug": true
# }
#
# Python:

with open("config.json", "w", encoding="utf-8") as file:
    json.dump(
        {
            "input_folder": "data",
            "output_folder": "results",
            "threshold": 0.75,
            "debug": True
        },
        file,
        indent=4
    )

with open("config.json", "r", encoding="utf-8") as file:
    config = json.load(file)

threshold = config["threshold"]


# ============================================================
# 18. CSV — THE BASIC IDEA
# ============================================================

# CSV = Comma-Separated Values
#
# Conceptually:
#
#     rows + columns
#
#
# Example:
#
# name,age,score
# Ali,25,88
# Sara,27,94
# Reza,31,76
#
#
# Think:
#
#       CSV FILE
#
#       ┌──────┬─────┬───────┐
#       │ name │ age │ score │
#       ├──────┼─────┼───────┤
#       │ Ali  │ 25  │ 88    │
#       │ Sara │ 27  │ 94    │
#       │ Reza │ 31  │ 76    │
#       └──────┴─────┴───────┘
#
#
# CSV is excellent for rectangular/tabular data.
#
# JSON is often better for nested/hierarchical data.


# ============================================================
# 19. CSV: WRITING ROWS
# ============================================================

rows = [
    ["name", "age", "score"],
    ["Ali", 25, 88],
    ["Sara", 27, 94],
    ["Reza", 31, 76]
]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    for row in rows:
        writer.writerow(row)

# csv.writer()
#
# creates a CSV writer object.
#
# writer.writerow()
#
# writes one row.


# ============================================================
# 20. CSV: WRITEROWS
# ============================================================

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerows([
        ["name", "age", "score"],
        ["Ali", 25, 88],
        ["Sara", 27, 94],
        ["Reza", 31, 76]
    ])

# writerows()
#
# writes many rows at once.


# ============================================================
# 21. CSV: READING ROWS
# ============================================================

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)

# Each row is normally a list of strings.
#
# IMPORTANT:
#
# CSV does NOT automatically know that:
#
#     "25"
#
# means integer 25.
#
# You may need conversion:
#
#     age = int(row[1])
#     score = float(row[2])


# ============================================================
# 22. CSV: DictWriter
# ============================================================

students = [
    {"name": "Ali", "age": 25, "score": 88},
    {"name": "Sara", "age": 27, "score": 94},
    {"name": "Reza", "age": 31, "score": 76}
]

fieldnames = ["name", "age", "score"]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(students)

# This is often much easier to understand than csv.writer()
# when your data naturally exists as dictionaries.


# ============================================================
# 23. CSV: DictReader
# ============================================================

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
        print(row["name"])
        print(row["score"])

# Each row behaves like a dictionary:
#
#     {
#         "name": "Ali",
#         "age": "25",
#         "score": "88"
#     }
#
# Again:
#
# VALUES ARE STRINGS.
#
# Convert when necessary.


# ============================================================
# 24. CSV READING FLOW
# ============================================================

#       students.csv
#            |
#            v
#       csv.DictReader
#            |
#            v
#       row dictionary
#            |
#            +── name
#            +── age
#            +── score
#            |
#            v
#       convert / clean
#            |
#            v
#       analysis


# ============================================================
# 25. CSV DELIMITER
# ============================================================

# CSV does not have to use a comma.
#
# Some files use:
#
#     ;
#     \t
#     |
#
# Example:

with open("data.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file, delimiter=";")

    for row in reader:
        print(row)

# delimiter=";"
#
# tells Python which character separates columns.


# ============================================================
# 26. CSV QUOTING
# ============================================================

# Consider:
#
# name,city,comment
# Ali,Karaj,"Good result, needs review"
#
# The comma inside the comment should NOT create another column.
#
# The csv module understands CSV quoting rules.
#
# This is why using csv.reader() is safer than simply:
#
#     line.split(",")
#
# Do not assume:
#
#     comma = always simple separator
#
# Real CSV files can contain:
#     commas inside fields
#     quotes
#     newlines inside quoted fields
#     missing values


# ============================================================
# 27. CSV PRACTICAL DATA CLEANING
# ============================================================

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    clean_students = []

    for row in reader:
        score_text = row["score"].strip()

        if score_text == "":
            continue

        row["name"] = row["name"].strip()
        row["age"] = int(row["age"])
        row["score"] = float(score_text)

        clean_students.append(row)

# Practical pipeline:
#
#     CSV
#      ↓
#     read
#      ↓
#     strip whitespace
#      ↓
#     validate missing values
#      ↓
#     convert types
#      ↓
#     clean records
#      ↓
#     analysis


# ============================================================
# 28. JSON vs CSV
# ============================================================

# ┌────────────────┬──────────────────────┬──────────────────────┐
# │ Feature        │ CSV                  │ JSON                 │
# ├────────────────┼──────────────────────┼──────────────────────┤
# │ Main structure │ rows/columns         │ nested structures    │
# │ Human readable │ yes                  │ yes                  │
# │ Nested data    │ poor                 │ excellent             │
# │ Tabular data   │ excellent            │ possible              │
# │ Data types     │ mostly text          │ richer                │
# │ APIs            │ less common          │ extremely common     │
# │ Excel-like     │ excellent            │ not ideal             │
# └────────────────┴──────────────────────┴──────────────────────┘
#
#
# Simple decision:
#
#     TABLE?
#       ↓
#     CSV
#
#     NESTED / HIERARCHICAL?
#       ↓
#     JSON


# ============================================================
# 29. DATETIME — THE BASIC IDEA
# ============================================================

# datetime is used for:
#
#     date
#     time
#     date + time
#     duration
#     timezone-aware date/time
#
#
# Main classes:
#
#     date
#     time
#     datetime
#     timedelta
#     timezone
#
#
# Mental model:
#
#
#     DATE
#       │
#       └── 2026-09-09
#
#     TIME
#       │
#       └── 15:30:45
#
#     DATETIME
#       │
#       └── 2026-09-09 15:30:45
#
#     TIME DIFFERENCE
#       │
#       └── 3 days, 2 hours
#              ↑
#          timedelta


# ============================================================
# 30. datetime.date
# ============================================================

from datetime import date

today = date.today()

print(today)
print(type(today))

# Example:
#
#     2026-09-09
#
# date contains:
#
#     year
#     month
#     day
#
# but NOT time of day.


# ============================================================
# 31. date() CONSTRUCTION
# ============================================================

birthday = date(1990, 5, 17)

print(birthday)
print(birthday.year)
print(birthday.month)
print(birthday.day)

# date(year, month, day)


# ============================================================
# 32. datetime.datetime.now()
# ============================================================

from datetime import datetime

now = datetime.now()

print(now)

# Contains:
#
#     year
#     month
#     day
#     hour
#     minute
#     second
#     microsecond


# ============================================================
# 33. DATETIME COMPONENTS
# ============================================================

now = datetime.now()

print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
print(now.microsecond)

# Mental model:
#
# datetime object
#
#     ┌───────────────────────────────────┐
#     │ 2026-09-09 15:30:45.123456        │
#     └───────────────────────────────────┘
#       │    │  │   │  │  │
#       │    │  │   │  │  └─ microsecond
#       │    │  │   │  └──── second
#       │    │  │   └─────── minute
#       │    │  └─────────── hour
#       │    └────────────── day/month
#       └────────────────── year


# ============================================================
# 34. datetime() CONSTRUCTION
# ============================================================

dt = datetime(2026, 9, 9, 15, 30, 45)

print(dt)

# datetime(
#     year,
#     month,
#     day,
#     hour,
#     minute,
#     second
# )


# ============================================================
# 35. datetime.today()
# ============================================================

today_datetime = datetime.today()

print(today_datetime)

# For most ordinary local-time usage:
#
#     datetime.now()
#
# and:
#
#     datetime.today()
#
# produce the current local datetime.
#
# datetime.now() is generally more flexible because it can
# accept a timezone.


# ============================================================
# 36. date.today() vs datetime.now()
# ============================================================

# date.today()
#
#     2026-09-09
#
# datetime.now()
#
#     2026-09-09 15:30:45.123456
#
#
# Use:
#
#     date
#
# when you only care about calendar date.
#
# Use:
#
#     datetime
#
# when time-of-day matters.


# ============================================================
# 37. datetime.combine()
# ============================================================

from datetime import time

d = date(2026, 9, 9)
t = time(14, 30)

dt = datetime.combine(d, t)

print(dt)

# Combines:
#
#     date + time
#          ↓
#       datetime


# ============================================================
# 38. datetime.fromisoformat()
# ============================================================

text = "2026-09-09 14:30:00"

dt = datetime.fromisoformat(text)

print(dt)

# ISO-style dates are extremely common in:
#
#     APIs
#     databases
#     logs
#     JSON
#     scientific datasets


# ============================================================
# 39. date.fromisoformat()
# ============================================================

d = date.fromisoformat("2026-09-09")

print(d)


# ============================================================
# 40. strptime() — STRING → DATETIME
# ============================================================

text = "09/09/2026"

dt = datetime.strptime(text, "%d/%m/%Y")

print(dt)

# strptime means:
#
#     string parse time
#
#
# STRING
#   ↓
# strptime()
#   ↓
# datetime object


# ============================================================
# 41. STRPTIME FORMAT CODES
# ============================================================

# ┌────────┬──────────────────────┐
# │ Code   │ Meaning              │
# ├────────┼──────────────────────┤
# │ %Y     │ 4-digit year         │
# │ %y     │ 2-digit year         │
# │ %m     │ month 01-12          │
# │ %d     │ day 01-31            │
# │ %H     │ hour 00-23           │
# │ %I     │ hour 01-12           │
# │ %M     │ minute               │
# │ %S     │ second               │
# │ %f     │ microsecond          │
# │ %p     │ AM/PM                │
# │ %a     │ abbreviated weekday  │
# │ %A     │ full weekday         │
# │ %b     │ abbreviated month    │
# │ %B     │ full month           │
# └────────┴──────────────────────┘


# ============================================================
# 42. MORE STRPTIME EXAMPLES
# ============================================================

datetime.strptime("2026-09-09", "%Y-%m-%d")

datetime.strptime("09/09/2026", "%d/%m/%Y")

datetime.strptime(
    "2026-09-09 15:30:45",
    "%Y-%m-%d %H:%M:%S"
)

datetime.strptime(
    "09-09-2026 03:30 PM",
    "%d-%m-%Y %I:%M %p"
)


# ============================================================
# 43. strftime() — DATETIME → STRING
# ============================================================

dt = datetime(2026, 9, 9, 15, 30, 45)

text = dt.strftime("%Y-%m-%d")

print(text)

# Opposite direction:
#
#
#     strptime
#     STRING → DATETIME
#
#     strftime
#     DATETIME → STRING
#
#
# MEMORY:
#
#     p = parse
#     f = format
#
#     strptime → parse a string
#     strftime → format a datetime


# ============================================================
# 44. DATETIME FORMATTING EXAMPLES
# ============================================================

dt = datetime(2026, 9, 9, 15, 30, 45)

print(dt.strftime("%Y-%m-%d"))
print(dt.strftime("%d/%m/%Y"))
print(dt.strftime("%Y/%m/%d %H:%M:%S"))
print(dt.strftime("%B %d, %Y"))
print(dt.strftime("%A"))
print(dt.strftime("%H:%M"))


# ============================================================
# 45. TIME OBJECT
# ============================================================

t = time(14, 30, 45)

print(t)
print(t.hour)
print(t.minute)
print(t.second)

# time represents:
#
#     hour
#     minute
#     second
#     microsecond
#
# without a calendar date.


# ============================================================
# 46. timedelta — DURATION
# ============================================================

from datetime import timedelta

duration = timedelta(days=3, hours=5, minutes=20)

print(duration)

# timedelta represents an amount of time.
#
#
# datetime
#     = a POINT in time
#
# timedelta
#     = DISTANCE between times


# ============================================================
# 47. DATETIME + TIMEDELTA
# ============================================================

now = datetime.now()

future = now + timedelta(days=7)

print(now)
print(future)

# Diagram:
#
# NOW
#  |
#  | + 7 days
#  v
# FUTURE


# ============================================================
# 48. DATETIME - TIMEDELTA
# ============================================================

deadline = datetime.now()

earlier = deadline - timedelta(days=30)

print(earlier)


# ============================================================
# 49. DATETIME - DATETIME
# ============================================================

start = datetime(2026, 9, 1)
end = datetime(2026, 9, 9)

difference = end - start

print(difference)
print(difference.days)

# Result:
#
#     8 days
#
#
# datetime
#     -
# datetime
#     =
# timedelta


# ============================================================
# 50. TIMEDELTA COMPONENTS
# ============================================================

duration = timedelta(
    days=2,
    hours=5,
    minutes=30,
    seconds=20
)

print(duration.days)
print(duration.seconds)
print(duration.microseconds)
print(duration.total_seconds())

# IMPORTANT:
#
# duration.seconds is NOT always "total seconds".
#
# It represents the seconds portion after normalization.
#
# For total duration:
#
#     duration.total_seconds()


# ============================================================
# 51. PRACTICAL: AGE CALCULATION
# ============================================================

birth_date = date(1990, 5, 17)
today = date.today()

age = today.year - birth_date.year

if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1

print(age)

# Why not simply:
#
#     today.year - birth_date.year
#
# Because the birthday may not have happened yet this year.


# ============================================================
# 52. PRACTICAL: DAYS UNTIL DEADLINE
# ============================================================

today = date.today()
deadline = date(2026, 12, 31)

remaining = deadline - today

print(remaining.days)

# Pattern:
#
#     deadline - today
#          ↓
#      timedelta
#          ↓
#      .days


# ============================================================
# 53. PRACTICAL: CHECK IF DATE IS IN THE PAST
# ============================================================

deadline = date(2026, 12, 31)

if deadline < date.today():
    print("Deadline has passed.")
else:
    print("Deadline is still ahead.")


# ============================================================
# 54. PRACTICAL: DATE RANGE
# ============================================================

start = date(2026, 9, 1)
end = date(2026, 9, 10)

current = start

while current <= end:
    print(current)
    current += timedelta(days=1)

# Diagram:
#
# Sep 1 → Sep 2 → Sep 3 → ... → Sep 10
#   ↑
# current
#
# Each step:
#
#     current += timedelta(days=1)


# ============================================================
# 55. COMPARING DATES
# ============================================================

date1 = date(2026, 1, 1)
date2 = date(2026, 12, 31)

print(date1 < date2)
print(date1 == date2)
print(date1 > date2)


# ============================================================
# 56. IMPORTANT DATETIME CONCEPT:
#    NAIVE vs AWARE
# ============================================================

# NAIVE datetime:
#
#     datetime.now()
#
# has no timezone information attached.
#
#
# AWARE datetime:
#
#     datetime.now(timezone.utc)
#
# contains timezone information.
#
#
# Mental model:
#
#     NAIVE
#       |
#       +-- "2026-09-09 15:30"
#       |
#       └── What timezone?
#              UNKNOWN
#
#
#     AWARE
#       |
#       +-- "2026-09-09 15:30 +00:00"
#       |
#       └── timezone known


# ============================================================
# 57. UTC
# ============================================================

from datetime import timezone

utc_now = datetime.now(timezone.utc)

print(utc_now)

# For systems, APIs, logs, distributed applications:
#
# UTC is often a very useful standard.
#
# General idea:
#
#     Store/communicate timestamps consistently
#     ↓
#     convert to local timezone for display


# ============================================================
# 58. timezone CONCEPT
# ============================================================

# The datetime module includes timezone for fixed offsets.
#
# Example:

offset = timezone(timedelta(hours=3, minutes=30))

dt = datetime.now(offset)

print(dt)

# For full named timezone databases and daylight-saving rules,
# Python applications commonly use:
#
#     zoneinfo
#
# which is part of the standard library.
#
# Example:
#
#     from zoneinfo import ZoneInfo
#
#     dt = datetime.now(ZoneInfo("Asia/Tehran"))
#
# This is preferable to manually applying offsets when real
# timezone rules matter.


# ============================================================
# 59. datetime IMPORTANT METHODS / ATTRIBUTES
# ============================================================

# datetime/date methods worth knowing:
#
#     .date()
#     .time()
#     .replace()
#     .strftime()
#     .isoformat()
#     .weekday()
#     .isoweekday()
#     .isocalendar()
#
#
# datetime constructors:
#
#     .now()
#     .today()
#     .fromisoformat()
#     .strptime()
#
#
# date constructors:
#
#     .today()
#     .fromisoformat()
#     .fromordinal()
#
# time:
#
#     .fromisoformat()


# ============================================================
# 60. datetime.date() and .time()
# ============================================================

dt = datetime(2026, 9, 9, 15, 30, 45)

print(dt.date())
print(dt.time())

# datetime
#    |
#    +── .date() → date
#    |
#    └── .time() → time


# ============================================================
# 61. replace()
# ============================================================

dt = datetime(2026, 9, 9, 15, 30)

new_dt = dt.replace(hour=18)

print(dt)
print(new_dt)

# IMPORTANT:
#
# datetime objects are immutable.
#
# replace() does NOT modify dt.
#
# It returns a new datetime object.


# ============================================================
# 62. isoformat()
# ============================================================

dt = datetime(2026, 9, 9, 15, 30, 45)

print(dt.isoformat())

# Example:
#
#     2026-09-09T15:30:45
#
# ISO 8601 format is excellent for:
#
#     APIs
#     JSON
#     databases
#     logs
#     machine-readable timestamps


# ============================================================
# 63. WEEKDAY METHODS
# ============================================================

d = date(2026, 9, 9)

print(d.weekday())
print(d.isoweekday())

# weekday():
#
#     Monday = 0
#     Tuesday = 1
#     ...
#     Sunday = 6
#
#
# isoweekday():
#
#     Monday = 1
#     Tuesday = 2
#     ...
#     Sunday = 7


# ============================================================
# 64. PRACTICAL: FILTER DATA BY DATE
# ============================================================

records = [
    {"name": "A", "date": "2026-09-01"},
    {"name": "B", "date": "2026-09-05"},
    {"name": "C", "date": "2026-09-12"}
]

cutoff = date(2026, 9, 5)

for record in records:
    record_date = date.fromisoformat(record["date"])

    if record_date >= cutoff:
        print(record)

# This pattern appears constantly in data analysis:
#
#     text date
#        ↓
#     date object
#        ↓
#     compare
#        ↓
#     filter


# ============================================================
# 65. PATHLIB — THE BASIC IDEA
# ============================================================

# pathlib represents filesystem paths as Python objects.
#
# Instead of thinking:
#
#     "I have a string containing a path."
#
# think:
#
#     "I have a Path object representing a location."
#
#
# Example:
#
#     Path("data/students.csv")
#
#
# Mental model:
#
#
#     Path object
#          |
#          +── location
#          +── name
#          +── suffix
#          +── parent
#          +── operations
#          |
#          +── exists()
#          +── is_file()
#          +── is_dir()
#          +── read_text()
#          +── write_text()
#          +── mkdir()
#          +── rename()
#          +── unlink()


# ============================================================
# 66. CREATE A PATH
# ============================================================

from pathlib import Path

path = Path("data/students.csv")

print(path)

# Path is an object:
#
print(type(path))


# ============================================================
# 67. ABSOLUTE PATH
# ============================================================

path = Path("data/students.csv")

print(path.absolute())

# absolute()
#
# gives an absolute path representation.
#
# Important:
#
# absolute() does not necessarily mean "resolve every
# symbolic link and verify that the path exists".
#
# For a fully resolved existing path, see:
#
#     path.resolve()


# ============================================================
# 68. resolve()
# ============================================================

path = Path("data/students.csv")

print(path.resolve())

# resolve() produces an absolute, resolved path.
#
# Depending on Python/platform/version and existence,
# resolution behavior can differ.
#
# It is useful when you need a canonical filesystem location.


# ============================================================
# 69. PATH COMPONENTS
# ============================================================

path = Path("data/reports/students.csv")

print(path.name)
print(path.stem)
print(path.suffix)
print(path.parent)

# Results conceptually:
#
# name:
#     students.csv
#
# stem:
#     students
#
# suffix:
#     .csv
#
# parent:
#     data/reports


# ============================================================
# 70. PATH PARTS
# ============================================================

path = Path("data/reports/students.csv")

print(path.parts)

# Conceptually:
#
#     ("data", "reports", "students.csv")
#
# This lets you think of a path as components:
#
#     data
#       ↓
#     reports
#       ↓
#     students.csv


# ============================================================
# 71. PATH JOINING — / OPERATOR
# ============================================================

base = Path("data")

file_path = base / "students" / "2026" / "scores.csv"

print(file_path)

# This:
#
#     base / "students" / "2026" / "scores.csv"
#
# is one of pathlib's nicest features.
#
# It avoids manually writing:
#
#     "data/students/2026/scores.csv"
#
#
# pathlib handles platform-specific path separators.


# ============================================================
# 72. Path.cwd()
# ============================================================

current_folder = Path.cwd()

print(current_folder)

# cwd = current working directory
#
#
# Program
#    |
#    v
# current working directory
#    |
#    +── relative paths are interpreted from here


# ============================================================
# 73. Path.home()
# ============================================================

home = Path.home()

print(home)

# Gives the user's home directory.


# ============================================================
# 74. EXISTS / FILE / DIRECTORY
# ============================================================

path = Path("students.csv")

print(path.exists())
print(path.is_file())
print(path.is_dir())

# Three very important questions:
#
#     exists?
#     file?
#     directory?


# ============================================================
# 75. mkdir() — CREATE DIRECTORY
# ============================================================

folder = Path("results")

folder.mkdir(exist_ok=True)

# exist_ok=True
#
# means:
#
#     if the folder already exists,
#     do not raise an error.
#
#
# Nested folders:
#
folder = Path("project/data/raw")

folder.mkdir(parents=True, exist_ok=True)

# parents=True:
#
# create missing parent directories too.


# ============================================================
# 76. WRITE TEXT WITH pathlib
# ============================================================

path = Path("notes.txt")

path.write_text(
    "Python\nJSON\nCSV\n",
    encoding="utf-8"
)

# Path.write_text()
#
# writes text directly to a file.


# ============================================================
# 77. READ TEXT WITH pathlib
# ============================================================

text = path.read_text(encoding="utf-8")

print(text)

# Very convenient for small/medium text files.
#
# For large files, streaming with open() may be more appropriate.


# ============================================================
# 78. WRITE BYTES / READ BYTES
# ============================================================

path = Path("data.bin")

path.write_bytes(b"ABC")

data = path.read_bytes()

print(data)

# Text:
#
#     read_text()
#     write_text()
#
# Binary:
#
#     read_bytes()
#     write_bytes()


# ============================================================
# 79. OPEN A FILE THROUGH pathlib
# ============================================================

path = Path("students.csv")

with path.open("r", encoding="utf-8") as file:
    content = file.read()

# Path.open()
#
# is conceptually similar to:
#
#     open(path, ...)
#
# and integrates naturally with pathlib.


# ============================================================
# 80. PATHLIB + JSON
# ============================================================

path = Path("data") / "students.json"

path.parent.mkdir(parents=True, exist_ok=True)

data = {
    "students": 100,
    "average": 86.5
}

with path.open("w", encoding="utf-8") as file:
    json.dump(data, file, indent=4)

# Excellent combination:
#
#     pathlib → where?
#     json    → how is the data structured?


# ============================================================
# 81. PATHLIB + CSV
# ============================================================

path = Path("data") / "students.csv"

path.parent.mkdir(parents=True, exist_ok=True)

with path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "score"])
    writer.writerow(["Ali", 88])
    writer.writerow(["Sara", 94])


# ============================================================
# 82. ITERDIR() — LIST DIRECTORY CONTENTS
# ============================================================

folder = Path("data")

if folder.exists():
    for item in folder.iterdir():
        print(item)

# iter = iterate
#
# Concept:
#
#     data/
#       |
#       +── a.csv
#       +── b.json
#       +── report.txt
#       +── results/
#
#
# folder.iterdir()
#       ↓
#   each Path object


# ============================================================
# 83. FILTER FILES WITH is_file()
# ============================================================

folder = Path("data")

if folder.exists():
    for item in folder.iterdir():
        if item.is_file():
            print(item)


# ============================================================
# 84. GLOB — FIND MATCHING FILES
# ============================================================

folder = Path("data")

for file in folder.glob("*.csv"):
    print(file)

# "*.csv"
#
# means:
#
#     any filename ending in .csv


# ============================================================
# 85. COMMON GLOB PATTERNS
# ============================================================

# "*.csv"
#     CSV files in current directory
#
# "*.json"
#     JSON files
#
# "report*.csv"
#     files beginning with report
#
# "*2026*.csv"
#     CSV files containing 2026
#
# "data_??.csv"
#     two-character wildcard position
#
#
# Recursive search:
#
# folder.rglob("*.csv")
#
# searches subdirectories too.


# ============================================================
# 86. GLOB vs RGLOB
# ============================================================

# glob:
#
#     folder/
#       ├── a.csv
#       ├── b.csv
#       └── sub/
#             └── c.csv
#
#     glob("*.csv")
#       ↓
#     a.csv
#     b.csv
#
#
# rglob:
#
#     rglob("*.csv")
#       ↓
#     a.csv
#     b.csv
#     c.csv
#
#
# r = recursive


# ============================================================
# 87. RENAME
# ============================================================

old = Path("old_name.txt")
new = Path("new_name.txt")

if old.exists():
    old.rename(new)

# rename()
#
# changes the path/name.


# ============================================================
# 88. REPLACE
# ============================================================

# Path.replace() also renames/replaces the target.
#
# It is useful when you deliberately want replacement semantics.
#
# Use carefully because an existing destination may be replaced.


# ============================================================
# 89. UNLINK — DELETE FILE
# ============================================================

path = Path("temporary.txt")

if path.exists():
    path.unlink()

# unlink()
#
# deletes a file.
#
# IMPORTANT:
#
# It is not a "move to recycle bin" operation.
#
# Be careful.


# ============================================================
# 90. RMDIR — REMOVE EMPTY DIRECTORY
# ============================================================

folder = Path("empty_folder")

if folder.exists() and folder.is_dir():
    folder.rmdir()

# rmdir() only removes an EMPTY directory.


# ============================================================
# 91. PATH STATISTICS
# ============================================================

path = Path("students.csv")

if path.exists():
    information = path.stat()

    print(information.st_size)
    print(information.st_mtime)

# stat() provides filesystem metadata.
#
# Useful fields:
#
#     st_size
#         size in bytes
#
#     st_mtime
#         modification time
#
#     st_ctime
#         platform-dependent creation/status-change information
#
#     st_atime
#         last access time


# ============================================================
# 92. FILE SIZE
# ============================================================

path = Path("students.csv")

if path.exists():
    size_bytes = path.stat().st_size

    print(size_bytes)


# ============================================================
# 93. FILE SUFFIXES
# ============================================================

path = Path("archive.tar.gz")

print(path.suffix)
print(path.suffixes)

# suffix:
#
#     .gz
#
# suffixes:
#
#     ['.tar', '.gz']


# ============================================================
# 94. WITH_NAME()
# ============================================================

path = Path("data/old.csv")

new_path = path.with_name("new.csv")

print(new_path)

# Keeps the parent:
#
#     data/old.csv
#          ↓
#     data/new.csv


# ============================================================
# 95. WITH_SUFFIX()
# ============================================================

path = Path("data/report.txt")

new_path = path.with_suffix(".csv")

print(new_path)

#     report.txt
#          ↓
#     report.csv


# ============================================================
# 96. AS_POSIX()
# ============================================================

path = Path("data") / "reports" / "file.csv"

print(path.as_posix())

# Gives a forward-slash representation.
#
# Useful when a system/API expects POSIX-style paths.
#
# But normally:
#
#     use Path objects
#
# instead of converting paths to strings unnecessarily.


# ============================================================
# 97. PATH RELATIVE_TO
# ============================================================

base = Path("/project")
path = Path("/project/data/file.csv")

print(path.relative_to(base))

# Result:
#
#     data/file.csv
#
# relative_to() asks:
#
#     "What is this path relative to that base?"


# ============================================================
# 98. PATH READ/WRITE SUMMARY
# ============================================================

# ┌────────────────────┬──────────────────────────────┐
# │ Method             │ Purpose                      │
# ├────────────────────┼──────────────────────────────┤
# │ exists()           │ does it exist?               │
# │ is_file()          │ is it a file?               │
# │ is_dir()           │ is it a directory?          │
# │ mkdir()            │ create directory             │
# │ iterdir()          │ list directory               │
# │ glob()             │ matching files               │
# │ rglob()            │ recursive matching           │
# │ open()             │ open file                    │
# │ read_text()        │ read text                    │
# │ write_text()       │ write text                   │
# │ read_bytes()       │ read binary                  │
# │ write_bytes()      │ write binary                 │
# │ rename()           │ rename/move                  │
# │ replace()          │ replace target               │
# │ unlink()           │ delete file                  │
# │ rmdir()            │ remove empty directory       │
# │ stat()             │ filesystem metadata           │
# │ resolve()          │ resolved absolute path        │
# └────────────────────┴──────────────────────────────┘


# ============================================================
# 99. PATH ATTRIBUTES SUMMARY
# ============================================================

# For:
#
#     Path("data/reports/result.csv")
#
#
# ┌──────────────┬─────────────────────────────┐
# │ Expression   │ Meaning                     │
# ├──────────────┼─────────────────────────────┤
# │ .name        │ result.csv                  │
# │ .stem        │ result                      │
# │ .suffix      │ .csv                        │
# │ .suffixes    │ ['.csv']                    │
# │ .parent      │ data/reports                │
# │ .parts       │ path components              │
# │ .anchor      │ root/drive part              │
# └──────────────┴─────────────────────────────┘


# ============================================================
# 100. PATHLIB PRACTICAL:
#     FIND ALL CSV FILES
# ============================================================

data_folder = Path("data")

csv_files = list(data_folder.rglob("*.csv"))

for file in csv_files:
    print(file)

# This is extremely useful for:
#
#     research projects
#     dataset collections
#     batch processing
#     ETL pipelines
#     ML experiments


# ============================================================
# 101. PATHLIB PRACTICAL:
#     PROCESS MANY CSV FILES
# ============================================================

data_folder = Path("data")

for file_path in data_folder.rglob("*.csv"):
    print("Processing:", file_path)

    with file_path.open(
        "r",
        newline="",
        encoding="utf-8"
    ) as file:
        reader = csv.DictReader(file)

        for row in reader:
            pass

# Architecture:
#
#     data/
#       |
#       +── file1.csv
#       +── file2.csv
#       +── sub/
#             +── file3.csv
#
#              ↓
#
#     rglob("*.csv")
#              ↓
#       all CSV Path objects
#              ↓
#       process each file


# ============================================================
# 102. OS — THE BASIC IDEA
# ============================================================

# os provides interaction with the operating system.
#
# Think:
#
#     Python
#       |
#       v
#     operating system
#       |
#       +── environment variables
#       +── current directory
#       +── process information
#       +── filesystem operations
#       +── OS-specific behavior
#
#
# pathlib overlaps with some filesystem operations.
#
# That is normal.
#
# Modern style:
#
#     pathlib → paths/filesystem
#
#     os → environment and OS-level functionality


# ============================================================
# 103. os.getcwd()
# ============================================================

print(os.getcwd())

# cwd = current working directory
#
# Equivalent conceptual idea:
#
#     Path.cwd()


# ============================================================
# 104. os.chdir()
# ============================================================

# os.chdir("some_folder")
#
# changes the process's current working directory.
#
# Example:
#
#     os.chdir("data")
#
#
# WARNING:
#
# Changing the working directory changes how relative paths
# are interpreted by your program.
#
# In larger applications, prefer explicit Path objects when
# possible instead of constantly changing cwd.


# ============================================================
# 105. os.listdir()
# ============================================================

items = os.listdir("data")

print(items)

# Returns names in a directory.
#
# Difference:
#
#     os.listdir()
#         → names, usually strings
#
#     Path.iterdir()
#         → Path objects
#
# For modern filesystem code, Path.iterdir() is often cleaner.


# ============================================================
# 106. os.mkdir()
# ============================================================

# os.mkdir("results")
#
# creates one directory.
#
# Parent directory must already exist.
#
#
# pathlib equivalent:
#
#     Path("results").mkdir()


# ============================================================
# 107. os.makedirs()
# ============================================================

# os.makedirs("project/data/raw", exist_ok=True)
#
# creates nested directories.
#
# pathlib equivalent:
#
#     Path("project/data/raw").mkdir(
#         parents=True,
#         exist_ok=True
#     )


# ============================================================
# 108. os.remove()
# ============================================================

# os.remove("temporary.txt")
#
# deletes a file.
#
# pathlib equivalent:
#
#     Path("temporary.txt").unlink()


# ============================================================
# 109. os.rmdir()
# ============================================================

# os.rmdir("empty_folder")
#
# removes an empty directory.
#
# pathlib equivalent:
#
#     Path("empty_folder").rmdir()


# ============================================================
# 110. os.rename()
# ============================================================

# os.rename("old.txt", "new.txt")
#
# pathlib equivalent:
#
#     Path("old.txt").rename("new.txt")


# ============================================================
# 111. os.replace()
# ============================================================

# os.replace("source.txt", "destination.txt")
#
# useful when replacement behavior is intended.
#
# pathlib has:
#
#     Path.replace()


# ============================================================
# 112. os.path — IMPORTANT
# ============================================================

# os.path contains many path-related functions.
#
# Common ones:
#
#     os.path.join()
#     os.path.exists()
#     os.path.isfile()
#     os.path.isdir()
#     os.path.abspath()
#     os.path.basename()
#     os.path.dirname()
#     os.path.splitext()
#     os.path.getsize()
#     os.path.getmtime()


# ============================================================
# 113. os.path.join()
# ============================================================

path = os.path.join(
    "data",
    "reports",
    "result.csv"
)

print(path)

# Older/common style:
#
#     os.path.join(...)
#
#
# Modern pathlib style:
#
#     Path("data") / "reports" / "result.csv"
#
#
# For new Python code, pathlib is generally easier to read.


# ============================================================
# 114. os.path.exists()
# ============================================================

print(os.path.exists("students.csv"))

# Equivalent:
#
#     Path("students.csv").exists()


# ============================================================
# 115. os.path.isfile()
# ============================================================

print(os.path.isfile("students.csv"))

# Equivalent:
#
#     Path("students.csv").is_file()


# ============================================================
# 116. os.path.isdir()
# ============================================================

print(os.path.isdir("data"))

# Equivalent:
#
#     Path("data").is_dir()


# ============================================================
# 117. os.path.basename()
# ============================================================

path = "/project/data/report.csv"

print(os.path.basename(path))

# Result:
#
#     report.csv
#
# pathlib equivalent:
#
#     Path(path).name


# ============================================================
# 118. os.path.dirname()
# ============================================================

path = "/project/data/report.csv"

print(os.path.dirname(path))

# Result:
#
#     /project/data
#
# pathlib equivalent:
#
#     Path(path).parent


# ============================================================
# 119. os.path.splitext()
# ============================================================

path = "report.csv"

name, extension = os.path.splitext(path)

print(name)
print(extension)

# Result:
#
#     report
#     .csv
#
#
# pathlib equivalents:
#
#     Path(path).stem
#     Path(path).suffix


# ============================================================
# 120. os.path.getsize()
# ============================================================

size = os.path.getsize("students.csv")

print(size)

# Size is returned in bytes.


# ============================================================
# 121. OS ENVIRONMENT VARIABLES
# ============================================================

# One of the most important practical uses of os.
#
# Environment variables store configuration outside your code.
#
#
# Example:
#
#     API_KEY
#     DATABASE_URL
#     USERNAME
#     PATH
#
#
# Access:
#
api_path = os.environ.get("PATH")

print(api_path)

# get() returns None if the variable does not exist.


# ============================================================
# 122. os.environ["VARIABLE"]
# ============================================================

# value = os.environ["PATH"]
#
# This raises KeyError if the variable does not exist.
#
#
# Difference:
#
#     os.environ.get("NAME")
#         → None if missing
#
#     os.environ["NAME"]
#         → KeyError if missing


# ============================================================
# 123. PRACTICAL ENVIRONMENT VARIABLE
# ============================================================

# Suppose the operating system contains:
#
#     DATA_FOLDER=data
#
# Then:

data_folder = os.environ.get("DATA_FOLDER", "data")

print(data_folder)

# Pattern:
#
#     environment variable
#             ↓
#     configuration
#             ↓
#     Python program
#
#
# This is better than hardcoding sensitive/configurable values
# into source code.


# ============================================================
# 124. NEVER PUT SECRETS DIRECTLY IN CODE
# ============================================================

# BAD:
#
#     API_KEY = "my-real-secret-key"
#
#
# Better:
#
#     API_KEY = os.environ.get("API_KEY")
#
#
# For real projects:
#
#     environment variables
#     secret managers
#     deployment configuration
#
# are preferred over committing secrets to Git.


# ============================================================
# 125. os.name
# ============================================================

print(os.name)

# Common values:
#
#     "nt"      → Windows
#     "posix"   → Unix-like systems
#
# It is useful for broad OS distinction.
#
# For more detailed information:
#
#     platform
#
# module can be useful.


# ============================================================
# 126. os.cpu_count()
# ============================================================

print(os.cpu_count())

# Returns the number of logical CPUs available/recognized.
#
# Useful when thinking about:
#
#     parallel processing
#     multiprocessing
#     workload sizing
#
# But:
#
#     logical CPU count ≠ guaranteed useful parallel capacity.


# ============================================================
# 127. os.getpid()
# ============================================================

print(os.getpid())

# PID = Process ID
#
# Useful for:
#
#     debugging
#     logging
#     process management


# ============================================================
# 128. os.walk()
# ============================================================

# os.walk() recursively walks through a directory tree.
#
# Conceptually:
#
#     project/
#       |
#       +── data/
#       |     +── a.csv
#       |     +── b.csv
#       |
#       +── reports/
#             +── report.csv
#
#
# os.walk("project")
#       ↓
#     root
#     directories
#     files
#
#
# Example:

for root, directories, files in os.walk("project"):
    print("ROOT:", root)
    print("DIRECTORIES:", directories)
    print("FILES:", files)

# os.walk() is still useful.
#
# But pathlib.rglob() is often simpler for "find files matching
# this pattern" tasks.


# ============================================================
# 129. PATHLIB vs OS
# ============================================================

# ┌──────────────────────────────┬───────────────────────────┐
# │ Task                         │ Preferred modern style    │
# ├──────────────────────────────┼───────────────────────────┤
# │ Build paths                 │ pathlib.Path               │
# │ Check file                  │ Path.is_file()              │
# │ Check directory             │ Path.is_dir()              │
# │ Create folder               │ Path.mkdir()               │
# │ Read text                   │ Path.read_text()            │
# │ Write text                  │ Path.write_text()           │
# │ Find files                  │ Path.glob()/rglob()         │
# │ Rename                      │ Path.rename()               │
# │ Delete file                 │ Path.unlink()               │
# │ Environment variables       │ os.environ                  │
# │ Current working directory   │ os.getcwd() / Path.cwd()     │
# │ OS name                     │ os.name                      │
# │ Process ID                  │ os.getpid()                  │
# │ Recursive directory walking │ os.walk()                   │
# └──────────────────────────────┴───────────────────────────┘


# ============================================================
# 130. THE MOST IMPORTANT FILE-WORKING ARCHITECTURE
# ============================================================

# A very useful mental model:
#
#
#                 FILE WORKFLOW
#
#                     Path
#                      |
#                      v
#               Does it exist?
#                      |
#             +--------+--------+
#             |                 |
#            NO                YES
#             |                 |
#             v                 v
#        create it?       file or folder?
#                               |
#                       +-------+-------+
#                       |               |
#                      file           folder
#                       |               |
#                       v               v
#                    open()         iterdir()
#                       |
#                       v
#                  read/write
#                       |
#                       v
#                    process
#                       |
#                       v
#                    output


# ============================================================
# 131. PRACTICAL PROJECT DIRECTORY
# ============================================================

# For a data-analysis project:
#
# project/
# │
# ├── data/
# │   ├── raw/
# │   ├── processed/
# │   └── external/
# │
# ├── notebooks/
# │
# ├── src/
# │
# ├── reports/
# │
# ├── outputs/
# │
# └── config/
#
#
# pathlib can create this structure:

project = Path("project")

directories = [
    project / "data" / "raw",
    project / "data" / "processed",
    project / "data" / "external",
    project / "notebooks",
    project / "src",
    project / "reports",
    project / "outputs",
    project / "config"
]

for directory in directories:
    directory.mkdir(parents=True, exist_ok=True)


# ============================================================
# 132. PRACTICAL:
#     FIND LATEST MODIFIED FILE
# ============================================================

folder = Path("data")

files = [
    path
    for path in folder.rglob("*")
    if path.is_file()
]

if files:
    latest = max(files, key=lambda path: path.stat().st_mtime)
    print("Latest file:", latest)

# Mental model:
#
# files
#   ↓
# stat().st_mtime
#   ↓
# max()
#   ↓
# latest file


# ============================================================
# 133. PRACTICAL:
#     GET ALL CSV FILES AND THEIR SIZES
# ============================================================

folder = Path("data")

for path in folder.rglob("*.csv"):
    size = path.stat().st_size

    print(
        path.name,
        size,
        "bytes"
    )


# ============================================================
# 134. PRACTICAL:
#     CONVERT BYTES TO MB
# ============================================================

size_bytes = 5_000_000

size_mb = size_bytes / (1024 ** 2)

print(size_mb)

# Common binary-style conversion:
#
#     1 KiB = 1024 bytes
#     1 MiB = 1024² bytes
#
# People often casually call this "MB", but technically
# MiB is the binary unit.


# ============================================================
# 135. PRACTICAL:
#     BACKUP A FILE
# ============================================================

source = Path("data/students.csv")
backup = Path("data/students_backup.csv")

if source.exists():
    backup.write_bytes(source.read_bytes())

# For large files, shutil.copy2() is usually a better choice
# because it is designed for file copying and can preserve
# metadata.


# ============================================================
# 136. SHUTIL — IMPORTANT CONNECTION
# ============================================================

# The user asked for os/pathlib, but one practical filesystem
# library belongs here conceptually:
#
#     shutil
#
# It handles higher-level file operations.
#
# Important functions:
#
#     shutil.copy()
#     shutil.copy2()
#     shutil.copytree()
#     shutil.move()
#     shutil.rmtree()
#
#
# Example:
#
#     import shutil
#     shutil.copy2(source, backup)
#
#
# Think:
#
#     pathlib → paths
#     os      → OS
#     shutil  → higher-level file operations


# ============================================================
# 137. PRACTICAL:
#     COPY AN ENTIRE DIRECTORY
# ============================================================

# import shutil
#
# shutil.copytree(
#     "data",
#     "data_backup",
#     dirs_exist_ok=True
# )
#
# dirs_exist_ok=True allows an existing destination directory.


# ============================================================
# 138. PRACTICAL:
#     MOVE FILES BY EXTENSION
# ============================================================

# import shutil
#
# source_folder = Path("downloads")
# target_folder = Path("documents")
#
# target_folder.mkdir(exist_ok=True)
#
# for file in source_folder.glob("*.csv"):
#     shutil.move(str(file), target_folder / file.name)
#
#
# Architecture:
#
#     downloads/
#        |
#        +── a.csv
#        +── b.txt
#        +── c.csv
#
#             ↓
#
#     "*.csv"
#             ↓
#
#     documents/
#        +── a.csv
#        +── c.csv


# ============================================================
# 139. JSON + PATHLIB + DATETIME
#     REAL-WORLD COMBINATION
# ============================================================

from datetime import datetime

output_folder = Path("outputs")
output_folder.mkdir(exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

output_file = output_folder / f"result_{timestamp}.json"

result = {
    "created_at": datetime.now().isoformat(),
    "records": 100,
    "accuracy": 0.91
}

with output_file.open("w", encoding="utf-8") as file:
    json.dump(result, file, indent=4)

print(output_file)

# This is a very useful research/ML pattern:
#
#     current time
#          ↓
#     unique filename
#          ↓
#     Path
#          ↓
#     JSON output
#
# Example:
#
#     result_20260909_153045.json


# ============================================================
# 140. CSV + PATHLIB + DATETIME
#     REAL-WORLD COMBINATION
# ============================================================

output_folder = Path("outputs")
output_folder.mkdir(exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

output_file = output_folder / f"results_{timestamp}.csv"

with output_file.open(
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=["name", "score"]
    )

    writer.writeheader()

    writer.writerow({
        "name": "Ali",
        "score": 88
    })

    writer.writerow({
        "name": "Sara",
        "score": 94
    })


# ============================================================
# 141. PRACTICAL:
#     LOGGING AN EXPERIMENT
# ============================================================

experiment = {
    "experiment_id": "exp_001",
    "created_at": datetime.now().isoformat(),
    "model": "XGBoost",
    "recall": 0.795,
    "f1": 0.730
}

experiment_folder = Path("experiments")
experiment_folder.mkdir(exist_ok=True)

file_path = experiment_folder / "exp_001.json"

with file_path.open("w", encoding="utf-8") as file:
    json.dump(experiment, file, indent=4)

# This pattern is highly relevant to ML/research:
#
#     experiment
#         |
#         +── parameters
#         +── metrics
#         +── timestamp
#         |
#         v
#     JSON
#         |
#         v
#     reproducible record


# ============================================================
# 142. PRACTICAL:
#     READ ALL JSON FILES
# ============================================================

folder = Path("experiments")

for file_path in folder.glob("*.json"):

    with file_path.open(
        "r",
        encoding="utf-8"
    ) as file:

        experiment = json.load(file)

    print(
        file_path.name,
        experiment
    )


# ============================================================
# 143. PRACTICAL:
#     MERGE DATA FROM MANY JSON FILES
# ============================================================

folder = Path("experiments")

all_experiments = []

for file_path in folder.glob("*.json"):

    with file_path.open(
        "r",
        encoding="utf-8"
    ) as file:

        data = json.load(file)

    all_experiments.append(data)

print(all_experiments)

# Architecture:
#
#     exp_001.json ─┐
#     exp_002.json ─┼──→ Python list
#     exp_003.json ─┘
#
# This is a simple foundation for building experiment
# tracking systems.


# ============================================================
# 144. PRACTICAL:
#     VALIDATE A FILE BEFORE READING
# ============================================================

file_path = Path("students.json")

if not file_path.exists():
    print("File does not exist.")

elif not file_path.is_file():
    print("Path is not a file.")

else:
    with file_path.open("r", encoding="utf-8") as file:
        data = json.load(file)

    print("Loaded successfully.")


# ============================================================
# 145. PRACTICAL:
#     SAFE JSON LOADER
# ============================================================

def load_json(path):
    path = Path(path)

    if not path.exists():
        raise FileNotFoundError(
            f"File not found: {path}"
        )

    if not path.is_file():
        raise IsADirectoryError(
            f"Expected a file: {path}"
        )

    try:
        with path.open(
            "r",
            encoding="utf-8"
        ) as file:
            return json.load(file)

    except json.JSONDecodeError as error:
        raise ValueError(
            f"Invalid JSON: {path}"
        ) from error


# Example:
#
#     data = load_json("students.json")


# ============================================================
# 146. PRACTICAL:
#     SAFE JSON SAVER
# ============================================================

def save_json(data, path):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with path.open(
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            data,
            file,
            indent=4,
            ensure_ascii=False
        )


# Example:
#
#     save_json(
#         {"name": "Ali"},
#         "outputs/person.json"
#     )


# ============================================================
# 147. PRACTICAL:
#     SAFE CSV READER
# ============================================================

def read_csv(path):
    path = Path(path)

    with path.open(
        "r",
        newline="",
        encoding="utf-8"
    ) as file:

        return list(csv.DictReader(file))


# Example:
#
#     records = read_csv("students.csv")


# ============================================================
# 148. PRACTICAL:
#     CREATE A CLEAN DATASET PIPELINE
# ============================================================

raw_folder = Path("data/raw")
processed_folder = Path("data/processed")

processed_folder.mkdir(
    parents=True,
    exist_ok=True
)

for input_file in raw_folder.glob("*.csv"):

    with input_file.open(
        "r",
        newline="",
        encoding="utf-8"
    ) as file:

        reader = csv.DictReader(file)

        cleaned = []

        for row in reader:

            name = row["name"].strip()

            score_text = row["score"].strip()

            if not name:
                continue

            if score_text == "":
                continue

            try:
                score = float(score_text)
            except ValueError:
                continue

            cleaned.append({
                "name": name,
                "score": score
            })

    output_file = processed_folder / input_file.name

    with output_file.open(
        "w",
        newline="",
        encoding="utf-8"
    ) as file:

        writer = csv.DictWriter(
            file,
            fieldnames=["name", "score"]
        )

        writer.writeheader()
        writer.writerows(cleaned)

# Complete architecture:
#
#
#     data/raw
#         |
#         | *.csv
#         v
#     read
#         |
#         v
#     clean
#         |
#         +── strip
#         +── missing-value check
#         +── type conversion
#         +── validation
#         |
#         v
#     data/processed
#
#
# This is the beginning of a real ETL pipeline.


# ============================================================
# 149. PRACTICAL:
#     DATE PARSING INSIDE CSV PROCESSING
# ============================================================

records = [
    {
        "name": "Ali",
        "date": "2026-09-01",
        "score": "88"
    },
    {
        "name": "Sara",
        "date": "2026-09-05",
        "score": "94"
    }
]

for record in records:

    record["date"] = date.fromisoformat(
        record["date"]
    )

    record["score"] = float(
        record["score"]
    )

print(records)

# Before:
#
#     "date"  → string
#     "score" → string
#
# After:
#
#     "date"  → date
#     "score" → float


# ============================================================
# 150. PRACTICAL:
#     FILTER RECORDS BY DATE
# ============================================================

cutoff = date(2026, 9, 3)

filtered = [
    record
    for record in records
    if record["date"] >= cutoff
]

print(filtered)


# ============================================================
# 151. PRACTICAL:
#     FIND FILES CREATED/MODIFIED RECENTLY
# ============================================================

folder = Path("data")

now = datetime.now()

for file_path in folder.rglob("*"):

    if not file_path.is_file():
        continue

    modified_timestamp = file_path.stat().st_mtime

    modified_datetime = datetime.fromtimestamp(
        modified_timestamp
    )

    age = now - modified_datetime

    if age <= timedelta(days=7):
        print(
            "Recently modified:",
            file_path
        )

# Architecture:
#
# filesystem timestamp
#        ↓
# datetime.fromtimestamp()
#        ↓
# datetime object
#        ↓
# now - modified
#        ↓
# timedelta
#        ↓
# filter


# ============================================================
# 152. IMPORTANT:
#     FILE TIMESTAMPS AND TIMEZONES
# ============================================================

# Filesystem timestamps are interpreted through the operating
# system/platform conventions.
#
# If timezone correctness matters:
#
#     use aware datetimes
#     define your timezone policy
#     be explicit about UTC/local time
#
# Do not blindly mix naive and aware datetime objects.


# ============================================================
# 153. PRACTICAL:
#     CREATE DATE-BASED FOLDERS
# ============================================================

today = date.today()

folder = Path(
    "reports"
) / str(today.year) / f"{today.month:02d}"

folder.mkdir(
    parents=True,
    exist_ok=True
)

print(folder)

# Example:
#
#     reports/
#       └── 2026/
#            └── 09/
#
#
# Useful for:
#
#     logs
#     reports
#     backups
#     experiment outputs
#     daily datasets


# ============================================================
# 154. PRACTICAL:
#     TIMESTAMPED FILE NAMES
# ============================================================

timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

file_path = Path("outputs") / f"report_{timestamp}.csv"

print(file_path)

# Example:
#
#     report_20260909_153045.csv
#
# Avoid characters such as ":" in filenames when you want
# maximum cross-platform compatibility.


# ============================================================
# 155. PRACTICAL:
#     CHECK FILE EXTENSION
# ============================================================

path = Path("dataset.csv")

if path.suffix.lower() == ".csv":
    print("CSV file")

elif path.suffix.lower() == ".json":
    print("JSON file")

else:
    print("Other file")


# ============================================================
# 156. PRACTICAL:
#     SELECT PROCESSOR BASED ON EXTENSION
# ============================================================

def load_data(path):
    path = Path(path)

    suffix = path.suffix.lower()

    if suffix == ".json":
        with path.open("r", encoding="utf-8") as file:
            return json.load(file)

    if suffix == ".csv":
        with path.open(
            "r",
            newline="",
            encoding="utf-8"
        ) as file:
            return list(csv.DictReader(file))

    raise ValueError(
        f"Unsupported file type: {suffix}"
    )


# Architecture:
#
#                file
#                  |
#                  v
#                suffix
#                  |
#           +------+------+
#           |             |
#         .json         .csv
#           |             |
#           v             v
#         json          csv
#         loader        loader


# ============================================================
# 157. PRACTICAL:
#     DATA EXPORTER
# ============================================================

def save_data(data, path):
    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    if path.suffix.lower() == ".json":

        with path.open(
            "w",
            encoding="utf-8"
        ) as file:

            json.dump(
                data,
                file,
                indent=4,
                ensure_ascii=False
            )

    elif path.suffix.lower() == ".csv":

        if not data:
            return

        fieldnames = data[0].keys()

        with path.open(
            "w",
            newline="",
            encoding="utf-8"
        ) as file:

            writer = csv.DictWriter(
                file,
                fieldnames=fieldnames
            )

            writer.writeheader()
            writer.writerows(data)

    else:
        raise ValueError(
            f"Unsupported extension: {path.suffix}"
        )


# ============================================================
# 158. PRACTICAL:
#     FILE ORGANIZATION AUTOMATION
# ============================================================

# Suppose:
#
# downloads/
#   report1.csv
#   report2.csv
#   image.png
#   notes.txt
#
# We want:
#
# organized/
#   csv/
#   images/
#   text/
#
#
# Example:

downloads = Path("downloads")

folders = {
    ".csv": downloads / "csv",
    ".png": downloads / "images",
    ".jpg": downloads / "images",
    ".jpeg": downloads / "images",
    ".txt": downloads / "text"
}

for folder in folders.values():
    folder.mkdir(
        parents=True,
        exist_ok=True
    )

# To actually move the files, shutil.move() is appropriate.


# ============================================================
# 159. JSON + CSV + DATETIME + PATHLIB + OS
#     COMPLETE PIPELINE
# ============================================================

# Imagine a real data project:
#
#
#                  operating system
#                         |
#                         v
#                     os.environ
#                         |
#                         v
#                    configuration
#                         |
#                         v
#                       Path
#                         |
#                         v
#                    input files
#                         |
#              +----------+----------+
#              |                     |
#             CSV                   JSON
#              |                     |
#              v                     v
#          csv.reader            json.load
#              |                     |
#              +----------+----------+
#                         |
#                         v
#                  Python objects
#                         |
#                         v
#                     cleaning
#                         |
#                         v
#                    datetime
#                         |
#                         v
#                    validation
#                         |
#                         v
#                     analysis
#                         |
#                         v
#                  timestamp output
#                         |
#                         v
#                       Path
#                         |
#              +----------+----------+
#              |                     |
#             CSV                   JSON
#              |                     |
#              v                     v
#          output file          output file


# ============================================================
# 160. REAL MINI PROJECT:
#     STUDENT DATA PROCESSOR
# ============================================================

# Raw CSV:
#
# name,department,date,score
#  Ali ,IT,2026-09-01,88
# Sara,IT,2026-09-02,94
# Reza,Sales,2026-09-03,76
#
#
# Goal:
#
#     read
#       ↓
#     clean
#       ↓
#     convert types
#       ↓
#     calculate summary
#       ↓
#     save JSON


def process_students(input_path, output_path):

    input_path = Path(input_path)
    output_path = Path(output_path)

    records = []

    with input_path.open(
        "r",
        newline="",
        encoding="utf-8"
    ) as file:

        reader = csv.DictReader(file)

        for row in reader:

            name = row["name"].strip()
            department = row["department"].strip()

            if not name:
                continue

            try:
                score = float(row["score"])
                record_date = date.fromisoformat(
                    row["date"].strip()
                )
            except (ValueError, TypeError):
                continue

            records.append({
                "name": name,
                "department": department,
                "date": record_date.isoformat(),
                "score": score
            })

    if records:
        average_score = (
            sum(record["score"] for record in records)
            / len(records)
        )
    else:
        average_score = None

    result = {
        "processed_at": datetime.now().isoformat(),
        "record_count": len(records),
        "average_score": average_score,
        "records": records
    }

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with output_path.open(
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            result,
            file,
            indent=4,
            ensure_ascii=False
        )

    return result


# This combines:
#
#     pathlib
#     csv
#     datetime
#     json
#
# into one realistic workflow.


# ============================================================
# 161. JSON FUNCTIONS — CHEAT SHEET
# ============================================================

# import json
#
#
# json.dumps(obj)
#     Python → JSON string
#
# json.loads(text)
#     JSON string → Python
#
# json.dump(obj, file)
#     Python → JSON file
#
# json.load(file)
#     JSON file → Python
#
#
# Common useful parameters:
#
#     indent=4
#     ensure_ascii=False
#     sort_keys=True
#     separators=...
#     default=...
#
#
# Important exception:
#
#     json.JSONDecodeError


# ============================================================
# 162. CSV FUNCTIONS — CHEAT SHEET
# ============================================================

# import csv
#
#
# csv.reader(file)
#     read rows as lists
#
# csv.writer(file)
#     write rows as lists
#
# csv.DictReader(file)
#     read rows as dictionaries
#
# csv.DictWriter(file, fieldnames=...)
#     write dictionaries
#
#
# writer.writerow(row)
# writer.writerows(rows)
#
# DictWriter:
#
#     writeheader()
#     writerow()
#     writerows()
#
#
# Important options:
#
#     delimiter=","
#     quotechar='"'
#     newline=""
#     encoding="utf-8"


# ============================================================
# 163. DATETIME FUNCTIONS / CLASSES — CHEAT SHEET
# ============================================================

# from datetime import (
#     date,
#     time,
#     datetime,
#     timedelta,
#     timezone
# )
#
#
# date:
#
#     date.today()
#     date(...)
#     date.fromisoformat(...)
#
#
# datetime:
#
#     datetime.now()
#     datetime.today()
#     datetime(...)
#     datetime.fromisoformat(...)
#     datetime.strptime(...)
#     datetime.combine(...)
#
#
# datetime instance methods:
#
#     .date()
#     .time()
#     .strftime()
#     .isoformat()
#     .replace()
#     .weekday()
#     .isoweekday()
#     .isocalendar()
#
#
# timedelta:
#
#     timedelta(...)
#     .days
#     .seconds
#     .microseconds
#     .total_seconds()
#
#
# timezone:
#
#     timezone.utc
#     timezone(offset)


# ============================================================
# 164. PATHLIB — CHEAT SHEET
# ============================================================

# from pathlib import Path
#
#
# Construction:
#
#     Path("data/file.csv")
#
# Location:
#
#     Path.cwd()
#     Path.home()
#     path.absolute()
#     path.resolve()
#
# Components:
#
#     path.name
#     path.stem
#     path.suffix
#     path.suffixes
#     path.parent
#     path.parts
#
# Filesystem:
#
#     path.exists()
#     path.is_file()
#     path.is_dir()
#
# Creation:
#
#     path.mkdir()
#
# Reading/writing:
#
#     path.open()
#     path.read_text()
#     path.write_text()
#     path.read_bytes()
#     path.write_bytes()
#
# Searching:
#
#     path.iterdir()
#     path.glob()
#     path.rglob()
#
# Modification:
#
#     path.rename()
#     path.replace()
#     path.unlink()
#     path.rmdir()
#
# Metadata:
#
#     path.stat()
#
# Path transformation:
#
#     path.with_name()
#     path.with_suffix()
#     path.relative_to()
#     path.as_posix()


# ============================================================
# 165. OS — CHEAT SHEET
# ============================================================

# import os
#
#
# Working directory:
#
#     os.getcwd()
#     os.chdir()
#
# Directory:
#
#     os.listdir()
#     os.mkdir()
#     os.makedirs()
#     os.rmdir()
#
# Files:
#
#     os.remove()
#     os.rename()
#     os.replace()
#
# Environment:
#
#     os.environ
#     os.environ.get()
#
# System:
#
#     os.name
#     os.cpu_count()
#     os.getpid()
#
# Recursive traversal:
#
#     os.walk()
#
# Path:
#
#     os.path.join()
#     os.path.exists()
#     os.path.isfile()
#     os.path.isdir()
#     os.path.abspath()
#     os.path.basename()
#     os.path.dirname()
#     os.path.splitext()
#     os.path.getsize()
#     os.path.getmtime()


# ============================================================
# 166. "WHICH FUNCTION DO I NEED?"
# ============================================================

# ┌──────────────────────────────┬─────────────────────────────┐
# │ I want to...                 │ Use                          │
# ├──────────────────────────────┼─────────────────────────────┤
# │ Python → JSON string         │ json.dumps()                 │
# │ JSON string → Python         │ json.loads()                 │
# │ Python → JSON file           │ json.dump()                  │
# │ JSON file → Python           │ json.load()                  │
# │ Read CSV rows                │ csv.reader()                 │
# │ Read CSV dictionaries        │ csv.DictReader()             │
# │ Write CSV rows               │ csv.writer()                 │
# │ Write CSV dictionaries       │ csv.DictWriter()             │
# │ Current date                 │ date.today()                 │
# │ Current date + time          │ datetime.now()               │
# │ String → datetime            │ datetime.strptime()          │
# │ Datetime → string            │ datetime.strftime()          │
# │ ISO string → datetime        │ datetime.fromisoformat()     │
# │ Datetime → ISO string        │ datetime.isoformat()         │
# │ Add/subtract time            │ timedelta                     │
# │ Build a path                │ Path()                       │
# │ Current folder              │ Path.cwd()                   │
# │ User home                   │ Path.home()                  │
# │ Check file                  │ Path.is_file()               │
# │ Check folder                │ Path.is_dir()                │
# │ Check existence             │ Path.exists()                │
# │ Make folder                 │ Path.mkdir()                 │
# │ Read text                   │ Path.read_text()              │
# │ Write text                  │ Path.write_text()             │
# │ Find files                  │ Path.glob()/rglob()           │
# │ Delete file                 │ Path.unlink()                 │
# │ Environment variable        │ os.environ.get()              │
# │ OS information              │ os.name / os.cpu_count()      │
# │ Recursive OS walk           │ os.walk()                     │
# └──────────────────────────────┴─────────────────────────────┘


# ============================================================
# 167. IMPORTANT MENTAL MODEL:
#     STRING vs OBJECT vs FILE
# ============================================================

# This distinction is fundamental.
#
#
#             STRING
#               |
#               | loads()
#               v
#          PYTHON OBJECT
#               |
#               | dumps()
#               v
#             STRING
#
#
# FILE:
#
#             FILE
#               |
#               | load()
#               v
#          PYTHON OBJECT
#               |
#               | dump()
#               v
#             FILE
#
#
# pathlib answers:
#
#     "Which FILE?"
#
# json/csv answer:
#
#     "How is the DATA inside that file represented?"


# ============================================================
# 168. THE MOST IMPORTANT DISTINCTION:
#     PATH ≠ FILE CONTENT
# ============================================================

# This:
#
#     Path("data/students.csv")
#
# represents a LOCATION.
#
#
# This:
#
#     "Ali,25,88"
#
# represents CONTENT.
#
#
# This:
#
#     {"name": "Ali", "age": 25}
#
# represents a PYTHON OBJECT.
#
#
# Think:
#
#     Path
#       ↓
#     location
#
#     open/read
#       ↓
#     content
#
#     parser
#       ↓
#     Python objects


# ============================================================
# 169. COMMON MISTAKE:
#     TREATING JSON AS A DICTIONARY
# ============================================================

# JSON text:
#
json_text = '{"name": "Ali", "age": 25}'

# WRONG conceptual assumption:
#
#     json_text["name"]
#
# because json_text is a string.
#
#
# Correct:
#
data = json.loads(json_text)

print(data["name"])


# ============================================================
# 170. COMMON MISTAKE:
#     USING json.loads() ON A FILE OBJECT
# ============================================================

# If you have a file:
#
#     file = open(...)
#
# use:
#
#     json.load(file)
#
# not:
#
#     json.loads(file)
#
#
# Remember:
#
#     load  → file
#     loads → string


# ============================================================
# 171. COMMON MISTAKE:
#     USING json.load() ON A STRING
# ============================================================

# If you have:
#
#     text = '{"name": "Ali"}'
#
# use:
#
#     json.loads(text)
#
# not:
#
#     json.load(text)


# ============================================================
# 172. COMMON MISTAKE:
#     CSV NUMBERS ARE STRINGS
# ============================================================

# CSV:
#
#     score
#     88
#
# csv.DictReader gives:
#
#     row["score"] == "88"
#
# NOT:
#
#     88
#
# Convert:
#
#     score = float(row["score"])


# ============================================================
# 173. COMMON MISTAKE:
#     SPLITTING CSV WITH split(",")
# ============================================================

# Avoid:
#
#     line.split(",")
#
# for general CSV parsing.
#
# Why?
#
# CSV supports quoting:
#
#     "Ali, Yasin"
#
# The comma may belong to the field itself.
#
# Use:
#
#     csv.reader()
#     csv.DictReader()


# ============================================================
# 174. COMMON MISTAKE:
#     datetime IS NOT A STRING
# ============================================================

dt = datetime.now()

# This:
#
#     dt
#
# is a datetime object.
#
#
# This:
#
#     str(dt)
#
# is a string.
#
#
# Conversion:
#
#     datetime → string
#         ↓
#     strftime() / isoformat()
#
#     string → datetime
#         ↓
#     strptime() / fromisoformat()


# ============================================================
# 175. COMMON MISTAKE:
#     MUTATING datetime
# ============================================================

dt = datetime(2026, 9, 9)

new_dt = dt + timedelta(days=1)

# dt did not change.
#
# datetime/date/time objects are immutable.
#
# Think:
#
#     old object
#          |
#          +── operation
#          |
#          v
#      new object


# ============================================================
# 176. COMMON MISTAKE:
#     MIXING NAIVE AND AWARE DATETIMES
# ============================================================

# Avoid operations such as:
#
#     naive_datetime < aware_datetime
#
# without deliberately normalizing the timezone semantics.
#
# Better:
#
#     choose a timezone policy
#     ↓
#     make timestamps consistently aware
#     ↓
#     compare


# ============================================================
# 177. COMMON MISTAKE:
#     STRING PATH MANIPULATION
# ============================================================

# Fragile:
#
#     path = "data" + "/" + "file.csv"
#
#
# Better:
#
#     path = Path("data") / "file.csv"
#
#
# Why?
#
# pathlib understands the platform's path rules.


# ============================================================
# 178. COMMON MISTAKE:
#     ASSUMING A PATH EXISTS
# ============================================================

path = Path("data/file.csv")

if path.exists():
    print(path.read_text(encoding="utf-8"))

# For important workflows:
#
#     verify
#     validate
#     then process


# ============================================================
# 179. COMMON MISTAKE:
#     DELETING WITHOUT THINKING
# ============================================================

# Dangerous:
#
#     path.unlink()
#
# blindly.
#
#
# Safer:
#
#     if path.exists() and path.is_file():
#         path.unlink()
#
#
# Still:
#
#     be deliberate before deletion.


# ============================================================
# 180. COMMON MISTAKE:
#     RELATIVE PATH CONFUSION
# ============================================================

# This:
#
#     Path("data/file.csv")
#
# depends on the current working directory.
#
#
# Therefore:
#
#     Path.cwd()
#
# is useful for debugging.
#
#
# Diagram:
#
#     CWD
#      |
#      +── data/
#            └── file.csv
#
# Change CWD:
#
#     relative paths may point somewhere else.


# ============================================================
# 181. WINDOWS PATHS
# ============================================================

# On Windows, you may see:
#
#     C:\Users\Yasin\data\file.csv
#
#
# pathlib:
#
#     Path(r"C:\Users\Yasin\data\file.csv")
#
#
# Raw string r"..." can prevent backslash escape issues.
#
# Even better for constructing paths:
#
#     Path("C:/Users/Yasin/data/file.csv")
#
# or:
#
#     Path("C:\\Users\\Yasin\\data\\file.csv")
#
#
# And for dynamic construction:
#
#     base / "data" / "file.csv"


# ============================================================
# 182. pathlib PATH JOINING
# ============================================================

base = Path("C:/Users/Yasin")

data = base / "data"
raw = data / "raw"
file = raw / "dataset.csv"

print(file)

# Read this as:
#
#     base
#       /
#     data
#       /
#     raw
#       /
#     dataset.csv


# ============================================================
# 183. PRACTICAL:
#     CHECK REQUIRED PROJECT FILES
# ============================================================

required_files = [
    Path("data/raw/train.csv"),
    Path("data/raw/test.csv"),
    Path("config/config.json")
]

for path in required_files:

    if not path.exists():
        print("Missing:", path)

    elif not path.is_file():
        print("Not a file:", path)

    else:
        print("OK:", path)


# ============================================================
# 184. PRACTICAL:
#     DATASET INVENTORY
# ============================================================

data_folder = Path("data")

inventory = []

for file_path in data_folder.rglob("*"):

    if not file_path.is_file():
        continue

    stat = file_path.stat()

    inventory.append({
        "name": file_path.name,
        "path": str(file_path),
        "suffix": file_path.suffix,
        "size_bytes": stat.st_size
    })

for item in inventory:
    print(item)

# This is the beginning of a useful dataset inventory tool.


# ============================================================
# 185. EXPORT THE INVENTORY TO JSON
# ============================================================

inventory_path = Path("outputs/file_inventory.json")

inventory_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

with inventory_path.open(
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        inventory,
        file,
        indent=4
    )


# ============================================================
# 186. EXPORT THE INVENTORY TO CSV
# ============================================================

if inventory:

    inventory_csv = Path(
        "outputs/file_inventory.csv"
    )

    with inventory_csv.open(
        "w",
        newline="",
        encoding="utf-8"
    ) as file:

        writer = csv.DictWriter(
            file,
            fieldnames=inventory[0].keys()
        )

        writer.writeheader()
        writer.writerows(inventory)


# ============================================================
# 187. PRACTICAL:
#     FIND DUPLICATE FILENAMES
# ============================================================

folder = Path("data")

seen = {}

for path in folder.rglob("*"):

    if not path.is_file():
        continue

    seen.setdefault(path.name, []).append(path)

for name, paths in seen.items():

    if len(paths) > 1:

        print("Duplicate filename:", name)

        for path in paths:
            print("   ", path)

# Important:
#
# Same filename does NOT necessarily mean same file contents.
#
# For actual duplicate-content detection, compare hashes
# rather than names.


# ============================================================
# 188. FILE HASHING — PRACTICAL CONNECTION
# ============================================================

# For large/research datasets, hashlib can identify identical
# file contents.
#
# Example concept:
#
#     import hashlib
#
#     digest = hashlib.sha256(
#         path.read_bytes()
#     ).hexdigest()
#
# For very large files, hash incrementally instead of loading
# the entire file into memory.
#
# This is useful for:
#
#     duplicate detection
#     dataset versioning
#     integrity checks


# ============================================================
# 189. DATA ANALYSIS APPLICATION MAP
# ============================================================

# These libraries fit naturally into a data workflow:
#
#
#        RAW FILES
#            |
#            v
#         pathlib
#            |
#            v
#      discover files
#            |
#       +----+----+
#       |         |
#      CSV       JSON
#       |         |
#       +----+----+
#            |
#            v
#        Python data
#            |
#            v
#       datetime
#            |
#            v
#      parse dates
#            |
#            v
#        validate
#            |
#            v
#       pandas / NumPy
#            |
#            v
#         analysis
#            |
#            v
#         results
#            |
#            v
#      CSV / JSON
#            |
#            v
#        reports


# ============================================================
# 190. WHEN TO USE PANDAS INSTEAD
# ============================================================

# The csv module is excellent for:
#
#     learning
#     simple files
#     streaming rows
#     lightweight scripts
#     standard-library-only projects
#
#
# pandas is often better for:
#
#     large tabular analysis
#     filtering
#     grouping
#     joins
#     missing-value handling
#     statistics
#     visualization preparation
#
#
# A common professional combination:
#
#     pathlib → discover files
#     pandas   → analyze tables
#     datetime → time logic
#     json     → configuration/API data
#     os       → environment


# ============================================================
# 191. PRACTICAL:
#     JSON API-STYLE DATA
# ============================================================

# Many APIs return structures resembling:
#
# {
#     "status": "success",
#     "data": [
#         {
#             "id": 1,
#             "name": "Ali"
#         },
#         {
#             "id": 2,
#             "name": "Sara"
#         }
#     ]
# }
#
#
# Typical workflow:
#
#     HTTP response text
#          ↓
#     response.json()
#          ↓
#     Python dict/list
#
#
# The requests library often performs the HTTP request,
# while json concepts explain the data representation.
#
# You do not necessarily call json.loads() yourself when an
# HTTP client already provides response.json().


# ============================================================
# 192. JSON SERIALIZATION OF datetime
# ============================================================

event = {
    "name": "meeting",
    "created_at": datetime.now().isoformat()
}

json_text = json.dumps(event, indent=4)

print(json_text)

# Excellent pattern:
#
#     datetime
#        ↓
#     isoformat()
#        ↓
#     string
#        ↓
#     JSON


# ============================================================
# 193. JSON DESERIALIZATION OF datetime
# ============================================================

json_text = """
{
    "name": "meeting",
    "created_at": "2026-09-09T15:30:00"
}
"""

data = json.loads(json_text)

created_at = datetime.fromisoformat(
    data["created_at"]
)

print(created_at)

# Reverse:
#
#     JSON
#      ↓
#     string
#      ↓
#     fromisoformat()
#      ↓
#     datetime


# ============================================================
# 194. PRACTICAL:
#     CONFIG + PATHLIB
# ============================================================

config_path = Path("config.json")

config = {
    "data_dir": "data",
    "output_dir": "outputs",
    "file_pattern": "*.csv"
}

config_path.write_text(
    json.dumps(config, indent=4),
    encoding="utf-8"
)

loaded_config = json.loads(
    config_path.read_text(encoding="utf-8")
)

data_dir = Path(
    loaded_config["data_dir"]
)

output_dir = Path(
    loaded_config["output_dir"]
)

pattern = loaded_config["file_pattern"]

print(data_dir)
print(output_dir)
print(pattern)


# ============================================================
# 195. THIS IS A REAL AUTOMATION PATTERN
# ============================================================

# Configuration:
#
#     config.json
#          |
#          v
#     json.load()
#          |
#          v
#     configuration object
#          |
#          +── input folder
#          +── output folder
#          +── file pattern
#          |
#          v
#       pathlib
#          |
#          v
#     discover files
#          |
#          v
#       process
#          |
#          v
#       outputs


# ============================================================
# 196. PRACTICAL:
#     CREATE A DAILY REPORT PATH
# ============================================================

today = date.today()

report_folder = (
    Path("reports")
    / str(today.year)
    / f"{today.month:02d}"
    / f"{today.day:02d}"
)

report_folder.mkdir(
    parents=True,
    exist_ok=True
)

report_file = report_folder / "daily_report.json"

report = {
    "date": today.isoformat(),
    "created_at": datetime.now().isoformat(),
    "status": "complete"
}

with report_file.open(
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        report,
        file,
        indent=4
    )

# Result conceptually:
#
# reports/
#   2026/
#     09/
#       09/
#         daily_report.json


# ============================================================
# 197. PRACTICAL:
#     DATE-BASED CSV LOG
# ============================================================

today = date.today()

log_folder = Path("logs")

log_folder.mkdir(
    parents=True,
    exist_ok=True
)

log_file = (
    log_folder
    / f"{today.isoformat()}.csv"
)

with log_file.open(
    "a",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    writer.writerow([
        datetime.now().isoformat(),
        "PROCESS",
        "completed"
    ])

# "a" means append.
#
# This is useful for logs:
#
#     don't erase existing records
#     add a new row


# ============================================================
# 198. FILE MODES — IMPORTANT
# ============================================================

# open() modes:
#
#     "r"   read
#     "w"   write / overwrite
#     "a"   append
#     "x"   create new file, fail if exists
#     "b"   binary mode modifier
#     "t"   text mode
#     "+"   update/read + write
#
#
# Common:
#
#     "r"
#     "w"
#     "a"
#     "rb"
#     "wb"
#
#
# WARNING:
#
#     "w"
#
# truncates/overwrites the file.


# ============================================================
# 199. WHY with open(...)?
# ============================================================

# Prefer:
#
#     with open(...) as file:
#         ...
#
#
# Because Python will properly close the file when leaving
# the context.
#
#
# Diagram:
#
#     with open()
#          |
#          v
#       file open
#          |
#          v
#       use file
#          |
#          v
#       leave block
#          |
#          v
#       file closed


# ============================================================
# 200. pathlib + with
# ============================================================

path = Path("notes.txt")

with path.open(
    "w",
    encoding="utf-8"
) as file:

    file.write("Hello")


# ============================================================
# 201. ENCODING — IMPORTANT
# ============================================================

# Prefer explicitly specifying:
#
#     encoding="utf-8"
#
# for text files when appropriate.
#
# This is especially important for:
#
#     Persian
#     Arabic
#     Unicode names
#     scientific text
#     international datasets
#
#
# Example:
#
path.write_text(
    "یادداشت Python",
    encoding="utf-8"
)


# ============================================================
# 202. NEWLINE AND CSV
# ============================================================

# When using csv with open(), the common recommended pattern is:
#
#     open(
#         file,
#         "w",
#         newline="",
#         encoding="utf-8"
#     )
#
#
# The newline="" setting lets the csv module handle newline
# behavior correctly across platforms.


# ============================================================
# 203. PRACTICAL:
#     CHECK FILE BEFORE OVERWRITING
# ============================================================

path = Path("important.json")

if path.exists():
    print("File already exists.")
else:
    save_json(
        {"status": "new"},
        path
    )


# ============================================================
# 204. PATH.exists() IS NOT ENOUGH
# ============================================================

path = Path("data")

if path.exists():
    if path.is_dir():
        print("Directory exists.")
    elif path.is_file():
        print("It is a file, not a directory.")

# A path can exist as:
#
#     file
#     directory
#     other filesystem object
#
# So ask the question you actually care about.


# ============================================================
# 205. PRACTICAL:
#     COUNT FILES BY EXTENSION
# ============================================================

folder = Path("data")

counts = {}

for path in folder.rglob("*"):

    if not path.is_file():
        continue

    extension = path.suffix.lower()

    counts[extension] = (
        counts.get(extension, 0) + 1
    )

print(counts)

# Example:
#
# {
#     ".csv": 12,
#     ".json": 5,
#     ".txt": 3
# }
#
#
# This combines:
#
#     pathlib
#     dictionaries
#     loops


# ============================================================
# 206. PRACTICAL:
#     FIND EMPTY FILES
# ============================================================

folder = Path("data")

for path in folder.rglob("*"):

    if not path.is_file():
        continue

    if path.stat().st_size == 0:
        print("Empty:", path)


# ============================================================
# 207. PRACTICAL:
#     FIND LARGE FILES
# ============================================================

folder = Path("data")

limit = 100 * 1024 * 1024  # 100 MiB

for path in folder.rglob("*"):

    if not path.is_file():
        continue

    size = path.stat().st_size

    if size > limit:
        print(
            "Large file:",
            path,
            size
        )


# ============================================================
# 208. PRACTICAL:
#     BUILD A FILE REPORT
# ============================================================

folder = Path("data")

report = []

for path in folder.rglob("*"):

    if not path.is_file():
        continue

    stat = path.stat()

    modified = datetime.fromtimestamp(
        stat.st_mtime
    )

    report.append({
        "name": path.name,
        "extension": path.suffix,
        "size_bytes": stat.st_size,
        "modified": modified.isoformat()
    })

print(report[:3])


# ============================================================
# 209. PRACTICAL:
#     SAVE FILE REPORT
# ============================================================

save_json(
    report,
    "outputs/file_report.json"
)


# ============================================================
# 210. THE FIVE LIBRARIES AS A SYSTEM
# ============================================================

# Think of them as different layers:
#
#
#             OPERATING SYSTEM
#                    |
#                   os
#                    |
#                    v
#             filesystem paths
#                    |
#                 pathlib
#                    |
#                    v
#                files
#                    |
#             +------+------+
#             |             |
#            CSV           JSON
#             |             |
#          csv module    json module
#             |             |
#             +------+------+
#                    |
#                    v
#              Python objects
#                    |
#                 datetime
#                    |
#                    v
#              date/time logic
#                    |
#                    v
#                 analysis


# ============================================================
# 211. THE "PYTHON FEELING" OF EACH LIBRARY
# ============================================================

# json
# ────
# "Turn structured Python data into portable text,
#  or turn JSON text back into Python data."
#
#
# csv
# ───
# "Read and write tables row by row."
#
#
# datetime
# ────────
# "Represent a moment, a calendar date, or a duration."
#
#
# pathlib
# ───────
# "Treat a filesystem path as a Python object."
#
#
# os
# ──
# "Talk to the operating system and its environment."


# ============================================================
# 212. MASTER DECISION TREE
# ============================================================

# I need to work with...
#
#                 FILE/DATA PROBLEM
#                       |
#         +-------------+-------------+
#         |             |             |
#       DATA           TIME          PATH/OS
#         |             |             |
#      +--+--+          |          +--+--+
#      |     |          |          |     |
#     CSV   JSON     datetime    pathlib  os
#      |     |          |          |     |
#    table nested     dates      paths  environment
#
#
# DATA:
#
#     rectangular table → CSV
#
#     nested records → JSON
#
# TIME:
#
#     calendar date → date
#
#     time of day → time
#
#     date + time → datetime
#
#     duration → timedelta
#
# PATH:
#
#     modern path manipulation → pathlib
#
# OS:
#
#     environment/process/system → os


# ============================================================
# 213. MASTER CONVERSION MAP
# ============================================================

# JSON:
#
#     Python
#       ↕
#     JSON string
#       ↕
#     JSON file
#
#
# DATETIME:
#
#     string
#       ↕
#     datetime
#
#     datetime
#       +
#     timedelta
#       ↓
#     datetime
#
#
# FILE:
#
#     Path
#       ↓
#     open
#       ↓
#     text/bytes
#       ↓
#     parser
#       ↓
#     Python objects


# ============================================================
# 214. PRACTICE LADDER
# ============================================================

# LEVEL 1 — BASIC JSON
#
# Must be able to:
#
#     json.dumps()
#     json.loads()
#     json.dump()
#     json.load()
#
#
# LEVEL 2 — BASIC CSV
#
# Must be able to:
#
#     csv.reader()
#     csv.writer()
#     csv.DictReader()
#     csv.DictWriter()
#
#
# LEVEL 3 — DATETIME
#
# Must be able to:
#
#     date.today()
#     datetime.now()
#     strptime()
#     strftime()
#     fromisoformat()
#     isoformat()
#     timedelta
#
#
# LEVEL 4 — PATHLIB
#
# Must be able to:
#
#     Path()
#     Path.cwd()
#     path.exists()
#     path.is_file()
#     path.is_dir()
#     path.mkdir()
#     path.open()
#     path.glob()
#     path.rglob()
#
#
# LEVEL 5 — OS
#
# Must be able to:
#
#     os.getcwd()
#     os.listdir()
#     os.environ
#     os.walk()
#     os.path basics
#
#
# LEVEL 6 — INTEGRATION
#
# Build:
#
#     CSV → clean → datetime conversion
#         → JSON output
#
#
# LEVEL 7 — AUTOMATION
#
# Build:
#
#     discover files
#         ↓
#     process all files
#         ↓
#     timestamp outputs
#         ↓
#     save reports
#
#
# LEVEL 8 — DATA ENGINEERING THINKING
#
# Build:
#
#     raw
#       ↓
#     validation
#       ↓
#     cleaning
#       ↓
#     processed
#       ↓
#     analysis
#       ↓
#     reproducible output


# ============================================================
# 215. EXERCISE IDEAS — PRACTICAL, NOT ABSTRACT
# ============================================================

# EXERCISE 1
#
# Create students.json containing:
#
#     name
#     age
#     score
#
# Read it and calculate the average score.
#
#
# EXERCISE 2
#
# Read a CSV dataset and:
#
#     strip whitespace
#     convert numeric columns
#     remove invalid rows
#     save cleaned CSV
#
#
# EXERCISE 3
#
# Read dates from:
#
#     "01/09/2026"
#
# Convert them to datetime objects.
#
#
# EXERCISE 4
#
# Find all CSV files under:
#
#     data/
#
# including subdirectories.
#
#
# EXERCISE 5
#
# Build:
#
#     data/raw/
#     data/processed/
#     reports/
#     outputs/
#
# automatically.
#
#
# EXERCISE 6
#
# Create a JSON experiment log containing:
#
#     experiment ID
#     model
#     parameters
#     metrics
#     timestamp
#
#
# EXERCISE 7
#
# Find the newest CSV file in a directory.
#
#
# EXERCISE 8
#
# Find files larger than 100 MiB.
#
#
# EXERCISE 9
#
# Create a daily report:
#
#     reports/YYYY/MM/DD/report.json
#
#
# EXERCISE 10
#
# Build a program that scans a directory and creates:
#
#     file_inventory.csv
#
# containing:
#
#     filename
#     extension
#     size
#     modification time


# ============================================================
# 216. FINAL CHEAT SHEET
# ============================================================

# JSON
# ────
#
#     json.dumps()
#     json.loads()
#     json.dump()
#     json.load()
#     JSONDecodeError
#
#
# CSV
# ───
#
#     csv.reader()
#     csv.writer()
#     csv.DictReader()
#     csv.DictWriter()
#     writerow()
#     writerows()
#     writeheader()
#
#
# DATETIME
# ────────
#
#     date
#     time
#     datetime
#     timedelta
#     timezone
#
#     today()
#     now()
#     strptime()
#     strftime()
#     fromisoformat()
#     isoformat()
#     combine()
#     replace()
#     weekday()
#     isoweekday()
#     total_seconds()
#
#
# PATHLIB
# ───────
#
#     Path()
#     cwd()
#     home()
#     absolute()
#     resolve()
#     exists()
#     is_file()
#     is_dir()
#     mkdir()
#     iterdir()
#     glob()
#     rglob()
#     open()
#     read_text()
#     write_text()
#     read_bytes()
#     write_bytes()
#     rename()
#     replace()
#     unlink()
#     rmdir()
#     stat()
#     with_name()
#     with_suffix()
#     relative_to()
#
#
# OS
# ──
#
#     getcwd()
#     chdir()
#     listdir()
#     mkdir()
#     makedirs()
#     remove()
#     rename()
#     replace()
#     rmdir()
#     environ
#     name
#     cpu_count()
#     getpid()
#     walk()
#
#     os.path.join()
#     os.path.exists()
#     os.path.isfile()
#     os.path.isdir()
#     os.path.abspath()
#     os.path.basename()
#     os.path.dirname()
#     os.path.splitext()
#     os.path.getsize()
#     os.path.getmtime()


# ============================================================
# 217. THE MOST IMPORTANT 20 THINGS TO REMEMBER
# ============================================================

# 1.  json.dumps()  → Python → JSON string
#
# 2.  json.loads()  → JSON string → Python
#
# 3.  json.dump()   → Python → JSON file
#
# 4.  json.load()   → JSON file → Python
#
# 5.  CSV is fundamentally row/column-oriented.
#
# 6.  CSV values are normally read as strings.
#
# 7.  DictReader/DictWriter are excellent for record-oriented
#     CSV processing.
#
# 8.  date = calendar date.
#
# 9.  datetime = date + time.
#
# 10. timedelta = duration/difference.
#
# 11. strptime() = string → datetime.
#
# 12. strftime() = datetime → formatted string.
#
# 13. isoformat() is excellent for machine-readable timestamps.
#
# 14. Use timezone-aware datetimes when timezone correctness
#     matters.
#
# 15. Path represents a filesystem location.
#
# 16. Path / "folder" / "file.csv" is the modern way to build
#     paths.
#
# 17. Path.glob()/rglob() are extremely useful for datasets.
#
# 18. Path.exists(), is_file(), and is_dir() answer different
#     filesystem questions.
#
# 19. os.environ is essential for environment configuration.
#
# 20. pathlib + csv/json + datetime is an extremely powerful
#     standard-library foundation for data automation.


# ============================================================
# 218. FINAL "PYTHON FEELING"
# ============================================================

#                         YOUR DATA PROJECT
#                                |
#                                v
#                         PATHLIB / OS
#                                |
#                         "Where is it?"
#                                |
#                                v
#                           FILE SYSTEM
#                                |
#                    +-----------+-----------+
#                    |                       |
#                   CSV                     JSON
#                    |                       |
#              "table data"            "structured data"
#                    |                       |
#                    +-----------+-----------+
#                                |
#                                v
#                         PYTHON OBJECTS
#                                |
#                                v
#                            DATETIME
#                                |
#                         "When was it?"
#                                |
#                                v
#                         CLEAN / ANALYZE
#                                |
#                                v
#                            RESULTS
#                                |
#                         +------+------+
#                         |             |
#                        CSV           JSON
#                         |             |
#                         +------+------+
#                                |
#                                v
#                             OUTPUT
#
#
# The deep idea is:
#
#     PATHLIB / OS
#         = location
#
#     CSV / JSON
#         = representation of data
#
#     DATETIME
#         = representation of time
#
#     Python
#         = processing and logic
#
#
# Once this mental model becomes natural, these libraries stop
# feeling like hundreds of unrelated functions.
#
#
# ============================================================
# END
# ============================================================
```
