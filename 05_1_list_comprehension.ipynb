```python
# ============================================================
# LIST COMPREHENSIONS
# ============================================================
#
# A list comprehension is a concise and Pythonic way to create
# a new list from an existing iterable.
#
# General Syntax:
#
""""
[expression for item in iterable]

"""
#
# It replaces many simple for-loops that append values to a list.
#
# Benefits
# --------
# ✔ Shorter code
# ✔ Easier to read
# ✔ Usually slightly faster
# ✔ Used everywhere in Data Science
#
# Example:
#
#     [x**2 for x in numbers]
#
# is equivalent to
#
#     result = []
#     for x in numbers:
#         result.append(x**2)
#
# ============================================================

numbers = [1, 2, 3, 4, 5]

squares = [x**2 for x in numbers] # it will square each number in the list

print(squares)
```

# Python List Comprehensions – A Data Science & ML Notebook


```python
# Python List Comprehensions – A Data Science & ML Notebook

# This notebook is structured like the example you shared. Each section explains **why** data scientists, data analysts, and machine learning practitioners use a particular pattern, gives concrete real‑world examples, and shows the code.


## 1. WHY DATA SCIENTISTS USE LIST COMPREHENSIONS

#
# Writing loops to build lists is common, but it adds clutter.
#
# List comprehensions are:
#
# • More concise – one line instead of several.
# • Faster – they run in C under the hood.
# • Easier to read – the intention is clear.
#
# In data work, you constantly create new lists from existing ones:
# normalising text, extracting features, filtering outliers, etc.
#
# List comprehensions make these tasks quick and elegant.
# ============================================================


# Traditional loop
squares = []
for x in range(10):
    squares.append(x**2)
print(squares)
```

```python

# List comprehension (same result)
squares_comp = [x**2 for x in range(10)]

print(squares_comp)
```

```python


# # 2. TEXT NORMALISATION (Categorical Variables)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================

# Text data is rarely consistent.

# Example:

#     Male
#     MALE
#     male

# Machine learning models treat these as different values.

# Before encoding categorical variables, we usually normalize
# text using lower().

# The same idea applies to:

# • Product names
# • Country names
# • User input
# • Categories

# This preprocessing step is extremely common in real projects.
# ============================================================
gender = ["Male", "MALE", "male"]
normalized = [g.lower() for g in gender]
print(normalized)

# Also remove extra whitespace
messy_names = ["  Alice ", "Bob\t", "Charlie\n", "  David  "]
clean_names = [name.strip().lower() for name in messy_names]
print(clean_names)
```

```python

## 3. HANDLING MISSING VALUES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Real datasets contain missing values (None, NaN, empty strings).
#
# Example:
#
#     age = [25, None, 30, None, 22]
#
# Many models cannot handle None/NaN directly.
#
# You often need to:
#
# • Replace missing values with a default (e.g., 0, mean, median)
# • Or remove rows with missing data
#
# List comprehensions give a quick way to do either.
# ============================================================


age = [25, None, 30, None, 22]

# Replace None with 0
filled_age = [a if a is not None else 0 for a in age]
print(filled_age)

# Remove missing values entirely
clean_age = [a for a in age if a is not None]
print(clean_age)
```

```python

## 4. EXTRACTING FEATURES FROM DATES OR STRINGS

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Dates, phone numbers, emails, etc. often come as strings.
#
# Example:
#
#     dates = ["2023-01-15", "2024-06-20", "2025-12-01"]
#
# You may want to extract the year, month, or day as a new feature.
#
# This is a common feature engineering step in:
#
# • Time series forecasting
# • Customer segmentation
# • Churn prediction
#
# List comprehensions let you split and parse in one line.
# ============================================================

dates = ["2023-01-15", "2024-06-20", "2025-12-01"]

# Extract year
years = [d.split('-')[0] for d in dates]
print(years)

# Convert to integer
years_int = [int(d.split('-')[0]) for d in dates]
print(years_int)

# Extract month and day as tuple
month_day = [(int(d.split('-')[1]), int(d.split('-')[2])) for d in dates]
print(month_day)
```

```python
## 5. FILTERING OUTLIERS

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Data often contains extreme values that distort models.
#
# Example:
#
#     prices = [120, 150, 130, 10000, 140, 110, 9000, 160]
#
# Outliers like 10000 can skew mean, variance, and regression.
#
# Common strategies:
#
# • Remove values outside a reasonable range
# • Cap (winsorize) extreme values
#
# A list comprehension with a condition does both cleanly.
# ============================================================

prices = [120, 150, 130, 10000, 140, 110, 9000, 160]

# Remove outliers > 500 (assume normal price range)
filtered_prices = [p for p in prices if p <= 500]
print(filtered_prices)

# Cap outliers at 500 (winsorizing)
capped_prices = [p if p <= 500 else 500 for p in prices]
print(capped_prices)
```

```python
## 6. CREATING NEW NUMERICAL FEATURES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Often you need to transform a feature to better fit a model.
#
# Examples:
#
# • Log transform to handle skewed distributions
# • Square root for count data
# • Length of text for NLP tasks
#
# In machine learning, feature engineering is key to performance.
#
# List comprehensions make applying transformations effortless.
# ============================================================

import math

values = [1, 10, 100, 0, 1000]

# Log transform (avoid log(0) by replacing with 0 or a small value)
log_values = [math.log(v) if v > 0 else 0 for v in values]
print([round(x, 2) for x in log_values])

# Length of product descriptions (for NLP)
descriptions = ['short', 'medium length', 'very long description']
desc_lengths = [len(d) for d in descriptions]
print(desc_lengths)
```

```python

## 7. COMBINING MULTIPLE FEATURES (PAIRWISE OPERATIONS)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# You often have two (or more) columns that together define a new feature.
#
# Example:
#
#     heights = [170, 160, 180]   # cm
#     weights = [70, 60, 80]      # kg
#
# Body Mass Index (BMI) = weight / (height in meters)^2
#
# This is a common derived feature in health analytics.
#
# List comprehensions with zip() let you combine multiple lists cleanly.
# ============================================================

heights = [170, 160, 180]  # cm
weights = [70, 60, 80]     # kg

# Compute BMI
bmi = [w / ((h/100)**2) for h, w in zip(heights, weights)]
print([round(b, 2) for b in bmi])

# Another example: price per unit
prices = [100, 200, 150]
quantities = [2, 3, 1]
price_per_unit = [p / q for p, q in zip(prices, quantities)]
print([round(x, 2) for x in price_per_unit])
```

```python

## 8. WORKING WITH PANDAS DATAFRAMES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# While pandas has vectorised operations, sometimes you need
# row‑wise transformations that are not easily vectorised.
#
# In such cases, apply() can be slow on large DataFrames.
#
# List comprehensions are often faster than apply() because
# they iterate in Python but with less overhead.
#
# Use them to:
#
# • Create new columns based on string operations
# • Extract parts of a string
# • Apply conditional logic
# ============================================================


import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'city': ['New York', 'Los Angeles', 'Chicago']
})

# Convert city to lower case (faster than df['city'].str.lower())
df['city_lower'] = [c.lower() for c in df['city']]

# Extract first letter of name
df['first_letter'] = [n[0] for n in df['name']]

# Flag if age >= 30
df['senior'] = [1 if a >= 30 else 0 for a in df['age']]

print(df)
```

```python

## 9. ONE‑HOT ENCODING (DUMMY VARIABLES)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Most ML models require numerical input.
#
# Categorical variables must be encoded.
#
# One‑hot encoding creates binary columns for each category.
#
# Instead of using pandas.get_dummies(), you can quickly
# generate dummy columns with a list comprehension if you
# need to control the output or integrate with other logic.
#
# This also demonstrates a nested comprehension.
# ============================================================

categories = ['A', 'B', 'A', 'C', 'B', 'A']
unique = sorted(set(categories))

# Create dummy matrix: each row is a list of 0/1
dummies = [[1 if cat == u else 0 for u in unique] for cat in categories]

# Display as a DataFrame for readability
dummy_df = pd.DataFrame(dummies, columns=unique)
print(dummy_df)
```

```python
## 10. PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# When dealing with millions of rows, performance matters.
#
# List comprehensions are implemented in C and are faster
# than explicit for‑loops with append().
#
# Let's measure the difference on a large list.
# ============================================================

import time

n = 10_000_000
data = list(range(n))

# Traditional loop
start = time.time()
squares_loop = []
for x in data:
    squares_loop.append(x**2)
end = time.time()
print(f"Loop time: {end - start:.4f} seconds")

# List comprehension
start = time.time()
squares_comp = [x**2 for x in data]
end = time.time()
print(f"Comprehension time: {end - start:.4f} seconds")
```

```python
# This is actually a very important concept that almost no Python tutorial explains well.

# When people say:

# "List comprehensions are optimized by CPython."

# many beginners think CPython is some special library. It isn't.

# What is CPython?

# CPython is the official and most widely used implementation of the Python programming language.

# When you install Python from:

# python.org
# Anaconda
# Miniconda

# you are almost certainly installing CPython.

# So when you type

# print("Hello")

# it is CPython that actually executes your code.

# Think of Python Like English

# Imagine English.

# Many people can speak English:

# Americans
# British
# Australians
# Canadians

# Same language.

# Different speakers.

# Python is similar.

# There are several programs that understand Python code.

# Implementation	Written In	Purpose
# CPython	C	Official implementation (most common)
# PyPy	Python + RPython	Faster execution using JIT compilation
# Jython	Java	Runs on the Java Virtual Machine (JVM)
# IronPython	C#	Runs on the .NET Framework
# MicroPython	C	Small devices and microcontrollers

# All of them understand Python syntax.

# Why is it Called CPython?

# Because it is written in the C programming language.

# Not C++.

# Not Python.

# Plain C.

# Millions of lines of C code implement Python.

# What Happens When You Run Python Code?

# Suppose you write

# numbers = [1, 2, 3]

# squares = [x**2 for x in numbers]

# You don't execute Python directly.

# The process is approximately:

# Your Python Code
#         │
#         ▼
# CPython Interpreter
#         │
#         ▼
# Bytecode
#         │
#         ▼
# Python Virtual Machine (PVM)
#         │
#         ▼
# CPU Executes Instructions

# CPython is the program that performs all these steps.

# Why Is It Written in C?

# C is:

# very fast
# close to the hardware
# efficient with memory
# portable

# Instead of rewriting Python for every operating system, the developers wrote CPython in C.

# Then they compile it for:

# Windows
# Linux
# macOS
# Why Does CPython Matter?

# Suppose you write

# result = []

# for x in numbers:
#     result.append(x)

# Every iteration CPython has to:

# Loop

# ↓

# Find the list object

# ↓

# Find append()

# ↓

# Call append()

# ↓

# Return

# ↓

# Continue loop

# That is quite a few operations for each element.

# Now compare

# result = [x for x in numbers]

# CPython recognizes this special syntax.

# Instead of repeatedly calling append(), it uses an internal, optimized mechanism to build the list more efficiently.

# Conceptually, it does something like:

# Allocate the result list

# ↓

# Iterate through the input

# ↓

# Insert each computed value directly

# ↓

# Return the finished list

# This reduces some of the overhead involved in repeated method lookups and calls.

# Why Is It Faster?

# Consider

# result = []

# for x in numbers:
#     result.append(x * 2)

# For every iteration, CPython must repeatedly execute Python-level operations such as:

# executing the loop
# looking up the append method
# calling append
# returning from that method

# Now compare

# result = [x * 2 for x in numbers]

# CPython knows you're constructing a list from the start, so it can execute the loop using a more specialized bytecode path with fewer interpreter-level operations.

# So the process is closer to:

# Allocate list

# ↓

# Loop internally

# ↓

# Compute value

# ↓

# Store value

# ↓

# Finish

# The reduction in interpreter overhead is why list comprehensions are often faster.

# Is It "Written in C" That Makes It Faster?

# Not exactly.

# A common misconception is:

# "List comprehensions are faster because they're written in C."

# That's only partly true.

# The entire CPython interpreter is written in C—including ordinary for loops.

# The reason list comprehensions are often faster is that CPython has a specialized implementation for them, allowing the interpreter to execute fewer bytecode instructions and avoid some of the overhead of repeated append() method calls.

# A More Accurate Performance Explanation for Your Notebook

# Instead of writing:

# List comprehensions are faster because they are written in C.

# I would write:

# # ============================================================
# # WHY ARE LIST COMPREHENSIONS OFTEN FASTER?
# # ============================================================
# #
# # Both regular for-loops and list comprehensions are executed
# # by CPython (the official Python interpreter).
# #
# # However, list comprehensions have a specialized implementation
# # inside CPython that reduces some of the overhead involved in
# # repeatedly calling list.append().
# #
# # This means that, for simple operations, list comprehensions
# # usually execute fewer interpreter-level instructions than an
# # equivalent for-loop.
# #
# # As a result, they are often faster while also producing
# # shorter and more readable code.
# #
# # Note:
# # For numerical computations on large arrays, NumPy's vectorized
# # operations are still much faster because they execute compiled
# # code over entire arrays rather than element by element.
# # ============================================================

# This explanation is technically accurate, explains what CPython is, and gives the real reason list comprehensions tend to outperform equivalent append() loops without oversimplifying the underlying implementation.
```
