##    PYTHON CONDITIONAL STATEMENTS
### How to Control with basic True/False Logic conditions
#### Basics of True/False Logic


```python
# ============================================================
#             ANY() AND ALL() FUNCTIONS
# ============================================================
#
# any() returns True if at least one item in an iterable is truthy.
#
# all() returns True if every item in an iterable is truthy.
#
# For strings:
#     ""       → False (empty string)
#     "Ali"    → True  (non-empty string)
#
# ============================================================


# ============================================================
# 1. CHECK IF ANY VALUE IS NOT EMPTY
# ============================================================

email = "Ali"              # Variable to store email address
phone = "0923456789"       # Variable to store phone number
username = "Ali001"        # Variable to store username

print("If any of email, phone, or username is not empty, print True:")
print(any([email, phone, username]))


# ============================================================
# 2. CHECK IF ALL VALUES ARE NOT EMPTY
# ============================================================

print("If all of email, phone, and username are not empty, print True:")
print(all([email, phone, username]))
```

```python
# ============================================================
#             ANY() AND ALL() WITH DIFFERENT DATA TYPES
# ============================================================
#
# any() → True if AT LEAST ONE item is truthy.
#
# all() → True if ALL items are truthy.
#
# They can work with:
#     - strings
#     - numbers
#     - booleans
#     - lists
#     - tuples
#     - sets
#     - dictionaries
#     - conditions / comparisons
#
# ============================================================


# ============================================================
# 1. WITH STRINGS
# ============================================================

names = ["Ali", "", "Yasin"]
print(names)
print("at least one string is non-empty? (any):", any(names))   # True  → at least one string is non-empty
print("False → not all strings are non-empty? (all):", all(names))   # False → not all strings are non-empty
```

```python
# ============================================================
# 2. WITH NUMBERS
# ============================================================

numbers = [0, 0, 5, 0]

print(any(numbers))   # True  → at least one number is non-zero
print(all(numbers))   # False → not all numbers are non-zero
```

```python
# ============================================================
# 3. WITH BOOLEANS
# ============================================================

values = [False, False, True]

print(any(values))   # True  → at least one value is True
print(all(values))   # False → not all values are True
```

```python
# ============================================================
# 4. WITH CONDITIONS
# ============================================================

age = 25
score = 85
balance = 100

print(any([
    age >= 18,
    score >= 90,
    balance > 500
]))
# True → age >= 18 is True


print(all([
    age >= 18,
    score >= 80,
    balance > 50
]))
# True → all three conditions are True
```

```python
# ============================================================
# 5. WITH A LIST
# ============================================================

items = ["Python", "SQL", "Power BI"]

print(any(items))   # True → at least one item exists
print(all(items))   # True → all items are non-empty
```

```python
# ============================================================
# 6. WITH AN EMPTY LIST
# ============================================================

empty_list = []

print(any(empty_list))   # False
print(all(empty_list))   # True
#
# IMPORTANT:
# any([]) → False
# all([]) → True
#
# This is a special rule of Python.
# ============================================================
```

```python
# ============================================================
# 7. WITH A DICTIONARY
# ============================================================

user = {
    "email": "ali@example.com",
    "phone": "",
    "username": "Ali001"
}

print(any(user.values()))
# True → at least one value is non-empty

print(all(user.values()))
# False → phone is empty
```

```python
# ============================================================
# 8. WITH CONDITIONS IN A LIST COMPREHENSION
# ============================================================

numbers = [10, 20, 30, 40, 50]

print(any(number > 40 for number in numbers))
# True → 50 is greater than 40

print(all(number > 5 for number in numbers))
# True → every number is greater than 5
```

```python
# ============================================================
# QUICK SUMMARY
# ============================================================
#
# any() → "Is there AT LEAST ONE True value?"
#
# all() → "Are ALL values True?"
#
# Examples:
#
# any([False, False, True]) → True
# any([False, False, False]) → False
#
# all([True, True, True])   → True
# all([True, False, True])  → False
#
# ============================================================
```

```python
# ============================================================
#                    isinstance()
# ============================================================
#
# isinstance() checks whether a value belongs to a specific
# data type (or class).
#
# Basic structure:
#
# isinstance(value, type)
#
# It returns:
#     True  → if the value is the specified type
#     False → if the value is not the specified type
#
# ============================================================


# ============================================================
# 1. CHECK IF A VALUE IS AN INTEGER
# ============================================================

age = 25

print(isinstance(age, int))
# True
```

```python
# ============================================================
# 2. CHECK IF A VALUE IS A STRING
# ============================================================
#
# isinstance(value, type) checks whether a value belongs to
# a specified type and returns True or False.
#
#     value
#       ↓
#     isinstance(value, type)
#       ↓
#     True / False
#
# Here, name refers to the string "Ali", so we check whether
# it is an instance of the str type.

name = "Ali"

print(isinstance(name, str))
# True
#
# HOW IT WORKS:
#
#     name → "Ali"
#          ↓
#     isinstance(name, str)
#          ↓
#     Is "Ali" a string?
#          ↓
#         True
#
# IMPORTANT:
# - str is Python's built-in string type.
# - isinstance() returns a Boolean value: True or False.
# - The check does not change the value stored in name.
#
# GENERAL PATTERN:
#
#     isinstance(value, type)
#
# Example:
#
#     isinstance(10, int)       # True
#     isinstance(3.14, float)   # True
#     isinstance("10", int)     # False
#
# CORE IDEA:
# Use isinstance() when a program needs to check the type
# of a value before deciding what to do with it.
```

```python
# ============================================================
# 3. CHECK IF A VALUE IS A FLOAT
# ============================================================

price = 19.99

print(isinstance(price, float))
# True
```

```python
# ============================================================
# 4. CHECK IF A VALUE IS A BOOLEAN
# ============================================================

is_logged_in = True

print(isinstance(is_logged_in, bool))
# True
```

```python
# ============================================================
# 5. CHECKING THE WRONG TYPE
# ============================================================

age = 25

print(isinstance(age, str))
# False

print(isinstance(age, float))
# False
```

```python
# ============================================================
# 6. CHECK IF A VALUE IS A LIST
# ============================================================
#
# isinstance(value, type) checks whether a value belongs to
# a specified type and returns True or False.
#
# Here, students refers to a list, so we check whether it is
# an instance of the list type.

students = ["Ali", "Yasin", "Sina"]

print(isinstance(students, list))
# True
#
# HOW IT WORKS:
#
#     students → ["Ali", "Yasin", "Sina"]
#             ↓
#     isinstance(students, list)
#             ↓
#     Is students a list?
#             ↓
#            True
#
# IMPORTANT:
# - list is Python's built-in list type.
# - isinstance() returns a Boolean value: True or False.
# - The check does not change the list.
#
# GENERAL PATTERN:
#
#     isinstance(value, type)
#
# Example:
#
#     isinstance([1, 2, 3], list)   # True
#     isinstance("Ali", list)       # False
#
# CORE IDEA:
# Use isinstance() to check whether a value is a particular
# type before performing type-dependent operations.
```

```python
# ============================================================
# 7. CHECK MULTIPLE TYPES
# ============================================================
#
# isinstance() can check whether a value matches ANY type
# listed in a tuple.
#
#     isinstance(value, (type1, type2, ...))
#
# Think:
#
#     value
#       ↓
#     int OR float?
#       ↓
#     True / False
# ============================================================

value = 25

print(isinstance(value, (int, float)))
# True → 25 is an int, so it matches one of the allowed types.


value = 19.5

print(isinstance(value, (int, float)))
# True → 19.5 is a float, so it matches one of the allowed types.


value = "25"

print(isinstance(value, (int, float)))
# False → "25" is a string, so it matches neither int nor float.


# ============================================================
# VALUE / TYPE TRACK
# ============================================================
#
# value = 25
#     → int
#     → True
#
# value = 19.5
#     → float
#     → True
#
# value = "25"
#     → str
#     → False
#
# IMPORTANT:
#
#     (int, float)
#
# means:
#
#     "int OR float"
#
# NOT:
#
#     "int AND float"
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Use isinstance() with a tuple when multiple types are
# acceptable:
#
#     isinstance(value, (int, float))
#
# It returns True when the value matches at least ONE
# of the specified types.
# ============================================================
```

```python
# ============================================================
# 8. USING isinstance() WITH IF
# ============================================================

value = 100

if isinstance(value, int):
    print("The value is an integer.")
else:
    print("The value is not an integer.")
```

```python
# ============================================================
# 8. CHECK IF A TUPLE IS A LIST
# ============================================================
#
# This example demonstrates two related ideas:
#
#     type(value)
#     → tells us the value's type
#
#     isinstance(value, list)
#     → checks whether the value is a list
#
# Here, value is a tuple, not a list.
# ============================================================

value = (100000, 321)

print(type(value))
# <class 'tuple'>
# Parentheses with comma-separated values create a tuple.
#
# VALUE:
#     (100000, 321)
#       ↓
#     tuple
#
# The variable 'value' refers to a tuple object.


if isinstance(value, list):
    # Python asks:
    #
    #     "Is value a list?"
    #
    # value → tuple
    #          ↓
    #     list? → NO
    #
    # Therefore, the condition is False and this block
    # is skipped.

    print("The value is a list.")

else:
    # Because isinstance(value, list) returned False,
    # Python executes the else block.

    print("The value is not a list.")


# ============================================================
# OUTPUT
# ============================================================
#
# <class 'tuple'>
# The value is not a list.
#
# IMPORTANT:
#
#     type(value)
#     → reports the exact type information.
#
#     isinstance(value, list)
#     → asks whether value is a list and returns True/False.
#
#
# ============================================================
# DECISION FLOW
# ============================================================
#
# value
#   ↓
# tuple
#   ↓
# isinstance(value, list)
#   ↓
# False
#   ↓
# else block
#   ↓
# "The value is not a list."
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Use isinstance() when you need to make a decision based
# on an object's type:
#
#     isinstance(value, list)
#
# returns True for a list and False for other types, such as
# tuple, string, integer, or float.
# ============================================================
```

```python
# ============================================================
# 4. CHECKING MULTIPLE POSSIBILITIES
# ============================================================
#
# endswith() can check whether a string ends with ANY of
# several specified values.
#
#     string.endswith((value1, value2, ...))
#
# The tuple contains the possible endings to check.
# ============================================================

filename = "photo.jpg"

print(filename.endswith((".jpg", ".png", ".gif")))
# True → "photo.jpg" ends with ".jpg".


filename = "document.pdf"

print(filename.endswith((".jpg", ".png", ".gif")))
# False → "document.pdf" ends with none of the specified endings.


# ============================================================
# HOW IT WORKS
# ============================================================
#
# "photo.jpg"
#     ↓
# endswith((".jpg", ".png", ".gif"))
#     ↓
# matches ".jpg"?
#     ↓
# YES → True
#
# "document.pdf"
#     ↓
# endswith((".jpg", ".png", ".gif"))
#     ↓
# matches any?
#     ↓
# NO → False
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A tuple passed to endswith() lets you test multiple
# possibilities in one operation.
#
#     filename.endswith((".jpg", ".png", ".gif"))
#
# means:
#
#     "Does filename end with .jpg OR .png OR .gif?"
# ============================================================
```

```python
filename = "photo.jpg"

# Check whether the filename ends with any of these image extensions.
print(filename.endswith((".jpg", ".png", ".gif")))
# True → "photo.jpg" ends with ".jpg".


# Check if the file has an image extension.
if filename.endswith((".jpg", ".png", ".gif")):
    # The condition is True because filename ends with ".jpg".
    print("This is an image file.")
else:
    # This block runs only when the filename ends with none
    # of the specified image extensions.
    print("This is not an image file.")


# ============================================================
# DECISION FLOW
# ============================================================
#
# "photo.jpg"
#      ↓
# endswith((".jpg", ".png", ".gif"))
#      ↓
# matches ".jpg"?
#      ↓
#     True
#      ↓
# execute if block
#      ↓
# "This is an image file."
#
#
# CORE IDEA:
# endswith() can test multiple possible endings at once.
# The if statement then uses the resulting True/False value
# to choose which block of code to execute.
# ============================================================
```

#             PYTHON CONDITIONAL STATEMENTS


##  IF

```python
# ============================================================
#             PYTHON CONDITIONAL STATEMENTS
# ============================================================
#
# Conditional statements allow Python to make decisions.
#
# Basic structure:
#
# if condition(True/False):
#     code
#
# The condition is evaluated first.
#
#     condition
#        ↓
#     True / False
#        ↓
#     True  → execute the indented block
#     False → skip the block
#
# elif and else can be used when multiple execution paths
# are needed.
#
# ============================================================


# ============================================================
# 1. BASIC IF STATEMENT
# ============================================================

age = 25

# Python evaluates the condition:
#
#     age >= 18
#
#     25 >= 18
#         ↓
#       True
#
# Because the condition is True, Python executes the
# indented print() statement.

if age >= 18:
    print("You are an adult.")


# OUTPUT:
# You are an adult.
#
# CORE IDEA:
# An if statement executes its block only when its condition
# evaluates to True.
# ============================================================
```

## IF + ELSE

```python
# ============================================================
# 2. IF + ELSE
# ============================================================
#
# if + else creates two possible execution paths:
#
#     condition
#          ↓
#     True / False
#      ↙       ↘
#    if        else
#
# True  → execute the if block
# False → execute the else block
# ============================================================

# Assign a value to the variable 'age'.
age = 16

# Python evaluates the condition:
#
#     age >= 18
#     16 >= 18
#         ↓
#       False
#
# Because the condition is False, Python skips the if block
# and executes the else block.

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")


# OUTPUT:
# You are a minor.
#
# CORE IDEA:
# The condition determines which branch Python executes:
#
#     True  → if
#     False → else
# ============================================================
```

## IF + ELIF + ELSE

```python
# ============================================================
# 3. IF + ELIF + ELSE
# ============================================================
#
# This structure provides multiple possible execution paths.
#
#     condition 1 → True  → execute its block
#            ↓ False
#     condition 2 → True  → execute its block
#            ↓ False
#        else block
#
# Python checks the conditions from top to bottom.
# Once one condition is True, its block executes and the
# remaining elif/else branches are skipped.
# ============================================================

age = 25

if age < 13:
    print("Child")

elif age < 18:
    print("Teenager")

else:
    print("Adult")


# ============================================================
# EXECUTION
# ============================================================
#
# age = 25
#
# Check 1:
#     age < 13
#     25 < 13 → False
#     ↓
#     continue to elif
#
# Check 2:
#     age < 18
#     25 < 18 → False
#     ↓
#     execute else
#
# OUTPUT:
#     Adult
#
# ============================================================
# CORE IDEA
# ============================================================
#
# An if/elif/else chain selects ONE execution path.
#
#     if    → first condition
#     elif  → additional conditions
#     else  → runs when all previous conditions are False
# ============================================================
```

```python
# ============================================================
# 4. MULTIPLE CONDITIONS
# ============================================================
#
# The 'and' operator requires ALL conditions to be True.
#
# condition 1 AND condition 2
#             ↓
#        both True?
#        ↙       ↘
#      YES        NO
#       ↓          ↓
#      if        else
# ============================================================

age = 30
has_license = True

# Python evaluates both conditions:
#
#     age >= 18
#     30 >= 18 → True
#
#     has_license
#     True → True
#
# 'and' requires both results to be True:
#
#     True and True → True
#
# Therefore, the if block executes.

if age >= 18 and has_license:
    print("You can drive.")
else:
    print("You cannot drive.")


# OUTPUT:
# You can drive.
#
# ============================================================
# CORE IDEA
# ============================================================
#
# 'and' combines conditions:
#
#     True  AND True  → True
#     True  AND False → False
#     False AND True  → False
#     False AND False → False
#
# The if block runs only when ALL required conditions are True.
# ============================================================
```

```python
# ============================================================
# 5. COMPARISON OPERATORS
# ============================================================
#
# Comparison operators compare two values and return a
# Boolean result: True or False.
# ============================================================

x = 10
y = 20

print("COMPARISON OPERATORS:\n")
print(f"is {x} == {y} ? : {x == y}")  # equal to
print(f"is {x} != {y} ? : {x != y}")  # not equal to
print(f"is {x} > {y} ? : {x > y}")    # greater than
print(f"is {x} < {y} ? : {x < y}")    # less than
print(f"is {x} >= {y} ? : {x >= y}")  # greater than or equal to
print(f"is {x} <= {y} ? : {x <= y}")  # less than or equal to


# ============================================================
# COMPARISON OPERATORS QUICK REFERENCE
# ============================================================
#
# | Operator | Meaning                  | Example   | Result |
# |----------|--------------------------|-----------|--------|
# | ==       | equal to                 | 10 == 20  | False  |
# | !=       | not equal to             | 10 != 20  | True   |
# | >        | greater than             | 10 > 20   | False  |
# | <        | less than                | 10 < 20   | True   |
# | >=       | greater than or equal to | 10 >= 20  | False  |
# | <=       | less than or equal to    | 10 <= 20  | True   |
#
# IMPORTANT:
#
#     =   → assignment
#     ==  → comparison
#
# For example:
#
#     x = 10
#     → assigns 10 to x
#
#     x == 10
#     → asks whether x equals 10
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Comparison:
#
#     value + operator + value
#                ↓
#           True / False
#
# These Boolean results are commonly used as conditions
# in if statements and loops.
# ============================================================
```

```python
# ============================================================
# 6. USING COMPARISON OPERATORS IN IF
# ============================================================
#
# A comparison operator produces True or False.
# An if statement then uses that Boolean result to make a
# decision.
#
#     score
#       ↓
#     score >= 50
#       ↓
#     True / False
#       ↓
#     choose if / else
# ============================================================

score = 85

# Python evaluates:
#
#     score >= 50
#     85 >= 50
#         ↓
#       True
#
# Because the condition is True, the if block executes.

if score >= 50:
    print("Pass")
else:
    print("Fail")


# ============================================================
# OUTPUT
# ============================================================
#
# Pass
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Comparison operators and conditional statements work
# together:
#
#     comparison
#         ↓
#     True / False
#         ↓
#     if makes a decision
#         ↓
#     selected block executes
#
# Here, 85 >= 50 is True, so "Pass" is printed.
# ============================================================
```

```python
# ============================================================
# 7. AND
# ============================================================
#
# The 'and' operator combines two or more conditions.
#
# ALL conditions must be True for the complete expression
# to be True.
#
#     condition 1
#          AND
#     condition 2
#          ↓
#     True only if BOTH are True
# ============================================================

age = 25
salary = 50000

# Python evaluates both comparisons:
#
#     age >= 18
#     25 >= 18 → True
#
#     salary >= 30000
#     50000 >= 30000 → True
#
# Then 'and' combines the results:
#
#     True and True → True
#
# Therefore, the if block executes.

if age >= 18 and salary >= 30000:
    print("Condition satisfied.")


# ============================================================
# AND QUICK REFERENCE
# ============================================================
#
# | Condition 1 | Condition 2 | Result |
# |-------------|-------------|--------|
# | True        | True        | True   |
# | True        | False       | False  |
# | False       | True        | False  |
# | False       | False       | False  |
#
# ============================================================
# OUTPUT
# ============================================================
#
# Condition satisfied.
#
# CORE IDEA:
#
# 'and' means ALL required conditions must be True.
#
#     condition 1 + condition 2
#                ↓
#          both must be True
#                ↓
#              True
# ============================================================
```

```python
# ============================================================
# 8. OR
# ============================================================
#
# The 'or' operator combines conditions where ANY one of them
# can be True.
#
#     condition 1
#          OR
#     condition 2
#          ↓
#     True if at least ONE is True
# ============================================================

# Assign the value "IT" to the variable 'department'.
department = "IT"

# Python evaluates the two comparisons:
#
#     department == "IT"
#     "IT" == "IT" → True
#
#     department == "Data Science"
#     "IT" == "Data Science" → False
#
# Then 'or' combines the results:
#
#     True or False → True
#
# Therefore, the if block executes.

if department == "IT" or department == "Data Science":
    print("Technology-related department.")


# ============================================================
# OR QUICK REFERENCE
# ============================================================
#
# | Condition 1 | Condition 2 | Result |
# |-------------|-------------|--------|
# | True        | True        | True   |
# | True        | False       | True   |
# | False       | True        | True   |
# | False       | False       | False  |
#
# ============================================================
# OUTPUT
# ============================================================
#
# Technology-related department.
#
# CORE IDEA:
#
# 'or' means at least ONE condition must be True.
#
#     condition 1 OR condition 2
#              ↓
#     at least one True?
#              ↓
#         True / False
# ============================================================
```

```python
# ============================================================
# 9. NOT
# ============================================================
#
# The 'not' operator reverses a Boolean value:
#
#     not True  → False
#     not False → True
#
# It is useful when we want to check that a condition is NOT
# true.
# ============================================================

# Store the Boolean value False in 'is_student'.
is_student = False

# Python evaluates:
#
#     is_student
#     → False
#
# Then 'not' reverses it:
#
#     not False
#     → True
#
# Because the condition is True, the if block executes.

if not is_student:
    print("The person is not a student.")


# ============================================================
# QUICK REFERENCE
# ============================================================
#
# | Original | With 'not' |
# |----------|------------|
# | True     | False      |
# | False    | True       |
#
# ============================================================
# CORE IDEA
# ============================================================
#
# 'not' means "reverse the Boolean result".
#
#     False
#       ↓
#      not
#       ↓
#     True
#       ↓
#     if block executes
# ============================================================
```

```python
# ============================================================
# 10. COMBINING AND / OR 
# ============================================================
#
# Logical operators can be combined to build more complex
# conditions.
#
#     and → ALL conditions must be True
#     or  → AT LEAST ONE condition must be True
#     not → reverses True / False
#
# Parentheses make the intended grouping clear.
# ============================================================


# Initialize variables with candidate information.
age = 30
experience = 6
has_degree = True


# The condition is:
#
#     (age >= 25 and experience >= 5) or has_degree
#
# First, Python evaluates the expression inside parentheses:
#
#     age >= 25
#     30 >= 25 → True
#
#     experience >= 5
#     6 >= 5 → True
#
#     True and True → True
#
# Then:
#
#     True or has_degree
#     True or True → True
#
# Therefore, the if block executes.

if (age >= 25 and experience >= 5) or has_degree:
    print("Candidate satisfies the requirement.")


# ============================================================
# LOGICAL FLOW
# ============================================================
#
#     age >= 25 ──┐
#                 AND → True ──┐
# experience >= 5 ─┘           │
#                              OR → True
#     has_degree ──────────────┘
#                                ↓
#                          execute if block
#
# NOTE:
# 'not' is not used in this particular example; the section
# demonstrates how AND and OR can be combined.
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Combine logical operators to express multi-part decisions.
#
# Parentheses help Python and the reader understand which
# conditions should be evaluated together.
# ============================================================
```

```python
# ============================================================
# 10.1 COMBINING AND / OR / NOT
# ============================================================
#
# We can combine AND, OR, and NOT to create a simple
# real-world decision.
#
# Example:
# A student can enter an advanced course if:
#
#   (They passed the exam AND have enough study hours)
#   OR
#   they are NOT a beginner.
#
# This demonstrates how several Boolean conditions can
# work together in one if statement.

# Student information
passed_exam = True
study_hours = 12
is_beginner = False

# Check whether the student can enter the advanced course
if (passed_exam and study_hours >= 10) or not is_beginner:
    print("Student can enter the advanced course.")
else:
    print("Student cannot enter the advanced course.")


# ------------------------------------------------------------
# HOW PYTHON EVALUATES THE CONDITION
# ------------------------------------------------------------
#
# First condition:
#
#     passed_exam
#     True
#
# Second condition:
#
#     study_hours >= 10
#     12 >= 10 → True
#
# AND:
#
#     True and True → True
#
# Third condition:
#
#     is_beginner
#     False
#
# NOT:
#
#     not False → True
#
# Finally, OR:
#
#     True or True → True
#
# The complete condition is True, so the if block runs.
#
#
# SIMPLE VALUE FLOW:
#
#   passed_exam ────────→ True ──┐
#                                AND → True ──┐
#   study_hours >= 10 → True ───┘            │
#                                             OR → True
#   is_beginner ──→ False ──→ NOT → True ────┘
#                                                   ↓
#                                              if branch
#
#
# ------------------------------------------------------------
# CORE IDEA
# ------------------------------------------------------------
#
# AND → requires ALL connected conditions to be True.
#
# OR  → requires AT LEAST ONE condition to be True.
#
# NOT → reverses a Boolean value.
#
# Parentheses group related conditions and make the logic
# easier for humans to understand.
#
# OUTPUT:
# Student can enter the advanced course.
```

```python
# ============================================================
# 11. NESTED IF
# ============================================================
#
# A nested if is an if statement inside another if statement.
# It is useful when decisions must be made in stages.
#
# This example first checks the age requirement.
# If the person is old enough, it then checks the license.
#
# Decision flow:
#
#     age >= 18?
#       ↓
#     YES → check license
#       ↓
#     has_license?
#       ↙        ↘
#     YES        NO
#      ↓          ↓
#   can drive   need license
#
#     NO → too young
# ============================================================


age = 30
has_license = True

# First, Python checks whether the person is at least 18.
#
#     age >= 18
#     30 >= 18
#     → True
#
# Because the condition is True, Python enters this block.
if age >= 18:

    # Now Python checks the second condition.
    #
    # 'has_license' contains the Boolean value True,
    # so this condition is satisfied.
    if has_license:
        print("You can drive.")

    # This else belongs to the NESTED if above.
    # It runs when the person is old enough but does not
    # have a license.
    else:
        print("You need a driving licence.")

# This else belongs to the OUTER if.
# It runs when age >= 18 is False.
else:
    print("You are too young to drive.")


# ============================================================
# HOW IT WORKS
# ============================================================
#
# Execution with the current values:
#
#     age = 30
#     has_license = True
#
#     30 >= 18
#         ↓
#       True
#         ↓
#     check has_license
#         ↓
#       True
#         ↓
#     print("You can drive.")
#
# The outer 'else' is skipped, and the nested 'else' is also
# skipped because both conditions are True.
#
#
# ============================================================
# STATE
# ============================================================
#
# BEFORE:
#     age = 30
#     has_license = True
#
# DURING:
#     age >= 18 → True
#     has_license → True
#
# AFTER:
#     No variables are changed.
#     The conditions only determine which code block executes.
#
#
# ============================================================
# WHY THIS WORKS
# ============================================================
#
# The outer condition controls whether Python reaches the
# nested condition.
#
# Therefore:
#
#     first condition
#         ↓
#     if True → evaluate second condition
#         ↓
#     second condition
#         ↓
#     choose final action
#
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# Pay attention to indentation and which 'else' belongs
# to which 'if'.
#
#     if age >= 18:
#         if has_license:
#             ...
#         else:
#             ...
#     else:
#         ...
#
# The inner 'else' belongs to 'if has_license'.
# The outer 'else' belongs to 'if age >= 18'.
#
#
# ============================================================
# OUTPUT
# ============================================================
#
# You can drive.
#
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Nested if statements allow Python to make decisions
# step-by-step: the inner condition is checked only when
# execution reaches that nested block.
# ============================================================
```

```python
# ============================================================
# 2. is CHECKS OBJECT IDENTITY
# ============================================================
#
# 'is' checks whether two variables refer to the SAME object.
#
# This is different from '==':
#
#     ==  → same VALUE?
#     is  → same OBJECT?
# ============================================================


x = 10

# x refers to the integer object 10.
#
#     x ───→ 10


y = x

# Assignment makes 'y' refer to the same object that 'x'
# currently refers to.
#
#     x ─────┐
#            ↓
#           10
#            ↑
#     y ─────┘
#
# No new value is created by 'y = x' here; both names now
# refer to the same object.

print(x is y)
# True
#
# Python asks:
#
#     "Do x and y refer to the same object?"
#
#     x ─────┐
#            ↓
#           10
#            ↑
#     y ─────┘
#
#     x is y → True
#
#
# ============================================================
# QUICK MEMORY CHECK
# ============================================================
#
#     y = x
#       ↓
#     y refers to the object x refers to
#       ↓
#     x is y
#       ↓
#     True
#
# Remember:
#
#     ==  → compares values
#     is  → compares object identity
# ============================================================
```

```python
# ============================================================
# 3. "is" NOT THE SAME AS ==
# ============================================================

x = [1, 2, 3]
y = [1, 2, 3]

print(x == y)
# True
#
# The lists contain the same values.


print(x is y)
# False
#
# They are two different list objects.
#
# x → [1, 2, 3]
# y → [1, 2, 3]
#
# Same contents, different objects.
```

```python
# ============================================================
# 3. == VS is WITH LISTS
# ============================================================
#
# With lists, == and is can be clearly different:
#
#     ==  → compares the CONTENTS of the lists
#     is  → checks whether both variables refer to the SAME object
#
# Important:
#     y = x does NOT create a copy.
#     It makes y refer to the same list as x.
# ============================================================


# Create a list object containing three integers.
x = [1, 2, 3]

# Make y refer to the SAME list object as x.
#
#     x ─────┐
#            ↓
#        [1, 2, 3]
#            ↑
#     y ─────┘
#
# No separate list is created.

y = x


# ============================================================
# == CHECKS VALUE / CONTENT EQUALITY
# ============================================================
#
# Python compares the contents of the two lists:
#
#     x → [1, 2, 3]
#     y → [1, 2, 3]
#              ↓
#     same contents → True
#
print(x == y)
# True


# ============================================================
# is CHECKS OBJECT IDENTITY
# ============================================================
#
# Python asks:
#
#     "Do x and y refer to the exact same list object?"
#
#     x ─────┐
#            ↓
#        [1, 2, 3]
#            ↑
#     y ─────┘
#
# Yes → True
print(x is y)
# True


# ============================================================
# QUICK MEMORY CHECK
# ============================================================
#
#     x = [1, 2, 3]
#     y = x
#
#     x == y
#         ↓
#     same contents? → True
#
#     x is y
#         ↓
#     same object? → True
#
# KEY IDEA:
#
#     == → "Do they contain equal values?"
#     is → "Are they the same object?"
#
# Because y = x makes both names refer to the same list,
# both checks return True in this example.
# ============================================================
```

```python
# ============================================================
# 4. ASSIGNING ONE VARIABLE TO ANOTHER
# ============================================================
#
# When assigning one list variable to another:
#
#     y = x
#
# Python makes 'y' refer to the SAME list object as 'x'.
# It does NOT create a copy.
#
# Therefore:
#
#     x ─────┐
#            ↓
#        [1, 2, 3]
#            ↑
#     y ─────┘
# ============================================================


x = [1, 2, 3]

# 'x' refers to a list object.
#
#     x → [1, 2, 3]


y = x

# 'y' now refers to the same list object.
#
#     x ─────┐
#            ↓
#        [1, 2, 3]
#            ↑
#     y ─────┘
#
# No second list was created.


print(x == y)
# True
#
# '==' compares the values/contents.
#
#     [1, 2, 3] == [1, 2, 3]
#              ↓
#             True


print(x is y)
# True
#
# 'is' checks object identity.
#
#     x and y
#        ↓
#     same object
#        ↓
#       True
#
#
# ============================================================
# QUICK MEMORY CHECK
# ============================================================
#
#     y = x
#       ↓
#     same list object
#
#     x == y  → True   (same contents)
#     x is y  → True   (same object)
#
# IMPORTANT:
#
#     y = x
#
# does not mean "make a copy of x".
# It means "make y refer to the object x refers to".
# ============================================================
```

```python
# ============================================================
# 5. is NOT
# ============================================================
#
# 'is not' checks whether two variables refer to DIFFERENT
# objects.
#
#     is not → different object?
#
# This is the logical opposite of 'is'.
# ============================================================


x = [1, 2, 3]
y = [1, 2, 3]

# Two separate list objects are created.
#
# Even though they contain the same values:
#
#     x → [1, 2, 3]
#     y → [1, 2, 3]
#
# they are different objects.
#
#     x → [1, 2, 3]
#
#     y → [1, 2, 3]
#
# Therefore:
#
#     x is not y → True

print(x is not y)
# True
#
# The lists have equal CONTENTS, but they are different OBJECTS.


x = [1, 2, 3]
y = x

# Here, 'y = x' makes y refer to the same list object as x.
#
#     x ─────┐
#            ↓
#        [1, 2, 3]
#            ↑
#     y ─────┘
#
# Therefore:
#
#     x is not y → False
#
# They are NOT different objects.

print(x is not y)
# False


# ============================================================
# QUICK MEMORY CHECK
# ============================================================
#
#     x = [1, 2, 3]
#     y = [1, 2, 3]
#
#     x is not y → True
#     Different list objects
#
#
#     x = [1, 2, 3]
#     y = x
#
#     x is not y → False
#     Same list object
#
# Remember:
#
#     ==      → same value/content?
#     is      → same object?
#     is not  → different object?
#
# Key distinction:
#
#     SAME CONTENT ≠ SAME OBJECT
# ============================================================
```

```python
# ============================================================
# 6. is NOT USED FOR NORMAL VALUE COMPARISON
# ============================================================
#
# Use the operator based on what you want to compare:
#
#     ==  → compares VALUE equality
#     is  → compares OBJECT identity
#
# For normal value comparisons, use ==.
# ============================================================


name = "Ali"

# 'name == "Ali"' asks:
#
#     "Does the value stored in name equal 'Ali'?"
#
#     "Ali" == "Ali"
#          ↓
#        True
#          ↓
#     execute the if block
#
if name == "Ali":
    print("The name is Ali.")


# ============================================================
# QUICK MEMORY CHECK
# ============================================================
#
#     name == "Ali"
#          ↓
#     compare values
#          ↓
#        True
#
# Use:
#
#     == → value comparison
#     is → object identity
#
# Do NOT normally use 'is' when you simply want to compare
# whether two values are equal.
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Ask yourself:
#
#     "Do I care about the VALUE?"
#         → use ==
#
#     "Do I care whether they are the SAME OBJECT?"
#         → use is
# ============================================================
```

```python
# ============================================================
# 7. CHECKING FOR None
# ============================================================
#
# None represents the absence of a value (no value / no object).
#
# When checking specifically for None, use:
#
#     is None      → the value is None
#     is not None  → the value is not None
#
# Use `is` for None checks because None is a singleton object.
# This is the standard and recommended Python pattern.
# ============================================================


email = None

# `email` refers to the None object.
# Therefore, `email is None` → True.
print(email is None)
# True


# `email` is None, so it is NOT a non-None value.
# Therefore, `email is not None` → False.
print(email is not None)
# False


# Quick mental model:
#
#     email
#       ↓
#     None
#
#     is None      → True
#     is not None  → False
#
# CORE IDEA:
# Use `is None` / `is not None` when checking for the
# absence or presence of a value.
```

## MULTIPLE ELIF CONDITIONS

```python
# ============================================================
# 12. MULTIPLE ELIF CONDITIONS
# ============================================================
#
# `if` / `elif` / `else` lets Python choose ONE branch
# from several possible conditions.
#
# Python checks conditions from TOP to BOTTOM.
# The first condition that is True is executed,
# and the remaining conditions are skipped.
#
# For grading, this creates ranges:
#
#     90+ → A
#     80+ → B
#     70+ → C
#     60+ → D
#     <60 → F
# ============================================================

score = 87

# Check the highest range first.
if score >= 90:
    grade = "A"

# Reached only if the previous condition was False.
# 87 >= 80 → True, so grade becomes "B".
elif score >= 80:
    grade = "B"

# These are skipped because a previous condition was True.
elif score >= 70:
    grade = "C"

elif score >= 60:
    grade = "D"

# Runs only if ALL previous conditions are False.
else:
    grade = "F"

print(grade)
# Output:
# B


# Execution flow:
#
#     score = 87
#         ↓
#     >= 90? → False
#         ↓
#     >= 80? → True
#         ↓
#     grade = "B"
#         ↓
#     skip remaining branches
#
#
# QUICK MEMORY:
#
#     if       → first condition
#     elif     → another condition
#     elif     → another condition
#     else     → nothing above was True
#
# CORE IDEA:
# `elif` allows multiple conditions to be checked in order.
# Only the FIRST True branch is executed.
```

```python
# ============================================================
# 13. CHECKING RANGES
# ============================================================
#
# Python allows chained comparisons to check whether a value
# falls within a range.
#
#     18 <= age <= 65
#
# means:
#
#     18 <= age AND age <= 65
#
# Both conditions must be True.
# ============================================================

age = 25

# 18 <= 25 → True
# 25 <= 65 → True
# Therefore, the whole condition is True.
if 18 <= age <= 65:
    print("Working-age adult.")


# Output:
# Working-age adult.
#
#
# Quick mental model:
#
#     18 ≤ age ≤ 65
#      ↑         ↑
#    minimum   maximum
#
#     age = 25
#        ↓
#     18 ≤ 25 ≤ 65 → True
#
# CORE IDEA:
# Chained comparisons provide a clean way to check
# whether a value is inside a range.
```

```python
# ============================================================
# 14. RANGE CONDITIONS
# ============================================================
#
# Multiple `elif` conditions can divide a value into ranges.
#
# Python checks the conditions from TOP to BOTTOM.
# The FIRST True condition is executed.
#
#     < 0       → Freezing
#     0–14      → Cold
#     15–24     → Moderate
#     25–34     → Warm
#     35+       → Hot
# ============================================================

temperature = 25

if temperature < 0:
    print("Freezing")

# 0 <= temperature < 15
# means: temperature is at least 0 AND less than 15.
elif 0 <= temperature < 15:
    print("Cold")

elif 15 <= temperature < 25:
    print("Moderate")

# 25 <= 25 < 35 → True
elif 25 <= temperature < 35:
    print("Warm")

else:
    # Runs only if none of the previous ranges matched.
    print("Hot")


# Output:
# Warm
#
#
# Execution flow:
#
#     temperature = 25
#           ↓
#     < 0?       → False
#           ↓
#     0 ≤ temp < 15? → False
#           ↓
#     15 ≤ temp < 25? → False
#           ↓
#     25 ≤ temp < 35? → True
#           ↓
#         "Warm"
#
# CORE IDEA:
# Chained comparisons make it easy to divide values into
# clear, non-overlapping ranges.
```

```python
# ============================================================
# 20. TRUTHY AND FALSY VALUES
# ============================================================
#
# In a condition, Python treats some values as False ("falsy")
# and most other values as True ("truthy").
#
# Common FALSY values:
#
#     False    → Boolean false
#     None     → no value
#     0        → zero
#     0.0      → zero
#     ""       → empty string
#     []       → empty list
#     {}       → empty dictionary
#     ()       → empty tuple
#     set()    → empty set
#
# Common TRUTHY values:
#
#     True
#     1
#     1.0
#     "Hello"
#     [1, 2, 3]
#     {"key": "value"}
#     (1, 2, 3)
#     {1, 2, 3}
#
# A useful beginner rule:
#
#     Empty / zero / None → usually FALSY
#     Non-empty / non-zero → usually TRUTHY
#
# ============================================================

value = 0

# Python evaluates `value` as a Boolean condition.
# 0 is falsy → the condition is False.
if value:
    print("Truthy")
else:
    print("Falsy")


# Output:
# Falsy
#
#
# Quick mental model:
#
#     value = 0
#          ↓
#      Boolean test
#          ↓
#        False
#          ↓
#       "Falsy"
#
# CORE IDEA:
# Python can use values directly in conditions.
# You do not always need to write `value != 0` or
# `value is not None`; Python automatically evaluates
# the value as truthy or falsy.

# Quick reference:
#
# | Value        | Truthy/Falsy | Why                  |
# |--------------|--------------|----------------------|
# | False        | Falsy        | Boolean false        |
# | None         | Falsy        | No value             |
# | 0            | Falsy        | Zero                 |
# | 0.0          | Falsy        | Zero                 |
# | ""           | Falsy        | Empty string         |
# | []           | Falsy        | Empty list           |
# | {}           | Falsy        | Empty dictionary     |
# | ()           | Falsy        | Empty tuple          |
# | set()        | Falsy        | Empty set            |
# | True         | Truthy       | Boolean true         |
# | 1            | Truthy       | Non-zero             |
# | "Hello"      | Truthy       | Non-empty string     |
# | [1, 2, 3]    | Truthy       | Non-empty list       |
# | {"key":"value"} | Truthy    | Non-empty dictionary |
#
# Easy rule:
# Empty / zero / None → Falsy
# Non-empty / non-zero → Truthy
```

```python
# ============================================================
# 26. CHECKING POSITIVE / NEGATIVE / ZERO
# ============================================================
#
# `if` / `elif` / `else` can classify a number into
# three possible categories:
#
#     number > 0  → Positive
#     number < 0  → Negative
#     otherwise   → Zero
#
# Python checks the conditions from top to bottom and
# executes only the FIRST matching branch.
# ============================================================

number = -5

if number > 0:
    print("Positive")

elif number < 0:
    # -5 < 0 → True
    print("Negative")

else:
    # Runs only when the number is neither positive nor negative.
    # Therefore, it must be zero.
    print("Zero")


# Output:
# Negative
#
#
# Quick mental model:
#
#     number = -5
#          ↓
#     > 0? → False
#          ↓
#     < 0? → True
#          ↓
#      "Negative"
#
# CORE IDEA:
# Use `if` / `elif` / `else` when a value can belong
# to one of several mutually exclusive categories.
```

```python
# ============================================================
# 27. EVEN OR ODD
# ============================================================
#
# `%` is the modulo operator. It gives the REMAINDER after
# division.
#
# For any integer:
#
#     number % 2 == 0 → Even
#     otherwise       → Odd
#
# Examples:
#
#     10 % 2 → 0  → Even
#     17 % 2 → 1  → Odd
# ============================================================

number = 17

# 17 % 2 → 1
# 1 == 0 → False, so the number is not even.
if number % 2 == 0:
    print("Even")

else:
    # If the remainder is not 0, the integer is odd.
    print("Odd")


# Output:
# Odd
#
#
# Quick mental model:
#
#     17 ÷ 2
#        ↓
#     remainder = 1
#        ↓
#     1 == 0? → False
#        ↓
#       Odd
#
# CORE IDEA:
# An integer is even when dividing it by 2 leaves
# a remainder of 0; otherwise, it is odd.
```

```python
# ============================================================
# 28. DIVISIBILITY
# ============================================================
#
# A number is divisible by another number when the division
# leaves a remainder of 0.
#
# Use `%` (modulo) to check the remainder:
#
#     number % 5 == 0 → divisible by 5
#     number % 5 != 0 → not divisible by 5
# ============================================================

number = 30

# 30 % 5 → 0
# The remainder is 0, so 30 is divisible by 5.
if number % 5 == 0:
    print("Divisible by 5.")

else:
    # Runs when the remainder is not 0.
    print("Not divisible by 5.")


# Output:
# Divisible by 5.
#
#
# Quick mental model:
#
#     30 ÷ 5
#        ↓
#     remainder = 0
#        ↓
#     30 % 5 == 0 → True
#        ↓
#     "Divisible by 5."
#
# CORE IDEA:
# To check divisibility, use `%` and test whether
# the remainder is 0.
```

```python
# ============================================================
# 29. MULTIPLE DIVISIBILITY CONDITIONS
# ============================================================
#
# `%` checks the remainder after division.
#
#     number % 3 == 0 → divisible by 3
#     number % 5 == 0 → divisible by 5
#
# `and` requires BOTH conditions to be True.
#
# The `if` / `elif` / `else` structure checks from top
# to bottom and executes only the FIRST matching branch.
# ============================================================

number = 30

# 30 % 3 == 0 → True
# 30 % 5 == 0 → True
# Both are True → the `and` condition is True.
if number % 3 == 0 and number % 5 == 0:
    print("Divisible by both 3 and 5.")

elif number % 3 == 0:
    print("Divisible by 3.")

elif number % 5 == 0:
    print("Divisible by 5.")

else:
    print("Not divisible by 3 or 5.")


# Output:
# Divisible by both 3 and 5.
#
#
# Quick mental model:
#
#     30 % 3 == 0 → True
#            AND
#     30 % 5 == 0 → True
#            ↓
#          True
#            ↓
#     "Divisible by both 3 and 5."
#
#
# Why check "both" first?
#
# A number divisible by both 3 and 5 also satisfies
# the individual conditions. Checking the combined case
# first ensures the more specific result is selected.
#
# CORE IDEA:
# Combine `%`, `and`, and `if`/`elif` to classify a number
# based on multiple divisibility conditions.
```

```python
# ============================================================
# ONE-LINE IF / ELSE (TERNARY - CONDITIONAL EXPRESSION)
# ============================================================
#
# Python has a shorter form for a simple if/else decision.
# It is called a CONDITIONAL EXPRESSION.
#
# It is often called a TERNARY expression because it has
# three parts:
#
#     1. value_if_true
#     2. condition
#     3. value_if_false
#
#
# The special syntax is:
#
#     value_if_true if condition else value_if_false
#
#
# IMPORTANT:
#
# This is different from writing a normal `if` statement
# on one line.
#
#     if condition:
#         ...
#
# The conditional expression PRODUCES A VALUE.
# That value can then be assigned to a variable.
#
# ============================================================
# HOW TO READ THE NEW FORM
# ============================================================
#
#     result = "Passed" if score >= 50 else "Failed"
#                 ↑              ↑             ↑
#            True value      condition     False value
#
# Read it from the CONDITION:
#
#     "If score >= 50 is True, use 'Passed';
#      otherwise, use 'Failed'."
#
#
# Another way to visualize it:
#
#                 score >= 50?
#                  /       \
#               True       False
#                ↓           ↓
#           "Passed"      "Failed"
#                \           /
#                 \         /
#                    result
#
# ============================================================
# 1. NORMAL INDENTED IF / ELSE
# ============================================================

score = 75

if score >= 50:
    result = "Passed"
else:
    result = "Failed"

print(result)

# Output:
# Passed
#
# Here, the `if` statement controls which assignment
# is executed.


# ============================================================
# 2. CONDITIONAL EXPRESSION
# ============================================================

score = 75

# The conditional expression chooses ONE of two values.
#
#     "Passed" → used when score >= 50 is True
#     "Failed" → used when score >= 50 is False
#
# The selected value is then assigned to `result`.
result = "Passed" if score >= 50 else "Failed"

print(result)

# Output:
# Passed


# ============================================================
# STEP-BY-STEP EXECUTION
# ============================================================
#
# Python evaluates:
#
#     "Passed" if score >= 50 else "Failed"
#
# Step 1:
#     Check the condition:
#
#     score >= 50
#     75 >= 50 → True
#
# Step 2:
#     Because the condition is True, choose:
#
#     "Passed"
#
# Step 3:
#     Assign that value to `result`:
#
#     result = "Passed"
#
#
# If score were 40:
#
#     40 >= 50 → False
#
# So Python would choose:
#
#     "Failed"
#
# and:
#
#     result = "Failed"
#
#
# ============================================================
# NORMAL IF/ELSE VS CONDITIONAL EXPRESSION
# ============================================================
#
# NORMAL:
#
#     if score >= 50:
#         result = "Passed"
#     else:
#         result = "Failed"
#
#
# CONDITIONAL EXPRESSION:
#
#     result = "Passed" if score >= 50 else "Failed"
#
#
# They produce the same result.
#
# The important difference is that the conditional expression
# directly PRODUCES A VALUE, which makes it useful for
# assignments and other expressions.
#
#
# ============================================================
# WHEN TO USE IT
# ============================================================
#
# Good for SIMPLE choices:
#
#     status = "Adult" if age >= 18 else "Minor"
#
#     result = "Even" if number % 2 == 0 else "Odd"
#
#     message = "Yes" if answer == "y" else "No"
#
# For complicated logic, a normal if/else is usually
# easier to read.
#
#
# QUICK MEMORY:
#
#     NORMAL:
#     if condition:
#         ...
#     else:
#         ...
#
#     CONDITIONAL EXPRESSION:
#     A if condition else B
#
#     Read as:
#     "Use A if condition is True, otherwise use B."
#
#
# CORE IDEA:
# A conditional expression is a compact way to choose
# BETWEEN TWO VALUES based on a condition.
#
#     A if condition else B
#       ↓          ↓
#     True       False
#       ↓          ↓
#       A          B
#
# ============================================================
```

```python
# ============================================================
# 3. COMPARE THEM
# ============================================================
#
# Both forms make the SAME decision and produce the SAME result.
#
# Normal if/else:
#
#     if score >= 50:
#         result = "Passed"
#     else:
#         result = "Failed"
#
#
# Conditional expression:
#
#     result = "Passed" if score >= 50 else "Failed"
#
#
# The difference is mainly STYLE:
#
#     Normal if/else       → better for multiple/complex actions
#     Conditional expression → better for one simple value choice
#
# ============================================================
# 4. ANOTHER EXAMPLE
# ============================================================

age = 20

# Normal if/else:
# If age is 18 or higher, assign "Adult".
# Otherwise, assign "Minor".
if age >= 18:
    status = "Adult"
else:
    status = "Minor"

print(status)
# Output:
# Adult


# One-line conditional expression:
#
#     "Adult" → value when condition is True
#     age >= 18 → condition
#     "Minor" → value when condition is False
#
status = "Adult" if age >= 18 else "Minor"

print(status)
# Output:
# Adult


# ============================================================
# QUICK COMPARISON
# ============================================================
#
# Normal:
#
#     if condition:
#         value = A
#     else:
#         value = B
#
#
# One-line:
#
#     value = A if condition else B
#
#
# Both:
#
#     condition True  → A
#     condition False → B
#
#
# CORE IDEA:
# The conditional expression is simply a shorter way to
# choose between TWO values.
#
#     A if condition else B
#     ↑       ↑        ↑
#   True   condition  False
#   value             value
#
# ============================================================
```

```python
# ============================================================
# 5. ONE-LINE IF WITHOUT ELSE
# ============================================================
#
# There is another short form for a simple `if` statement
# when you only need to do something when the condition is True.
#
# Syntax:
#
#     if condition: code
#
# This is called a ONE-LINE IF statement.
#
# IMPORTANT:
# This is NOT a ternary/conditional expression.
#
#     One-line if:
#         if condition: action
#
#     Ternary expression:
#         value_if_true if condition else value_if_false
#
# The one-line if does NOT have an `else`.
# ============================================================

score = 90

# 90 >= 50 → True
# Therefore, Python executes the print statement.
if score >= 50:
    print("Passed")


# Output:
# Passed
#
#
# ------------------------------------------------------------
# Same logic using the normal form:
# ------------------------------------------------------------
#
# if score >= 50:
#     print("Passed")
#
# Both forms do exactly the same thing.
#
#
# Quick mental model:
#
#     score = 90
#          ↓
#     score >= 50?
#       ↙       ↘
#    True      False
#      ↓          ↓
#    print     do nothing
#
#
# IMPORTANT:
# Use this short form only for very simple actions.
# For longer or more complicated code, the normal indented
# form is clearer and easier to read.
#
#
# CORE IDEA:
# A one-line `if` is a compact way to execute ONE simple
# statement only when a condition is True.
#
#     if condition: action
#
# True  → action runs
# False → nothing happens
#
# ============================================================
```

```python
# ============================================================
# 6. WHEN TO USE EACH ONE
# ============================================================
#
# Use NORMAL INDENTED `if` / `else` when:
#
#     - The logic is more complicated
#     - You have multiple statements
#     - Readability is important
#
#
# Use a ONE-LINE CONDITIONAL EXPRESSION when:
#
#     - The decision is very simple
#     - You need to choose between TWO values
#
#
# Example:
#
#     result = "Pass" if score >= 50 else "Fail"
#
#     "Pass" → if condition is True
#     "Fail" → if condition is False
#
#
# ============================================================
# QUICK SUMMARY
# ============================================================
#
# NORMAL IF / ELSE:
#
#     if condition:
#         value = A
#     else:
#         value = B
#
#
# ONE-LINE CONDITIONAL EXPRESSION:
#
#     value = A if condition else B
#
#
# ONE-LINE IF WITHOUT ELSE:
#
#     if condition: action
#
#
# Remember the difference:
#
#     if/else statement
#         → controls which code runs
#
#     conditional expression
#         → produces one of two values
#
#
# Example:
#
#     result = "Pass" if score >= 50 else "Fail"
#
#     condition True  → result = "Pass"
#     condition False → result = "Fail"
#
#
# CORE IDEA:
# Use the shortest form that keeps the code clear.
# Simple decisions can be written compactly; complicated
# decisions are usually clearer with normal indentation.
#
# ============================================================
```

## Independent IFs (Sepatare ifs)

```python
# ============================================================
# INDEPENDENT ifs VS if / elif / else
# ============================================================
#
# These two structures are useful for DIFFERENT situations.
#
# Independent `if`s:
#     → Every condition is checked.
#     → Several actions can happen.
#
# `if / elif / else`:
#     → Conditions are checked from top to bottom.
#     → The FIRST True branch runs.
#     → The remaining branches are skipped.
#
#
# ============================================================
# 1. INDEPENDENT ifs — MULTIPLE ACTIONS CAN HAPPEN
# ============================================================
#
# Example: checking properties of a number.
#
# A number can be:
#
#     - positive
#     - even
#     - greater than 100
#
# These properties are NOT mutually exclusive.
# One number can satisfy several conditions at the same time.
# ============================================================

number = 120

if number > 0:
    print("The number is positive.")

if number % 2 == 0:
    print("The number is even.")

if number > 100:
    print("The number is greater than 100.")


# Output:
# The number is positive.
# The number is even.
# The number is greater than 100.
#
#
# Execution:
#
#     number = 120
#          │
#          ├── > 0? ────── True  → Run
#          │
#          ├── % 2 == 0? ─ True  → Run
#          │
#          └── > 100? ──── True  → Run
#                             
#                         ALL run
#
# Notice:
# Python does NOT stop after the first True condition.
# It continues checking every independent `if`.
#
#
# ============================================================
# 2. if / elif / else — ONLY ONE OUTCOME
# ============================================================
#
# Example: classifying a person's age group.
#
# A person should belong to ONE age category:
#
#     under 13  → Child
#     13–17     → Teenager
#     18–64     → Adult
#     65+       → Senior
#
# These categories are mutually exclusive.
# A person should receive only ONE classification.
# ============================================================

age = 25

if age < 13:
    print("Child")

elif age < 18:
    print("Teenager")

elif age < 65:
    print("Adult")

else:
    print("Senior")


# Output:
# Adult
#
#
# Execution:
#
#     age = 25
#          │
#          ▼
#      age < 13?
#       False
#          │
#          ▼
#      age < 18?
#       False
#          │
#          ▼
#      age < 65?
#       True
#          │
#          ▼
#       "Adult"
#          │
#          ▼
#      STOP CHECKING
#
# The `else` and remaining branches are skipped.
#
#
# ============================================================
# 3. WHY THE TWO EXAMPLES ARE DIFFERENT
# ============================================================
#
# NUMBER PROPERTIES:
#
#     120 can be:
#
#     Positive  ✓
#     Even      ✓
#     > 100     ✓
#
#     Therefore, independent `if`s are appropriate.
#
#
# AGE CATEGORY:
#
#     25 cannot be both:
#
#     Child     ✗
#     Teenager  ✗
#     Adult     ✓
#     Senior    ✗
#
#     Therefore, `if / elif / else` is appropriate.
#
#
# ============================================================
# VISUAL COMPARISON
# ============================================================
#
# INDEPENDENT ifs:
#
#     Start
#       │
#       ├── Check A → Run? ──┐
#       │                    │
#       ├── Check B → Run? ──┤
#       │                    │
#       └── Check C → Run? ──┘
#                            │
#                            ▼
#                           End
#
#     Every condition is checked.
#
#
# if / elif / else:
#
#     Start
#       │
#       ▼
#     Check A?
#      /    \
#   True   False
#    │       │
#   Run A    ▼
#          Check B?
#           /    \
#        True   False
#         │       │
#        Run B    ▼
#               Check C?
#                /    \
#             True   False
#              │       │
#             Run C   Else
#                      │
#                      ▼
#                     End
#
#     FIRST TRUE → run its branch → stop.
#
#
# ============================================================
# QUICK MEMORY
# ============================================================
#
#     Independent `if`s:
#         CHECK ALL
#         ↓
#         Multiple can run
#
#
#     `if / elif / else`:
#         FIRST TRUE WINS
#         ↓
#         Only one branch runs
#
#
# ============================================================
# WHEN SHOULD YOU USE EACH?
# ============================================================
#
# Use independent `if`s when:
#
#     "Can several conditions be true at the same time?"
#
#     YES → independent `if`s may be appropriate.
#
#
# Use `if / elif / else` when:
#
#     "Do I want exactly one result?"
#
#     YES → use `if / elif / else`.
#
#
# CORE IDEA:
#
#     Independent `if`:
#         Several conditions → several possible actions
#
#     if / elif / else:
#         Several conditions → ONE selected action
#
# ============================================================
```
