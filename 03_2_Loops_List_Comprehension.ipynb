expression	"This is what I do to each thing."

```python


# # 2. TEXT NORMALISATION (Categorical Variables)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================

# Text data is rarely consistent.

# Example:

#     Male
#     MALE
#     male

# Machine learning models treat these as different values.

# Before encoding categorical variables, we usually normalize
# text using lower().

# The same idea applies to:

# • Product names
# • Country names
# • User input
# • Categories

# This preprocessing step is extremely common in real projects.
# ============================================================
gender = ["Male", "MALE", "male"]
normalized = [g.lower() for g in gender]
print(normalized)

# Also remove extra whitespace
messy_names = ["  Alice ", "Bob\t", "Charlie\n", "  David  "]
clean_names = [name.strip().lower() for name in messy_names]
print(clean_names)
```

```python

## 3. HANDLING MISSING VALUES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Real datasets contain missing values (None, NaN, empty strings).
#
# Example:
#
#     age = [25, None, 30, None, 22]
#
# Many models cannot handle None/NaN directly.
#
# You often need to:
#
# • Replace missing values with a default (e.g., 0, mean, median)
# • Or remove rows with missing data
#
# List comprehensions give a quick way to do either.
# ============================================================


age = [25, None, 30, None, 22]

# Replace None with 0
filled_age = [a if a is not None else 0 for a in age]
print(filled_age)

# Remove missing values entirely
clean_age = [a for a in age if a is not None]
print(clean_age)
```

```python

## 4. EXTRACTING FEATURES FROM DATES OR STRINGS

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Dates, phone numbers, emails, etc. often come as strings.
#
# Example:
#
#     dates = ["2023-01-15", "2024-06-20", "2025-12-01"]
#
# You may want to extract the year, month, or day as a new feature.
#
# This is a common feature engineering step in:
#
# • Time series forecasting
# • Customer segmentation
# • Churn prediction
#
# List comprehensions let you split and parse in one line.
# ============================================================

dates = ["2023-01-15", "2024-06-20", "2025-12-01"]

# Extract year
years = [d.split('-')[0] for d in dates]
print(years)

# Convert to integer
years_int = [int(d.split('-')[0]) for d in dates]
print(years_int)

# Extract month and day as tuple
month_day = [(int(d.split('-')[1]), int(d.split('-')[2])) for d in dates]
print(month_day)
```

```python
## 5. FILTERING OUTLIERS

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Data often contains extreme values that distort models.
#
# Example:
#
#     prices = [120, 150, 130, 10000, 140, 110, 9000, 160]
#
# Outliers like 10000 can skew mean, variance, and regression.
#
# Common strategies:
#
# • Remove values outside a reasonable range
# • Cap (winsorize) extreme values
#
# A list comprehension with a condition does both cleanly.
# ============================================================

prices = [120, 150, 130, 10000, 140, 110, 9000, 160]

# Remove outliers > 500 (assume normal price range)
filtered_prices = [p for p in prices if p <= 500]
print(filtered_prices)

# Cap outliers at 500 (winsorizing)
capped_prices = [p if p <= 500 else 500 for p in prices]
print(capped_prices)
```

```python
## 6. CREATING NEW NUMERICAL FEATURES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Often you need to transform a feature to better fit a model.
#
# Examples:
#
# • Log transform to handle skewed distributions
# • Square root for count data
# • Length of text for NLP tasks
#
# In machine learning, feature engineering is key to performance.
#
# List comprehensions make applying transformations effortless.
# ============================================================

import math

values = [1, 10, 100, 0, 1000]

# Log transform (avoid log(0) by replacing with 0 or a small value)
log_values = [math.log(v) if v > 0 else 0 for v in values]
print([round(x, 2) for x in log_values])

# Length of product descriptions (for NLP)
descriptions = ['short', 'medium length', 'very long description']
desc_lengths = [len(d) for d in descriptions]
print(desc_lengths)
```

```python

## 7. COMBINING MULTIPLE FEATURES (PAIRWISE OPERATIONS)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# You often have two (or more) columns that together define a new feature.
#
# Example:
#
#     heights = [170, 160, 180]   # cm
#     weights = [70, 60, 80]      # kg
#
# Body Mass Index (BMI) = weight / (height in meters)^2
#
# This is a common derived feature in health analytics.
#
# List comprehensions with zip() let you combine multiple lists cleanly.
# ============================================================

heights = [170, 160, 180]  # cm
weights = [70, 60, 80]     # kg

# Compute BMI
bmi = [w / ((h/100)**2) for h, w in zip(heights, weights)]
print([round(b, 2) for b in bmi])

# Another example: price per unit
prices = [100, 200, 150]
quantities = [2, 3, 1]
price_per_unit = [p / q for p, q in zip(prices, quantities)]
print([round(x, 2) for x in price_per_unit])
```

```python

## 8. WORKING WITH PANDAS DATAFRAMES

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# While pandas has vectorised operations, sometimes you need
# row‑wise transformations that are not easily vectorised.
#
# In such cases, apply() can be slow on large DataFrames.
#
# List comprehensions are often faster than apply() because
# they iterate in Python but with less overhead.
#
# Use them to:
#
# • Create new columns based on string operations
# • Extract parts of a string
# • Apply conditional logic
# ============================================================


import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'city': ['New York', 'Los Angeles', 'Chicago']
})

# Convert city to lower case (faster than df['city'].str.lower())
df['city_lower'] = [c.lower() for c in df['city']]

# Extract first letter of name
df['first_letter'] = [n[0] for n in df['name']]

# Flag if age >= 30
df['senior'] = [1 if a >= 30 else 0 for a in df['age']]

print(df)
```

```python

## 9. ONE‑HOT ENCODING (DUMMY VARIABLES)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Most ML models require numerical input.
#
# Categorical variables must be encoded.
#
# One‑hot encoding creates binary columns for each category.
#
# Instead of using pandas.get_dummies(), you can quickly
# generate dummy columns with a list comprehension if you
# need to control the output or integrate with other logic.
#
# This also demonstrates a nested comprehension.
# ============================================================

categories = ['A', 'B', 'A', 'C', 'B', 'A']
unique = sorted(set(categories))

# Create dummy matrix: each row is a list of 0/1
dummies = [[1 if cat == u else 0 for u in unique] for cat in categories]

# Display as a DataFrame for readability
dummy_df = pd.DataFrame(dummies, columns=unique)
print(dummy_df)
```

```python
## 10. PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# When dealing with millions of rows, performance matters.
#
# List comprehensions are implemented in C and are faster
# than explicit for‑loops with append().
#
# Let's measure the difference on a large list.
# ============================================================

import time

n = 10_000_000
data = list(range(n))

# Traditional loop
start = time.time()
squares_loop = []
for x in data:
    squares_loop.append(x**2)
end = time.time()
print(f"Loop time: {end - start:.4f} seconds")

# List comprehension
start = time.time()
squares_comp = [x**2 for x in data]
end = time.time()
print(f"Comprehension time: {end - start:.4f} seconds")
```

```python
# ============================================================
# 62. LIST COMPREHENSION
# ============================================================
#
# A list comprehension is a compact way to create a NEW list
# by processing each item from an iterable.
#
# Basic pattern:
#
#     [expression for item in iterable]
#
# Think:
#
#     INPUT
#       ↓
#     take each item
#       ↓
#     transform it
#       ↓
#     collect the result
#       ↓
#     NEW LIST
#
# The important idea is the DATA TRANSFORMATION pattern:
#
#     one input item → one output item
#
# ============================================================
# CREATE A LIST
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# LIST COMPREHENSION
# ============================================================
#
# This creates a NEW list called squares.
#
# General structure:
#
#     [expression for item in iterable]
#      ↑             ↑         ↑
#      |             |         |
#      |             |         input collection
#      |             loop variable
#      output expression
#
# Here:
#
#     expression = number ** 2
#     item       = number
#     iterable   = numbers
#
# Meaning:
#
#     "For each number in numbers,
#      calculate its square,
#      and collect that result into a new list."
#
# ============================================================
# VALUE FLOW
# ============================================================
#
#     numbers
#        ↓
#     1 → 1 ** 2 → 1
#     2 → 2 ** 2 → 4
#     3 → 3 ** 2 → 9
#     4 → 4 ** 2 → 16
#     5 → 5 ** 2 → 25
#        ↓
#     [1, 4, 9, 16, 25]
#
# ============================================================
# PROCESS TABLE
# ============================================================
#
#     Step | Current number | Operation    | Result | squares
#     -----|----------------|---------------|--------|-----------------------
#       1  |       1        | 1 ** 2        |   1    | [1]
#       2  |       2        | 2 ** 2        |   4    | [1, 4]
#       3  |       3        | 3 ** 2        |   9    | [1, 4, 9]
#       4  |       4        | 4 ** 2        |  16    | [1, 4, 9, 16]
#       5  |       5        | 5 ** 2        |  25    | [1, 4, 9, 16, 25]
#
# The table shows the important process:
#
#     get item
#         ↓
#     apply expression
#         ↓
#     produce result
#         ↓
#     collect result
#
# ============================================================

squares = [
    number ** 2
    for number in numbers
]


# ============================================================
# PROGRAM STATE
# ============================================================
#
# BEFORE:
#
#     numbers = [1, 2, 3, 4, 5]
#
# DURING:
#
#     number receives one item at a time.
#
#     1 → calculate 1² → collect 1
#     2 → calculate 2² → collect 4
#     3 → calculate 3² → collect 9
#     4 → calculate 4² → collect 16
#     5 → calculate 5² → collect 25
#
# AFTER:
#
#     numbers = [1, 2, 3, 4, 5]
#     squares = [1, 4, 9, 16, 25]
#
# Notice:
#
#     numbers does NOT change.
#
#     squares is the NEW list created by the comprehension.
#
# ============================================================
# HOW IT WORKS
# ============================================================
#
#     numbers
#        ↓
#     get next item
#        ↓
#     number receives item
#        ↓
#     evaluate number ** 2
#        ↓
#     result is collected into squares
#        ↓
#     get next item
#        ↓
#     repeat
#        ↓
#     no more items
#        ↓
#     squares is complete
#
# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The comprehension is essentially a compact form of:
#
#     squares = []
#
#     for number in numbers:
#         squares.append(number ** 2)
#
# Conceptually:
#
#     create empty list
#          ↓
#     get item
#          ↓
#     calculate result
#          ↓
#     append result
#          ↓
#     repeat
#
# The list comprehension expresses the same common pattern
# more compactly.
#
# ============================================================
# PRINT RESULT
# ============================================================

print(squares)

# Output:
#
#     [1, 4, 9, 16, 25]
#
# Why?
#
# Each number is squared:
#
#     1² = 1
#     2² = 4
#     3² = 9
#     4² = 16
#     5² = 25
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, transform it and create a new list."
#
# Think:
#
#     [expression for item in iterable]
#
# Examples:
#
#     [number * 2 for number in numbers]
#     [number ** 2 for number in numbers]
#     [word.upper() for word in words]
#
# General pattern:
#
#     INPUT COLLECTION
#          ↓
#     FOR EACH ITEM
#          ↓
#     TRANSFORM
#          ↓
#     COLLECT RESULT
#          ↓
#     NEW LIST
#
# ============================================================
# MENTAL MODEL
# ============================================================
#
# Do not think of a list comprehension as mysterious syntax.
#
# Mentally expand it into:
#
#     "Take each item,
#      do something with it,
#      collect the result."
#
# For this example:
#
#     take number
#         ↓
#     calculate number ** 2
#         ↓
#     collect result
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
#     [TRANSFORMATION for ITEM in COLLECTION]
#
# Examples:
#
#     [x * 10 for x in numbers]
#
#     [x ** 2 for x in numbers]
#
#     [word.upper() for word in words]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A list comprehension combines:
#
#     ITERATION
#          +
#     TRANSFORMATION
#          +
#     LIST CREATION
#
# into one compact expression.
#
# The key mental pattern is:
#
#     "For each input item
#      → calculate an output
#      → collect it into a new list."
#
# ============================================================
```

```python
# ============================================================
# 63. LIST COMPREHENSION WITH CONDITION
# ============================================================
#
# A conditional list comprehension creates a new list by
# iterating through an iterable and keeping only the items
# that satisfy a condition.
#
# GENERAL PATTERN:
#
#     [expression for item in iterable if condition]
#
# Think:
#
#     collection
#          ↓
#     current item
#          ↓
#     condition
#      ↙       ↘
#   False      True
#    ↓           ↓
#   skip       include
#      \         /
#       ↓       ↓
#        new list
#
# This combines:
#
#     ITERATION + FILTERING + LIST CREATION
#
# ============================================================
# CONCEPTUAL BREAKDOWN
# ============================================================
#
# numbers
#     → source collection
#
# number
#     → loop variable representing the current item
#
# number % 2 == 0
#     → condition that tests whether the number is even
#
# even_numbers
#     → new list containing the values that pass the condition
#
# RELATIONSHIP:
#
#     numbers
#         ↓
#     get current number
#         ↓
#     test condition
#         ↓
#     True → include number
#     False → skip number
#         ↓
#     even_numbers
#
# ============================================================
# SYNTAX → MEANING
# ============================================================
#
#     number
#     → value placed into the new list
#
#     for number in numbers
#     → process every item in numbers
#
#     if number % 2 == 0
#     → include the item only when the condition is True
#
# IMPORTANT:
#
# The if condition FILTERS the items.
#
# It does not modify number.
#
# ============================================================
# HOW THE CONDITION WORKS
# ============================================================
#
#     number % 2
#
# gives the remainder after division by 2.
#
#     1 % 2 → 1
#     2 % 2 → 0
#     3 % 2 → 1
#     4 % 2 → 0
#     5 % 2 → 1
#     6 % 2 → 0
#
# Therefore:
#
#     number % 2 == 0
#
#     even number → True
#     odd number  → False
#
# ============================================================
# VALUE FLOW
# ============================================================
#
#     1 → 1 % 2 → 1 == 0 → False → skip
#     2 → 2 % 2 → 0 == 0 → True  → keep
#     3 → 3 % 2 → 1 == 0 → False → skip
#     4 → 4 % 2 → 0 == 0 → True  → keep
#     5 → 5 % 2 → 1 == 0 → False → skip
#     6 → 6 % 2 → 0 == 0 → True  → keep
#
# Final:
#
#     [2, 4, 6]
#
# ============================================================
# VARIABLE / VALUE / TYPE TRACKING
# ============================================================
#
# numbers
#     → [1, 2, 3, 4, 5, 6]
#     → list
#     → source collection
#
# number
#     → 1, then 2, then 3, ..., then 6
#     → int
#     → current item
#
# even_numbers
#     → list
#     → contains accepted values
#
# IMPORTANT:
#
# numbers stays unchanged.
#
# number changes on each iteration.
#
# even_numbers grows as qualifying values are included.
#
# ============================================================
# HOW PYTHON EXECUTES THE CODE
# ============================================================
#
# Python conceptually does:
#
#     create new list
#          ↓
#     get next item
#          ↓
#     assign it to number
#          ↓
#     evaluate the condition
#          ↓
#     True → include number
#     False → skip it
#          ↓
#     get next item
#          ↓
#     repeat
#          ↓
#     finish the list
#
# ============================================================
# PROCESS TABLE
# ============================================================
#
# Step | number | % 2 | Condition | Action | even_numbers
# -----|--------|-----|-----------|--------|--------------
#  1   |   1    |  1  | False     | Skip   | []
#  2   |   2    |  0  | True      | Keep   | [2]
#  3   |   3    |  1  | False     | Skip   | [2]
#  4   |   4    |  0  | True      | Keep   | [2, 4]
#  5   |   5    |  1  | False     | Skip   | [2, 4]
#  6   |   6    |  0  | True      | Keep   | [2, 4, 6]
#
# ============================================================
# EXECUTION TRACE
# ============================================================
#
# Start:
#
#     even_numbers = []
#
# Iteration 1:
#
#     number = 1
#     1 % 2 == 0 → False
#     → skip
#
# Iteration 2:
#
#     number = 2
#     2 % 2 == 0 → True
#     → include 2
#
#     even_numbers = [2]
#
# Iteration 3:
#
#     number = 3
#     3 % 2 == 0 → False
#     → skip
#
# Iteration 4:
#
#     number = 4
#     4 % 2 == 0 → True
#     → include 4
#
#     even_numbers = [2, 4]
#
# Iteration 5:
#
#     number = 5
#     5 % 2 == 0 → False
#     → skip
#
# Iteration 6:
#
#     number = 6
#     6 % 2 == 0 → True
#     → include 6
#
#     even_numbers = [2, 4, 6]
#
# No more items → comprehension ends.
#
# ============================================================
# EQUIVALENT BEGINNER-FRIENDLY VERSION
# ============================================================
#
# The same basic logic can be written explicitly:
#
#     even_numbers = []
#
#     for number in numbers:
#
#         if number % 2 == 0:
#             even_numbers.append(number)
#
# The loop version makes the individual steps visible.
#
# The comprehension combines those steps into one compact
# list-creation expression.
#
# ============================================================
# DATA TRANSFORMATION
# ============================================================
#
# INPUT:
#
#     [1, 2, 3, 4, 5, 6]
#
#         ↓
#
# RULE:
#
#     Keep values where number % 2 == 0
#
#         ↓
#
#     1 → reject
#     2 → keep
#     3 → reject
#     4 → keep
#     5 → reject
#     6 → keep
#
#         ↓
#
# OUTPUT:
#
#     [2, 4, 6]
#
# This is primarily FILTERING, not value transformation.
#
# ============================================================
# INVARIANT / ALWAYS-TRUE RELATIONSHIP
# ============================================================
#
# After each iteration:
#
#     even_numbers contains exactly the even values
#     encountered so far.
#
# This explains why the final result is correct.
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# The condition:
#
#     number % 2 == 0
#
# does NOT change number.
#
# It only decides whether number is included.
#
# Example:
#
#     number = 3
#     3 % 2 == 0 → False
#     → 3 is skipped
#
#     number = 4
#     4 % 2 == 0 → True
#     → 4 is included as 4
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
#     [item for item in collection if condition]
#
# means:
#
#     "For every item in the collection,
#      keep the item if the condition is True."
#
# Examples:
#
#     [x for x in numbers if x > 10]
#     → keep values greater than 10
#
#     [word for word in words if len(word) > 5]
#     → keep words longer than 5 characters
#
#     [sale for sale in sales if sale >= 1000]
#     → keep sales of at least 1000
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A conditional list comprehension combines:
#
#     ITERATION
#         +
#     CONDITION
#         +
#     LIST CREATION
#
# The key mental model is:
#
#     collection
#         ↓
#     current item
#         ↓
#     test condition
#       ↙       ↘
#    False      True
#      ↓          ↓
#    skip       include
#         \      /
#          ↓    ↓
#          new list
#
# For this example:
#
#     [number for number in numbers if number % 2 == 0]
#
# means:
#
#     "Take each number, keep it if it is even,
#      and build a new list from those values."
#
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]  # Original list of numbers

# Create a new list containing only even numbers.
# For each number in numbers, Python evaluates the condition.
# If the condition is True, number is included in even_numbers.
even_numbers = [
    number
    for number in numbers
    if number % 2 == 0
]

print(even_numbers)  # Output: [2, 4, 6]
```

```python
# ============================================================
# 64. LIST COMPREHENSION — DATA TRANSFORMATION
# ============================================================
#
# A list comprehension can do more than filter data.
# It can TRANSFORM each item and create a new list.
#
# General pattern:
#
#     [expression for item in iterable]
#
# Meaning:
#
#     Take each item
#         ↓
#     apply an expression/transformation
#         ↓
#     store the result in a new list
#
# Here, the transformation is:
#
#     price * 1.20
#
# This means:
#
#     original price + 20% tax
#
# Mathematical view:
#
#     price_with_tax = price × 1.20
#
# Python view:
#
#     price_with_tax = price * 1.20
#
# IMPORTANT:
#     x² is mathematical notation.
#     x**2 is Python syntax.
#
# ------------------------------------------------------------
# CONCEPTUAL BREAKDOWN
# ------------------------------------------------------------
#
# Original data:
#
#     prices = [100, 200, 300, 400]
#
# Transformation:
#
#     price × 1.20
#
# Result:
#
#     [120.0, 240.0, 360.0, 480.0]
#
# The original list is NOT modified.
# A completely new list is created.
#
# ------------------------------------------------------------
# COMPONENT ROLES
# ------------------------------------------------------------
#
# prices
#     → original iterable containing the input values
#
# price
#     → temporary loop variable
#     → represents one price at a time
#
# price * 1.20
#     → transformation expression
#     → calculates the tax-included price
#
# prices_with_tax
#     → new list containing all transformed results
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     prices
#       ↓
#     100 → 100 × 1.20 → 120.0
#     200 → 200 × 1.20 → 240.0
#     300 → 300 × 1.20 → 360.0
#     400 → 400 × 1.20 → 480.0
#       ↓
#     prices_with_tax
#
# ------------------------------------------------------------
# PROCESS TABLE
# ------------------------------------------------------------
#
# | Step | price | Operation       | Result | New List              |
# |------|-------|-----------------|--------|------------------------|
# | 1    | 100   | 100 × 1.20      | 120.0  | [120.0]                |
# | 2    | 200   | 200 × 1.20      | 240.0  | [120.0, 240.0]        |
# | 3    | 300   | 300 × 1.20      | 360.0  | [120.0, 240.0, 360.0] |
# | 4    | 400   | 400 × 1.20      | 480.0  | [120.0, 240.0, 360.0, 480.0] |
#
# ------------------------------------------------------------
# EXECUTION TRACE
# ------------------------------------------------------------
#
# Python effectively performs:
#
#     price = 100
#     100 * 1.20 → 120.0
#
#     price = 200
#     200 * 1.20 → 240.0
#
#     price = 300
#     300 * 1.20 → 360.0
#
#     price = 400
#     400 * 1.20 → 480.0
#
# Then the new list becomes:
#
#     [120.0, 240.0, 360.0, 480.0]
#
# ------------------------------------------------------------
# VARIABLE / VALUE / TYPE TRACKING
# ------------------------------------------------------------
#
# prices
#     → value: [100, 200, 300, 400]
#     → type: list
#     → role: input data
#
# price
#     → values during iteration: 100 → 200 → 300 → 400
#     → type: int
#     → role: current item
#
# prices_with_tax
#     → value: [120.0, 240.0, 360.0, 480.0]
#     → type: list
#     → role: transformed output
#
# ------------------------------------------------------------
# WHY 1.20?
# ------------------------------------------------------------
#
# A 20% tax means:
#
#     20% = 20 / 100 = 0.20
#
# Original price:
#
#     100% = 1.00
#
# Price + tax:
#
#     100% + 20% = 120%
#
#     120% = 1.20
#
# Therefore:
#
#     price × 1.20
#
# calculates the final price including 20% tax.
#
# Example:
#
#     100 × 1.20 = 120
#
#     Tax = 100 × 0.20 = 20
#     Final = 100 + 20 = 120
#
# ------------------------------------------------------------
# DATA TRANSFORMATION
# ------------------------------------------------------------
#
# This is a simple DATA TRANSFORMATION pipeline:
#
#     INPUT
#       ↓
#     [100, 200, 300, 400]
#       ↓
#     TRANSFORM EACH ITEM
#       ↓
#     price × 1.20
#       ↓
#     OUTPUT
#       ↓
#     [120.0, 240.0, 360.0, 480.0]
#
# The number of elements stays the same.
# Only the values are transformed.
#
# ------------------------------------------------------------
# WHAT CHANGES / WHAT STAYS THE SAME
# ------------------------------------------------------------
#
# Changes:
#     Each price value
#
# Stays the same:
#     Number of elements → 4
#     Order of elements → preserved
#     Original prices → unchanged
#
# So this is a 1-to-1 transformation:
#
#     1 input item → 1 output item
#
# ------------------------------------------------------------
# EQUIVALENT BEGINNER-FRIENDLY VERSION
# ------------------------------------------------------------
#
# The list comprehension is conceptually equivalent to:
#
#     prices_with_tax = []
#
#     for price in prices:
#         taxed_price = price * 1.20
#         prices_with_tax.append(taxed_price)
#
# The comprehension simply expresses this pattern more compactly.
#
# ------------------------------------------------------------
# GENERALIZABLE PATTERN
# ------------------------------------------------------------
#
# Whenever you hear:
#
#     "For every item, calculate a new value."
#
# Think:
#
#     [transformation for item in collection]
#
# Examples:
#
#     [price * 1.20 for price in prices]
#     → add 20% tax
#
#     [price * 0.90 for price in prices]
#     → apply a 10% discount
#
#     [number ** 2 for number in numbers]
#     → square every number
#
#     [name.upper() for name in names]
#     → convert every name to uppercase
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
# List comprehension = "transform every item and collect the results."
#
#     collection
#          ↓
#     one item
#          ↓
#     transformation
#          ↓
#     result
#          ↓
#     new list
#
# ------------------------------------------------------------
# CORE IDEA
# ------------------------------------------------------------
#
# A list comprehension is especially useful when the task is:
#
#     "Take every item from a collection,
#      perform the same transformation,
#      and build a new list from the results."
#
# Here:
#
#     prices
#       → apply 20% tax to each price
#       → prices_with_tax
#
# ============================================================

# Define a list of prices
prices = [100, 200, 300, 400]

# Use list comprehension to create a new list with prices including 20% tax
# Each price is multiplied by 1.20 to add the tax
prices_with_tax = [
    price * 1.20  # Apply 20% tax to each price
    for price in prices  # Iterate through each price in the original list
]

# Print the new list with tax-included prices
print(prices_with_tax)
```

```python
# ============================================================
# 65. LIST COMPREHENSION — IF ELSE
# ============================================================
#
# A list comprehension can contain an if-else expression to
# produce different results depending on a condition.
#
# The basic pattern is:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# IMPORTANT:
#
# In this form, the if-else expression comes BEFORE the
# for loop.
#
# Think:
#
#     For each item:
#         if condition → return this
#         else         → return that
#
# ============================================================


# ============================================================
# CREATE A LIST OF SCORES
# ============================================================
#
# Each score will be checked to determine whether the student
# passed or failed.
#
# Passing score:
#
#     50 or higher → Pass
#
# Below 50:
#
#     Fail
#
# ============================================================

scores = [45, 67, 82, 91, 38]


# ============================================================
# LIST COMPREHENSION WITH IF ELSE
# ============================================================
#
# For each score:
#
#     score >= 50 → "Pass"
#     otherwise   → "Fail"
#
# Step by step:
#
#     45 → 45 >= 50 → False → "Fail"
#     67 → 67 >= 50 → True  → "Pass"
#     82 → 82 >= 50 → True  → "Pass"
#     91 → 91 >= 50 → True  → "Pass"
#     38 → 38 >= 50 → False → "Fail"
#
# Result:
#
#     ["Fail", "Pass", "Pass", "Pass", "Fail"]
#
# ============================================================

results = [
    "Pass" if score >= 50 else "Fail"
    for score in scores
]


# ============================================================
# PRINT RESULT
# ============================================================

print(results)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The list comprehension above is equivalent to:
#
#     results = []
#
#     for score in scores:
#
#         if score >= 50:
#             results.append("Pass")
#         else:
#             results.append("Fail")
#
# The comprehension combines the same logic into one
# expression.
#
# ============================================================


# ============================================================
# IMPORTANT: IF-ELSE VS IF-ONLY
# ============================================================
#
# IF-ONLY:
#
#     [expression
#      for item in iterable
#      if condition]
#
# The if comes AFTER the for because it FILTERS items.
#
#
# IF-ELSE:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# The if-else comes BEFORE the for because it TRANSFORMS
# every item into one of two possible values.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, choose A or B."
#
# Think:
#
#     [A if condition else B for item in iterable]
#
#
# When you think:
#
#     "Keep only the items that satisfy a condition."
#
# Think:
#
#     [item for item in iterable if condition]
#
# The key difference:
#
#     IF ONLY → FILTER
#     IF ELSE → CHOOSE / TRANSFORM
#
# ============================================================
```

```python
# ============================================================
# 66. DICTIONARY COMPREHENSION
# ============================================================
#
# A dictionary comprehension is a compact way to create a
# dictionary by generating KEY-VALUE pairs from an iterable.
#
# The basic pattern is:
#
#     {key: value for item in iterable}
#
# It is similar to a list comprehension, but instead of
# producing individual values, it produces:
#
#     KEY → VALUE
#
# Dictionary comprehensions are especially useful when you
# want to transform data into a dictionary in a single,
# readable expression.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains the numbers that we want to process.
#
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# DICTIONARY COMPREHENSION
# ============================================================
#
# For each number:
#
#     number       → KEY
#     number ** 2  → VALUE
#
# Step by step:
#
#     1 → 1² → 1  → 1: 1
#     2 → 2² → 4  → 2: 4
#     3 → 3² → 9  → 3: 9
#     4 → 4² → 16 → 4: 16
#     5 → 5² → 25 → 5: 25
#
# Result:
#
#     {
#         1: 1,
#         2: 4,
#         3: 9,
#         4: 16,
#         5: 25
#     }
#
# ============================================================

squares = {
    number: number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The dictionary comprehension above is equivalent to:
#
#     squares = {}
#
#     for number in numbers:
#         squares[number] = number ** 2
#
# The comprehension simply expresses the same operation
# in a more compact form.
#
# ============================================================


# ============================================================
# DICTIONARY COMPREHENSION WITH A CONDITION
# ============================================================
#
# You can also filter items using an if condition.
#
# Example:
#
#     even_squares = {
#         number: number ** 2
#         for number in numbers
#         if number % 2 == 0
#     }
#
# Result:
#
#     {2: 4, 4: 16}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, create a KEY → VALUE pair."
#
# Think:
#
#     {key: value for item in iterable}
#
# The three main comprehension types are:
#
#     [expression for item in iterable]
#         → LIST
#
#     {expression for item in iterable}
#         → SET
#
#     {key: value for item in iterable}
#         → DICTIONARY
#
# ============================================================
```

```python
# ============================================================
# 67. SET COMPREHENSION
# ============================================================
#
# A set comprehension is a compact way to create a set by
# applying an expression to each item in an iterable.
#
# The basic pattern is:
#
#     {expression for item in iterable}
#
# It is similar to a list comprehension, but uses { } and
# produces a SET instead of a LIST.
#
# Important:
#
#     List → ordered collection, allows duplicates
#     Set  → unique elements, removes duplicates
#
# This makes set comprehensions useful when you want to
# transform data and automatically keep only unique results.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains duplicate values:
#
#     2 appears twice
#     3 appears twice
#
# ============================================================

numbers = [1, 2, 2, 3, 3, 4, 5]


# ============================================================
# SET COMPREHENSION
# ============================================================
#
# For each number:
#
#     number ** 2
#
# calculates its square.
#
# The resulting values are stored in a SET.
#
# Because a set contains only unique elements, duplicate
# results are automatically removed.
#
# Step by step:
#
#     1 → 1² → 1
#     2 → 2² → 4
#     2 → 2² → 4  ← duplicate
#     3 → 3² → 9
#     3 → 3² → 9  ← duplicate
#     4 → 4² → 16
#     5 → 5² → 25
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================

squares = {
    number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# SET COMPREHENSION VS LIST COMPREHENSION
# ============================================================
#
# List comprehension:
#
#     squares = [
#         number ** 2
#         for number in numbers
#     ]
#
# Result:
#
#     [1, 4, 4, 9, 9, 16, 25]
#
#
# Set comprehension:
#
#     squares = {
#         number ** 2
#         for number in numbers
#     }
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "Transform every item and keep only unique results."
#
# Think:
#
#     {expression for item in iterable}
#
# The key difference is:
#
#     [ ] → list comprehension
#     { } → set comprehension
#
# ============================================================
```

```python
# ============================================================
# 68. NESTED LIST COMPREHENSION
# ============================================================
#
# A nested list comprehension is a list comprehension that
# contains more than one for loop.
#
# It is useful when working with nested structures such as:
#
#     list of lists
#     2D matrices
#     tables
#     nested data
#
# The general pattern is:
#
#     [value for outer_item in collection
#            for inner_item in outer_item]
#
# Think:
#
#     Outer loop → go through each row
#     Inner loop → go through each value in the row
#
# ============================================================


# ============================================================
# CREATE A 2D MATRIX
# ============================================================
#
# A matrix can be represented as a list containing multiple
# lists.
#
#     Row 1 → [1, 2]
#     Row 2 → [3, 4]
#     Row 3 → [5, 6]
#
# ============================================================

matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]


# ============================================================
# NESTED LIST COMPREHENSION
# ============================================================
#
# We want to flatten the matrix:
#
#     [[1, 2], [3, 4], [5, 6]]
#
# into:
#
#     [1, 2, 3, 4, 5, 6]
#
# The nested comprehension:
#
#     [
#         value
#         for row in matrix
#         for value in row
#     ]
#
# is equivalent to:
#
#     flattened = []
#
#     for row in matrix:
#
#         for value in row:
#
#             flattened.append(value)
#
# ============================================================

flattened = [
    value
    for row in matrix
    for value in row
]


# ============================================================
# PRINT RESULT
# ============================================================
#
# Output:
#
#     [1, 2, 3, 4, 5, 6]
#
# ============================================================

print(flattened)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see nested data:
#
#     [
#         [1, 2],
#         [3, 4],
#         [5, 6]
#     ]
#
# Think:
#
#     for each ROW
#         for each VALUE in that ROW
#
# In normal loops:
#
#     for row in matrix:
#         for value in row:
#             ...
#
# In a nested list comprehension:
#
#     [value for row in matrix for value in row]
#
# IMPORTANT:
#
# The order of the for clauses is the same as the order of
# the normal nested loops.
#
# ============================================================
```

```python
# ============================================================
# 69. NUMPY LOOP
# ============================================================
#
# A NumPy array can be iterated over using a normal Python
# for loop.
#
# The loop takes each element from the NumPy array one by one.
#
# Think:
#
#     NumPy array → one element → process it → next element
#
# Note:
#
# NumPy is designed for vectorized operations, so explicit
# Python loops are often not the best choice for numerical
# operations on large arrays.
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# Import NumPy so we can create and work with NumPy arrays.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE A NUMPY ARRAY
# ============================================================
#
# Create a NumPy array containing four numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40])


# ============================================================
# ITERATE THROUGH THE ARRAY
# ============================================================
#
# The for loop visits each element one at a time:
#
#     10 → number → print
#     20 → number → print
#     30 → number → print
#     40 → number → print
#
# ============================================================

for number in numbers:
    print(number)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every element in this array, do something."
#
# Think:
#
#     for number in numbers:
#
# This works because a NumPy array is iterable.
#
# However, if the operation can be expressed as a NumPy
# vectorized operation, prefer vectorization for numerical
# computations, especially with large arrays.
#
# Example:
#
#     # Loop
#     for number in numbers:
#         print(number ** 2)
#
#     # Vectorized
#     print(numbers ** 2)
#
# ============================================================
```

```python
# ============================================================
# 71. NUMPY BOOLEAN MASKING
# ============================================================
#
# Boolean masking is a powerful NumPy technique for filtering
# elements of an array based on a condition.
#
# The basic idea is:
#
#     array[condition]
#
# NumPy first evaluates the condition for every element and
# creates a Boolean mask containing True or False values.
#
#     True  → keep the element
#     False → exclude the element
#
# Example:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# This mask is then used to select the corresponding elements.
#
# Think:
#
#     Condition → Boolean mask → Filtered array
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# NumPy provides arrays and efficient operations for numerical
# and data-science computations.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE NUMPY ARRAY
# ============================================================
#
# Create an array containing five numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40, 50])


# ============================================================
# CREATE A BOOLEAN MASK
# ============================================================
#
# The condition:
#
#     numbers > 25
#
# is evaluated element by element:
#
#     10 > 25 → False
#     20 > 25 → False
#     30 > 25 → True
#     40 > 25 → True
#     50 > 25 → True
#
# Therefore:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# ============================================================

mask = numbers > 25

print(mask)


# ============================================================
# APPLY THE BOOLEAN MASK
# ============================================================
#
# Use the Boolean mask to select only the elements where
# the mask is True.
#
#     numbers[mask]
#
# returns:
#
#     [30, 40, 50]
#
# ============================================================

print(numbers[mask])


# ============================================================
# SHORT FORM
# ============================================================
#
# The mask does not need to be stored in a separate variable.
#
# We can directly write:
#
#     numbers[numbers > 25]
#
# This means:
#
#     "Select elements from numbers where numbers > 25."
#
# ============================================================

print(numbers[numbers > 25])


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see:
#
#     "Keep only the values that satisfy a condition."
#
# Think:
#
#     Boolean masking
#
# The general pattern is:
#
#     array[array condition]
#
# Examples:
#
#     numbers[numbers > 25]
#     numbers[numbers < 25]
#     numbers[numbers == 30]
#     numbers[numbers != 30]
#
# Boolean masking is extremely important in Data Science
# because the same idea appears in NumPy, Pandas, and data
# filtering.
#
# ============================================================
```

```python
# ============================================================
# 72. LOOP VS NUMPY VECTORIZATION
# ============================================================
#
# There are two common ways to perform the same operation
# on multiple values:
#
# 1. Python loop
# 2. NumPy vectorization
#
# A loop processes the elements one by one.
# Vectorization applies the operation to the entire array
# at once.
#
# In Data Science, vectorization is usually preferred because
# NumPy operations are implemented in optimized low-level code
# and are generally faster for large numerical datasets.
#
# Think:
#
#     Loop          → one element at a time
#     Vectorization → whole array at once
#
# ============================================================

numbers = np.array([1, 2, 3, 4, 5])


# ============================================================
# LOOP
# ============================================================
#
# The loop takes each number individually:
#
#     1 → 1²
#     2 → 2²
#     3 → 3²
#     4 → 4²
#     5 → 5²
#
# The result is gradually built inside a new list.
#
# ============================================================

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)


# ============================================================
# NUMPY VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to apply the operation directly to the
# entire array.
#
# No explicit Python loop is needed.
#
#     [1, 2, 3, 4, 5] ** 2
#              ↓
#     [1, 4, 9, 16, 25]
#
# ============================================================

squares = numbers ** 2

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every number, do this..."
#
# A Python loop is the natural starting point.
#
# But when working with NumPy arrays, ask:
#
#     "Can NumPy do this operation on the whole array?"
#
# If yes, prefer the vectorized NumPy operation in most
# data-processing situations.
#
# ============================================================
```

```python
# ============================================================
# VECTORWISE / VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to perform an operation on an entire array
# without writing an explicit Python for loop.
#
# This is called VECTORIZATION.
#
# Think:
#
#     Python loop:
#         one element → operation
#         one element → operation
#         one element → operation
#
#     NumPy vectorization:
#         entire array → operation
#
# NumPy performs the operation element by element internally,
# but the iteration is handled by optimized NumPy code rather
# than an explicit Python loop.
#
# ============================================================


# ============================================================
# CALCULATE SQUARES
# ============================================================
#
# The ** operator is applied to every element of the NumPy
# array.
#
# If:
#
#     numbers = [1, 2, 3, 4, 5]
#
# Then:
#
#     numbers ** 2
#
# produces:
#
#     [1, 4, 9, 16, 25]
#
# No explicit for loop is needed.
#
# ============================================================

squares = numbers ** 2


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with NumPy arrays, think:
#
#     "Can I apply this operation to the whole array?"
#
# If yes, NumPy often allows you to write:
#
#     array ** 2
#     array + 10
#     array * 5
#     array / 2
#
# instead of explicitly writing a Python loop.
#
# This usually makes numerical code:
#
#     ✓ shorter
#     ✓ clearer
#     ✓ easier to read
#     ✓ often faster for large numerical arrays
#
# ============================================================
```

```python
# ============================================================
# 73. LOOP VS PANDAS VECTORIZATION
# ============================================================
#
# Pandas provides vectorized operations that allow us to
# perform calculations on an entire column without explicitly
# writing a Python loop.
#
# There are two approaches:
#
#     1. Loop through rows
#     2. Vectorized column operation
#
# The loop processes data one row at a time.
# Vectorization applies the operation to the entire column.
#
# For typical numerical and column-based operations, the
# vectorized approach is usually preferred.
#
# ============================================================


# ============================================================
# CREATE DATAFRAME
# ============================================================
#
# Create a DataFrame containing employee salaries.
#
# ============================================================

df = pd.DataFrame({
    "Salary": [30000, 50000, 70000]
})


# ============================================================
# LOOP — PROCESS ROW BY ROW
# ============================================================
#
# iterrows() allows us to iterate through a DataFrame
# one row at a time.
#
#     index → current row index
#     row   → current row
#
# For each row, increase the salary by 10%.
#
#     Salary × 1.10
#
# ============================================================

for index, row in df.iterrows():

    print(row["Salary"] * 1.10)


# ============================================================
# VECTORIZED OPERATION
# ============================================================
#
# Pandas can perform the same calculation directly on the
# entire Salary column.
#
#     df["Salary"] * 1.10
#
# is applied element by element to the whole Series.
#
#     30000 × 1.10 → 33000
#     50000 × 1.10 → 55000
#     70000 × 1.10 → 77000
#
# The results are stored in a new column.
#
# ============================================================

df["New_Salary"] = df["Salary"] * 1.10


# ============================================================
# PRINT RESULT
# ============================================================

print(df)


# ============================================================
# LOOP VS VECTORIZATION
# ============================================================
#
# LOOP:
#
#     for index, row in df.iterrows():
#         ...
#
#     → processes one row at a time
#
#
# VECTORIZATION:
#
#     df["Salary"] * 1.10
#
#     → processes the entire column
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with Pandas and you think:
#
#     "For every row, calculate something."
#
# Don't immediately write a loop.
#
# First ask:
#
#     "Can Pandas do this to the entire column?"
#
# If yes, use the vectorized operation.
#
# Think:
#
#     Python loop → one row at a time
#     Pandas       → entire column
#
# This is an important shift from basic Python programming
# toward efficient Data Science programming.
#
# ============================================================
```

```python
# ============================================================
# 74. PRACTICAL ML EXAMPLE
# ============================================================
#
# A common Machine Learning task is converting predicted
# probabilities into class labels.
#
# For binary classification, a model may output a probability
# between 0 and 1 instead of directly returning 0 or 1.
#
# Example:
#
#     0.2 → low probability  → class 0
#     0.7 → high probability → class 1
#
# Here, we use 0.5 as the classification threshold:
#
#     probability >= 0.5 → prediction = 1
#     probability <  0.5 → prediction = 0
#
# Think:
#
#     Model probability
#            ↓
#       Apply threshold
#            ↓
#       Class prediction
#
# ============================================================


# ============================================================
# MODEL PREDICTION PROBABILITIES
# ============================================================
#
# These values represent the predicted probability of the
# positive class for five samples.
#
# In a real ML project, these probabilities might come from:
#
#     model.predict_proba(X)
#
# ============================================================

# List of predicted probabilities from a machine learning model
# Each value represents the model's confidence that a sample belongs to the positive class
probabilities = [0.2, 0.7, 0.9, 0.3, 0.8]


# ============================================================
# CREATE AN EMPTY LIST
# ============================================================
#
# We will store the final class predictions in this list.
#
# ============================================================

# Initialize an empty list to store the binary predictions
# This will hold the final class labels (0 or 1) after applying the threshold
predictions = []


# ============================================================
# LOOP THROUGH PROBABILITIES
# ============================================================
#
# Process each probability one at a time.
#
# The condition determines which class the probability
# belongs to.
#
#     probability >= 0.5 → 1
#     probability <  0.5 → 0
#
# Step by step:
#
#     0.2 → < 0.5 → 0
#     0.7 → >= 0.5 → 1
#     0.9 → >= 0.5 → 1
#     0.3 → < 0.5 → 0
#     0.8 → >= 0.5 → 1
#
# ============================================================

# Iterate through each probability in the list
for probability in probabilities:

    # Check if the probability is greater than or equal to the threshold (0.5)
    if probability >= 0.5:
        # If probability is high, assign class 1 (positive class)
        predictions.append(1)

    # Otherwise, the probability is below the threshold
    else:
        # If probability is low, assign class 0 (negative class)
        predictions.append(0)


# ============================================================
# PRINT PREDICTIONS
# ============================================================
#
# Output:
#
#     [0, 1, 1, 0, 1]
#
# ============================================================

# Print the final predictions
# This converts the probabilities into binary class labels based on the threshold
print(predictions)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a classic:
#
#     LOOP + CONDITION + RESULT COLLECTION
#
# Pattern:
#
#     result = []
#
#     for item in data:
#
#         if condition:
#             result.append(value_1)
#
#         else:
#             result.append(value_2)
#
# In Machine Learning, this pattern appears in tasks such as:
#
#     probability → class label
#     score       → decision
#     prediction  → category
#
# ============================================================
```

```python
# ============================================================
# 75. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================
#
# This example shows how a Machine Learning prediction can
# support a real-world engineering decision.
#
# Suppose an ML model predicts the probability that each water
# pipeline will fail.
#
# Instead of using only two classes:
#
#     0 → No failure
#     1 → Failure
#
# we can use the predicted probability to define different
# maintenance actions.
#
# Higher probability → higher risk → more urgent action.
#
# This is an example of:
#
#     ML prediction → Risk level → Engineering action
#
# ============================================================


# ============================================================
# FAILURE PROBABILITIES
# ============================================================
#
# Each value represents the predicted probability of pipeline
# failure.
#
#     0.15 → 15% probability
#     0.42 → 42% probability
#     0.73 → 73% probability
#     0.91 → 91% probability
#     0.31 → 31% probability
#
# In a real ML project, these values could come from a model
# such as:
#
#     model.predict_proba(X)
#
# ============================================================

failure_probabilities = [
    0.15,
    0.42,
    0.73,
    0.91,
    0.31
]


# ============================================================
# RISK-BASED DECISION LOGIC
# ============================================================
#
# Each probability is checked against several thresholds.
#
#     >= 0.80 → Immediate inspection
#     >= 0.60 → Schedule inspection
#     >= 0.40 → Monitor
#     <  0.40 → Normal operation
#
# Notice that the conditions are checked from HIGH to LOW.
#
# Once a condition is True, Python executes that block and
# skips the remaining elif / else conditions.
#
# ============================================================

for probability in failure_probabilities:

    if probability >= 0.80:
        action = "Immediate inspection"

    elif probability >= 0.60:
        action = "Schedule inspection"

    elif probability >= 0.40:
        action = "Monitor"

    else:
        action = "Normal operation"


    # ========================================================
    # DISPLAY RESULT
    # ========================================================
    #
    # Show the predicted probability together with the
    # corresponding engineering action.
    #
    # ========================================================

    print(
        probability,
        "->",
        action
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     0.15 -> Normal operation
#     0.42 -> Monitor
#     0.73 -> Schedule inspection
#     0.91 -> Immediate inspection
#     0.31 -> Normal operation
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a very useful pattern:
#
#     FOR each prediction:
#         IF high risk:
#             urgent action
#         ELIF medium-high risk:
#             planned action
#         ELIF moderate risk:
#             monitoring
#         ELSE:
#             normal operation
#
# In Data Science and ML, this connects:
#
#     DATA
#       ↓
#     ML MODEL
#       ↓
#     PROBABILITY
#       ↓
#     THRESHOLD / DECISION RULE
#       ↓
#     REAL-WORLD ACTION
#
# ============================================================


# ============================================================
# IMPORTANT ML INSIGHT
# ============================================================
#
# The thresholds (0.80, 0.60, 0.40) are only example values.
#
# In a real pipeline-failure prediction system, thresholds
# should ideally be determined using:
#
#     • historical failure data
#     • cost of missed failures
#     • cost of unnecessary inspections
#     • precision / recall
#     • ROC / PR analysis
#     • engineering constraints
#     • domain expertise
#
# Therefore:
#
#     Probability ≠ automatic decision
#
# A probability becomes an operational decision only after
# an appropriate decision threshold or risk policy is defined.
#
# ============================================================
```

```python
# ============================================================
# 76. PRACTICAL ML METRICS
# ============================================================
#
# Machine Learning models are often evaluated using multiple
# metrics rather than relying on a single score.
#
# Common classification metrics include:
#
#     Accuracy  → overall proportion of correct predictions
#     Precision → proportion of predicted positives that are
#                 actually positive
#     Recall    → proportion of actual positives that are found
#     F1        → harmonic mean of Precision and Recall
#
# Storing metrics in a dictionary makes it easy to organize
# and iterate through them.
#
# Think:
#
#     Metric name → Metric value
#
# ============================================================


# ============================================================
# CREATE A DICTIONARY OF METRICS
# ============================================================
#
# The dictionary stores each metric as a KEY-VALUE pair:
#
#     KEY   → metric name
#     VALUE → metric score
#
# The scores are represented as decimal values:
#
#     0.91 → 91%
#     0.84 → 84%
#     0.76 → 76%
#     0.80 → 80%
#
# ============================================================

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}


# ============================================================
# LOOP THROUGH THE DICTIONARY
# ============================================================
#
# .items() gives us both the key and value:
#
#     metric → "Accuracy"
#     value  → 0.91
#
#     metric → "Precision"
#     value  → 0.84
#
#     ...
#
# ============================================================

for metric, value in metrics.items():

    print(
        metric,
        ":",
        value
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     Accuracy : 0.91
#     Precision : 0.84
#     Recall : 0.76
#     F1 : 0.8
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you have a dictionary and need both:
#
#     KEY + VALUE
#
# Think:
#
#     for key, value in dictionary.items():
#
# This pattern is extremely common in Data Science and
# Machine Learning.
#
# For example:
#
#     for metric, value in metrics.items():
#         print(metric, value)
#
#     for parameter, value in model_params.items():
#         print(parameter, value)
#
#     for feature, importance in feature_importance.items():
#         print(feature, importance)
#
# ============================================================
```

```python
# ============================================================
# 77. CHECKING ML METRICS
# ============================================================
#
# This example evaluates several ML performance metrics using
# the same threshold-based decision rules.
#
# CORE PATTERN:
#
# metrics
#    ↓
# iterate through each metric
#    ↓
# evaluate its value
#    ↓
# choose a performance category
#    ↓
# print the result
#
# The important idea is that the SAME decision logic is applied
# independently to every metric.

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}

# CONCEPTUAL ROLES:
#
# metrics
#    → dictionary containing metric names and their scores
#
# metric
#    → current metric name (str)
#
# value
#    → current metric score (float)
#
# performance
#    → classification assigned according to the thresholds
#
# for
#    → processes each key-value pair in the dictionary
#
# if / elif / else
#    → converts a numeric score into a performance category

for metric, value in metrics.items():

    # Python evaluates the conditions from top to bottom.
    # Once one condition is True, its block executes and the
    # remaining elif/else branches are skipped.

    if value >= 0.90:
        performance = "Excellent"      # Score is at least 0.90

    elif value >= 0.80:
        performance = "Good"           # Score is 0.80–0.899...

    elif value >= 0.70:
        performance = "Acceptable"     # Score is 0.70–0.799...

    else:
        performance = "Poor"           # Score is below 0.70

    print(
        metric,
        "->",
        performance
    )

# ============================================================
# VALUE FLOW
# ============================================================
#
# Each iteration follows:
#
# metric + value
#       ↓
# compare value with thresholds
#       ↓
# True condition
#       ↓
# assign performance
#       ↓
# print metric + performance
#
# Example:
#
# Accuracy + 0.91
#       ↓
# 0.91 >= 0.90
#       ↓
# True
#       ↓
# performance = "Excellent"
#       ↓
# Accuracy -> Excellent

# ============================================================
# PROCESS TABLE
# ============================================================
#
# | Step | Metric    | Value | Selected Condition | Performance |
# |------|-----------|-------|--------------------|-------------|
# | 1    | Accuracy  | 0.91  | >= 0.90            | Excellent   |
# | 2    | Precision | 0.84  | >= 0.80            | Good        |
# | 3    | Recall    | 0.76  | >= 0.70            | Acceptable  |
# | 4    | F1        | 0.80  | >= 0.80            | Good        |
#
# Notice that 0.80 satisfies >= 0.80, so F1 is "Good".
#
# ============================================================
# EXECUTION TRACE
# ============================================================
#
# Iteration 1:
# metric = "Accuracy", value = 0.91
# 0.91 >= 0.90 → True
# performance = "Excellent"
# → Accuracy -> Excellent
#
# Iteration 2:
# metric = "Precision", value = 0.84
# 0.84 >= 0.90 → False
# 0.84 >= 0.80 → True
# performance = "Good"
# → Precision -> Good
#
# Iteration 3:
# metric = "Recall", value = 0.76
# 0.76 >= 0.90 → False
# 0.76 >= 0.80 → False
# 0.76 >= 0.70 → True
# performance = "Acceptable"
# → Recall -> Acceptable
#
# Iteration 4:
# metric = "F1", value = 0.80
# 0.80 >= 0.90 → False
# 0.80 >= 0.80 → True
# performance = "Good"
# → F1 -> Good
#
# No more dictionary items → loop ends.

# ============================================================
# STATE
# ============================================================
#
# The dictionary remains unchanged.
#
# performance is reassigned on every iteration:
#
# "Excellent" → "Good" → "Acceptable" → "Good"
#
# The loop variable values also change:
#
# metric:
# "Accuracy" → "Precision" → "Recall" → "F1"
#
# value:
# 0.91 → 0.84 → 0.76 → 0.80
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# This is a common DATA ANALYTICS / ML pattern:
#
# data
#   ↓
# iterate through observations
#   ↓
# apply rules
#   ↓
# classify each value
#   ↓
# report results
#
# The same structure can be used for:
# - model quality checks
# - data validation
# - risk levels
# - customer segmentation
# - alert systems
# - KPI classification
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A conditional chain converts numeric metric values into
# meaningful categories.
#
# The for loop allows the same rules to be applied consistently
# to every metric.
```
