## LIST COMPREHENSIONS


```python
# ============================================================
# LIST COMPREHENSIONS
# ============================================================
#
# A list comprehension is a concise way to create a NEW list
# by processing each item in an iterable.
#
# BASIC SYNTAX
# ------------------------------------------------------------
#
# [expression for item in iterable]
#      (3)       (1)       (2)
#
# Written syntax:
#
#     [ expression  for item  in iterable ]
#          │            │          │
#          │            │          └── source iterable (2)
#          │            │
#          │            └───────────── current item (1)
#          │
#          └────────────────────────── operation (3)
#
# Conceptual execution:
#
#     iterable
#         ↓
#     get next item
#         ↓
#     item
#         ↓
#     expression
#         ↓
#     result
#         ↓
#     collect result
#         ↓
#     repeat
#
# Example:
#
#     [x**2 for x in numbers]
#
# Means:
#
#     "For each x in numbers, calculate x**2
#      and collect the results in a new list."
#
# ============================================================
# THE THREE PARTS
# ============================================================
#
#     [ x**2  for x  in numbers ]
#        │       │        │
#        │       │        └── source iterable → provides the values (2)
#        │       │            
#        │       │
#        │       └─────────── loop variable → receives each item   (1)
#        │                    
#        │
#        └────────────────── expression → processes the current item (3)
#                             
#
# Conceptual order:
#
#     numbers → x → x**2 → result
#
# ============================================================
# EXAMPLE
# ============================================================

numbers = [1, 2, 3, 4, 5]

squares = [x**2 for x in numbers]

# Value flow:
#
#     numbers
#     [1, 2, 3, 4, 5]
#           ↓
#        iterate
#           ↓
#     x = 1 → 1**2 → 1
#     x = 2 → 2**2 → 4
#     x = 3 → 3**2 → 9
#     x = 4 → 4**2 → 16
#     x = 5 → 5**2 → 25
#           ↓
#     collect results
#           ↓
#     squares
#     [1, 4, 9, 16, 25]
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
# Step | x | x**2 | squares
# -----|---|------|---------------------------
#  1   | 1 |  1   | [1]
#  2   | 2 |  4   | [1, 4]
#  3   | 3 |  9   | [1, 4, 9]
#  4   | 4 | 16   | [1, 4, 9, 16]
#  5   | 5 | 25   | [1, 4, 9, 16, 25]
#
# ============================================================
# EQUIVALENT FOR-LOOP
# ============================================================
#
#     squares = []
#
#     for x in numbers:
#         squares.append(x**2)
#
# Both follow the same pattern:
#
#     take item → process item → collect result
#
# ============================================================
# CORE IDEA
# ============================================================
#
# Written:
#
#     [expression for item in iterable]
#
# Think:
#
#     iterable → item → expression → result
#
# In this example:
#
#     numbers → x → x**2 → squares
#
# ============================================================
```

```python
#  IMPORTANT:
# The expression is evaluated once for each item.
#
# For example:
#
# [x**2 for x in numbers]
#
# means:
#
# "For every x in numbers, calculate x**2,
#  and put that result into a new list."
# 
# [x**2 for x in numbers]
#   │       │       │
#   │       │       └── source iterable 2
#   │       └────────── loop variable   1
#   └────────────────── expression      3
#     
#     
#     
#[ x**2  for x  in numbers ]
#   3       1       2  
#   │       │       │
#   │       │       └── source iterable → 2. in numbers   → obtain items from numbers
#   │       └────────── loop variable   → 1. for x        → iterate / get the next item
#   └────────────────── expression      → 3. x**2         → process the current x





# 1. for x        → iterate / get the next item
# 2. in numbers   → obtain items from numbers
# 3. x**2         → process the current x

# WRITTEN SYNTAX
# ────────────────────────────

# [ expression  for item in iterable ]
#      3             1         2

```

```python
# ============================================================
# THE MENTAL READING OF A LIST COMPREHENSION
# ============================================================
#
# [expression for item in iterable]
#
# Read it conceptually as:
#
# [ (3) DO WHAT IS IN expression
#       (1) for each item
#       in (2) iterable ]
#
# Execution idea:
#
# (2) iterable
#       ↓
# (1) get each item
#       ↓
# (3) do what is in expression
#       ↓
#     collect the result
#       ↓
#     repeat
#
```

```python
numbers = range(1, 10)

# For each number in numbers, calculate number * number.
squares = [number * number for number in numbers]

print(list(numbers))
print(squares)

# ============================================================
# THE "ENGLISH-TO-PYTHON" TRANSLATION
# ============================================================
#
# [expression for item in iterable]
#
# Mentally read it as:
#
# [ (3) DO expression
#       (1) FOR EACH item
#       (2) IN iterable ]
#
# Example:
#
# [number * number for number in numbers]
#
# numbers           → (2) source values
# number            → (1) current value
# number * number   → (3) operation
#
# Execution:
#
# numbers → number → number * number → result
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
# Iteration | number | result | squares
# ----------|--------|--------|---------------------------
# Start     |   -    |   -    | []
#     1     |   1    |   1    | [1]
#     2     |   2    |   4    | [1, 4]
#     3     |   3    |   9    | [1, 4, 9]
#     4     |   4    |  16    | [1, 4, 9, 16]
#     5     |   5    |  25    | [1, 4, 9, 16, 25]
#     6     |   6    |  36    | [1, 4, 9, 16, 25, 36]
#     7     |   7    |  49    | [1, 4, 9, 16, 25, 36, 49]
#     8     |   8    |  64    | [1, 4, 9, 16, 25, 36, 49, 64]
#     9     |   9    |  81    | [1, 4, 9, 16, 25, 36, 49, 64, 81]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# The syntax is written:
#
#     [expression for item in iterable]
#
# But mentally execute it as:
#
#     iterable → item → expression → result
#
# In this example:
#
#     numbers → number → number * number → squares
# ============================================================
```

```python
# ============================================================
# LIST COMPREHENSION — SIMPLE MENTAL MODEL
# ============================================================
#
# [expression for item in iterable]
#
# The simplest way to understand it:
#
# "Build a new list. For each item in an existing iterable,
#  do something to it and put the result into the new list."
#
# ============================================================
# ENGLISH → PYTHON
# ============================================================
#
# Written syntax:
#
#     [expression   for item   in iterable]
#          │            │           │
#          │            │           └── source
#          │            └────────────── current item
#          └─────────────────────────── operation
#
# Do NOT think of this as Python executing right-to-left.
# The right side provides the items; the expression is then
# evaluated for each item.
#
# Conceptual execution:
#
#     iterable
#         ↓
#     get an item
#         ↓
#     assign it to item
#         ↓
#     evaluate expression
#         ↓
#     collect result
#         ↓
#     repeat
#
# In plain English:
#
#     in iterable → "Where do the items come from?"
#     for item    → "Take one item at a time."
#     expression  → "What should I do with it?"
#
# ============================================================
# THE HAMBURGER-BUN ANALOGY
# ============================================================
#
# Imagine a basket of plain buns:
#
#     buns = ["bun1", "bun2", "bun3"]
#
# You want a new tray where every bun gets cheese:
#
#     cheeseburgers = ["🍔🧀" for bun in buns]
#
# Conceptually:
#
#     buns
#       ↓
#     take one bun
#       ↓
#     add cheese
#       ↓
#     collect result
#       ↓
#     repeat
#
# Result:
#
#     ["🍔🧀", "🍔🧀", "🍔🧀"]
#
# ============================================================
# NORMAL WAY → SHORTCUT
# ============================================================
#
# List comprehension:
#
#     squares = [x * x for x in [1, 2, 3, 4]]
#
# Equivalent for-loop:
#
#     squares = []
#
#     for x in [1, 2, 3, 4]:
#         squares.append(x * x)
#
# Both follow the same pattern:
#
#     create list
#         ↓
#     get each item
#         ↓
#     process item
#         ↓
#     add result
#
# The comprehension simply expresses this pattern more
# compactly.
#
# ============================================================
# GOLDEN RULE
# ============================================================
#
#     [expression for item in iterable]
#
#     expression  → "What do I do to each item?"
#     for item    → "Take each item."
#     in iterable  → "From this existing collection."
#
# Read it naturally as:
#
#     "A list of expression for each item in iterable."
#
# Example:
#
#     [x * 2 for x in range(5)]
#
# Think:
#
#     "A list of x times 2 for each x in range(5)."
#
# Result:
#
#     [0, 2, 4, 6, 8]
#
# ============================================================
# OPTIONAL IF CONDITION
# ============================================================
#
# If an if appears at the end:
#
#     [x for x in numbers if x > 5]
#
# Think:
#
#     "Take each x from numbers,
#      but only keep it if x > 5."
#
# So the basic pattern is:
#
#     take → process → collect
#
# And with a condition:
#
#     take → test → process/keep → collect
#
# ============================================================
```

```python
# ============================================================
# LIST COMPREHENSION — STRING REPETITION
# ============================================================
#
# A list comprehension can apply an operation to every item
# and collect the results into a NEW list.
#
# Here, * with a string repeats that string.
#
# ============================================================

# Source list containing emoji strings.
emoji = ['😊', '😂', '😍', '🤔', '😎']

# For each x in emoji, repeat x twice and collect the results.
doubbled = [x * 2 for x in emoji]

# Conceptual execution:
#
#     emoji
#       ↓
#     get x
#       ↓
#     x * 2
#       ↓
#     collect result
#       ↓
#     repeat
#
# Iteration | x    | x * 2
# ----------|------|------------
#     1     | 😊   | 😊😊
#     2     | 😂   | 😂😂
#     3     | 😍   | 😍😍
#     4     | 🤔   | 🤔🤔
#     5     | 😎   | 😎😎
#
# Final value:
#
#     doubbled = ['😊😊', '😂😂', '😍😍', '🤔🤔', '😎😎']

print(doubbled)

# Output:
#
# ['😊😊', '😂😂', '😍😍', '🤔🤔', '😎😎']
#
# CORE IDEA:
#
#     [expression for item in iterable]
#
#     source → current item → operation → new list
#
# Here:
#
#     emoji → x → x * 2 → doubbled
#
# IMPORTANT:
#
# x * 2 means "repeat the string twice" because x is a string.
# With numbers, * 2 would instead perform multiplication.
```

```python
# ============================================================
# LIST COMPREHENSION + enumerate()
# ============================================================
#
# enumerate() gives us TWO values for each item:
#
#     index → position of the item
#     item  → actual value
#
# Example:
#
#     enumerate(emoji)
#
#     0 → '😊'
#     1 → '😂'
#     2 → '😍'
#     3 → '🤔'
#     4 → '😎'
#
# ============================================================

# List of emojis.
emoji = ['😊', '😂', '😍', '🤔', '😎']

# For each item:
#
#     i → index
#     e → emoji
#
# Then repeat the emoji i times.
doubled = [e * i for i, e in enumerate(emoji)]

# Conceptual execution:
#
#     enumerate(emoji)
#          ↓
#     get index + emoji
#          ↓
#     e * i
#          ↓
#     collect result
#          ↓
#     repeat
#
# Iteration | i | e  | e * i
# ----------|---|----|--------
#     1     | 0 | 😊 | ''
#     2     | 1 | 😂 | '😂'
#     3     | 2 | 😍 | '😍😍'
#     4     | 3 | 🤔 | '🤔🤔🤔'
#     5     | 4 | 😎 | '😎😎😎😎'
#
# Notice:
#
#     e * i
#
# repeats the string e, i times.
#
# Therefore, the first emoji appears zero times because
# its index is 0.

print(doubled)

# Output:
#
# ['', '😂', '😍😍', '🤔🤔🤔', '😎😎😎😎']
#
# CORE IDEA:
#
#     enumerate() → index + item
#
#     index → controls repetition
#     item  → provides the string
#
#     enumerate → i, e → e * i → new list
```

```python
# ============================================================
# LIST COMPREHENSION + print()
# ============================================================
#
# This example demonstrates an important point:
#
# A list comprehension collects the VALUE returned by the
# expression — it does not collect whatever the expression
# happens to display.
#
# print() displays a value, but returns None.
#
# ============================================================

# Define a list of numbers.
numbers = [1, 2, 3, 4, 5]

# For each item in numbers:
#
#     i → index
#     x → value
#
# print(i, x) displays both values.
#
# BUT:
#
#     print() returns None
#
# Therefore, the comprehension collects None each time.
writer = [print(i, x) for i, x in enumerate(numbers)]

# Conceptual execution:
#
#     enumerate(numbers)
#          ↓
#     i, x
#          ↓
#     print(i, x)
#          ↓
#     displays the values
#          ↓
#     returns None
#          ↓
#     None is collected
#
# Iteration | i | x | print(i, x) | collected
# ----------|---|---|-------------|----------
#     1     | 0 | 1 |    0 1      | None
#     2     | 1 | 2 |    1 2      | None
#     3     | 2 | 3 |    2 3      | None
#     4     | 3 | 4 |    3 4      | None
#     5     | 4 | 5 |    4 5      | None
#
# So writer becomes:
#
#     [None, None, None, None, None]

print("\nwhat is inside writer now?")

print(writer)

# Output:
#
#     0 1
#     1 2
#     2 3
#     3 4
#     4 5
#
#     what is inside writer now?
#     [None, None, None, None, None]
#
# CORE IDEA:
#
#     display ≠ return value
#
# print() displays information but returns None.
#
# Therefore:
#
#     [print(...) for ...]
#
# is usually NOT useful when the goal is to create a list.
# ============================================================
```

```python
things = ["toy", "game", "book", "pen", "pencil"]

writer = [print(i, x) for i, x in enumerate(things)]

writer

# Output:
#
# 0 toy
# 1 game
# 2 book
# 3 pen
# 4 pencil
#
# writer contains:
#
# [None, None, None, None, None]
#
# Why?
#
#     print(i, x)
#          ↓
#     displays i and x
#          ↓
#     returns None
#          ↓
#     None is collected by the comprehension
#
# So:
#
#     display → print()
#     return  → None
#
# CORE IDEA:
#
# The list comprehension stores the RETURN VALUE of its
# expression, not what the expression displays.
```

# DATA SCIENCE EXAMPLE

```python
# ============================================================
# DATA SCIENCE EXAMPLE — FILTERING WITH A LIST COMPREHENSION
# ============================================================
#
# Suppose a dataset contains some invalid age values.
# We can keep only ages that satisfy a condition.
#
# ============================================================

ages = [18, 19, 20, 19, 16, 32, -3, 25, -1, 40]

# Keep each age only if age >= 0.
valid_ages = [age for age in ages if age >= 0]

# Conceptual execution:
#
#     ages
#       ↓
#     take age
#       ↓
#     age >= 0 ?
#      ↙     ↘
#    True   False
#      ↓       ↓
#    keep    discard
#      ↓
#    repeat
#
# Iteration | age | age >= 0 | Result
# ----------|-----|----------|--------
#     1     |  18 |   True   | keep 18
#     2     |  19 |   True   | keep 19
#     3     |  20 |   True   | keep 20
#     4     |  19 |   True   | keep 19
#     5     |  16 |   True   | keep 16
#     6     |  32 |   True   | keep 32
#     7     |  -3 |   False  | discard
#     8     |  25 |   True   | keep 25
#     9     |  -1 |   False  | discard
#    10     |  40 |   True   | keep 40
#
# Result:
#
#     valid_ages
#     [18, 19, 20, 19, 16, 32, 25, 40]
#
# ============================================================
# IMPORTANT: FILTERING
# ============================================================
#
# Basic transformation:
#
#     [expression for item in iterable]
#
# With a condition:
#
#     [expression for item in iterable if condition]
#
# Think:
#
#     take → test → keep/discard → repeat
#
# Here:
#
#     age → age >= 0 → keep/discard
#
# The original ages list remains unchanged.
# ============================================================

print(valid_ages)
```

```python
ages = [18, 19, 20, 19, 16, 32, 13, 25, 11, 40]

# Keep only ages that are 18 or older.
# Ages below 18 are excluded from the new list.
valid_ages = [age for age in ages if age >= 18]

# Conceptual flow:
#
#     ages
#       ↓
#     take age
#       ↓
#     age >= 18 ?
#      ↙      ↘
#    True    False
#     ↓        ↓
#   keep     discard
#
# Result:
#
#     [18, 19, 20, 19, 32, 25, 40]

print(valid_ages)
```

```python
numbers = range(10)

# Keep only the even numbers.
# Odd numbers are excluded because x % 2 == 0 is False for them.
results = [x for x in numbers if x % 2 == 0]

# Conceptual flow:
#
#     numbers
#        ↓
#     take x
#        ↓
#     x % 2 == 0 ?
#      ↙         ↘
#    True       False
#     ↓           ↓
#   keep        discard
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
# Step | x | x % 2 == 0 | Result
# -----|---|------------|--------
#  1   | 0 |    True    | keep 0
#  2   | 1 |    False   | discard
#  3   | 2 |    True    | keep 2
#  4   | 3 |    False   | discard
#  5   | 4 |    True    | keep 4
#  6   | 5 |    False   | discard
#  7   | 6 |    True    | keep 6
#  8   | 7 |    False   | discard
#  9   | 8 |    True    | keep 8
# 10   | 9 |    False   | discard
#
# Result:
#
#     [0, 2, 4, 6, 8]

results
```

```python
numbers = range(10)

# For each number:
#     if it is even → keep the number
#     if it is odd  → replace it with "Odd"
result = [x if x % 2 == 0 else "Odd" for x in numbers]

# Conceptual flow:
#
#     numbers
#        ↓
#     take x
#        ↓
#     x % 2 == 0 ?
#      ↙         ↘
#    True       False
#     ↓           ↓
#   keep x     "Odd"
#      \         /
#       ↓       ↓
#        collect
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
# x | x % 2 == 0 | Result
# --|------------|--------
# 0 |    True    |    0
# 1 |    False   |  "Odd"
# 2 |    True    |    2
# 3 |    False   |  "Odd"
# 4 |    True    |    4
# 5 |    False   |  "Odd"
# 6 |    True    |    6
# 7 |    False   |  "Odd"
# 8 |    True    |    8
# 9 |    False   |  "Odd"
#
# Result:
#
# [0, 'Odd', 2, 'Odd', 4, 'Odd', 6, 'Odd', 8, 'Odd']
#
# IMPORTANT:
#
# This does NOT multiply even numbers by -1.
# It keeps even numbers unchanged.
#
# The general conditional-comprehension pattern is:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# Here:
#
#     x       → if even, keep x
#     "Odd"   → if odd, use "Odd"

print(result)
```

```python
# ============================================================
# CONDITIONAL LIST COMPREHENSION
# ============================================================
#
# A conditional list comprehension lets us choose WHICH VALUE
# to put into the new list for each item.
#
# It follows this pattern:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# In simple English:
#
#     "For each item, check the condition.
#      If True, use this value.
#      If False, use that value."
#
# Unlike a filtering comprehension, no items are removed here.
# Every item produces one result.
#
# ============================================================
# EXAMPLE
# ============================================================

numbers = range(10)

# For each number:
#
#     even → "Even"
#     odd  → "Odd"
#
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]

# Conceptual flow:
#
#     numbers
#        ↓
#     take x
#        ↓
#     x % 2 == 0 ?
#      ↙         ↘
#    True       False
#     ↓           ↓
#  "Even"      "Odd"
#      \         /
#       ↓       ↓
#        collect
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
# x | condition | Result
# --|-----------|--------
# 0 |   True    | "Even"
# 1 |   False   | "Odd"
# 2 |   True    | "Even"
# 3 |   False   | "Odd"
# 4 |   True    | "Even"
# 5 |   False   | "Odd"
# 6 |   True    | "Even"
# 7 |   False   | "Odd"
# 8 |   True    | "Even"
# 9 |   False   | "Odd"
#
# Result:
#
# ["Even", "Odd", "Even", "Odd", "Even",
#  "Odd", "Even", "Odd", "Even", "Odd"]
#
# ============================================================
# FILTERING vs CONDITIONAL
# ============================================================
#
# Filtering:
#
#     [x for x in numbers if x % 2 == 0]
#
# → keeps only even numbers.
#
# Conditional:
#
#     ["Even" if x % 2 == 0 else "Odd" for x in numbers]
#
# → keeps EVERY item but chooses a result for each one.
#
# CORE IDEA:
#
#     condition → choose value → collect
#
# ============================================================

print(result)
```

```python
# ============================================================
# CONDITIONAL LIST COMPREHENSION
# ============================================================
#
# This comprehension processes EVERY number in 'numbers'.
#
# Pattern:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# Here:
#
#     x if x % 2 == 0 else -x
#
# means:
#
#     If x is even → keep x unchanged
#     If x is odd  → change x to -x
#
# IMPORTANT:
# It does NOT remove the odd numbers.
# Every number produces exactly one result.
#
# Example:
#
#     numbers
#     [0, 1, 2, 3, 4, 5]
#          ↓
#       check x
#          ↓
#         even?
#      ↙       ↘
#    True     False
#      ↓         ↓
#     x         -x
#      \         /
#       ↓       ↓
#        collect
#           ↓
#     [0, -1, 2, -3, 4, -5]
#
# ------------------------------------------------------------
# ITERATION TRACKING
# ------------------------------------------------------------
#
# Step | x | x % 2 == 0 | Result
# -----|---|------------|-------
#  1   | 0 | True       |  0
#  2   | 1 | False      | -1
#  3   | 2 | True       |  2
#  4   | 3 | False      | -3
#  5   | 4 | True       |  4
#  6   | 5 | False      | -5
#
# ------------------------------------------------------------

numbers = range(10)

result = [x if x % 2 == 0 else -x for x in numbers]

print(result)
```

```python
# ============================================================
# DATA SCIENCE EXAMPLE — FILTERING INVALID VALUES
# ============================================================
#
# Imagine we imported a CSV file and some ages are invalid.
#
# Before training a model, we may want to remove
# impossible or invalid values.
#
# This is a FILTERING list comprehension.
#
# Pattern:
#
#     [item for item in iterable if condition]
#
# It means:
#
#     "For each age, keep it only if age >= 0."
#
# Unlike a conditional expression, this comprehension
# can REMOVE items from the resulting list.
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
# ages
# [18, 22, -3, 25, -1, 40]
#          ↓
#       take age
#          ↓
#      age >= 0 ?
#       ↙      ↘
#    True     False
#      ↓         ↓
#    keep      discard
#      \         /
#       ↓       ↓
#        collect
#           ↓
# [18, 22, 25, 40]
#
# ------------------------------------------------------------
# ITERATION TRACKING
# ------------------------------------------------------------
#
# Step | age | age >= 0 | Action
# -----|-----|----------|---------
#  1   | 18  | True     | keep 18
#  2   | 22  | True     | keep 22
#  3   | -3  | False    | discard
#  4   | 25  | True     | keep 25
#  5   | -1  | False    | discard
#  6   | 40  | True     | keep 40
#
# Result:
#
#     [18, 22, 25, 40]
#
# ------------------------------------------------------------

ages = [18, 22, -3, 25, -1, 40]

valid_ages = [age for age in ages if age >= 0]

print(valid_ages)
```

```python
# ============================================================
# LIST COMPREHENSION — TRANSFORMING STRINGS
# ============================================================
#
# This comprehension applies .lower() to EVERY name.
#
# Pattern:
#
#     [expression for item in iterable]
#
# Here:
#
#     x.lower()
#
# converts each string to lowercase.
#
# IMPORTANT:
# This is a TRANSFORMATION, not a filter.
# No names are removed. Each name produces one new result.
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
# names
# ["Ali", "hasan", "hussein"]
#          ↓
#       take x
#          ↓
#       x.lower()
#          ↓
#   "ali", "hasan", "hussein"
#          ↓
#       collect
#          ↓
# ["ali", "hasan", "hussein"]
#
# ------------------------------------------------------------
# ITERATION TRACKING
# ------------------------------------------------------------
#
# Step | x        | x.lower() | Result
# -----|----------|-----------|-------------------
#  1   | "Ali"    | "ali"     | "ali"
#  2   | "hasan"  | "hasan"   | "hasan"
#  3   | "hussein"| "hussein" | "hussein"
#
# ------------------------------------------------------------

names = ["Ali", "hasan", "hussein"]

result = [x.lower() for x in names]

print(result)
```

```python
# ============================================================
# DATA SCIENCE EXAMPLE — TEXT NORMALIZATION
# ============================================================
#
# Text data is often inconsistent:
#
#     "Male"
#     "MALE"
#     "male"
#
# These are different strings to Python.
#
# Before encoding categorical variables, we can normalize
# the text so that different letter cases become consistent.
#
#     "Male"  → "male"
#     "MALE"  → "male"
#     "male"  → "male"
#
# This same idea is commonly used for:
#
# • Product names
# • Country names
# • User input
# • Categories
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
# gender
# ["Male", "MALE", "male"]
#          ↓
#       take g
#          ↓
#       g.lower()
#          ↓
#        "male"
#          ↓
#        collect
#          ↓
# ["male", "male", "male"]
#
# ------------------------------------------------------------
# ITERATION TRACKING
# ------------------------------------------------------------
#
# Step | g        | g.lower() | Result
# -----|----------|-----------|----------------
#  1   | "Male"   | "male"    | "male"
#  2   | "MALE"   | "male"    | "male"
#  3   | "male"   | "male"    | "male"
#
# ------------------------------------------------------------
# KEY IDEA
# ------------------------------------------------------------
#
# [expression for item in iterable]
#
# Here:
#
#     expression → g.lower()
#     item       → g
#     iterable   → gender
#
# Every item is transformed, so no items are removed.
# ============================================================

gender = ["Male", "MALE", "male"]

normalized = [g.lower() for g in gender]

print(normalized)
```

# Python List Comprehensions – A Data Science & ML Notebook


```python
# ============================================================
# PYTHON LIST COMPREHENSIONS
# DATA SCIENCE & MACHINE LEARNING
# ============================================================
#
# List comprehensions provide a concise way to create a NEW
# list by processing items from an existing iterable.
#
# In data science, we often create new lists by:
#
# • Transforming values
# • Normalizing text
# • Extracting features
# • Filtering data
# • Cleaning data
#
# ------------------------------------------------------------
# 1. WHY DATA SCIENTISTS USE LIST COMPREHENSIONS
# ------------------------------------------------------------
#
# A traditional loop is often used to build a new list.
# It works well, but requires several lines:
#
#     create empty list
#           ↓
#       get an item
#           ↓
#       process item
#           ↓
#       append result
#           ↓
#        repeat
#
# A list comprehension expresses the same idea more compactly:
#
#     get item → process item → collect result
#
# Main advantages:
#
# • More concise — less code for simple operations.
# • Readable   — the transformation is visible in one line.
# • Convenient — useful for everyday data preparation.
#
# IMPORTANT:
# A list comprehension is NOT automatically better than a
# traditional loop. For complex logic, a normal loop is often
# easier to read and maintain.
#
# ------------------------------------------------------------
# BASIC PATTERN
# ------------------------------------------------------------
#
#     [expression for item in iterable]
#
# Simple English:
#
#     "Build a new list. For each item in the iterable,
#      do something to it and collect the result."
#
# ------------------------------------------------------------
# EXAMPLE — SQUARING NUMBERS
# ------------------------------------------------------------
#
# Traditional loop:
#
#     squares = []
#     for x in range(10):
#         squares.append(x**2)
#
# Conceptually:
#
#     range(10)
#         ↓
#      take x
#         ↓
#       x**2
#         ↓
#      append
#         ↓
#     repeat
#
# ============================================================


# ------------------------------------------------------------
# TRADITIONAL LOOP
# ------------------------------------------------------------

squares = []

for x in range(10):
    squares.append(x**2)

print(squares)
```

```python
# ------------------------------------------------------------
# LIST COMPREHENSION — SAME RESULT
# ------------------------------------------------------------
#
# This is the list comprehension version of the loop above.
#
#     [x**2 for x in range(10)]
#        │       │
#        │       └── get each x from range(10)
#        └────────── square x
#
# Conceptual execution:
#
#     range(10)
#         ↓
#       take x
#         ↓
#       x**2
#         ↓
#      collect
#         ↓
#     repeat
#         ↓
# [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
#
# It produces exactly the same values as the traditional loop.
#
# ------------------------------------------------------------

squares_comp = [x**2 for x in range(10)]

print(squares_comp)
```

```python
# ============================================================
# FORGET THE COMPLICATED DEFINITION
# ============================================================
#
# List comprehension = "make a new list from existing data"
#
# The simplest idea is:
#
#     OLD DATA → process each item → NEW LIST
#
#
# ------------------------------------------------------------
# NORMAL WAY
# ------------------------------------------------------------
#
# numbers = [1, 2, 3, 4]
#
# squares = []
#
# for x in numbers:
#     squares.append(x * x)
#
# The loop does this:
#
#     numbers
#        ↓
#      take x
#        ↓
#      x * x
#        ↓
#    append result
#        ↓
#      repeat
#        ↓
#    squares
#
#
# ------------------------------------------------------------
# LIST COMPREHENSION
# ------------------------------------------------------------
#
# The same operation in one line:
#
# squares = [x * x for x in numbers]
#
# Read it like English:
#
#     "Put x * x into the new list,
#      for every x in numbers."
#
#
# ------------------------------------------------------------
# ANOTHER SIMPLE EXAMPLE
# ------------------------------------------------------------
#
# names = ["ali", "reza", "sara"]
#
# upper_names = [name.upper() for name in names]
#
# Think:
#
#     "ali"  → .upper() → "ALI"
#     "reza" → .upper() → "REZA"
#     "sara" → .upper() → "SARA"
#
# Result:
#
#     ["ALI", "REZA", "SARA"]
#
#
# ------------------------------------------------------------
# THE MENTAL MODEL
# ------------------------------------------------------------
#
# Don't think of list comprehension as complicated
# Python magic.
#
# Just remember:
#
#     [WHAT_I_WANT for EACH_ITEM in OLD_DATA]
#
#
# For example:
#
#     [x + 10 for x in numbers]
#
# Means:
#
#     "Take each x from numbers,
#      add 10,
#      and put the result into a new list."
#
#
# Value flow:
#
#     numbers
#        ↓
#     take each x
#        ↓
#      x + 10
#        ↓
#     collect result
#        ↓
#     NEW LIST
#
#
# ------------------------------------------------------------
# THE CORE PATTERN
# ------------------------------------------------------------
#
#     [expression for item in iterable]
#
# Think:
#
#     OLD DATA
#        ↓
#     take one item
#        ↓
#     do something to it
#        ↓
#     collect the result
#        ↓
#      repeat
#        ↓
#     NEW LIST
#
# That's basically it.
# ============================================================
```

```python
# ============================================================
# 2. TEXT NORMALIZATION — CATEGORICAL VARIABLES
# ============================================================
#
# WHY DATA SCIENTISTS USE THIS
# ------------------------------------------------------------
#
# Text data is often inconsistent.
#
# For example, these three values mean the same thing:
#
#     "Male"
#     "MALE"
#     "male"
#
# But Python treats them as three different strings.
#
#     "Male" != "MALE" != "male"
#
# If we use these values as categorical features, inconsistent
# text can create unnecessary categories.
#
# A common preprocessing step is to normalize the text before
# encoding or analyzing categorical variables.
#
# Common normalization operations include:
#
#     .lower()  → convert text to lowercase
#     .strip()  → remove extra whitespace
#
# The same idea is useful for:
#
# • Gender / category labels
# • Product names
# • Country names
# • User input
# • Survey responses
# • Text-based features
#
# IMPORTANT:
# Normalization should match the meaning of the data.
# Do not blindly modify text when capitalization or whitespace
# carries meaningful information.
#
# ============================================================


# ------------------------------------------------------------
# EXAMPLE 1 — CONVERT TEXT TO LOWERCASE
# ------------------------------------------------------------

gender = ["Male", "MALE", "male"]

normalized = [g.lower() for g in gender]

print(normalized)

# Result:
#
# ["male", "male", "male"]
#
# Value flow:
#
#     "Male" → .lower() → "male"
#     "MALE" → .lower() → "male"
#     "male" → .lower() → "male"
#
# Every item is transformed.
# Nothing is removed.
#
# Pattern:
#
#     [expression for item in iterable]
#
# Here:
#
#     expression → g.lower()
#     item       → g
#     iterable   → gender
```

```python
# ------------------------------------------------------------
# REMOVE WHITESPACE + CONVERT TO LOWERCASE
# ------------------------------------------------------------
#
# Real-world text can contain unwanted spaces, tabs, or
# newline characters.
#
# .strip() removes whitespace from the beginning and end.
#
# .lower() converts the remaining text to lowercase.
#
# We can combine both operations:
#
#     name.strip().lower()
#
# The operations happen from LEFT to RIGHT:
#
#     name
#       ↓
#     .strip()
#       ↓
#     .lower()
#       ↓
#     cleaned name


messy_names = ["  Alice ", "Bob\t", "Charlie\n", "  David  "]

clean_names = [name.strip().lower() for name in messy_names]

print(clean_names)

# Result:
#
# ["alice", "bob", "charlie", "david"]
#
#
# Value flow:
#
#     "  Alice "  → strip() → "Alice"   → lower() → "alice"
#     "Bob\t"     → strip() → "Bob"     → lower() → "bob"
#     "Charlie\n" → strip() → "Charlie" → lower() → "charlie"
#     "  David  " → strip() → "David"   → lower() → "david"
#
#
# KEY IDEA:
#
#     [name.strip().lower() for name in messy_names]
#
# means:
#
#     "Take each name, remove surrounding whitespace,
#      convert it to lowercase, and collect the results
#      into a new list."
# ============================================================
```

```python
# ============================================================
# 3. HANDLING MISSING VALUES
# ============================================================
#
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Real-world datasets often contain missing values.
#
# Common examples:
#
#     None
#     NaN
#     empty strings ("")
#
# Example:
#
#     age = [25, None, 30, None, 22]
#
# Here, None means that the age is missing.
#
# Missing values can cause problems during data analysis,
# preprocessing, or machine learning.
#
# Common strategies include:
#
# • Replace missing values with a suitable value
#   (for example, 0, mean, or median)
#
# • Remove records containing missing values
#
# List comprehensions provide a simple way to demonstrate
# these operations.
#
# IMPORTANT:
# In real ML projects, replacing missing values with 0 is
# NOT always appropriate. The replacement should make sense
# for the meaning and distribution of the feature.
#
# ============================================================


age = [25, None, 30, None, 22]


# ------------------------------------------------------------
# METHOD 1 — REPLACE MISSING VALUES
# ------------------------------------------------------------
#
# Pattern:
#
#     [value_if_true if condition else replacement
#      for item in iterable]
#
# Here:
#
#     a if a is not None else 0
#
# means:
#
#     If a contains a value → keep a
#     If a is None           → replace it with 0
#
# Every item produces one result.
# Therefore, the length of the list stays the same.
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     age
#      ↓
#    take a
#      ↓
#   a is not None?
#     ↙          ↘
#   True        False
#     ↓            ↓
#   keep a       use 0
#     \            /
#      ↓          ↓
#       collect
#          ↓
#   [25, 0, 30, 0, 22]
#
# ------------------------------------------------------------

filled_age = [a if a is not None else 0 for a in age]

print(filled_age)

# Result:
#
#     [25, 0, 30, 0, 22]


# ------------------------------------------------------------
# METHOD 2 — REMOVE MISSING VALUES
# ------------------------------------------------------------
#
# Pattern:
#
#     [item for item in iterable if condition]
#
# Here:
#
#     a for a in age if a is not None
#
# means:
#
#     "Keep a only when a is not None."
#
# Unlike Method 1, this is FILTERING.
#
# Missing values are discarded.
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     age
#      ↓
#    take a
#      ↓
#   a is not None?
#     ↙          ↘
#   True        False
#     ↓            ↓
#   keep a       discard
#     \            /
#      ↓          ↓
#       collect
#          ↓
#      [25, 30, 22]
#
# ------------------------------------------------------------

clean_age = [a for a in age if a is not None]

print(clean_age)

# Result:
#
#     [25, 30, 22]
#
#
# ============================================================
# KEY DIFFERENCE
# ============================================================
#
# REPLACE:
#
#     [a if a is not None else 0 for a in age]
#
#     → every item stays
#     → missing values are replaced
#
#
# FILTER:
#
#     [a for a in age if a is not None]
#
#     → missing items are removed
#
#
# Remember:
#
#     CONDITIONAL EXPRESSION
#     → choose a value
#
#     FILTERING
#     → keep or remove an item
# ============================================================
```

```python
# ============================================================
# 4. EXTRACTING FEATURES FROM DATES OR STRINGS
# ============================================================
#
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Real-world datasets often store dates, phone numbers, emails,
# IDs, and other structured information as strings.
#
# Example:
#
#     "2023-01-15"
#
# This single string contains three useful pieces of information:
#
#     Year  → 2023
#     Month → 01
#     Day   → 15
#
# We can extract these parts and use them as separate features.
#
# This is a common feature-engineering technique in:
#
# • Time-series analysis
# • Customer segmentation
# • Churn prediction
# • Sales analysis
# • Demand forecasting
#
# List comprehensions are useful when the extraction rule is
# simple and needs to be applied to every value.
#
# ============================================================


dates = ["2023-01-15", "2024-06-20", "2025-12-01"]


# ------------------------------------------------------------
# EXAMPLE 1 — EXTRACT THE YEAR
# ------------------------------------------------------------
#
# .split("-") breaks each date into three parts:
#
#     "2023-01-15"
#          ↓
#     ["2023", "01", "15"]
#
# Index [0] gives the year.
#
#     d.split("-")[0]
#             ↑
#           year
#
# ------------------------------------------------------------

years = [d.split("-")[0] for d in dates]

print(years)

# Result:
#
#     ["2023", "2024", "2025"]
#
# Notice:
# The results are still STRINGS because split() returns strings.


# ------------------------------------------------------------
# EXAMPLE 2 — CONVERT THE YEAR TO INTEGER
# ------------------------------------------------------------
#
# If we need the years as numbers, use int().
#
#     int(d.split("-")[0])
#     ↑
#     convert "2023" → 2023
#
# ------------------------------------------------------------

years_int = [int(d.split("-")[0]) for d in dates]

print(years_int)

# Result:
#
#     [2023, 2024, 2025]


# ------------------------------------------------------------
# EXAMPLE 3 — EXTRACT MONTH AND DAY
# ------------------------------------------------------------
#
# Index positions:
#
#     "2023-01-15"
#        │  │  │
#        │  │  └── [2] → day
#        │  └───── [1] → month
#        └──────── [0] → year
#
# We can create a tuple containing month and day:
#
#     (month, day)
#
# ------------------------------------------------------------

month_day = [
    (int(d.split("-")[1]), int(d.split("-")[2]))
    for d in dates
]

print(month_day)

# Result:
#
#     [(1, 15), (6, 20), (12, 1)]
#
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     "2023-01-15"
#           ↓
#      split("-")
#           ↓
#     ["2023", "01", "15"]
#           ↓
#       [0] [1] [2]
#        ↓   ↓   ↓
#      year month day
#
# The comprehension repeats this process for every date.
#
#
# ============================================================
# KEY IDEA
# ============================================================
#
# List comprehensions can apply the same extraction or
# transformation to every value in a dataset.
#
#     [expression for item in iterable]
#
# Here:
#
#     iterable  → dates
#     item      → d
#     expression → d.split("-")[0]
#
# In simple English:
#
#     "For every date, extract the part I need
#      and collect the results into a new list."
# ============================================================
```

```python
# ============================================================
# EXTRACT YEAR, MONTH, AND DAY
# ============================================================
#
# Example:
#
#     "2023-01-15"
#
# The date contains three parts separated by "-":
#
#     "2023-01-15"
#       │    │  │
#       │    │  └── day
#       │    └───── month
#       └────────── year
#
# ------------------------------------------------------------
# HOW split("-") WORKS
# ------------------------------------------------------------
#
#     "2023-01-15"
#            ↓
#        split("-")
#            ↓
#     ["2023", "01", "15"]
#        │      │     │
#       [0]    [1]   [2]
#        │      │     │
#      year   month  day
#
# ============================================================

dates = ["2023-01-15", "2024-06-20", "2025-12-01"]


# ------------------------------------------------------------
# EXTRACT YEAR
# ------------------------------------------------------------
#
#     d.split("-")[0]
#             │     │
#             │     └── take the first part
#             └──────── split the date
#
# Diagram:
#
#     "2023-01-15"
#           ↓
#       split("-")
#           ↓
#     ["2023", "01", "15"]
#        ↓
#       [0] = "2023"
#        ↓
#      int()
#        ↓
#       2023
#
# ------------------------------------------------------------

years = [int(d.split("-")[0]) for d in dates]

print(years)

# Result:
# [2023, 2024, 2025]


# ------------------------------------------------------------
# EXTRACT MONTH
# ------------------------------------------------------------
#
# Diagram:
#
#     "2023-01-15"
#           ↓
#       split("-")
#           ↓
#     ["2023", "01", "15"]
#                 ↓
#             [1] = "01"
#                 ↓
#               int()
#                 ↓
#                 1
#
# ------------------------------------------------------------

months = [int(d.split("-")[1]) for d in dates]

print(months)

# Result:
# [1, 6, 12]


# ------------------------------------------------------------
# EXTRACT DAY
# ------------------------------------------------------------
#
# Diagram:
#
#     "2023-01-15"
#           ↓
#       split("-")
#           ↓
#       ["2023", "01", "15"]
#                        ↓
#                       [2] = "15"
#                            ↓
#                          int()
#                           ↓
#                           15
#
# ------------------------------------------------------------

days = [int(d.split("-")[2]) for d in dates]

print(days)

# Result:
# [15, 20, 1]


# ============================================================
# EXTRACT YEAR + MONTH + DAY TOGETHER
# ============================================================
#
# We can extract all three parts during the same iteration.
#
# Diagram:
#
#             dates
#               ↓
#        take one date (d)
#               ↓
#          split("-")
#               ↓
#       ["2023", "01", "15"]
#          ↓       ↓      ↓
#        [0]     [1]    [2]
#          ↓       ↓      ↓
#        year    month   day
#          ↓       ↓      ↓
#        2023      1      15
#           \       |     /
#            \      |    /
#             └─────┴───┘
#                   ↓
#          (2023, 1, 15)
#                   ↓
#               collect
#
# Repeat for every date.
#
# ============================================================

date_parts = [
    (
        int(d.split("-")[0]),  # year
        int(d.split("-")[1]),  # month
        int(d.split("-")[2])   # day
    )
    for d in dates
]

print(date_parts)

# Result:
#
# [
#     (2023, 1, 15),
#     (2024, 6, 20),
#     (2025, 12, 1)
# ]


# ============================================================
# COMPLETE VALUE FLOW
# ============================================================
#
# dates
#   ↓
# "2023-01-15"
#   ↓
# split("-")
#   ↓
# ["2023", "01", "15"]
#   ↓       ↓      ↓
# year    month    day
#   ↓       ↓      ↓
# 2023      1      15
#   ↓       ↓      ↓
#       collect
#          ↓
#    (2023, 1, 15)
#
# Then the same process happens for:
#
#     "2024-06-20" → (2024, 6, 20)
#     "2025-12-01" → (2025, 12, 1)
#
#
# ============================================================
# KEY IDEA
# ============================================================
#
#     [expression for item in iterable]
#
# Here:
#
#     iterable  → dates
#     item      → d
#     expression → extract year/month/day
#
# In simple English:
#
#     "For every date, split it into its parts,
#      convert the parts to integers,
#      and collect them."
# ============================================================
```

```python
# ============================================================
# 5. FILTERING OUTLIERS
# ============================================================
#
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Real-world datasets can contain extreme values.
#
# Example:
#
#     prices = [120, 150, 130, 10000, 140, 110, 9000, 160]
#
# Most prices are between 100 and 160, but 9000 and 10000
# are much larger than the rest.
#
# Such extreme values can strongly affect statistics such as:
#
# • Mean
# • Variance
# • Standard deviation
#
# They can also influence some machine-learning models.
#
# Common strategies include:
#
# • Remove extreme values
# • Replace extreme values with a boundary value
#   (capping / winsorization)
#
# A list comprehension can express both operations clearly.
#
# IMPORTANT:
# An outlier is not automatically a bad or invalid value.
# In real projects, the threshold should be based on domain
# knowledge or a statistical method such as IQR or z-score.
#
# ============================================================


prices = [120, 150, 130, 10000, 140, 110, 9000, 160]


# ------------------------------------------------------------
# METHOD 1 — REMOVE OUTLIERS
# ------------------------------------------------------------
#
# Keep only prices that are 500 or less.
#
# Pattern:
#
#     [item for item in iterable if condition]
#
# This is FILTERING.
#
#     p <= 500 ?
#       ↙       ↘
#    True      False
#      ↓          ↓
#    keep       remove
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     prices
#        ↓
#     take p
#        ↓
#    p <= 500 ?
#      ↙     ↘
#   True    False
#     ↓        ↓
#   keep    discard
#     \        /
#      ↓      ↓
#      collect
#         ↓
# [120, 150, 130, 140, 110, 160]
#
# ------------------------------------------------------------

filtered_prices = [p for p in prices if p <= 500]

print(filtered_prices)

# Result:
#
#     [120, 150, 130, 140, 110, 160]
#
# The values 10000 and 9000 were removed.


# ------------------------------------------------------------
# METHOD 2 — CAP OUTLIERS
# ------------------------------------------------------------
#
# Instead of removing extreme values, we can replace them
# with a maximum allowed value of 500.
#
# Pattern:
#
#     [value_if_true if condition else replacement
#      for item in iterable]
#
# Here:
#
#     p if p <= 500 else 500
#
# means:
#
#     If p <= 500 → keep p
#     Otherwise   → use 500
#
# Every item produces a result.
# Therefore, no items are removed.
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     prices
#        ↓
#     take p
#        ↓
#    p <= 500 ?
#      ↙     ↘
#   True    False
#     ↓        ↓
#   keep p    use 500
#      \       /
#       ↓     ↓
#       collect
#          ↓
# [120, 150, 130, 500, 140, 110, 500, 160]
#
# ------------------------------------------------------------

capped_prices = [p if p <= 500 else 500 for p in prices]

print(capped_prices)

# Result:
#
#     [120, 150, 130, 500, 140, 110, 500, 160]


# ============================================================
# KEY DIFFERENCE
# ============================================================
#
# FILTERING:
#
#     [p for p in prices if p <= 500]
#
#     → keep valid values
#     → remove outliers
#     → list becomes shorter
#
#
# CAPPING:
#
#     [p if p <= 500 else 500 for p in prices]
#
#     → keep normal values
#     → replace outliers with 500
#     → list keeps the same length
#
#
# Remember:
#
#     FILTER
#     → "Should I KEEP this item?"
#
#     CONDITIONAL EXPRESSION
#     → "WHAT VALUE should I put in the new list?"
# ============================================================
```

```python
# ============================================================
# 6. CREATING NEW NUMERICAL FEATURES
# ============================================================
#
# Feature engineering often means creating a new feature from
# an existing feature.
#
# Common examples:
#
#     Original feature → New feature
#
#     values          → log(values)
#     counts          → sqrt(counts)
#     text            → text length
#
# The goal is to create a representation that may be more
# useful for analysis or machine-learning models.
#
# A list comprehension is useful when the same transformation
# needs to be applied to every item.
#
# ============================================================
# 1. LOG TRANSFORMATION
# ============================================================
#
# A logarithmic transformation can reduce the effect of very
# large values and is often useful for right-skewed data.
#
# Important:
#
#     math.log(0)
#
# is undefined, so we handle zero separately.
#
# Here:
#
#     v > 0 → calculate log(v)
#     v <= 0 → use 0
#
# ============================================================

import math

values = [1, 10, 100, 0, 1000]

log_values = [math.log(v) if v > 0 else 0 for v in values]

print([round(x, 2) for x in log_values])


# ============================================================
# HOW THE COMPREHENSION WORKS
# ============================================================
#
#     values
#       ↓
#     take one value
#       ↓
#     v > 0?
#       ↓
#    ┌───────┐
#   YES      NO
#    ↓        ↓
# log(v)      0
#    │        │
#    └───┬────┘
#        ↓
#   add to new list
#        ↓
#    next value
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
#     Step | v    | v > 0? | Result
#     -----|------|--------|--------
#       1  |   1  | True   | log(1)     = 0.00
#       2  |  10  | True   | log(10)    = 2.30
#       3  | 100  | True   | log(100)   = 4.61
#       4  |   0  | False  | 0
#       5  | 1000 | True   | log(1000)  = 6.91
#
# Result:
#
#     log_values
#     → [0.00, 2.30, 4.61, 0, 6.91]
#
# ============================================================
# IMPORTANT: THIS IS NOT JUST A FILTER
# ============================================================
#
# The "if ... else" here is a CONDITIONAL EXPRESSION:
#
#     value_if_true if condition else value_if_false
#
# It produces a result for EVERY input item.
#
#     v > 0 → calculate log(v)
#     v <= 0 → produce 0
#
# Therefore, the output still contains 5 values.
#
# Compare this with a FILTER:
#
#     [math.log(v) for v in values if v > 0]
#
# This would REMOVE the zero instead of replacing it.
#
# ============================================================
# 2. CREATING A TEXT-LENGTH FEATURE
# ============================================================
#
# Text can also be transformed into a numerical feature.
#
# Here, each description is converted into its number of
# characters using len().
#
#     string → len(string) → integer
#
# ============================================================

```

```python
descriptions = ["short", "medium length", "very long description"]

desc_lengths = [len(d) for d in descriptions]

print(desc_lengths)


# ============================================================
# HOW THE COMPREHENSION WORKS
# ============================================================
#
# Read it as:
#
#     For each description "d" in descriptions,
#     calculate len(d), then put the result in a new list.
#
# Flow:
#
#     descriptions
#          ↓
#     take one description
#          ↓
#       d = item
#          ↓
#       len(d)
#          ↓
#     store result
#          ↓
#      next item
#          ↓
#        repeat
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
#     Step | d                       | len(d)
#     -----|-------------------------|--------
#       1  | "short"                 |    5
#       2  | "medium length"         |   13
#       3  | "very long description" |   21
#
# Result:
#
#     desc_lengths = [5, 13, 21]
#
# ============================================================
# MEMORY MODEL
# ============================================================
#
# The comprehension creates a NEW list.
#
#     descriptions
#          │
#          │  transform each string
#          ↓
#     desc_lengths
#
# The original list is unchanged:
#
#     descriptions
#     → ["short", "medium length", "very long description"]
#
# The new list contains the transformed values:
#
#     desc_lengths
#     → [5, 13, 21]
#
# During iteration, "d" is one temporary loop variable:
#
#     d = "short"
#         ↓
#     d = "medium length"
#         ↓
#     d = "very long description"
#
# ============================================================
# WHAT DOES len(d) DO?
# ============================================================
#
#     len(d)
#
# returns the number of characters in the string.
#
# Examples:
#
#     len("short")                 → 5
#     len("medium length")         → 13
#     len("very long description") → 21
#
# Spaces are characters too:
#
#     "medium length"
#              ↑
#          one space
#
# ============================================================
# EQUIVALENT FOR LOOP
# ============================================================
#
# The list comprehension:
#
#     desc_lengths = [len(d) for d in descriptions]
#
# is equivalent to:
#
#     desc_lengths = []
#
#     for d in descriptions:
#         desc_lengths.append(len(d))
#
# Both produce:
#
#     [5, 13, 21]
#
# The comprehension simply expresses the same process more
# compactly.
#
# ============================================================
# GENERAL TRANSFORMATION PATTERN
# ============================================================
#
#     new_list = [expression for item in collection]
#
# Here:
#
#     expression = len(d)
#     item       = d
#     collection = descriptions
#
# Therefore:
#
#     desc_lengths = [len(d) for d in descriptions]
#
# ============================================================
# FEATURE-ENGINEERING VIEW
# ============================================================
#
# Original feature:
#
#     descriptions
#          ↓
#     ["short", "medium length", "very long description"]
#
#          ↓  transform
#
# New numerical feature:
#
#     desc_lengths
#          ↓
#     [5, 13, 21]
#
# The text feature has been converted into a numerical feature
# that can be used in further analysis or modeling.
#
# ============================================================
# TRANSFORMATION VS FILTERING
# ============================================================
#
# TRANSFORMATION:
#
#     [len(d) for d in descriptions]
#
# Produces one output for every input:
#
#     3 descriptions → 3 lengths
#
#
# FILTERING:
#
#     [d for d in descriptions if len(d) > 10]
#
# Keeps only descriptions satisfying the condition:
#
#     description → check condition → KEEP or DISCARD
#
#
# CONDITIONAL TRANSFORMATION:
#
#     ["long" if len(d) > 10 else "short"
#      for d in descriptions]
#
# Produces one output for every input, but the result depends
# on a condition.
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you hear:
#
#     "Apply the same transformation to every item and create
#      a new list."
#
# Think:
#
#     new_list = [transform(item) for item in collection]
#
# For this example:
#
#     descriptions → len() → desc_lengths
#
# ============================================================
# CORE IDEA
# ============================================================
#
#     OLD LIST
#        ↓
#     take one item
#        ↓
#     transform item
#        ↓
#     store result
#        ↓
#     repeat
#        ↓
#     NEW LIST
#
#     descriptions
#          ↓
#       len(d)
#          ↓
#     desc_lengths
#
# Python feeling:
#
#     "For every description, calculate its length and collect
#      those lengths into a new list."
#
# ============================================================
```

```python
## 7. COMBINING MULTIPLE FEATURES (PAIRWISE OPERATIONS)

# ============================================================
# WHY DATA SCIENTISTS USE THIS
# ============================================================
#
# Sometimes two or more features must be combined to create
# a new feature.
#
# Example:
#
#     heights = [170, 160, 180]   # cm
#     weights = [70, 60, 80]      # kg
#
# Each height must be paired with the corresponding weight:
#
#     170 ↔ 70
#     160 ↔ 60
#     180 ↔ 80
#
# zip() creates these pairs.
#
# List comprehensions can then perform the same calculation
# for every pair.
#
# General pattern:
#
#     [operation(a, b) for a, b in zip(list_A, list_B)]
#
# ============================================================


heights = [170, 160, 180]  # cm
weights = [70, 60, 80]     # kg


# ============================================================
# 1. CALCULATING BMI
# ============================================================
#
# BMI formula:
#
#     BMI = weight / height²
#
# Height must be expressed in meters.
#
# Therefore:
#
#     height in cm → height / 100 → height in meters
#
# Example:
#
#     170 cm → 1.70 m
#
# The corresponding height and weight are paired using zip():
#
#     zip(heights, weights)
#          ↓
#     (170, 70)
#     (160, 60)
#     (180, 80)
#
# Then tuple unpacking gives:
#
#     h = height
#     w = weight
#
# ============================================================

bmi = [w / ((h / 100) ** 2) for h, w in zip(heights, weights)]

print([round(b, 2) for b in bmi])


# ============================================================
# HOW THE COMPREHENSION WORKS
# ============================================================
#
#     heights ───┐
#                ├──→ zip() → (h, w)
#     weights ───┘
#                         ↓
#                   w / ((h/100)²)
#                         ↓
#                    BMI value
#                         ↓
#                    new list
#                         ↓
#                      repeat
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
#     Step | h (cm) | w (kg) | h/100 | BMI
#     -----|--------|--------|-------|--------
#       1  |  170   |   70   |  1.70 | 24.22
#       2  |  160   |   60   |  1.60 | 23.44
#       3  |  180   |   80   |  1.80 | 24.69
#
# Result:
#
#     bmi → [24.22, 23.44, 24.69]
#
# ============================================================
# MEMORY MODEL
# ============================================================
#
# zip() produces one pair at a time.
#
#     (170, 70)
#         ↓
#     h = 170
#     w = 70
#         ↓
#     calculate BMI
#
#     (160, 60)
#         ↓
#     h = 160
#     w = 60
#         ↓
#     calculate BMI
#
#     (180, 80)
#         ↓
#     h = 180
#     w = 80
#         ↓
#     calculate BMI
#
# The original lists are not changed.
#
# ============================================================
# 2. PRICE PER UNIT
# ============================================================
#
# Another common pairwise operation is calculating a derived
# feature from price and quantity.
#
#     price_per_unit = price / quantity
#
# Example:
#
#     $100 / 2 units → $50 per unit
#
# ============================================================
```

```python
prices = [100, 200, 150]
quantities = [2, 3, 1]

price_per_unit = [p / q for p, q in zip(prices, quantities)]

print([round(x, 2) for x in price_per_unit])


# ============================================================
# HOW IT WORKS
# ============================================================
#
#     prices      quantities
#       ↓              ↓
#     100   ───────→   2
#     200   ───────→   3
#     150   ───────→   1
#
# zip() creates:
#
#     (100, 2)
#     (200, 3)
#     (150, 1)
#
# Then:
#
#     100 / 2 → 50
#     200 / 3 → 66.67
#     150 / 1 → 150
#
# Result:
#
#     price_per_unit → [50.0, 66.67, 150.0]
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
#     Step | p   | q | p / q
#     -----|-----|---|--------
#       1  | 100 | 2 |  50.00
#       2  | 200 | 3 |  66.67
#       3  | 150 | 1 | 150.00
#
# ============================================================
# THE GENERAL PATTERN
# ============================================================
#
# When two features must be combined pair by pair:
#
#     new_feature = [
#         operation(a, b)
#         for a, b in zip(feature_A, feature_B)
#     ]
#
# Conceptually:
#
#     feature_A ──┐
#                 ├──→ zip() → pair → operation → result
#     feature_B ──┘
#
# ============================================================
# WHY zip()?
# ============================================================
#
# zip() matches values by POSITION:
#
#     first  ↔ first
#     second ↔ second
#     third  ↔ third
#
# Example:
#
#     heights = [170, 160, 180]
#     weights = [70, 60, 80]
#
#     170 ↔ 70
#     160 ↔ 60
#     180 ↔ 80
#
# The lists are not matched by value.
# They are matched by position.
#
# ============================================================
# DIFFERENT LIST LENGTHS
# ============================================================
#
# zip() stops when the shortest iterable is exhausted.
#
# Example:
#
#     prices     = [100, 200, 150]
#     quantities = [2, 3]
#
#     zip(prices, quantities)
#
# produces only:
#
#     (100, 2)
#     (200, 3)
#
# The third price has no corresponding quantity, so it is
# not processed.
#
# ============================================================
# IMPORTANT: DATA ALIGNMENT
# ============================================================
#
# Pairwise operations assume that corresponding positions
# actually represent corresponding observations.
#
# For example:
#
#     heights[0] ↔ weights[0]
#     heights[1] ↔ weights[1]
#     heights[2] ↔ weights[2]
#
# If the lists are misaligned, the calculation can produce
# incorrect features even though the Python code is valid.
#
# ============================================================
# TRANSFORMATION VS PAIRWISE TRANSFORMATION
# ============================================================
#
# ONE FEATURE:
#
#     [transform(x) for x in values]
#
# Example:
#
#     [len(text) for text in descriptions]
#
#
# TWO FEATURES:
#
#     [transform(a, b) for a, b in zip(A, B)]
#
# Example:
#
#     [p / q for p, q in zip(prices, quantities)]
#
# The important difference is that each output depends on
# TWO corresponding input values.
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you hear:
#
#     "Combine two lists element by element and create a
#      new feature."
#
# Think:
#
#     [operation(a, b) for a, b in zip(A, B)]
#
# Read the process as:
#
#     A + B
#       ↓
#     zip into pairs
#       ↓
#     unpack pair
#       ↓
#     perform operation
#       ↓
#     collect result
#       ↓
#     repeat
#
# ============================================================
# CORE IDEA
# ============================================================
#
#     FEATURE A ──┐
#                 ├──→ zip() → pair → calculation → NEW FEATURE
#     FEATURE B ──┘
#
# Python feeling:
#
#     "Take corresponding values from multiple features,
#      combine them, and store the result."
#
# ============================================================
```

## WORKING WITH PANDAS DATAFRAMES

```python
# ============================================================
# 8. WORKING WITH PANDAS DATAFRAMES
# ============================================================
#
# List comprehensions can also be used to create new pandas
# DataFrame columns from existing columns.
#
# Common examples:
#
#     • transform strings
#     • extract parts of strings
#     • apply simple conditions
#     • create derived features
#
# For simple column operations, pandas vectorized methods are
# usually preferred because they operate efficiently on the
# whole Series.
#
# List comprehensions are useful when the transformation is
# simple and you want to work with each value directly.
#
# ============================================================


import pandas as pd


# Create a DataFrame.
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["New York", "Los Angeles", "Chicago"]
})


# ============================================================
# 1. CREATE A LOWERCASE CITY FEATURE
# ============================================================
#
# df["city"] is a pandas Series.
#
# The comprehension takes one city at a time:
#
#     "New York"    → "new york"
#     "Los Angeles" → "los angeles"
#     "Chicago"     → "chicago"
#
# The resulting list becomes a new DataFrame column.
#
# ============================================================

df["city_lower"] = [c.lower() for c in df["city"]]


# ============================================================
# HOW IT WORKS
# ============================================================
#
#     df["city"]
#          ↓
#     take one value
#          ↓
#       c.lower()
#          ↓
#     new value
#          ↓
#     collect into list
#          ↓
#     assign to new column
#
# Result:
#
#     city          city_lower
#     -----------   -----------
#     New York      new york
#     Los Angeles   los angeles
#     Chicago       chicago
#
# ============================================================
# 2. EXTRACT THE FIRST LETTER
# ============================================================
#
# Each name is converted into its first character.
#
#     "Alice"   → "A"
#     "Bob"     → "B"
#     "Charlie" → "C"
#
# ============================================================

df["first_letter"] = [n[0] for n in df["name"]]


# ============================================================
# HOW IT WORKS
# ============================================================
#
#     df["name"]
#         ↓
#     take one name
#         ↓
#        n[0]
#         ↓
#     first character
#         ↓
#     new column
#
# Result:
#
#     name      first_letter
#     --------  ------------
#     Alice     A
#     Bob       B
#     Charlie   C
#
# ============================================================
# 3. CREATE A CONDITIONAL FEATURE
# ============================================================
#
# Create a new feature called "senior".
#
#     age >= 30 → 1
#     age < 30  → 0
#
# The conditional expression is:
#
#     1 if a >= 30 else 0
#
# ============================================================

df["senior"] = [1 if a >= 30 else 0 for a in df["age"]]


# ============================================================
# HOW IT WORKS
# ============================================================
#
#     age
#      ↓
#     a >= 30?
#      ↓
#    ┌───────┐
#   YES      NO
#    ↓        ↓
#    1        0
#    │        │
#    └───┬────┘
#        ↓
#      senior
#
# ============================================================
# ITERATION TRACKING
# ============================================================
#
#     Step | name    | age | city        | senior
#     -----|---------|-----|-------------|--------
#       1  | Alice   | 25  | New York    |   0
#       2  | Bob     | 30  | Los Angeles |   1
#       3  | Charlie | 35  | Chicago     |   1
#
# ============================================================
# FINAL DATAFRAME
# ============================================================
#
#     name      age  city          city_lower     first_letter  senior
#     --------  ---  ------------  -------------  ------------  ------
#     Alice      25  New York      new york       A             0
#     Bob        30  Los Angeles   los angeles    B             1
#     Charlie    35  Chicago       chicago        C             1
#
# print(df) displays the complete DataFrame.
#
# ============================================================
# MEMORY MODEL
# ============================================================
#
# The original columns remain unchanged.
#
#     name ──────────────→ first_letter
#       │
#       └── each value → n[0]
#
#     city ──────────────→ city_lower
#       │
#       └── each value → c.lower()
#
#     age ───────────────→ senior
#       │
#       └── each value → condition → 1 or 0
#
# Each comprehension creates a new list, and pandas assigns
# that list as a new column.
#
# ============================================================
# IMPORTANT: DATAFRAME COLUMNS ARE SERIES
# ============================================================
#
# When you write:
#
#     df["age"]
#
# you get a pandas Series containing the age values.
#
# The comprehension then iterates over that Series:
#
#     df["age"]
#         ↓
#     25 → 30 → 35
#
# ============================================================
# LIST COMPREHENSION VS PANDAS METHODS
# ============================================================
#
# For many simple operations, pandas already provides
# vectorized methods.
#
# For example:
#
#     df["city"].str.lower()
#
# can perform the lowercase transformation directly.
#
# Similarly:
#
#     df["name"].str[0]
#
# can extract the first character.
#
# And:
#
#     (df["age"] >= 30).astype(int)
#
# can create the 0/1 feature.
#
# These pandas expressions are often preferable for
# DataFrame work because they operate on the Series directly.
#
# ============================================================
# WHEN LIST COMPREHENSIONS ARE USEFUL
# ============================================================
#
# List comprehensions are particularly convenient when the
# transformation is:
#
#     • simple
#     • value-by-value
#     • awkward to express with existing pandas operations
#
# Example:
#
#     df["label"] = [
#         "adult" if age >= 30 else "young"
#         for age in df["age"]
#     ]
#
# ============================================================
# WHEN NOT TO USE THEM
# ============================================================
#
# For large DataFrames, avoid assuming that a list
# comprehension is automatically faster than pandas.
#
# Performance depends on the operation.
#
# Prefer pandas vectorized operations when a suitable
# vectorized method exists.
#
# Use list comprehensions when they make a particular
# value-by-value transformation clear or when no convenient
# vectorized operation is available.
#
# ============================================================
# GENERAL DATAFRAME PATTERN
# ============================================================
#
#     df["new_column"] = [
#         transform(value)
#         for value in df["old_column"]
#     ]
#
# With a condition:
#
#     df["new_column"] = [
#         A if condition(value) else B
#         for value in df["old_column"]
#     ]
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you hear:
#
#     "Create a new DataFrame column by transforming each
#      value from an existing column."
#
# Think:
#
#     df["new_column"] = [
#         transform(value)
#         for value in df["old_column"]
#     ]
#
# ============================================================
# CORE IDEA
# ============================================================
#
#     EXISTING COLUMN
#            ↓
#      take one value
#            ↓
#        transform
#            ↓
#       collect result
#            ↓
#      NEW COLUMN
#
# A list comprehension performs the value-by-value
# transformation, and pandas stores the resulting list
# as a new DataFrame column.
#
# ============================================================
```

## ONE‑HOT ENCODING (DUMMY VARIABLES)

```python
# ============================================================
# 9. ONE-HOT ENCODING (DUMMY VARIABLES)
# ============================================================
#
# Categorical variables contain labels rather than numerical
# values.
#
# Example:
#
#     ["London", "Paris", "London", "Tokyo", "Paris", "London"]
#
# One-hot encoding converts each category into a binary column:
#
#     London  Paris  Tokyo
#       1       0      0
#       0       1      0
#       1       0      0
#       0       0      1
#       0       1      0
#       1       0      0
#
# Each dummy column represents one category.
#
# ============================================================


import pandas as pd


cities = ["London", "Paris", "London", "Tokyo", "Paris", "London"]


# ============================================================
# WAY 1 — pandas.get_dummies()
# ============================================================
#
# This is the practical pandas approach.
#
#     pd.get_dummies()
#
# automatically creates one binary column for each category.
#
# ============================================================

dummy_df = pd.get_dummies(cities, dtype=int)

print(dummy_df)


# ============================================================
# WAY 2 — SIMPLE LIST COMPREHENSION
# ============================================================
#
# We can create ONE dummy variable for a specific category.
#
# Example:
#
#     Is the city "London"?
#
#     London → 1
#     Paris  → 0
#     London → 1
#     Tokyo  → 0
#     Paris  → 0
#     London → 1
#
# This is a simple list-comprehension pattern:
#
#     [1 if condition else 0 for item in iterable]
#
# ============================================================

is_london = [1 if city == "London" else 0 for city in cities]

print(is_london)


# ============================================================
# HOW THE SIMPLE COMPREHENSION WORKS
# ============================================================
#
# cities
#    ↓
# take one city
#    ↓
# city == "London"?
#    ↓
#   ┌──YES──→ 1
#   │
#   └──NO───→ 0
#    ↓
# store result
#    ↓
# next city
#
#
# Iteration:
#
# Step | city   | city == "London" | Result
# -----|--------|-------------------|-------
#  1   | London | True              | 1
#  2   | Paris  | False             | 0
#  3   | London | True              | 1
#  4   | Tokyo  | False             | 0
#  5   | Paris  | False             | 0
#  6   | London | True              | 1
#
# Result:
#
#     [1, 0, 1, 0, 0, 1]
#
# ============================================================


# ============================================================
# WHY WE SHOW BOTH WAYS
# ============================================================
#
# These examples teach DIFFERENT things.
#
# pd.get_dummies()
#     ↓
# Actual pandas/data-science operation
#     ↓
# Creates all dummy variables automatically
#
#
# List comprehension
#     ↓
# Python programming pattern
#     ↓
# Performs a simple element-by-element transformation
#
# That is much more consistent with the way we are building
# these notes:
#
# The comprehension should demonstrate a SIMPLE Python pattern,
# while pd.get_dummies() demonstrates the ACTUAL pandas/
# data-science operation.
#
# ============================================================


# ============================================================
# IMPORTANT — DON'T FORCE A NESTED COMPREHENSION
# ============================================================
#
# A nested comprehension could technically construct the entire
# dummy matrix, but it would make this example unnecessarily
# complicated.
#
# When pandas already provides get_dummies(), use it for the
# actual one-hot encoding task.
#
# The goal is NOT:
#
#     "Use list comprehensions everywhere."
#
# The goal is:
#
#     "Use a list comprehension when it makes a simple
#      transformation clear and readable."
#
# ============================================================


# ============================================================
# MACHINE LEARNING CONNECTION
# ============================================================
#
# Suppose:
#
#     cities = ["London", "Paris", "Tokyo"]
#
# We should not simply assign:
#
#     London → 1
#     Paris  → 2
#     Tokyo  → 3
#
# because those numbers can imply an artificial ordering.
#
# One-hot encoding represents the categories independently:
#
#     London → [1, 0, 0]
#     Paris  → [0, 1, 0]
#     Tokyo  → [0, 0, 1]
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# Simple list comprehension:
#
#     item
#       ↓
#     condition / transformation
#       ↓
#     result
#       ↓
#     next item
#
#
# One-hot encoding:
#
#     category
#         ↓
#     binary representation
#         ↓
#       0 or 1
#
#
# CORE IDEA:
#
#     List comprehension → simple Python transformation
#
#     pd.get_dummies()   → practical one-hot encoding
#
# ============================================================
```

## PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)

```python
# ============================================================
# 10. PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)
# ============================================================
#
# WHY DATA SCIENTISTS CARE ABOUT THIS
# ============================================================
#
# When processing large amounts of data, execution time can
# become important.
#
# A list comprehension is often faster than an equivalent
# Python for-loop that repeatedly uses append().
#
# IMPORTANT:
#
# A list comprehension is NOT "implemented in C" while a
# for-loop is "implemented in Python."
#
# Both execute Python-level iteration and expression evaluation.
# The comprehension mainly has less Python-level overhead for
# building the list, because the list construction is handled
# more efficiently internally.
#
# Also, performance depends on the operation and data.
# Do not assume that a comprehension is ALWAYS faster.
#
# ============================================================


import time


n = 10_000_000
data = list(range(n))


# ============================================================
# WAY 1 — TRADITIONAL FOR LOOP
# ============================================================
#
# Pattern:
#
#     create empty list
#          ↓
#     get one item
#          ↓
#     calculate result
#          ↓
#     append result
#          ↓
#     repeat
#
# ============================================================

start = time.time()

squares_loop = []

for x in data:
    squares_loop.append(x ** 2)

end = time.time()

print(f"Loop time: {end - start:.4f} seconds")


# ============================================================
# WAY 2 — LIST COMPREHENSION
# ============================================================
#
# Pattern:
#
#     data
#       ↓
#     take x
#       ↓
#     x ** 2
#       ↓
#     store result
#       ↓
#     repeat
#
# ============================================================

start = time.time()

squares_comp = [x ** 2 for x in data]

end = time.time()

print(f"Comprehension time: {end - start:.4f} seconds")


# ============================================================
# WHAT ARE WE ACTUALLY COMPARING?
# ============================================================
#
# Both methods perform the SAME transformation:
#
#     x → x ** 2
#
#
# Traditional loop:
#
#     squares_loop = []
#     for x in data:
#         squares_loop.append(x ** 2)
#
#
# List comprehension:
#
#     squares_comp = [x ** 2 for x in data]
#
#
# The main difference is HOW the result list is constructed.
#
# ============================================================


# ============================================================
# MEMORY / RESULT MODEL
# ============================================================
#
# data
#   ↓
# [0, 1, 2, 3, ..., 9,999,999]
#   │
#   ├──────────────→ for loop ──────────────→ squares_loop
#   │
#   └──────────────→ comprehension ─────────→ squares_comp
#
#
# Both should produce equivalent results:
#
#     squares_loop == squares_comp
#
# Both contain:
#
#     [0, 1, 4, 9, 16, ...]
#
# ============================================================


# ============================================================
# SIMPLE PERFORMANCE CHECK
# ============================================================
#
# We can verify that both approaches produced the same data:
#
# ============================================================

print(squares_loop == squares_comp)


# ============================================================
# PERFORMANCE IDEA
# ============================================================
#
# For a simple transformation such as:
#
#     [x ** 2 for x in data]
#
# a list comprehension is often:
#
#     shorter
#     + easier to read
#     + often somewhat faster
#
# than:
#
#     squares = []
#     for x in data:
#         squares.append(x ** 2)
#
#
# But the performance difference is usually NOT the main
# reason to prefer comprehensions.
#
# Readability should come first.
#
# ============================================================


# ============================================================
# IMPORTANT — DATA SCIENCE CONTEXT
# ============================================================
#
# For large datasets, the bigger performance question is often
# whether we should use Python iteration at all.
#
# For example, with pandas:
#
#     df["squared"] = df["value"] ** 2
#
# pandas/NumPy vectorized operations can operate much more
# efficiently on columnar numerical data than Python-level
# loops or comprehensions.
#
# Therefore, a useful priority is:
#
#     vectorized pandas/NumPy operation
#                 ↓
#          when appropriate
#
#     simple list comprehension
#                 ↓
#          when working with Python
#          collections or awkward operations
#
#     traditional for-loop
#                 ↓
#          when logic is more complex
#          or readability requires multiple steps
#
# ============================================================


# ============================================================
# WHEN TO USE EACH
# ============================================================
#
# LIST COMPREHENSION
# ------------------------------------------------------------
# Best when the operation is:
#
#     simple
#     readable
#     element-by-element
#
# Example:
#
#     squares = [x ** 2 for x in data]
#
#
# FOR LOOP
# ------------------------------------------------------------
# Better when the logic becomes more complicated:
#
#     results = []
#
#     for x in data:
#         if x < 0:
#             ...
#         elif ...:
#             ...
#         else:
#             ...
#
# Forcing complicated logic into one comprehension can make
# the code harder to understand.
#
# ============================================================


# ============================================================
# HOW TO MEASURE PERFORMANCE
# ============================================================
#
# time.time() can be used for a simple timing experiment:
#
#     start = time.time()
#     code_to_test()
#     end = time.time()
#
#     elapsed = end - start
#
# For more reliable Python benchmarking, timeit is generally
# preferred because it performs repeated measurements and is
# designed specifically for timing small pieces of code.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# Do NOT learn:
#
#     "Comprehensions are always faster."
#
# Learn:
#
#     "For a simple list transformation, a comprehension is
#      concise and often performs well."
#
#
# And remember:
#
#     SIMPLE + READABLE
#             ↓
#     list comprehension
#
#     COMPLEX LOGIC
#             ↓
#     for loop
#
#     NUMERICAL DATA / DATAFRAME
#             ↓
#     consider pandas / NumPy vectorization
#
# ============================================================
#
# CORE IDEA:
#
# Choose the clearest appropriate operation first.
# Optimize performance when measurement shows that it matters.
#
# ============================================================
```

## PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)

```python
# ============================================================
# 10. PERFORMANCE COMPARISON (LOOP vs. COMPREHENSION)
# ============================================================
#
# WHY THIS MATTERS
# ============================================================
#
# List comprehensions are often faster than a traditional
# for-loop when both perform the same simple transformation.
#
# But "comprehension = faster" is NOT a rule.
#
# Performance depends on:
#
#     - the operation being performed
#     - how the result is constructed
#     - the amount of Python-level work
#     - the data structure
#     - the libraries/functions being called
#
# The first question should always be:
#
#     "Are these two implementations actually doing
#      the same work?"
#
# ============================================================


import time


n = 10_000_000
data = list(range(n))


# ============================================================
# EXAMPLE 1 — SIMPLE TRANSFORMATION
# ============================================================
#
# Both approaches perform:
#
#     x → x ** 2
#
# ------------------------------------------------------------
# FOR LOOP
# ------------------------------------------------------------

start = time.time()

squares_loop = []

for x in data:
    squares_loop.append(x ** 2)

loop_time = time.time() - start


# ------------------------------------------------------------
# LIST COMPREHENSION
# ------------------------------------------------------------

start = time.time()

squares_comp = [x ** 2 for x in data]

comp_time = time.time() - start


print(f"Loop time:          {loop_time:.4f} seconds")
print(f"Comprehension time: {comp_time:.4f} seconds")


# ============================================================
# COMPARE THE RESULTS
# ============================================================

print(squares_loop == squares_comp)


# ============================================================
# WHAT ARE WE MEASURING?
# ============================================================
#
# The important thing is that the two versions do the same job:
#
#     data
#       ↓
#     x ** 2
#       ↓
#     new list
#
#
# FOR LOOP:
#
#     create list
#          ↓
#     calculate
#          ↓
#     append
#          ↓
#     repeat
#
#
# COMPREHENSION:
#
#     calculate
#          ↓
#     build list
#          ↓
#     repeat
#
#
# The comprehension has less explicit Python-level list-building
# code, so it is often faster for this kind of simple operation.
#
# ============================================================


# ============================================================
# ASPECT 1 — READABILITY
# ============================================================
#
# Performance is NOT the only consideration.
#
# Simple:
#
#     squares = [x ** 2 for x in data]
#
# This is both concise and readable.
#
#
# More complicated:
#
#     results = []
#
#     for x in data:
#         if x > 0:
#             if x % 2 == 0:
#                 value = x ** 2
#                 if value < 1000:
#                     results.append(value)
#
# Trying to compress this into one huge comprehension can make
# the code harder to understand.
#
# Therefore:
#
#     readable comprehension → GOOD
#     complicated comprehension → often BAD
#
# ============================================================


# ============================================================
# ASPECT 2 — MEMORY
# ============================================================
#
# Both of the examples above create a complete list in memory.
#
#     data       → original list
#     squares    → new list
#
# For very large data, the memory cost can matter more than the
# small speed difference between the two approaches.
#
# A list comprehension does NOT automatically save memory.
#
# ============================================================


# ============================================================
# ASPECT 3 — GENERATOR EXPRESSIONS
# ============================================================
#
# If we do not need the entire result list immediately, a
# generator expression can avoid creating the whole list.
#
# List comprehension:
#
#     squares = [x ** 2 for x in data]
#
# Generator expression:
#
#     squares = (x ** 2 for x in data)
#
#
# Conceptually:
#
# LIST COMPREHENSION
#     data → calculate everything → store everything
#
#
# GENERATOR
#     data → calculate one → produce one
#             ↓
#          next value
#             ↓
#          calculate one
#             ↓
#          ...
#
# This is primarily a MEMORY/LAZY-EVALUATION distinction,
# not simply a "faster version" of a comprehension.
#
# ============================================================


# ============================================================
# ASPECT 4 — NUMPY / PANDAS
# ============================================================
#
# In data science, an even more important question is:
#
#     "Should I use Python iteration at all?"
#
# For numerical data, vectorized NumPy/pandas operations can
# often be much faster than either a Python loop or a list
# comprehension.
#
# Example:
#
#     df["squared"] = df["value"] ** 2
#
# Instead of:
#
#     df["squared"] = [x ** 2 for x in df["value"]]
#
# Therefore:
#
#     Python loop
#          ↓
#     list comprehension
#          ↓
#     NumPy / pandas vectorization
#
# is NOT a universal speed ranking.
#
# The operation and implementation matter.
#
# ============================================================


# ============================================================
# EXAMPLE 2 — A COMPREHENSION IS NOT NECESSARILY FASTER
# ============================================================
#
# We should not conclude:
#
#     "A comprehension is always faster."
#
# For example, suppose the work is dominated by a function call.
#
# ------------------------------------------------------------
# FUNCTION
# ------------------------------------------------------------

def calculate(x):
    return x ** 2


# ------------------------------------------------------------
# FOR LOOP
# ------------------------------------------------------------

start = time.time()

results_loop = []

for x in data:
    results_loop.append(calculate(x))

loop_time = time.time() - start


# ------------------------------------------------------------
# LIST COMPREHENSION
# ------------------------------------------------------------

start = time.time()

results_comp = [calculate(x) for x in data]

comp_time = time.time() - start


print(f"Function loop time:          {loop_time:.4f} seconds")
print(f"Function comprehension time: {comp_time:.4f} seconds")


# ============================================================
# WHAT THIS DEMONSTRATES
# ============================================================
#
# The comprehension is NOT guaranteed to be dramatically faster.
#
# Both versions still have to perform:
#
#     Python iteration
#          ↓
#     function call
#          ↓
#     calculate(x)
#          ↓
#     result
#
# The function call itself can become an important part of the
# total execution time.
#
# So the difference between the two ways of constructing the
# result can become relatively small.
#
# The exact timing depends on the Python version, hardware,
# operation, and workload.
#
# ============================================================


# ============================================================
# EXAMPLE 3 — SOMETIMES THE "FASTER" CODE IS NOT THE BETTER CODE
# ============================================================
#
# Suppose we need several operations:
#
#     results = []
#
#     for x in data:
#         value = x ** 2
#
#         if value > 100:
#             value = value / 10
#
#         if value < 1000:
#             results.append(value)
#
#
# This is easy to extend, debug, and inspect.
#
# A compressed comprehension could become difficult to read:
#
#     results = [
#         (x ** 2 / 10 if x ** 2 > 100 else x ** 2)
#         for x in data
#         if (x ** 2 / 10 if x ** 2 > 100 else x ** 2) < 1000
#     ]
#
# Even if the comprehension happened to be faster, it may be
# the worse engineering choice because the logic is duplicated
# and harder to understand.
#
# ============================================================


# ============================================================
# ASPECT 5 — BENCHMARKING MUST BE FAIR
# ============================================================
#
# When comparing performance:
#
#     1. Make both versions produce the same result.
#     2. Make them perform equivalent work.
#     3. Run the test more than once.
#     4. Avoid relying on one timing measurement.
#     5. Use timeit for serious micro-benchmarking.
#
# time.time() is useful for a simple demonstration.
#
# timeit is generally better for controlled Python benchmarks.
#
# ============================================================


# ============================================================
# ASPECT 6 — DON'T OPTIMIZE WITHOUT MEASURING
# ============================================================
#
# A common mistake is:
#
#     "This looks faster, so I'll use it."
#
# Better approach:
#
#     write clear code
#          ↓
#     measure performance
#          ↓
#     identify the actual bottleneck
#          ↓
#     optimize that part
#          ↓
#     measure again
#
# This is especially important in data science because the
# bottleneck may actually be:
#
#     disk I/O
#     database queries
#     network requests
#     pandas operations
#     model training
#     memory usage
#
# rather than the loop itself.
#
# ============================================================


# ============================================================
# PERFORMANCE SUMMARY
# ============================================================
#
# SIMPLE PYTHON TRANSFORMATION
#
#     for loop + append
#            ↓
#     list comprehension
#
# A comprehension is often somewhat faster.
#
#
# COMPLEX PYTHON LOGIC
#
#     for loop
#
# Often preferable because readability and maintainability
# become more important than a small speed difference.
#
#
# NUMERICAL DATA
#
#     consider NumPy / pandas vectorization
#
# because avoiding Python-level iteration can provide a much
# larger performance improvement.
#
#
# LARGE DATA + MEMORY CONSTRAINT
#
#     consider generators / chunking / streaming
#
# rather than automatically creating another huge list.
#
# ============================================================


# ============================================================
# THE IMPORTANT MENTAL MODEL
# ============================================================
#
# DO NOT THINK:
#
#     comprehension = fast
#     loop          = slow
#
#
# THINK:
#
#     simple comprehension
#          → often efficient
#
#     explicit loop
#          → more flexible for complex logic
#
#     vectorization
#          → often better for numerical/dataframe operations
#
#     generator
#          → useful when lazy evaluation saves memory
#
#     benchmark
#          → tells you what is actually faster
#
# ============================================================
#
# CORE IDEA:
#
# A list comprehension is primarily a concise way to express
# a simple transformation or filtering operation.
#
# Its speed advantage is real in many cases, but it is NOT a
# universal rule.
#
# Choose for clarity first.
# Measure when performance matters.
#
# ============================================================
```

```python
# ============================================================
# 11. BEST PRACTICES & READABILITY
# ============================================================
#
# WHY THIS MATTERS
# ============================================================
#
# List comprehensions are useful for simple transformations and
# filtering, but readability should come before compactness.
#
# A good comprehension should be easy to understand at a glance.
#
# DO NOT think:
#
#     "If I can put it in one line, I should."
#
# Instead:
#
#     "If the one-line version is clear, use it."
#
# ============================================================


# ============================================================
# 1. KEEP THE LOGIC SIMPLE
# ============================================================
#
# GOOD:
#
#     squares = [x ** 2 for x in numbers]
#
# The pattern is immediately clear:
#
#     item → transform → result
#
#
# ALSO GOOD:
#
#     even_numbers = [x for x in numbers if x % 2 == 0]
#
# Simple transformation + simple condition.
#
#
# If a comprehension requires several conditions, nested logic,
# or difficult expressions, a regular for-loop may be clearer.
#
# ============================================================


# ============================================================
# 2. USE DESCRIPTIVE VARIABLE NAMES
# ============================================================
#
# LESS CLEAR:
#
#     c = [x.lower() for x in data]
#
#
# CLEARER:
#
#     city_names = ["London", "Paris", "Tokyo"]
#     lowercase_cities = [city.lower() for city in city_names]
#
# The variable names communicate what the data represents.
#
# ============================================================


# ============================================================
# 3. USE A HELPER FUNCTION WHEN THE TRANSFORMATION IS COMPLEX
# ============================================================
#
# Instead of putting several string operations directly inside
# the comprehension, move the logic into a function.
#
# This separates:
#
#     WHAT ARE WE DOING?
#          ↓
#     clean_text()
#
# from:
#
#     HOW ARE WE ITERATING?
#          ↓
#     [clean_text(text) for text in raw]
#
# ============================================================


def clean_text(text):
    return text.strip().lower().replace("-", "_")


raw = ["Hello-World ", "  foo-bar", "baz"]

cleaned = [clean_text(text) for text in raw]

print(cleaned)


# Result:
#
#     ['hello_world', 'foo_bar', 'baz']
#
#
# HOW IT WORKS:
#
# raw
#   ↓
# take one text
#   ↓
# clean_text(text)
#   ↓
# cleaned value
#   ↓
# store in new list
#   ↓
# next text
#
# ============================================================


# ============================================================
# 4. WHEN A REGULAR LOOP IS CLEARER, USE A REGULAR LOOP
# ============================================================
#
# Suppose the processing requires several decisions and steps.
#
# A regular loop can make the sequence of operations much easier
# to understand:
#
#     cleaned = []
#
#     for text in raw:
#         text = text.strip()
#
#         if not text:
#             continue
#
#         text = text.lower()
#
#         if "-" in text:
#             text = text.replace("-", "_")
#
#         cleaned.append(text)
#
# This is longer than a comprehension, but the logic is explicit.
#
# ============================================================
#
# CORE RULE:
#
#     Simple logic
#         → comprehension
#
#     Complex logic
#         → regular loop / helper function
#
# ============================================================


# ============================================================
# 5. NESTING — DON'T FORCE TOO MUCH INTO ONE COMPREHENSION
# ============================================================
#
# A comprehension can contain another comprehension:
#
#     [[x * y for y in values] for x in values]
#
# This is valid Python, but nested comprehensions quickly become
# harder to read.
#
# The important lesson is NOT:
#
#     "Never use nested comprehensions."
#
# It is:
#
#     "Use them only when the resulting code remains clear."
#
# If the reader has to mentally decode several loops and
# conditions, use a regular loop instead.
#
# ============================================================


# ============================================================
# 6. GENERATOR EXPRESSIONS — SAVE MEMORY
# ============================================================
#
# A list comprehension creates the entire list immediately:
#
#     squares = [x ** 2 for x in range(10)]
#
#
# A generator expression produces values lazily:
#
#     squares = (x ** 2 for x in range(10))
#
#
# Notice:
#
#     [ ] → list comprehension
#     ( ) → generator expression
#
# ============================================================


gen = (x ** 2 for x in range(10))


# At this point, the generator has NOT created a complete list
# containing all the squared values.
#
# It produces values as they are requested.
#
#
# Conceptual model:
#
#     generator
#         ↓
#     request next value
#         ↓
#     calculate one value
#         ↓
#     return it
#         ↓
#     request next value
#         ↓
#     repeat
#
# ============================================================


print(list(gen))


# list(gen) consumes the generator and materialises its values
# into a list.
#
# Result:
#
#     [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
#
# After the generator is exhausted, it cannot be restarted:
#
#     gen → values → exhausted
#
# ============================================================


# ============================================================
# LIST COMPREHENSION vs GENERATOR EXPRESSION
# ============================================================
#
# LIST COMPREHENSION
#
#     squares = [x ** 2 for x in range(10)]
#
#     → creates the complete list immediately
#     → values are stored in memory
#     → can be indexed and reused
#
#
# GENERATOR EXPRESSION
#
#     squares = (x ** 2 for x in range(10))
#
#     → produces values lazily
#     → does not create the complete result list immediately
#     → useful when values can be processed one at a time
#
#
# Visual:
#
#     LIST
#
#     data → calculate → store ALL results
#                         ↓
#                  [0, 1, 4, 9, ...]
#
#
#     GENERATOR
#
#     data → calculate one → yield
#                       ↓
#                  calculate next
#                       ↓
#                  yield
#                       ↓
#                     ...
#
# ============================================================


# ============================================================
# 7. DON'T ASSUME GENERATORS ARE ALWAYS FASTER
# ============================================================
#
# Generators primarily solve a MEMORY / LAZY-EVALUATION problem.
#
# They are not simply:
#
#     "a faster list comprehension."
#
# If you actually need every result as a list, creating a list
# directly may be more appropriate.
#
# ============================================================


# ============================================================
# 8. DON'T OPTIMIZE ONLY FOR FEWER CHARACTERS
# ============================================================
#
# This:
#
#     result = [f(x) for x in data]
#
# is not automatically better than:
#
#     result = []
#     for x in data:
#         result.append(f(x))
#
# The comprehension is better when its compact form remains
# readable.
#
# Code quality includes:
#
#     readability
#     maintainability
#     correctness
#     performance
#
# Performance is only one part of the decision.
#
# ============================================================


# ============================================================
# PRACTICAL DECISION GUIDE
# ============================================================
#
# Ask these questions:
#
#     Is the transformation simple?
#             ↓
#           YES
#             ↓
#     Is the comprehension easy to read?
#             ↓
#           YES → use comprehension
#
#
#     Is the logic complicated?
#             ↓
#           YES
#             ↓
#     Use a regular loop and/or helper function.
#
#
#     Do I need the entire result in memory?
#             ↓
#           YES → list comprehension may be appropriate
#
#
#     Can I process values one at a time?
#             ↓
#           YES → consider a generator expression
#
#
#     Am I working with numerical pandas/NumPy data?
#             ↓
#           YES → first consider vectorized operations
#
# ============================================================


# ============================================================
# THE CORE RULE
# ============================================================
#
# LIST COMPREHENSION
#     → simple transformation/filter
#     → concise and readable
#
# HELPER FUNCTION + COMPREHENSION
#     → transformation has meaningful logic
#     → keeps the comprehension clean
#
# REGULAR FOR LOOP
#     → multiple steps, decisions, or side effects
#     → clearer than forcing everything into one line
#
# GENERATOR EXPRESSION
#     → lazy processing
#     → useful when the complete result does not need to exist
#       in memory at once
#
# ============================================================
#
# PYTHON FEELING:
#
#     Don't ask:
#         "Can I write this as a comprehension?"
#
#     Ask:
#         "Does a comprehension make this code clearer?"
#
# ============================================================
#
# BEST PRACTICE:
#
#     SIMPLE + CLEAR
#          ↓
#     comprehension
#
#     COMPLEX
#          ↓
#     loop / helper function
#
#     LARGE + LAZY
#          ↓
#     generator
#
#     NUMERICAL DATA
#          ↓
#     consider NumPy / pandas vectorization
#
# ============================================================
```

```python
# ============================================================
# CPYTHON — WHAT ACTUALLY RUNS YOUR PYTHON CODE?
# ============================================================
#
# This is an important concept for understanding why some Python
# constructs, such as list comprehensions, can perform differently
# from equivalent code written with explicit loops.
#
# ============================================================


# ============================================================
# 1. WHAT IS CPYTHON?
# ============================================================
#
# Python is a PROGRAMMING LANGUAGE.
#
# CPython is an IMPLEMENTATION of that language.
#
# In other words:
#
#     Python
#       ↓
#     language / syntax / rules
#
#     CPython
#       ↓
#     a program that implements those rules
#
#
# CPython is the standard and most widely used implementation
# of Python.
#
# When you install the standard Python distribution from:
#
#     python.org
#
# you normally get CPython.
#
# ============================================================


# ============================================================
# 2. PYTHON vs CPYTHON
# ============================================================
#
# Think of a language such as English.
#
# English
#    ↓
# language
#
# Different people can speak it:
#
#     American
#     British
#     Australian
#     Canadian
#
# Same language, different speakers.
#
#
# Python works similarly:
#
#     Python
#       ↓
#     programming language
#
# Different implementations can execute Python code:
#
#     CPython
#     PyPy
#     Jython
#     IronPython
#     MicroPython
#
#
# They aim to implement Python, but they use different
# technologies and have different goals.
#
# ============================================================


# ============================================================
# 3. COMMON PYTHON IMPLEMENTATIONS
# ============================================================
#
# Implementation    Main Technology       Main Purpose
# ------------------------------------------------------------
# CPython           C                     Standard implementation
# PyPy              RPython + Python      JIT-optimized execution
# Jython            Java                  JVM integration
# IronPython        C# / .NET             .NET integration
# MicroPython       C                     Microcontrollers/devices
#
#
# IMPORTANT:
#
# "Python" does not mean that there is only one program capable
# of executing Python code.
#
# ============================================================


# ============================================================
# 4. WHY IS IT CALLED "CPYTHON"?
# ============================================================
#
# The "C" refers to the C programming language.
#
# CPython's implementation is primarily written in C.
#
# It is NOT:
#
#     C++
#     Python
#
# It is primarily:
#
#     C
#
#
# The C implementation is compiled for different platforms,
# such as:
#
#     Windows
#     Linux
#     macOS
#
# ============================================================


# ============================================================
# 5. WHAT HAPPENS WHEN YOU RUN PYTHON?
# ============================================================
#
# Suppose you write:
#
#     numbers = [1, 2, 3]
#
#     squares = [x ** 2 for x in numbers]
#
#
# A simplified mental model is:
#
#
#     YOUR PYTHON CODE
#            ↓
#     CPython implementation
#            ↓
#     Python bytecode
#            ↓
#     CPython execution machinery
#            ↓
#     CPU / operating system
#
#
# This is a simplified model, not a literal complete description
# of every internal step.
#
# The important idea is:
#
#     Python source code
#            ↓
#     CPython interprets/executes it
#
# ============================================================


# ============================================================
# 6. WHAT IS BYTECODE?
# ============================================================
#
# CPython does not normally execute your source code directly as
# CPU machine instructions.
#
# It first compiles Python source into BYTECODE.
#
# Example:
#
#     Python source
#          ↓
#     compiler
#          ↓
#     bytecode
#          ↓
#     CPython execution
#
#
# Bytecode is an intermediate instruction format designed for
# the Python virtual machine/execution machinery.
#
# It is NOT the same thing as CPU machine code.
#
# ============================================================


# ============================================================
# 7. WHY DOES CPYTHON MATTER FOR LIST COMPREHENSIONS?
# ============================================================
#
# Consider these two equivalent operations:
#
#
# ------------------------------------------------------------
# REGULAR FOR LOOP
# ------------------------------------------------------------

numbers = [1, 2, 3, 4, 5]

result = []

for x in numbers:
    result.append(x * 2)


# ------------------------------------------------------------
# LIST COMPREHENSION
# ------------------------------------------------------------

result = [x * 2 for x in numbers]


# Both produce:
#
#     [2, 4, 6, 8, 10]
#
#
# But the two pieces of Python source code are not identical
# internally.
#
# CPython has specialized execution mechanisms for list
# comprehensions.
#
# This can reduce some interpreter overhead involved in
# repeatedly using list.append().
#
# ============================================================


# ============================================================
# 8. WHY CAN THE COMPREHENSION BE FASTER?
# ============================================================
#
# REGULAR LOOP:
#
#     result = []
#          ↓
#     get next x
#          ↓
#     calculate x * 2
#          ↓
#     access result.append
#          ↓
#     call append()
#          ↓
#     add result
#          ↓
#     repeat
#
#
# LIST COMPREHENSION:
#
#     create result list
#          ↓
#     get next x
#          ↓
#     calculate x * 2
#          ↓
#     add result through specialized
#     comprehension machinery
#          ↓
#     repeat
#
#
# The comprehension can avoid some of the overhead associated
# with repeatedly performing an explicit method call such as:
#
#     result.append(...)
#
# ============================================================


# ============================================================
# 9. THE IMPORTANT CORRECTION
# ============================================================
#
# DO NOT SAY:
#
#     "List comprehensions are faster because they are written
#      in C."
#
#
# This is misleading.
#
# Ordinary Python for-loops are ALSO executed by CPython, whose
# implementation is written primarily in C.
#
# The important difference is that CPython has specialized
# execution machinery for different Python constructs.
#
# Therefore:
#
#     Python for-loop
#          ↓
#     CPython
#
#     List comprehension
#          ↓
#     CPython
#
# Both use CPython.
#
# The comprehension can nevertheless have lower interpreter
# overhead for certain simple list-building operations.
#
# ============================================================


# ============================================================
# 10. BYTECODE VIEW
# ============================================================
#
# You can actually inspect the bytecode generated by CPython.
#
# The dis module is useful for learning this:
#
# ============================================================

import dis


def loop_version(numbers):
    result = []

    for x in numbers:
        result.append(x * 2)

    return result


def comprehension_version(numbers):
    return [x * 2 for x in numbers]


# Display the bytecode:
#
#     dis.dis(loop_version)
#     dis.dis(comprehension_version)
#
# Uncomment the following lines to inspect it:
#
# dis.dis(loop_version)
# dis.dis(comprehension_version)
#
#
# You will see that CPython generates different bytecode for
# the two implementations.
#
# The exact instructions can change between Python versions.
#
# Therefore, do not memorize particular bytecode instructions.
#
# Learn the bigger concept:
#
#     different Python constructs
#             ↓
#     different bytecode / execution paths
#             ↓
#     potentially different performance
#
# ============================================================


# ============================================================
# 11. WHY "FEWER INSTRUCTIONS" IS A USEFUL IDEA
# ============================================================
#
# At a high level:
#
#     more interpreter work
#             ↓
#     more overhead
#
#     less interpreter work
#             ↓
#     potentially less overhead
#
#
# For a very simple operation repeated millions of times,
# small overhead differences can become measurable.
#
# That is one reason a comprehension can outperform:
#
#     result = []
#     for x in data:
#         result.append(x * 2)
#
# ============================================================


# ============================================================
# 12. BUT THIS DOES NOT MEAN "COMPREHENSION = FAST"
# ============================================================
#
# Performance depends on the actual workload.
#
# For example:
#
#     [x * 2 for x in data]
#
# performs very little work per iteration.
#
# The overhead of the loop itself therefore matters relatively
# more.
#
#
# But if each iteration performs an expensive operation:
#
#     [expensive_function(x) for x in data]
#
# then the expensive function may dominate the total runtime.
#
#
# Conceptually:
#
#     simple operation
#          ↓
#     loop overhead is significant
#
#
#     expensive operation
#          ↓
#     operation itself dominates
#
# ============================================================


# ============================================================
# 13. NUMPY IS A DIFFERENT PERFORMANCE STORY
# ============================================================
#
# Consider numerical data.
#
# Python-level:
#
#     squares = [x ** 2 for x in numbers]
#
#
# NumPy:
#
#     squares = array ** 2
#
#
# With NumPy, the operation can be performed by optimized
# compiled numerical code over an entire array.
#
#
# Conceptually:
#
#     Python loop
#          ↓
#     Python-level iteration
#          ↓
#     one element at a time
#
#
#     NumPy vectorization
#          ↓
#     compiled numerical operation
#          ↓
#     operate over array data efficiently
#
#
# Therefore, when working with large numerical datasets, the
# important optimization may be to MOVE THE WORK OUT OF THE
# PYTHON-LEVEL LOOP entirely.
#
# ============================================================


# ============================================================
# 14. THE PERFORMANCE HIERARCHY IS NOT ABSOLUTE
# ============================================================
#
# Do NOT memorize:
#
#     comprehension > loop > NumPy
#
# That is too simplistic.
#
# Instead think:
#
#     What operation am I performing?
#              ↓
#     Where is the work happening?
#              ↓
#     How much Python-level overhead exists?
#              ↓
#     How much work does each iteration perform?
#              ↓
#     Can a specialized library perform the operation?
#
# ============================================================


# ============================================================
# 15. THE MOST ACCURATE NOTE FOR YOUR PERFORMANCE SECTION
# ============================================================
#
# If you want one concise statement to remember:
#
# ------------------------------------------------------------
#
# List comprehensions are often faster than equivalent
# for-loops that repeatedly use append() because CPython has
# specialized execution machinery for comprehensions, which can
# reduce some interpreter overhead.
#
# This does NOT mean comprehensions are "written in C" while
# ordinary loops are not.
#
# Both are executed by CPython.
#
# For numerical data, NumPy/pandas vectorization can often
# provide a much larger performance advantage by reducing
# Python-level iteration.
#
# ------------------------------------------------------------
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# Python
#   ↓
# language
#
# CPython
#   ↓
# implementation of that language
#
# Your code
#   ↓
# CPython
#   ↓
# bytecode / execution machinery
#   ↓
# CPU
#
#
# And for performance:
#
#     Python code
#          ↓
#     how does CPython execute it?
#          ↓
#     how much interpreter overhead is involved?
#          ↓
#     can specialized machinery or compiled libraries
#     perform the work more efficiently?
#
# ============================================================
#
# CORE IDEA:
#
# CPython is not a special Python library.
#
# It is the implementation that normally runs your Python code.
#
# List comprehensions can be faster because CPython gives them
# specialized execution treatment and can avoid some overhead
# associated with explicit append() calls.
#
# ============================================================
```

# 62. LIST COMPREHENSION 2

```python
# ============================================================
# 62. LIST COMPREHENSION
# ============================================================
#
# A list comprehension is a compact way to create a NEW list
# by processing each item from an iterable.
#
# Basic pattern:
#
#     [expression for item in iterable]
#
# Think:
#
#     INPUT
#       ↓
#     take each item
#       ↓
#     transform it
#       ↓
#     collect the result
#       ↓
#     NEW LIST
#
# The important idea is the DATA TRANSFORMATION pattern:
#
#     one input item → one output item
#
# ============================================================
# CREATE A LIST
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# LIST COMPREHENSION
# ============================================================
#
# This creates a NEW list called squares.
#
# General structure:
#
#     [expression for item in iterable]
#      ↑             ↑         ↑
#      |             |         |
#      |             |         input collection
#      |             loop variable
#      output expression
#
# Here:
#
#     expression = number ** 2
#     item       = number
#     iterable   = numbers
#
# Meaning:
#
#     "For each number in numbers,
#      calculate its square,
#      and collect that result into a new list."
#
# ============================================================
# VALUE FLOW
# ============================================================
#
#     numbers
#        ↓
#     1 → 1 ** 2 → 1
#     2 → 2 ** 2 → 4
#     3 → 3 ** 2 → 9
#     4 → 4 ** 2 → 16
#     5 → 5 ** 2 → 25
#        ↓
#     [1, 4, 9, 16, 25]
#
# ============================================================
# PROCESS TABLE
# ============================================================
#
#     Step | Current number | Operation    | Result | squares
#     -----|----------------|---------------|--------|-----------------------
#       1  |       1        | 1 ** 2        |   1    | [1]
#       2  |       2        | 2 ** 2        |   4    | [1, 4]
#       3  |       3        | 3 ** 2        |   9    | [1, 4, 9]
#       4  |       4        | 4 ** 2        |  16    | [1, 4, 9, 16]
#       5  |       5        | 5 ** 2        |  25    | [1, 4, 9, 16, 25]
#
# The table shows the important process:
#
#     get item
#         ↓
#     apply expression
#         ↓
#     produce result
#         ↓
#     collect result
#
# ============================================================

squares = [
    number ** 2
    for number in numbers
]


# ============================================================
# PROGRAM STATE
# ============================================================
#
# BEFORE:
#
#     numbers = [1, 2, 3, 4, 5]
#
# DURING:
#
#     number receives one item at a time.
#
#     1 → calculate 1² → collect 1
#     2 → calculate 2² → collect 4
#     3 → calculate 3² → collect 9
#     4 → calculate 4² → collect 16
#     5 → calculate 5² → collect 25
#
# AFTER:
#
#     numbers = [1, 2, 3, 4, 5]
#     squares = [1, 4, 9, 16, 25]
#
# Notice:
#
#     numbers does NOT change.
#
#     squares is the NEW list created by the comprehension.
#
# ============================================================
# HOW IT WORKS
# ============================================================
#
#     numbers
#        ↓
#     get next item
#        ↓
#     number receives item
#        ↓
#     evaluate number ** 2
#        ↓
#     result is collected into squares
#        ↓
#     get next item
#        ↓
#     repeat
#        ↓
#     no more items
#        ↓
#     squares is complete
#
# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The comprehension is essentially a compact form of:
#
#     squares = []
#
#     for number in numbers:
#         squares.append(number ** 2)
#
# Conceptually:
#
#     create empty list
#          ↓
#     get item
#          ↓
#     calculate result
#          ↓
#     append result
#          ↓
#     repeat
#
# The list comprehension expresses the same common pattern
# more compactly.
#
# ============================================================
# PRINT RESULT
# ============================================================

print(squares)

# Output:
#
#     [1, 4, 9, 16, 25]
#
# Why?
#
# Each number is squared:
#
#     1² = 1
#     2² = 4
#     3² = 9
#     4² = 16
#     5² = 25
#
# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, transform it and create a new list."
#
# Think:
#
#     [expression for item in iterable]
#
# Examples:
#
#     [number * 2 for number in numbers]
#     [number ** 2 for number in numbers]
#     [word.upper() for word in words]
#
# General pattern:
#
#     INPUT COLLECTION
#          ↓
#     FOR EACH ITEM
#          ↓
#     TRANSFORM
#          ↓
#     COLLECT RESULT
#          ↓
#     NEW LIST
#
# ============================================================
# MENTAL MODEL
# ============================================================
#
# Do not think of a list comprehension as mysterious syntax.
#
# Mentally expand it into:
#
#     "Take each item,
#      do something with it,
#      collect the result."
#
# For this example:
#
#     take number
#         ↓
#     calculate number ** 2
#         ↓
#     collect result
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
#     [TRANSFORMATION for ITEM in COLLECTION]
#
# Examples:
#
#     [x * 10 for x in numbers]
#
#     [x ** 2 for x in numbers]
#
#     [word.upper() for word in words]
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A list comprehension combines:
#
#     ITERATION
#          +
#     TRANSFORMATION
#          +
#     LIST CREATION
#
# into one compact expression.
#
# The key mental pattern is:
#
#     "For each input item
#      → calculate an output
#      → collect it into a new list."
#
# ============================================================
```

```python
# ============================================================
# 63. LIST COMPREHENSION WITH CONDITION
# ============================================================
#
# A conditional list comprehension creates a new list by
# iterating through an iterable and keeping only the items
# that satisfy a condition.
#
# GENERAL PATTERN:
#
#     [expression for item in iterable if condition]
#
# Think:
#
#     collection
#          ↓
#     current item
#          ↓
#     condition
#      ↙       ↘
#   False      True
#    ↓           ↓
#   skip       include
#      \         /
#       ↓       ↓
#        new list
#
# This combines:
#
#     ITERATION + FILTERING + LIST CREATION
#
# ============================================================
# CONCEPTUAL BREAKDOWN
# ============================================================
#
# numbers
#     → source collection
#
# number
#     → loop variable representing the current item
#
# number % 2 == 0
#     → condition that tests whether the number is even
#
# even_numbers
#     → new list containing the values that pass the condition
#
# RELATIONSHIP:
#
#     numbers
#         ↓
#     get current number
#         ↓
#     test condition
#         ↓
#     True → include number
#     False → skip number
#         ↓
#     even_numbers
#
# ============================================================
# SYNTAX → MEANING
# ============================================================
#
#     number
#     → value placed into the new list
#
#     for number in numbers
#     → process every item in numbers
#
#     if number % 2 == 0
#     → include the item only when the condition is True
#
# IMPORTANT:
#
# The if condition FILTERS the items.
#
# It does not modify number.
#
# ============================================================
# HOW THE CONDITION WORKS
# ============================================================
#
#     number % 2
#
# gives the remainder after division by 2.
#
#     1 % 2 → 1
#     2 % 2 → 0
#     3 % 2 → 1
#     4 % 2 → 0
#     5 % 2 → 1
#     6 % 2 → 0
#
# Therefore:
#
#     number % 2 == 0
#
#     even number → True
#     odd number  → False
#
# ============================================================
# VALUE FLOW
# ============================================================
#
#     1 → 1 % 2 → 1 == 0 → False → skip
#     2 → 2 % 2 → 0 == 0 → True  → keep
#     3 → 3 % 2 → 1 == 0 → False → skip
#     4 → 4 % 2 → 0 == 0 → True  → keep
#     5 → 5 % 2 → 1 == 0 → False → skip
#     6 → 6 % 2 → 0 == 0 → True  → keep
#
# Final:
#
#     [2, 4, 6]
#
# ============================================================
# VARIABLE / VALUE / TYPE TRACKING
# ============================================================
#
# numbers
#     → [1, 2, 3, 4, 5, 6]
#     → list
#     → source collection
#
# number
#     → 1, then 2, then 3, ..., then 6
#     → int
#     → current item
#
# even_numbers
#     → list
#     → contains accepted values
#
# IMPORTANT:
#
# numbers stays unchanged.
#
# number changes on each iteration.
#
# even_numbers grows as qualifying values are included.
#
# ============================================================
# HOW PYTHON EXECUTES THE CODE
# ============================================================
#
# Python conceptually does:
#
#     create new list
#          ↓
#     get next item
#          ↓
#     assign it to number
#          ↓
#     evaluate the condition
#          ↓
#     True → include number
#     False → skip it
#          ↓
#     get next item
#          ↓
#     repeat
#          ↓
#     finish the list
#
# ============================================================
# PROCESS TABLE
# ============================================================
#
# Step | number | % 2 | Condition | Action | even_numbers
# -----|--------|-----|-----------|--------|--------------
#  1   |   1    |  1  | False     | Skip   | []
#  2   |   2    |  0  | True      | Keep   | [2]
#  3   |   3    |  1  | False     | Skip   | [2]
#  4   |   4    |  0  | True      | Keep   | [2, 4]
#  5   |   5    |  1  | False     | Skip   | [2, 4]
#  6   |   6    |  0  | True      | Keep   | [2, 4, 6]
#
# ============================================================
# EXECUTION TRACE
# ============================================================
#
# Start:
#
#     even_numbers = []
#
# Iteration 1:
#
#     number = 1
#     1 % 2 == 0 → False
#     → skip
#
# Iteration 2:
#
#     number = 2
#     2 % 2 == 0 → True
#     → include 2
#
#     even_numbers = [2]
#
# Iteration 3:
#
#     number = 3
#     3 % 2 == 0 → False
#     → skip
#
# Iteration 4:
#
#     number = 4
#     4 % 2 == 0 → True
#     → include 4
#
#     even_numbers = [2, 4]
#
# Iteration 5:
#
#     number = 5
#     5 % 2 == 0 → False
#     → skip
#
# Iteration 6:
#
#     number = 6
#     6 % 2 == 0 → True
#     → include 6
#
#     even_numbers = [2, 4, 6]
#
# No more items → comprehension ends.
#
# ============================================================
# EQUIVALENT BEGINNER-FRIENDLY VERSION
# ============================================================
#
# The same basic logic can be written explicitly:
#
#     even_numbers = []
#
#     for number in numbers:
#
#         if number % 2 == 0:
#             even_numbers.append(number)
#
# The loop version makes the individual steps visible.
#
# The comprehension combines those steps into one compact
# list-creation expression.
#
# ============================================================
# DATA TRANSFORMATION
# ============================================================
#
# INPUT:
#
#     [1, 2, 3, 4, 5, 6]
#
#         ↓
#
# RULE:
#
#     Keep values where number % 2 == 0
#
#         ↓
#
#     1 → reject
#     2 → keep
#     3 → reject
#     4 → keep
#     5 → reject
#     6 → keep
#
#         ↓
#
# OUTPUT:
#
#     [2, 4, 6]
#
# This is primarily FILTERING, not value transformation.
#
# ============================================================
# INVARIANT / ALWAYS-TRUE RELATIONSHIP
# ============================================================
#
# After each iteration:
#
#     even_numbers contains exactly the even values
#     encountered so far.
#
# This explains why the final result is correct.
#
# ============================================================
# COMMON MISTAKE
# ============================================================
#
# The condition:
#
#     number % 2 == 0
#
# does NOT change number.
#
# It only decides whether number is included.
#
# Example:
#
#     number = 3
#     3 % 2 == 0 → False
#     → 3 is skipped
#
#     number = 4
#     4 % 2 == 0 → True
#     → 4 is included as 4
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
#     [item for item in collection if condition]
#
# means:
#
#     "For every item in the collection,
#      keep the item if the condition is True."
#
# Examples:
#
#     [x for x in numbers if x > 10]
#     → keep values greater than 10
#
#     [word for word in words if len(word) > 5]
#     → keep words longer than 5 characters
#
#     [sale for sale in sales if sale >= 1000]
#     → keep sales of at least 1000
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A conditional list comprehension combines:
#
#     ITERATION
#         +
#     CONDITION
#         +
#     LIST CREATION
#
# The key mental model is:
#
#     collection
#         ↓
#     current item
#         ↓
#     test condition
#       ↙       ↘
#    False      True
#      ↓          ↓
#    skip       include
#         \      /
#          ↓    ↓
#          new list
#
# For this example:
#
#     [number for number in numbers if number % 2 == 0]
#
# means:
#
#     "Take each number, keep it if it is even,
#      and build a new list from those values."
#
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]  # Original list of numbers

# Create a new list containing only even numbers.
# For each number in numbers, Python evaluates the condition.
# If the condition is True, number is included in even_numbers.
even_numbers = [
    number
    for number in numbers
    if number % 2 == 0
]

print(even_numbers)  # Output: [2, 4, 6]
```

```python
# ============================================================
# 64. LIST COMPREHENSION — DATA TRANSFORMATION
# ============================================================
#
# A list comprehension can do more than filter data.
# It can TRANSFORM each item and create a new list.
#
# General pattern:
#
#     [expression for item in iterable]
#
# Meaning:
#
#     Take each item
#         ↓
#     apply an expression/transformation
#         ↓
#     store the result in a new list
#
# Here, the transformation is:
#
#     price * 1.20
#
# This means:
#
#     original price + 20% tax
#
# Mathematical view:
#
#     price_with_tax = price × 1.20
#
# Python view:
#
#     price_with_tax = price * 1.20
#
# IMPORTANT:
#     x² is mathematical notation.
#     x**2 is Python syntax.
#
# ------------------------------------------------------------
# CONCEPTUAL BREAKDOWN
# ------------------------------------------------------------
#
# Original data:
#
#     prices = [100, 200, 300, 400]
#
# Transformation:
#
#     price × 1.20
#
# Result:
#
#     [120.0, 240.0, 360.0, 480.0]
#
# The original list is NOT modified.
# A completely new list is created.
#
# ------------------------------------------------------------
# COMPONENT ROLES
# ------------------------------------------------------------
#
# prices
#     → original iterable containing the input values
#
# price
#     → temporary loop variable
#     → represents one price at a time
#
# price * 1.20
#     → transformation expression
#     → calculates the tax-included price
#
# prices_with_tax
#     → new list containing all transformed results
#
# ------------------------------------------------------------
# VALUE FLOW
# ------------------------------------------------------------
#
#     prices
#       ↓
#     100 → 100 × 1.20 → 120.0
#     200 → 200 × 1.20 → 240.0
#     300 → 300 × 1.20 → 360.0
#     400 → 400 × 1.20 → 480.0
#       ↓
#     prices_with_tax
#
# ------------------------------------------------------------
# PROCESS TABLE
# ------------------------------------------------------------
#
# | Step | price | Operation       | Result | New List              |
# |------|-------|-----------------|--------|------------------------|
# | 1    | 100   | 100 × 1.20      | 120.0  | [120.0]                |
# | 2    | 200   | 200 × 1.20      | 240.0  | [120.0, 240.0]        |
# | 3    | 300   | 300 × 1.20      | 360.0  | [120.0, 240.0, 360.0] |
# | 4    | 400   | 400 × 1.20      | 480.0  | [120.0, 240.0, 360.0, 480.0] |
#
# ------------------------------------------------------------
# EXECUTION TRACE
# ------------------------------------------------------------
#
# Python effectively performs:
#
#     price = 100
#     100 * 1.20 → 120.0
#
#     price = 200
#     200 * 1.20 → 240.0
#
#     price = 300
#     300 * 1.20 → 360.0
#
#     price = 400
#     400 * 1.20 → 480.0
#
# Then the new list becomes:
#
#     [120.0, 240.0, 360.0, 480.0]
#
# ------------------------------------------------------------
# VARIABLE / VALUE / TYPE TRACKING
# ------------------------------------------------------------
#
# prices
#     → value: [100, 200, 300, 400]
#     → type: list
#     → role: input data
#
# price
#     → values during iteration: 100 → 200 → 300 → 400
#     → type: int
#     → role: current item
#
# prices_with_tax
#     → value: [120.0, 240.0, 360.0, 480.0]
#     → type: list
#     → role: transformed output
#
# ------------------------------------------------------------
# WHY 1.20?
# ------------------------------------------------------------
#
# A 20% tax means:
#
#     20% = 20 / 100 = 0.20
#
# Original price:
#
#     100% = 1.00
#
# Price + tax:
#
#     100% + 20% = 120%
#
#     120% = 1.20
#
# Therefore:
#
#     price × 1.20
#
# calculates the final price including 20% tax.
#
# Example:
#
#     100 × 1.20 = 120
#
#     Tax = 100 × 0.20 = 20
#     Final = 100 + 20 = 120
#
# ------------------------------------------------------------
# DATA TRANSFORMATION
# ------------------------------------------------------------
#
# This is a simple DATA TRANSFORMATION pipeline:
#
#     INPUT
#       ↓
#     [100, 200, 300, 400]
#       ↓
#     TRANSFORM EACH ITEM
#       ↓
#     price × 1.20
#       ↓
#     OUTPUT
#       ↓
#     [120.0, 240.0, 360.0, 480.0]
#
# The number of elements stays the same.
# Only the values are transformed.
#
# ------------------------------------------------------------
# WHAT CHANGES / WHAT STAYS THE SAME
# ------------------------------------------------------------
#
# Changes:
#     Each price value
#
# Stays the same:
#     Number of elements → 4
#     Order of elements → preserved
#     Original prices → unchanged
#
# So this is a 1-to-1 transformation:
#
#     1 input item → 1 output item
#
# ------------------------------------------------------------
# EQUIVALENT BEGINNER-FRIENDLY VERSION
# ------------------------------------------------------------
#
# The list comprehension is conceptually equivalent to:
#
#     prices_with_tax = []
#
#     for price in prices:
#         taxed_price = price * 1.20
#         prices_with_tax.append(taxed_price)
#
# The comprehension simply expresses this pattern more compactly.
#
# ------------------------------------------------------------
# GENERALIZABLE PATTERN
# ------------------------------------------------------------
#
# Whenever you hear:
#
#     "For every item, calculate a new value."
#
# Think:
#
#     [transformation for item in collection]
#
# Examples:
#
#     [price * 1.20 for price in prices]
#     → add 20% tax
#
#     [price * 0.90 for price in prices]
#     → apply a 10% discount
#
#     [number ** 2 for number in numbers]
#     → square every number
#
#     [name.upper() for name in names]
#     → convert every name to uppercase
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
# List comprehension = "transform every item and collect the results."
#
#     collection
#          ↓
#     one item
#          ↓
#     transformation
#          ↓
#     result
#          ↓
#     new list
#
# ------------------------------------------------------------
# CORE IDEA
# ------------------------------------------------------------
#
# A list comprehension is especially useful when the task is:
#
#     "Take every item from a collection,
#      perform the same transformation,
#      and build a new list from the results."
#
# Here:
#
#     prices
#       → apply 20% tax to each price
#       → prices_with_tax
#
# ============================================================

# Define a list of prices
prices = [100, 200, 300, 400]

# Use list comprehension to create a new list with prices including 20% tax
# Each price is multiplied by 1.20 to add the tax
prices_with_tax = [
    price * 1.20  # Apply 20% tax to each price
    for price in prices  # Iterate through each price in the original list
]

# Print the new list with tax-included prices
print(prices_with_tax)
```

```python
# ============================================================
# 65. LIST COMPREHENSION — IF ELSE
# ============================================================
#
# A list comprehension can contain an if-else expression to
# produce different results depending on a condition.
#
# The basic pattern is:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# IMPORTANT:
#
# In this form, the if-else expression comes BEFORE the
# for loop.
#
# Think:
#
#     For each item:
#         if condition → return this
#         else         → return that
#
# ============================================================


# ============================================================
# CREATE A LIST OF SCORES
# ============================================================
#
# Each score will be checked to determine whether the student
# passed or failed.
#
# Passing score:
#
#     50 or higher → Pass
#
# Below 50:
#
#     Fail
#
# ============================================================

scores = [45, 67, 82, 91, 38]


# ============================================================
# LIST COMPREHENSION WITH IF ELSE
# ============================================================
#
# For each score:
#
#     score >= 50 → "Pass"
#     otherwise   → "Fail"
#
# Step by step:
#
#     45 → 45 >= 50 → False → "Fail"
#     67 → 67 >= 50 → True  → "Pass"
#     82 → 82 >= 50 → True  → "Pass"
#     91 → 91 >= 50 → True  → "Pass"
#     38 → 38 >= 50 → False → "Fail"
#
# Result:
#
#     ["Fail", "Pass", "Pass", "Pass", "Fail"]
#
# ============================================================

results = [
    "Pass" if score >= 50 else "Fail"
    for score in scores
]


# ============================================================
# PRINT RESULT
# ============================================================

print(results)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The list comprehension above is equivalent to:
#
#     results = []
#
#     for score in scores:
#
#         if score >= 50:
#             results.append("Pass")
#         else:
#             results.append("Fail")
#
# The comprehension combines the same logic into one
# expression.
#
# ============================================================


# ============================================================
# IMPORTANT: IF-ELSE VS IF-ONLY
# ============================================================
#
# IF-ONLY:
#
#     [expression
#      for item in iterable
#      if condition]
#
# The if comes AFTER the for because it FILTERS items.
#
#
# IF-ELSE:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# The if-else comes BEFORE the for because it TRANSFORMS
# every item into one of two possible values.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, choose A or B."
#
# Think:
#
#     [A if condition else B for item in iterable]
#
#
# When you think:
#
#     "Keep only the items that satisfy a condition."
#
# Think:
#
#     [item for item in iterable if condition]
#
# The key difference:
#
#     IF ONLY → FILTER
#     IF ELSE → CHOOSE / TRANSFORM
#
# ============================================================
```

```python
# ============================================================
# 66. DICTIONARY COMPREHENSION
# ============================================================
#
# A dictionary comprehension is a compact way to create a
# dictionary by generating KEY-VALUE pairs from an iterable.
#
# The basic pattern is:
#
#     {key: value for item in iterable}
#
# It is similar to a list comprehension, but instead of
# producing individual values, it produces:
#
#     KEY → VALUE
#
# Dictionary comprehensions are especially useful when you
# want to transform data into a dictionary in a single,
# readable expression.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains the numbers that we want to process.
#
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# DICTIONARY COMPREHENSION
# ============================================================
#
# For each number:
#
#     number       → KEY
#     number ** 2  → VALUE
#
# Step by step:
#
#     1 → 1² → 1  → 1: 1
#     2 → 2² → 4  → 2: 4
#     3 → 3² → 9  → 3: 9
#     4 → 4² → 16 → 4: 16
#     5 → 5² → 25 → 5: 25
#
# Result:
#
#     {
#         1: 1,
#         2: 4,
#         3: 9,
#         4: 16,
#         5: 25
#     }
#
# ============================================================

squares = {
    number: number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The dictionary comprehension above is equivalent to:
#
#     squares = {}
#
#     for number in numbers:
#         squares[number] = number ** 2
#
# The comprehension simply expresses the same operation
# in a more compact form.
#
# ============================================================


# ============================================================
# DICTIONARY COMPREHENSION WITH A CONDITION
# ============================================================
#
# You can also filter items using an if condition.
#
# Example:
#
#     even_squares = {
#         number: number ** 2
#         for number in numbers
#         if number % 2 == 0
#     }
#
# Result:
#
#     {2: 4, 4: 16}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, create a KEY → VALUE pair."
#
# Think:
#
#     {key: value for item in iterable}
#
# The three main comprehension types are:
#
#     [expression for item in iterable]
#         → LIST
#
#     {expression for item in iterable}
#         → SET
#
#     {key: value for item in iterable}
#         → DICTIONARY
#
# ============================================================
```

```python
# ============================================================
# 67. SET COMPREHENSION
# ============================================================
#
# A set comprehension is a compact way to create a set by
# applying an expression to each item in an iterable.
#
# The basic pattern is:
#
#     {expression for item in iterable}
#
# It is similar to a list comprehension, but uses { } and
# produces a SET instead of a LIST.
#
# Important:
#
#     List → ordered collection, allows duplicates
#     Set  → unique elements, removes duplicates
#
# This makes set comprehensions useful when you want to
# transform data and automatically keep only unique results.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains duplicate values:
#
#     2 appears twice
#     3 appears twice
#
# ============================================================

numbers = [1, 2, 2, 3, 3, 4, 5]


# ============================================================
# SET COMPREHENSION
# ============================================================
#
# For each number:
#
#     number ** 2
#
# calculates its square.
#
# The resulting values are stored in a SET.
#
# Because a set contains only unique elements, duplicate
# results are automatically removed.
#
# Step by step:
#
#     1 → 1² → 1
#     2 → 2² → 4
#     2 → 2² → 4  ← duplicate
#     3 → 3² → 9
#     3 → 3² → 9  ← duplicate
#     4 → 4² → 16
#     5 → 5² → 25
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================

squares = {
    number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# SET COMPREHENSION VS LIST COMPREHENSION
# ============================================================
#
# List comprehension:
#
#     squares = [
#         number ** 2
#         for number in numbers
#     ]
#
# Result:
#
#     [1, 4, 4, 9, 9, 16, 25]
#
#
# Set comprehension:
#
#     squares = {
#         number ** 2
#         for number in numbers
#     }
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "Transform every item and keep only unique results."
#
# Think:
#
#     {expression for item in iterable}
#
# The key difference is:
#
#     [ ] → list comprehension
#     { } → set comprehension
#
# ============================================================
```

```python
# ============================================================
# 68. NESTED LIST COMPREHENSION
# ============================================================
#
# A nested list comprehension is a list comprehension that
# contains more than one for loop.
#
# It is useful when working with nested structures such as:
#
#     list of lists
#     2D matrices
#     tables
#     nested data
#
# The general pattern is:
#
#     [value for outer_item in collection
#            for inner_item in outer_item]
#
# Think:
#
#     Outer loop → go through each row
#     Inner loop → go through each value in the row
#
# ============================================================


# ============================================================
# CREATE A 2D MATRIX
# ============================================================
#
# A matrix can be represented as a list containing multiple
# lists.
#
#     Row 1 → [1, 2]
#     Row 2 → [3, 4]
#     Row 3 → [5, 6]
#
# ============================================================

matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]


# ============================================================
# NESTED LIST COMPREHENSION
# ============================================================
#
# We want to flatten the matrix:
#
#     [[1, 2], [3, 4], [5, 6]]
#
# into:
#
#     [1, 2, 3, 4, 5, 6]
#
# The nested comprehension:
#
#     [
#         value
#         for row in matrix
#         for value in row
#     ]
#
# is equivalent to:
#
#     flattened = []
#
#     for row in matrix:
#
#         for value in row:
#
#             flattened.append(value)
#
# ============================================================

flattened = [
    value
    for row in matrix
    for value in row
]


# ============================================================
# PRINT RESULT
# ============================================================
#
# Output:
#
#     [1, 2, 3, 4, 5, 6]
#
# ============================================================

print(flattened)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see nested data:
#
#     [
#         [1, 2],
#         [3, 4],
#         [5, 6]
#     ]
#
# Think:
#
#     for each ROW
#         for each VALUE in that ROW
#
# In normal loops:
#
#     for row in matrix:
#         for value in row:
#             ...
#
# In a nested list comprehension:
#
#     [value for row in matrix for value in row]
#
# IMPORTANT:
#
# The order of the for clauses is the same as the order of
# the normal nested loops.
#
# ============================================================
```

```python
# ============================================================
# 69. NUMPY LOOP
# ============================================================
#
# A NumPy array can be iterated over using a normal Python
# for loop.
#
# The loop takes each element from the NumPy array one by one.
#
# Think:
#
#     NumPy array → one element → process it → next element
#
# Note:
#
# NumPy is designed for vectorized operations, so explicit
# Python loops are often not the best choice for numerical
# operations on large arrays.
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# Import NumPy so we can create and work with NumPy arrays.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE A NUMPY ARRAY
# ============================================================
#
# Create a NumPy array containing four numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40])


# ============================================================
# ITERATE THROUGH THE ARRAY
# ============================================================
#
# The for loop visits each element one at a time:
#
#     10 → number → print
#     20 → number → print
#     30 → number → print
#     40 → number → print
#
# ============================================================

for number in numbers:
    print(number)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every element in this array, do something."
#
# Think:
#
#     for number in numbers:
#
# This works because a NumPy array is iterable.
#
# However, if the operation can be expressed as a NumPy
# vectorized operation, prefer vectorization for numerical
# computations, especially with large arrays.
#
# Example:
#
#     # Loop
#     for number in numbers:
#         print(number ** 2)
#
#     # Vectorized
#     print(numbers ** 2)
#
# ============================================================
```

```python
# ============================================================
# 71. NUMPY BOOLEAN MASKING
# ============================================================
#
# Boolean masking is a powerful NumPy technique for filtering
# elements of an array based on a condition.
#
# The basic idea is:
#
#     array[condition]
#
# NumPy first evaluates the condition for every element and
# creates a Boolean mask containing True or False values.
#
#     True  → keep the element
#     False → exclude the element
#
# Example:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# This mask is then used to select the corresponding elements.
#
# Think:
#
#     Condition → Boolean mask → Filtered array
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# NumPy provides arrays and efficient operations for numerical
# and data-science computations.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE NUMPY ARRAY
# ============================================================
#
# Create an array containing five numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40, 50])


# ============================================================
# CREATE A BOOLEAN MASK
# ============================================================
#
# The condition:
#
#     numbers > 25
#
# is evaluated element by element:
#
#     10 > 25 → False
#     20 > 25 → False
#     30 > 25 → True
#     40 > 25 → True
#     50 > 25 → True
#
# Therefore:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# ============================================================

mask = numbers > 25

print(mask)


# ============================================================
# APPLY THE BOOLEAN MASK
# ============================================================
#
# Use the Boolean mask to select only the elements where
# the mask is True.
#
#     numbers[mask]
#
# returns:
#
#     [30, 40, 50]
#
# ============================================================

print(numbers[mask])


# ============================================================
# SHORT FORM
# ============================================================
#
# The mask does not need to be stored in a separate variable.
#
# We can directly write:
#
#     numbers[numbers > 25]
#
# This means:
#
#     "Select elements from numbers where numbers > 25."
#
# ============================================================

print(numbers[numbers > 25])


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see:
#
#     "Keep only the values that satisfy a condition."
#
# Think:
#
#     Boolean masking
#
# The general pattern is:
#
#     array[array condition]
#
# Examples:
#
#     numbers[numbers > 25]
#     numbers[numbers < 25]
#     numbers[numbers == 30]
#     numbers[numbers != 30]
#
# Boolean masking is extremely important in Data Science
# because the same idea appears in NumPy, Pandas, and data
# filtering.
#
# ============================================================
```

```python
# ============================================================
# 72. LOOP VS NUMPY VECTORIZATION
# ============================================================
#
# There are two common ways to perform the same operation
# on multiple values:
#
# 1. Python loop
# 2. NumPy vectorization
#
# A loop processes the elements one by one.
# Vectorization applies the operation to the entire array
# at once.
#
# In Data Science, vectorization is usually preferred because
# NumPy operations are implemented in optimized low-level code
# and are generally faster for large numerical datasets.
#
# Think:
#
#     Loop          → one element at a time
#     Vectorization → whole array at once
#
# ============================================================

numbers = np.array([1, 2, 3, 4, 5])


# ============================================================
# LOOP
# ============================================================
#
# The loop takes each number individually:
#
#     1 → 1²
#     2 → 2²
#     3 → 3²
#     4 → 4²
#     5 → 5²
#
# The result is gradually built inside a new list.
#
# ============================================================

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)


# ============================================================
# NUMPY VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to apply the operation directly to the
# entire array.
#
# No explicit Python loop is needed.
#
#     [1, 2, 3, 4, 5] ** 2
#              ↓
#     [1, 4, 9, 16, 25]
#
# ============================================================

squares = numbers ** 2

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every number, do this..."
#
# A Python loop is the natural starting point.
#
# But when working with NumPy arrays, ask:
#
#     "Can NumPy do this operation on the whole array?"
#
# If yes, prefer the vectorized NumPy operation in most
# data-processing situations.
#
# ============================================================
```

```python
# ============================================================
# VECTORWISE / VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to perform an operation on an entire array
# without writing an explicit Python for loop.
#
# This is called VECTORIZATION.
#
# Think:
#
#     Python loop:
#         one element → operation
#         one element → operation
#         one element → operation
#
#     NumPy vectorization:
#         entire array → operation
#
# NumPy performs the operation element by element internally,
# but the iteration is handled by optimized NumPy code rather
# than an explicit Python loop.
#
# ============================================================


# ============================================================
# CALCULATE SQUARES
# ============================================================
#
# The ** operator is applied to every element of the NumPy
# array.
#
# If:
#
#     numbers = [1, 2, 3, 4, 5]
#
# Then:
#
#     numbers ** 2
#
# produces:
#
#     [1, 4, 9, 16, 25]
#
# No explicit for loop is needed.
#
# ============================================================

squares = numbers ** 2


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with NumPy arrays, think:
#
#     "Can I apply this operation to the whole array?"
#
# If yes, NumPy often allows you to write:
#
#     array ** 2
#     array + 10
#     array * 5
#     array / 2
#
# instead of explicitly writing a Python loop.
#
# This usually makes numerical code:
#
#     ✓ shorter
#     ✓ clearer
#     ✓ easier to read
#     ✓ often faster for large numerical arrays
#
# ============================================================
```

```python
# ============================================================
# 73. LOOP VS PANDAS VECTORIZATION
# ============================================================
#
# Pandas provides vectorized operations that allow us to
# perform calculations on an entire column without explicitly
# writing a Python loop.
#
# There are two approaches:
#
#     1. Loop through rows
#     2. Vectorized column operation
#
# The loop processes data one row at a time.
# Vectorization applies the operation to the entire column.
#
# For typical numerical and column-based operations, the
# vectorized approach is usually preferred.
#
# ============================================================


# ============================================================
# CREATE DATAFRAME
# ============================================================
#
# Create a DataFrame containing employee salaries.
#
# ============================================================

df = pd.DataFrame({
    "Salary": [30000, 50000, 70000]
})


# ============================================================
# LOOP — PROCESS ROW BY ROW
# ============================================================
#
# iterrows() allows us to iterate through a DataFrame
# one row at a time.
#
#     index → current row index
#     row   → current row
#
# For each row, increase the salary by 10%.
#
#     Salary × 1.10
#
# ============================================================

for index, row in df.iterrows():

    print(row["Salary"] * 1.10)


# ============================================================
# VECTORIZED OPERATION
# ============================================================
#
# Pandas can perform the same calculation directly on the
# entire Salary column.
#
#     df["Salary"] * 1.10
#
# is applied element by element to the whole Series.
#
#     30000 × 1.10 → 33000
#     50000 × 1.10 → 55000
#     70000 × 1.10 → 77000
#
# The results are stored in a new column.
#
# ============================================================

df["New_Salary"] = df["Salary"] * 1.10


# ============================================================
# PRINT RESULT
# ============================================================

print(df)


# ============================================================
# LOOP VS VECTORIZATION
# ============================================================
#
# LOOP:
#
#     for index, row in df.iterrows():
#         ...
#
#     → processes one row at a time
#
#
# VECTORIZATION:
#
#     df["Salary"] * 1.10
#
#     → processes the entire column
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with Pandas and you think:
#
#     "For every row, calculate something."
#
# Don't immediately write a loop.
#
# First ask:
#
#     "Can Pandas do this to the entire column?"
#
# If yes, use the vectorized operation.
#
# Think:
#
#     Python loop → one row at a time
#     Pandas       → entire column
#
# This is an important shift from basic Python programming
# toward efficient Data Science programming.
#
# ============================================================
```

```python
# ============================================================
# 74. PRACTICAL ML EXAMPLE
# ============================================================
#
# A common Machine Learning task is converting predicted
# probabilities into class labels.
#
# For binary classification, a model may output a probability
# between 0 and 1 instead of directly returning 0 or 1.
#
# Example:
#
#     0.2 → low probability  → class 0
#     0.7 → high probability → class 1
#
# Here, we use 0.5 as the classification threshold:
#
#     probability >= 0.5 → prediction = 1
#     probability <  0.5 → prediction = 0
#
# Think:
#
#     Model probability
#            ↓
#       Apply threshold
#            ↓
#       Class prediction
#
# ============================================================


# ============================================================
# MODEL PREDICTION PROBABILITIES
# ============================================================
#
# These values represent the predicted probability of the
# positive class for five samples.
#
# In a real ML project, these probabilities might come from:
#
#     model.predict_proba(X)
#
# ============================================================

# List of predicted probabilities from a machine learning model
# Each value represents the model's confidence that a sample belongs to the positive class
probabilities = [0.2, 0.7, 0.9, 0.3, 0.8]


# ============================================================
# CREATE AN EMPTY LIST
# ============================================================
#
# We will store the final class predictions in this list.
#
# ============================================================

# Initialize an empty list to store the binary predictions
# This will hold the final class labels (0 or 1) after applying the threshold
predictions = []


# ============================================================
# LOOP THROUGH PROBABILITIES
# ============================================================
#
# Process each probability one at a time.
#
# The condition determines which class the probability
# belongs to.
#
#     probability >= 0.5 → 1
#     probability <  0.5 → 0
#
# Step by step:
#
#     0.2 → < 0.5 → 0
#     0.7 → >= 0.5 → 1
#     0.9 → >= 0.5 → 1
#     0.3 → < 0.5 → 0
#     0.8 → >= 0.5 → 1
#
# ============================================================

# Iterate through each probability in the list
for probability in probabilities:

    # Check if the probability is greater than or equal to the threshold (0.5)
    if probability >= 0.5:
        # If probability is high, assign class 1 (positive class)
        predictions.append(1)

    # Otherwise, the probability is below the threshold
    else:
        # If probability is low, assign class 0 (negative class)
        predictions.append(0)


# ============================================================
# PRINT PREDICTIONS
# ============================================================
#
# Output:
#
#     [0, 1, 1, 0, 1]
#
# ============================================================

# Print the final predictions
# This converts the probabilities into binary class labels based on the threshold
print(predictions)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a classic:
#
#     LOOP + CONDITION + RESULT COLLECTION
#
# Pattern:
#
#     result = []
#
#     for item in data:
#
#         if condition:
#             result.append(value_1)
#
#         else:
#             result.append(value_2)
#
# In Machine Learning, this pattern appears in tasks such as:
#
#     probability → class label
#     score       → decision
#     prediction  → category
#
# ============================================================
```

```python
# ============================================================
# 75. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================
#
# This example shows how a Machine Learning prediction can
# support a real-world engineering decision.
#
# Suppose an ML model predicts the probability that each water
# pipeline will fail.
#
# Instead of using only two classes:
#
#     0 → No failure
#     1 → Failure
#
# we can use the predicted probability to define different
# maintenance actions.
#
# Higher probability → higher risk → more urgent action.
#
# This is an example of:
#
#     ML prediction → Risk level → Engineering action
#
# ============================================================


# ============================================================
# FAILURE PROBABILITIES
# ============================================================
#
# Each value represents the predicted probability of pipeline
# failure.
#
#     0.15 → 15% probability
#     0.42 → 42% probability
#     0.73 → 73% probability
#     0.91 → 91% probability
#     0.31 → 31% probability
#
# In a real ML project, these values could come from a model
# such as:
#
#     model.predict_proba(X)
#
# ============================================================

failure_probabilities = [
    0.15,
    0.42,
    0.73,
    0.91,
    0.31
]


# ============================================================
# RISK-BASED DECISION LOGIC
# ============================================================
#
# Each probability is checked against several thresholds.
#
#     >= 0.80 → Immediate inspection
#     >= 0.60 → Schedule inspection
#     >= 0.40 → Monitor
#     <  0.40 → Normal operation
#
# Notice that the conditions are checked from HIGH to LOW.
#
# Once a condition is True, Python executes that block and
# skips the remaining elif / else conditions.
#
# ============================================================

for probability in failure_probabilities:

    if probability >= 0.80:
        action = "Immediate inspection"

    elif probability >= 0.60:
        action = "Schedule inspection"

    elif probability >= 0.40:
        action = "Monitor"

    else:
        action = "Normal operation"


    # ========================================================
    # DISPLAY RESULT
    # ========================================================
    #
    # Show the predicted probability together with the
    # corresponding engineering action.
    #
    # ========================================================

    print(
        probability,
        "->",
        action
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     0.15 -> Normal operation
#     0.42 -> Monitor
#     0.73 -> Schedule inspection
#     0.91 -> Immediate inspection
#     0.31 -> Normal operation
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a very useful pattern:
#
#     FOR each prediction:
#         IF high risk:
#             urgent action
#         ELIF medium-high risk:
#             planned action
#         ELIF moderate risk:
#             monitoring
#         ELSE:
#             normal operation
#
# In Data Science and ML, this connects:
#
#     DATA
#       ↓
#     ML MODEL
#       ↓
#     PROBABILITY
#       ↓
#     THRESHOLD / DECISION RULE
#       ↓
#     REAL-WORLD ACTION
#
# ============================================================


# ============================================================
# IMPORTANT ML INSIGHT
# ============================================================
#
# The thresholds (0.80, 0.60, 0.40) are only example values.
#
# In a real pipeline-failure prediction system, thresholds
# should ideally be determined using:
#
#     • historical failure data
#     • cost of missed failures
#     • cost of unnecessary inspections
#     • precision / recall
#     • ROC / PR analysis
#     • engineering constraints
#     • domain expertise
#
# Therefore:
#
#     Probability ≠ automatic decision
#
# A probability becomes an operational decision only after
# an appropriate decision threshold or risk policy is defined.
#
# ============================================================
```

```python
# ============================================================
# 76. PRACTICAL ML METRICS
# ============================================================
#
# Machine Learning models are often evaluated using multiple
# metrics rather than relying on a single score.
#
# Common classification metrics include:
#
#     Accuracy  → overall proportion of correct predictions
#     Precision → proportion of predicted positives that are
#                 actually positive
#     Recall    → proportion of actual positives that are found
#     F1        → harmonic mean of Precision and Recall
#
# Storing metrics in a dictionary makes it easy to organize
# and iterate through them.
#
# Think:
#
#     Metric name → Metric value
#
# ============================================================


# ============================================================
# CREATE A DICTIONARY OF METRICS
# ============================================================
#
# The dictionary stores each metric as a KEY-VALUE pair:
#
#     KEY   → metric name
#     VALUE → metric score
#
# The scores are represented as decimal values:
#
#     0.91 → 91%
#     0.84 → 84%
#     0.76 → 76%
#     0.80 → 80%
#
# ============================================================

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}


# ============================================================
# LOOP THROUGH THE DICTIONARY
# ============================================================
#
# .items() gives us both the key and value:
#
#     metric → "Accuracy"
#     value  → 0.91
#
#     metric → "Precision"
#     value  → 0.84
#
#     ...
#
# ============================================================

for metric, value in metrics.items():

    print(
        metric,
        ":",
        value
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     Accuracy : 0.91
#     Precision : 0.84
#     Recall : 0.76
#     F1 : 0.8
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you have a dictionary and need both:
#
#     KEY + VALUE
#
# Think:
#
#     for key, value in dictionary.items():
#
# This pattern is extremely common in Data Science and
# Machine Learning.
#
# For example:
#
#     for metric, value in metrics.items():
#         print(metric, value)
#
#     for parameter, value in model_params.items():
#         print(parameter, value)
#
#     for feature, importance in feature_importance.items():
#         print(feature, importance)
#
# ============================================================
```

```python
# ============================================================
# 77. CHECKING ML METRICS
# ============================================================
#
# This example evaluates several ML performance metrics using
# the same threshold-based decision rules.
#
# CORE PATTERN:
#
# metrics
#    ↓
# iterate through each metric
#    ↓
# evaluate its value
#    ↓
# choose a performance category
#    ↓
# print the result
#
# The important idea is that the SAME decision logic is applied
# independently to every metric.

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}

# CONCEPTUAL ROLES:
#
# metrics
#    → dictionary containing metric names and their scores
#
# metric
#    → current metric name (str)
#
# value
#    → current metric score (float)
#
# performance
#    → classification assigned according to the thresholds
#
# for
#    → processes each key-value pair in the dictionary
#
# if / elif / else
#    → converts a numeric score into a performance category

for metric, value in metrics.items():

    # Python evaluates the conditions from top to bottom.
    # Once one condition is True, its block executes and the
    # remaining elif/else branches are skipped.

    if value >= 0.90:
        performance = "Excellent"      # Score is at least 0.90

    elif value >= 0.80:
        performance = "Good"           # Score is 0.80–0.899...

    elif value >= 0.70:
        performance = "Acceptable"     # Score is 0.70–0.799...

    else:
        performance = "Poor"           # Score is below 0.70

    print(
        metric,
        "->",
        performance
    )

# ============================================================
# VALUE FLOW
# ============================================================
#
# Each iteration follows:
#
# metric + value
#       ↓
# compare value with thresholds
#       ↓
# True condition
#       ↓
# assign performance
#       ↓
# print metric + performance
#
# Example:
#
# Accuracy + 0.91
#       ↓
# 0.91 >= 0.90
#       ↓
# True
#       ↓
# performance = "Excellent"
#       ↓
# Accuracy -> Excellent

# ============================================================
# PROCESS TABLE
# ============================================================
#
# | Step | Metric    | Value | Selected Condition | Performance |
# |------|-----------|-------|--------------------|-------------|
# | 1    | Accuracy  | 0.91  | >= 0.90            | Excellent   |
# | 2    | Precision | 0.84  | >= 0.80            | Good        |
# | 3    | Recall    | 0.76  | >= 0.70            | Acceptable  |
# | 4    | F1        | 0.80  | >= 0.80            | Good        |
#
# Notice that 0.80 satisfies >= 0.80, so F1 is "Good".
#
# ============================================================
# EXECUTION TRACE
# ============================================================
#
# Iteration 1:
# metric = "Accuracy", value = 0.91
# 0.91 >= 0.90 → True
# performance = "Excellent"
# → Accuracy -> Excellent
#
# Iteration 2:
# metric = "Precision", value = 0.84
# 0.84 >= 0.90 → False
# 0.84 >= 0.80 → True
# performance = "Good"
# → Precision -> Good
#
# Iteration 3:
# metric = "Recall", value = 0.76
# 0.76 >= 0.90 → False
# 0.76 >= 0.80 → False
# 0.76 >= 0.70 → True
# performance = "Acceptable"
# → Recall -> Acceptable
#
# Iteration 4:
# metric = "F1", value = 0.80
# 0.80 >= 0.90 → False
# 0.80 >= 0.80 → True
# performance = "Good"
# → F1 -> Good
#
# No more dictionary items → loop ends.

# ============================================================
# STATE
# ============================================================
#
# The dictionary remains unchanged.
#
# performance is reassigned on every iteration:
#
# "Excellent" → "Good" → "Acceptable" → "Good"
#
# The loop variable values also change:
#
# metric:
# "Accuracy" → "Precision" → "Recall" → "F1"
#
# value:
# 0.91 → 0.84 → 0.76 → 0.80
#
# ============================================================
# GENERALIZABLE PATTERN
# ============================================================
#
# This is a common DATA ANALYTICS / ML pattern:
#
# data
#   ↓
# iterate through observations
#   ↓
# apply rules
#   ↓
# classify each value
#   ↓
# report results
#
# The same structure can be used for:
# - model quality checks
# - data validation
# - risk levels
# - customer segmentation
# - alert systems
# - KPI classification
#
# ============================================================
# CORE IDEA
# ============================================================
#
# A conditional chain converts numeric metric values into
# meaningful categories.
#
# The for loop allows the same rules to be applied consistently
# to every metric.
```
