

Example:

    # Heading 1

````python
# ============================================================
# 2. EMPHASIS
# ============================================================

# Bold and italic can be combined.

# Bold:
# **bold text** or __bold text__

# Italic:
# *italic text* or _italic text_

# Bold and italic:
# ***bold and italic*** or ___bold and italic___

````

````python
# ============================================================
# 11. MATHEMATICAL FORMULATION (LATEX)
# ============================================================
#
# Supported by: Obsidian, Typora, Jupyter, GitLab, Notion, 
#               Hugo, and editors with MathJax/KaTeX.
# NOT supported by: Plain GitHub READMEs (without plugins).
#
# Two main modes:
#
# Inline math  (within a sentence)
# Display math (centered on its own line)
#
# ============================================================


# ============================================================
# 11a. INLINE MATH
# ============================================================

# # Wrap math in single dollar signs or \( \)
# # Example:
# The area of a circle is $A = \pi r^2$.

# # Or using the LaTeX-style parentheses:
# The quadratic formula is \(x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}\).
````

>The quadratic formula is \(x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}\).
