**Bias-variance decomposition**

Expected test error:

$\mathbb{E}\left[(y - \hat{f}(x))^2\right] = \text{Bias}^2 + \text{Variance} + \text{Noise}$

Where:
- Bias² = $(\mathbb{E}[\hat{f}(x)] - f(x))^2$
- Variance = $\mathbb{E}\left[(\hat{f}(x) - \mathbb{E}[\hat{f}(x)])^2\right]$
- Noise = irreducible error
