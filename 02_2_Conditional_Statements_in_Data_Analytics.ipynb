### Independent IFs (Sepatare ifs)

```python
# ============================================================
#                    INDEPENDENT IF STATEMENTS
# ============================================================
#
# Independent if statements are multiple separate if statements.
#
# Each if is checked separately.
#
# IMPORTANT:
#
# All conditions are tested,
# even if one condition is already True.
#
# ============================================================


# ============================================================
# 1. BASIC EXAMPLE
# ============================================================

age = 25

if age >= 18:
    print("You are an adult.")

if age >= 21:
    print("You can legally drive a Car.")

if age >= 65:
    print("You are a senior.")


# Output:
#
# You are an adult.
# You can legally drive in some countries.
#
# Each if was checked separately.
```

```python
# ============================================================
# 2. MULTIPLE CONDITIONS CAN BE TRUE
# ============================================================

score = 95

if score >= 50:
    print("You passed.")

if score >= 70:
    print("Good score. Highter than Middle!")

if score >= 90:
    print("Excellent score.!!!")


# Output:
#
# You passed.
# Good score. Highter than Middle!
# Excellent score.!!!
#
# ALL three conditions are True,
# so ALL three blocks execute.
```

```python
# ============================================================
# 3. ALL CONDITIONS ARE TESTED
# ============================================================

x = 10

if x > 5:
    print("Condition 1 is True.")

if x > 8:
    print("Condition 2 is True.")

if x > 20:
    print("Condition 3 is True.")


# Output:
#
# Condition 1 is True.
# Condition 2 is True.
#
# Python still checks Condition 3,
# even though the previous conditions were True.
```

```python
x = 10

if x > 5:
    print("Condition 1 Tested and it is True.")
else:
    print("Condition 1 Tested and it is False.")


if x > 8:
    print("Condition 2 Tested and it is True.")
else:
    print("Condition 2 Tested and it is False.")

if x > 20:
    print("Condition 3 Tested and it is True.")
else:
    print("Condition 3 Tested and it is False.")
```

```python
# ============================================================
# 4.1 INDEPENDENT if VS if / elif / else
# ============================================================

score = 95


# ------------------------------------------------------------
# Independent if statements
# ------------------------------------------------------------

if score >= 50:
    print("Passed.")

if score >= 70:
    print("Good.")

if score >= 90:
    print("Excellent.")


# Output:
#
# Passed.
# Good.
# Excellent.
# ALL block execute.
# Independent ifs


# ────────────────────────────────
# if A → check
# if B → check
# if C → check

# A, B, and C can ALL execute.
```

```python
# ============================================================
# 4.2 INDEPENDENT if VS if / elif / else
# ============================================================

# ------------------------------------------------------------
# if / elif / else
# ------------------------------------------------------------

if score >= 90:
    print("Excellent.")
elif score >= 70:
    print("Good.")
elif score >= 50:
    print("Passed.")
else:
    print("Failed.")


# Output:
#
# Excellent.
#
# Only ONE block executes.
#
# Once Python finds the first True condition,
# it skips the remaining elif/else conditions.


# if / elif / else
# ────────────────────────────────
# if A → check
#    ↓ False
# elif B → check
#    ↓ True
# elif C → SKIP
# else → SKIP

# Only ONE branch executes.
```

```python
# Independent ifs
# ────────────────────────────────
# if A → check
# if B → check
# if C → check

# A, B, and C can ALL execute.


# if / elif / else
# ────────────────────────────────
# if A → check
#    ↓ False
# elif B → check
#    ↓ True
# elif C → SKIP
# else → SKIP

# Only ONE branch executes.
```

```python
# ============================================================
# INDEPENDENT ifs VS if / elif / else
# ============================================================
#
# The main difference:
#
#     Independent `if`s       → CHECK EVERY condition
#     `if / elif / else`      → STOP after the FIRST True
#
#
# ============================================================
# 1. INDEPENDENT ifs
# ============================================================
#
# Each `if` is separate and is checked independently.
#
# Diagram:
#
#     Start
#       │
#       ▼
#     Check A ── True ──→ Run A
#       │
#       ▼
#     Check B ── True ──→ Run B
#       │
#       ▼
#     Check C ── False
#       │
#       ▼
#      End
#
# A, B, and C can ALL be True.
# Python does NOT stop after one `if` runs.
# ============================================================

score = 85

if score >= 50:
    print("Passed")

if score >= 80:
    print("Good score")

if score >= 90:
    print("Excellent score")


# Output:
# Passed
# Good score
#
# Execution:
#
#     85 >= 50 → True  → Run
#     85 >= 80 → True  → Run
#     85 >= 90 → False → Skip
#
#
# ============================================================
# 2. if / elif / else
# ============================================================
#
# These conditions belong to ONE decision structure.
#
# Diagram:
#
#             Start
#               │
#               ▼
#            Check A?
#           /        \
#        True        False
#         │            │
#       Run A          ▼
#                  Check B?
#                 /        \
#              True        False
#               │            │
#             Run B          ▼
#                        Check C?
#                       /        \
#                    True        False
#                     │            │
#                   Run C        else
#                                  │
#                                  ▼
#                                 End
#
# IMPORTANT:
# As soon as one condition is True, Python runs that branch
# and skips all remaining `elif` / `else` branches.
# ============================================================

score = 85

if score >= 90:
    print("Excellent score")

elif score >= 80:
    print("Good score")

elif score >= 50:
    print("Passed")

else:
    print("Failed")


# Output:
# Good score
#
# Execution:
#
#     85 >= 90 → False
#          ↓
#     85 >= 80 → True
#          ↓
#     Run "Good score"
#          ↓
#     Skip remaining branches
#
#
# ============================================================
# QUICK COMPARISON
# ============================================================
#
# Independent `if`s:
#
#     if A → check
#     if B → check
#     if C → check
#
#     → Multiple conditions can run.
#
#
# if / elif / else:
#
#     if A
#       ↓ False
#     elif B
#       ↓ True
#     Run B
#       ↓
#     Skip everything below
#
#     → Only ONE branch can run.
#
#
# ============================================================
# EASY WAY TO REMEMBER
# ============================================================
#
#     Independent if:
#         "CHECK ALL"
#
#     if / elif / else:
#         "FIRST TRUE WINS"
#
#
# CORE IDEA:
# Use separate `if`s when several conditions may need to run.
# Use `if / elif / else` when you want exactly ONE outcome.
#
# ============================================================
```

#### ⚠ A good rule to put in your notes is:

- Use separate ifs when More than ONE condition can be True.
- Use if/elif/else when Only ONE condition should be True.

```python
# ============================================================
# 5. IMPORTANT DIFFERENCE
# ============================================================
#
# Independent if:
#
#     if condition_1:
#         code
#
#     if condition_2:
#         code
#
#     if condition_3:
#         code
#
# Multiple conditions can execute.
#
#
# if / elif / else:
#
#     if condition_1:
#         code
#     elif condition_2:
#         code
#     elif condition_3:
#         code
#     else:
#         code
#
# Only ONE branch executes.
#
# ============================================================


# ============================================================
# QUICK SUMMARY
# ============================================================
#
# INDEPENDENT ifs:
#
#     Each if is checked separately.
#     All conditions are tested.
#     Multiple conditions can be True.
#     Multiple blocks can execute.
#
#
# if / elif / else:
#
#     Conditions are checked from top to bottom.
#     Once a condition is True, the remaining branches
#     are skipped.
#     Only one branch executes.
#
# ============================================================
```

```python
# ============================================================
# 35. DATA ANALYTICS — OUTLIER LOGIC
# ============================================================
#
# This example uses the IQR (Interquartile Range) method
# to identify a POTENTIAL outlier.
#
# Key concepts:
#
#     Q1  → 25th percentile
#     Q3  → 75th percentile
#     IQR → Q3 - Q1
#
# Common outlier rule:
#
#     Lower limit = Q1 - 1.5 × IQR
#     Upper limit = Q3 + 1.5 × IQR
#
# A value is a potential outlier if it is:
#
#     below the lower limit OR above the upper limit
#
# ============================================================

salary = 250000

# Q1: 25% of the data is below this value.
Q1 = 40000

# Q3: 75% of the data is below this value.
Q3 = 80000

# IQR measures the spread of the middle 50% of the data.
IQR = Q3 - Q1
# IQR = 80000 - 40000 = 40000


# Calculate the boundaries for potential outliers.
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Values:
#
#     lower_limit = 40000 - 1.5 × 40000 = -20000
#     upper_limit = 80000 + 1.5 × 40000 = 140000
#
# Therefore, the expected range according to this rule is:
#
#     -20,000 ≤ value ≤ 140,000


# Check whether salary falls outside the boundaries.
#
# `or` means at least ONE condition must be True.
if salary < lower_limit or salary > upper_limit:

    # 250000 > 140000 → True
    # Therefore, salary is outside the upper boundary.
    print("Potential outlier.")

else:
    print("Within expected range.")


# Output:
# Potential outlier.


# ============================================================
# QUICK VISUAL
# ============================================================
#
#     Lower limit                         Upper limit
#          │                                  │
#          ▼                                  ▼
#     -20,000 ─────────────────────────── 140,000
#          │                                  │
#          │       Expected range             │
#          │                                  │
#          └──────────────────────────────────┘
#                                               ↑
#                                           250,000
#                                           OUTLIER
#
#
# ============================================================
# EXECUTION FLOW
# ============================================================
#
#     Q1 = 40,000
#     Q3 = 80,000
#          ↓
#     IQR = 40,000
#          ↓
#     Calculate limits
#          ↓
#     Lower = -20,000
#     Upper = 140,000
#          ↓
#     Is salary < -20,000?
#         False
#          ↓
#     Is salary > 140,000?
#         True
#          ↓
#     "Potential outlier."
#
#
# CORE IDEA:
# The IQR method identifies potential outliers by checking
# whether a value falls unusually far below Q1 or above Q3.
#
# IMPORTANT:
# An "outlier" here means a value flagged by the chosen rule.
# It does NOT automatically mean the value is wrong or invalid.
#
# ============================================================
```

```python
# ============================================================
# 42. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [10, 20, 30, 40, 50, 60, 7, 18, 90, 100]  # Define a list of numbers

# Create a new list containing only numbers greater than 25 using list comprehension
large_numbers = [
    number      # Each number from the original list that meets the condition
    for number in numbers  # Iterate through each number in the original list
    if number > 25  # Condition to filter numbers
]

print(large_numbers)  # Print the filtered list containing only numbers > 25
```

```python
# ============================================================
# 44. CONDITIONAL LOGIC WITH PANDAS
# ============================================================

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sina", "Yasin", "John"],
    "Age": [25, 32, 41, 19],
    "Salary": [35000, 55000, 90000, 28000]
})

print(df)
```

```python

# ============================================================
# 45. PANDAS BOOLEAN FILTERING
# ============================================================
print("show people with age greater than 30:")
print(df[df["Age"] >= 30])


print("\nshow people with salary greater than 50000:")
print(df[df["Salary"] > 50000])
```

```python
# ============================================================
# 46. PANDAS AND CONDITION
# ============================================================

# This code prints a subset of a pandas DataFrame (df) that meets specific conditions
# The conditions are:
# 1. Age is greater than or equal to 30
# 2. Salary is greater than 50000
# The '&' operator is used to combine these conditions with a logical AND
print(
    df[
        (df["Age"] >= 30) &  # Filter rows where Age is 30 or more
        (df["Salary"] > 50000)  # AND Salary is greater than 50000
    ]
)
```

```python
# ============================================================
# 47. PANDAS OR CONDITION
# ============================================================

print(
    df[
        (df["Age"] < 25) |
        (df["Salary"] > 80000)
    ]
)
```

```python
# ============================================================
# 48. PANDAS NOT CONDITION
# ============================================================

print(
    df[
        ~(df["Age"] >= 30)
    ]
)
```

```python
# ============================================================
# 49. PANDAS np.where()
# ============================================================

df["Age_Group"] = np.where(
    df["Age"] >= 30,
    "30+",
    "Under 30"
)

print(df)
```

```python
# ============================================================
# 50. PANDAS MULTI
# PLE CONDITIONS WITH np.select()
# ============================================================

conditions = [
    df["Salary"] < 40000,
    df["Salary"] < 70000,
    df["Salary"] >= 70000
]

choices = [
    "Low",
    "Medium",
    "High"
]

df["Salary_Level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df)
```

```python
# ============================================================
# 51. PRACTICAL ANALYTICS EXAMPLE
# ============================================================

df["Performance"] = np.where(
    df["Salary"] >= 50000,
    "High",
    "Low"
)

print(df)
```

```python
# ============================================================
# 52. PRACTICAL DATA QUALITY EXAMPLE
# ============================================================

df["Age_Status"] = np.where(
    (df["Age"] >= 18) & (df["Age"] <= 100),
    "Valid",
    "Check"
)

print(df)
```

```python
# ============================================================
# 53. CONDITIONS WITH FUNCTIONS
# ============================================================

def classify_salary(salary):

    if salary < 40000:
        return "Low"

    elif salary < 70000:
        return "Medium"

    else:
        return "High"


print(classify_salary(35000))
print(classify_salary(60000))
print(classify_salary(90000))
```

```python
# ============================================================
# 54. APPLYING A FUNCTION TO A DATAFRAME
# ============================================================

df["Salary_Level"] = df["Salary"].apply(
    classify_salary
)

print(df)
```
