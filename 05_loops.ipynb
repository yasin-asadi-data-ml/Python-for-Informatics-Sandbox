============================================================
                        PYTHON LOOPS
============================================================

A LOOP is a programming structure used to REPEAT code.

Instead of writing the same instructions again and again,
we write them once and tell Python when and how to repeat
them.

The fundamental idea is:

    "Repeat this operation."

There are two main questions when thinking about loops:

    FOR   → "For **each** item, what should I do?"

    WHILE → "While this condition is True, what should I do?"


> learn to say **FOR EACH** instead of **FOR**
============================================================

1. FOR LOOP
============================================================

A for loop is mainly used to **ITERATE** over an **iterable**.

An **iterable** is something Python can go through one item
at a time, such as:

    list
    tuple
    string
    dictionary
    set
    range()
    NumPy array
    Pandas Series

Basic pattern:

    for item in iterable:
        do_something()

Think:

    COLLECTION
        ↓
    ITEM 1 → ACTION
    ITEM 2 → ACTION
    ITEM 3 → ACTION
    ...

============================================================

2. WHILE LOOP
============================================================

A while loop repeats code as long as a condition is True.

Basic pattern:

    while condition:
        do_something()

Think:

    CONDITION?
       ↓
     True  → repeat
       ↓
    CONDITION?
       ↓
     False → stop

3. IMPORTANT LOOP TOOLS
============================================================

    range(5)

enumerate()

    for index, value in enumerate(data):

else

4. NESTED LOOPS
============================================================

A nested loop is a loop inside another loop.

    for i in collection:

        for j in collection:
            ...

Think:

    FOR every A
        FOR every B
            do something

Common uses:

    • Matrices
    • Tables
    • Rows and columns
    • Images and pixels
    • Combinations
    • Parameter combinations

5. COMPREHENSIONS
============================================================

loops.

DICTIONARY COMPREHENSION:

    {key: value for item in iterable}


Comprehensions are especially useful when the loop is simple
and its purpose is to CREATE A NEW COLLECTION.

9. LOOPS VS VECTORIZATION
============================================================

In Data Science, an important skill is knowing when NOT to
use a Python loop.

For numerical data, NumPy often allows vectorized operations:

    numbers ** 2

instead of:

    for number in numbers:
        ...

Similarly, Pandas can operate on entire columns:

    df["Salary"] * 1.10

instead of processing every row individually.

Think:

    Python loop
        → one item / row at a time

    NumPy / Pandas vectorization
        → operate on the whole array / column

Vectorization is generally preferred for operations that
NumPy or Pandas can express directly.

```python
# ============================================================
# 34. MULTIPLICATION TABLE
# ============================================================
#
# Nested loops mean:
#
#     "Put one loop inside another loop."
#
# The OUTER loop controls the first value.
# The INNER loop controls the second value.
#
# The important idea is:
#
#     For every i
#         run the ENTIRE j loop
#
# Therefore, if the outer loop runs 5 times and the inner
# loop runs 5 times for each outer iteration:
#
#     5 × 5 = 25 iterations
#
# ============================================================


# ============================================================
# OUTER LOOP
# ============================================================
#
# range(1, 6) produces:
#
#     1, 2, 3, 4, 5
#
# The variable i takes each value one at a time.
#
# ============================================================

for i in range(1, 6):


    # ========================================================
    # INNER LOOP
    # ========================================================
    #
    # For EACH value of i, the inner loop runs completely.
    #
    # range(1, 6) produces:
    #
    #     1, 2, 3, 4, 5
    #
    # The variable j takes each value one at a time.
    #
    # ========================================================

    for j in range(1, 6):


        # ====================================================
        # MULTIPLY AND PRINT
        # ====================================================
        #
        # i and j represent the current values from the two
        # loops.
        #
        #     i * j
        #
        # calculates their product.
        #
        # ====================================================

        print(
            i,
            "*",
            j,
            "=",
            i * j
        )


# ============================================================
# HOW THE NESTED LOOP MOVES
# ============================================================
#
# The execution happens like this:
#
#     i = 1
#         j = 1 → 1 × 1
#         j = 2 → 1 × 2
#         j = 3 → 1 × 3
#         j = 4 → 1 × 4
#         j = 5 → 1 × 5
#
#     i = 2
#         j = 1 → 2 × 1
#         j = 2 → 2 × 2
#         j = 3 → 2 × 3
#         j = 4 → 2 × 4
#         j = 5 → 2 × 5
#
#     i = 3
#         ...
#
# The inner loop completes ALL of its iterations before the
# outer loop moves to its next value.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every A, do something with every B."
#
# Think:
#
#     for A in collection_A:
#         for B in collection_B:
#             ...
#
# In this example:
#
#     FOR each i
#         FOR each j
#             multiply i and j
#
# ============================================================
#
# A useful mental picture:
#
#     i = 1 ──→ j = 1, 2, 3, 4, 5
#     i = 2 ──→ j = 1, 2, 3, 4, 5
#     i = 3 ──→ j = 1, 2, 3, 4, 5
#     i = 4 ──→ j = 1, 2, 3, 4, 5
#     i = 5 ──→ j = 1, 2, 3, 4, 5
#
# Total iterations:
#
#     5 × 5 = 25
#
# ============================================================
#
# REAL-WORLD CONNECTION
#
# Nested loops are useful when working with two-dimensional
# structures, such as:
#
#     • rows and columns
#     • matrices
#     • tables
#     • images (pixels)
#     • combinations of two datasets
#     • hyperparameter combinations in ML
#
# ============================================================
```

```python
# ============================================================
# 34. MULTIPLICATION TABLE
# ============================================================
#
# Nested loops are used to repeat one loop inside another.
#
# The outer loop controls the first number.
# The inner loop controls the second number.
#
# For every value of i, the inner loop runs through all values
# of j.
#
# ============================================================

# Outer loop for the first number of multiplication (1-5)
for i in range(1, 6):

    # Inner loop for the second number of multiplication (1-5)
    for j in range(1, 6):

        # Print the multiplication expression and result
        print(i, "*", j, "=", i * j)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# Think:
#
#     FOR each i
#         FOR each j
#             do something
#
# Example:
#
#     i = 1 → j = 1, 2, 3, 4, 5
#     i = 2 → j = 1, 2, 3, 4, 5
#     i = 3 → j = 1, 2, 3, 4, 5
#     ...
#
# ============================================================
```

```python
# Outer loop for the first number of multiplication (1-5)
mult_matrix = []

# Iterate through rows (first multiplication factor)
for i in range(1, 6):
    row = []

    # Inner loop for the second number of multiplication (1-5)
    # Iterate through columns (second multiplication factor)
    for j in range(1, 6):
        # Calculate the product and add it to the current row
        row.append(i * j)

    # Add the completed row to the matrix
    mult_matrix.append(row)

# Print the resulting 5x5 multiplication matrix
print(f"mult_matrix: {mult_matrix}")

for row in mult_matrix:
    for value in row:
        print("\n",value)
```

```python
# ============================================================
# 41. LOOP ELSE
# ============================================================

numbers = [1, 2, 3, 4, 5]  # Initialize a list of numbers

for number in numbers:  # Iterate through each number in the list

    print(number)  # Print the current number

else:

    # This else block executes after the loop completes normally (without break)
    print("Loop completed.")  # Indicate successful loop completion

```

```python
# ============================================================
# 42. LOOP ELSE WITH BREAK
# ============================================================

numbers = [1, 2, 3, 4, 5] #  List of numbers to iterate through

for number in numbers: #  Iterate through each number in the list

    if number == 3: #  Check if current number is 3
        break #  Exit the loop if number is 3
    
    print(number) #  Print the current number if not 3

else:
    print("Loop completed without break.") #  This else block executes only if the loop completes without break

```

```python
# ============================================================
# 46. DATA ANALYTICS — CLASSIFYING VALUES
# ============================================================

salaries = [30000, 45000, 70000, 90000, 120000]

for salary in salaries:

    if salary < 40000:
        category = "Low"

    elif salary < 70000:
        category = "Medium"

    else:
        category = "High"

    print(salary, category)
```

```python
# ============================================================
# 59. LOOPING THROUGH COLUMNS FOR DATA QUALITY
# ============================================================
#
# Before analyzing or building a Machine Learning model, we
# need to check the quality of our data.
#
# One important data-quality check is finding MISSING VALUES.
#
# Pandas provides:
#
#     .isnull()
#
# to identify missing values.
#
#     .sum()
#
# then counts how many missing values exist.
#
# The pattern is:
#
#     df[column].isnull().sum()
#
# Think:
#
#     DataFrame
#         ↓
#     One column at a time
#         ↓
#     Find missing values
#         ↓
#     Count them
#
# ============================================================


# ============================================================
# LOOP THROUGH COLUMNS
# ============================================================
#
#     df.columns
#
# provides the names of all columns in the DataFrame.
#
# The loop processes each column one at a time.
#
# ============================================================

for column in df.columns:


    # ========================================================
    # COUNT MISSING VALUES
    # ========================================================
    #
    #     df[column]
    #
    # selects the current column.
    #
    #     .isnull()
    #
    # checks every value and returns:
    #
    #     True  → value is missing
    #     False → value is not missing
    #
    #     .sum()
    #
    # counts the True values because:
    #
    #     True  = 1
    #     False = 0
    #
    # Therefore:
    #
    #     df[column].isnull().sum()
    #
    # gives the total number of missing values.
    #
    # ========================================================

    missing = df[column].isnull().sum()


    # ========================================================
    # PRINT RESULT
    # ========================================================
    #
    # Display the column name together with its number of
    # missing values.
    #
    # ========================================================

    print(
        column,
        "missing values:",
        missing
    )


# ============================================================
# EXAMPLE
# ============================================================
#
# Suppose we have:
#
#     Name      → 0 missing values
#     Age       → 2 missing values
#     Salary    → 5 missing values
#     Department → 1 missing value
#
# The output might be:
#
#     Name missing values: 0
#     Age missing values: 2
#     Salary missing values: 5
#     Department missing values: 1
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every column, check something."
#
# Think:
#
#     for column in df.columns:
#         ...
#
# Then access the current column:
#
#     df[column]
#
# In this example:
#
#     df[column].isnull().sum()
#
# means:
#
#     "For the current column, find and count its missing
#      values."
#
# ============================================================


# ============================================================
# DATA SCIENCE TIP
# ============================================================
#
# Pandas can check ALL columns at once:
#
#     df.isnull().sum()
#
# This is usually simpler and faster than explicitly writing
# a Python loop.
#
# The loop is useful when you want to add custom logic, such as:
#
#     • reporting only columns with missing values
#     • calculating missing-value percentages
#     • applying different rules to different columns
#     • automatically deciding how to handle missing data
#
# ============================================================
```

```python
# ============================================================
# 60. LOOPING THROUGH COLUMNS FOR UNIQUE VALUES
# ============================================================
#
# In Data Science, it is often useful to inspect each column
# and find out how many different values it contains.
#
# Pandas provides:
#
#     .nunique()
#
# to count the number of unique values in a column.
#
# The pattern is:
#
#     for column in df.columns:
#         df[column].nunique()
#
# Think:
#
#     DataFrame
#         ↓
#     Get each column
#         ↓
#     Count unique values
#         ↓
#     Print the result
#
# ============================================================


# ============================================================
# LOOP THROUGH COLUMNS
# ============================================================
#
#     df.columns
#
# provides the names of all columns in the DataFrame.
#
# The loop processes one column at a time.
#
# ============================================================

for column in df.columns:

    # ========================================================
    # COUNT UNIQUE VALUES
    # ========================================================
    #
    #     df[column]
    #
    # selects the current column.
    #
    #     .nunique()
    #
    # counts how many distinct values exist in that column.
    #
    # ========================================================

    print(
        column,
        "unique values:",
        df[column].nunique()
    )


# ============================================================
# EXAMPLE
# ============================================================
#
# Suppose the DataFrame contains:
#
#     Gender  → Male, Female
#     Age     → many different values
#     City    → Tehran, Karaj, Shiraz
#
# The output might be:
#
#     Gender unique values: 2
#     Age unique values: 35
#     City unique values: 3
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every column, perform an operation."
#
# Think:
#
#     for column in df.columns:
#         ...
#
# Then access the current column using:
#
#     df[column]
#
# In this example:
#
#     df[column].nunique()
#
# means:
#
#     "For the current column, count its distinct values."
#
# ============================================================


# ============================================================
# DATA SCIENCE TIP
# ============================================================
#
# Pandas can also calculate the number of unique values for
# ALL columns at once:
#
#     df.nunique()
#
# So the loop is mainly useful when you want to perform
# additional customized operations for each column.
#
# ============================================================
```

```python
# ============================================================
# 62. LIST COMPREHENSION
# ============================================================
#
# A list comprehension is a concise way to create a new list
# by applying an expression to each item in an iterable.
#
# The basic pattern is:
#
#     [expression for item in iterable]
#
# Think:
#
#     "For each item, calculate something and put the result
#      into a new list."
#
# List comprehensions are essentially a compact form of a
# simple for loop.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This is the original list that we want to process.
#
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# LIST COMPREHENSION
# ============================================================
#
# For each number in numbers:
#
#     number ** 2
#
# calculates its square.
#
# Step by step:
#
#     1 → 1² → 1
#     2 → 2² → 4
#     3 → 3² → 9
#     4 → 4² → 16
#     5 → 5² → 25
#
# The results are automatically collected into a new list.
#
# ============================================================

squares = [
    number ** 2
    for number in numbers
]


# ============================================================
# PRINT RESULT
# ============================================================
#
# Output:
#
#     [1, 4, 9, 16, 25]
#
# ============================================================

print(squares)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The list comprehension above is equivalent to:
#
#     squares = []
#
#     for number in numbers:
#         squares.append(number ** 2)
#
# List comprehension simply expresses this common pattern
# in a more compact way.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, transform it and create a new list."
#
# Think:
#
#     [expression for item in iterable]
#
# Examples:
#
#     [number * 2 for number in numbers]
#     [number ** 2 for number in numbers]
#     [word.upper() for word in words]
#
# The key idea:
#
#     INPUT LIST
#          ↓
#     FOR EACH ITEM
#          ↓
#     TRANSFORM
#          ↓
#     NEW LIST
#
# ============================================================

```

```python
# ============================================================
# 63. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]  # Original list of numbers

# Using list comprehension to create a new list containing only even numbers
# The list comprehension iterates through each number in the original list
# and includes it in the new list only if it satisfies the condition (number % 2 == 0)
even_numbers = [
    number  # Current number being evaluated
    for number in numbers  # Iterate through each number in the original list
    if number % 2 == 0  # Condition to include only even numbers
]

print(even_numbers)

```

```python
# ============================================================
# 65. LIST COMPREHENSION — IF ELSE
# ============================================================
#
# A list comprehension can contain an if-else expression to
# produce different results depending on a condition.
#
# The basic pattern is:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# IMPORTANT:
#
# In this form, the if-else expression comes BEFORE the
# for loop.
#
# Think:
#
#     For each item:
#         if condition → return this
#         else         → return that
#
# ============================================================


# ============================================================
# CREATE A LIST OF SCORES
# ============================================================
#
# Each score will be checked to determine whether the student
# passed or failed.
#
# Passing score:
#
#     50 or higher → Pass
#
# Below 50:
#
#     Fail
#
# ============================================================

scores = [45, 67, 82, 91, 38]


# ============================================================
# LIST COMPREHENSION WITH IF ELSE
# ============================================================
#
# For each score:
#
#     score >= 50 → "Pass"
#     otherwise   → "Fail"
#
# Step by step:
#
#     45 → 45 >= 50 → False → "Fail"
#     67 → 67 >= 50 → True  → "Pass"
#     82 → 82 >= 50 → True  → "Pass"
#     91 → 91 >= 50 → True  → "Pass"
#     38 → 38 >= 50 → False → "Fail"
#
# Result:
#
#     ["Fail", "Pass", "Pass", "Pass", "Fail"]
#
# ============================================================

results = [
    "Pass" if score >= 50 else "Fail"
    for score in scores
]


# ============================================================
# PRINT RESULT
# ============================================================

print(results)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The list comprehension above is equivalent to:
#
#     results = []
#
#     for score in scores:
#
#         if score >= 50:
#             results.append("Pass")
#         else:
#             results.append("Fail")
#
# The comprehension combines the same logic into one
# expression.
#
# ============================================================


# ============================================================
# IMPORTANT: IF-ELSE VS IF-ONLY
# ============================================================
#
# IF-ONLY:
#
#     [expression
#      for item in iterable
#      if condition]
#
# The if comes AFTER the for because it FILTERS items.
#
#
# IF-ELSE:
#
#     [value_if_true if condition else value_if_false
#      for item in iterable]
#
# The if-else comes BEFORE the for because it TRANSFORMS
# every item into one of two possible values.
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, choose A or B."
#
# Think:
#
#     [A if condition else B for item in iterable]
#
#
# When you think:
#
#     "Keep only the items that satisfy a condition."
#
# Think:
#
#     [item for item in iterable if condition]
#
# The key difference:
#
#     IF ONLY → FILTER
#     IF ELSE → CHOOSE / TRANSFORM
#
# ============================================================
```

```python
# ============================================================
# 66. DICTIONARY COMPREHENSION
# ============================================================
#
# A dictionary comprehension is a compact way to create a
# dictionary by generating KEY-VALUE pairs from an iterable.
#
# The basic pattern is:
#
#     {key: value for item in iterable}
#
# It is similar to a list comprehension, but instead of
# producing individual values, it produces:
#
#     KEY → VALUE
#
# Dictionary comprehensions are especially useful when you
# want to transform data into a dictionary in a single,
# readable expression.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains the numbers that we want to process.
#
# ============================================================

numbers = [1, 2, 3, 4, 5]


# ============================================================
# DICTIONARY COMPREHENSION
# ============================================================
#
# For each number:
#
#     number       → KEY
#     number ** 2  → VALUE
#
# Step by step:
#
#     1 → 1² → 1  → 1: 1
#     2 → 2² → 4  → 2: 4
#     3 → 3² → 9  → 3: 9
#     4 → 4² → 16 → 4: 16
#     5 → 5² → 25 → 5: 25
#
# Result:
#
#     {
#         1: 1,
#         2: 4,
#         3: 9,
#         4: 16,
#         5: 25
#     }
#
# ============================================================

squares = {
    number: number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# EQUIVALENT NORMAL FOR LOOP
# ============================================================
#
# The dictionary comprehension above is equivalent to:
#
#     squares = {}
#
#     for number in numbers:
#         squares[number] = number ** 2
#
# The comprehension simply expresses the same operation
# in a more compact form.
#
# ============================================================


# ============================================================
# DICTIONARY COMPREHENSION WITH A CONDITION
# ============================================================
#
# You can also filter items using an if condition.
#
# Example:
#
#     even_squares = {
#         number: number ** 2
#         for number in numbers
#         if number % 2 == 0
#     }
#
# Result:
#
#     {2: 4, 4: 16}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "For every item, create a KEY → VALUE pair."
#
# Think:
#
#     {key: value for item in iterable}
#
# The three main comprehension types are:
#
#     [expression for item in iterable]
#         → LIST
#
#     {expression for item in iterable}
#         → SET
#
#     {key: value for item in iterable}
#         → DICTIONARY
#
# ============================================================
```

```python
# ============================================================
# 67. SET COMPREHENSION
# ============================================================
#
# A set comprehension is a compact way to create a set by
# applying an expression to each item in an iterable.
#
# The basic pattern is:
#
#     {expression for item in iterable}
#
# It is similar to a list comprehension, but uses { } and
# produces a SET instead of a LIST.
#
# Important:
#
#     List → ordered collection, allows duplicates
#     Set  → unique elements, removes duplicates
#
# This makes set comprehensions useful when you want to
# transform data and automatically keep only unique results.
#
# ============================================================


# ============================================================
# CREATE A LIST
# ============================================================
#
# This list contains duplicate values:
#
#     2 appears twice
#     3 appears twice
#
# ============================================================

numbers = [1, 2, 2, 3, 3, 4, 5]


# ============================================================
# SET COMPREHENSION
# ============================================================
#
# For each number:
#
#     number ** 2
#
# calculates its square.
#
# The resulting values are stored in a SET.
#
# Because a set contains only unique elements, duplicate
# results are automatically removed.
#
# Step by step:
#
#     1 → 1² → 1
#     2 → 2² → 4
#     2 → 2² → 4  ← duplicate
#     3 → 3² → 9
#     3 → 3² → 9  ← duplicate
#     4 → 4² → 16
#     5 → 5² → 25
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================

squares = {
    number ** 2
    for number in numbers
}


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# SET COMPREHENSION VS LIST COMPREHENSION
# ============================================================
#
# List comprehension:
#
#     squares = [
#         number ** 2
#         for number in numbers
#     ]
#
# Result:
#
#     [1, 4, 4, 9, 9, 16, 25]
#
#
# Set comprehension:
#
#     squares = {
#         number ** 2
#         for number in numbers
#     }
#
# Result:
#
#     {1, 4, 9, 16, 25}
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you think:
#
#     "Transform every item and keep only unique results."
#
# Think:
#
#     {expression for item in iterable}
#
# The key difference is:
#
#     [ ] → list comprehension
#     { } → set comprehension
#
# ============================================================
```

```python
# ============================================================
# 68. NESTED LIST COMPREHENSION
# ============================================================
#
# A nested list comprehension is a list comprehension that
# contains more than one for loop.
#
# It is useful when working with nested structures such as:
#
#     list of lists
#     2D matrices
#     tables
#     nested data
#
# The general pattern is:
#
#     [value for outer_item in collection
#            for inner_item in outer_item]
#
# Think:
#
#     Outer loop → go through each row
#     Inner loop → go through each value in the row
#
# ============================================================


# ============================================================
# CREATE A 2D MATRIX
# ============================================================
#
# A matrix can be represented as a list containing multiple
# lists.
#
#     Row 1 → [1, 2]
#     Row 2 → [3, 4]
#     Row 3 → [5, 6]
#
# ============================================================

matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]


# ============================================================
# NESTED LIST COMPREHENSION
# ============================================================
#
# We want to flatten the matrix:
#
#     [[1, 2], [3, 4], [5, 6]]
#
# into:
#
#     [1, 2, 3, 4, 5, 6]
#
# The nested comprehension:
#
#     [
#         value
#         for row in matrix
#         for value in row
#     ]
#
# is equivalent to:
#
#     flattened = []
#
#     for row in matrix:
#
#         for value in row:
#
#             flattened.append(value)
#
# ============================================================

flattened = [
    value
    for row in matrix
    for value in row
]


# ============================================================
# PRINT RESULT
# ============================================================
#
# Output:
#
#     [1, 2, 3, 4, 5, 6]
#
# ============================================================

print(flattened)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see nested data:
#
#     [
#         [1, 2],
#         [3, 4],
#         [5, 6]
#     ]
#
# Think:
#
#     for each ROW
#         for each VALUE in that ROW
#
# In normal loops:
#
#     for row in matrix:
#         for value in row:
#             ...
#
# In a nested list comprehension:
#
#     [value for row in matrix for value in row]
#
# IMPORTANT:
#
# The order of the for clauses is the same as the order of
# the normal nested loops.
#
# ============================================================
```

```python
# ============================================================
# 69. NUMPY LOOP
# ============================================================
#
# A NumPy array can be iterated over using a normal Python
# for loop.
#
# The loop takes each element from the NumPy array one by one.
#
# Think:
#
#     NumPy array → one element → process it → next element
#
# Note:
#
# NumPy is designed for vectorized operations, so explicit
# Python loops are often not the best choice for numerical
# operations on large arrays.
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# Import NumPy so we can create and work with NumPy arrays.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE A NUMPY ARRAY
# ============================================================
#
# Create a NumPy array containing four numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40])


# ============================================================
# ITERATE THROUGH THE ARRAY
# ============================================================
#
# The for loop visits each element one at a time:
#
#     10 → number → print
#     20 → number → print
#     30 → number → print
#     40 → number → print
#
# ============================================================

for number in numbers:
    print(number)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every element in this array, do something."
#
# Think:
#
#     for number in numbers:
#
# This works because a NumPy array is iterable.
#
# However, if the operation can be expressed as a NumPy
# vectorized operation, prefer vectorization for numerical
# computations, especially with large arrays.
#
# Example:
#
#     # Loop
#     for number in numbers:
#         print(number ** 2)
#
#     # Vectorized
#     print(numbers ** 2)
#
# ============================================================
```

```python
# ============================================================
# 70. NUMPY ARRAY CONDITION
# ============================================================

# Import numpy library (assuming it's already imported as np)
# Create a numpy array named 'numbers' with values [10, 20, 30, 40, 50]
numbers = np.array([10, 20, 30, 40, 50])

# Iterate through each element in the 'numbers' array
for number in numbers:

    # Check if the current number is greater than 25
    if number > 25:
        # Print the number if it meets the condition
        print(number)
```

```python
# ============================================================
# 71. NUMPY BOOLEAN MASKING
# ============================================================
#
# Boolean masking is a powerful NumPy technique for filtering
# elements of an array based on a condition.
#
# The basic idea is:
#
#     array[condition]
#
# NumPy first evaluates the condition for every element and
# creates a Boolean mask containing True or False values.
#
#     True  → keep the element
#     False → exclude the element
#
# Example:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# This mask is then used to select the corresponding elements.
#
# Think:
#
#     Condition → Boolean mask → Filtered array
#
# ============================================================


# ============================================================
# IMPORT NUMPY
# ============================================================
#
# NumPy provides arrays and efficient operations for numerical
# and data-science computations.
#
# ============================================================

import numpy as np


# ============================================================
# CREATE NUMPY ARRAY
# ============================================================
#
# Create an array containing five numbers.
#
# ============================================================

numbers = np.array([10, 20, 30, 40, 50])


# ============================================================
# CREATE A BOOLEAN MASK
# ============================================================
#
# The condition:
#
#     numbers > 25
#
# is evaluated element by element:
#
#     10 > 25 → False
#     20 > 25 → False
#     30 > 25 → True
#     40 > 25 → True
#     50 > 25 → True
#
# Therefore:
#
#     numbers > 25
#
# produces:
#
#     [False, False, True, True, True]
#
# ============================================================

mask = numbers > 25

print(mask)


# ============================================================
# APPLY THE BOOLEAN MASK
# ============================================================
#
# Use the Boolean mask to select only the elements where
# the mask is True.
#
#     numbers[mask]
#
# returns:
#
#     [30, 40, 50]
#
# ============================================================

print(numbers[mask])


# ============================================================
# SHORT FORM
# ============================================================
#
# The mask does not need to be stored in a separate variable.
#
# We can directly write:
#
#     numbers[numbers > 25]
#
# This means:
#
#     "Select elements from numbers where numbers > 25."
#
# ============================================================

print(numbers[numbers > 25])


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you see:
#
#     "Keep only the values that satisfy a condition."
#
# Think:
#
#     Boolean masking
#
# The general pattern is:
#
#     array[array condition]
#
# Examples:
#
#     numbers[numbers > 25]
#     numbers[numbers < 25]
#     numbers[numbers == 30]
#     numbers[numbers != 30]
#
# Boolean masking is extremely important in Data Science
# because the same idea appears in NumPy, Pandas, and data
# filtering.
#
# ============================================================
```

```python
# ============================================================
# 72. LOOP VS NUMPY VECTORIZATION
# ============================================================
#
# There are two common ways to perform the same operation
# on multiple values:
#
# 1. Python loop
# 2. NumPy vectorization
#
# A loop processes the elements one by one.
# Vectorization applies the operation to the entire array
# at once.
#
# In Data Science, vectorization is usually preferred because
# NumPy operations are implemented in optimized low-level code
# and are generally faster for large numerical datasets.
#
# Think:
#
#     Loop          → one element at a time
#     Vectorization → whole array at once
#
# ============================================================

numbers = np.array([1, 2, 3, 4, 5])


# ============================================================
# LOOP
# ============================================================
#
# The loop takes each number individually:
#
#     1 → 1²
#     2 → 2²
#     3 → 3²
#     4 → 4²
#     5 → 5²
#
# The result is gradually built inside a new list.
#
# ============================================================

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)


# ============================================================
# NUMPY VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to apply the operation directly to the
# entire array.
#
# No explicit Python loop is needed.
#
#     [1, 2, 3, 4, 5] ** 2
#              ↓
#     [1, 4, 9, 16, 25]
#
# ============================================================

squares = numbers ** 2

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# If you think:
#
#     "For every number, do this..."
#
# A Python loop is the natural starting point.
#
# But when working with NumPy arrays, ask:
#
#     "Can NumPy do this operation on the whole array?"
#
# If yes, prefer the vectorized NumPy operation in most
# data-processing situations.
#
# ============================================================
```

```python
# ============================================================
# VECTORWISE / VECTORIZED OPERATION
# ============================================================
#
# NumPy allows us to perform an operation on an entire array
# without writing an explicit Python for loop.
#
# This is called VECTORIZATION.
#
# Think:
#
#     Python loop:
#         one element → operation
#         one element → operation
#         one element → operation
#
#     NumPy vectorization:
#         entire array → operation
#
# NumPy performs the operation element by element internally,
# but the iteration is handled by optimized NumPy code rather
# than an explicit Python loop.
#
# ============================================================


# ============================================================
# CALCULATE SQUARES
# ============================================================
#
# The ** operator is applied to every element of the NumPy
# array.
#
# If:
#
#     numbers = [1, 2, 3, 4, 5]
#
# Then:
#
#     numbers ** 2
#
# produces:
#
#     [1, 4, 9, 16, 25]
#
# No explicit for loop is needed.
#
# ============================================================

squares = numbers ** 2


# ============================================================
# PRINT RESULT
# ============================================================

print(squares)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with NumPy arrays, think:
#
#     "Can I apply this operation to the whole array?"
#
# If yes, NumPy often allows you to write:
#
#     array ** 2
#     array + 10
#     array * 5
#     array / 2
#
# instead of explicitly writing a Python loop.
#
# This usually makes numerical code:
#
#     ✓ shorter
#     ✓ clearer
#     ✓ easier to read
#     ✓ often faster for large numerical arrays
#
# ============================================================
```

```python
# ============================================================
# 73. LOOP VS PANDAS VECTORIZATION
# ============================================================
#
# Pandas provides vectorized operations that allow us to
# perform calculations on an entire column without explicitly
# writing a Python loop.
#
# There are two approaches:
#
#     1. Loop through rows
#     2. Vectorized column operation
#
# The loop processes data one row at a time.
# Vectorization applies the operation to the entire column.
#
# For typical numerical and column-based operations, the
# vectorized approach is usually preferred.
#
# ============================================================


# ============================================================
# CREATE DATAFRAME
# ============================================================
#
# Create a DataFrame containing employee salaries.
#
# ============================================================

df = pd.DataFrame({
    "Salary": [30000, 50000, 70000]
})


# ============================================================
# LOOP — PROCESS ROW BY ROW
# ============================================================
#
# iterrows() allows us to iterate through a DataFrame
# one row at a time.
#
#     index → current row index
#     row   → current row
#
# For each row, increase the salary by 10%.
#
#     Salary × 1.10
#
# ============================================================

for index, row in df.iterrows():

    print(row["Salary"] * 1.10)


# ============================================================
# VECTORIZED OPERATION
# ============================================================
#
# Pandas can perform the same calculation directly on the
# entire Salary column.
#
#     df["Salary"] * 1.10
#
# is applied element by element to the whole Series.
#
#     30000 × 1.10 → 33000
#     50000 × 1.10 → 55000
#     70000 × 1.10 → 77000
#
# The results are stored in a new column.
#
# ============================================================

df["New_Salary"] = df["Salary"] * 1.10


# ============================================================
# PRINT RESULT
# ============================================================

print(df)


# ============================================================
# LOOP VS VECTORIZATION
# ============================================================
#
# LOOP:
#
#     for index, row in df.iterrows():
#         ...
#
#     → processes one row at a time
#
#
# VECTORIZATION:
#
#     df["Salary"] * 1.10
#
#     → processes the entire column
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When working with Pandas and you think:
#
#     "For every row, calculate something."
#
# Don't immediately write a loop.
#
# First ask:
#
#     "Can Pandas do this to the entire column?"
#
# If yes, use the vectorized operation.
#
# Think:
#
#     Python loop → one row at a time
#     Pandas       → entire column
#
# This is an important shift from basic Python programming
# toward efficient Data Science programming.
#
# ============================================================
```

```python
# ============================================================
# 74. PRACTICAL ML EXAMPLE
# ============================================================
#
# A common Machine Learning task is converting predicted
# probabilities into class labels.
#
# For binary classification, a model may output a probability
# between 0 and 1 instead of directly returning 0 or 1.
#
# Example:
#
#     0.2 → low probability  → class 0
#     0.7 → high probability → class 1
#
# Here, we use 0.5 as the classification threshold:
#
#     probability >= 0.5 → prediction = 1
#     probability <  0.5 → prediction = 0
#
# Think:
#
#     Model probability
#            ↓
#       Apply threshold
#            ↓
#       Class prediction
#
# ============================================================


# ============================================================
# MODEL PREDICTION PROBABILITIES
# ============================================================
#
# These values represent the predicted probability of the
# positive class for five samples.
#
# In a real ML project, these probabilities might come from:
#
#     model.predict_proba(X)
#
# ============================================================

# List of predicted probabilities from a machine learning model
# Each value represents the model's confidence that a sample belongs to the positive class
probabilities = [0.2, 0.7, 0.9, 0.3, 0.8]


# ============================================================
# CREATE AN EMPTY LIST
# ============================================================
#
# We will store the final class predictions in this list.
#
# ============================================================

# Initialize an empty list to store the binary predictions
# This will hold the final class labels (0 or 1) after applying the threshold
predictions = []


# ============================================================
# LOOP THROUGH PROBABILITIES
# ============================================================
#
# Process each probability one at a time.
#
# The condition determines which class the probability
# belongs to.
#
#     probability >= 0.5 → 1
#     probability <  0.5 → 0
#
# Step by step:
#
#     0.2 → < 0.5 → 0
#     0.7 → >= 0.5 → 1
#     0.9 → >= 0.5 → 1
#     0.3 → < 0.5 → 0
#     0.8 → >= 0.5 → 1
#
# ============================================================

# Iterate through each probability in the list
for probability in probabilities:

    # Check if the probability is greater than or equal to the threshold (0.5)
    if probability >= 0.5:
        # If probability is high, assign class 1 (positive class)
        predictions.append(1)

    # Otherwise, the probability is below the threshold
    else:
        # If probability is low, assign class 0 (negative class)
        predictions.append(0)


# ============================================================
# PRINT PREDICTIONS
# ============================================================
#
# Output:
#
#     [0, 1, 1, 0, 1]
#
# ============================================================

# Print the final predictions
# This converts the probabilities into binary class labels based on the threshold
print(predictions)


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a classic:
#
#     LOOP + CONDITION + RESULT COLLECTION
#
# Pattern:
#
#     result = []
#
#     for item in data:
#
#         if condition:
#             result.append(value_1)
#
#         else:
#             result.append(value_2)
#
# In Machine Learning, this pattern appears in tasks such as:
#
#     probability → class label
#     score       → decision
#     prediction  → category
#
# ============================================================
```

```python
# ============================================================
# 75. PRACTICAL WATER PIPELINE EXAMPLE
# ============================================================
#
# This example shows how a Machine Learning prediction can
# support a real-world engineering decision.
#
# Suppose an ML model predicts the probability that each water
# pipeline will fail.
#
# Instead of using only two classes:
#
#     0 → No failure
#     1 → Failure
#
# we can use the predicted probability to define different
# maintenance actions.
#
# Higher probability → higher risk → more urgent action.
#
# This is an example of:
#
#     ML prediction → Risk level → Engineering action
#
# ============================================================


# ============================================================
# FAILURE PROBABILITIES
# ============================================================
#
# Each value represents the predicted probability of pipeline
# failure.
#
#     0.15 → 15% probability
#     0.42 → 42% probability
#     0.73 → 73% probability
#     0.91 → 91% probability
#     0.31 → 31% probability
#
# In a real ML project, these values could come from a model
# such as:
#
#     model.predict_proba(X)
#
# ============================================================

failure_probabilities = [
    0.15,
    0.42,
    0.73,
    0.91,
    0.31
]


# ============================================================
# RISK-BASED DECISION LOGIC
# ============================================================
#
# Each probability is checked against several thresholds.
#
#     >= 0.80 → Immediate inspection
#     >= 0.60 → Schedule inspection
#     >= 0.40 → Monitor
#     <  0.40 → Normal operation
#
# Notice that the conditions are checked from HIGH to LOW.
#
# Once a condition is True, Python executes that block and
# skips the remaining elif / else conditions.
#
# ============================================================

for probability in failure_probabilities:

    if probability >= 0.80:
        action = "Immediate inspection"

    elif probability >= 0.60:
        action = "Schedule inspection"

    elif probability >= 0.40:
        action = "Monitor"

    else:
        action = "Normal operation"


    # ========================================================
    # DISPLAY RESULT
    # ========================================================
    #
    # Show the predicted probability together with the
    # corresponding engineering action.
    #
    # ========================================================

    print(
        probability,
        "->",
        action
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     0.15 -> Normal operation
#     0.42 -> Monitor
#     0.73 -> Schedule inspection
#     0.91 -> Immediate inspection
#     0.31 -> Normal operation
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# This is a very useful pattern:
#
#     FOR each prediction:
#         IF high risk:
#             urgent action
#         ELIF medium-high risk:
#             planned action
#         ELIF moderate risk:
#             monitoring
#         ELSE:
#             normal operation
#
# In Data Science and ML, this connects:
#
#     DATA
#       ↓
#     ML MODEL
#       ↓
#     PROBABILITY
#       ↓
#     THRESHOLD / DECISION RULE
#       ↓
#     REAL-WORLD ACTION
#
# ============================================================


# ============================================================
# IMPORTANT ML INSIGHT
# ============================================================
#
# The thresholds (0.80, 0.60, 0.40) are only example values.
#
# In a real pipeline-failure prediction system, thresholds
# should ideally be determined using:
#
#     • historical failure data
#     • cost of missed failures
#     • cost of unnecessary inspections
#     • precision / recall
#     • ROC / PR analysis
#     • engineering constraints
#     • domain expertise
#
# Therefore:
#
#     Probability ≠ automatic decision
#
# A probability becomes an operational decision only after
# an appropriate decision threshold or risk policy is defined.
#
# ============================================================
```

```python
# ============================================================
# 76. PRACTICAL ML METRICS
# ============================================================
#
# Machine Learning models are often evaluated using multiple
# metrics rather than relying on a single score.
#
# Common classification metrics include:
#
#     Accuracy  → overall proportion of correct predictions
#     Precision → proportion of predicted positives that are
#                 actually positive
#     Recall    → proportion of actual positives that are found
#     F1        → harmonic mean of Precision and Recall
#
# Storing metrics in a dictionary makes it easy to organize
# and iterate through them.
#
# Think:
#
#     Metric name → Metric value
#
# ============================================================


# ============================================================
# CREATE A DICTIONARY OF METRICS
# ============================================================
#
# The dictionary stores each metric as a KEY-VALUE pair:
#
#     KEY   → metric name
#     VALUE → metric score
#
# The scores are represented as decimal values:
#
#     0.91 → 91%
#     0.84 → 84%
#     0.76 → 76%
#     0.80 → 80%
#
# ============================================================

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}


# ============================================================
# LOOP THROUGH THE DICTIONARY
# ============================================================
#
# .items() gives us both the key and value:
#
#     metric → "Accuracy"
#     value  → 0.91
#
#     metric → "Precision"
#     value  → 0.84
#
#     ...
#
# ============================================================

for metric, value in metrics.items():

    print(
        metric,
        ":",
        value
    )


# ============================================================
# EXPECTED OUTPUT
# ============================================================
#
#     Accuracy : 0.91
#     Precision : 0.84
#     Recall : 0.76
#     F1 : 0.8
#
# ============================================================


# ============================================================
# PYTHON FEELING
# ============================================================
#
# When you have a dictionary and need both:
#
#     KEY + VALUE
#
# Think:
#
#     for key, value in dictionary.items():
#
# This pattern is extremely common in Data Science and
# Machine Learning.
#
# For example:
#
#     for metric, value in metrics.items():
#         print(metric, value)
#
#     for parameter, value in model_params.items():
#         print(parameter, value)
#
#     for feature, importance in feature_importance.items():
#         print(feature, importance)
#
# ============================================================
```

```python
# ============================================================
# 77. CHECKING ML METRICS
# ============================================================

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}

for metric, value in metrics.items():

    if value >= 0.90:
        performance = "Excellent"

    elif value >= 0.80:
        performance = "Good"

    elif value >= 0.70:
        performance = "Acceptable"

    else:
        performance = "Poor"

    print(
        metric,
        "->",
        performance
    )
```
