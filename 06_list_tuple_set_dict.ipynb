```python
# ============================================================
#       PYTHON COLLECTIONS
# ============================================================
#
# Python has four important built-in collection types:
#
# 1. LIST        -> ordered, mutable
# 2. TUPLE       -> ordered, immutable
# 3. DICTIONARY  -> key-value pairs, mutable
# 4. SET         -> unordered collection of unique values
#
# ============================================================


# ============================================================
# 1. LISTS
# ============================================================

# A list stores multiple values in one variable.

numbers = [10, 20, 30, 40, 50]

print(numbers)


# ============================================================
# 2. LIST CAN CONTAIN DIFFERENT DATA TYPES
# ============================================================

data = [
    "Yasin",
    30,
    3.14,
    True
]

print(data)


# ============================================================
# 3. LIST INDEXING
# ============================================================

names = ["Ali", "Sara", "Yasin", "John"]

print(names[0])       # first item
print(names[1])       # second item
print(names[2])       # third item
print(names[-1])      # last item
print(names[-2])      # second-last item


# ============================================================
# 4. LIST SLICING
# ============================================================

numbers = [10, 20, 30, 40, 50]

print(numbers[0:3])
print(numbers[:3])
print(numbers[2:])
print(numbers[:])
print(numbers[-3:])
print(numbers[::2])
print(numbers[::-1])


# ============================================================
# 5. LENGTH OF A LIST
# ============================================================

numbers = [10, 20, 30, 40, 50]

print(len(numbers))


# ============================================================
# 6. CHANGE A LIST ITEM
# ============================================================

numbers = [10, 20, 30, 40]

numbers[0] = 100

print(numbers)


# ============================================================
# 7. CHANGE MULTIPLE LIST ITEMS
# ============================================================

numbers = [10, 20, 30, 40, 50]

numbers[1:3] = [200, 300]

print(numbers)


# ============================================================
# 8. APPEND()
# ============================================================

numbers = [10, 20, 30]

numbers.append(40)

print(numbers)


# ============================================================
# 9. INSERT()
# ============================================================

numbers = [10, 20, 40]

numbers.insert(2, 30)

print(numbers)


# ============================================================
# 10. EXTEND()
# ============================================================

numbers = [10, 20, 30]

numbers.extend([40, 50, 60])

print(numbers)


# ============================================================
# 11. APPEND VS EXTEND
# ============================================================

numbers = [10, 20, 30]

numbers.append([40, 50])

print(numbers)

# Result:
# [10, 20, 30, [40, 50]]


numbers = [10, 20, 30]

numbers.extend([40, 50])

print(numbers)

# Result:
# [10, 20, 30, 40, 50]


# ============================================================
# 12. REMOVE()
# ============================================================

numbers = [10, 20, 30, 40]

numbers.remove(30)

print(numbers)


# ============================================================
# 13. POP()
# ============================================================

numbers = [10, 20, 30, 40]

removed = numbers.pop()

print(numbers)
print(removed)


# Remove by index
numbers = [10, 20, 30, 40]

removed = numbers.pop(1)

print(numbers)
print(removed)


# ============================================================
# 14. DEL
# ============================================================

numbers = [10, 20, 30, 40]

del numbers[1]

print(numbers)


# ============================================================
# 15. CLEAR()
# ============================================================

numbers = [10, 20, 30]

numbers.clear()

print(numbers)


# ============================================================
# 16. INDEX()
# ============================================================

names = ["Ali", "Sara", "Yasin", "John"]

print(names.index("Yasin"))


# ============================================================
# 17. COUNT()
# ============================================================

numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))


# ============================================================
# 18. IN / NOT IN
# ============================================================

names = ["Ali", "Sara", "Yasin"]

print("Yasin" in names)
print("John" in names)

print("John" not in names)


# ============================================================
# 19. SORT()
# ============================================================

numbers = [50, 10, 40, 20, 30]

numbers.sort()

print(numbers)


# Descending
numbers.sort(reverse=True)

print(numbers)


# ============================================================
# 20. SORTED()
# ============================================================

numbers = [50, 10, 40, 20, 30]

sorted_numbers = sorted(numbers)

print(sorted_numbers)

# Original remains unchanged
print(numbers)


# ============================================================
# 21. REVERSE()
# ============================================================

numbers = [10, 20, 30, 40]

numbers.reverse()

print(numbers)


# ============================================================
# 22. COPY()
# ============================================================

numbers = [10, 20, 30]

copy_numbers = numbers.copy()

print(copy_numbers)


# ============================================================
# 23. LIST CONCATENATION
# ============================================================

list1 = [1, 2, 3]
list2 = [4, 5, 6]

result = list1 + list2

print(result)


# ============================================================
# 24. LIST REPETITION
# ============================================================

numbers = [1, 2, 3]

print(numbers * 3)


# ============================================================
# 25. LOOPING THROUGH A LIST
# ============================================================

names = ["Ali", "Sara", "Yasin"]

for name in names:
    print(name)


# ============================================================
# 26. LIST + CONDITION
# ============================================================

numbers = [10, 15, 20, 25, 30]

for number in numbers:

    if number > 20:
        print(number)


# ============================================================
# 27. LIST COMPREHENSION
# ============================================================

numbers = [1, 2, 3, 4, 5]

squares = [
    number ** 2
    for number in numbers
]

print(squares)


# ============================================================
# 28. LIST COMPREHENSION WITH CONDITION
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [
    number
    for number in numbers
    if number % 2 == 0
]

print(even_numbers)


# ============================================================
# 29. LIST COMPREHENSION WITH IF / ELSE
# ============================================================

scores = [45, 67, 82, 91, 38]

results = [
    "Pass" if score >= 50 else "Fail"
    for score in scores
]

print(results)


# ============================================================
# 30. NESTED LIST
# ============================================================

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix)

print(matrix[0])
print(matrix[0][1])


# ============================================================
# 31. LOOP THROUGH NESTED LIST
# ============================================================

for row in matrix:

    for value in row:

        print(value)


# ============================================================
# 32. LIST OF DICTIONARIES
# ============================================================

employees = [
    {
        "name": "Ali",
        "age": 25,
        "salary": 35000
    },

    {
        "name": "Sara",
        "age": 30,
        "salary": 50000
    },

    {
        "name": "Yasin",
        "age": 35,
        "salary": 70000
    }
]

print(employees)


# ============================================================
# 33. DATA ANALYTICS — FILTERING A LIST
# ============================================================

sales = [100, 250, 500, 750, 1200]

high_sales = [
    sale
    for sale in sales
    if sale >= 500
]

print(high_sales)


# ============================================================
# 34. DATA ANALYTICS — TRANSFORMING A LIST
# ============================================================

prices = [100, 200, 300, 400]

prices_with_tax = [
    price * 1.20
    for price in prices
]

print(prices_with_tax)


# ============================================================
# 35. TUPLES
# ============================================================
#
# Tuple:
# - ordered
# - indexed
# - immutable
#
# ============================================================

coordinates = (10, 20)

print(coordinates)


# ============================================================
# 36. TUPLE INDEXING
# ============================================================

coordinates = (10, 20, 30)

print(coordinates[0])
print(coordinates[1])
print(coordinates[-1])


# ============================================================
# 37. TUPLE SLICING
# ============================================================

numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[:3])
print(numbers[-2:])


# ============================================================
# 38. TUPLE IS IMMUTABLE
# ============================================================

numbers = (10, 20, 30)

# This causes an error:
#
# numbers[0] = 100


# ============================================================
# 39. TUPLE COUNT()
# ============================================================

numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))


# ============================================================
# 40. TUPLE INDEX()
# ============================================================

names = ("Ali", "Sara", "Yasin")

print(names.index("Yasin"))


# ============================================================
# 41. TUPLE MEMBERSHIP
# ============================================================

names = ("Ali", "Sara", "Yasin")

print("Yasin" in names)
print("John" not in names)


# ============================================================
# 42. LOOPING THROUGH TUPLE
# ============================================================

numbers = (10, 20, 30, 40)

for number in numbers:
    print(number)


# ============================================================
# 43. TUPLE UNPACKING
# ============================================================

person = ("Yasin", 30, "Data Science")

name, age, field = person

print(name)
print(age)
print(field)


# ============================================================
# 44. SWAPPING VARIABLES
# ============================================================

x = 10
y = 20

x, y = y, x

print(x)
print(y)


# ============================================================
# 45. ZIP RETURNS TUPLES
# ============================================================

names = ["Ali", "Sara", "Yasin"]
ages = [25, 30, 35]

for item in zip(names, ages):

    print(item)


# ============================================================
# 46. DICTIONARIES
# ============================================================
#
# Dictionary stores:
#
# key -> value
#
# ============================================================

student = {
    "name": "Yasin",
    "age": 30,
    "field": "Data Science"
}

print(student)


# ============================================================
# 47. ACCESS DICTIONARY VALUES
# ============================================================

print(student["name"])
print(student["age"])
print(student["field"])


# ============================================================
# 48. GET()
# ============================================================

print(student.get("name"))

print(student.get("salary"))

print(student.get("salary", 0))


# ============================================================
# 49. ADD A NEW KEY
# ============================================================

student["salary"] = 70000

print(student)


# ============================================================
# 50. MODIFY A VALUE
# ============================================================

student["age"] = 31

print(student)


# ============================================================
# 51. UPDATE()
# ============================================================

student.update({
    "age": 32,
    "salary": 80000
})

print(student)


# ============================================================
# 52. REMOVE WITH POP()
# ============================================================

student = {
    "name": "Yasin",
    "age": 30,
    "field": "Data Science"
}

field = student.pop("field")

print(student)
print(field)


# ============================================================
# 53. POPITEM()
# ============================================================

student = {
    "name": "Yasin",
    "age": 30,
    "field": "Data Science"
}

item = student.popitem()

print(student)
print(item)


# ============================================================
# 54. DEL DICTIONARY ITEM
# ============================================================

student = {
    "name": "Yasin",
    "age": 30
}

del student["age"]

print(student)


# ============================================================
# 55. CLEAR DICTIONARY
# ============================================================

student = {
    "name": "Yasin",
    "age": 30
}

student.clear()

print(student)


# ============================================================
# 56. DICTIONARY KEYS
# ============================================================

student = {
    "name": "Yasin",
    "age": 30,
    "field": "Data Science"
}

print(student.keys())


# ============================================================
# 57. DICTIONARY VALUES
# ============================================================

print(student.values())


# ============================================================
# 58. DICTIONARY ITEMS
# ============================================================

print(student.items())


# ============================================================
# 59. LOOP THROUGH DICTIONARY KEYS
# ============================================================

for key in student:

    print(key)


# ============================================================
# 60. LOOP THROUGH DICTIONARY VALUES
# ============================================================

for value in student.values():

    print(value)


# ============================================================
# 61. LOOP THROUGH KEYS AND VALUES
# ============================================================

for key, value in student.items():

    print(key, ":", value)


# ============================================================
# 62. CHECKING DICTIONARY KEYS
# ============================================================

if "name" in student:
    print("Name exists.")


if "salary" not in student:
    print("Salary does not exist.")


# ============================================================
# 63. DICTIONARY LENGTH
# ============================================================

print(len(student))


# ============================================================
# 64. DICTIONARY COMPREHENSION
# ============================================================

numbers = [1, 2, 3, 4, 5]

squares = {
    number: number ** 2
    for number in numbers
}

print(squares)


# ============================================================
# 65. DICTIONARY COMPREHENSION WITH CONDITION
# ============================================================

numbers = [1, 2, 3, 4, 5, 6]

even_squares = {
    number: number ** 2
    for number in numbers
    if number % 2 == 0
}

print(even_squares)


# ============================================================
# 66. NESTED DICTIONARY
# ============================================================

students = {

    "student1": {
        "name": "Ali",
        "age": 25
    },

    "student2": {
        "name": "Sara",
        "age": 30
    }
}

print(students)


# Access nested values
print(students["student1"]["name"])
print(students["student2"]["age"])


# ============================================================
# 67. LIST OF DICTIONARIES
# ============================================================

employees = [

    {
        "name": "Ali",
        "salary": 35000
    },

    {
        "name": "Sara",
        "salary": 50000
    },

    {
        "name": "Yasin",
        "salary": 70000
    }
]

for employee in employees:

    print(
        employee["name"],
        employee["salary"]
    )


# ============================================================
# 68. DATA ANALYTICS — DICTIONARY OF METRICS
# ============================================================

metrics = {

    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80
}

for metric, value in metrics.items():

    print(metric, ":", value)


# ============================================================
# 69. DATA ANALYTICS — CUSTOMER DATA
# ============================================================

customer = {

    "customer_id": 101,
    "age": 35,
    "income": 65000,
    "purchase_count": 12,
    "total_spent": 2500
}

print(customer["income"])
print(customer["total_spent"])


# ============================================================
# 70. DATA ANALYTICS — CALCULATING A METRIC
# ============================================================

sales = {
    "January": 12000,
    "February": 15000,
    "March": 18000,
    "April": 14000
}

total_sales = sum(sales.values())

print(total_sales)


# ============================================================
# 71. DATA ANALYTICS — FINDING BEST MONTH
# ============================================================

best_month = max(
    sales,
    key=sales.get
)

print(best_month)
print(sales[best_month])


# ============================================================
# 72. DATA ANALYTICS — FILTER DICTIONARY
# ============================================================

sales = {
    "January": 12000,
    "February": 15000,
    "March": 18000,
    "April": 14000
}

high_sales = {
    month: value
    for month, value in sales.items()
    if value > 14000
}

print(high_sales)


# ============================================================
# 73. SETS
# ============================================================
#
# A set:
# - stores unique values
# - does not support normal indexing
# - is mutable
#
# ============================================================

numbers = {10, 20, 30, 40}

print(numbers)


# ============================================================
# 74. DUPLICATES ARE REMOVED
# ============================================================

numbers = {10, 20, 20, 30, 30, 30}

print(numbers)


# ============================================================
# 75. CREATE EMPTY SET
# ============================================================

empty_set = set()

print(empty_set)

# IMPORTANT:
# {} creates an empty dictionary, NOT an empty set.


# ============================================================
# 76. ADD()
# ============================================================

numbers = {10, 20, 30}

numbers.add(40)

print(numbers)


# ============================================================
# 77. ADDING AN EXISTING VALUE
# ============================================================

numbers = {10, 20, 30}

numbers.add(20)

print(numbers)

# Nothing changes because 20 already exists.


# ============================================================
# 78. UPDATE()
# ============================================================

numbers = {10, 20, 30}

numbers.update([40, 50, 60])

print(numbers)


# ============================================================
# 79. REMOVE()
# ============================================================

numbers = {10, 20, 30}

numbers.remove(20)

print(numbers)


# ============================================================
# 80. DISCARD()
# ============================================================

numbers = {10, 20, 30}

numbers.discard(50)

print(numbers)

# discard() does not raise an error
# if the value doesn't exist.


# ============================================================
# 81. POP()
# ============================================================

numbers = {10, 20, 30}

item = numbers.pop()

print(item)
print(numbers)


# ============================================================
# 82. CLEAR()
# ============================================================

numbers = {10, 20, 30}

numbers.clear()

print(numbers)


# ============================================================
# 83. SET MEMBERSHIP
# ============================================================

numbers = {10, 20, 30}

print(20 in numbers)
print(50 in numbers)


# ============================================================
# 84. SET UNION
# ============================================================

set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1 | set2)

print(set1.union(set2))


# ============================================================
# 85. SET INTERSECTION
# ============================================================

set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1 & set2)

print(set1.intersection(set2))


# ============================================================
# 86. SET DIFFERENCE
# ============================================================

set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1 - set2)

print(set1.difference(set2))


# ============================================================
# 87. SYMMETRIC DIFFERENCE
# ============================================================

set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1 ^ set2)

print(set1.symmetric_difference(set2))


# ============================================================
# 88. SUBSET
# ============================================================

set1 = {1, 2}
set2 = {1, 2, 3, 4}

print(set1.issubset(set2))


# ============================================================
# 89. SUPERSET
# ============================================================

set1 = {1, 2, 3, 4}
set2 = {1, 2}

print(set1.issuperset(set2))


# ============================================================
# 90. DISJOINT SETS
# ============================================================

set1 = {1, 2, 3}
set2 = {4, 5, 6}

print(set1.isdisjoint(set2))


# ============================================================
# 91. SET COMPREHENSION
# ============================================================

numbers = [1, 2, 2, 3, 3, 4, 5]

squares = {
    number ** 2
    for number in numbers
}

print(squares)


# ============================================================
# 92. DATA ANALYTICS — REMOVE DUPLICATES
# ============================================================

values = [
    "Python",
    "SQL",
    "Python",
    "Power BI",
    "SQL"
]

unique_values = set(values)

print(unique_values)


# ============================================================
# 93. DATA ANALYTICS — UNIQUE CATEGORIES
# ============================================================

departments = [
    "IT",
    "HR",
    "IT",
    "Finance",
    "HR",
    "IT"
]

unique_departments = set(departments)

print(unique_departments)


# ============================================================
# 94. DATA ANALYTICS — FIND COMMON SKILLS
# ============================================================

skills_yasin = {
    "Python",
    "SQL",
    "Machine Learning",
    "Power BI"
}

skills_job = {
    "Python",
    "SQL",
    "Power BI",
    "Excel"
}

common_skills = skills_yasin & skills_job

print(common_skills)


# ============================================================
# 95. DATA ANALYTICS — MISSING SKILLS
# ============================================================

missing_skills = skills_job - skills_yasin

print(missing_skills)


# ============================================================
# 96. LIST VS TUPLE VS SET
# ============================================================

my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_set = {1, 2, 3}

print(type(my_list))
print(type(my_tuple))
print(type(my_set))


# ============================================================
# 97. CONVERT LIST TO SET
# ============================================================

numbers = [1, 2, 2, 3, 3, 4]

unique_numbers = set(numbers)

print(unique_numbers)


# ============================================================
# 98. CONVERT SET TO LIST
# ============================================================

numbers = {1, 2, 3, 4}

numbers_list = list(numbers)

print(numbers_list)


# ============================================================
# 99. CONVERT LIST TO TUPLE
# ============================================================

numbers = [1, 2, 3, 4]

numbers_tuple = tuple(numbers)

print(numbers_tuple)


# ============================================================
# 100. CONVERT TUPLE TO LIST
# ============================================================

numbers = (1, 2, 3, 4)

numbers_list = list(numbers)

print(numbers_list)


# ============================================================
# 101. PRACTICAL DATA ANALYTICS EXAMPLE
# ============================================================

employees = [
    {
        "name": "Ali",
        "department": "IT",
        "salary": 45000
    },

    {
        "name": "Sara",
        "department": "HR",
        "salary": 50000
    },

    {
        "name": "Yasin",
        "department": "IT",
        "salary": 70000
    }
]


# Get all salaries
salaries = [
    employee["salary"]
    for employee in employees
]

print(salaries)


# Calculate average salary
average_salary = sum(salaries) / len(salaries)

print(average_salary)


# Get unique departments
departments = {
    employee["department"]
    for employee in employees
}

print(departments)


# ============================================================
# 102. PRACTICAL DATA ANALYTICS — GROUPING
# ============================================================

employees = [
    {"name": "Ali", "department": "IT", "salary": 45000},
    {"name": "Sara", "department": "HR", "salary": 50000},
    {"name": "Yasin", "department": "IT", "salary": 70000},
    {"name": "John", "department": "HR", "salary": 55000}
]

department_salaries = {}

for employee in employees:

    department = employee["department"]
    salary = employee["salary"]

    if department not in department_salaries:
        department_salaries[department] = []

    department_salaries[department].append(salary)


print(department_salaries)


# ============================================================
# 103. CALCULATE AVERAGE BY GROUP
# ============================================================

for department, salaries in department_salaries.items():

    average = sum(salaries) / len(salaries)

    print(
        department,
        "Average Salary:",
        average
    )


# ============================================================
# 104. PRACTICAL ML DATA
# ============================================================

features = [
    "Age",
    "Income",
    "Experience",
    "Education"
]

target = "Churn"

print(features)
print(target)


# ============================================================
# 105. LIST OF FEATURES
# ============================================================

feature_values = [
    [25, 40000, 2, 1],
    [35, 70000, 8, 2],
    [45, 90000, 15, 3]
]

for row in feature_values:

    print(row)


# ============================================================
# 106. DICTIONARY OF ML METRICS
# ============================================================

metrics = {
    "Accuracy": 0.91,
    "Precision": 0.84,
    "Recall": 0.76,
    "F1": 0.80,
    "AUC": 0.87
}

for metric, value in metrics.items():

    print(
        f"{metric}: {value:.2f}"
    )
```
