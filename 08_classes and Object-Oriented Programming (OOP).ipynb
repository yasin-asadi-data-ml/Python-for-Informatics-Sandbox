```python
# ============================================================
#       PYTHON CLASSES & OBJECT-ORIENTED PROGRAMMING
# ============================================================
#
# OOP = Object-Oriented Programming
#
# The four major ideas commonly associated with OOP are:
#
# 1. Encapsulation
# 2. Inheritance
# 3. Polymorphism
# 4. Abstraction
#
# A CLASS is a blueprint.
# An OBJECT is an instance created from a class.
#
# ============================================================


# ============================================================
# 1. SIMPLE CLASS
# ============================================================

class Student:
    pass


# Create an object
student1 = Student()

print(student1)


# ============================================================
# 2. CLASS WITH ATTRIBUTES
# ============================================================

class Student:

    name = "Yasin"
    age = 35
    field = "Data Science"


student1 = Student()

print(student1.name)
print(student1.age)
print(student1.field)


# ============================================================
# 3. CREATING MULTIPLE OBJECTS
# ============================================================

class Student:

    name = "Unknown"
    age = 0


student1 = Student()
student2 = Student()

student1.name = "Ali"
student1.age = 25

student2.name = "Sara"
student2.age = 30

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)


# ============================================================
# 4. __init__()
# ============================================================
#
# __init__() is called automatically when an object is created.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age


student1 = Student("Yasin", 35)

print(student1.name)
print(student1.age)


# ============================================================
# 5. UNDERSTANDING self
# ============================================================
#
# self refers to the current object.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age


student1 = Student("Ali", 25)
student2 = Student("Sara", 30)

print(student1.name)
print(student2.name)


# ============================================================
# 6. MULTIPLE ATTRIBUTES
# ============================================================

class Researcher:

    def __init__(
        self,
        name,
        field,
        experience
    ):

        self.name = name
        self.field = field
        self.experience = experience


researcher1 = Researcher(
    "Yasin",
    "Machine Learning",
    8
)

print(researcher1.name)
print(researcher1.field)
print(researcher1.experience)


# ============================================================
# 7. METHODS
# ============================================================
#
# A method is a function defined inside a class.
#
# ============================================================

class Student:

    def __init__(self, name):

        self.name = name

    def introduce(self):

        print(
            "My name is",
            self.name
        )


student1 = Student("Yasin")

student1.introduce()


# ============================================================
# 8. METHOD WITH PARAMETERS
# ============================================================

class Calculator:

    def add(self, a, b):

        return a + b

    def subtract(self, a, b):

        return a - b


calculator = Calculator()

print(
    calculator.add(10, 20)
)

print(
    calculator.subtract(20, 10)
)


# ============================================================
# 9. OBJECT ATTRIBUTES
# ============================================================

class Person:

    def __init__(self, name, age):

        self.name = name
        self.age = age


person = Person(
    "Yasin",
    35
)

print(person.name)
print(person.age)


# ============================================================
# 10. MODIFY OBJECT ATTRIBUTES
# ============================================================

person.age = 36

print(person.age)


# ============================================================
# 11. ADD NEW ATTRIBUTE
# ============================================================

person.city = "Karaj"

print(person.city)


# ============================================================
# 12. DELETE ATTRIBUTE
# ============================================================

del person.city

# print(person.city)
# This would produce an AttributeError.


# ============================================================
# 13. DELETE OBJECT
# ============================================================

student = Student("Ali")

del student

# The object no longer exists.


# ============================================================
# 14. CHECK OBJECT TYPE
# ============================================================

class Student:
    pass


student = Student()

print(type(student))


# ============================================================
# 15. isinstance()
# ============================================================

class Student:
    pass


student = Student()

print(
    isinstance(
        student,
        Student
    )
)


# ============================================================
# 16. CLASS ATTRIBUTES
# ============================================================
#
# A class attribute belongs to the class.
#
# ============================================================

class Employee:

    company = "Data Analytics Lab"

    def __init__(self, name):

        self.name = name


employee1 = Employee("Ali")
employee2 = Employee("Sara")

print(employee1.company)
print(employee2.company)


# ============================================================
# 17. INSTANCE ATTRIBUTES
# ============================================================

class Employee:

    company = "Data Analytics Lab"

    def __init__(self, name, salary):

        self.name = name
        self.salary = salary


employee1 = Employee(
    "Ali",
    50000
)

employee2 = Employee(
    "Sara",
    60000
)

print(employee1.name)
print(employee1.salary)

print(employee2.name)
print(employee2.salary)


# ============================================================
# 18. CLASS ATTRIBUTE VS INSTANCE ATTRIBUTE
# ============================================================

class Employee:

    company = "AI Lab"

    def __init__(self, name):

        self.name = name


employee1 = Employee("Ali")

print(employee1.company)

Employee.company = "Data Science Lab"

print(employee1.company)


# ============================================================
# 19. METHOD USING OBJECT ATTRIBUTES
# ============================================================

class Rectangle:

    def __init__(self, width, height):

        self.width = width
        self.height = height

    def area(self):

        return self.width * self.height


rectangle = Rectangle(
    10,
    5
)

print(rectangle.area())


# ============================================================
# 20. MULTIPLE METHODS
# ============================================================

class Rectangle:

    def __init__(self, width, height):

        self.width = width
        self.height = height

    def area(self):

        return self.width * self.height

    def perimeter(self):

        return (
            2 * self.width
            + 2 * self.height
        )


rectangle = Rectangle(
    10,
    5
)

print(rectangle.area())
print(rectangle.perimeter())


# ============================================================
# 21. METHOD MODIFYING AN ATTRIBUTE
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.balance = balance

    def deposit(self, amount):

        self.balance += amount


account = BankAccount(1000)

account.deposit(500)

print(account.balance)


# ============================================================
# 22. METHOD WITH CONDITION
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.balance = balance

    def withdraw(self, amount):

        if amount <= self.balance:

            self.balance -= amount

            return "Withdrawal successful."

        else:

            return "Insufficient balance."


account = BankAccount(1000)

print(
    account.withdraw(300)
)

print(account.balance)


# ============================================================
# 23. METHOD WITH LOOP
# ============================================================

class DataAnalyzer:

    def __init__(self, values):

        self.values = values

    def calculate_total(self):

        total = 0

        for value in self.values:

            total += value

        return total


analyzer = DataAnalyzer(
    [10, 20, 30, 40]
)

print(
    analyzer.calculate_total()
)


# ============================================================
# 24. DATA ANALYTICS CLASS
# ============================================================

class DataAnalyzer:

    def __init__(self, values):

        self.values = values

    def mean(self):

        return (
            sum(self.values)
            / len(self.values)
        )

    def minimum(self):

        return min(self.values)

    def maximum(self):

        return max(self.values)

    def count(self):

        return len(self.values)


data = DataAnalyzer(
    [10, 20, 30, 40, 50]
)

print("Mean:", data.mean())
print("Minimum:", data.minimum())
print("Maximum:", data.maximum())
print("Count:", data.count())


# ============================================================
# 25. DATA ANALYTICS CLASS — DESCRIPTIVE STATISTICS
# ============================================================

class Statistics:

    def __init__(self, values):

        self.values = values

    def mean(self):

        return (
            sum(self.values)
            / len(self.values)
        )

    def range(self):

        return (
            max(self.values)
            - min(self.values)
        )

    def summary(self):

        return {
            "count": len(self.values),
            "mean": self.mean(),
            "minimum": min(self.values),
            "maximum": max(self.values),
            "range": self.range()
        }


data = Statistics(
    [10, 20, 30, 40, 50]
)

print(data.summary())


# ============================================================
# 26. __str__()
# ============================================================
#
# __str__() controls how an object is represented
# when print() is used.
#
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

    def __str__(self):

        return (
            f"Student: {self.name}, "
            f"Age: {self.age}"
        )


student = Student(
    "Yasin",
    35
)

print(student)


# ============================================================
# 27. __repr__()
# ============================================================

class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

    def __repr__(self):

        return (
            f"Student("
            f"name='{self.name}', "
            f"age={self.age})"
        )


student = Student(
    "Yasin",
    35
)

print(student)


# ============================================================
# 28. __len__()
# ============================================================

class Dataset:

    def __init__(self, data):

        self.data = data

    def __len__(self):

        return len(self.data)


dataset = Dataset(
    [10, 20, 30, 40]
)

print(len(dataset))


# ============================================================
# 29. __getitem__()
# ============================================================
#
# Allows object[index].
#
# ============================================================

class Dataset:

    def __init__(self, data):

        self.data = data

    def __getitem__(self, index):

        return self.data[index]


dataset = Dataset(
    [10, 20, 30, 40]
)

print(dataset[0])
print(dataset[2])


# ============================================================
# 30. ENCAPSULATION
# ============================================================
#
# Encapsulation means controlling how internal
# data is accessed or modified.
#
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self._balance = balance

    def get_balance(self):

        return self._balance


account = BankAccount(1000)

print(
    account.get_balance()
)


# ============================================================
# 31. PRIVATE-STYLE ATTRIBUTE
# ============================================================
#
# __name uses name mangling.
#
# ============================================================

class BankAccount:

    def __init__(self, balance):

        self.__balance = balance

    def get_balance(self):

        return self.__balance


account = BankAccount(1000)

print(
    account.get_balance()
)


# ============================================================
# 32. PROPERTY
# ============================================================
#
# @property allows a method to behave like an attribute.
#
# ============================================================

class Employee:

    def __init__(self, salary):

        self._salary = salary

    @property
    def salary(self):

        return self._salary


employee = Employee(70000)

print(employee.salary)


# ============================================================
# 33. PROPERTY WITH SETTER
# ============================================================

class Employee:

    def __init__(self, salary):

        self._salary = salary

    @property
    def salary(self):

        return self._salary

    @salary.setter
    def salary(self, value):

        if value >= 0:

            self._salary = value

        else:

            raise ValueError(
                "Salary cannot be negative."
            )


employee = Employee(70000)

employee.salary = 80000

print(employee.salary)


# ============================================================
# 34. INHERITANCE
# ============================================================
#
# A child class can inherit from a parent class.
#
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name

    def introduce(self):

        return f"My name is {self.name}."


class Student(Person):

    pass


student = Student("Yasin")

print(student.name)
print(student.introduce())


# ============================================================
# 35. CHILD CLASS WITH ADDITIONAL ATTRIBUTE
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name


class Student(Person):

    def __init__(
        self,
        name,
        field
    ):

        super().__init__(name)

        self.field = field


student = Student(
    "Yasin",
    "Machine Learning"
)

print(student.name)
print(student.field)


# ============================================================
# 36. super()
# ============================================================
#
# super() accesses functionality from the parent class.
#
# ============================================================

class Person:

    def __init__(self, name):

        self.name = name


class Researcher(Person):

    def __init__(
        self,
        name,
        field
    ):

        super().__init__(name)

        self.field = field


researcher = Researcher(
    "Yasin",
    "Artificial Intelligence"
)

print(researcher.name)
print(researcher.field)


# ============================================================
# 37. METHOD OVERRIDING
# ============================================================

class Animal:

    def sound(self):

        return "Some sound"


class Dog(Animal):

    def sound(self):

        return "Bark"


class Cat(Animal):

    def sound(self):

        return "Meow"


dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound())


# ============================================================
# 38. POLYMORPHISM
# ============================================================
#
# Different objects can respond to the same method name
# in different ways.
#
# ============================================================

animals = [
    Dog(),
    Cat()
]

for animal in animals:

    print(animal.sound())


# ============================================================
# 39. POLYMORPHISM — DATA ANALYTICS EXAMPLE
# ============================================================

class CSVData:

    def load(self):

        return "Loading CSV data"


class SQLData:

    def load(self):

        return "Loading SQL data"


class APIData:

    def load(self):

        return "Loading API data"


data_sources = [
    CSVData(),
    SQLData(),
    APIData()
]

for source in data_sources:

    print(source.load())


# ============================================================
# 40. ABSTRACT BASE CLASS
# ============================================================
#
# An abstract class defines a common interface.
#
# ============================================================

from abc import ABC, abstractmethod


class DataSource(ABC):

    @abstractmethod
    def load(self):

        pass


class CSVData(DataSource):

    def load(self):

        return "CSV data loaded"


class SQLData(DataSource):

    def load(self):

        return "SQL data loaded"


csv = CSVData()
sql = SQLData()

print(csv.load())
print(sql.load())


# ============================================================
# 41. CLASS METHOD
# ============================================================
#
# @classmethod receives cls instead of self.
#
# ============================================================

class Employee:

    company = "AI Lab"

    def __init__(self, name):

        self.name = name

    @classmethod
    def change_company(
        cls,
        new_company
    ):

        cls.company = new_company


print(Employee.company)

Employee.change_company(
    "Data Science Lab"
)

print(Employee.company)


# ============================================================
# 42. STATIC METHOD
# ============================================================
#
# A static method does not need self or cls.
#
# ============================================================

class Calculator:

    @staticmethod
    def add(a, b):

        return a + b


print(
    Calculator.add(
        10,
        20
    )
)


# ============================================================
# 43. CLASS METHOD VS STATIC METHOD
# ============================================================

class Example:

    class_variable = 100

    def __init__(self, value):

        self.value = value

    @classmethod
    def show_class_variable(cls):

        return cls.class_variable

    @staticmethod
    def multiply(a, b):

        return a * b


example = Example(50)

print(
    example.show_class_variable()
)

print(
    example.multiply(
        10,
        20
    )
)


# ============================================================
# 44. COMPOSITION
# ============================================================
#
# One class can contain an object of another class.
#
# ============================================================

class Engine:

    def start(self):

        return "Engine started"


class Car:

    def __init__(self):

        self.engine = Engine()

    def start(self):

        return self.engine.start()


car = Car()

print(car.start())


# ============================================================
# 45. DATA ANALYTICS — DATASET CLASS
# ============================================================

class Dataset:

    def __init__(self, name, data):

        self.name = name
        self.data = data

    def count(self):

        return len(self.data)

    def mean(self):

        return (
            sum(self.data)
            / len(self.data)
        )

    def minimum(self):

        return min(self.data)

    def maximum(self):

        return max(self.data)


dataset = Dataset(
    "Sales",
    [100, 200, 300, 400, 500]
)

print("Dataset:", dataset.name)
print("Count:", dataset.count())
print("Mean:", dataset.mean())
print("Minimum:", dataset.minimum())
print("Maximum:", dataset.maximum())


# ============================================================
# 46. DATA ANALYTICS — DATA CLEANING CLASS
# ============================================================

class DataCleaner:

    def __init__(self, data):

        self.data = data

    def remove_missing(self):

        self.data = [
            value
            for value in self.data
            if value is not None
        ]

        return self.data

    def remove_duplicates(self):

        self.data = list(
            set(self.data)
        )

        return self.data


data = [
    10,
    20,
    None,
    20,
    30,
    None
]

cleaner = DataCleaner(data)

print(
    cleaner.remove_missing()
)

print(
    cleaner.remove_duplicates()
)


# ============================================================
# 47. DATA ANALYTICS — DATASET SUMMARY CLASS
# ============================================================

class DatasetSummary:

    def __init__(self, values):

        self.values = values

    def summary(self):

        return {

            "count": len(self.values),

            "mean": (
                sum(self.values)
                / len(self.values)
            ),

            "minimum": min(self.values),

            "maximum": max(self.values)
        }


summary = DatasetSummary(
    [10, 20, 30, 40, 50]
)

print(
    summary.summary()
)


# ============================================================
# 48. MACHINE LEARNING MODEL CLASS
# ============================================================

class SimpleModel:

    def __init__(self, name):

        self.name = name
        self.is_trained = False

    def train(self, X, y):

        print(
            "Training",
            self.name
        )

        self.is_trained = True

    def predict(self, X):

        if not self.is_trained:

            raise ValueError(
                "Model has not been trained."
            )

        return [
            0
            for _ in X
        ]


model = SimpleModel(
    "My ML Model"
)

X_train = [
    [1, 2],
    [3, 4],
    [5, 6]
]

y_train = [
    0,
    1,
    1
]

model.train(
    X_train,
    y_train
)

predictions = model.predict(
    [
        [7, 8],
        [9, 10]
    ]
)

print(predictions)


# ============================================================
# 49. MACHINE LEARNING — MODEL INHERITANCE
# ============================================================

class Model:

    def train(self, X, y):

        raise NotImplementedError


class ClassificationModel(Model):

    def train(self, X, y):

        return "Classification model trained"


class RegressionModel(Model):

    def train(self, X, y):

        return "Regression model trained"


classification = ClassificationModel()
regression = RegressionModel()

print(
    classification.train(
        X_train,
        y_train
    )
)

print(
    regression.train(
        X_train,
        y_train
    )
)


# ============================================================
# 50. OOP DATA ANALYTICS PROJECT STRUCTURE
# ============================================================

class DataPipeline:

    def __init__(self, data):

        self.data = data

    def clean(self):

        # Cleaning operations
        return self

    def transform(self):

        # Transformation operations
        return self

    def analyze(self):

        # Analysis operations
        return self

    def report(self):

        return "Analysis completed"


pipeline = DataPipeline(
    [10, 20, 30, 40]
)

result = (
    pipeline
    .clean()
    .transform()
    .analyze()
    .report()
)

print(result)
```
