```python
# ============================================================
# 41. IQR — INTERQUARTILE RANGE
# ============================================================

# Calculate the 25th percentile (Q1) of the 'salary' column
Q1 = df["salary"].quantile(0.25)  # 25th percentile
# Calculate the 75th percentile (Q3) of the 'salary' column
Q3 = df["salary"].quantile(0.75)  # 75th percentile

# Calculate the Interquartile Range (IQR) by subtracting Q1 from Q3
IQR = Q3 - Q1  # interquartile range

# Print the calculated values of Q1, Q3, and IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# ============================================================
# 42. OUTLIER DETECTION USING IQR
# ============================================================

# Calculate the lower and upper bounds for outlier detection
# using the Interquartile Range (IQR) method
lower_limit = Q1 - 1.5 * IQR  # lower bound for outliers
upper_limit = Q3 + 1.5 * IQR  # upper bound for outliers

# Print the calculated lower and upper bounds
print("Lower:", lower_limit)
print("Upper:", upper_limit)

# Identify outliers in the dataset
# An outlier is any data point that falls below the lower limit
# or above the upper limit
# Identify outliers in the salary column
# These are the data points that fall below the lower limit or above the upper limit
outliers = df[
    (df["salary"] < lower_limit) |  # Filter for salaries below the lower limit
    (df["salary"] > upper_limit)   # Filter for salaries above the upper limit
]  # identify outliers

# Print the outliers
print("outliers: ", outliers)

```

```python
# ============================================================
# 55. BASIC BUSINESS QUESTION
# ============================================================

# Question:
# Which department has the highest average salary?

# Group the data by department and calculate the mean salary for each department
# Then sort the results in descending order to find the department with the highest average salary
department_salary = (
    df.groupby("Department")["salary"]  # Group the DataFrame by the "Department" column and select the "salary" column
    .mean()  # Calculate the mean salary for each department
    .sort_values(ascending=False)  # Sort the results in descending order
)  # average salary by department, sorted descending

# Print the series showing each department and its average salary
print(department_salary)

# Print the department with the highest average salary
# The index[0] contains the department name with the highest average salary since it's sorted in descending order
print(
    "Highest average salary department:",
    department_salary.index[0]  # department with highest average salary
)

```

```python
# ============================================================
# 57. ANALYTICAL INTERPRETATION
# ============================================================

# Check if correlation is positive
if correlation > 0:
    print(
        "There is a positive relationship "
        "between experience and salary."
    )  # positive correlation

# Check if correlation is negative
elif correlation < 0:
    print(
        "There is a negative relationship "
        "between experience and salary."
    )  # negative correlation

# If correlation is zero
else:
    print(
        "There is no linear relationship "
        "between experience and salary."
    )  # no correlation
```
